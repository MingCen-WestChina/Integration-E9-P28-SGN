## Upstream analysis definitions
The earlier analysis functions are retained in this file for provenance and explicit replay. Defining them performs no download or analysis; the archived CLI dispatcher is intentionally omitted. The current delivery cells do not rerun these expensive stages.


```python
"""Retained Fig4E upstream data-audit and analysis definitions.

The original acquisition, colocalization, orthology and source-aware RNA functions
are retained in this single notebook. Defining them performs no download or
analysis. The current delivery cells use audited upstream inputs and write only
the two specified compact result files and six independent figures.
"""

#1. =========== Shared inputs and verified reference snapshots
from __future__ import annotations

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")

# Cleanup must never interpret the project/current directory as a temporary target.
def _release_cleanup_target(path):
    candidate = Path(path).resolve()
    current = Path.cwd().resolve()
    if candidate == current or candidate in current.parents or candidate == Path(candidate.anchor):
        raise RuntimeError("Configure the owned temporary cleanup path before running this section.")
    return path


import gzip
import hashlib
import io
import json
import pickle
import re
import os
import sys
from pathlib import Path

for _thread_setting in ["OMP_NUM_THREADS", "OPENBLAS_NUM_THREADS", "MKL_NUM_THREADS", "NUMEXPR_NUM_THREADS", "NUMBA_NUM_THREADS"]:
    os.environ[_thread_setting] = "1"
sys.dont_write_bytecode = True

import h5py
import numpy as np
import pandas as pd
import urllib.request
import urllib.parse

# Set this storage path independently before running.
ROOT = ROOT
# Set this storage path independently before running.
TASK = ROOT
# Set this storage path independently before running.
os.environ["MPLCONFIGDIR"] = str(ROOT)
# Set this storage path independently before running.
os.environ["NUMBA_CACHE_DIR"] = str(ROOT)
# Set this storage path independently before running. Whole_CSS.h5ad.
WHOLE = ROOT
# Set this storage path independently before running. CSS_3000_Embryonic_Inte_adata.h5ad.
EMBRYO = ROOT
# Set this storage path independently before running. tree_results-1.pkl. Accepted tree input produced by the released Figure 3 analysis.
TREE = ROOT
# Set this storage path independently before running. Trajectory dot plot_results-2.h5.
HI = ROOT
# Set this storage path independently before running. Deve_regulon_dynamics.h5. Accepted primary result produced by fig4A_regulon_dynamics_analysis.py.
REGULONS = ROOT
# Set this storage path independently before running. 5477fbe6.h5ad.
GEAR = ROOT
# Set this storage path independently before running. gencode.vM25.primary_assembly.annotation.gtf.gz.
GTF = ROOT
EXPECTED_HASHES = {
    WHOLE: "57217aabefc823bf64fa9f6db5e9cd5c2314d7164f3a758d7458650bd36c24dc",
    EMBRYO: "8625e64dfb80313661f2258c1363e63ebe3e6f5446c4e2da3c920e46137e1e2c",
    TREE: "3b0a882539816def5e8cd17f6b445bd75e617ea8ecc3687ec5bf65bd6e56c0a5",
}
HOMOLOG_URL = "https://raw.githubusercontent.com/martinjzhang/scDRS/master/scdrs/data/mouse_human_homologs.txt"
HOMOLOG_SHA256 = "2d0ef4953bc1654f3835dba02c15c836ffe9f2162631069a766d898959016fb3"
COLOC_URL = "https://media.springernature.com/original/springer-static/esm/art%3A10.1038%2Fs42003-022-03408-7/MediaObjects/42003_2022_3408_MOESM4_ESM.xlsx"
COLOC_SHA256 = "d8e771b76a86d13726dc42b53c01378b35e9ca7f67ee48ad3a47eecfba2e98cf"
NEURAL_TISSUES = {"Spinal": ("Brain_Spinal_cord_cervical_c-1", 126),
                  "Tibial": ("Nerve_Tibial", 532)}


def digest(path):
    value = hashlib.sha256()
    with path.open("rb") as stream:
        for block in iter(lambda: stream.read(8 * 1024 * 1024), b""):
            value.update(block)
    return value.hexdigest()


def vector(node):
    if isinstance(node, h5py.Dataset):
        return node.asstr()[()] if node.dtype.kind in "OSU" else node[()]
    encoding = node.attrs.get("encoding-type", "")
    if isinstance(encoding, bytes):
        encoding = encoding.decode()
    if encoding == "categorical":
        return pd.Categorical.from_codes(node["codes"][()], vector(node["categories"]))
    if "values" in node and "mask" in node:
        values = np.asarray(vector(node["values"]), dtype=object)
        values[node["mask"][()]] = None
        return values
    raise ValueError(f"Unsupported vector encoding: {node.name}: {encoding}")


def index(group):
    key = group.attrs.get("_index", "_index")
    if isinstance(key, bytes):
        key = key.decode()
    return pd.Index(vector(group[key]).astype(str))


def columns(group, names):
    return pd.DataFrame({name: vector(group[name]) for name in names if name in group}, index=index(group))


def standard_stage(value):
    text = str(value)
    return text + ".5" if re.fullmatch(r"E\d+", text) else text


def fetch_verified(url, expected):
    cached = {
        # Set this storage path independently before running. mouse_human_homologs.txt.
        HOMOLOG_URL: ROOT,
        # Set this storage path independently before running.
        COLOC_URL: ROOT,
    }.get(url)
    if cached is not None and cached.is_file():
        content = cached.read_bytes()
    else:
        with urllib.request.urlopen(url, timeout=60) as response:
            content = response.read()
    actual = hashlib.sha256(content).hexdigest()
    if actual != expected:
        raise ValueError(f"Public reference changed: {url}: {actual}")
    return content


def read_reference_sheet(content, sheet_name, header=3):
    """Read reference XLSX values with stdlib XML; preserve merged-cell blanks."""
    import posixpath
    import zipfile
    import xml.etree.ElementTree as ET
    ns = "{http://schemas.openxmlformats.org/spreadsheetml/2006/main}"
    relationship = "{http://schemas.openxmlformats.org/officeDocument/2006/relationships}id"
    with zipfile.ZipFile(io.BytesIO(content)) as archive:
        shared = []
        if "xl/sharedStrings.xml" in archive.namelist():
            shared = ["".join(node.text or "" for node in item.iter(ns + "t")) for item in ET.fromstring(archive.read("xl/sharedStrings.xml"))]
        workbook = ET.fromstring(archive.read("xl/workbook.xml"))
        sheets = [node for node in workbook.iter(ns + "sheet") if node.attrib["name"] == sheet_name]
        if len(sheets) != 1:
            raise ValueError("Reference worksheet is missing or ambiguous")
        links = {node.attrib["Id"]: node.attrib["Target"] for node in ET.fromstring(archive.read("xl/_rels/workbook.xml.rels"))}
        target = links[sheets[0].attrib[relationship]]
        target = target.lstrip("/") if target.startswith("/") else posixpath.normpath("xl/" + target)
        rows = {}
        for row in ET.fromstring(archive.read(target)).iter(ns + "row"):
            values = {}
            for cell in row.findall(ns + "c"):
                letters = re.match(r"[A-Z]+", cell.attrib["r"]).group()
                column = 0
                for letter in letters:
                    column = column * 26 + ord(letter) - 64
                kind = cell.attrib.get("t", "n")
                node = cell.find(ns + "v")
                value = np.nan
                if kind == "inlineStr":
                    value = "".join(part.text or "" for part in cell.iter(ns + "t"))
                elif node is not None and node.text is not None:
                    value = shared[int(node.text)] if kind == "s" else (node.text if kind in ["str", "e"] else float(node.text))
                if isinstance(value, str) and value in {"", "NA", "N/A", "#N/A", "nan", "NaN", "NULL", "null"}:
                    value = np.nan
                values[column - 1] = value
            rows[int(row.attrib["r"]) - 1] = values
    heading = rows[header]
    width = max(heading) + 1
    columns = [heading.get(i, f"Unnamed: {i}") for i in range(width)]
    return pd.DataFrame([[rows.get(r, {}).get(i, np.nan) for i in range(width)] for r in range(header + 1, max(rows) + 1)], columns=columns)


#2. =========== Preserve accepted identities and tree correspondence
def audit_fixed_inputs():
    for path, expected in EXPECTED_HASHES.items():
        if path != TREE and digest(path) != expected:
            raise ValueError(f"Accepted input fingerprint changed: {path}")
    with h5py.File(WHOLE, "r") as store:
        obs = columns(store["obs"], ["input_stage", "source_stage_celltype", "library_id", "study_id", "original_cell_id", "source_cell_id", "cell_id_global", "cell_id_unique"])
        var = columns(store["var"], ["gene_symbol", "canonical_gene_id", "target_id"])
        assert len(obs) == 8340 and len(var) == 9174 and obs.index.is_unique and var.index.is_unique
        layers = list(store["layers"])
        counts_dtype = str(store["layers/counts/data"].dtype)
    with TREE.open("rb") as stream:
        tree = pickle.load(stream)
    tree_container_hash = digest(TREE)
    fixed_core_hash = "3d92516cebc6e70093163b945667d2847be5037c1a4ff01812f4fafae58cecd5"
    if tree_container_hash != EXPECTED_HASHES[TREE]:
        import ast
        # Set this storage path independently before running. 04_Trajectory_dot_plot-2.ipynb. Use fig3_segment_gene_analysis.ipynb; original numeric cell positions are retained.
        notebook_path = ROOT
        notebook = json.loads(notebook_path.read_text(encoding="utf-8"))
        functions = []
        for cell in notebook["cells"]:
            source = "".join(cell.get("source", []))
            if "def fixed_tree_core_hash" in source:
                functions.extend(node for node in ast.parse(source).body if isinstance(node, ast.FunctionDef)
                    and node.name in ["canonical_tree_value", "fixed_tree_core_hash"])
        assert len(functions) == 2
        namespace = {"np": np, "pd": pd, "hashlib": hashlib, "json": json,
                     "TREE_CORE_SCHEMA": "CVG-nine-natural-segments-core-v1"}
        exec(compile(ast.Module(body=functions, type_ignores=[]), str(notebook_path), "exec"), namespace)
        if namespace["fixed_tree_core_hash"](tree) != fixed_core_hash:
            raise ValueError("Accepted numeric tree core or annotations changed")
        with h5py.File(HI, "r") as store:
            assert json.loads(store["meta/input"].asstr()[()])["tree_core_sha256"] == fixed_core_hash
    accepted = pd.Index(tree["obs"].index.astype(str))
    with h5py.File(EMBRYO, "r") as store:
        embryo = index(store["obs"])
    with h5py.File(HI, "r") as store:
        hi = index(store["cells"])
    with h5py.File(REGULONS, "r") as store:
        regulons = pd.Index(vector(store["cells"]).astype(str))
    assert len(accepted) == 7355
    assert accepted.equals(embryo) and accepted.equals(hi) and accepted.equals(regulons)
    assert accepted.equals(obs.index[:7355])
    obs["stage"] = obs.input_stage.astype(str).map(standard_stage)
    status = {
        "cells": len(obs), "genes": len(var), "embryo_cells": len(accepted),
        "accepted_cell_order_equal": True, "tree_sha256": tree_container_hash,
        "tree_analysis_container_sha256": EXPECTED_HASHES[TREE],
        "tree_container_compatibility": ("original_container" if tree_container_hash == EXPECTED_HASHES[TREE]
                                          else "updated_container_with_exact_frozen_numeric_core"),
        "tree_core_sha256": fixed_core_hash,
        "master_key": "obs_names", "counts_dtype": counts_dtype, "layers": layers,
        "stage_counts": obs.stage.value_counts().to_dict(),
        "observed_stage_state_groups": int(obs.groupby(["stage", "source_stage_celltype"], observed=True).ngroups),
        "missing_cell_id_global": int(obs.cell_id_global.isna().sum()),
        "missing_cell_id_unique": int(obs.cell_id_unique.isna().sum()),
    }
    return obs, var, status


#3. =========== Audit the local cochlear reference without assuming its scale
def audit_gear():
    sample_blocks = 100       # Dispersed contiguous blocks; avoids large fancy indexing.
    values_per_block = 1000   # Numerical diagnostic only; not a count-origin certificate.
    with h5py.File(GEAR, "r") as store:
        obs = columns(store["obs"], ["P8-12-20", "Age", "Round", "Sample name"])
        var = columns(store["var"], ["gene_symbol"])
        data = store["X/data"]
        starts = np.linspace(0, max(0, len(data) - values_per_block), sample_blocks, dtype=np.int64)
        values = np.concatenate([data[int(start):int(start) + values_per_block] for start in starts])
        valid = np.isfinite(values)
        result = {
            "cells": len(obs), "genes": len(var), "sparse_encoding": str(store["X"].attrs.get("encoding-type")),
            "stored_values": len(data), "dtype": str(data.dtype), "raw_present": "raw" in store,
            "layers": list(store["layers"]), "uns_keys": list(store["uns"]),
            "type_counts": obs["P8-12-20"].value_counts(dropna=False).to_dict(),
            "age_counts_recorded_not_resolved": obs.Age.value_counts(dropna=False).to_dict(),
            "numerical_sample_n": len(values), "sample_all_finite": bool(valid.all()),
            "sample_min": float(values.min()), "sample_max": float(values.max()),
            "sample_noninteger_fraction_atol_1e_3": float((~np.isclose(values, np.rint(values), atol=1e-3, rtol=0)).mean()),
            "expression_scale_status": "gEAR Raw Data entry: unnormalized Partek count-scale values; retain fractional assignments",
            "portal_metadata_url": "https://umgear.org/cgi/search_datasets.cgi?search_terms=Michalski&limit=50&page=1",
            "portal_share_id": "5477fbe6",
            "portal_uuid": "3fc9de44-8bdf-4b85-b52a-1868749d0f3c",
            "portal_title": "ScRNA-seq - P8, P12, P20 mouse cochlea (Michalski 2023) - Raw Data",
            "full_cochlea_status": "scRNA subset lacks outer hair cells; companion snRNA required",
            "age_status": "P8/P12 labels conflict with paper; no age-specific comparison",
            "source_doi": "10.1073/pnas.2221744120",
        }
    return var, result


#4. =========== Audit measured source feature axes, preserving missingness
def read_m25():
    symbol_by_id = {}
    with gzip.open(GTF, "rt", encoding="utf-8") as stream:
        for line in stream:
            if line.startswith("#"):
                continue
            fields = line.rstrip("\n").split("\t")
            if fields[2] == "gene":
                attrs = dict(re.findall(r'(\w+) "([^"]+)"', fields[8]))
                symbol_by_id[attrs["gene_id"].split(".")[0]] = attrs["gene_name"]
    assert len(symbol_by_id) == 55487
    return symbol_by_id


def audit_source_axes(symbol_by_id):
    h5_sources = {
        # Set this storage path independently before running. E9_GSE178931__GSM5401072.h5ad.
        "GSE178931": 'ROOT',
        # Set this storage path independently before running. E14_E15_E16_E17_E18_P3_unresolved_GSE165502__all_4992_raw_cells.h5ad.
        "GSE165502": 'ROOT',
        # Set this storage path independently before running. E14_E16_E18_P1_GSE195500__all_5441_local_cells.h5ad.
        "GSE195500": 'ROOT',
        # Set this storage path independently before running. P7_Local_P7_SET7__SET7.h5ad.
        "P7": 'ROOT',
    }
    axes, summary = {}, {}
    for name, relative in h5_sources.items():
        with h5py.File(ROOT / relative, "r") as store:
            var = columns(store["var"], ["gene_symbol", "canonical_gene_id", "target_id"])
        symbols = var.gene_symbol.astype(str).tolist() if "gene_symbol" in var else var.index.astype(str).tolist()
        stable = var.canonical_gene_id if "canonical_gene_id" in var else var.index.to_series()
        mapped = pd.Series(stable, index=var.index).astype(str).str.split(".").str[0].map(symbol_by_id)
        available = set(symbols) | set(mapped.dropna())
        axes[name] = available
        summary[name] = {"features": len(var), "axis_path": relative, "m25_stable_id_matches": int(mapped.notna().sum())}
    # Set this storage path independently before running. counts_unstranded.tsv.gz.
    path = ROOT
    with gzip.open(path, "rt", encoding="utf-8") as stream:
        header = stream.readline().rstrip("\n").split("\t")
        assert header[0] == "cell_id_global"
        gene_ids = [value.split(".")[0] for value in header[1:]]
        cell_ids = [line.split("\t", 1)[0] for line in stream]
    assert len(gene_ids) == 55487 and len(set(gene_ids)) == 55487
    assert len(cell_ids) == 73 and len(set(cell_ids)) == 73
    axes["E10.5"] = {symbol_by_id[g] for g in gene_ids if g in symbol_by_id}
    summary["E10.5"] = {"features": len(gene_ids), "cells": len(cell_ids), "axis_path": str(path.relative_to(ROOT))}
    for sample, prefix in [("P28-1", "GSM6138444_p28_1"), ("P28-2", "GSM6138445_p28_2")]:
        relative = f"Data/P28_GSM6138444_GSM6138445/{prefix}_features.tsv.gz"
        features = pd.read_csv(ROOT / relative, sep="\t", header=None)
        stable = features[0].astype(str).str.split(".").str[0]
        axes[sample] = set(features[1].astype(str)) | set(stable.map(symbol_by_id).dropna())
        summary[sample] = {"features": len(features), "axis_path": relative, "m25_stable_id_matches": int(stable.isin(symbol_by_id).sum())}
    return axes, summary


#5. =========== Public ortholog and published colocalization coverage
def audit_genetic_coverage(common_var, gear_var, axes, summary):
    homolog_bytes = fetch_verified(HOMOLOG_URL, HOMOLOG_SHA256)
    homologs = pd.read_csv(io.BytesIO(homolog_bytes), sep="\t")
    assert len(homologs) == 16466 and all(homologs[col].is_unique for col in homologs)
    human_to_mouse = dict(zip(homologs.HUMAN_GENE_SYM, homologs.MOUSE_GENE_SYM))
    workbook = fetch_verified(COLOC_URL, COLOC_SHA256)
    coloc = read_reference_sheet(workbook, "Data 4", header=3)
    coloc = coloc.loc[coloc["Gene symbol"].notna()].copy()
    coloc["gene"] = coloc["Gene symbol"].astype(str).str.replace(r"\*+$", "", regex=True).str.strip()
    coloc["PPH4"] = pd.to_numeric(coloc.PPH4, errors="raise")
    assert len(coloc) == 55 and coloc.gene.nunique() == 19 and coloc.PPH4.ge(0.5).all()
    common = set(common_var.gene_symbol.dropna().astype(str))
    gear = set(gear_var.gene_symbol.dropna().astype(str))
    source_union = set().union(*axes.values())
    source_intersection = set.intersection(*[set(values) for values in axes.values()])
    records = []
    for gene, frame in coloc.groupby("gene", sort=True):
        mouse = human_to_mouse.get(gene)
        records.append({
            "human_gene": gene, "mouse_symbol_scdrs": mouse,
            "published_max_PPH4": float(frame.PPH4.max()), "tissue_count": int(frame.GTEx_tissue.nunique()),
            "in_common_axis": bool(mouse in common) if mouse else None,
            "in_gear_axis": bool(mouse in gear) if mouse else None,
            "in_native_sources": [name for name, available in axes.items() if mouse in available] if mouse else None,
        })
    for name, available in axes.items():
        summary[name]["scdrs_mouse_homologs_available"] = len(set(homologs.MOUSE_GENE_SYM) & available)
    report = {
        "ortholog_reference": {"url": HOMOLOG_URL, "sha256": HOMOLOG_SHA256, "pairs": len(homologs), "both_columns_unique": True,
            "limitation": "scDRS symbol reference; unmapped/renamed genes require stable-ID validation, not case conversion"},
        "published_coloc_reference": {"doi": "10.1038/s42003-022-03408-7", "url": COLOC_URL, "sha256": COLOC_SHA256,
            "sheet": "Data 4", "gene_tissue_rows": len(coloc), "genes": int(coloc.gene.nunique()), "author_PPH4_threshold": 0.5},
        "common_axis_homolog_pairs": len(set(homologs.MOUSE_GENE_SYM) & common),
        "gear_axis_homolog_pairs": len(set(homologs.MOUSE_GENE_SYM) & gear),
        "native_union_homolog_pairs_diagnostic_only": len(set(homologs.MOUSE_GENE_SYM) & source_union),
        "native_intersection_homolog_pairs_diagnostic_only": len(set(homologs.MOUSE_GENE_SYM) & source_intersection),
        "native_union_not_an_analysis_matrix": True, "candidate_coverage": records,
        "gene_set_policy": "Rank the full valid MAGMA Entrez universe before HGNC and RNA mapping; 19 published coloc genes are a separate candidate set",
    }
    # Set this storage path independently before running. gene-result-QC.json.
    qc_path = ROOT
    if qc_path.exists():
        qc = json.loads(qc_path.read_text(encoding="utf-8"))
        report["GWAS_gene_statistics"] = {key: qc[key] for key in ["gene_count", "original_rank_n", "weighted_gs_n", "bonferroni_genes"]}
    return report


#6. =========== Execute only this audit and emit reviewable diagnostics
def run_audit():
    obs, common_var, fixed = audit_fixed_inputs()
    gear_var, gear = audit_gear()
    axes, source_summary = audit_source_axes(read_m25())
    genetics = audit_genetic_coverage(common_var, gear_var, axes, source_summary)
    result = {"stage": "input_and_gene_coverage_audit", "fixed_inputs": fixed, "gear": gear,
        "source_axes": source_summary, "genetics": genetics,
        "counts_extraction_routes": {
            "GSE195500": "Use original CSV: its first header field is a barcode; current QC misses 14 accepted P1 cells",
            "P28": "Use original MTX/barcodes/features: all 284 accepted cells match; current QC is not the accepted cohort",
            "P7": "Use canonical Ensembl IDs; four legacy symbols differ from accepted symbols",
            "postnatal": "Retain accepted Whole axis including 18 restored IDs; do not recompute intersection from a newer source object"},
        "input_files_modified": False, "analysis_scores_computed": False}
    print(json.dumps(result, ensure_ascii=True, indent=2, default=lambda x: int(x) if isinstance(x, np.integer) else str(x)))

#7. =========== Native RNA and frozen developmental evidence
def native_rna_tools():
    """Build the closed-over native RNA tools without running an analysis."""
    #1. =========== Fixed inputs and scratch outputs

    import ast
    import argparse
    import csv
    import gzip
    import hashlib
    import json
    import re
    from collections import defaultdict
    from pathlib import Path

    import h5py
    import numpy as np
    import pandas as pd
    from scipy import sparse

    # Set this storage path independently before running.
    TASK = ROOT
    # Set this storage path independently before running. Whole_CSS.h5ad.
    WHOLE = ROOT
    # Set this storage path independently before running. gencode.vM25.primary_assembly.annotation.gtf.gz.
    GTF = ROOT
    # Set this storage path independently before running. MRK_List1.rpt.gz.
    MGI = ROOT
    GTF_SHA256 = "b3ceeef170048fc86c5dead33ba7af61d71e50c9a19ca5b4b9784b830802f5cd"
    MGI_SHA256 = "714e4390ae85807d4640802ab7303ebc4040d8b4e5a83f2b47e0be49c64b762f"
    # Set this storage path independently before running.
    QC = ROOT
    SOURCES = {
        # Set this storage path independently before running. E9_GSE178931__GSM5401072.h5ad.
        "GSE178931": ROOT,
        # Set this storage path independently before running. E14_E16_E18_P1_GSE195500__all_5441_local_cells.h5ad.
        "GSE195500": ROOT,
        # Set this storage path independently before running. E14_E15_E16_E17_E18_P3_unresolved_GSE165502__all_4992_raw_cells.h5ad.
        "GSE165502": ROOT,
        # Set this storage path independently before running. P7_Local_P7_SET7__SET7.h5ad.
        "Local_P7_SET7": ROOT,
        # Set this storage path independently before running. counts_unstranded.tsv.gz.
        "GSE231513": ROOT,
        # Set this storage path independently before running. GSM6138444_p28_1_features.tsv.gz.
        "GSE202920": ROOT,
    }
    CANDIDATES = "Acadvl Ccdc68 Cldn7 Crip3 Ctdnep1 Cyp26c1 Dlg4 Eaf2 Lmo7 Mmab Nol12 Nucks1 Ptk2 Rab29 Sptbn1 Tcf19 Ythdf1".split()


    def require(condition, message):
        if not condition:
            raise ValueError(message)


    def file_sha256(path):
        with Path(path).open("rb") as stream:
            return hashlib.file_digest(stream, "sha256").hexdigest()


    def vector(node):
        if isinstance(node, h5py.Dataset):
            return node.asstr()[()] if node.dtype.kind in "OSU" else node[()]
        if "categories" in node and "codes" in node:
            return pd.Categorical.from_codes(node["codes"][()], vector(node["categories"]))
        if "values" in node and "mask" in node:
            values = np.asarray(vector(node["values"]), dtype=object)
            values[node["mask"][()]] = None
            return values
        raise ValueError(node.name)


    def frame(group):
        key = group.attrs.get("_index", "_index")
        if isinstance(key, bytes):
            key = key.decode()
        return pd.DataFrame({k: vector(group[k]) for k in group if k != key},
                            index=pd.Index(vector(group[key]).astype(str)))


    def load_functions(path, names, namespace):
        """Reuse existing pure identity functions without importing analysis runtimes."""
        tree = ast.parse(path.read_text(encoding="utf-8-sig"))
        selected = [node for node in tree.body if isinstance(node, ast.FunctionDef) and node.name in names]
        require(len(selected) == len(names), "Missing established identity function.")
        exec(compile(ast.Module(body=selected, type_ignores=[]), str(path), "exec"), namespace)


    #2. =========== Established M25 and MGI identity decisions
    def prepare_mapping():
        namespace = {**globals(), **native_rna_tools()}
        # Set this storage path independently before running. Integration-genes.py. Use the supplied fig4_Integration_genes.py.
        load_functions(ROOT,
                       {"read_reference", "read_alias_evidence", "map_features"}, namespace)
        namespace["GENES"] = namespace
        # Set this storage path independently before running. Integration-axis.py. Use the supplied fig4_Integration_axis.py.
        load_functions(ROOT, {"resolve_mapping"}, namespace)
        reference, symbols = namespace["read_reference"]()
        with h5py.File(WHOLE, "r") as handle:
            obs, var = frame(handle["obs"]), frame(handle["var"])
            protected = set(var.index)
            restored = set(vector(handle["uns/restored_gene_mapping/restored_ids"]))
        axes = {}
        for study, path in SOURCES.items():
            if path.suffix == ".h5ad":
                with h5py.File(path, "r") as handle:
                    source_var = frame(handle["var"])
                names = np.asarray(source_var.index.astype(str))
                source_symbols = source_var.gene_symbol.astype(str).to_numpy() if "gene_symbol" in source_var else names
                native_ids = source_var.source_feature_id.astype(str).to_numpy() if study == "Local_P7_SET7" else None
                flags = np.zeros(len(names), dtype=bool)
                for key in ("mt_feature", "ERCC_feature", "reporter_or_vector_feature", "technical_feature"):
                    flags |= source_var[key].to_numpy(dtype=bool)
            elif study == "GSE231513":
                with gzip.open(path, "rt", encoding="utf-8") as stream:
                    names = np.asarray(next(stream).rstrip("\n").split("\t")[1:])
                native_ids = names
                source_symbols = np.asarray([reference.at[name.split(".")[0], "target_symbol"] for name in names])
                flags = np.zeros(len(names), dtype=bool)
            else:
                features = pd.read_csv(path, sep="\t", header=None, names=["id", "symbol", "type"])
                names = features.id.to_numpy()
                native_ids, source_symbols = names, features.symbol.to_numpy()
                flags = features.type.ne("Gene Expression").to_numpy()
                other = path.with_name("GSM6138445_p28_2_features.tsv.gz")
                other_features = pd.read_csv(other, sep="\t", header=None, names=["id", "symbol", "type"])
                require(features.equals(other_features), "P28 source gene axes differ.")
            flags |= pd.Series(source_symbols).str.match(r"(?i)^mt-|^Hb[ab]-|^ERCC[-_]").to_numpy()
            axes[study] = dict(names=names, source_symbols=source_symbols, native_ids=native_ids, excluded=flags)
        queries = set().union(*(set(x["source_symbols"]) for k, x in axes.items() if k in {"GSE178931", "GSE195500", "GSE165502"}))
        aliases, mgi_counts = namespace["read_alias_evidence"](queries, reference)
        tables, accepted, summary = [], {}, {}
        for study, axis in axes.items():
            if study in {"GSE178931", "GSE195500", "GSE165502"}:
                result = namespace["resolve_mapping"](axis["names"], "gene_symbol", reference, symbols, aliases)
                targets = result.accepted_gene_id.to_numpy().copy()
                decisions = result.decision.to_numpy().copy()
                route = "established_exact_M25_symbol_plus_MGI_identity_review"
            elif study == "GSE231513":
                result = namespace["resolve_mapping"](axis["names"], "ensembl_gene_id_versioned", reference, symbols, {})
                targets = result.accepted_gene_id.to_numpy().copy()
                decisions = result.decision.to_numpy().copy()
                route = "exact_M25_Ensembl_ID_and_gene_model_version"
            else:
                raw = np.asarray(axis["native_ids"], dtype=str)
                valid = np.asarray([bool(re.fullmatch(r"ENSMUSG\d{11}", name)) and name in reference.index for name in raw])
                excluded = np.asarray([bool(reference.at[name, "target_excluded"]) if ok else False for name, ok in zip(raw, valid)])
                targets = np.where(valid & ~excluded, raw, "").astype(object)
                decisions = np.where(valid & ~excluded, "accepted_source_bound_Ensembl_stable_ID_version_unprovided", "excluded_native_ID_not_in_M25_or_reference_class").astype(object)
                duplicates = pd.Series(targets).ne("") & pd.Series(targets).duplicated(keep=False)
                targets[duplicates.to_numpy()] = ""
                decisions[duplicates.to_numpy()] = "excluded_accepted_target_collision"
                route = "native_feature_bound_Ensembl_stable_ID; source_model_version_not_provided"
            decisions[axis["excluded"]] = "excluded_source_technical_mt_or_Hb_feature"
            targets[axis["excluded"]] = ""
            good = targets != ""
            require(pd.Index(targets[good]).is_unique, f"Accepted targets collide: {study}")
            accepted[study] = set(targets[good])
            table = pd.DataFrame({"study": study, "source_column": np.arange(len(targets)),
                                  "source_feature": axis["names"], "source_symbol": axis["source_symbols"],
                                  "target_id": targets, "decision": decisions})
            tables.append(table)
            summary[study] = dict(source_features=len(targets), accepted_unique=int(good.sum()),
                                  protected9174_covered=len(accepted[study] & protected),
                                  route=route, decision_counts=table.decision.value_counts().to_dict())
            print(study, summary[study]["source_features"], summary[study]["accepted_unique"],
                  "protected", summary[study]["protected9174_covered"], flush=True)
        common = set.intersection(*accepted.values())
        missing = protected - common
        report = dict(schema="Fig4_native_RNA_mapping_v1", sources=summary, reference="GENCODE M25 / Ensembl100 + fixed MGI snapshot",
                      reference_sha256=file_sha256(GTF), MGI_sha256=file_sha256(MGI), MGI_status_counts=mgi_counts,
                      native_common=len(common), common_beyond9174=len(common - protected), protected_missing=sorted(missing),
                      restored18_preserved=restored <= common, union=len(set.union(*accepted.values())),
                      filtering="Exclude technical, mitochondrial and Hb source/reference classes; do not inherit HVG or glial-marker exclusion",
                      no_alias_only_acceptance=True, no_target_aggregation=True,
                      stable_ID_version_policy="P7/P28 original feature-bound stable IDs are accepted; gene model versions were not supplied",
                      approved_cells=8340, current_Whole_sha256=file_sha256(WHOLE))
        TASK.mkdir(parents=True, exist_ok=True)
        merged = pd.concat(tables, ignore_index=True)
        # Set this storage path independently before running. native_mapping.tsv.gz.
        merged.to_csv(ROOT, sep="\t", index=False)
        # Set this storage path independently before running. mapping.json.
        (ROOT).write_text(json.dumps(report, indent=2, ensure_ascii=False), encoding="utf-8")
        pd.DataFrame({"gene_id": sorted(common), "gene_symbol": [reference.at[g, "target_symbol"] for g in sorted(common)],
                      # Set this storage path independently before running. common_genes.tsv.
                      "in_protected9174": [g in protected for g in sorted(common)]}).to_csv(ROOT, sep="\t", index=False)
        candidate_rows = []
        for name in CANDIDATES:
            candidates = symbols.get(name, ())
            require(len(candidates) == 1, f"Candidate M25 identity is not unique: {name}")
            gene = candidates[0]
            candidate_rows.append(dict(gene_symbol=name, gene_id=gene, in_common=gene in common,
                                       **{source: gene in available for source, available in accepted.items()}))
        # Set this storage path independently before running. candidate_measured_mask.tsv.
        pd.DataFrame(candidate_rows).to_csv(ROOT, sep="\t", index=False)
        print(json.dumps({k: report[k] for k in ["native_common", "common_beyond9174", "restored18_preserved", "protected_missing", "union"]}), flush=True)
        require(not missing, "Strict native intersection omitted protected Whole identities; inspect evidence before extracting.")
        return report


    #3. =========== Extract only real source columns for accepted fixed cells
    def h5_rows(path, requested_ids, source_to_common, n_genes, endogenous_eligible=None, return_totals=False):
        with h5py.File(path, "r") as handle:
            obs = frame(handle["obs"])
            rows = obs.index.get_indexer(requested_ids)
            require(np.all(rows >= 0), f"Fixed cells absent from source: {path}")
            matrix = handle["X"]
            require(matrix.attrs["encoding-type"] == "csr_matrix", "Source matrix is not CSR.")
            ptr = matrix["indptr"][()]
            parts, columns, output_ptr, totals = [], [], [0], []
            for row in rows:
                start, end = int(ptr[row]), int(ptr[row + 1])
                raw_columns = matrix["indices"][start:end]
                mapped = source_to_common[raw_columns]
                keep = mapped >= 0
                raw_values = matrix["data"][start:end]
                if return_totals:
                    eligible = np.ones(len(raw_values), dtype=bool) if endogenous_eligible is None else endogenous_eligible[raw_columns]
                    totals.append(float(raw_values[eligible].sum(dtype=np.float64)))
                values = raw_values[keep]
                require(np.isfinite(values).all() and (values >= 0).all() and np.equal(values, np.rint(values)).all(), "Source counts are not nonnegative integers.")
                parts.append(values.astype(np.int32))
                columns.append(mapped[keep])
                output_ptr.append(output_ptr[-1] + int(keep.sum()))
        result = sparse.csr_matrix((np.concatenate(parts), np.concatenate(columns), np.asarray(output_ptr)),
                                   shape=(len(rows), n_genes))
        result.sort_indices()
        return (result, np.asarray(totals, dtype=np.float64)) if return_totals else result


    def csv195500(obs, selected, mapping, n_genes):
        # Set this storage path independently before running. GSE195500_SGN_Combined_counts.csv.gz.
        path = ROOT
        require(obs.iloc[selected].original_cell_id.is_unique, "195500 original cell IDs collide.")
        values = np.empty((len(selected), n_genes), dtype=np.int32)
        gene_to_common = dict(zip(mapping.source_feature, mapping.common_column))
        observed_features = []
        with gzip.open(path, "rt", encoding="utf-8") as stream:
            cells = pd.Index(next(stream).rstrip("\n").split(","))
            require(len(cells) == 5441 and cells.is_unique, "195500 header format changed.")
            cell_columns = cells.get_indexer(obs.iloc[selected].original_cell_id)
            require(np.all(cell_columns >= 0), "Accepted 195500 cell absent from original CSV.")
            assigned = np.zeros(n_genes, dtype=bool)
            for line in stream:
                name, raw = line.rstrip("\n").split(",", 1)
                name = name.strip('"')
                observed_features.append(name)
                destination = gene_to_common.get(name, -1)
                if destination < 0:
                    continue
                counts = np.fromstring(raw, sep=",", dtype=np.int64)
                require(len(counts) == len(cells) and (counts >= 0).all() and counts.max() <= np.iinfo(np.int32).max, "195500 count row invalid.")
                values[:, destination] = counts[cell_columns]
                require(not assigned[destination], "Repeated 195500 gene assignment.")
                assigned[destination] = True
        require(observed_features == mapping.source_feature.tolist(), "195500 original CSV differs from verified QC gene axis.")
        require(assigned.all(), "Unmeasured 195500 gene would enter common matrix.")
        return sparse.csr_matrix(values), path


    def mtx_p28(obs, selected, study_mapping, n_genes, gsm, endogenous_eligible=None, return_totals=False):
        # Set this storage path independently before running.
        directory = ROOT
        prefix = "GSM6138444_p28_1" if gsm == "GSM6138444" else "GSM6138445_p28_2"
        with gzip.open(directory / f"{prefix}_barcodes.tsv.gz", "rt") as stream:
            barcodes = pd.Index([line.strip() for line in stream])
        positions = barcodes.get_indexer(obs.iloc[selected].original_cell_id)
        require(np.all(positions >= 0) and len(set(positions)) == len(selected), "P28 accepted barcodes are missing or repeated.")
        cell_map = np.full(len(barcodes), -1, dtype=np.int64)
        cell_map[positions] = np.arange(len(selected))
        gene_map = study_mapping.common_column.to_numpy(dtype=np.int64)
        row_parts, col_parts, data_parts = [], [], []
        path = directory / f"{prefix}_matrix.mtx.gz"
        parsed = 0
        totals = np.zeros(len(selected), dtype=np.float64)
        with gzip.open(path, "rt") as stream:
            first = next(stream)
            require(first.strip() == "%%MatrixMarket matrix coordinate integer general", "P28 matrix contract changed.")
            line = next(stream)
            while line.startswith("%"):
                line = next(stream)
            n_features, n_cells, n_entries = map(int, line.split())
            require(n_features == len(gene_map) and n_cells == len(barcodes), "P28 matrix axes differ.")
            while lines := stream.readlines(16 * 1024 * 1024):
                triplets = np.fromstring("".join(lines), sep=" ", dtype=np.int64).reshape(-1, 3)
                parsed += len(triplets)
                mapped_cells = cell_map[triplets[:, 1] - 1]
                keep = mapped_cells >= 0
                if return_totals:
                    total_keep = keep.copy()
                    if endogenous_eligible is not None:
                        total_keep &= endogenous_eligible[triplets[:, 0] - 1]
                    totals += np.bincount(mapped_cells[total_keep], weights=triplets[total_keep, 2], minlength=len(selected))
                mapped_genes = gene_map[triplets[:, 0] - 1]
                keep &= mapped_genes >= 0
                chosen = triplets[keep]
                require((chosen[:, 2] >= 0).all() and chosen[:, 2].max(initial=0) <= np.iinfo(np.int32).max, "P28 counts invalid.")
                row_parts.append(mapped_cells[keep])
                col_parts.append(mapped_genes[keep])
                data_parts.append(chosen[:, 2].astype(np.int32))
        require(parsed == n_entries, "P28 matrix entries differ from header.")
        result = sparse.coo_matrix((np.concatenate(data_parts), (np.concatenate(row_parts), np.concatenate(col_parts))),
                                   shape=(len(selected), n_genes)).tocsr()
        return (result, path, totals) if return_totals else (result, path)


    #4. =========== Assemble and verify the expanded common matrix
    def prepare_counts():
        # Set this storage path independently before running. mapping.json.
        mapping_report = json.loads((ROOT).read_text(encoding="utf-8"))
        require(mapping_report["protected_missing"] == [], "Mapping audit did not preserve protected genes.")
        require(mapping_report["current_Whole_sha256"] == file_sha256(WHOLE), "Protected Whole file changed.")
        # Set this storage path independently before running. common_genes.tsv.
        genes = pd.read_csv(ROOT, sep="\t")
        gene_index = pd.Index(genes.gene_id)
        n_genes = len(genes)
        # Set this storage path independently before running. native_mapping.tsv.gz.
        mapping = pd.read_csv(ROOT, sep="\t", keep_default_na=False)
        mapping["common_column"] = gene_index.get_indexer(mapping.target_id)
        with h5py.File(WHOLE, "r") as handle:
            obs, var = frame(handle["obs"]), frame(handle["var"])
            whole = sparse.csr_matrix((handle["layers/counts/data"][()], handle["layers/counts/indices"][()],
                                       handle["layers/counts/indptr"][()]), shape=(len(obs), len(var)))
        matrices, row_indices, provenance = [], [], []
        for study in SOURCES:
            selected = np.flatnonzero(obs.study_id.astype(str).eq(study).to_numpy())
            table = mapping[mapping.study.eq(study)].sort_values("source_column")
            require((table.source_column.to_numpy() == np.arange(len(table))).all(), "Source mapping column order changed.")
            require(set(gene_index) <= set(table.target_id), f"Unmeasured genes in common axis: {study}")
            if study == "GSE178931":
                parts, order = [], []
                for library in obs.iloc[selected].library_id.astype(str).unique():
                    indices = selected[obs.iloc[selected].library_id.astype(str).eq(library).to_numpy()]
                    stage = obs.iloc[indices[0]].stage_standardized.split(".")[0]
                    path = QC / f"{stage}_GSE178931__{library}.h5ad"
                    with h5py.File(path, "r") as handle:
                        require(frame(handle["var"]).index.tolist() == table.source_feature.tolist(), "178931 library gene axes differ.")
                    part = h5_rows(path, obs.index[indices], table.common_column.to_numpy(dtype=np.int64), n_genes)
                    parts.append(part)
                    order.extend(indices.tolist())
                    provenance.append(dict(path=str(path.relative_to(ROOT)), sha256=file_sha256(path), cells=len(indices)))
                combined = sparse.vstack(parts, format="csr")
                rows = np.asarray(order)
            elif study in {"GSE165502", "Local_P7_SET7"}:
                path = SOURCES[study]
                combined = h5_rows(path, obs.index[selected], table.common_column.to_numpy(dtype=np.int64), n_genes)
                rows = selected
                provenance.append(dict(path=str(path.relative_to(ROOT)), sha256=file_sha256(path), cells=len(rows)))
            elif study == "GSE195500":
                combined, path = csv195500(obs, selected, table, n_genes)
                rows = selected
                provenance.append(dict(path=str(path.relative_to(ROOT)), sha256=file_sha256(path), cells=len(rows)))
            elif study == "GSE231513":
                path = SOURCES[study]
                raw = pd.read_csv(path, sep="\t", index_col=0, dtype={"cell_id_global": str})
                require(raw.columns.tolist() == table.source_feature.tolist(), "E10 source gene axis changed.")
                require(set(raw.index) == set(obs.index[selected]), "E10 fixed cells changed.")
                positions = table.loc[table.common_column.ge(0)].sort_values("common_column").source_column.to_numpy()
                combined = sparse.csr_matrix(raw.loc[obs.index[selected]].iloc[:, positions].to_numpy(dtype=np.int32))
                rows = selected
                provenance.append(dict(path=str(path.relative_to(ROOT)), sha256=file_sha256(path), cells=len(rows)))
            else:
                parts, order = [], []
                for library in obs.iloc[selected].library_id.astype(str).unique():
                    indices = selected[obs.iloc[selected].library_id.astype(str).eq(library).to_numpy()]
                    part, path = mtx_p28(obs, indices, table, n_genes, library)
                    parts.append(part)
                    order.extend(indices.tolist())
                    provenance.append(dict(path=str(path.relative_to(ROOT)), sha256=file_sha256(path), cells=len(indices)))
                combined = sparse.vstack(parts, format="csr")
                rows = np.asarray(order)
            common_columns = gene_index.get_indexer(var.index)
            difference = combined[:, common_columns].astype(np.int64) - whole[rows].astype(np.int64)
            difference.eliminate_zeros()
            require(difference.nnz == 0, f"Original 9174 counts differ from Whole in {study}: {difference.nnz} entries")
            require(combined.data.min(initial=0) >= 0, "Negative source counts.")
            matrices.append(combined)
            row_indices.extend(rows.tolist())
            print(study, "cells", len(rows), "nnz", combined.nnz, "protected9174_exact", True, flush=True)
        require(sorted(row_indices) == list(range(len(obs))), "Fixed cells were lost or repeated.")
        counts = sparse.vstack(matrices, format="csr")[np.argsort(row_indices)].tocsr()
        counts.sort_indices()
        require(counts.shape == (8340, n_genes) and counts.has_canonical_format, "Expanded matrix shape or canonical format invalid.")
        difference = counts[:, gene_index.get_indexer(var.index)].astype(np.int64) - whole.astype(np.int64)
        difference.eliminate_zeros()
        require(difference.nnz == 0, "Expanded common counts are not an exact extension of Whole counts.")
        report = dict(schema="Fig4_native_common_counts_v1", cells=8340, genes=n_genes, nnz=counts.nnz,
                      protected_genes=9174, extra_genes=n_genes - 9174, protected_counts_elementwise_equal=True,
                      protected_cell_ID_order_equal=True, counts="source integer raw counts", normalized=False,
                      missingness="All genes have one accepted native source column in every source; no missing feature zero imputation",
                      sources=provenance, mapping=mapping_report,
                      original_cell_ids_sha256_LF=hashlib.sha256("\n".join(obs.index).encode()).hexdigest(),
                      gene_ids_sha256_LF=hashlib.sha256("\n".join(gene_index).encode()).hexdigest())
        # Set this storage path independently before running. common_counts.h5.
        path = ROOT
        with h5py.File(path, "w") as handle:
            group = handle.create_group("counts")
            group.attrs["shape"] = counts.shape
            group.attrs["encoding-type"] = "csr_matrix"
            for name, data in [("data", counts.data), ("indices", counts.indices), ("indptr", counts.indptr)]:
                group.create_dataset(name, data=data, compression="gzip", shuffle=True)
            string = h5py.string_dtype("utf-8")
            handle.create_dataset("cell_ids", data=np.asarray(obs.index, dtype=object), dtype=string)
            handle.create_dataset("gene_ids", data=np.asarray(gene_index, dtype=object), dtype=string)
            handle.create_dataset("gene_symbols", data=genes.gene_symbol.to_numpy(dtype=object), dtype=string)
            handle.create_dataset("protected9174", data=genes.in_protected9174.to_numpy(dtype=bool))
            for name in ["study_id", "library_id", "stage_standardized"]:
                handle.create_dataset(name, data=obs[name].astype(str).to_numpy(dtype=object), dtype=string)
            handle.create_dataset("manifest_json", data=json.dumps(report), dtype=string)
        report["cache_sha256"] = file_sha256(path)
        report["cache_bytes"] = path.stat().st_size
        # Set this storage path independently before running. counts.json.
        (ROOT).write_text(json.dumps(report, indent=2), encoding="utf-8")
        print(json.dumps({key: report[key] for key in ["cells", "genes", "nnz", "protected_counts_elementwise_equal", "cache_bytes", "cache_sha256"]}), flush=True)
    return dict(locals(), ROOT=ROOT)

def prepare_rna_inputs():
    """Rebuild the strict native common axis and verified fixed-cell count cache."""
    tools = native_rna_tools()
    tools["prepare_mapping"]()
    tools["prepare_counts"]()

def prepare_genetic_candidates():
    """Prepare source-aware candidate expression and true-library summaries."""
    #1. =========== Reuse native readers and fixed source identity evidence

    import json
    import runpy
    from pathlib import Path

    import h5py
    import numpy as np
    import pandas as pd
    from scipy import sparse

    # Set this storage path independently before running.
    TASK = ROOT
    NATIVE = native_rna_tools()
    WHOLE = NATIVE["WHOLE"]
    frame, require, digest = NATIVE["frame"], NATIVE["require"], NATIVE["file_sha256"]
    CANDIDATES = NATIVE["CANDIDATES"]
    # Set this storage path independently before running. Trajectory dot plot_results-2.h5.
    HI = ROOT


    def endogenous_axis(path):
        with h5py.File(path, "r") as handle:
            var = frame(handle["var"])
        excluded = np.zeros(len(var), dtype=bool)
        for key in ["technical_feature", "ERCC_feature", "reporter_or_vector_feature", "external_spike_feature"]:
            if key in var:
                excluded |= var[key].to_numpy(dtype=bool)
        return ~excluded, var


    def read_h5_candidate_rows(path, obs, selected, table, extra_gene_ids):
        target = pd.Index(extra_gene_ids)
        eligible, source_var = endogenous_axis(path)
        require(source_var.index.tolist() == table.source_feature.tolist(), "Candidate source feature order changed.")
        columns = target.get_indexer(table.target_id)
        result, totals = NATIVE["h5_rows"](path, obs.index[selected], columns, len(target),
                                          endogenous_eligible=eligible, return_totals=True)
        return result.toarray().astype(np.float64), totals, int(eligible.sum())


    #2. =========== Resolve the complete source denominators before normalization
    def prepare_candidates():
        # Set this storage path independently before running. candidate_measured_mask.tsv.
        candidate_mask = pd.read_csv(ROOT, sep="\t")
        require(candidate_mask.gene_symbol.tolist() == CANDIDATES, "Candidate order changed.")
        genes = pd.Index(candidate_mask.gene_id)
        extra_positions = np.flatnonzero(~candidate_mask.in_common.to_numpy())
        common_positions = np.flatnonzero(candidate_mask.in_common.to_numpy())
        extra_genes = genes[extra_positions]
        # Set this storage path independently before running. native_mapping.tsv.gz.
        mapping = pd.read_csv(ROOT, sep="\t", keep_default_na=False)
        with h5py.File(WHOLE, "r") as handle:
            obs, whole_var = frame(handle["obs"]), frame(handle["var"])
            original = sparse.csr_matrix((handle["layers/counts/data"][()], handle["layers/counts/indices"][()],
                                         handle["layers/counts/indptr"][()]), shape=(len(obs), len(whole_var)))
        # Set this storage path independently before running. common_counts.h5.
        with h5py.File(ROOT, "r") as handle:
            require(np.array_equal(handle["cell_ids"].asstr()[()], obs.index.to_numpy()), "Common-cache cell order changed.")
            common_genes = pd.Index(handle["gene_ids"].asstr()[()])
            common = sparse.csr_matrix((handle["counts/data"][()], handle["counts/indices"][()],
                                       handle["counts/indptr"][()]), shape=tuple(handle["counts"].attrs["shape"]))
            counts = np.full((len(obs), len(genes)), np.nan, dtype=np.float64)
            counts[:, common_positions] = common[:, common_genes.get_indexer(genes[common_positions])].toarray()
        require(np.array_equal(counts[:, common_positions],
                               original[:, whole_var.index.get_indexer(genes[common_positions])].toarray()),
                "Nine shared candidate numerators differ from Whole.")
        measured = np.zeros(counts.shape, dtype=bool)
        for study in NATIVE["SOURCES"]:
            selected = obs.study_id.astype(str).eq(study).to_numpy()
            measured[selected] = candidate_mask[study].to_numpy(dtype=bool)
        denominator = np.full(len(obs), np.nan, dtype=np.float64)
        denominator_source = np.full(len(obs), "", dtype=object)
        with h5py.File(HI, "r") as handle:
            embryo_ids = pd.Index(handle["cells/_index"].asstr()[()])
            embryo_denominator = handle["cells/library_total"][()]
            require(np.array_equal(embryo_denominator, handle["normalization/library_total"][()]), "HI denominator archives differ.")
            hi_input = json.loads(handle["meta/input"].asstr()[()])
            require(hi_input["calculation"] == "log10(1 + common_axis_raw_count / full_available_source_endogenous_gene_counts * 1e6)", "HI normalization contract changed.")
        embryo_positions = obs.index.get_indexer(embryo_ids)
        require(np.array_equal(embryo_positions, np.arange(7355)), "HI and Whole embryo cell order differs.")
        denominator[embryo_positions] = embryo_denominator
        denominator_source[embryo_positions] = "Fig3HI audited full_available_source_endogenous_counts"
        source_reports = []

        # Native full-axis denominators retain mitochondrial/Hb expression where supplied.
        # Technical, reporter/vector and external spike features remain excluded.
        for study, path in NATIVE["SOURCES"].items():
            selected = np.flatnonzero(obs.study_id.astype(str).eq(study).to_numpy())
            table = mapping[mapping.study.eq(study)].sort_values("source_column").copy()
            table["common_column"] = pd.Index(extra_genes).get_indexer(table.target_id)
            report = dict(study=study, cells=len(selected), extra_candidates_measured=bool(study != "GSE195500"))
            if study == "GSE178931":
                checks = []
                for library in obs.iloc[selected].library_id.astype(str).unique():
                    rows = selected[obs.iloc[selected].library_id.astype(str).eq(library).to_numpy()]
                    stage = obs.iloc[rows[0]].stage_standardized.split(".")[0]
                    source = NATIVE["QC"] / f"{stage}_GSE178931__{library}.h5ad"
                    values, totals, n_endogenous = read_h5_candidate_rows(source, obs, rows, table, extra_genes)
                    counts[np.ix_(rows, extra_positions)] = values
                    require(np.array_equal(totals, denominator[rows]), "178931 full-source totals differ from HI.")
                    checks.append(dict(path=str(source.relative_to(ROOT)), sha256=digest(source),
                                       cells=len(rows), endogenous_features=n_endogenous, full_totals_equal_HI=True))
                report["native_checks"] = checks
            elif study in {"GSE165502", "Local_P7_SET7"}:
                values, totals, n_endogenous = read_h5_candidate_rows(path, obs, selected, table, extra_genes)
                counts[np.ix_(selected, extra_positions)] = values
                if study == "GSE165502":
                    require(np.array_equal(totals, denominator[selected]), "165502 full-source totals differ from HI.")
                else:
                    require(np.array_equal(totals, obs.iloc[selected].total_counts_endogenous.to_numpy(dtype=float)), "P7 full-source totals differ from saved endogenous totals.")
                    denominator[selected] = totals
                    denominator_source[selected] = "P7_QC_full29267_minus_EGFP2_technical"
                report.update(path=str(path.relative_to(ROOT)), sha256=digest(path), endogenous_features=n_endogenous,
                              totals_checked_against="Fig3HI" if study == "GSE165502" else "Whole_total_counts_endogenous", totals_equal=True)
            elif study == "GSE231513":
                raw = pd.read_csv(path, sep="\t", index_col=0)
                require(raw.columns.tolist() == table.source_feature.tolist(), "E10 native columns changed.")
                columns = table.loc[table.common_column.ge(0)].sort_values("common_column").source_column.to_numpy()
                counts[np.ix_(selected, extra_positions)] = raw.loc[obs.index[selected]].iloc[:, columns].to_numpy(dtype=float)
                totals = raw.loc[obs.index[selected]].sum(axis=1).to_numpy(dtype=float)
                require(np.array_equal(totals, denominator[selected]), "E10 assigned-pair totals differ from HI.")
                report.update(path=str(path.relative_to(ROOT)), sha256=digest(path), endogenous_features=55487,
                              totals_checked_against="Fig3HI embedded E10 assigned-pair totals", totals_equal=True)
            elif study == "GSE195500":
                p1 = selected[obs.iloc[selected].stage_standardized.astype(str).eq("P1").to_numpy()]
                # Set this storage path independently before running. P1_cluster_intergration.h5ad.
                stage_path = ROOT
                with h5py.File(stage_path, "r") as handle:
                    stage_obs = frame(handle["obs"])
                    provenance = handle["uns/source_provenance"]
                    definition = provenance["qc_metric_definitions/total_counts_source"].asstr()[()]
                    raw_hash = provenance["source_file_sha256/raw_counts"].asstr()[()]
                    require(definition == "Source count matrix before feature exclusion", "P1 saved denominator meaning is not verified.")
                    require(raw_hash == "fb70f0f47d74b4866f3f7a4376bf6eb6bd82c7e585854f10c486152a9ac1cdfe", "P1 original source fingerprint changed.")
                totals = stage_obs.loc[obs.index[p1]].total_counts_endogenous.to_numpy(dtype=float)
                require(np.array_equal(totals, stage_obs.loc[obs.index[p1]].total_counts_source.to_numpy(dtype=float)), "P1 source/endogenous saved totals differ.")
                denominator[p1] = totals
                denominator_source[p1] = "P1_saved_full_source_totals_bound_to_original195500_CSV_SHA256"
                with h5py.File(path, "r") as handle:
                    qc_obs = frame(handle["obs"])
                rows = obs.index[selected].intersection(qc_obs.index)
                positions = obs.index.get_indexer(rows)
                eligible, source_var = endogenous_axis(path)
                empty_map = np.full(len(source_var), -1, dtype=np.int64)
                _, qc_totals = NATIVE["h5_rows"](path, rows, empty_map, 0, endogenous_eligible=eligible, return_totals=True)
                require(np.array_equal(qc_totals, denominator[positions]), "195500 native totals differ from HI/P1 saved source denominators.")
                report.update(path=str(path.relative_to(ROOT)), sha256=digest(path), endogenous_features=int(eligible.sum()),
                              native_QC_totals_verified_cells=len(rows), P1_native_QC_matched_cells=int(obs.loc[rows].stage_standardized.astype(str).eq("P1").sum()),
                              P1_preserved_original_source_metadata_only_cells=int(len(p1)-obs.loc[rows].stage_standardized.astype(str).eq("P1").sum()),
                              P1_source_metadata_definition=definition, P1_original_CSV_sha256=raw_hash,
                              P1_stagewise_source=str(stage_path.relative_to(ROOT)), original_CSV_rescanned=False)
            else:
                checks = []
                for library in obs.iloc[selected].library_id.astype(str).unique():
                    rows = selected[obs.iloc[selected].library_id.astype(str).eq(library).to_numpy()]
                    qc = NATIVE["QC"] / f"P28_GSE202920__{library}.h5ad"
                    eligible, native_var = endogenous_axis(qc)
                    require(len(native_var) == len(table), "P28 denominator source axis length differs.")
                    values, source, totals = NATIVE["mtx_p28"](obs, rows, table, len(extra_genes), library,
                                                             endogenous_eligible=eligible, return_totals=True)
                    counts[np.ix_(rows, extra_positions)] = values.toarray()
                    require(np.array_equal(totals, obs.iloc[rows].total_counts_endogenous.to_numpy(dtype=float)), "P28 full-source MTX totals differ from saved endogenous totals.")
                    denominator[rows] = totals
                    denominator_source[rows] = "P28_full32285_native_MTX_endogenous_counts_duplicate_symbols_do_not_change_total"
                    checks.append(dict(path=str(source.relative_to(ROOT)), sha256=digest(source), cells=len(rows),
                                       endogenous_features=int(eligible.sum()), saved_total_equal=True))
                report["native_checks"] = checks
            source_reports.append(report)
            print(study, "candidate_counts_ready", "denominator_verified", True, flush=True)
        require(np.isfinite(denominator).all() and (denominator > 0).all(), "Invalid full-source denominators.")
        require(np.array_equal(np.isfinite(counts), measured), "Missing candidate features were converted to observed values.")
        require(np.equal(counts[measured], np.rint(counts[measured])).all() and (counts[measured] >= 0).all(), "Measured candidate counts are invalid.")
        detection = np.full_like(counts, np.nan)
        detection[measured] = (counts[measured] > 0).astype(float)
        cpm = counts / denominator[:, None] * 1e6
        expression = np.log10(1 + cpm)
        embryo_again = np.log10(1 + original[:7355, whole_var.index.get_indexer(genes[common_positions])].toarray()
                                / embryo_denominator[:, None] * 1e6)
        require(np.array_equal(expression[:7355, common_positions], embryo_again), "Nine embryo candidate expressions do not use the HI denominator exactly.")

        #3. =========== Summarize cells within each true library, then weight libraries equally
        obs["state"] = obs.source_stage_celltype.astype(str).str.split(" | ", n=1, regex=False).str[1]
        group_order = list(dict.fromkeys(obs.source_stage_celltype.astype(str)))
        require(len(group_order) == 36 and obs.state.notna().all(), "Expected fixed 36 stage-state groups changed.")
        library_rows = []
        grouped = obs.groupby(["source_stage_celltype", "stage_standardized", "state", "study_id", "library_id"], observed=True, sort=False)
        for key, rows in grouped.indices.items():
            group, stage, state, study, library = key
            for j, name in enumerate(CANDIDATES):
                valid = measured[rows, j]
                selected_rows = rows[valid]
                library_rows.append(dict(group=str(group), stage=str(stage), state=str(state), study_id=str(study), library_id=str(library),
                                         gene_symbol=name, gene_id=genes[j], n_cells=len(rows), n_measured_cells=int(valid.sum()),
                                         measured=bool(valid.any()),
                                         mean_raw_count=float(counts[selected_rows, j].mean()) if valid.any() else np.nan,
                                         detection_fraction=float(detection[selected_rows, j].mean()) if valid.any() else np.nan,
                                         mean_CPM=float(cpm[selected_rows, j].mean()) if valid.any() else np.nan,
                                         mean_log10_1p_CPM=float(expression[selected_rows, j].mean()) if valid.any() else np.nan,
                                         denominator_min=float(denominator[rows].min()), denominator_max=float(denominator[rows].max()),
                                         source_count_unit=";".join(sorted(set(obs.iloc[rows].count_unit_verified.astype(str))))))
        libraries = pd.DataFrame(library_rows)
        group_rows = []
        for group in group_order:
            for name in CANDIDATES:
                sub = libraries[libraries.group.eq(group) & libraries.gene_symbol.eq(name)]
                valid = sub[sub.measured]
                group_rows.append(dict(group=group, stage=sub.stage.iloc[0], state=sub.state.iloc[0], gene_symbol=name, gene_id=sub.gene_id.iloc[0],
                                       n_cells=int(sub.n_cells.sum()), n_measured_cells=int(sub.n_measured_cells.sum()),
                                       n_libraries=len(sub), n_measured_libraries=len(valid),
                                       min_measured_cells_per_library=int(valid.n_measured_cells.min()) if len(valid) else 0,
                                       sources=";".join(sorted(sub.study_id.unique())), measured_sources=";".join(sorted(valid.study_id.unique())),
                                       mean_log10_1p_CPM_equal_library=float(valid.mean_log10_1p_CPM.mean()) if len(valid) else np.nan,
                                       mean_CPM_equal_library=float(valid.mean_CPM.mean()) if len(valid) else np.nan,
                                       detection_fraction_equal_library=float(valid.detection_fraction.mean()) if len(valid) else np.nan,
                                       library_weighting="equal across measured libraries after within-library cell mean"))
        groups = pd.DataFrame(group_rows)
        require(len(groups) == 36 * 17 and groups.groupby("gene_symbol").n_cells.sum().eq(8340).all(), "Group summaries lost fixed cells.")
        # Set this storage path independently before running. candidate_library_summary.tsv.
        libraries.to_csv(ROOT, sep="\t", index=False)
        # Set this storage path independently before running. candidate_group_summary.tsv.
        groups.to_csv(ROOT, sep="\t", index=False)
        report = dict(schema="Fig4_E4_source_aware_candidates_v1", cells=len(obs), genes=CANDIDATES,
                      common_candidates=len(common_positions), source_specific_candidates=len(extra_positions),
                      missing_values=int((~measured).sum()), normalization="log10(1 + native integer count / full available source endogenous count * 1e6)",
                      denominator_feature_scope="All supplied source endogenous features; retain mitochondrial/Hb when supplied; exclude technical/reporter/external spikes. Source missing genes are not imputed.",
                      embryo_denominator_source=str(HI.relative_to(ROOT)), embryo_denominator_sha256=digest(HI),
                      embryo7355_denominator_arrays_agree=True, embryo9candidate_expression_reconstructed_from_Whole_and_HI_denominator_exact=True,
                      HI_selected_candidate_overlap=0, sources=source_reports,
                      aggregation="Within stage-state-study-library, average cells; then average measured true-library summaries equally",
                      groups=36, true_libraries=int(obs.library_id.nunique()), stage_state_libraries=len(libraries)//17,
                      limitations=["Library IDs are sampling/technical units; not verified animal replicates.",
                                   "No age P values treating individual cells as animal replicates.",
                                   "Eight candidates are unmeasured in GSE195500; groups containing this source have gene-specific measured source/library counts.",
                                   "E9.5, E17.5 and P7 have one true library per stage-state; summaries are descriptive.",
                                   "UMI, Smart-seq2 read counts and E10 assigned-pair counts remain platform/source dependent after CPM normalization.",
                                   "GSE195500 mitochondrial features are absent from the supplied source; full available source is not a reconstructed full transcriptome.",
                                   "Fourteen P1 totals use source-bound archived full-source metadata; 300 P1 totals were independently checked in the current native QC."])
        # Set this storage path independently before running. candidates.h5.
        output = ROOT
        with h5py.File(output, "w") as handle:
            text = h5py.string_dtype("utf-8")
            for key, value in [("counts", counts), ("measured", measured), ("detection", detection), ("denominator", denominator),
                               ("CPM", cpm), ("log10_1p_CPM", expression)]:
                handle.create_dataset(key, data=value, compression="gzip", shuffle=True)
            for key, value in [("cell_ids", obs.index.to_numpy(dtype=object)), ("gene_ids", np.asarray(genes, dtype=object)),
                               ("gene_symbols", np.asarray(CANDIDATES, dtype=object)), ("denominator_source", denominator_source)]:
                handle.create_dataset(key, data=value, dtype=text)
            for key in ["stage_standardized", "source_stage_celltype", "state", "study_id", "library_id", "count_unit_verified"]:
                handle.create_dataset(key, data=obs[key].astype(str).to_numpy(dtype=object), dtype=text)
            handle.create_dataset("manifest_json", data=json.dumps(report), dtype=text)
        report["cache_sha256"] = digest(output)
        report["cache_bytes"] = output.stat().st_size
        # Set this storage path independently before running. candidates.json.
        (ROOT).write_text(json.dumps(report, indent=2), encoding="utf-8")
        print(json.dumps({key: report[key] for key in ["cells", "common_candidates", "source_specific_candidates", "missing_values", "groups", "true_libraries", "stage_state_libraries", "cache_bytes", "cache_sha256"]}), flush=True)
    prepare_candidates()

def score_development(argv=None):
    """Check or run fixed external-trait SGN scDRS and its technical sensitivity."""
    #1. =========== Shared configuration and pinned inputs

    import argparse
    import functools
    import hashlib
    import inspect
    import json
    import os
    import sys
    from pathlib import Path

    for setting in ["OMP_NUM_THREADS", "OPENBLAS_NUM_THREADS", "MKL_NUM_THREADS", "NUMEXPR_NUM_THREADS"]:
        os.environ[setting] = "1"
    os.environ["TQDM_DISABLE"] = "1"
    sys.dont_write_bytecode = True

    import h5py
    import numpy as np
    import pandas as pd
    from scipy import sparse

    # Set this storage path independently before running.
    TASK = ROOT
    # Set this storage path independently before running.
    PACKAGES = ROOT
    # Set this storage path independently before running. common_counts.h5.
    COUNTS = ROOT
    # Set this storage path independently before running. Whole_CSS.h5ad.
    WHOLE = ROOT
    # Set this storage path independently before running. mouse_human_homologs.txt.
    HOMOLOGS = ROOT
    # Set this storage path independently before running. score.h5.
    SCORES = ROOT
    EXPECTED_COUNTS_SHA256 = "c02e0728d46848552e59476994c708109d82229a61c587a92bd2b8a09b55f0be"
    EXPECTED_WHOLE_SHA256 = "57217aabefc823bf64fa9f6db5e9cd5c2314d7164f3a758d7458650bd36c24dc"
    HOMOLOG_SHA256 = "2d0ef4953bc1654f3835dba02c15c836ffe9f2162631069a766d898959016fb3"
    SCDRS_COMMIT = "644a5c7900d3c8d84c068725b47dd81e0db22ffb"
    RANDOM_SEED = 42
    N_CTRL = 1000
    TARGET_SUM = 10000.0
    N_MEAN_BIN = 20
    N_VAR_BIN = 20


    def require(condition, message):
        if not condition:
            raise ValueError(message)


    def digest(path):
        with Path(path).open("rb") as stream:
            return hashlib.file_digest(stream, "sha256").hexdigest()


    def vector(node):
        if isinstance(node, h5py.Dataset):
            return node.asstr()[()] if node.dtype.kind in "OSU" else node[()]
        if "categories" in node and "codes" in node:
            return pd.Categorical.from_codes(node["codes"][()], vector(node["categories"]))
        if "values" in node and "mask" in node:
            result = np.asarray(vector(node["values"]), dtype=object)
            result[node["mask"][()]] = None
            return result
        raise ValueError(node.name)


    def frame(group):
        key = group.attrs.get("_index", "_index")
        if isinstance(key, bytes):
            key = key.decode()
        return pd.DataFrame({name: vector(group[name]) for name in group if name != key},
                            index=pd.Index(vector(group[key]).astype(str)))


    def check_package():
        # Set this storage path independently before running. source_manifest.json.
        manifest = json.loads((ROOT).read_text(encoding="utf-8"))
        require(manifest["github_commit"] == SCDRS_COMMIT, "scDRS source commit changed.")
        for record in manifest["files"]:
            path = PACKAGES / record["path"]
            require(path.stat().st_size == record["bytes"] and digest(path) == record["sha256"],
                    f"Pinned scDRS source changed: {path}")
        require(digest(HOMOLOGS) == HOMOLOG_SHA256, "scDRS human-mouse reference changed.")
        return manifest


    #2. =========== Fixed cells and the scoring-only RNA copy
    def load_inputs():
        require(digest(COUNTS) == EXPECTED_COUNTS_SHA256, "Verified native count cache changed.")
        require(digest(WHOLE) == EXPECTED_WHOLE_SHA256, "Accepted Whole file changed.")
        with h5py.File(COUNTS, "r") as handle:
            cells = pd.Index(handle["cell_ids"].asstr()[()])
            genes = pd.DataFrame({"gene_id": handle["gene_ids"].asstr()[()],
                                  "gene_symbol": handle["gene_symbols"].asstr()[()]})
            counts = sparse.csr_matrix((handle["counts/data"][()], handle["counts/indices"][()],
                                        handle["counts/indptr"][()]), shape=tuple(handle["counts"].attrs["shape"]))
        with h5py.File(WHOLE, "r") as handle:
            obs = frame(handle["obs"])
        require(cells.equals(obs.index) and len(cells) == 8340 and cells.is_unique, "Fixed accepted cell IDs/order changed.")
        require(counts.shape == (8340, 11149) and genes.gene_id.is_unique and genes.gene_symbol.is_unique,
                "Native common counts or gene identities changed.")
        require(counts.has_canonical_format and np.isfinite(counts.data).all() and (counts.data >= 0).all(),
                "Count cache is not canonical nonnegative CSR.")
        require(np.equal(counts.data, np.rint(counts.data)).all(), "Count cache has noninteger values.")
        keep = np.asarray(counts.sum(axis=0)).ravel() > 0
        require(int((~keep).sum()) == 8, "The expected eight all-zero genes changed; review before scoring.")
        omitted = genes.loc[~keep].to_dict(orient="records")
        counts = counts[:, keep].tocsr()
        genes = genes.loc[keep].reset_index(drop=True)
        require(counts.shape == (8340, 11141), "Unexpected scoring-only gene count.")
        obs["detected_genes_common_axis"] = np.asarray((counts > 0).sum(axis=1)).ravel().astype(np.int32)
        complexity = np.log1p(obs.detected_genes_common_axis.astype(np.float64))
        centered = complexity - complexity.groupby(obs.library_id, observed=True).transform("mean")
        obs["within_library_log1p_detected_genes"] = centered.to_numpy(dtype=np.float64)
        require(obs.source_stage_celltype.nunique() == 36 and obs.library_id.nunique() == 41,
                "Accepted stage-state groups or true-library identities changed.")
        require(centered.groupby(obs.library_id, observed=True).mean().abs().max() < 1e-12,
                "Technical complexity covariate is not centered within real libraries.")
        return counts, genes, obs, omitted


    def normalize_counts(counts):
        totals = np.asarray(counts.sum(axis=1)).ravel().astype(np.float64)
        require((totals > 0).all(), "Cannot normalize an empty accepted cell.")
        expression = counts.astype(np.float32, copy=True)
        expression.data *= np.repeat((TARGET_SUM / totals).astype(np.float32), np.diff(expression.indptr))
        require(np.allclose(np.asarray(expression.sum(axis=1)).ravel(), TARGET_SUM, rtol=2e-6), "CP10k row sums differ.")
        np.log1p(expression.data, out=expression.data)
        require(expression.dtype == np.float32 and sparse.isspmatrix_csr(expression) and np.isfinite(expression.data).all(),
                "Scoring expression is not finite float32 CSR.")
        return expression


    #3. =========== External top-1000 human genes are mapped after ranking
    def load_trait(path, trait, genes, rank_manifest=None):
        table = pd.read_csv(path, sep="\t")
        require(table.columns.tolist() == ["TRAIT", "GENESET"], "Expected weighted .gs columns TRAIT and GENESET.")
        if trait is None:
            require(len(table) == 1, "Multiple external traits: select an exact --trait value.")
            row = table.iloc[0]
        else:
            selected = table[table.TRAIT.eq(trait)]
            require(len(selected) == 1, "Requested external trait is missing or duplicated.")
            row = selected.iloc[0]
        tokens = [token.split(":") for token in row.GENESET.split(",")]
        require(0 < len(tokens) <= 1000 and all(len(token) == 2 for token in tokens),
                "External trait must contain weighted mapped symbols from the original top-1000 Entrez ranks.")
        weighted = pd.DataFrame({"human_gene": [token[0] for token in tokens], "MAGMA_Z": [float(token[1]) for token in tokens]})
        require(weighted.human_gene.is_unique and weighted.human_gene.str.len().gt(0).all(), "External weighted human names collide.")
        rank_path = Path(rank_manifest) if rank_manifest is not None else Path(path).with_suffix(".tsv")
        require(rank_path.is_file(), "Original complete top-1000 Entrez rank/mapping manifest is required; do not refill unmapped ranks.")
        ranking = pd.read_csv(rank_path, sep="\t", dtype={"GENE": str}, keep_default_na=False)
        required = {"original_rank", "GENE", "symbol", "mapping_status", "ZSTAT"}
        require(required <= set(ranking), "Original top-1000 rank manifest columns are incomplete.")
        require(len(ranking) == 1000 and ranking.GENE.is_unique and np.array_equal(ranking.original_rank, np.arange(1, 1001)),
                "Original fixed Entrez ranks must contain exactly 1..1000 once each.")
        ranking["ZSTAT"] = pd.to_numeric(ranking.ZSTAT, errors="raise")
        ordered = ranking.assign(_entrez_sort=pd.to_numeric(ranking.GENE, errors="raise")).sort_values(
            ["ZSTAT", "_entrez_sort"], ascending=[False, True], kind="mergesort")
        require(ordered.GENE.tolist() == ranking.GENE.tolist(), "Top-1000 Entrez rank order differs from MAGMA Z and its numeric-ID tie rule.")
        expected = ranking.loc[ranking.mapping_status.eq("HGNC_unique_entrez") & ranking.symbol.ne("")]
        require(expected.symbol.is_unique and set(expected.symbol) == set(weighted.human_gene),
                "Weighted symbols differ from the uniquely HGNC-mapped original top-1000; ranking must not be refilled.")
        expected_z = expected.set_index("symbol").ZSTAT.loc[weighted.human_gene].to_numpy(dtype=float)
        require(np.allclose(weighted.MAGMA_Z.to_numpy(), expected_z, rtol=1e-14, atol=1e-14), "Weighted .gs values differ from the original MAGMA Z.")
        external = ranking.rename(columns={"original_rank": "external_Entrez_rank", "GENE": "human_entrez_id", "symbol": "human_gene", "ZSTAT": "MAGMA_Z"}).copy()
        external["HGNC_mapping_status"] = external.mapping_status
        require(np.isfinite(external.MAGMA_Z).all() and external.MAGMA_Z.gt(0).all(),
                "Top-1000 MAGMA Z weights must be finite and positive; do not alter the external weights silently.")
        homologs = pd.read_csv(HOMOLOGS, sep="\t")
        require(len(homologs) == 16466 and all(homologs[column].is_unique for column in homologs), "Expected one-to-one scDRS homolog snapshot changed.")
        human_to_mouse = dict(zip(homologs.HUMAN_GENE_SYM, homologs.MOUSE_GENE_SYM))
        external["mouse_gene"] = external.human_gene.map(human_to_mouse)
        available = genes.set_index("gene_symbol").gene_id
        external["mouse_gene_id"] = external.mouse_gene.map(available)
        external["in_scoring_axis"] = external.mouse_gene_id.notna()
        external["mapping_status"] = np.where(external.HGNC_mapping_status.ne("HGNC_unique_entrez"), "unresolved_original_Entrez_to_HGNC_symbol",
                                              np.where(external.mouse_gene.isna(), "unresolved_in_fixed_homolog_table",
                                              np.where(external.in_scoring_axis, "mapped_and_in_detected_native_common_axis",
                                                       "mapped_but_outside_detected_native_common_axis")))
        mapped = external[external.in_scoring_axis].copy()
        require(mapped.mouse_gene.is_unique and len(mapped) > 0, "No unique trait genes available for scoring.")
        report = dict(trait=str(row.TRAIT), external_human_Entrez_genes=1000, weighted_human_symbols=len(weighted),
                      original_top1000_HGNC_unresolved=1000-len(weighted), mapped_mouse_genes=int(external.mouse_gene.notna().sum()),
                      genes_in_scoring_axis=len(mapped), MAGMA_Z_weight_fraction_in_scoring_axis=float(mapped.MAGMA_Z.sum()/external.MAGMA_Z.sum()),
                      external_gs_path=str(Path(path).resolve()), external_gs_sha256=digest(path),
                      original_rank_manifest=str(rank_path.resolve()), original_rank_manifest_sha256=digest(rank_path),
                      fixed_gene_set_policy="Human MAGMA top1000 Entrez genes ranked externally before HGNC, one-to-one homolog mapping and RNA intersection; never replace unresolved ranks or refill from expressed genes",
                      weight_policy="Original external human MAGMA Z; scDRS variance-stabilizing weight_opt=vs remains standard")
        return str(row.TRAIT), mapped, external, report


    #4. =========== Pre-score stage, accepted category and source coverage
    def coverage_table(counts, genes, obs, mapped=None):
        rows = []
        trait_columns = None if mapped is None else pd.Index(genes.gene_symbol).get_indexer(mapped.mouse_gene)
        trait_detected = None if mapped is None else counts[:, trait_columns] > 0
        grouping = [("stage", ["stage_standardized"]), ("category", ["source_stage_celltype"]),
                    ("source", ["study_id"]), ("stage_category_source", ["stage_standardized", "source_stage_celltype", "study_id"])]
        for level, columns in grouping:
            for key, indices in obs.groupby(columns, observed=True, sort=False).indices.items():
                if not isinstance(key, tuple):
                    key = (key,)
                sub = obs.iloc[indices]
                row = dict(level=level, **dict(zip(columns, map(str, key))), n_cells=len(indices),
                           n_libraries=int(sub.library_id.nunique()), sources=";".join(sorted(sub.study_id.astype(str).unique())),
                           measured_background_genes=counts.shape[1], detected_background_genes=int((np.asarray(counts[indices].sum(axis=0)).ravel()>0).sum()),
                           common_detected_genes_per_cell_median=float(sub.detected_genes_common_axis.median()))
                if mapped is not None:
                    available = np.asarray(trait_detected[indices].sum(axis=0)).ravel() > 0
                    per_cell = np.asarray(trait_detected[indices].sum(axis=1)).ravel()
                    row.update(fixed_trait_genes_in_common_axis=len(mapped), fixed_trait_genes_detected_in_group=int(available.sum()),
                               fixed_trait_genes_detected_per_cell_mean=float(per_cell.mean()),
                               fixed_trait_genes_detected_per_cell_median=float(np.median(per_cell)),
                               fixed_trait_genes_detected_per_cell_min=int(per_cell.min()), fixed_trait_genes_detected_per_cell_max=int(per_cell.max()),
                               fraction_trait_Z_weights_detected_in_group=float(mapped.MAGMA_Z.to_numpy()[available].sum()/mapped.MAGMA_Z.sum()))
                rows.append(row)
        return pd.DataFrame(rows)


    #5. =========== Standard scDRS and a within-library complexity sensitivity
    def write_frame(group, name, table):
        child = group.create_group(name)
        text = h5py.string_dtype("utf-8")
        child.create_dataset("index", data=np.asarray(table.index.astype(str), dtype=object), dtype=text)
        for column in table:
            values = table[column].to_numpy()
            if values.dtype.kind in "biufc":
                child.create_dataset(str(column), data=values, compression="gzip", shuffle=True)
            else:
                child.create_dataset(str(column), data=table[column].astype(str).to_numpy(dtype=object), dtype=text)


    @functools.lru_cache(maxsize=1)
    def association_helper():
        from types import SimpleNamespace
        from statsmodels.stats.multitest import multipletests
        helper = SimpleNamespace(association_only=association_only, multipletests=multipletests)
        evidence = dict(
            source="Inline E1-validated association_only; same formulas and columns as the verified 30-cell/2500-gene/1000-control official-function comparison",
            function_sha256=hashlib.sha256(inspect.getsource(association_only).encode()).hexdigest(),
            scope="95th-quantile matched-control MC association and global cell-FDR counts; no graph or heterogeneity",
        )
        return helper, evidence



    from statsmodels.stats.multitest import multipletests
    FDR_THRESHOLDS = [0.05, 0.1, 0.2]

    def association_only(adata, full_score, group_col="cell_type_author"):
        """Reproduce method.py lines 763-817 at COMMIT; no graph or heterogeneity.

        The official function has no heterogeneity switch. This explicit function
        retains its sorted-cell alignment, cell FDR counts and 95th-quantile MC test.
        It is tested against the complete official function on its 30-cell toy data.
        """
        assert group_col in adata.obs
        cell_list = sorted(set(adata.obs_names) & set(full_score.index))
        assert len(cell_list) == adata.n_obs == len(full_score)
        control_list = [key for key in full_score if key.startswith("ctrl_norm_score")]
        n_ctrl = len(control_list)
        assert n_ctrl == N_CTRL
        reg = adata.obs.loc[cell_list, [group_col]].copy()
        reg = reg.join(full_score.loc[cell_list, ["norm_score"] + control_list + ["pval"]])
        group_list = sorted(set(adata.obs[group_col]))
        columns = ["n_cell", "n_ctrl", "assoc_mcp", "assoc_mcz", "hetero_mcp", "hetero_mcz"]
        columns.extend(f"n_fdr_{threshold}" for threshold in FDR_THRESHOLDS)
        result = pd.DataFrame(index=group_list, columns=columns, dtype=np.float32)
        result.index.name = "group"
        cell_fdr = pd.DataFrame({"fdr": multipletests(reg["pval"].values, method="fdr_bh")[1]}, index=reg.index)
        for group in group_list:
            cells = list(reg.index[reg[group_col] == group])
            result.loc[group, ["n_cell", "n_ctrl"]] = [len(cells), n_ctrl]
            for threshold in FDR_THRESHOLDS:
                result.loc[group, f"n_fdr_{threshold}"] = (cell_fdr.loc[cells, "fdr"].values < threshold).sum()
            score_q95 = np.quantile(reg.loc[cells, "norm_score"], 0.95)
            controls_q95 = np.quantile(reg.loc[cells, control_list], 0.95, axis=0)
            mc_p = ((controls_q95 >= score_q95).sum() + 1) / (controls_q95.shape[0] + 1)
            mc_z = (score_q95 - controls_q95.mean()) / controls_q95.std()
            result.loc[group, ["assoc_mcp", "assoc_mcz"]] = [mc_p, mc_z]
        return result

    def run_model(expression, genes, obs, mapped, trait, mode, handle, shared_report):
        import anndata as ad
        import scdrs

        technical = mode == "technical"
        data = ad.AnnData(X=expression.copy(), obs=obs.copy(), var=genes.set_index("gene_symbol").copy())
        covariance = obs[["within_library_log1p_detected_genes"]].copy() if technical else None
        scdrs.pp.preprocess(data, cov=covariance, adj_prop=None, n_mean_bin=N_MEAN_BIN, n_var_bin=N_VAR_BIN, n_chunk=None, copy=False)
        require(sparse.issparse(data.X) and data.X.dtype == np.float32, "Preprocessing lost sparse float32 expression.")
        gene_stats = data.uns["SCDRS_PARAM"]["GENE_STATS"]
        require(gene_stats.index.equals(data.var_names), "scDRS gene-statistics axis changed.")
        require(np.isfinite(gene_stats[["mean", "var", "var_tech"]].to_numpy(dtype=float)).all(), "scDRS preprocessing returned nonfinite gene statistics.")
        trait_genes = mapped.mouse_gene.tolist()
        trait_weights = mapped.MAGMA_Z.to_numpy(dtype=np.float64)
        result = scdrs.score_cell(data, gene_list=trait_genes, gene_weight=trait_weights, ctrl_match_key="mean_var", n_ctrl=N_CTRL,
                                 n_genebin=200, weight_opt="vs", copy=False, return_ctrl_raw_score=False,
                                 return_ctrl_norm_score=True, random_seed=RANDOM_SEED, verbose=False, save_intermediate=None)
        require(result.index.equals(obs.index), "Scoring changed accepted cell IDs/order.")
        control_columns = [f"ctrl_norm_score_{i}" for i in range(N_CTRL)]
        controls = result[control_columns].to_numpy(dtype=np.float32)
        score_columns = [column for column in result if column not in control_columns]
        scores = result[score_columns].to_numpy(dtype=np.float32)
        require(controls.shape == (len(obs), N_CTRL) and np.isfinite(controls).all() and np.isfinite(scores).all(), "Nonfinite scDRS scores or normalized controls.")
        helper, association_evidence = association_helper()
        association = helper.association_only(data, result, group_col="source_stage_celltype")
        association["assoc_fdr_bh"] = helper.multipletests(association.assoc_mcp.to_numpy(dtype=float), method="fdr_bh")[1]
        require(association.n_cell.sum() == len(obs) and set(association.index) == set(obs.source_stage_celltype),
                "Association groups differ from the accepted categories.")
        group = handle.create_group(mode)
        group.create_dataset("norm_control", data=controls, compression="gzip", shuffle=True)
        group.create_dataset("scores", data=scores, compression="gzip", shuffle=True)
        text = h5py.string_dtype("utf-8")
        group.create_dataset("score_columns", data=np.asarray(score_columns, dtype=object), dtype=text)
        group.create_dataset("cell_ids", data=np.asarray(obs.index, dtype=object), dtype=text)
        group.create_dataset("gene_symbols", data=genes.gene_symbol.to_numpy(dtype=object), dtype=text)
        write_frame(group, "gene_statistics", gene_stats)
        write_frame(group, "cell_statistics", data.uns["SCDRS_PARAM"]["CELL_STATS"])
        write_frame(group, "trait_genes", mapped)
        write_frame(group, "group_association", association)
        if technical:
            parameters = data.uns["SCDRS_PARAM"]
            write_frame(group, "covariates", parameters["COV_MAT"])
            write_frame(group, "covariate_beta", parameters["COV_BETA"])
            group.create_dataset("covariate_gene_mean", data=parameters["COV_GENE_MEAN"].to_numpy(dtype=np.float32))
        report = dict(shared_report, mode=mode, cells=len(obs), genes=len(genes), trait_genes=len(mapped),
                      covariates=[] if covariance is None else list(covariance),
                      correction="None" if covariance is None else "Only within-library centered log1p(number of detected common-axis genes); package adds intercept",
                      expression_sparse=True, expression_dtype=str(expression.dtype), normalized_controls_retained=N_CTRL,
                      scdrs_version=scdrs.__version__, score_columns=score_columns,
                      group_association=association_evidence,
                      no_age_cell_as_animal_test=True)
        group.create_dataset("manifest_json", data=json.dumps(report), dtype=text)
        handle.flush()
        print(json.dumps(dict(model=mode, cells=len(obs), genes=len(genes), trait_genes=len(mapped), controls=N_CTRL, complete=True)), flush=True)


    #6. =========== Explicit execution gate and separate within-study scales
    def main():
        parser = argparse.ArgumentParser(description=__doc__)
        parser.add_argument("--gs", type=Path, help="Externally fixed human top-1000 MAGMA-Z weighted .gs")
        parser.add_argument("--trait", help="Exact TRAIT label when the .gs contains multiple traits")
        parser.add_argument("--rank-manifest", type=Path, help="Complete original 1000 Entrez ranks with HGNC mapping status; default: .gs sibling .tsv")
        parser.add_argument("--run", action="store_true", help="Run scoring only after the human MAGMA trait has been accepted")
        parser.add_argument("--mode", choices=["both", "primary", "technical", "by-study"], default="both")
        parser.add_argument("--output", type=Path, default=SCORES)
        args = parser.parse_args(argv)
        require(args.output.resolve().parent == TASK.resolve(), "Score caches must stay inside Tmp/Fig4-genetics/rna.")
        source_manifest = check_package()
        counts, genes, obs, omitted = load_inputs()
        mapped, external, trait, trait_report = None, None, None, None
        if args.gs is not None:
            require(args.gs.is_file(), "External human MAGMA top-1000 weighted .gs is not ready.")
            trait, mapped, external, trait_report = load_trait(args.gs, args.trait, genes, args.rank_manifest)
        coverage = coverage_table(counts, genes, obs, mapped)
        # Set this storage path independently before running. score_coverage.tsv.
        coverage.to_csv(ROOT, sep="\t", index=False)
        if external is not None:
            # Set this storage path independently before running. score_trait_mapping.tsv.
            external.to_csv(ROOT, sep="\t", index=False)
        report = dict(schema="Fig4_SGN_scDRS_preflight_v1", cells=8340, original_native_genes=11149,
                      scoring_genes=len(genes), omitted_all_zero_genes=omitted, accepted_stage_state_categories=36, real_libraries=41,
                      counts_path=str(COUNTS), counts_sha256=EXPECTED_COUNTS_SHA256, Whole_sha256=EXPECTED_WHOLE_SHA256,
                      normalization="Fresh CP10k then natural log1p from integer native-common counts; Whole.X is never reused",
                      denominator="Per-cell subtotal over the complete shared 11149 native-assay genes (eight zero genes have no effect on totals); specific to scDRS preprocessing",
                      scdrs_commit=SCDRS_COMMIT, homolog_reference_sha256=HOMOLOG_SHA256, controls=N_CTRL, random_seed=RANDOM_SEED,
                      ctrl_match_key="mean_var", weight_opt="vs", preprocess_mean_bins=N_MEAN_BIN, preprocess_var_bins=N_VAR_BIN,
                      main_covariates=None, sensitivity_covariate="Within-library centered log1p(detected genes on common native axis)",
                      within_library_covariate_mean_abs_max=float(obs.groupby("library_id", observed=True).within_library_log1p_detected_genes.mean().abs().max()),
                      trait=trait_report, scoring_started=False,
                      limitations=["Accepted categories are preserved; stage and source are confounded in several periods.",
                                   "Libraries are not verified animal replicates; no cell-as-animal age P values are computed.",
                                   "Study-specific scores have independent normalized scales and cannot be compared directly between studies.",
                                   "Unmapped human genes are unresolved in the fixed homolog table, not proven absent from mouse."])
        # Set this storage path independently before running. score_preflight.json.
        (ROOT).write_text(json.dumps(report, indent=2), encoding="utf-8")
        print(json.dumps({key: report[key] for key in ["cells", "scoring_genes", "accepted_stage_state_categories", "real_libraries", "trait", "scoring_started"]}), flush=True)
        if not args.run:
            print("Input and coverage checks complete; scDRS has not been run.", flush=True)
            return
        require(mapped is not None, "No external MAGMA trait supplied; scoring is blocked without substitution.")
        require(not args.output.exists(), "Score output already exists; inspect it before explicitly replacing a current-task cache.")
        sys.path.insert(0, str(PACKAGES))
        import scdrs
        require(scdrs.__version__ == "1.0.4" and Path(scdrs.__file__).resolve().is_relative_to(PACKAGES.resolve()), "Wrong scDRS runtime imported.")
        scdrs.method.tqdm = functools.partial(scdrs.method.tqdm, disable=True)
        expression = normalize_counts(counts)
        shared_report = dict(report, scoring_started=True, script_sha256=digest(Path(__file__)))
        with h5py.File(args.output, "w") as handle:
            text = h5py.string_dtype("utf-8")
            write_frame(handle, "accepted_observations", obs[["stage_standardized", "source_stage_celltype", "study_id", "library_id",
                                                              "count_unit_verified", "detected_genes_common_axis", "within_library_log1p_detected_genes"]])
            write_frame(handle, "external_human_trait", external)
            handle.create_dataset("manifest_json", data=json.dumps(shared_report), dtype=text)
            if args.mode == "by-study":
                for study in dict.fromkeys(obs.study_id.astype(str)):
                    selected = np.flatnonzero(obs.study_id.astype(str).eq(study).to_numpy())
                    keep = np.asarray(counts[selected].sum(axis=0)).ravel() > 0
                    study_genes = genes.loc[keep].reset_index(drop=True)
                    study_mapped = mapped[mapped.mouse_gene.isin(study_genes.gene_symbol)].copy()
                    require(len(study_mapped) > 0, f"No fixed trait genes detected in study {study}.")
                    model_report = dict(shared_report, independent_study=study,
                                        within_study_all_zero_background_genes_dropped=int((~keep).sum()),
                                        independent_score_scale=True, cross_study_direct_comparison_allowed=False,
                                        within_study_trait_coverage=len(study_mapped), external_top1000_unchanged=True)
                    run_model(expression[selected][:, keep].tocsr(), study_genes, obs.iloc[selected], study_mapped, trait,
                              f"study_{study}", handle, model_report)
            else:
                modes = ["primary", "technical"] if args.mode == "both" else [args.mode]
                for mode in modes:
                    run_model(expression, genes, obs, mapped, trait, mode, handle, shared_report)
        summary = dict(shared_report, score_output=str(args.output), score_output_sha256=digest(args.output),
                       score_output_bytes=args.output.stat().st_size, mode=args.mode, scoring_complete=True)
        (TASK / f"{args.output.stem}.json").write_text(json.dumps(summary, indent=2), encoding="utf-8")
        print(json.dumps(dict(scoring_complete=True, output=str(args.output), bytes=args.output.stat().st_size)), flush=True)
    main()

def prepare_developmental_sets():
    """Export frozen full-direction sets and their unique Entrez background."""
    #1. =========== Frozen local identity references
    from pathlib import Path
    import hashlib
    import json

    import h5py
    import numpy as np
    import pandas as pd

    # Set this storage path independently before running.
    TASK = ROOT
    # Set this storage path independently before running. Trajectory dot plot_results-2.h5.
    HI = ROOT
    # Set this storage path independently before running. mouse_human_homologs.txt.
    HOMOLOGS = ROOT
    # Set this storage path independently before running. human-gene-map.tsv.
    HUMAN_MAP = ROOT
    # Set this storage path independently before running. hgnc_complete_set.txt.
    HGNC = ROOT
    # Set this storage path independently before running.
    NCBI = ROOT
    HOMOLOG_SHA256 = "2d0ef4953bc1654f3835dba02c15c836ffe9f2162631069a766d898959016fb3"
    HGNC_SHA256 = "12363916be36f8f6f91879b0faade4eeaac973722de7136ec61329b77f2c2f91"
    NCBI_SHA256 = "657b6c6b43726fd4946bd301728b57bba87230b113c87771f52bffc30837bc46"
    SEGMENTS = ["CVG", "VGN", "T", "AC", "Ib", "Ic", "AB", "Ia", "II"]


    def digest(path):
        with path.open("rb") as stream:
            return hashlib.file_digest(stream, "sha256").hexdigest()


    #2. =========== Unique mouse-symbol to human-symbol to Entrez identities
    def main():
        assert digest(HOMOLOGS) == HOMOLOG_SHA256
        assert digest(HGNC) == HGNC_SHA256
        # Set this storage path independently before running. MHC-audit.json.
        mhc_audit = json.loads((ROOT).read_text(encoding="utf-8"))
        assert mhc_audit["dependency_sha256"] == NCBI_SHA256 and mhc_audit["exact_gene_rows_equal"]
        homologs = pd.read_csv(HOMOLOGS, sep="\t", dtype=str)
        assert len(homologs) == 16466 and all(homologs[column].is_unique for column in homologs)
        human = pd.read_csv(HUMAN_MAP, sep="\t", dtype=str, keep_default_na=False)
        reliable = human.loc[human.mapping_status.eq("HGNC_unique_entrez") & human.symbol.ne("")].copy()
        assert reliable.entrez_id.is_unique and reliable.symbol.is_unique
        hgnc = pd.read_csv(HGNC, sep="\t", dtype=str, keep_default_na=False)
        approved = hgnc.loc[hgnc.status.eq("Approved") & hgnc.entrez_id.ne("")].copy()
        assert approved.symbol.is_unique
        approved = approved.set_index("symbol")
        assert all(approved.at[row.symbol, "entrez_id"] == row.entrez_id for row in reliable.itertuples())
        assert all(approved.at[row.symbol, "hgnc_id"] == row.hgnc_id for row in reliable.itertuples())
        location = pd.read_csv(NCBI, sep="\t", header=None, dtype=str)
        # Set this storage path independently before running.
        complete_location = pd.read_csv(ROOT, sep=r"\s+", header=None, dtype=str)
        reconstructed = complete_location.loc[~complete_location[0].isin(mhc_audit["excluded_entrez_ids"])].reset_index(drop=True)
        assert reconstructed.equals(location.reset_index(drop=True))
        assert location[0].is_unique and set(reliable.entrez_id) <= set(location[0])
        mouse_to_human = dict(zip(homologs.MOUSE_GENE_SYM, homologs.HUMAN_GENE_SYM))
        symbol_to_entrez = dict(zip(reliable.symbol, reliable.entrez_id))
        with h5py.File(HI, "r") as handle:
            tested_ids = handle["genes/gene_id"].asstr()[()]
            tested_symbols = handle["genes/gene_symbol"].asstr()[()]
            assert len(tested_ids) == 11467 and len(set(tested_ids)) == len(set(tested_symbols)) == 11467
            assert np.array_equal(tested_ids, handle["association/gene_id"].asstr()[()])
            candidate_ids = handle["candidate/gene_ids"].asstr()[()]
            candidate_symbols = handle["candidate/gene_symbols"].asstr()[()]
            assert len(candidate_ids) == 3876 and len(set(candidate_ids)) == 3876
            assert set(candidate_ids) == set(tested_ids[handle["association/signi"][()]])
            orders = {(segment, direction): handle[f"candidate/{direction}_order/{segment}"][()]
                      for segment in SEGMENTS for direction in ["up", "down"]}
            input_manifest = json.loads(handle["meta/input"].asstr()[()])
        mapping = pd.DataFrame({"mouse_gene_id": tested_ids, "mouse_symbol": tested_symbols})
        mapping["human_symbol"] = mapping.mouse_symbol.map(mouse_to_human)
        mapping["entrez_id"] = mapping.human_symbol.map(symbol_to_entrez)
        mapping["status"] = np.where(mapping.human_symbol.isna(), "unresolved_in_fixed_one_to_one_homolog_table",
                                     np.where(mapping.entrez_id.isna(), "human_symbol_has_no_unique_current_HGNC_Entrez_in_MHC_excluded_NCBI37_axis", "accepted_unique_chain"))
        collision = mapping.entrez_id.notna() & mapping.entrez_id.duplicated(keep=False)
        mapping.loc[collision, "status"] = "excluded_multiple_mouse_genes_to_one_Entrez"
        mapping.loc[collision, "entrez_id"] = np.nan
        accepted = mapping[mapping.entrez_id.notna()].copy()
        assert accepted.entrez_id.is_unique
        background = set(accepted.entrez_id)
        lookup = mapping.set_index("mouse_gene_id")
        rows, lines, summaries = [], [], []
        for segment in SEGMENTS:
            up = set(map(int, orders[(segment, "up")]))
            down = set(map(int, orders[(segment, "down")]))
            assert not up & down
            assert up | down <= set(range(3876))
            for direction in ["up", "down"]:
                positions = orders[(segment, direction)]
                assert len(set(positions)) == len(positions) and np.all((positions >= 0) & (positions < 3876))
                ids = candidate_ids[positions]
                sub = lookup.loc[ids].reset_index()
                sub.insert(0, "set_name", f"{segment}_{direction}")
                sub.insert(1, "original_order", np.arange(1, len(sub) + 1))
                sub.insert(2, "candidate_position", positions)
                sub["in_mapped_tested_background"] = sub.entrez_id.isin(background)
                kept = sub.loc[sub.in_mapped_tested_background, "entrez_id"].tolist()
                assert len(kept) == len(set(kept)) and set(kept) <= background
                lines.append(" ".join([f"{segment}_{direction}"] + kept))
                summaries.append(dict(set_name=f"{segment}_{direction}", full_HI_candidate_members=len(positions),
                                      mapped_Entrez_in_tested_background=len(kept), dropped=len(positions)-len(kept),
                                      mapping_status_counts=sub.status.value_counts().to_dict(),
                                      zero_fitted_change_candidates_not_in_either_direction=3876-len(up|down)))
                rows.append(sub)
        # Set this storage path independently before running. E3-mapping.tsv.
        mapping.to_csv(ROOT, sep="\t", index=False)
        # Set this storage path independently before running. E3-membership.tsv.gz.
        pd.concat(rows, ignore_index=True).to_csv(ROOT, sep="\t", index=False)
        # Set this storage path independently before running. E3-background.txt.
        (ROOT).write_text("\n".join(accepted.entrez_id) + "\n", encoding="utf-8")
        # Set this storage path independently before running. E3-sets.txt.
        (ROOT).write_text("\n".join(lines) + "\n", encoding="utf-8")
        report = dict(schema="Fig4_E3_frozen_HI_direction_sets_v1", tested_mouse_genes=11467, full_significant_candidates=3876,
                      mapped_Entrez_background=len(background), unmapped_background_genes=11467-len(background),
                      background_status_counts=mapping.status.value_counts().to_dict(), sets=summaries,
                      use_72_display_genes=False, set_policy="All frozen candidate/up_order and candidate/down_order entries; no display-gene subset and no new gene ranking",
                      direction_policy="Retain the existing fitted end-minus-start directions and order; do not assign an artificial sign to exact-zero change",
                      mapping_policy="Fixed unique scDRS mouse/human homolog, exact current Approved HGNC symbol and unique Entrez, on the MHC-excluded NCBI37.3 MAGMA axis; no alias rescue or ambiguous merging",
                      sources={str(path.relative_to(ROOT)): digest(path) for path in [HI, HOMOLOGS, HUMAN_MAP, HGNC, NCBI]},
                      HI_tree_sha256=input_manifest["tree_bundle_sha256"],
                      MAGMA_gene_result_overlap="Not yet checked here; final GSA must report overlap with valid estimated MAGMA genes",
                      output_sha256={name: digest(TASK/name) for name in ["E3-background.txt", "E3-sets.txt", "E3-mapping.tsv", "E3-membership.tsv.gz"]})
        # Set this storage path independently before running. E3-mapping.json.
        (ROOT).write_text(json.dumps(report, indent=2), encoding="utf-8")
        print(json.dumps(dict(tested=11467, mapped_background=len(background), set_count=len(lines),
                             sets=[(row["set_name"], row["full_HI_candidate_members"], row["mapped_Entrez_in_tested_background"]) for row in summaries])), flush=True)
    main()

def prepare_developmental_covariates():
    """Compute full-axis library-balanced expression and dynamic covariates."""
    #1. =========== Full embryonic counts, source denominators and unique Entrez axis
    from pathlib import Path
    import hashlib
    import json

    import h5py
    import numpy as np
    import pandas as pd
    from scipy import sparse

    # Set this storage path independently before running.
    TASK = ROOT
    # Set this storage path independently before running. CSS_3000_Embryonic_Inte_adata.h5ad.
    EMBRYO = ROOT
    # Set this storage path independently before running. Trajectory dot plot_results-2.h5.
    HI = ROOT


    def digest(path):
        with path.open("rb") as stream:
            return hashlib.file_digest(stream, "sha256").hexdigest()


    def main():
        with h5py.File(HI, "r") as handle:
            cells = handle["cells/_index"].asstr()[()]
            denominator = handle["cells/library_total"][()]
            libraries = handle["cells/library_id"].asstr()[()]
            genes = handle["genes/gene_id"].asstr()[()]
            dynamic = set(handle["candidate/gene_ids"].asstr()[()])
            assert np.array_equal(denominator, handle["normalization/library_total"][()])
            assert len(cells) == 7355 and len(genes) == 11467 and len(dynamic) == 3876
            assert np.isfinite(denominator).all() and (denominator > 0).all()
        assert digest(EMBRYO) == "8625e64dfb80313661f2258c1363e63ebe3e6f5446c4e2da3c920e46137e1e2c"
        with h5py.File(EMBRYO, "r") as handle:
            assert np.array_equal(cells, handle["obs/_index"].asstr()[()])
            assert np.array_equal(genes, handle["var/_index"].asstr()[()])
            counts = sparse.csr_matrix((handle["layers/counts/data"][()], handle["layers/counts/indices"][()],
                                        handle["layers/counts/indptr"][()]), shape=(7355, 11467))
        assert np.isfinite(counts.data).all() and (counts.data >= 0).all()
        expression = counts.astype(np.float64)
        expression.data *= np.repeat(1e6 / denominator, np.diff(expression.indptr))
        np.log10(1 + expression.data, out=expression.data)
        library_order = list(dict.fromkeys(libraries))
        aggregation = sparse.lil_matrix((len(library_order), len(cells)), dtype=np.float64)
        for row, library in enumerate(library_order):
            selected = np.flatnonzero(libraries == library)
            aggregation[row, selected] = 1.0 / len(selected)
        library_means = (aggregation.tocsr() @ expression).toarray()
        mean_expression = library_means.mean(axis=0)
        assert np.isfinite(mean_expression).all() and (mean_expression >= 0).all()
        tested = pd.DataFrame({"mouse_gene_id": genes, "mean_expression": mean_expression,
                               "dynamic_candidate": [int(gene in dynamic) for gene in genes]})
        assert tested.dynamic_candidate.sum() == 3876
        # Set this storage path independently before running. E3-mapping.tsv.
        mapping = pd.read_csv(ROOT, sep="\t", dtype={"entrez_id": str})
        accepted = mapping.loc[mapping.status.eq("accepted_unique_chain")]
        aligned = accepted[["mouse_gene_id", "entrez_id"]].merge(tested, on="mouse_gene_id", how="left", validate="one_to_one")
        # Set this storage path independently before running. E3-background.txt.
        background = (ROOT).read_text().splitlines()
        assert aligned.entrez_id.tolist() == background and len(aligned) == 10186
        output = aligned.rename(columns={"entrez_id": "GENE"})[["GENE", "mean_expression", "dynamic_candidate"]]
        assert output.GENE.is_unique and not output.isna().any().any()
        # Set this storage path independently before running. E3-gene-covar.txt.
        output.to_csv(ROOT, sep="\t", index=False, float_format="%.17g")
        report = dict(schema="Fig4_E3_conditioning_covariates_v1", background_genes=len(output), full_tested_genes=11467,
                      full_frozen_dynamic_candidates=3876, mapped_dynamic_candidates=int(output.dynamic_candidate.sum()),
                      embryonic_cells=7355, real_libraries=len(library_order), library_order=library_order,
                      expression="Full 11467-axis raw counts / per-cell latest-HI audited full available source endogenous count * 1e6; log10(1+CPM); mean within each true library, then equal mean over all embryonic libraries",
                      dynamic="Frozen full3876 candidate membership; not 72 display genes or segment-local membership",
                      no_MAGMA_result_read=True, primary_GSA="Condition each of 18 segment-direction sets on global dynamic_candidate and full-tested-axis mean_expression",
                      sensitivity_GSA="Unconditioned competitive GSA retained separately", BH_family=18,
                      mean_expression_range=[float(output.mean_expression.min()), float(output.mean_expression.max())],
                      # Set this storage path independently before running. E3-mapping.tsv.
                      sources={str(path.relative_to(ROOT)): digest(path) for path in [EMBRYO, HI, ROOT]},
                      # Set this storage path independently before running. E3-gene-covar.txt.
                      output_sha256=digest(ROOT))
        # Set this storage path independently before running. E3-gene-covar.json.
        (ROOT).write_text(json.dumps(report, indent=2), encoding="utf-8")
        print(json.dumps({key: report[key] for key in ["background_genes", "full_tested_genes", "mapped_dynamic_candidates", "embryonic_cells", "real_libraries", "mean_expression_range", "output_sha256"]}), flush=True)
    main()

def analyze_developmental_trajectories(argv=None):
    """Prepare or analyze frozen native-t weighted metacell designs; candidate WLS is diagnostic."""
    #1. =========== Frozen native pseudotime and declared fitting settings
    from pathlib import Path
    import argparse
    import hashlib
    import json

    import h5py
    import numpy as np
    import pandas as pd
    from patsy import dmatrix, build_design_matrices
    from scipy import sparse
    from statsmodels.stats.multitest import multipletests

    # Set this storage path independently before running.
    TASK = ROOT
    # Set this storage path independently before running. Deve_regulon_dynamics.h5. Accepted primary result produced by fig4A_regulon_dynamics_analysis.py.
    REGULONS = ROOT
    # Set this storage path independently before running. Trajectory dot plot_results-2.h5.
    HI = ROOT
    PATHS = {"Ia": "path_Ia", "Ib": "path_Ib", "Ic": "path_Ic", "II": "path_II", "VGN_candidate": "path_VGN_candidate"}
    N_BINS = 20
    MIN_CELLS = 3
    SPLINE_DF = 4
    N_GRID = 200
    N_CTRL = 1000
    MIN_STUDY_METACELLS = 8
    MIN_STUDY_BINS = 5
    MIN_STUDY_LIBRARIES = 2
    MIN_STUDY_SPAN_FRACTION = 0.5  # Coverage-only working criterion; review before source sensitivity.


    def require(condition, message):
        if not condition:
            raise ValueError(message)


    def digest(path):
        with path.open("rb") as stream:
            return hashlib.file_digest(stream, "sha256").hexdigest()


    def frozen_cells():
        with h5py.File(REGULONS, "r") as handle:
            cells = pd.Index(handle["cells"].asstr()[()])
            native = pd.DataFrame({"t": handle["obs/t"][()], "library": handle["obs/library"].asstr()[()],
                                   "study": handle["obs/study"].asstr()[()], "stage": handle["obs/stage"].asstr()[()]}, index=cells)
            masks = {name: handle[f"obs/{key}"][()].astype(bool) for name, key in PATHS.items()}
        with h5py.File(HI, "r") as handle:
            require(np.array_equal(cells, handle["cells/_index"].asstr()[()]), "Frozen path and HI cell orders differ.")
            require(np.array_equal(native.t, handle["cells/t"][()]), "Frozen native t differs from HI original metric pseudotime.")
        require(len(native) == 7355 and cells.is_unique and np.isfinite(native.t).all(), "Native embryo cells/t are invalid.")
        return native, masks


    #2. =========== Library-by-quantile-bin metacells and equal total library weight
    def metacell_design(native, member):
        positions = np.flatnonzero(member)
        require(len(positions) >= N_BINS, "Path has too few fixed cells for quantile bins.")
        selected = native.iloc[positions].copy()
        selected["bin"] = pd.qcut(selected.t, q=N_BINS, labels=False, duplicates="drop").astype(int)
        rows, membership = [], []
        for (library, bin_id), local in selected.groupby(["library", "bin"], observed=True, sort=False).indices.items():
            if len(local) < MIN_CELLS:
                continue
            cells = positions[local]
            sub = native.iloc[cells]
            require(sub.study.nunique() == 1, "A library ID spans multiple studies.")
            membership.append(cells)
            rows.append(dict(library=str(library), bin=int(bin_id), study=str(sub.study.iloc[0]), cells=len(cells),
                             t=float(sub.t.mean()), t_min=float(sub.t.min()), t_max=float(sub.t.max()),
                             stage_counts=json.dumps(sub.stage.value_counts().to_dict(), sort_keys=True)))
        metadata = pd.DataFrame(rows)
        require(len(metadata) >= SPLINE_DF + 1 and metadata.t.nunique() >= SPLINE_DF + 1, "Too few effective metacells to fit the natural cubic model.")
        valid_bins = metadata.groupby("library", observed=True).bin.transform("count")
        metadata["weight"] = 1.0 / valid_bins.to_numpy(dtype=float)
        require(np.allclose(metadata.groupby("library", observed=True).weight.sum(), 1), "Library total weights differ.")
        indptr = np.r_[0, np.cumsum([len(part) for part in membership])]
        indices = np.concatenate(membership)
        values = np.concatenate([np.full(len(part), 1.0/len(part)) for part in membership])
        aggregation = sparse.csr_matrix((values, indices, indptr), shape=(len(metadata), len(native)))
        basis = dmatrix(f"cr(t, df={SPLINE_DF}) - 1", {"t": metadata.t.to_numpy()}, return_type="dataframe")
        design = np.asarray(basis)
        require(np.linalg.matrix_rank(design) == SPLINE_DF, "Natural cubic design is rank deficient.")
        require(np.max(np.abs(design @ np.linalg.lstsq(design, np.ones(len(design)), rcond=None)[0] - 1)) < 1e-10,
                "Natural cubic basis does not contain the constant span.")
        grid = np.linspace(metadata.t.min(), metadata.t.max(), N_GRID)
        grid_basis = np.asarray(build_design_matrices([basis.design_info], {"t": grid})[0])
        return metadata, aggregation, design, grid, grid_basis


    def fit_values(values, metadata, design, grid_basis):
        weight = metadata.weight.to_numpy(dtype=float)
        root_weight = np.sqrt(weight)
        coefficient = np.linalg.lstsq(design * root_weight[:, None], values * root_weight[:, None], rcond=None)[0]
        fitted = design @ coefficient
        weighted_mean = (values * weight[:, None]).sum(axis=0) / weight.sum()
        total = ((values - weighted_mean)**2 * weight[:, None]).sum(axis=0)
        residual = ((values - fitted)**2 * weight[:, None]).sum(axis=0)
        require(np.isfinite(total).all() and (total > 0).all(), "A constant metacell score has undefined R2; inspect instead of fabricating a trend test.")
        r2 = 1 - residual / total
        require((r2 >= -1e-9).all() and (r2 <= 1 + 1e-9).all(), "Weighted R2 is outside its constant-span bounds.")
        return fitted, grid_basis @ coefficient, r2


    def text_frame(group, frame):
        text = h5py.string_dtype("utf-8")
        for name in frame:
            values = frame[name].to_numpy()
            if values.dtype.kind in "biuf":
                group.create_dataset(name, data=values)
            else:
                group.create_dataset(name, data=frame[name].astype(str).to_numpy(dtype=object), dtype=text)


    def score_values(handle, native):
        cells = pd.Index(handle["cell_ids"].asstr()[()])
        positions = cells.get_indexer(native.index)
        require((positions >= 0).all(), "Some native embryo cells have no genetic score.")
        names = handle["score_columns"].asstr()[()].tolist()
        observed = handle["scores"][()][:, names.index("norm_score")][positions]
        controls = handle["norm_control"][()][positions]
        require(controls.shape == (len(native), N_CTRL), "Expected 1000 matched normalized control score sets.")
        return np.column_stack([observed, controls])


    #3. =========== Main five-path matched-control tests and descriptive candidate curves
    def run_paths(native, masks, scores, output):
        results = []
        for path, member in masks.items():
            metadata, aggregation, design, grid, grid_basis = metacell_design(native, member)
            values = aggregation @ scores
            fitted, curves, r2 = fit_values(values, metadata, design, grid_basis)
            p_mc = (1 + int((r2[1:] >= r2[0]).sum())) / (1 + N_CTRL)
            group = output.create_group(path)
            text_frame(group.create_group("metacells"), metadata)
            for name, data in [("values", values), ("fitted", fitted), ("t_grid", grid), ("curves", curves), ("R2", r2),
                               ("control_fit_null_envelope95", np.quantile(curves[:, 1:], [0.025, 0.975], axis=1).T)]:
                group.create_dataset(name, data=data, compression="gzip")
            results.append(dict(path=path, cells=int(member.sum()), retained_cells=int(metadata.cells.sum()), metacells=len(metadata),
                                libraries=int(metadata.library.nunique()), effective_bins=int(metadata.bin.nunique()),
                                valid_t_min=float(grid[0]), valid_t_max=float(grid[-1]), R2=float(r2[0]), p_MC=p_mc,
                                end_minus_start=float(curves[-1, 0]-curves[0, 0]), peak_t=float(grid[np.argmax(curves[:, 0])]),
                                MC_interpretation="Matched-gene-set null; not independent-animal age inference"))
        results = pd.DataFrame(results)
        results["FDR_BH_5paths"] = multipletests(results.p_MC, method="fdr_bh")[1]
        return results


    def candidate_curves(native, masks, output):
        # Set this storage path independently before running. candidates.h5.
        with h5py.File(ROOT, "r") as handle:
            positions = pd.Index(handle["cell_ids"].asstr()[()]).get_indexer(native.index)
            symbols = handle["gene_symbols"].asstr()[()]
            measured = handle["measured"][()][positions]
            common = measured.all(axis=0)
            require(int(common.sum()) == 9, "Candidate shared-source list changed.")
            expression = handle["log10_1p_CPM"][()][positions][:, common]
            symbols = symbols[common]
        for path, member in masks.items():
            metadata, aggregation, design, grid, grid_basis = metacell_design(native, member)
            values = aggregation @ expression
            # Constant genes can be described without inventing an R2 or gene/age P value.
            weight = np.sqrt(metadata.weight.to_numpy())
            coefficient = np.linalg.lstsq(design*weight[:, None], values*weight[:, None], rcond=None)[0]
            group = output.create_group(path)
            text_frame(group.create_group("metacells"), metadata)
            group.create_dataset("gene_symbols", data=np.asarray(symbols, dtype=object), dtype=h5py.string_dtype("utf-8"))
            for name, data in [("observed_metacell_log10_1p_CPM", values), ("fitted_metacell_log10_1p_CPM", design@coefficient),
                               ("t_grid", grid), ("fitted_curve_log10_1p_CPM", grid_basis@coefficient)]:
                group.create_dataset(name, data=data, compression="gzip")
            points = metadata[["bin", "t"]].copy()
            for j, gene in enumerate(symbols):
                points[str(gene)] = values[:, j]
            text_frame(group.create_group("bin_equal_library_points"), points.groupby("bin", sort=True).mean().reset_index())


    #4. =========== Design audit first; scoring curves require an explicit run
    def main():
        parser = argparse.ArgumentParser(description=__doc__)
        parser.add_argument("--run", action="store_true")
        # Set this storage path independently before running. score.h5.
        parser.add_argument("--score", type=Path, default=ROOT)
        parser.add_argument("--model", choices=["primary", "technical"], default="primary")
        parser.add_argument("--candidate-curves", action="store_true")
        parser.add_argument("--independent-score", type=Path)
        args = parser.parse_args(argv)
        args.score = args.score.resolve()
        if args.independent_score is not None:
            args.independent_score = args.independent_score.resolve()
        native, masks = frozen_cells()
        designs, study_rows = [], []
        for path, member in masks.items():
            metadata, _, _, grid, _ = metacell_design(native, member)
            full_span = float(native.loc[member, "t"].max()-native.loc[member, "t"].min())
            designs.append(dict(path=path, native_cells=int(member.sum()), retained_cells=int(metadata.cells.sum()), metacells=len(metadata),
                                libraries=int(metadata.library.nunique()), bins=int(metadata.bin.nunique()), t_min=float(grid[0]), t_max=float(grid[-1]),
                                per_library_total_weight_min=float(metadata.groupby("library").weight.sum().min()),
                                per_library_total_weight_max=float(metadata.groupby("library").weight.sum().max())))
            for study in native.study.unique():
                study_member = member & native.study.eq(study).to_numpy()
                try:
                    sub, _, _, sub_grid, _ = metacell_design(native, study_member)
                    fraction = float((sub_grid[-1]-sub_grid[0])/full_span)
                    adequate = len(sub)>=MIN_STUDY_METACELLS and sub.bin.nunique()>=MIN_STUDY_BINS and sub.library.nunique()>=MIN_STUDY_LIBRARIES and fraction>=MIN_STUDY_SPAN_FRACTION
                    study_rows.append(dict(path=path, study=study, cells=int(study_member.sum()), metacells=len(sub), bins=int(sub.bin.nunique()),
                                           libraries=int(sub.library.nunique()), span_fraction=fraction, adequate_coverage_working_rule=bool(adequate)))
                except ValueError as error:
                    study_rows.append(dict(path=path, study=study, cells=int(study_member.sum()), adequate_coverage_working_rule=False, reason=str(error)))
        # Set this storage path independently before running. continuous_design.tsv.
        pd.DataFrame(designs).to_csv(ROOT, sep="\t", index=False)
        # Set this storage path independently before running. continuous_study_coverage.tsv.
        pd.DataFrame(study_rows).to_csv(ROOT, sep="\t", index=False)
        report = dict(schema="Fig4_frozen_path_continuous_v1", requested_quantile_bins=N_BINS, minimum_cells_per_library_bin=MIN_CELLS,
                      frozen_native_t=True, display_log_compression_used=False, weight="Each metacell 1/number_of_valid_bins_in_its_path_and_library; every library total weight=1",
                      spline="patsy cr(t,df=4)-1 with verified constant span; weighted least squares",
                      statistic="Weighted R2 versus weighted constant; controls use identical cells, bins, libraries, weights and model",
                      MC="(1 + number of 1000 control R2 >= observed R2)/1001; BH over five frozen paths",
                      envelope="2.5th/97.5th percentiles of fitted null controls; not an animal confidence interval",
                      extrapolation=False, group_points="Within each quantile bin, equal average of library metacells",
                      independent_study_working_coverage=dict(min_metacells=MIN_STUDY_METACELLS,min_bins=MIN_STUDY_BINS,min_libraries=MIN_STUDY_LIBRARIES,min_native_t_span_fraction=MIN_STUDY_SPAN_FRACTION),
                      source_coverage_rule_needs_review_before_execution=True, fits_run=False,
                      sources={str(path.relative_to(ROOT)):digest(path) for path in [REGULONS,HI]})
        # Set this storage path independently before running. continuous_preflight.json.
        (ROOT).write_text(json.dumps(report,indent=2),encoding="utf-8")
        print(pd.DataFrame(designs).to_string(index=False), flush=True)
        if not args.run:
            print("Frozen designs checked; no observed/control or candidate curves were fitted.",flush=True)
            return
        # Set this storage path independently before running. continuous_candidate9.h5.
        path = ROOT if args.candidate_curves else TASK/f"continuous_{args.model}.h5"
        require(not path.exists(), "Continuous result already exists; inspect before replacing a current-task cache.")
        with h5py.File(path,"w") as output:
            if args.candidate_curves:
                candidate_curves(native,masks,output.create_group("candidate9"))
            else:
                with h5py.File(args.score,"r") as scores:
                    values=score_values(scores[args.model],native)
                results=run_paths(native,masks,values,output.create_group(args.model))
                results.to_csv(TASK/f"continuous_{args.model}.tsv",sep="\t",index=False)
            if args.independent_score is not None:
                require(not args.candidate_curves,"Independent genetic-score sensitivity is separate from candidate expression curves.")
                sensitivity=[]
                with h5py.File(args.independent_score,"r") as scores:
                    for study in native.study.unique():
                        keep=native.study.eq(study).to_numpy();sub_native=native.loc[keep]
                        mode=scores[f"study_{study}"]
                        values=score_values(mode,sub_native)
                        for path,member in masks.items():
                            decision=next(row for row in study_rows if row["path"]==path and row["study"]==study)
                            if not decision["adequate_coverage_working_rule"]:continue
                            metadata,aggregation,design,grid,grid_basis=metacell_design(sub_native,member[keep])
                            _,curves,r2=fit_values(aggregation@values,metadata,design,grid_basis)
                            sensitivity.append(dict(path=path,study=study,R2=float(r2[0]),p_MC=(1+int((r2[1:]>=r2[0]).sum()))/1001,
                                                    end_minus_start=float(curves[-1,0]-curves[0,0]),peak_t=float(grid[np.argmax(curves[:,0])]),
                                                    t_min=float(grid[0]),t_max=float(grid[-1]),independent_scale=True,
                                                    interpretation="Direction/peak within this study only; no direct score-level comparison between studies"))
                sensitivity_table = pd.DataFrame(sensitivity)
                sensitivity_table["Q_BH_13"] = multipletests(sensitivity_table.p_MC, method="fdr_bh")[1]
                require(len(sensitivity_table) == 13, "Prespecified independent-study testing family changed.")
                # Set this storage path independently before running. continuous_study_sensitivity.tsv.
                sensitivity_table.to_csv(ROOT,sep="\t",index=False)
            result_report=dict(report,fits_run=True,
                               score_source=None if args.candidate_curves else {"path":str(args.score),"sha256":digest(args.score)},
                               independent_score_source=None if args.independent_score is None else {"path":str(args.independent_score),"sha256":digest(args.independent_score)},
                               analysis_type="candidate9_descriptive_mean_log10_1p_CPM" if args.candidate_curves else "genetic_score_matched_control_R2",
                               independent_source_scores_used=bool(args.independent_score is not None),
                               source_coverage_rule_needs_review_before_execution=False,
                               no_gene_or_age_P_values=bool(args.candidate_curves))
            output.create_dataset("manifest_json",data=json.dumps(result_report),dtype=h5py.string_dtype("utf-8"))
    main()

def prepare_candidate_segment_bins(argv=None):
    """Export observed candidate segment means; adaptive resolution is the main display."""
    #1. =========== Fixed shared-source candidates and nine frozen segments
    from pathlib import Path
    import argparse
    import hashlib
    import json

    import h5py
    import numpy as np
    import pandas as pd

    # Set this storage path independently before running.
    TASK = ROOT
    # Set this storage path independently before running. Trajectory dot plot_results-2.h5.
    HI = ROOT
    N_BINS = 20
    MIN_CELLS = 3


    def digest(path):
        with path.open("rb") as stream:
            return hashlib.file_digest(stream,"sha256").hexdigest()


    def main():
        parser=argparse.ArgumentParser(description=__doc__)
        parser.add_argument("--adaptive",action="store_true")
        args=parser.parse_args(argv)
        with h5py.File(HI,"r") as handle:
            cells=pd.Index(handle["cells/_index"].asstr()[()])
            segments=handle["cells/seg"].asstr()[()]
            t=handle["cells/t"][()]
            libraries=handle["cells/library_id"].asstr()[()]
            studies=handle["cells/study_id"].asstr()[()]
            segment_order=handle["candidate/segment_order"].asstr()[()]
        # Set this storage path independently before running. candidates.h5.
        with h5py.File(ROOT,"r") as handle:
            rows=pd.Index(handle["cell_ids"].asstr()[()]).get_indexer(cells)
            measured=handle["measured"][()][rows]
            common=measured.all(axis=0)
            genes=handle["gene_symbols"].asstr()[()][common]
            expression=handle["log10_1p_CPM"][()][rows][:,common]
        assert len(cells)==7355 and len(segment_order)==9 and len(genes)==9 and np.isfinite(expression).all()
        mean=np.full((9,N_BINS,9),np.nan); n_cells=np.zeros((9,N_BINS),dtype=np.int32)
        n_used=np.zeros_like(n_cells); n_libraries=np.zeros_like(n_cells); n_libraries_used=np.zeros_like(n_cells)
        mean_t=np.full((9,N_BINS),np.nan); bounds=np.full((9,N_BINS+1),np.nan); summary=[]; library_rows=[]
        requested_bins=np.full(9,N_BINS,dtype=np.int32);actual_bins=np.full(9,N_BINS,dtype=np.int32)
        for k,segment in enumerate(segment_order):
            positions=np.flatnonzero(segments==segment)
            if args.adaptive:
                requested_bins[k]=min(20,max(3,len(positions)//50))
                bin_id,finite_edges=pd.qcut(t[positions],q=int(requested_bins[k]),labels=False,retbins=True,duplicates="drop")
                bin_id=np.asarray(bin_id,dtype=int);actual_bins[k]=len(finite_edges)-1
                bounds[k,:len(finite_edges)]=finite_edges
            else:
                finite_edges=np.quantile(t[positions],np.linspace(0,1,N_BINS+1));bounds[k]=finite_edges
                # Preserve the original 20-slot diagnostic; exact repeated edges can be empty.
                bin_id=np.searchsorted(finite_edges[1:-1],t[positions],side="left")
            edges=bounds[k]
            for b in range(N_BINS):
                selected=positions[bin_id==b];n_cells[k,b]=len(selected)
                means=[];library_t=[];used_sources=[]
                for library in dict.fromkeys(libraries[selected]):
                    part=selected[libraries[selected]==library]
                    eligible=len(part)>=MIN_CELLS
                    record=dict(segment=str(segment),bin=b,library_id=str(library),study_id=str(studies[part[0]]),
                                n_cells=len(part),eligible=eligible,t_mean=float(t[part].mean()))
                    values=expression[part].mean(axis=0) if eligible else np.full(9,np.nan)
                    record.update(dict(zip(genes,values)))
                    library_rows.append(record)
                    if eligible:
                        means.append(values);library_t.append(float(t[part].mean()));used_sources.extend(studies[part].tolist())
                        n_used[k,b]+=len(part)
                n_libraries[k,b]=len(set(libraries[selected]))
                n_libraries_used[k,b]=len(means)
                if means:
                    mean[k,b]=np.asarray(means).mean(axis=0);mean_t[k,b]=np.mean(library_t)
                row=dict(segment=str(segment),bin=b,n_bins_requested=int(requested_bins[k]),n_bins_after_ties=int(actual_bins[k]),
                         t_interval_min=float(edges[b]),t_interval_max=float(edges[b+1]),
                         mean_t_equal_library=float(mean_t[k,b]),n_cells=int(n_cells[k,b]),n_used_cells=int(n_used[k,b]),
                         n_libraries=int(n_libraries[k,b]),n_used_libraries=int(n_libraries_used[k,b]),
                         valid=bool(means),sources=";".join(sorted(set(studies[selected]))),
                         measured_sources_used=";".join(sorted(set(used_sources))))
                row.update(dict(zip(genes,mean[k,b])));summary.append(row)
        assert n_cells.sum()==7355 and np.array_equal(np.isfinite(mean).all(axis=2),n_libraries_used>0)
        prefix="candidate9_segment_adaptive" if args.adaptive else "candidate9_segment"
        pd.DataFrame(summary).to_csv(TASK/f"{prefix}_summary.tsv",sep="\t",index=False)
        pd.DataFrame(library_rows).to_csv(TASK/f"{prefix}_libraries.tsv",sep="\t",index=False)
        report=dict(schema="Fig4_E4_candidate9_frozen_segment_bins_v1",segments=segment_order.tolist(),genes=genes.tolist(),
                    bins_per_segment=20,cells=7355,minimum_cells_per_library_bin=3,
                    t="Frozen original native metric pseudotime; no display compression",
                    binning="n_bins=min(20,max(3,n_segment_cells//50)); qcut duplicate boundaries drop; unused padded bins NaN" if args.adaptive else "20 nominal native-t quantile intervals per frozen segment; exact repeated boundaries retained as empty slots; ties never split/jittered",
                    adaptive_resolution=args.adaptive,requested_bins_by_segment=dict(zip(segment_order.tolist(),requested_bins.tolist())),
                    effective_quantile_bins_by_segment=dict(zip(segment_order.tolist(),actual_bins.tolist())),
                    expression="Saved candidate9 log10(1+CPM) using audited full available source endogenous totals",
                    aggregation="Within each segment/library/bin average cells (>=3), then equal average of eligible libraries within bin",
                    empty_bins="NaN expression and mean_t; zero eligible-library/cell counts",
                    no_new_gene_sets=True,no_gene_or_age_P_values=True,shared_trunk_displayed_once=True,
                    valid_bins_by_segment={str(segment):int((n_libraries_used[k]>0).sum()) for k,segment in enumerate(segment_order)},
                    retained_cells_by_segment={str(segment):int(n_used[k].sum()) for k,segment in enumerate(segment_order)},
                    # Set this storage path independently before running. candidates.h5.
                    sources={str(path.relative_to(ROOT)):digest(path) for path in [HI,ROOT]})
        # Set this storage path independently before running. candidate9_segment_adaptive.h5.
        path=ROOT if args.adaptive else ROOT
        with h5py.File(path,"w") as handle:
            text=h5py.string_dtype("utf-8")
            for name,value in [("segment_order",segment_order),("gene_symbols",genes)]:handle.create_dataset(name,data=np.asarray(value,dtype=object),dtype=text)
            for name,value in [("mean_log10_1p_CPM",mean),("native_t_quantile_edges",bounds),("mean_t_equal_library",mean_t),
                               ("n_cells",n_cells),("n_used_cells",n_used),("n_libraries",n_libraries),("n_used_libraries",n_libraries_used)]:
                handle.create_dataset(name,data=value)
            handle.create_dataset("requested_n_bins",data=requested_bins)
            handle.create_dataset("effective_quantile_bins",data=actual_bins)
            handle.create_dataset("manifest_json",data=json.dumps(report),dtype=text)
        report["output_sha256"]=digest(path);report["output_bytes"]=path.stat().st_size
        manifest_name="candidate9_segment_adaptive.json" if args.adaptive else "candidate9_segment_bins.json"
        (TASK/manifest_name).write_text(json.dumps(report,indent=2),encoding="utf-8")
        print(json.dumps({key:report[key] for key in ["valid_bins_by_segment","retained_cells_by_segment","output_sha256","output_bytes"]}),flush=True)
    main()

def prepare_genetic_segment_profile(argv=None):
    """Export observed and matched-control means on the same adaptive segment bins."""
    #1. =========== Accepted observed/control score caches; no bin-level P values
    from pathlib import Path
    import argparse
    import hashlib
    import json

    import h5py
    import numpy as np
    import pandas as pd

    # Set this storage path independently before running.
    TASK = ROOT
    # Set this storage path independently before running. Trajectory dot plot_results-2.h5.
    HI = ROOT


    def digest(path):
        with path.open("rb") as stream:
            return hashlib.file_digest(stream,"sha256").hexdigest()


    def main():
        parser=argparse.ArgumentParser(description=__doc__)
        # Set this storage path independently before running. score.h5.
        parser.add_argument("--score",type=Path,default=ROOT)
        parser.add_argument("--model",choices=["primary","technical"],default="primary")
        parser.add_argument("--run",action="store_true")
        args=parser.parse_args(argv)
        args.score=args.score.resolve()
        # Set this storage path independently before running. candidate9_segment_adaptive.h5.
        bin_cache=ROOT
        with h5py.File(bin_cache,"r") as f:
            order=f["segment_order"].asstr()[()];edges=f["native_t_quantile_edges"][()]
            requested_bins=f["requested_n_bins"][()];effective_bins=f["effective_quantile_bins"][()]
            expected_cells=f["n_used_cells"][()];expected_libraries=f["n_used_libraries"][()]
        with h5py.File(HI,"r") as f:
            cells=pd.Index(f["cells/_index"].asstr()[()]);segment=f["cells/seg"].asstr()[()]
            t=f["cells/t"][()];library=f["cells/library_id"].asstr()[()]
        if not args.run:
            print("Frozen nine-segment adaptive quantile-bin design available; no score profile computed.",flush=True)
            return
        with h5py.File(args.score,"r") as f:
            g=f[args.model];indices=pd.Index(g["cell_ids"].asstr()[()]).get_indexer(cells)
            assert (indices>=0).all()
            columns=g["score_columns"].asstr()[()].tolist()
            observed=g["scores"][()][:,columns.index("norm_score")][indices]
            controls=g["norm_control"][()][indices]
        assert controls.shape==(7355,1000) and np.isfinite(controls).all() and np.isfinite(observed).all()
        values=np.column_stack([observed,controls]);mean=np.full((9,20,1001),np.nan)
        n_cells=np.zeros((9,20),dtype=np.int32);n_libraries=np.zeros_like(n_cells)
        libraries=[];summary=[]
        for k,name in enumerate(order):
            positions=np.flatnonzero(segment==name);finite_edges=edges[k,np.isfinite(edges[k])]
            assert len(finite_edges)==effective_bins[k]+1
            bin_id=np.searchsorted(finite_edges[1:-1],t[positions],side="left")
            for b in range(20):
                selected=positions[bin_id==b];averages=[]
                for lib in dict.fromkeys(library[selected]):
                    part=selected[library[selected]==lib]
                    if len(part)<3:continue
                    averages.append(values[part].mean(axis=0));n_cells[k,b]+=len(part)
                    libraries.append(dict(segment=str(name),bin=b,library=str(lib),cells=len(part),t=float(t[part].mean())))
                n_libraries[k,b]=len(averages)
                if averages:mean[k,b]=np.asarray(averages).mean(axis=0)
                summary.append(dict(segment=str(name),bin=b,n_used_cells=int(n_cells[k,b]),n_used_libraries=int(n_libraries[k,b]),
                                    observed_equal_library_mean=float(mean[k,b,0]),
                                    null_mean=float(mean[k,b,1:].mean()) if averages else np.nan,
                                    null_q025=float(np.quantile(mean[k,b,1:],.025)) if averages else np.nan,
                                    null_q975=float(np.quantile(mean[k,b,1:],.975)) if averages else np.nan))
        assert np.array_equal(n_cells,expected_cells) and np.array_equal(n_libraries,expected_libraries)
        envelope=np.full((9,20,2),np.nan);valid=n_libraries>0
        envelope[valid]=np.quantile(mean[valid,1:],[.025,.975],axis=1).T
        path=TASK/f"segment_score_adaptive_{args.model}.h5"
        assert not path.exists(),"Inspect the existing current-task score profile before replacing it"
        manifest=dict(schema="Fig4_frozen_segment_genetic_profile_v1",model=args.model,segments=order.tolist(),cells=7355,
                      observed="Mean of cell norm_score within each segment/library/bin with >=3 cells, then equal library mean per bin",
                      controls="Each of the same 1000 normalized controls aggregated with identical cells/libraries/bins/weights",
                      null_envelope="95% fitted-free matched-control mean envelope, not animal CI",bin_P_values_computed=False,
                      shared_trunk_displayed_once=True,unconstrained_smooth_used=False,
                      adaptive_resolution="min(20,max(3,n_segment_cells//50)), qcut duplicate edges dropped, unused padded bins NaN",
                      sources={str(path.relative_to(ROOT)):digest(path) for path in [args.score,HI,bin_cache]})
        with h5py.File(path,"w") as f:
            text=h5py.string_dtype("utf-8");f.create_dataset("segment_order",data=np.asarray(order,dtype=object),dtype=text)
            for key,val in [("observed_equal_library_mean",mean[:,:,0]),("control_equal_library_mean",mean[:,:,1:]),
                            ("null_envelope95",envelope),("n_used_cells",n_cells),("n_used_libraries",n_libraries),
                            ("native_t_quantile_edges",edges)]:f.create_dataset(key,data=val,compression="gzip")
            f.create_dataset("requested_n_bins",data=requested_bins);f.create_dataset("effective_quantile_bins",data=effective_bins)
            f.create_dataset("manifest_json",data=json.dumps(manifest),dtype=text)
        pd.DataFrame(summary).to_csv(TASK/f"segment_score_adaptive_{args.model}.tsv",sep="\t",index=False)
        pd.DataFrame(libraries).to_csv(TASK/f"segment_score_adaptive_{args.model}_libraries.tsv",sep="\t",index=False)
        print(json.dumps(dict(output=str(path),valid_bins=int(valid.sum()),bin_P_values_computed=False,sha256=digest(path))),flush=True)
    main()

#7. =========== Separate cochlear assay preparation and scoring
def prepare_cochlear_inputs():
    """Prepare separate, traceable cochlear references without normalization or scoring."""

    #1. =========== Shared configuration
    from pathlib import Path
    import hashlib
    import json
    import time

    import h5py
    import numpy as np
    import pandas as pd
    from scipy import sparse

    # Set this storage path independently before running.
    SCRATCH = ROOT
    # Set this storage path independently before running. mouse_human_homologs.txt.
    HOMOLOGS = ROOT
    HOMOLOG_SHA256 = "2d0ef4953bc1654f3835dba02c15c836ffe9f2162631069a766d898959016fb3"
    GENE_BLOCK = 512
    RANDOM_SEED = 42
    SOURCES = {
        "snRNA": {
            # Set this storage path independently before running. b0de5e98.h5ad.
            "path": ROOT,
            "share_id": "b0de5e98",
            "label_key": "Classification",
            "expected_cells": 21442,
        },
        "scRNA": {
            # Set this storage path independently before running. 5477fbe6.h5ad.
            "path": ROOT,
            "share_id": "5477fbe6",
            "label_key": "P8-12-20",
            "expected_cells": 36893,
        },
    }
    BLOOD = {
        "B cells", "Erythrocytes", "Neutrophils", "Macrophages", "Monocytes",
        "T cells", "NK cells", "Mast cells",
    }
    UNKNOWN = {"SS1", "SS2", "New cell type", "NA circulating", "NA SCs", "NA (sc)"}
    UTF8 = h5py.string_dtype("utf-8")


    def read_column(obj):
        if isinstance(obj, h5py.Dataset):
            return obj.asstr()[:] if obj.dtype.kind in "OSU" else obj[:]
        categories = read_column(obj["categories"])
        codes = obj["codes"][:]
        return np.asarray([categories[k] if k >= 0 else None for k in codes], dtype=object)


    def write_array(group, name, values):
        values = np.asarray(values)
        if values.dtype.kind in "OSU":
            assert not pd.isna(values).any(), name
            group.create_dataset(name, data=values.astype(object), dtype=UTF8)
        else:
            group.create_dataset(name, data=values, compression="gzip", compression_opts=1, shuffle=True)


    def write_table(group, frame):
        group.attrs["column_order"] = json.dumps(list(frame.columns))
        for name in frame:
            write_array(group, name, frame[name].to_numpy())


    def digest_file(path):
        digest = hashlib.sha256()
        with path.open("rb") as stream:
            for block in iter(lambda: stream.read(4 * 1024 * 1024), b""):
                digest.update(block)
        return digest.hexdigest()


    #2. =========== Establish the measured and one-to-one homologous axes
    assert digest_file(HOMOLOGS) == HOMOLOG_SHA256
    homolog = pd.read_csv(HOMOLOGS, sep="\t")
    assert len(homolog) == 16466 and not homolog.isna().any().any()
    assert homolog.MOUSE_GENE_SYM.is_unique and homolog.HUMAN_GENE_SYM.is_unique
    mouse_to_human = dict(zip(homolog.MOUSE_GENE_SYM, homolog.HUMAN_GENE_SYM))
    gene_axes = {}
    for technology, config in SOURCES.items():
        with h5py.File(config["path"], "r") as source:
            frame = pd.DataFrame({
                "source_gene_id": read_column(source["var/ensembl_ID"]),
                "mouse_symbol": read_column(source["var/gene_symbol"]),
            })
            assert frame.source_gene_id.is_unique and frame.mouse_symbol.is_unique
            frame[f"{technology}_var_index"] = np.arange(len(frame), dtype=np.int64)
            gene_axes[technology] = frame

    common = gene_axes["scRNA"].merge(
        gene_axes["snRNA"], on=["source_gene_id", "mouse_symbol"],
        how="inner", sort=False, validate="one_to_one",
    )
    common = common.loc[~common.mouse_symbol.str.lower().str.startswith("mt-")].reset_index(drop=True)
    assert len(common) == 36888
    common["human_symbol"] = common.mouse_symbol.map(mouse_to_human)
    mapped_mask = common.human_symbol.notna().to_numpy()
    mapped = common.loc[mapped_mask].reset_index(drop=True)
    assert len(mapped) == 15722 and mapped.human_symbol.is_unique
    assert mapped.mouse_symbol.is_unique and mapped.source_gene_id.is_unique
    print("GENE_AXES", json.dumps({"common_nonmt": len(common), "mapped_one_to_one": len(mapped)}), flush=True)


    #3. =========== Filter only explicit author classes and stream CSC blocks
    for technology, config in SOURCES.items():
        started = time.monotonic()
        destination = SCRATCH / f"E1-{technology}.h5"
        partial = destination.with_suffix(".h5.part")
        assert not destination.exists() and not partial.exists(), destination
        with h5py.File(config["path"], "r") as source:
            obs_original = pd.DataFrame({key: read_column(source["obs"][key]) for key in source["obs"]})
            label_raw = obs_original[config["label_key"]]
            label = label_raw.fillna("<missing>")
            is_sc = label.str.endswith(" (sc)")
            is_blood = label.isin(BLOOD)
            is_unknown = label_raw.isna() | label.isin(UNKNOWN) | label.str.strip().eq("")
            reason = np.select([is_sc, is_blood, is_unknown], ["sc_contaminant", "blood", "unknown"], default="retain")
            keep = np.flatnonzero(reason == "retain")
            assert len(keep) == config["expected_cells"]
            obs = obs_original.iloc[keep].reset_index(drop=True).copy()
            assert obs[config["label_key"]].notna().all() and obs["_index"].is_unique
            obs["source_obs_index"] = keep
            obs["cell_type_author"] = obs[config["label_key"]]
            obs["sample_id_author"] = obs["Sample name"]
            obs["technology"] = technology
            obs["source_share_id"] = config["share_id"]
            obs["stage_source"] = "P8" if technology == "snRNA" else obs["Age"]

            x = source["X"]
            assert x.attrs["encoding-type"] == "csc_matrix"
            source_shape = tuple(x.attrs["shape"])
            ptr = x["indptr"][:]
            source_columns = common[f"{technology}_var_index"].to_numpy()
            read_order = np.argsort(source_columns)
            read_columns = source_columns[read_order]
            assert np.all(np.diff(read_columns) > 0)
            total_common = np.zeros(len(keep), dtype=np.float64)
            detected_common = np.zeros(len(keep), dtype=np.int64)
            mapped_blocks = []
            mapped_positions = []
            for start in range(0, len(common), GENE_BLOCK):
                stop = min(start + GENE_BLOCK, len(common))
                columns = read_columns[start:stop]
                background_positions = read_order[start:stop]
                first, last = int(columns[0]), int(columns[-1])
                low, high = int(ptr[first]), int(ptr[last + 1])
                block = sparse.csc_matrix((
                    x["data"][low:high], x["indices"][low:high],
                    ptr[first:last + 2] - low,
                ), shape=(source_shape[0], last - first + 1))
                assert block.has_canonical_format
                block = block[keep, :][:, columns - first]
                block.eliminate_zeros()
                assert np.isfinite(block.data).all() and (block.data >= 0).all()
                total_common += np.asarray(block.sum(axis=1)).ravel()
                detected_common += block.getnnz(axis=1)
                local_mapped = mapped_mask[background_positions]
                if local_mapped.any():
                    mapped_blocks.append(block[:, local_mapped].tocsr())
                    mapped_positions.append(background_positions[local_mapped])
            matrix = sparse.hstack(mapped_blocks, format="csr")
            mapped_positions = np.concatenate(mapped_positions)
            assert np.array_equal(np.sort(mapped_positions), np.flatnonzero(mapped_mask))
            matrix = matrix[:, np.argsort(mapped_positions)].tocsr()
            del mapped_blocks, mapped_positions, block
            matrix.sort_indices()
            assert matrix.shape == (len(keep), len(mapped)) and matrix.has_canonical_format
            assert matrix.dtype == np.float64 and (total_common > 0).all()
            total_mapped = np.asarray(matrix.sum(axis=1)).ravel()
            detected_mapped = matrix.getnnz(axis=1)
            assert np.all(total_mapped <= total_common + 1e-6)
            obs["total_common_nonmt"] = total_common
            obs["detected_common_nonmt"] = detected_common
            obs["total_mapped_common_nonmt"] = total_mapped
            obs["detected_mapped_common_nonmt"] = detected_mapped

            #4. =========== Preserve every mapped measured gene, including all-zero genes
            gene_detected = matrix.getnnz(axis=0)
            gene_total = np.asarray(matrix.sum(axis=0)).ravel()
            var = pd.DataFrame({
                "mouse_symbol": mapped.mouse_symbol,
                "human_symbol": mapped.human_symbol,
                "source_gene_id": mapped.source_gene_id,
                "source_var_index": mapped[f"{technology}_var_index"],
                "common_nonmt_position": np.flatnonzero(mapped_mask),
                "detected_cells": gene_detected,
                "gene_total": gene_total,
                "measured_zero_after_cell_filter": gene_detected == 0,
            })
            count_table = pd.crosstab(obs.cell_type_author, obs.sample_id_author)
            filter_table = pd.DataFrame({"label": label, "reason": reason}).groupby(["reason", "label"], dropna=False).size().rename("n").reset_index()
            sample_filter = pd.crosstab(obs_original["Sample name"], reason)

            #5. =========== Write an explicit custom HDF5 schema
            with h5py.File(partial, "x") as output:
                output.attrs.update({
                    "format": "fig4_e1_reference_v1", "technology": technology,
                    "source_path": str(config["path"]), "source_share_id": config["share_id"],
                    "author_label_key": config["label_key"],
                    "expression_scale": "Partek raw estimated counts; float64; unnormalized and not log-transformed",
                    "common_nonmt_genes": len(common), "mapped_one_to_one_genes": len(mapped),
                    "normalization_denominator": "obs/total_common_nonmt",
                    "homolog_path": str(HOMOLOGS), "homolog_sha256": HOMOLOG_SHA256,
                    "zero_gene_policy": "Preserved measured zeros; no gene detection intersection or all-zero filtering",
                    "cell_filter_policy": "Exclude author (sc), blood/immune, SS1/SS2/New cell type/NA/missing; no re-clustering",
                    "normalization_status": "not_applied", "scdrs_status": "not_run",
                })
                counts = output.create_group("X")
                counts.attrs["encoding"] = "csr_matrix"
                counts.attrs["shape"] = matrix.shape
                for key, values in [("data", matrix.data), ("indices", matrix.indices), ("indptr", matrix.indptr)]:
                    write_array(counts, key, values)
                write_table(output.create_group("obs"), obs)
                write_table(output.create_group("var"), var)
                background = common[["source_gene_id", "mouse_symbol", f"{technology}_var_index"]].copy()
                background = background.rename(columns={f"{technology}_var_index": "source_var_index"})
                background["mappable_one_to_one"] = mapped_mask
                write_table(output.create_group("common_background"), background)
                diagnostic = output.create_group("diagnostics")
                write_table(diagnostic.create_group("filter_label_counts"), filter_table)
                cross = diagnostic.create_group("type_by_sample")
                write_array(cross, "cell_type", count_table.index.to_numpy())
                write_array(cross, "sample", count_table.columns.to_numpy())
                write_array(cross, "counts", count_table.to_numpy())
                sample_status = diagnostic.create_group("sample_filter_counts")
                write_array(sample_status, "sample", sample_filter.index.to_numpy())
                write_array(sample_status, "status", sample_filter.columns.to_numpy())
                write_array(sample_status, "counts", sample_filter.to_numpy())

            #6. =========== Independently verify saved CSR, original metadata and source columns
            with h5py.File(partial, "r") as output:
                saved = sparse.csr_matrix((output["X/data"][:], output["X/indices"][:], output["X/indptr"][:]), shape=tuple(output["X"].attrs["shape"]))
                assert saved.has_canonical_format and saved.shape == matrix.shape
                assert np.array_equal(saved.data, matrix.data) and np.array_equal(saved.indices, matrix.indices) and np.array_equal(saved.indptr, matrix.indptr)
                for key in obs_original:
                    expected = obs_original[key].to_numpy()[keep]
                    actual = read_column(output["obs"][key])
                    assert np.array_equal(expected, actual), (technology, key)
                assert np.array_equal(read_column(output["var/human_symbol"]), mapped.human_symbol.to_numpy())
                assert np.allclose(np.asarray(saved.sum(axis=1)).ravel(), read_column(output["obs/total_mapped_common_nonmt"]), rtol=1e-12, atol=1e-8)
                assert np.array_equal(saved.getnnz(axis=0) == 0, read_column(output["var/measured_zero_after_cell_filter"]))
                rng = np.random.default_rng(RANDOM_SEED)
                probes = set(rng.choice(len(mapped), size=12, replace=False).tolist())
                probes.update(np.flatnonzero(mapped.mouse_symbol.isin(["Slc26a5", "Ikzf2", "Otof", "Slc17a8", "Tubb3"])).tolist())
                probes = np.asarray(sorted(probes))
                observed = saved[:, probes].toarray()
                for local, feature in enumerate(probes):
                    source_feature = int(mapped.iloc[feature][f"{technology}_var_index"])
                    low, high = int(ptr[source_feature]), int(ptr[source_feature + 1])
                    expected = np.zeros(source_shape[0], dtype=np.float64)
                    expected[x["indices"][low:high]] = x["data"][low:high]
                    assert np.array_equal(observed[:, local], expected[keep]), (technology, feature)
                assert int(output["diagnostics/type_by_sample/counts"][:].sum()) == len(keep)
            del saved, matrix
        partial.rename(destination)
        summary = {
            "path": str(destination), "bytes": destination.stat().st_size,
            "sha256": digest_file(destination), "cells": len(obs), "types": len(count_table),
            "samples": len(count_table.columns), "mapped_genes": len(mapped),
            "measured_zero_genes": int((gene_detected == 0).sum()),
            "common_total_min_median_max": np.quantile(total_common, [0, .5, 1]).tolist(),
            "retained_by_sample": obs.sample_id_author.value_counts().sort_index().to_dict(),
            "excluded_labels": filter_table.loc[filter_table.reason.ne("retain")].to_dict("records"),
            "seconds": round(time.monotonic() - started, 2),
        }
        print(technology, json.dumps(summary, ensure_ascii=False), flush=True)

    print("PREPARATION_COMPLETE_NO_NORMALIZATION_OR_SCDRS", flush=True)

def score_cochlear():
    """Separate E1 scDRS analyses; requires the supplied fixed MAGMA top-1000 .gs."""

    #1. =========== Shared configuration and pinned imports
    from pathlib import Path
    import argparse
    import hashlib
    import json
    import os
    import sys
    import time

    # Set this storage path independently before running.
    SCRATCH = ROOT
    # Set this storage path independently before running.
    PACKAGES = ROOT
    COMMIT = "644a5c7900d3c8d84c068725b47dd81e0db22ffb"
    SEED = 42
    N_CTRL = 1000
    CP_TARGET = 10000
    FDR_THRESHOLDS = [0.05, 0.1, 0.2]
    INPUT_SHA = {
        "snRNA": "6ff2183d095ed0ba7f0bda9585228b2683a5dc5f67433f7321fb1211fde33440",
        "scRNA": "623f368712224dcda92830b6debd54b0bfa593f7a2666b9cce6b05da15249fe2",
    }
    os.environ["TQDM_DISABLE"] = "1"
    os.environ["PYTHONDONTWRITEBYTECODE"] = "1"
    # Set this storage path independently before running.
    os.environ["NUMBA_CACHE_DIR"] = str(ROOT)
    # Set this storage path independently before running.
    os.environ["MPLCONFIGDIR"] = str(ROOT)
    for name in ["OMP_NUM_THREADS", "MKL_NUM_THREADS", "OPENBLAS_NUM_THREADS", "NUMEXPR_NUM_THREADS"]:
        os.environ[name] = "1"
    os.environ["NUMBA_NUM_THREADS"] = "1"
    sys.dont_write_bytecode = True
    sys.path.insert(0, str(PACKAGES))

    import anndata
    import h5py
    import numpy as np
    import pandas as pd
    from scipy import sparse
    from statsmodels.stats.multitest import multipletests
    from threadpoolctl import threadpool_limits
    from tqdm import tqdm
    import scdrs


    def sha256(path):
        digest = hashlib.sha256()
        with Path(path).open("rb") as stream:
            for block in iter(lambda: stream.read(4 * 1024 * 1024), b""):
                digest.update(block)
        return digest.hexdigest()


    def validate_sources():
        # Set this storage path independently before running. source_manifest.json.
        manifest = json.loads((ROOT).read_text())
        assert manifest["github_commit"] == COMMIT
        assert scdrs.__version__ == "1.0.4"
        # Set this storage path independently before running.
        assert Path(scdrs.__file__).resolve().parent == ROOT
        for record in manifest["files"]:
            assert sha256(PACKAGES / record["path"]) == record["sha256"], record["path"]
        bar = tqdm(range(0))
        assert bar.disable, "TQDM_DISABLE was not honored by this runtime"
        bar.close()


    def read_frame(group):
        columns = json.loads(group.attrs["column_order"])
        return pd.DataFrame({
            key: group[key].asstr()[:] if group[key].dtype.kind == "O" else group[key][:]
            for key in columns
        })


    def write_frame(group, frame):
        group.attrs["column_order"] = json.dumps(list(frame.columns))
        group.attrs["index_name"] = str(frame.index.name or "index")
        values = frame.copy()
        if "_index" in values:
            assert np.array_equal(values["_index"].astype(str).to_numpy(), frame.index.astype(str).to_numpy())
        else:
            values["_index"] = frame.index.astype(str)
        for key in values:
            array = values[key].to_numpy()
            if array.dtype.kind in "OSU":
                assert not pd.isna(array).any(), key
                group.create_dataset(key, data=np.asarray([str(x) for x in array], dtype=object), dtype=h5py.string_dtype("utf-8"))
            else:
                group.create_dataset(key, data=array, compression="gzip", compression_opts=1, shuffle=True)


    #2. =========== Association-only implementation of the pinned official formulas
    def association_only(adata, full_score, group_col="cell_type_author"):
        """Reproduce method.py lines 763-817 at COMMIT; no graph or heterogeneity.

        The official function has no heterogeneity switch. This explicit function
        retains its sorted-cell alignment, cell FDR counts and 95th-quantile MC test.
        It is tested against the complete official function on its 30-cell toy data.
        """
        assert group_col in adata.obs
        cell_list = sorted(set(adata.obs_names) & set(full_score.index))
        assert len(cell_list) == adata.n_obs == len(full_score)
        control_list = [key for key in full_score if key.startswith("ctrl_norm_score")]
        n_ctrl = len(control_list)
        assert n_ctrl == N_CTRL
        reg = adata.obs.loc[cell_list, [group_col]].copy()
        reg = reg.join(full_score.loc[cell_list, ["norm_score"] + control_list + ["pval"]])
        group_list = sorted(set(adata.obs[group_col]))
        columns = ["n_cell", "n_ctrl", "assoc_mcp", "assoc_mcz", "hetero_mcp", "hetero_mcz"]
        columns.extend(f"n_fdr_{threshold}" for threshold in FDR_THRESHOLDS)
        result = pd.DataFrame(index=group_list, columns=columns, dtype=np.float32)
        result.index.name = "group"
        cell_fdr = pd.DataFrame({"fdr": multipletests(reg["pval"].values, method="fdr_bh")[1]}, index=reg.index)
        for group in group_list:
            cells = list(reg.index[reg[group_col] == group])
            result.loc[group, ["n_cell", "n_ctrl"]] = [len(cells), n_ctrl]
            for threshold in FDR_THRESHOLDS:
                result.loc[group, f"n_fdr_{threshold}"] = (cell_fdr.loc[cells, "fdr"].values < threshold).sum()
            score_q95 = np.quantile(reg.loc[cells, "norm_score"], 0.95)
            controls_q95 = np.quantile(reg.loc[cells, control_list], 0.95, axis=0)
            mc_p = ((controls_q95 >= score_q95).sum() + 1) / (controls_q95.shape[0] + 1)
            mc_z = (score_q95 - controls_q95.mean()) / controls_q95.std()
            result.loc[group, ["assoc_mcp", "assoc_mcz"]] = [mc_p, mc_z]
        return result


    def self_test():
        """Use a real kNN only on the official toy; no project objects are loaded."""
        from sklearn.neighbors import kneighbors_graph

        # Set this storage path independently before running. toydata_mouse.h5ad.
        toy = anndata.read_h5ad(ROOT)
        assert toy.shape == (30, 2500)
        toy.X = toy.X.astype(np.float32).tocsr()
        # Set this storage path independently before running.
        gene_sets = scdrs.util.load_gs(str(ROOT))
        genes, weights = next(iter(gene_sets.values()))
        scdrs.preprocess(toy, cov=None, adj_prop=None)
        full = scdrs.score_cell(toy, genes, gene_weight=weights, n_ctrl=N_CTRL, random_seed=SEED,
                                return_ctrl_norm_score=True, verbose=False)
        graph = kneighbors_graph(toy.X, n_neighbors=5, mode="connectivity", include_self=False, n_jobs=1)
        toy.obsp["connectivities"] = graph.maximum(graph.T).tocsr()
        assert toy.obsp["connectivities"].nnz > 0
        official = scdrs.method.downstream_group_analysis(toy, full, ["cell_type"], fdr_thresholds=FDR_THRESHOLDS)["cell_type"]
        ours = association_only(toy, full, group_col="cell_type")
        columns = [key for key in official if not key.startswith("hetero_")]
        pd.testing.assert_frame_equal(ours[columns], official[columns], check_exact=True)
        assert ours[["hetero_mcp", "hetero_mcz"]].isna().all().all()
        print("TOY_ASSOCIATION_EXACT_MATCH", json.dumps({"cells": 30, "genes": 2500, "controls": N_CTRL,
              "groups": len(ours), "columns": columns, "real_knn_edges": int(graph.maximum(graph.T).nnz)}), flush=True)


    #3. =========== Require the fixed supplied weighted human top-1000 list
    def load_trait(path, requested_trait=None, rank_manifest=None, magma_qc=None):
        path = Path(path).resolve(strict=True)
        rank_path = Path(rank_manifest).resolve(strict=True) if rank_manifest else path.with_suffix(".tsv").resolve(strict=True)
        # Set this storage path independently before running. gene-result-QC.json.
        qc_path = Path(magma_qc).resolve(strict=True) if magma_qc else ROOT
        qc = json.loads(qc_path.read_text(encoding="utf-8"))
        assert int(qc["original_rank_n"]) == 1000
        file_records = {Path(record["path"]).resolve(): record for record in qc["files"].values()}
        for checked_path in [path, rank_path]:
            record = file_records[checked_path]
            assert checked_path.stat().st_size == record["bytes"]
            assert sha256(checked_path) == record["sha256"]
        ranked = pd.read_csv(rank_path, sep="\t", dtype={"GENE": str}, keep_default_na=False, float_precision="round_trip")
        required = {"original_rank", "GENE", "symbol", "mapping_status", "ZSTAT"}
        assert required <= set(ranked)
        assert len(ranked) == 1000 and ranked.GENE.is_unique
        assert ranked.GENE.str.fullmatch(r"\d+").all()
        assert np.array_equal(ranked.original_rank.to_numpy(), np.arange(1, 1001))
        ordered = ranked.assign(entrez_numeric=ranked.GENE.astype(np.int64)).sort_values(
            ["ZSTAT", "entrez_numeric"], ascending=[False, True], kind="stable",
        )
        assert ranked.GENE.tolist() == ordered.GENE.tolist()
        native_candidates = [entry for entry in file_records if entry.name.endswith((".genes.out", ".genes.out.txt"))]
        assert len(native_candidates) == 1, "Require the native complete MAGMA gene result in QC"
        native_path = native_candidates[0]
        assert sha256(native_path) == file_records[native_path]["sha256"]
        native = pd.read_csv(native_path, sep=r"\s+", comment="#", dtype={"GENE": str}, float_precision="round_trip")
        native["ZSTAT"] = pd.to_numeric(native.ZSTAT, errors="coerce")
        native = native.loc[native.GENE.str.fullmatch(r"\d+") & np.isfinite(native.ZSTAT)].copy()
        assert native.GENE.is_unique
        native["entrez_numeric"] = native.GENE.astype(np.int64)
        native_top = native.sort_values(["ZSTAT", "entrez_numeric"], ascending=[False, True], kind="stable").head(1000)
        assert ranked.GENE.tolist() == native_top.GENE.tolist(), "The manifest must be the full-MAGMA top 1000 before mapping"
        assert np.array_equal(ranked.ZSTAT.to_numpy(), native_top.ZSTAT.to_numpy())
        mapped = ranked.loc[ranked.mapping_status.eq("HGNC_unique_entrez")]
        assert len(mapped) > 0 and mapped.symbol.ne("").all() and mapped.symbol.is_unique
        raw = pd.read_csv(path, sep="\t")
        assert list(raw.columns) == ["TRAIT", "GENESET"] and raw.TRAIT.is_unique
        assert not raw.isna().any().any()
        if requested_trait is None:
            assert len(raw) == 1, "Specify --trait when the supplied file contains multiple traits"
            trait = str(raw.TRAIT.iloc[0])
        else:
            trait = requested_trait
            assert trait in set(raw.TRAIT)
        tokens = raw.loc[raw.TRAIT.eq(trait), "GENESET"].iloc[0].split(",")
        assert 0 < len(tokens) <= 1000, "HGNC mapping may reduce the original fixed top-1000; never refill"
        parsed = [token.split(":") for token in tokens]
        assert all(len(token) == 2 for token in parsed), "Require weighted human gene:weight entries"
        genes = [token[0] for token in parsed]
        weights = np.asarray([float(token[1]) for token in parsed], dtype=np.float64)
        assert len(set(genes)) == len(genes) and all(genes)
        assert np.isfinite(weights).all() and (weights > 0).all()
        assert len(genes) == int(qc["weighted_gs_n"]) == len(mapped)
        assert genes == mapped.symbol.tolist(), "Only uniquely mapped genes from the original top 1000 are allowed"
        assert np.array_equal(weights, mapped.ZSTAT.to_numpy()), "Weights must retain the native MAGMA ZSTAT"
        official_genes, official_weights = scdrs.util.load_gs(str(path))[trait]
        assert genes == official_genes and np.array_equal(weights, official_weights)
        provenance = {
            "rank_manifest_path": str(rank_path), "rank_manifest_sha256": sha256(rank_path),
            "magma_qc_path": str(qc_path), "magma_qc_sha256": sha256(qc_path),
            "native_genes_path": str(native_path), "native_genes_sha256": sha256(native_path),
            "magma_qc_json": json.dumps(qc),
        }
        return trait, pd.DataFrame({"human_symbol": genes, "input_weight": weights}), path, ranked, provenance


    #4. =========== Normalize each technology on its measured common-gene totals
    def load_reference(technology):
        path = SCRATCH / f"E1-{technology}.h5"
        assert sha256(path) == INPUT_SHA[technology]
        with h5py.File(path, "r") as source:
            assert source.attrs["normalization_status"] == "not_applied"
            assert source.attrs["common_nonmt_genes"] == 36888
            assert source.attrs["normalization_denominator"] == "obs/total_common_nonmt"
            matrix = sparse.csr_matrix((source["X/data"].astype(np.float32)[:], source["X/indices"][:], source["X/indptr"][:]),
                                       shape=tuple(source["X"].attrs["shape"]))
            obs = read_frame(source["obs"]).set_index("_index", drop=False)
            measured = read_frame(source["var"])
        assert matrix.dtype == np.float32 and matrix.has_canonical_format
        denominator = obs.total_common_nonmt.to_numpy(dtype=np.float64)
        assert np.isfinite(denominator).all() and (denominator > 0).all()
        factor = (CP_TARGET / denominator).astype(np.float32)
        matrix.data *= np.repeat(factor, np.diff(matrix.indptr))
        np.log1p(matrix.data, out=matrix.data)
        expressed = matrix.getnnz(axis=0) > 0
        assert np.array_equal(~expressed, measured.measured_zero_after_cell_filter.to_numpy())
        measured["used_for_technology_background"] = expressed
        measured = measured.set_index("human_symbol", drop=False)
        logged_detected = pd.Series(np.log1p(obs.detected_common_nonmt.to_numpy(dtype=np.float32)), index=obs.index)
        centered = logged_detected - logged_detected.groupby(obs.sample_id_author, observed=True).transform("mean")
        obs["cov_library_centered_log1p_detected"] = centered.astype(np.float32)
        assert centered.groupby(obs.sample_id_author, observed=True).mean().abs().max() < 1e-5
        adata = anndata.AnnData(matrix[:, expressed].tocsr(), obs=obs.copy(), var=measured.loc[expressed].copy())
        assert adata.X.dtype == np.float32 and not adata.obsp
        return adata, measured, path


    #5. =========== Score, group MC testing and descriptive library consistency
    def score_reference(technology, mode, trait, trait_table, gs_path, ranked, provenance):
        started = time.monotonic()
        stem = SCRATCH / f"E1-{technology}-{mode}"
        paths = {
            "h5": Path(str(stem) + ".h5"),
            "group": Path(str(stem) + "-groups.tsv"),
            "sample": Path(str(stem) + "-sample-type.tsv"),
            "consistency": Path(str(stem) + "-library-consistency.tsv"),
        }
        assert all(not path.exists() for path in paths.values()), "Refuse to overwrite existing results"
        adata, measured, input_path = load_reference(technology)
        coverage = trait_table.copy()
        coverage["in_measured_mapped_axis"] = coverage.human_symbol.isin(measured.index)
        coverage["used_for_score"] = coverage.human_symbol.isin(adata.var_names)
        coverage["exclusion_reason"] = np.select(
            [~coverage.in_measured_mapped_axis, ~coverage.used_for_score],
            ["not_in_measured_mapped_axis", "all_zero_in_technology"], default="used",
        )
        selected = coverage.loc[coverage.used_for_score]
        assert len(selected) > 0
        cov = None if mode == "primary" else adata.obs[["cov_library_centered_log1p_detected"]].copy()
        print("START", technology, mode, json.dumps({"cells": adata.n_obs, "background_genes": adata.n_vars,
              "ranked_entrez_genes": len(ranked), "mapped_human_genes": len(trait_table),
              "trait_measured": int(coverage.in_measured_mapped_axis.sum()), "trait_used": len(selected)}), flush=True)
        scdrs.preprocess(adata, cov=cov, adj_prop=None)
        full = scdrs.score_cell(adata, selected.human_symbol.tolist(), gene_weight=selected.input_weight.tolist(),
                                ctrl_match_key="mean_var", n_ctrl=N_CTRL, n_genebin=200, weight_opt="vs",
                                random_seed=SEED, return_ctrl_norm_score=True, return_ctrl_raw_score=False, verbose=False)
        assert full.dtypes.eq(np.dtype("float32")).all() and np.isfinite(full.to_numpy()).all()
        assert full.index.equals(adata.obs_names)
        assert len([key for key in full if key.startswith("ctrl_norm_score_")]) == N_CTRL
        groups = association_only(adata, full)
        assert groups.assoc_mcp.notna().all()
        groups["assoc_fdr_bh"] = multipletests(groups.assoc_mcp.to_numpy(), method="fdr_bh")[1]
        groups["technology"] = technology
        groups["trait"] = trait
        groups["analysis_mode"] = mode
        groups["heterogeneity_computed"] = False
        groups["mean_norm_score"] = adata.obs.cell_type_author.to_frame().join(full[["norm_score"]]).groupby("cell_type_author", observed=True).norm_score.mean()
        metadata_columns = ["cell_type_author", "sample_id_author", "technology", "stage_source"]
        library = adata.obs[metadata_columns].join(full[["norm_score", "zscore"]])
        library = library.groupby(metadata_columns, observed=True).agg(
            n_cells=("norm_score", "size"), mean_norm_score=("norm_score", "mean"), mean_zscore=("zscore", "mean"),
        ).reset_index()
        consistency = library.groupby("cell_type_author", observed=True).agg(
            n_libraries_observed=("sample_id_author", "nunique"),
            equal_library_mean=("mean_norm_score", "mean"), sd_library_means=("mean_norm_score", "std"),
            min_library_mean=("mean_norm_score", "min"), max_library_mean=("mean_norm_score", "max"),
            positive_fraction_libraries=("mean_norm_score", lambda values: float((values > 0).mean())),
        )
        consistency["summary_unit"] = "observed library; biological independence not established"
        metadata = {
            "format": "fig4_e1_scdrs_v1", "technology": technology, "trait": trait, "analysis_mode": mode,
            "scdrs_version": scdrs.__version__, "source_commit": COMMIT, "gs_path": str(gs_path), "gs_sha256": sha256(gs_path),
            "input_path": str(input_path), "input_sha256": INPUT_SHA[technology], "script_sha256": sha256(__file__),
            "seed": SEED, "n_ctrl": N_CTRL, "ranked_entrez_genes": len(ranked),
            "mapped_human_genes": len(trait_table), "used_trait_genes": len(selected),
            "measured_genes": len(measured), "background_genes": adata.n_vars,
            "normalization": "CP10k using total_common_nonmt (36888 measured genes), then log1p; float32",
            "covariates": "None" if cov is None else "within-library centered log1p(detected_common_nonmt); sensitivity only",
            "group_test": f"Association-only formulas from scdrs/method.py lines 763-817 at {COMMIT}; no graph",
            "group_fdr_family": f"All {len(groups)} author cell types within this technology and mode",
            "heterogeneity_computed": False, "library_summary": "Descriptive library means; no animal-replicate inference",
            **{key: value for key, value in provenance.items() if key != "magma_qc_json"},
        }
        with h5py.File(paths["h5"], "x") as output:
            output.attrs.update(metadata)
            output.create_dataset("magma_qc_json", data=provenance["magma_qc_json"], dtype=h5py.string_dtype("utf-8"))
            write_frame(output.create_group("obs"), adata.obs)
            write_frame(output.create_group("measurement_mask"), measured)
            write_frame(output.create_group("trait_gene_coverage"), coverage)
            write_frame(output.create_group("rank_manifest"), ranked)
            write_frame(output.create_group("gene_stats"), adata.uns["SCDRS_PARAM"]["GENE_STATS"])
            score = output.create_group("score")
            score.create_dataset("columns", data=np.asarray(full.columns, dtype=object), dtype=h5py.string_dtype("utf-8"))
            score.create_dataset("values", data=full.to_numpy(dtype=np.float32), compression="gzip", compression_opts=1, shuffle=True)
            write_frame(output.create_group("group_association"), groups)
            write_frame(output.create_group("sample_type_means"), library)
            write_frame(output.create_group("library_consistency"), consistency)
        groups.to_csv(paths["group"], sep="\t", index=True)
        library.to_csv(paths["sample"], sep="\t", index=False)
        consistency.to_csv(paths["consistency"], sep="\t", index=True)
        with h5py.File(paths["h5"], "r") as output:
            assert output["score/values"].shape == full.shape
            assert np.array_equal(output["score/values"][:], full.to_numpy())
            assert output["measurement_mask/used_for_technology_background"][:].sum() == adata.n_vars
            assert int(output["sample_type_means/n_cells"][:].sum()) == adata.n_obs
        print("COMPLETE", technology, mode, json.dumps({"groups_bh_0.05": int((groups.assoc_fdr_bh < .05).sum()),
              "groups_total": len(groups), "sample_type_rows": len(library), "seconds": round(time.monotonic()-started, 2),
              "output_h5": str(paths["h5"]), "sha256": sha256(paths["h5"])}), flush=True)


    #6. =========== Explicit execution gate; no substitute trait input
    def main():
        parser = argparse.ArgumentParser(description=__doc__)
        parser.add_argument("--gs", type=Path, help="The supplied human MAGMA fixed top-1000 weighted.gs")
        parser.add_argument("--trait", default=None)
        parser.add_argument("--rank-manifest", type=Path, help="Original 1000 Entrez ranks; default: .gs same-stem .tsv")
        parser.add_argument("--magma-qc", type=Path, help="Default: gene-result-QC.json next to the supplied .gs")
        parser.add_argument("--technology", choices=["snRNA", "scRNA", "both"], default="both")
        parser.add_argument("--mode", choices=["primary", "tech", "both", "detected-sensitivity"], default="primary")
        parser.add_argument("--self-test", action="store_true", help="Official toy only; creates no result files")
        args = parser.parse_args()
        validate_sources()
        if args.self_test:
            assert args.gs is None, "Do not combine a toy test with real trait scoring"
            self_test()
            return
        if args.gs is None:
            parser.error("--gs is required; no substitute gene set will be selected")
        trait, table, path, ranked, provenance = load_trait(args.gs, args.trait, args.rank_manifest, args.magma_qc)
        technologies = ["snRNA", "scRNA"] if args.technology == "both" else [args.technology]
        modes = ["primary", "tech"] if args.mode == "both" else ["tech" if args.mode == "detected-sensitivity" else args.mode]
        with threadpool_limits(limits=1):
            for technology in technologies:
                for mode in modes:
                    score_reference(technology, mode, trait, table, path, ranked, provenance)

    main()

#7. =========== Cochlear sensitivity summaries
def summarize_cochlear():
    """Summarize separate E1 assays and modes; ROOT is supplied by the caller."""
    #1. =========== Result locations and fixed reference sizes
    from pathlib import Path
    import hashlib
    import json
    import h5py
    import numpy as np
    import pandas as pd

    # Set this storage path independently before running.
    scratch = ROOT
    expected = {"snRNA": {"cells": 21442, "types": 21, "samples": 8},
                "scRNA": {"cells": 36893, "types": 25, "samples": 20}}
    output_paths = [scratch / f"E1-{tech}-comparison.tsv" for tech in expected]
    # Set this storage path independently before running. E1-cross-assay.tsv.
    output_paths += [ROOT, ROOT]
    assert all(not path.exists() for path in output_paths), "Refuse to overwrite an existing summary"

    def sha256(path):
        digest = hashlib.sha256()
        with Path(path).open("rb") as source:
            for block in iter(lambda: source.read(4 * 1024 * 1024), b""):
                digest.update(block)
        return digest.hexdigest()

    def finite_or_none(value):
        value = float(value)
        return value if np.isfinite(value) else None

    def read_result(tech, mode):
        stem = scratch / f"E1-{tech}-{mode}"
        group = pd.read_csv(str(stem) + "-groups.tsv", sep="\t", index_col=0)
        library = pd.read_csv(str(stem) + "-sample-type.tsv", sep="\t")
        consistency = pd.read_csv(str(stem) + "-library-consistency.tsv", sep="\t", index_col=0)
        assert len(group) == expected[tech]["types"] and group.index.is_unique
        assert int(group.n_cell.sum()) == expected[tech]["cells"]
        assert group.technology.eq(tech).all() and group.analysis_mode.eq(mode).all()
        assert group.n_ctrl.eq(1000).all()
        assert group.hetero_mcp.isna().all() and group.hetero_mcz.isna().all()
        assert not group.heterogeneity_computed.any()
        assert group.assoc_mcp.between(0, 1).all() and group.assoc_fdr_bh.between(0, 1).all()
        assert library.sample_id_author.nunique() == expected[tech]["samples"]
        assert int(library.n_cells.sum()) == expected[tech]["cells"]
        assert library.groupby(["cell_type_author", "sample_id_author"], observed=True).size().eq(1).all()
        np.testing.assert_array_equal(library.groupby("cell_type_author", observed=True).n_cells.sum().reindex(group.index), group.n_cell)
        with h5py.File(Path(str(stem) + ".h5"), "r") as h5:
            assert h5.attrs["technology"] == tech and h5.attrs["analysis_mode"] == mode
            assert "36888" in h5.attrs["normalization"] and h5.attrs["n_ctrl"] == 1000
            assert h5.attrs["measured_genes"] == 15722
            assert len(h5["rank_manifest/GENE"]) == h5.attrs["ranked_entrez_genes"] == 1000
            meta = {key: h5.attrs[key].item() if isinstance(h5.attrs[key], np.generic) else h5.attrs[key]
                    for key in ["trait", "gs_sha256", "input_sha256", "rank_manifest_sha256", "ranked_entrez_genes",
                                "mapped_human_genes", "used_trait_genes", "measured_genes", "background_genes", "covariates"]}
            columns = h5["score/columns"].asstr()[:].tolist()
            assert sum(key.startswith("ctrl_norm_score_") for key in columns) == 1000
            assert h5["score/values"].shape == (expected[tech]["cells"], len(columns))
            scores = h5["score/values"][:, columns.index("norm_score")]
            cell_ids = h5["obs/_index"].asstr()[:]
            human_symbols = h5["measurement_mask/human_symbol"].asstr()[:]
            background_mask = h5["measurement_mask/used_for_technology_background"][:]
            assert len(human_symbols) == 15722 and background_mask.sum() == meta["background_genes"]
            assert np.isfinite(scores).all()
        return group, library, consistency, meta, scores, cell_ids, human_symbols, background_mask

    #2. =========== Keep every author cell type in each assay
    report = {"assay_policy": "Separate assays; no pooled score, combined P value or meta-analysis",
              "normalization_denominator": "Complete 36888-gene common non-MT axis",
              "library_policy": "Descriptive observed-library means; no animal-replicate inference", "assays": {}}
    primary_tables = {}
    trait_hashes = set()
    for tech in expected:
        primary, library_p, consistency_p, meta_p, scores_p, ids_p, genes_p, mask_p = read_result(tech, "primary")
        technical, library_t, consistency_t, meta_t, scores_t, ids_t, genes_t, mask_t = read_result(tech, "tech")
        assert primary.index.equals(technical.index)
        np.testing.assert_array_equal(primary.n_cell, technical.n_cell)
        np.testing.assert_array_equal(ids_p, ids_t)
        np.testing.assert_array_equal(genes_p, genes_t)
        np.testing.assert_array_equal(mask_p, mask_t)
        for key in ["trait", "gs_sha256", "input_sha256", "rank_manifest_sha256", "ranked_entrez_genes",
                    "mapped_human_genes", "used_trait_genes", "measured_genes", "background_genes"]:
            assert meta_p[key] == meta_t[key], (tech, key)
        assert meta_p["covariates"] == "None" and "within-library centered" in meta_t["covariates"]
        trait_hashes.add(meta_p["gs_sha256"])
        a = primary.copy()
        b = technical.copy()
        for frame in [a, b]:
            frame["fraction_cell_fdr_0.1"] = frame["n_fdr_0.1"] / frame.n_cell
        keep_columns = ["assoc_mcp", "assoc_mcz", "assoc_fdr_bh", "n_fdr_0.05", "n_fdr_0.1", "n_fdr_0.2",
                        "fraction_cell_fdr_0.1", "mean_norm_score"]
        combined = primary[["n_cell"]].join(a[keep_columns].add_prefix("primary_")).join(b[keep_columns].add_prefix("tech_"))
        library_columns = ["n_libraries_observed", "equal_library_mean", "sd_library_means", "min_library_mean",
                           "max_library_mean", "positive_fraction_libraries"]
        combined = combined.join(consistency_p[library_columns].add_prefix("primary_"))
        combined = combined.join(consistency_t[library_columns].add_prefix("tech_"))
        sig_p = combined.primary_assoc_fdr_bh.lt(0.05)
        sig_t = combined.tech_assoc_fdr_bh.lt(0.05)
        combined["sensitivity_status"] = np.select(
            [sig_p & sig_t, sig_p & ~sig_t, ~sig_p & sig_t], ["both_modes", "primary_only", "tech_only"], default="neither")
        combined["technology"] = tech
        combined["trait"] = meta_p["trait"]
        combined.to_csv(scratch / f"E1-{tech}-comparison.tsv", sep="\t", index=True)
        cell_correlation = finite_or_none(np.corrcoef(scores_p.astype(np.float64), scores_t.astype(np.float64))[0, 1])
        group_rank_correlation = finite_or_none(primary.assoc_mcz.corr(technical.assoc_mcz, method="spearman"))
        report["assays"][tech] = {
            **expected[tech], **meta_p,
            "primary_significant_types": primary.index[primary.assoc_fdr_bh.lt(0.05)].tolist(),
            "tech_significant_types": technical.index[technical.assoc_fdr_bh.lt(0.05)].tolist(),
            "both_modes_significant_types": combined.index[sig_p & sig_t].tolist(),
            "primary_only_types": combined.index[sig_p & ~sig_t].tolist(),
            "tech_only_types": combined.index[~sig_p & sig_t].tolist(),
            "primary_vs_tech_cell_score_pearson_descriptive": cell_correlation,
            "primary_vs_tech_group_mcz_spearman_descriptive": group_rank_correlation,
            "correlation_inference": "Descriptive only; no correlation P values or animal-replicate claims",
            "comparison_sha256": sha256(scratch / f"E1-{tech}-comparison.tsv"),
        }
        primary_tables[tech] = primary
    assert len(trait_hashes) == 1

    #3. =========== Side-by-side exact-label validation, without combining assays
    cross_columns = ["n_cell", "assoc_mcp", "assoc_mcz", "assoc_fdr_bh", "n_fdr_0.1"]
    cross = primary_tables["snRNA"][cross_columns].add_prefix("snRNA_").join(
        primary_tables["scRNA"][cross_columns].add_prefix("scRNA_"), how="outer")
    sn_present = cross.snRNA_n_cell.notna()
    sc_present = cross.scRNA_n_cell.notna()
    cross["availability"] = np.select([sn_present & sc_present, sn_present], ["both_assays", "snRNA_only"], default="scRNA_only")
    cross["primary_validation"] = np.select(
        [~(sn_present & sc_present), cross.snRNA_assoc_fdr_bh.lt(.05) & cross.scRNA_assoc_fdr_bh.lt(.05)],
        ["not_comparable_exact_author_label", "both_assays_BH_lt_0.05"], default="not_both_assays_significant")
    assert int(sn_present.sum()) == 21 and int(sc_present.sum()) == 25
    # Set this storage path independently before running. E1-cross-assay.tsv.
    cross.to_csv(ROOT, sep="\t", index=True)
    report["cross_assay"] = {
        "exact_shared_author_labels": cross.index[sn_present & sc_present].tolist(),
        "snRNA_only_labels": cross.index[sn_present & ~sc_present].tolist(),
        "scRNA_only_labels": cross.index[~sn_present & sc_present].tolist(),
        "missing_policy": "Unavailable assay/type values remain missing, never zero",
        # Set this storage path independently before running. E1-cross-assay.tsv.
        "cross_assay_sha256": sha256(ROOT),
    }
    # Set this storage path independently before running. E1-summary.json.
    (ROOT).write_text(json.dumps(report, ensure_ascii=False, indent=2, allow_nan=False), encoding="utf-8")
    print("E1_SUMMARY_COMPLETE", json.dumps({tech: {
        "all_types": expected[tech]["types"], "primary_significant": len(report["assays"][tech]["primary_significant_types"]),
        "tech_significant": len(report["assays"][tech]["tech_significant_types"])} for tech in expected}), flush=True)

#7. =========== Locked public GWAS and official MAGMA workflow
def make_fig4_magma_workflow(ROOT):
    from pathlib import Path
    from collections import Counter, deque
    import gzip
    import datetime
    import hashlib
    import json
    import os
    import re
    import shutil
    import stat
    import subprocess
    import zipfile
    import zlib
    import threading
    from concurrent.futures import ThreadPoolExecutor, as_completed
    import numpy as np
    import pandas as pd
    from scipy.stats import norm

    # Set this storage path independently before running.
    ROOT = ROOT
    # Set this storage path independently before running.
    WORK = ROOT
    # Set this storage path independently before running.
    REF = ROOT
    # Set this storage path independently before running.
    LD = ROOT
    # Set this storage path independently before running.
    RNA = ROOT
    # Set this storage path independently before running. hl_ma_nov18_v2_summarystats_09122021.txt.gz.
    GWAS = ROOT
    # Set this storage path independently before running.
    PREFIX = ROOT
    SEED = 42
    CHUNK_SIZE = 500000
    MAF_MIN = 0.01
    N_TOP = 1000
    TRAIT = "ARHL_Trpchevska2022"
    BASES = {"A": 0, "C": 1, "G": 2, "T": 3}
    COMPLEMENT = np.array([3, 2, 1, 0], dtype=np.uint8)
    REF_DTYPE = np.dtype([("rsid", "<i8"), ("chr", "u1"), ("bp", "<i4"), ("a1", "u1"), ("a2", "u1")])
    RESOURCES = [
        ("hl_ma_nov18_v2_summarystats_09122021.txt.gz", 297729431,
         "77e6b9e07435f0dfdc999d15511260dd923986fd99229006928bb693e577a6fe",
         "9cf9069eeef9ce6c6de7f00fd07ec212",
         "https://zenodo.org/api/records/5769707/files/hl_ma_nov18_v2_summarystats_09122021.txt.gz/content"),
        ("magma_v1.10_win.zip", 2151875,
         "953a2a88eb3a7047b2963712a5fce5d74ab82ca63c9ac32bc3184b3055020507", None,
         "https://vu.data.surf.nl/index.php/s/TOH4SuvczAKE29d/download"),
        ("g1000_eur.zip", 511626945,
         "83a48fd9dcaa0b9a874b18c63143a4ede93f05505b215b0bd8790130a0d7a954", None,
         "https://vu.data.surf.nl/index.php/s/VZNByNwpD8qqINe/download"),
        ("NCBI37.3.zip", 355626,
         "8c2bc9cba581819b2df5a63e741a360d92efef7d819f553a06e2d946ae60a90b", None,
         "https://vu.data.surf.nl/index.php/s/Pj2orwuF2JYyKxq/download"),
    ]
    LOCKED_AUXILIARY = {
        "reference/hgnc_complete_set.txt": "12363916be36f8f6f91879b0faade4eeaac973722de7136ec61329b77f2c2f91",
        "reference/MHC-audit.json": "22a857279546cae35f17649052197766eed71263ef03f22dbabc8653a0376f19",
        "Trpchevska2022.leads.tsv": "fc55f7e97ae724b2cf187e0a614d5eafff513226e360a374c7534846f85a3620",
        "lead-source.json": "ec5288d9cad9c45d0b9b0e33f20eaa4c532ce4b70331f07fb17b1a38b3ce0b4f",
    }

    def digest(path, algorithm="sha256"):
        result = hashlib.new(algorithm)
        with Path(path).open("rb") as stream:
            for block in iter(lambda: stream.read(4 * 1024 * 1024), b""):
                result.update(block)
        return result.hexdigest()

    def file_manifest(paths):
        return {p.name: {"path": str(p.resolve()), "bytes": p.stat().st_size, "sha256": digest(p)} for p in paths}

    def dump_json(path, value):
        Path(path).write_text(json.dumps(value, indent=2), encoding="utf-8")

    def native_file(suffix, prefix=PREFIX):
        paths = [p for p in [Path(str(prefix) + suffix), Path(str(prefix) + suffix + ".txt")] if p.exists()]
        if len(paths) != 1:
            raise RuntimeError(f"Expected exactly one native file for {prefix.name}{suffix}")
        return paths[0]

    def run_native(args, cwd=None):
        tail, warnings = deque(maxlen=32), []
        process = subprocess.Popen([str(x) for x in args], cwd=WORK if cwd is None else cwd, stdout=subprocess.PIPE,
                                   stderr=subprocess.STDOUT, text=True, encoding="utf-8", errors="replace")
        for line in process.stdout:
            tail.append(line.rstrip())
            if "WARNING" in line or "ERROR" in line:
                warnings.append(line.rstrip())
                if len(warnings) <= 8:
                    print(line.rstrip(), flush=True)
            if "SNP location must be a positive integer" in line:
                process.terminate()
                process.wait()
                raise RuntimeError("Invalid SNP coordinate representation")
        code = process.wait()
        print("\n".join(tail), flush=True)
        if code:
            raise RuntimeError(f"MAGMA failed with return code {code}")
        return warnings


    #2. =========== Validate retained downloads and extract only within this task
    def validate_resources(extract_missing=True):
        if not WORK.is_relative_to(ROOT) or not WORK.is_dir():
            raise RuntimeError("Expected existing task directory under ROOT")
        # Set this storage path independently before running. resource-manifest.json.
        manifest = json.loads((ROOT).read_text())
        by_name = {entry["name"]: entry for entry in manifest}
        for name, size, sha, author_md5, url in RESOURCES:
            path = WORK / name
            if not path.is_file() or path.stat().st_size != size or digest(path) != sha:
                raise RuntimeError(f"Missing or changed public resource: {path}")
            if author_md5 is not None and digest(path, "md5") != author_md5:
                raise RuntimeError("GWAS does not match the author's published MD5")
            record = by_name.get(name, {})
            if record.get("sha256") != sha or record.get("url") != url or record.get("bytes") != size:
                raise RuntimeError("Resource manifest differs from the locked source contract")
        for relative, expected_sha in LOCKED_AUXILIARY.items():
            path = WORK / relative
            if not path.is_file() or digest(path) != expected_sha:
                raise RuntimeError(f"Missing or changed frozen annotation input: {relative}")
        # Set this storage path independently before running.
        for name, destination in [("magma_v1.10_win.zip", ROOT),
                                  ("NCBI37.3.zip", REF), ("g1000_eur.zip", LD)]:
            destination.mkdir(parents=True, exist_ok=True)
            with zipfile.ZipFile(WORK / name) as archive:
                missing_size = sum(x.file_size for x in archive.infolist() if not (destination / x.filename).exists())
                if missing_size > shutil.disk_usage(WORK).free:
                    raise RuntimeError("Insufficient space for verified resource extraction")
                for info in archive.infolist():
                    target = (destination / info.filename).resolve()
                    if not target.is_relative_to(destination.resolve()) or ":" in info.filename or stat.S_ISLNK(info.external_attr >> 16):
                        raise RuntimeError(f"Unsafe ZIP member: {info.filename}")
                    if info.is_dir():
                        target.mkdir(parents=True, exist_ok=True)
                        continue
                    if target.exists():
                        if target.stat().st_size != info.file_size:
                            raise RuntimeError(f"Existing reference size mismatch: {target}")
                        crc = 0
                        with target.open("rb") as stream:
                            for block in iter(lambda: stream.read(4 * 1024 * 1024), b""):
                                crc = zlib.crc32(block, crc)
                        if (crc & 0xffffffff) != info.CRC:
                            raise RuntimeError(f"Existing reference CRC mismatch: {target}")
                    elif extract_missing:
                        target.parent.mkdir(parents=True, exist_ok=True)
                        with archive.open(info) as source, target.open("wb") as output:
                            shutil.copyfileobj(source, output, 4 * 1024 * 1024)
                    else:
                        raise RuntimeError(f"Missing extracted reference: {target}")
        # Set this storage path independently before running. magma.exe.
        if digest(ROOT) != "e05e3d963215a973064852ec33640118d65745884cc800a5119ca7cffe6cb859":
            raise RuntimeError("Unexpected executable")
        # Set this storage path independently before running. magma.exe.
        version = subprocess.run([str(ROOT), "--version"], cwd=WORK,
                                 capture_output=True, text=True, check=True)
        if "v1.10" not in version.stdout:
            raise RuntimeError("Unexpected MAGMA version")
        return {"version": version.stdout.strip(), "resources": manifest}


    #3. =========== Recreate the exact MHC-excluded axis and unique HGNC mapping
    def prepare_annotations():
        for relative, expected_sha in LOCKED_AUXILIARY.items():
            if digest(WORK / relative) != expected_sha:
                raise RuntimeError(f"Frozen annotation source changed: {relative}")
        # Set this storage path independently before running.
        source = pd.read_csv(ROOT, sep=r"\s+", header=None, dtype=str)
        # Set this storage path independently before running. MHC-audit.json.
        mhc = json.loads((ROOT).read_text())
        if len(source) != 19427 or len(mhc["excluded_entrez_ids"]) != 252:
            raise RuntimeError("Unexpected original MHC exclusion axis")
        derived = source.loc[~source[0].isin(mhc["excluded_entrez_ids"])].copy()
        if len(derived) != 19175:
            raise RuntimeError("Expected 19,175 retained gene locations")
        # Set this storage path independently before running.
        gene_loc = ROOT
        derived.to_csv(gene_loc, sep="\t", header=False, index=False, lineterminator="\n")
        # Set this storage path independently before running. hgnc_complete_set.txt.
        hgnc_file = ROOT
        hgnc = pd.read_csv(hgnc_file, sep="\t", dtype=str, keep_default_na=False)
        approved = hgnc.loc[hgnc["status"].eq("Approved")]
        with_entrez = approved.loc[approved["entrez_id"].ne("")]
        duplicate_entrez = with_entrez.loc[with_entrez["entrez_id"].duplicated(keep=False)]
        unique = with_entrez.loc[~with_entrez["entrez_id"].duplicated(keep=False)]
        mapping = derived.copy()
        mapping.columns = ["entrez_id", "CHR", "START", "STOP", "STRAND", "symbol_ncbi37"]
        mapping = mapping.merge(unique[["entrez_id", "symbol", "hgnc_id", "ensembl_gene_id"]],
                                on="entrez_id", how="left", validate="one_to_one")
        mapping["mapping_status"] = np.where(mapping["symbol"].notna(), "HGNC_unique_entrez", "unmapped_entrez")
        if mapping.loc[mapping["symbol"].notna(), "symbol"].duplicated().any():
            raise RuntimeError("Approved-symbol mapping is not unique")
        # Set this storage path independently before running. human-gene-map.tsv.
        mapping.to_csv(ROOT, sep="\t", index=False, lineterminator="\n")
        audit = {"source_url": "https://storage.googleapis.com/public-download-files/hgnc/tsv/tsv/hgnc_complete_set.txt",
                 "source_sha256": digest(hgnc_file), "approved_rows": len(approved),
                 "approved_entrez_duplicate_rows": duplicate_entrez[["entrez_id", "symbol"]].values.tolist(),
                 "map_status_counts": mapping["mapping_status"].value_counts().to_dict(),
                 "unmapped": mapping.loc[mapping["symbol"].isna(), ["entrez_id", "symbol_ncbi37"]].values.tolist()}
        # Set this storage path independently before running. symbol-map-audit.json.
        dump_json(ROOT, audit)
        return audit


    #4. =========== Match the complete GWAS and audit all exclusions in one pass
    def prepare_inputs():
        """Explicit recomputation step. Do not call while annotation/MAGMA is running."""
        # Set this storage path independently before running.
        ref_cache = ROOT
        # Set this storage path independently before running. reference-index-audit.json.
        ref_audit_file = ROOT
        # Set this storage path independently before running.
        bim = ROOT
        if not ref_cache.exists():
            arrays, ref_counts = [], Counter()
            for chunk in pd.read_csv(bim, sep=r"\s+", header=None, usecols=[0, 1, 3, 4, 5],
                                     dtype={0: "uint8", 1: str, 3: "int32", 4: str, 5: str}, chunksize=CHUNK_SIZE):
                ref_counts["bim_rows"] += len(chunk)
                valid_id = chunk[1].str.fullmatch(r"rs[0-9]+")
                ref_counts["non_rs_reference_ids"] += int((~valid_id).sum())
                chunk = chunk.loc[valid_id]
                array = np.empty(len(chunk), dtype=REF_DTYPE)
                array["rsid"] = pd.to_numeric(chunk[1].str.slice(2)).to_numpy(dtype=np.int64)
                array["chr"], array["bp"] = chunk[0].to_numpy(), chunk[3].to_numpy()
                array["a1"] = chunk[4].str.upper().map(BASES).fillna(255).to_numpy(dtype=np.uint8)
                array["a2"] = chunk[5].str.upper().map(BASES).fillna(255).to_numpy(dtype=np.uint8)
                arrays.append(array)
            reference = np.concatenate(arrays)
            del arrays
            reference.sort(order="rsid")
            duplicate = np.zeros(len(reference), dtype=bool)
            equal = reference["rsid"][1:] == reference["rsid"][:-1]
            duplicate[:-1] |= equal
            duplicate[1:] |= equal
            ref_counts["duplicate_reference_id_rows_excluded"] = int(duplicate.sum())
            reference = reference[~duplicate]
            ref_counts["unique_reference_rsids"] = len(reference)
            np.save(ref_cache, reference)
            dump_json(ref_audit_file, dict(ref_counts))
        reference = np.load(ref_cache, mmap_mode="r")
        ref_counts = Counter(json.loads(ref_audit_file.read_text()))
        if len(reference) != 22665064 or reference.dtype != REF_DTYPE or digest(ref_cache) != "0a005379b67e2592efdff2907a9463f8db81646c478b40a460f0c5578779ef70":
            raise RuntimeError("Unexpected cached reference index")
        ref_ids = np.ascontiguousarray(reference["rsid"])
        if not np.all(ref_ids[1:] > ref_ids[:-1]):
            raise RuntimeError("Reference rs IDs must be strictly increasing and unique")
        ref_counts["reference_individuals"] = sum(1 for _ in bim.with_suffix(".fam").open())
        if ref_counts["reference_individuals"] != 503:
            raise RuntimeError("Unexpected EUR reference sample count")
        # Set this storage path independently before running. Trpchevska2022.leads.tsv.
        leads = pd.read_csv(ROOT, sep="\t")
        lead_ids = set(leads["SNP"]) | {"rs6545432"}
        counts, exclusion_stats, preview = Counter(), {}, {}
        matched_ids, sample_sizes, lead_frames, region_frames = [], [], [], []
        p_min, p_max, exclusion_header = 1.0, 0.0, True
        audit_cols = ["SNP", "CHR", "BP", "Allele1", "Allele2", "Freq1", "Effect", "StdErr", "P.value", "N"]
        # Set this storage path independently before running.
        with (ROOT).open("w", newline="") as p_output, \
             (ROOT).open("w", newline="") as loc_output, \
             gzip.open(ROOT, "wt", newline="") as excluded_output:
            p_output.write("SNP\tP\tN\n")
            for d in pd.read_csv(GWAS, sep=r"\s+", chunksize=CHUNK_SIZE):
                counts["gwas_rows"] += len(d)
                numeric = d[["CHR", "BP", "Freq1", "Effect", "StdErr", "P.value", "N"]].to_numpy(dtype=float)
                valid = np.isfinite(numeric).all(axis=1) & d["P.value"].gt(0) & d["P.value"].le(1) & d["N"].gt(0) & d["StdErr"].gt(0)
                if not np.equal(d.loc[valid, "N"], np.rint(d.loc[valid, "N"])).all():
                    raise RuntimeError("Per-SNP total N is noninteger")
                status = pd.Series("retained", index=d.index)
                status.loc[~valid] = "invalid_statistics"
                maf = d["Freq1"].gt(MAF_MIN) & d["Freq1"].lt(1 - MAF_MIN)
                status.loc[valid & ~maf] = "failed_maf_1pct"
                valid_id = d["SNP"].str.fullmatch(r"rs[0-9]+")
                valid_allele = d["Allele1"].str.upper().isin(BASES) & d["Allele2"].str.upper().isin(BASES)
                status.loc[valid & maf & ~(valid_id & valid_allele)] = "invalid_id_or_alleles"
                c = d.loc[valid & maf & valid_id & valid_allele]
                ids = pd.to_numeric(c["SNP"].str.slice(2)).to_numpy(dtype=np.int64)
                idx = np.searchsorted(ref_ids, ids)
                matched = idx < len(ref_ids)
                matched[matched] &= ref_ids[idx[matched]] == ids[matched]
                status.loc[c.index[~matched]] = "missing_reference"
                c, ids = c.loc[matched], ids[matched]
                rr = reference[idx[matched]]
                coords = (c["CHR"].to_numpy() == rr["chr"]) & (c["BP"].to_numpy() == rr["bp"])
                a1 = c["Allele1"].str.upper().map(BASES).to_numpy(dtype=np.uint8)
                a2 = c["Allele2"].str.upper().map(BASES).to_numpy(dtype=np.uint8)
                direct = ((a1 == rr["a1"]) & (a2 == rr["a2"])) | ((a1 == rr["a2"]) & (a2 == rr["a1"]))
                complement = ((COMPLEMENT[a1] == rr["a1"]) & (COMPLEMENT[a2] == rr["a2"])) | ((COMPLEMENT[a1] == rr["a2"]) & (COMPLEMENT[a2] == rr["a1"]))
                allele_ok = direct | complement
                status.loc[c.index[~coords]] = "coordinate_mismatch"
                status.loc[c.index[~allele_ok]] = "allele_mismatch"
                status.loc[c.index[~coords & ~allele_ok]] = "coordinate_and_allele_mismatch"
                bad = d.loc[status.ne("retained"), audit_cols].copy()
                bad.insert(0, "reason", status.loc[bad.index])
                bad.to_csv(excluded_output, sep="\t", index=False, header=exclusion_header, float_format="%.17g", lineterminator="\n")
                exclusion_header = False
                for reason, group in bad.groupby("reason"):
                    entry = exclusion_stats.setdefault(reason, {"n": 0, "P_lt_5e8": 0, "min_P": 1.0})
                    entry["n"] += len(group)
                    entry["P_lt_5e8"] += int(group["P.value"].lt(5e-8).sum())
                    entry["min_P"] = float(min(entry["min_P"], group["P.value"].min()))
                keep = coords & allele_ok
                accepted = c.loc[keep].copy()
                counts["matched_rows"] += len(accepted)
                counts["direct_or_swapped_alleles"] += int((direct & keep).sum())
                counts["complement_only_alleles"] += int((~direct & complement & keep).sum())
                matched_ids.append(ids[keep])
                sample_sizes.append(accepted["N"].to_numpy(dtype=np.int32))
                p_min, p_max = min(p_min, float(accepted["P.value"].min())), max(p_max, float(accepted["P.value"].max()))
                pvals = accepted[["SNP", "P.value", "N"]].copy()
                pvals["N"] = pvals["N"].astype(np.int32)
                pvals.to_csv(p_output, sep="\t", index=False, header=False, float_format="%.17g", lineterminator="\n")
                locs = accepted[["SNP", "CHR", "BP"]].copy()
                locs[["CHR", "BP"]] = locs[["CHR", "BP"]].astype(np.int64)
                locs.to_csv(loc_output, sep="\t", index=False, header=False, lineterminator="\n")
                region = d.loc[d["CHR"].eq(2) & d["BP"].between(54317683, 55317683)].copy()
                if len(region):
                    region_frames.append(region)
                found = d.loc[d["SNP"].isin(lead_ids)].copy()
                if len(found):
                    found["status"] = status.loc[found.index]
                    lead_frames.append(found)
                print(f"GWAS rows checked={counts['gwas_rows']}; retained={counts['matched_rows']}", flush=True)
        ids = np.sort(np.concatenate(matched_ids))
        ns = np.concatenate(sample_sizes)
        if counts["gwas_rows"] != 8244937 or counts["matched_rows"] != 7518665 or np.any(ids[1:] == ids[:-1]):
            raise RuntimeError("Accepted SNP universe changed or contains duplicate IDs")
        expected_exclusions = {"failed_maf_1pct": 561398, "missing_reference": 164715, "coordinate_mismatch": 2, "allele_mismatch": 157}
        if {key: value["n"] for key, value in exclusion_stats.items()} != expected_exclusions:
            raise RuntimeError("Exclusion audit differs from the reviewed analysis")
        qc = {"reference": dict(ref_counts), "gwas": dict(counts), "exclusion_stats": exclusion_stats,
              "matched_duplicate_id_pairs": 0, "N_min": int(ns.min()), "N_max": int(ns.max()), "N_median": float(np.median(ns)),
              "P_min": p_min, "P_max": p_max, "bfile": str(bim.with_suffix("")),
              "N_definition": "published per-SNP total sample size; MAGMA v1.10 manual page 8",
              "allele_note": "P-only analysis; effect reversal is invariant, palindromic SNPs retained only after exact build/coordinate and unordered-allele match"}
        # Set this storage path independently before running. input-QC.json.
        dump_json(ROOT, qc)
        found = pd.concat(lead_frames, ignore_index=True)
        if found["SNP"].duplicated().any():
            raise RuntimeError("Duplicate published lead variant")
        lead_report = leads[["SNP"]].merge(found, on="SNP", how="left", validate="one_to_one")
        lead_report["status"] = lead_report["status"].fillna("not_in_public_GWAS")
        # Set this storage path independently before running. lead-retention.tsv.
        lead_report.to_csv(ROOT, sep="\t", index=False, float_format="%.17g", lineterminator="\n")
        region = pd.concat(region_frames, ignore_index=True).sort_values("BP", kind="mergesort")
        region[["CHR", "BP", "N"]] = region[["CHR", "BP", "N"]].astype(np.int64)
        # Set this storage path independently before running. SPTBN1-region.tsv.gz.
        region.to_csv(ROOT, sep="\t", index=False, float_format="%.17g",
                      compression={"method": "gzip", "mtime": 0}, lineterminator="\n")
        exclusion_qc = {"total_GWAS_rows": counts["gwas_rows"], "retained_rows": counts["matched_rows"], "exclusions": exclusion_stats,
                        "lead_status_counts": lead_report["status"].value_counts().to_dict(),
                        "nonretained_leads": lead_report.loc[lead_report["status"].ne("retained"), ["SNP", "Freq1", "P.value", "status"]].to_dict("records"),
                        "SPTBN1_region": {"build": "hg19", "chromosome": 2, "start": 54317683, "stop": 55317683, "rows": len(region), "all_original_columns": region.columns.tolist()},
                        "rs6545432": found.loc[found["SNP"].eq("rs6545432"), audit_cols + ["status"]].to_dict("records")}
        # Set this storage path independently before running. exclusion-QC.json.
        dump_json(ROOT, exclusion_qc)
        return qc, exclusion_qc


    #5. =========== Annotate the accepted integer SNP coordinates
    def annotate():
        # Set this storage path independently before running. magma.exe.
        warnings = run_native([ROOT, "--annotate", "window=35,10", "--snp-loc", ROOT,
                               # Set this storage path independently before running.
                               "--gene-loc", ROOT, "--out", ROOT])
        # Set this storage path independently before running.
        log = (ROOT).read_text()
        match = re.search(r"([0-9]+) SNP locations read", log)
        if warnings or match is None or int(match.group(1)) != 7518665:
            raise RuntimeError("Annotation did not read all 7,518,665 accepted SNPs without warnings")
        # Set this storage path independently before running.
        return native_file(".genes.annot", ROOT)


    #6. =========== Calculate all eligible gene statistics and preserve LD-aware raw
    def analyze_genes():
        # Set this storage path independently before running.
        if any(p.exists() for p in [ROOT, ROOT]):
            raise RuntimeError("Full gene results already exist; validate/reuse them rather than rerunning implicitly")
        # Set this storage path independently before running.
        annotation = native_file(".genes.annot", ROOT)
        # Set this storage path independently before running. magma.exe.
        warnings = run_native([ROOT, "--bfile", ROOT, "--pval", ROOT, "ncol=N",
                               "--gene-annot", annotation, "--out", PREFIX, "--seed", SEED])
        return {"warnings": warnings, "gene_results": str(native_file(".genes.out")), "gene_raw": str(native_file(".genes.raw"))}


    #6b. =========== Official eight-batch recovery with all native gene correlations
    def analyze_gene_batches():
        """Run in the hosting script/process; do not invoke if a batch driver is active."""
        # Set this storage path independently before running.
        OUT = ROOT
        OUT.mkdir(exist_ok=True)
        # Set this storage path independently before running.
        BATCH_PREFIX = ROOT
        # Set this storage path independently before running. batch-state.json.
        STATE = ROOT
        if STATE.exists():
            previous = json.loads(STATE.read_text())
            if previous.get("status") == "batches_complete":
                return previous
            raise RuntimeError("Existing batch state needs inspection; no implicit restart")
        lock = threading.Lock()
        def now():
            return datetime.datetime.now(datetime.timezone.utc).isoformat()
        def save_state():
            temporary = STATE.with_suffix(".json.tmp")
            dump_json(temporary, state)
            temporary.replace(STATE)
        # Set this storage path independently before running.
        inputs = [ROOT, native_file(".genes.annot", ROOT), ROOT]
        state = {"status": "running", "started_utc": now(), "driver_pid": os.getpid(),
                 "batch_count": 8, "workers": 8, "seed": SEED,
                 "strategy": "Official MAGMA v1.10 --batch k 8 / --merge; preserves native gene correlations",
                 "input_files": {str(p.relative_to(WORK)): {"bytes": p.stat().st_size, "sha256": digest(p)} for p in inputs}, "jobs": {}}
        # Set this storage path independently before running. magma.exe.
        base = [str(ROOT), "--bfile", str(ROOT), "--pval", str(ROOT),
                "ncol=N", "--gene-annot", str(inputs[1]), "--out", str(BATCH_PREFIX), "--seed", str(SEED)]
        state["common_command"] = base
        save_state()
        environment = os.environ.copy()
        environment.update({"OMP_NUM_THREADS": "1", "MKL_NUM_THREADS": "1", "OPENBLAS_NUM_THREADS": "1", "NUMEXPR_NUM_THREADS": "1",
                            "TMP": str(OUT), "TEMP": str(OUT), "PYTHONDONTWRITEBYTECODE": "1"})
        flags = subprocess.CREATE_NO_WINDOW if os.name == "nt" else 0
        def one_batch(number):
            command = base + ["--batch", str(number), "8"]
            console = OUT / f"batch-{number}.console.log"
            with console.open("w", encoding="utf-8", newline="") as output:
                process = subprocess.Popen(command, cwd=WORK, stdout=output, stderr=subprocess.STDOUT,
                                           env=environment, creationflags=flags)
                with lock:
                    state["jobs"][str(number)] = {"status": "running", "pid": process.pid, "started_utc": now(),
                                                   "command": command, "console": str(console)}
                    save_state()
                code = process.wait()
            native_file(".genes.out", Path(str(BATCH_PREFIX) + f".batch{number}_8"))
            native_file(".genes.raw", Path(str(BATCH_PREFIX) + f".batch{number}_8"))
            with lock:
                state["jobs"][str(number)].update({"status": "complete" if code == 0 else "failed", "returncode": code, "ended_utc": now()})
                save_state()
            if code:
                raise RuntimeError(f"Batch {number} failed with code {code}")
            return number
        try:
            with ThreadPoolExecutor(max_workers=8) as pool:
                futures = [pool.submit(one_batch, number) for number in range(1, 9)]
                for future in as_completed(futures):
                    print(f"Completed batch {future.result()}/8", flush=True)
            state["status"], state["ended_utc"] = "batches_complete", now()
            save_state()
            return state
        except BaseException as error:
            state["status"], state["error"], state["ended_utc"] = "failed", repr(error), now()
            save_state()
            raise

    def merge_gene_batches():
        # Set this storage path independently before running.
        OUT = ROOT
        # Set this storage path independently before running.
        BATCH_PREFIX = ROOT
        # Set this storage path independently before running. batch-state.json.
        state = json.loads((ROOT).read_text())
        if state.get("status") != "batches_complete" or len(state["jobs"]) != 8:
            raise RuntimeError("All eight original batch jobs must finish successfully before merging")
        for relative, details in state["input_files"].items():
            path = WORK / relative
            if path.stat().st_size != details["bytes"] or digest(path) != details["sha256"]:
                raise RuntimeError("Reviewed analysis input changed during batch execution")
        batch_results, paths = [], []
        for number in range(1, 9):
            job = state["jobs"][str(number)]
            expected = state["common_command"] + ["--batch", str(number), "8"]
            if job["status"] != "complete" or job["returncode"] != 0 or job["command"] != expected:
                raise RuntimeError("Batch command/return code differs from the common analysis contract")
            prefix = Path(str(BATCH_PREFIX) + f".batch{number}_8")
            gene_file, raw_file = native_file(".genes.out", prefix), native_file(".genes.raw", prefix)
            if min(gene_file.stat().st_size, raw_file.stat().st_size) == 0:
                raise RuntimeError("Empty native batch result")
            batch_results.append(pd.read_csv(gene_file, sep=r"\s+", comment="#", dtype={"GENE": str}))
            paths.extend([gene_file, raw_file])
        concatenated = pd.concat(batch_results, ignore_index=True)
        if len(concatenated) != 18078 or concatenated["GENE"].duplicated().any():
            raise RuntimeError("Batch union does not cover all 18,078 annotated valid genes exactly once")
        annotation_ids = []
        # Set this storage path independently before running.
        for line in native_file(".genes.annot", ROOT).read_text().splitlines():
            if line.strip() and not line.startswith("#"):
                annotation_ids.append(line.split()[0])
        if set(concatenated["GENE"]) != set(annotation_ids):
            raise RuntimeError("Batch gene universe differs from the accepted annotation")
        for suffix in [".log", ".log.suppl"]:
            original = Path(str(PREFIX) + suffix)
            backup = OUT / ("interrupted-monolithic" + suffix)
            if original.exists() and not backup.exists():
                shutil.copyfile(original, backup)
        # Set this storage path independently before running.
        if any(p.exists() for p in [ROOT, ROOT]):
            raise RuntimeError("Merged result already exists; inspect and reuse rather than overwrite")
        # The Windows writer appends .txt to .genes.out, whereas --merge looks
        # only for the documented suffix. Preserve originals and create verified,
        # byte-identical conventional-name inputs in a separate merge directory.
        # Set this storage path independently before running.
        merge_directory = ROOT
        merge_directory.mkdir(exist_ok=True)
        # Set this storage path independently before running.
        merge_prefix = ROOT
        compatibility = []
        for original in paths:
            target = merge_directory / (original.name[:-4] if original.name.endswith(".genes.out.txt") else original.name)
            if not target.exists():
                shutil.copyfile(original, target)
            if digest(original) != digest(target):
                raise RuntimeError("Conventional-suffix merge input differs from the native original")
            compatibility.append({"native": str(original), "merge_input": str(target), "sha256": digest(target)})
        prior_merge_log = Path(str(PREFIX) + ".log")
        if prior_merge_log.exists() and "ERROR - merging gene results files" in prior_merge_log.read_text():
            counter = 1
            while (OUT / f"failed-merge-{counter}.log").exists():
                counter += 1
            shutil.copyfile(prior_merge_log, OUT / f"failed-merge-{counter}.log")
        # Windows batch-file discovery also requires a relative prefix when the
        # absolute task path contains spaces; all files remain in this directory.
        # Set this storage path independently before running. magma.exe.
        merge_command = [ROOT, "--merge", merge_prefix.name, "8", "--out", PREFIX, "--seed", SEED]
        warnings = run_native(merge_command, cwd=merge_directory)
        merged_path, raw_path = native_file(".genes.out"), native_file(".genes.raw")
        merged = pd.read_csv(merged_path, sep=r"\s+", comment="#", dtype={"GENE": str})
        left = concatenated.sort_values("GENE").reset_index(drop=True)
        right = merged.sort_values("GENE").reset_index(drop=True)
        pd.testing.assert_frame_equal(left, right, check_exact=True)
        audit = {"batch_count": 8, "gene_count": len(merged), "batch_union_equals_accepted_annotation": True,
                 "merged_gene_table_equals_batch_union_exactly": True, "warnings": warnings,
                 "Windows_output_suffix_compatibility": compatibility,
                 "merge_command": [str(x) for x in merge_command], "merge_cwd": str(merge_directory),
                 "correlations": "Kept native .genes.raw in every batch and official merge; no genes-only or big-data mode",
                 "files": file_manifest(paths + [merged_path, raw_path])}
        # Set this storage path independently before running. batch-merge-QC.json.
        dump_json(ROOT, audit)
        return audit


    #7. =========== Freeze top 1000 Entrez before any symbol/ortholog/expression mapping
    # The validated full-result export and E3 functions are appended below.

    def export_gene_statistics():
        GENE_RESULTS = native_file(".genes.out")
        # Validate all gene-level results
        genes = pd.read_csv(GENE_RESULTS, sep=r"\s+", comment="#", dtype={"GENE": str})
        required = ["GENE", "CHR", "START", "STOP", "NSNPS", "NPARAM", "N", "ZSTAT", "P"]
        if list(genes.columns) != required:
            raise RuntimeError(f"Unexpected MAGMA columns: {genes.columns.tolist()}")
        if genes["GENE"].duplicated().any() or len(genes) < 15000:
            raise RuntimeError("Unexpected gene coverage or duplicate Entrez IDs")
        if not np.isfinite(genes[["NSNPS", "NPARAM", "N", "ZSTAT", "P"]]).all().all():
            raise RuntimeError("Nonfinite MAGMA results")
        if not (genes["P"].gt(0) & genes["P"].le(1) & genes["NSNPS"].gt(0) & genes["N"].gt(0)).all():
            raise RuntimeError("Invalid MAGMA P, SNP count, or sample size")
        # Native ZSTAT and P both have five significant digits. Check overlapping
        # rounding intervals, including the amplified inverse-Z error near P=1.
        z = genes["ZSTAT"].to_numpy()
        p = genes["P"].to_numpy()
        z_half = np.zeros(len(z))
        nonzero = z != 0
        z_half[nonzero] = 0.5001 * np.power(10.0, np.floor(np.log10(np.abs(z[nonzero]))) - 4)
        p_half = 0.5001 * np.power(10.0, np.floor(np.log10(p)) - 4)
        compatible = (norm.sf(z + z_half) <= np.minimum(1.0, p + p_half)) & (norm.sf(z - z_half) >= np.maximum(0.0, p - p_half))
        if not compatible.all():
            raise RuntimeError("Native five-significant-digit ZSTAT and P rounding intervals are inconsistent")
        inverse_error = np.abs(norm.isf(p) - z)
        max_z_error = float(np.max(inverse_error)) if np.isfinite(inverse_error).all() else None
        # Set this storage path independently before running. MHC-audit.json.
        mhc = json.loads((ROOT).read_text())
        if set(genes["GENE"]) & set(mhc["excluded_entrez_ids"]):
            raise RuntimeError("Excluded MHC gene appears in output")
        # Set this storage path independently before running. human-gene-map.tsv.
        mapping = pd.read_csv(ROOT, sep="\t", dtype=str)
        mapped = genes.merge(mapping, left_on="GENE", right_on="entrez_id", how="left", validate="one_to_one", indicator=True)
        if mapped["_merge"].ne("both").any():
            raise RuntimeError("MAGMA output cannot be mapped to original gene locations")
        eligible = mapped.loc[mapped["mapping_status"].eq("HGNC_unique_entrez")].copy()
        if eligible["symbol"].isna().any() or eligible["symbol"].duplicated().any():
            raise RuntimeError("Ambiguous approved human symbols")
        eligible["entrez_sort"] = pd.to_numeric(eligible["GENE"])
        eligible = eligible.sort_values(["ZSTAT", "entrez_sort"], ascending=[False, True], kind="mergesort")
        mapped["entrez_sort"] = pd.to_numeric(mapped["GENE"])
        raw_top = mapped.sort_values(["ZSTAT", "entrez_sort"], ascending=[False, True], kind="mergesort").head(N_TOP).copy()
        raw_top.insert(0, "original_rank", np.arange(1, len(raw_top) + 1))
        top = raw_top.loc[raw_top["mapping_status"].eq("HGNC_unique_entrez")].copy()

        #3. =========== Export the weighted human input for scDRS
        human_z = eligible[["symbol", "ZSTAT"]].copy()
        human_z.columns = ["GENE", TRAIT]
        # Set this storage path independently before running. Trpchevska2022.gene-zscore.tsv.
        human_z.to_csv(ROOT, sep="\t", index=False, float_format="%.17g")
        # Set this storage path independently before running. Trpchevska2022.top1000.tsv.
        raw_top[["original_rank", "GENE", "symbol_ncbi37", "symbol", "mapping_status", "ZSTAT", "P", "NSNPS", "N"]].to_csv(ROOT, sep="\t", index=False, float_format="%.17g")
        weighted = ",".join(f"{r.symbol}:{r.ZSTAT:.17g}" for r in top.itertuples())
        # Set this storage path independently before running.
        (ROOT).write_text(f"TRAIT\tGENESET\n{TRAIT}\t{weighted}\n", encoding="utf-8")
        audit = {"gene_count": len(genes), "unique_approved_symbol_count": len(eligible),
                 "unmapped_entrez_count": int(mapped["symbol"].isna().sum()),
                 "raw_top1000_unmapped": raw_top.loc[raw_top["symbol"].isna(), ["GENE", "symbol_ncbi37", "ZSTAT", "P"]].to_dict("records"),
                 "top1000_selection": "Fix top 1000 Entrez genes among all valid MAGMA genes before HGNC, ortholog, or expression mapping; no replacement of unmapped genes; ties resolved by numeric Entrez ID",
                 "original_rank_n": len(raw_top), "weighted_gs_n": len(top),
                 "dropped_IDs": raw_top.loc[raw_top["symbol"].isna(), "GENE"].tolist(),
                 "top1000_min_ZSTAT": float(raw_top["ZSTAT"].min()),
                 "P_bonferroni_threshold": 0.05 / len(genes), "bonferroni_genes": int(genes["P"].lt(0.05 / len(genes)).sum()),
                 "max_abs_ZSTAT_vs_norm_isf_P": max_z_error,
                 "Z_P_consistency": "All five-significant-digit native ZSTAT/P rounding intervals overlap; native values remain unchanged",
                 "Z_P_rounding_interval_violations": int((~compatible).sum()), "output_columns": required,
                 "top10": top[["GENE", "symbol", "ZSTAT", "P"]].head(10).to_dict("records"),
                 "method": {"MAGMA": "1.10 win/s", "model": "SNP-wise mean", "gene_window_kb_up_down": [35, 10],
                            "GWAS_MAF": ">0.01 and <0.99 effect-allele frequency", "seed": 42,
                            "N": "published per-SNP total sample size", "MHC": "252 excluded gene IDs from the original research dependency; 19175 retained input genes"}}
        audit["files"] = {}
        # Set this storage path independently before running. Trpchevska2022.top1000.tsv.
        for path in [GENE_RESULTS, ROOT, ROOT, ROOT] + list(WORK.glob("Trpchevska2022.genes.raw*")):
            digest = hashlib.sha256()
            with path.open("rb") as stream:
                for block in iter(lambda: stream.read(4 * 1024 * 1024), b""):
                    digest.update(block)
            audit["files"][path.name] = {"path": str(path.resolve()), "bytes": path.stat().st_size, "sha256": digest.hexdigest()}
        # Set this storage path independently before running. gene-result-QC.json.
        (ROOT).write_text(json.dumps(audit, indent=2), encoding="utf-8")
        print(json.dumps(audit), flush=True)

        return audit


    #8. =========== Prespecified conditional competitive E3 and technical-only sensitivity
    def analyze_gene_sets(mode):
        MODES = {"primary": ["condition=mean_expression,dynamic_candidate"], "sensitivity": []}
        if mode not in MODES:
            raise ValueError("Expected primary or sensitivity")
        # Validate the locked RNA-derived background and covariates
        # Set this storage path independently before running. E3-background.txt.
        bg_path = ROOT
        # Set this storage path independently before running. E3-sets.txt.
        set_path = ROOT
        # Set this storage path independently before running. E3-gene-covar.txt.
        cov_path = ROOT
        expected_cov_sha = "cffecc54316ceb83384e63158450e3d1250674a4015f86230f0659a0d332931e"
        expected_background_sha = "9a955814ac5431c37948d9c4afc32a62dfdf78ec604b34516c9f85cf99268b2f"
        expected_sets_sha = "27b7b0fccb7f069b00459eb058437ff9b7b6cef99379fa7f99ef2eaea340122c"
        if digest(bg_path) != expected_background_sha or digest(set_path) != expected_sets_sha:
            raise RuntimeError("The frozen tested background or 18 direction sets have changed")
        if digest(cov_path) != expected_cov_sha:
            raise RuntimeError("The pre-specified covariate table has changed")
        background = bg_path.read_text().split()
        if len(background) != 10186 or len(set(background)) != len(background):
            raise RuntimeError("Unexpected tested-gene background")
        sets = {}
        for line in set_path.read_text().splitlines():
            fields = line.split()
            if len(fields) < 2 or fields[0] in sets or len(set(fields[1:])) != len(fields[1:]):
                raise RuntimeError("Malformed, empty, or duplicated gene set")
            if not set(fields[1:]).issubset(background):
                raise RuntimeError("Gene set is not contained in the tested background")
            sets[fields[0]] = set(fields[1:])
        if len(sets) != 18:
            raise RuntimeError("The prespecified family must contain 18 sets")
        cov = pd.read_csv(cov_path, sep=r"\s+", dtype={"GENE": str})
        if list(cov.columns) != ["GENE", "mean_expression", "dynamic_candidate"]:
            raise RuntimeError("Unexpected conditional-model variables")
        if cov["GENE"].tolist() != background or not np.isfinite(cov.iloc[:, 1:]).all().all():
            raise RuntimeError("Covariates must completely and uniquely cover the background")
        if not cov["dynamic_candidate"].isin([0, 1]).all() or int(cov["dynamic_candidate"].sum()) != 3632:
            raise RuntimeError("Unexpected frozen global-dynamic membership")
        dynamic = set(cov.loc[cov["dynamic_candidate"].eq(1), "GENE"])
        if any(not members.issubset(dynamic) for members in sets.values()):
            raise RuntimeError("Direction sets must come from frozen globally dynamic genes")
        gene_result = native_file(".genes.out")
        raw_result = native_file(".genes.raw")
        genes = pd.read_csv(gene_result, sep=r"\s+", comment="#", dtype={"GENE": str})
        if genes["GENE"].duplicated().any() or len(genes) < 15000:
            raise RuntimeError("Invalid full MAGMA gene result")
        valid_ids = set(genes["GENE"])
        effective = set(background) & valid_ids
        effective_cov = cov.loc[cov["GENE"].isin(effective)]
        if len(effective_cov) != len(effective) or np.linalg.matrix_rank(np.column_stack([np.ones(len(effective_cov)), effective_cov.iloc[:, 1:].to_numpy()])) != 3:
            raise RuntimeError("Effective conditional-model covariates are incomplete or collinear")
        sizes = {name: {"input_n": len(members), "effective_n": len(members & effective)} for name, members in sets.items()}
        for name, members in sets.items():
            indicator = effective_cov["GENE"].isin(members).astype(int).to_numpy()
            design = np.column_stack([np.ones(len(effective_cov)), effective_cov.iloc[:, 1:].to_numpy(), indicator])
            if np.linalg.matrix_rank(design) != 4:
                raise RuntimeError(f"Gene set is collinear with the prespecified covariates: {name}")
        audit = {"method": "MAGMA v1.10 competitive gene-set regression", "mode": mode,
                 "prespecified_tests": 18, "technical_correction": "default correct=all (gene size, density, inverse MAC and per-gene N)",
                 "conditioned_covariates": ["mean_expression", "dynamic_candidate"] if mode == "primary" else [],
                 "set_direction": "one-sided positive", "BH_family": "18 gene-set coefficients only; separately within each model specification",
                 "background_input_n": len(background), "background_effective_n": len(effective),
                 "background_missing_gene_result": sorted(set(background) - valid_ids, key=int),
                 "dynamic_input_n": len(dynamic), "dynamic_effective_n": len(dynamic & effective), "sets": sizes,
                 "definition": "Globally dynamic genes with a positive or negative modeled endpoint difference in the specified segment; not independently significant within each segment",
                 "input_files": {p.name: {"path": str(p.resolve()), "bytes": p.stat().st_size, "sha256": digest(p)}
                                 for p in [bg_path, set_path, cov_path, gene_result, raw_result]}}


        #3. =========== Run the official, prespecified competitive model
        prefix = WORK / ("E3-" + mode)
        # Set this storage path independently before running. magma.exe.
        args = [str(ROOT), "--gene-results", str(raw_result), "--set-annot", str(set_path),
                "--gene-covar", str(cov_path), "--settings", "gene-include=" + str(bg_path), "abbreviate=0",
                "--model", "analyse=sets", "direction-sets=pos", "correct=all"] + MODES[mode] + ["--out", str(prefix), "--seed", "42"]
        audit["command"] = args
        tail = deque(maxlen=32)
        warnings = []
        process = subprocess.Popen(args, cwd=WORK, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                                   text=True, encoding="utf-8", errors="replace")
        for line in process.stdout:
            tail.append(line.rstrip())
            if "WARNING" in line or "ERROR" in line:
                warnings.append(line.rstrip())
                if len(warnings) <= 8:
                    print(line.rstrip(), flush=True)
        code = process.wait()
        print("\n".join(tail), flush=True)
        if code:
            raise RuntimeError(f"MAGMA {mode} gene-set model failed with code {code}")
        audit["warnings"] = warnings
        native_candidates = [p for p in [Path(str(prefix) + ".gsa.out"), Path(str(prefix) + ".gsa.out.txt")] if p.exists()]
        if len(native_candidates) != 1:
            raise RuntimeError("Missing or ambiguous native GSA result")
        native = native_candidates[0]
        text = native.read_text()
        total = re.search(r"TOTAL_GENES\s*=\s*(\d+)", text)
        if total is None or int(total.group(1)) != len(effective):
            raise RuntimeError("Native TOTAL_GENES differs from the prespecified effective background")
        result = pd.read_csv(native, sep=r"\s+", comment="#")
        if not {"VARIABLE", "TYPE", "NGENES", "BETA", "BETA_STD", "SE", "P"}.issubset(result.columns):
            raise RuntimeError("Unexpected native GSA output columns")
        set_result = result.loc[result["TYPE"].eq("SET")].copy()
        if len(set_result) != 18 or set(set_result["VARIABLE"]) != set(sets) or set_result["VARIABLE"].duplicated().any():
            raise RuntimeError("Native gene-set testing family changed")
        if not np.isfinite(set_result[["BETA", "BETA_STD", "SE", "P"]]).all().all() or not set_result["P"].between(0, 1).all() or not set_result["SE"].gt(0).all():
            raise RuntimeError("Invalid competitive regression coefficients or P values")
        for row in set_result.itertuples():
            if int(row.NGENES) != sizes[row.VARIABLE]["effective_n"]:
                raise RuntimeError(f"Native gene-set coverage differs: {row.VARIABLE}")
        if mode == "primary":
            cov_rows = result.loc[result["TYPE"].eq("COVAR")]
            if len(cov_rows) != 36 or cov_rows["VARIABLE"].value_counts().to_dict() != {"mean_expression": 18, "dynamic_candidate": 18}:
                raise RuntimeError("Conditional regressions did not include both covariates in all 18 models")
        order = np.argsort(set_result["P"].to_numpy(), kind="stable")
        sorted_p = set_result["P"].to_numpy()[order]
        adjusted = np.minimum(1.0, np.minimum.accumulate((sorted_p * len(sorted_p) / np.arange(1, len(sorted_p) + 1))[::-1])[::-1])
        q = np.empty(len(adjusted))
        q[order] = adjusted
        set_result["Q_BH_18"] = q
        set_result["model"] = mode
        set_result["background_n"] = len(effective)
        set_result["input_set_n"] = set_result["VARIABLE"].map(lambda x: sizes[x]["input_n"])
        set_result["NGENES"] = set_result["NGENES"].astype(int)
        set_result.to_csv(Path(str(prefix) + ".tsv"), sep="\t", index=False, float_format="%.17g", lineterminator="\n")
        audit["native_TOTAL_GENES"] = int(total.group(1))
        audit["native_result_columns"] = result.columns.tolist()
        audit["BH_significant_sets"] = set_result.loc[set_result["Q_BH_18"].lt(0.05), "VARIABLE"].tolist()
        audit["files"] = {p.name: {"path": str(p.resolve()), "bytes": p.stat().st_size, "sha256": digest(p)}
                          for p in [native, Path(str(prefix) + ".tsv")]}
        Path(str(prefix) + ".json").write_text(json.dumps(audit, indent=2), encoding="utf-8")
        print(json.dumps(audit), flush=True)

        return set_result, audit


    return {"validate_resources": validate_resources, "prepare_annotations": prepare_annotations,
            "prepare_inputs": prepare_inputs, "annotate": annotate, "analyze_genes": analyze_genes,
            "analyze_gene_batches": analyze_gene_batches, "merge_gene_batches": merge_gene_batches,
            "export_gene_statistics": export_gene_statistics, "analyze_gene_sets": analyze_gene_sets}

#7. =========== Representative public variant-to-gene evidence
def prepare_sptbn1_evidence():
    # Set this storage path independently before running.
    work = ROOT
    # Set this storage path independently before running.
    reference = ROOT
    lo, hi = 54317683, 55317683
    # Set this storage path independently before running. SPTBN1-region.tsv.gz.
    region_file = ROOT
    if not region_file.exists():
        # Set this storage path independently before running. hl_ma_nov18_v2_summarystats_09122021.txt.gz.
        source = ROOT
        with source.open("rb") as stream:
            actual = hashlib.file_digest(stream, "sha256").hexdigest()
        assert actual == "77e6b9e07435f0dfdc999d15511260dd923986fd99229006928bb693e577a6fe"
        parts = []
        for chunk in pd.read_csv(source, sep=r"\s+", chunksize=250000):
            parts.append(chunk.loc[chunk.CHR.eq(2) & chunk.BP.between(lo, hi)])
        pd.concat(parts).to_csv(region_file, sep="\t", index=False)
    gwas = pd.read_csv(region_file, sep="\t")
    assert len(gwas) == 3257 and not gwas.SNP.duplicated().any()
    identifiers = set(gwas.SNP)
    parts, offset = [], 0
    for chunk in pd.read_csv(reference.with_suffix(".bim"), sep=r"\s+", names=["chr", "snp", "cm", "bp", "a1", "a2"], chunksize=1000000):
        positions = np.flatnonzero(chunk.snp.isin(identifiers).to_numpy())
        selected = chunk.iloc[positions].copy()
        selected["bed_index"] = positions + offset
        parts.append(selected)
        offset += len(chunk)
    bim = pd.concat(parts, ignore_index=True)
    assert not bim.snp.duplicated().any() and len(bim) == 3198
    coordinate_check = gwas.merge(bim, left_on="SNP", right_on="snp", how="inner", validate="one_to_one")
    assert coordinate_check.CHR.eq(coordinate_check.chr).all() and coordinate_check.BP.eq(coordinate_check.bp).all()
    lead = bim.loc[bim.snp.eq("rs6545432")]
    assert len(lead) == 1
    n_people = sum(1 for _ in reference.with_suffix(".fam").open())
    assert n_people == 503
    stride = (n_people + 3) // 4
    shifts = np.arange(4, dtype="uint8") * 2
    lookup = np.asarray([0, np.nan, 1, 2])
    with reference.with_suffix(".bed").open("rb") as stream:
        assert stream.read(3) == bytes([0x6c, 0x1b, 0x01])

        def dose(index):
            stream.seek(3 + int(index) * stride)
            encoded = np.frombuffer(stream.read(stride), dtype="uint8")
            return lookup[((encoded[:, None] >> shifts) & 3).ravel()[:n_people]]

        lead_dose = dose(lead.bed_index.iloc[0])
        correlations, denominators = [], []
        for index in bim.bed_index:
            genotype = dose(index)
            keep = np.isfinite(genotype) & np.isfinite(lead_dose)
            denominators.append(int(keep.sum()))
            correlations.append(np.corrcoef(genotype[keep], lead_dose[keep])[0, 1] ** 2 if np.std(genotype[keep]) > 0 else np.nan)
    bim["r2"], bim["n_pairwise"] = correlations, denominators
    result = gwas.merge(bim[["snp", "bed_index", "r2", "n_pairwise"]], left_on="SNP", right_on="snp", how="left", validate="one_to_one").drop(columns="snp")
    assert result.loc[result.SNP.eq("rs6545432"), "r2"].iloc[0] > .999999
    # Set this storage path independently before running. SPTBN1-locus.tsv.gz.
    result.to_csv(ROOT, sep="\t", index=False)
    # Set this storage path independently before running. SPTBN1-Thyroid-GTEx-v8.json.
    portal_file = ROOT
    if not portal_file.exists():
        url = "https://gtexportal.org/api/v2/association/dyneqtl"
        query = dict(gencodeId="ENSG00000115306.15", variantId="chr2_54590546_A_G_b38", tissueSiteDetailId="Thyroid", datasetId="gtex_v8")
        with urllib.request.urlopen(url + "?" + urllib.parse.urlencode(query), timeout=60) as response:
            payload = {"url": url, "query": query, "response": json.load(response)}
        portal_file.write_text(json.dumps(payload, indent=2), encoding="utf-8")
    portal = json.loads(portal_file.read_text(encoding="utf-8"))["response"]
    assert portal["gencodeId"] == "ENSG00000115306.15" and portal["variantId"] == "chr2_54590546_A_G_b38"
    assert portal["tissueSiteDetailId"] == "Thyroid" and len(portal["data"]) == len(portal["genotypes"]) == 574
    assert np.isclose(portal["nes"], -.16865015096079863) and np.isclose(portal["pValue"], 1.3366281383919387e-13, rtol=1e-6, atol=0)
    metadata = dict(region="GRCh37/hg19 chr2:54317683-55317683", GWAS_SNPs=len(gwas), EUR_LD_SNPs=len(bim), coordinate_conflicts=0,
                    reference_individuals=n_people, missing_reference_SNPs=len(gwas)-len(bim), lead="rs6545432",
                    LD="Squared Pearson dosage correlation; pairwise-complete genotypes; SNP-major BED codes 00/01/10/11 = 0/missing/1/2",
                    reference="Official MAGMA g1000_eur; resource-manifest.json records source and hashes",
                    GTEx_plot="Public normalized expression is plotted; P and NES come from the covariate-adjusted GTEx model and are not re-estimated from displayed expression. This does not reproduce colocalization or establish a cochlear eQTL.")
    # Set this storage path independently before running. SPTBN1-locus.tsv.gz.
    for path in [region_file, ROOT, portal_file]:
        with path.open("rb") as stream:
            metadata[path.name] = {"sha256": hashlib.file_digest(stream, "sha256").hexdigest(), "bytes": path.stat().st_size}
    # Set this storage path independently before running. SPTBN1-evidence.json.
    (ROOT).write_text(json.dumps(metadata, indent=2), encoding="utf-8")
    print(json.dumps({k: metadata[k] for k in ["GWAS_SNPs", "EUR_LD_SNPs", "coordinate_conflicts", "reference_individuals"]}))

#8. =========== Scientific previews in memory
def figure_tools():
    import os
    # Set this storage path independently before running.
    os.environ.setdefault("MPLCONFIGDIR", str(ROOT))
    import matplotlib
    matplotlib.use("Agg")
    import matplotlib.pyplot as plt
    from matplotlib.colors import ListedColormap, BoundaryNorm
    # Set this storage path independently before running.
    TASK = ROOT
    plt.rcParams.update({"font.family":"DejaVu Sans", "font.size":10,
                         "axes.spines.top":False, "axes.spines.right":False})
    def candidate_coverage():
        # Set this storage path independently before running. candidate_group_summary.tsv.
        table = pd.read_csv(ROOT, sep="\t")
        groups = table[["group", "stage", "state"]].drop_duplicates()
        stages = ["E9.5", "E10.5", "E11.5", "E13.5", "E14.5", "E15.5", "E16.5", "E17.5", "E18.5", "P1", "P7", "P28"]
        groups = groups.assign(rank=groups.stage.map(dict(zip(stages, range(len(stages)))))).sort_values("rank", kind="stable")
        common = ["Acadvl", "Cldn7", "Ctdnep1", "Dlg4", "Mmab", "Nucks1", "Rab29", "Sptbn1", "Ythdf1"]
        genes = common + [g for g in table.gene_symbol.unique() if g not in common]
        coverage = table.assign(coverage=table.n_measured_cells / table.n_cells).pivot(index="group", columns="gene_symbol", values="coverage").reindex(index=groups.group, columns=genes)
        fig, ax = plt.subplots(figsize=(11, 11), layout="constrained")
        im = ax.imshow(coverage.values, vmin=0, vmax=1, cmap="Blues", aspect="auto", interpolation="none")
        ax.set_xticks(range(len(genes)), genes, rotation=55, ha="right", fontstyle="italic")
        ax.set_yticks(range(len(groups)), groups.group, fontsize=8)
        ax.axvline(8.5, color="black", lw=1)
        for i in range(1, len(groups)):
            if groups.iloc[i].stage != groups.iloc[i - 1].stage:
                ax.axhline(i - .5, color="white", lw=1.3)
        ax.set_title("Genetically supported candidates: measurement coverage", loc="left", fontweight="bold", pad=15)
        ax.set_xlabel("9 genes measured in every source  |  8 genes with source-dependent coverage", labelpad=10)
        cb = fig.colorbar(im, ax=ax, shrink=.5, pad=.02)
        cb.set_label("Fraction of cells whose native assay contains the gene")
        return fig

    def sptbn1_evidence():
        # Set this storage path independently before running. SPTBN1-locus.tsv.gz.
        locus = pd.read_csv(ROOT, sep="\t")
        # Set this storage path independently before running. human-gene-map.tsv.
        genes = pd.read_csv(ROOT, sep="\t")
        lo, hi = 54317683, 55317683
        genes = genes[genes.CHR.astype(str).eq("2") & genes.STOP.ge(lo) & genes.START.le(hi)].sort_values("START")
        # Set this storage path independently before running. SPTBN1-Thyroid-GTEx-v8.json.
        response = json.loads((ROOT).read_text())["response"]
        expression = np.asarray(response["data"], dtype=float)
        genotypes = np.asarray(response["genotypes"], dtype=int)
        assert len(expression) == len(genotypes) == 574
        assert [int((genotypes == i).sum()) for i in range(3)] == [154, 291, 129]
        lead = locus[locus.SNP.eq("rs6545432")].iloc[0]
        assert int(lead.BP) == 54817683 and float(lead.r2) > .999999
        fig = plt.figure(figsize=(12, 5.5), layout="constrained")
        grid = fig.add_gridspec(2, 2, width_ratios=[1.7, 1], height_ratios=[3.4, 1.25], wspace=.13)
        ax = fig.add_subplot(grid[0, 0])
        track = fig.add_subplot(grid[1, 0], sharex=ax)
        box = fig.add_subplot(grid[:, 1])
        cmap = ListedColormap(["#2166ac", "#67a9cf", "#fee08b", "#f46d43", "#a50026"])
        norm = BoundaryNorm([0, .2, .4, .6, .8, 1.000001], cmap.N)
        observed = locus.r2.notna()
        ax.scatter(locus.loc[~observed, "BP"] / 1e6, -np.log10(locus.loc[~observed, "P.value"]), color=".75", s=11, linewidths=0, zorder=1)
        points = ax.scatter(locus.loc[observed, "BP"] / 1e6, -np.log10(locus.loc[observed, "P.value"]), c=locus.loc[observed, "r2"], norm=norm, cmap=cmap, s=13, linewidths=0, zorder=2)
        ax.scatter([lead.BP / 1e6], [-np.log10(lead["P.value"])], marker="D", s=55, c="#72246c", edgecolors="black", linewidths=.6, zorder=5)
        ax.annotate("rs6545432", (lead.BP / 1e6, -np.log10(lead["P.value"])), xytext=(12, 12), textcoords="offset points", fontsize=9, color="#72246c")
        ax.axhline(-np.log10(5e-8), ls="--", color=".6", lw=.7)
        ax.set_ylim(-.2, 15.8)
        ax.set_ylabel("Hearing-loss GWAS −log₁₀(P)")
        ax.set_title("SPTBN1 locus: adult hearing-loss association", loc="left", fontweight="bold", fontsize=11)
        prior = "Published GWAS–thyroid eQTL colocalization\nPPH4 = 0.928 (Praveen et al., Data 4)"
        # Set this storage path independently before running. Trpchevska2022.top1000.tsv.
        rank_file = ROOT
        if rank_file.exists():
            ranked = pd.read_csv(rank_file, sep="\t")
            gene_row = ranked.loc[ranked.symbol.eq("SPTBN1")].iloc[0]
            prior += f"\nMAGMA gene P = {gene_row.P:.2g}; genome-wide rank {int(gene_row.original_rank)}"
        ax.text(.015, .96, prior, transform=ax.transAxes, va="top", fontsize=8, bbox={"facecolor":"white", "alpha":.85, "edgecolor":"none"})
        ax.text(.99, -np.log10(5e-8)+.1, "P = 5 × 10⁻⁸", transform=ax.get_yaxis_transform(), ha="right", va="bottom", color=".4", fontsize=7)
        ax.tick_params(labelbottom=False)
        cb = fig.colorbar(points, ax=ax, fraction=.035, pad=.015, ticks=[0, .2, .4, .6, .8, 1])
        cb.set_label("EUR LD r² with rs6545432", fontsize=9)
        for row, gene in enumerate(genes.itertuples()):
            color = "#72246c" if gene.symbol == "SPTBN1" else ".35"
            y = 5 - row
            track.plot([max(gene.START, lo) / 1e6, min(gene.STOP, hi) / 1e6], [y, y], c=color, lw=3, solid_capstyle="butt")
            track.text((max(gene.START, lo) + min(gene.STOP, hi)) / 2e6, y + .2, str(gene.symbol) + (" →" if gene.STRAND == "+" else " ←"), ha="center", va="bottom", fontsize=8, color=color, fontstyle="italic")
        track.set_ylim(-.5, 6.1)
        track.set_xlim(lo / 1e6, hi / 1e6)
        track.set_yticks([])
        for side in ["left", "right", "top"]:
            track.spines[side].set_visible(False)
        track.set_xlabel("Chromosome 2 position (Mb; GRCh37/hg19)")
        data = [expression[genotypes == i] for i in range(3)]
        artists = box.boxplot(data, positions=[0, 1, 2], widths=.5, patch_artist=True, showfliers=False, medianprops={"color":"black", "lw":1.2}, boxprops={"lw":.8})
        rng = np.random.default_rng(42)
        for i, values in enumerate(data):
            artists["boxes"][i].set_facecolor(["#b8d9e8", "#76aecb", "#3e7f9f"][i])
            box.scatter(i + rng.uniform(-.16, .16, len(values)), values, s=5, color=".25", alpha=.28, linewidths=0, zorder=1)
        box.set_xticks([0, 1, 2], ["AA\nn = 154", "AG\nn = 291", "GG\nn = 129"])
        box.set_xlabel("rs6545432 genotype")
        box.set_ylabel("Portal normalized SPTBN1 expression")
        box.set_title("GTEx v8: adult human thyroid", loc="left", fontweight="bold", fontsize=11)
        box.text(.04, .97, "GTEx covariate-adjusted eQTL model\nP = 1.34 × 10⁻¹³; NES per G = −0.169", transform=box.transAxes, va="top", fontsize=9)
        box.margins(y=.18)
        fig.suptitle("Genetic evidence for a candidate gene", x=.015, ha="left", fontweight="bold", fontsize=14)
        return fig

    def candidate_development():
        """One copy of each accepted segment; postnatal stages remain discrete."""
        # Set this storage path independently before running. candidate9_segment_adaptive.h5.
        with h5py.File(ROOT, "r") as handle:
            names = handle["gene_symbols"].asstr()[()]
            segments = handle["segment_order"].asstr()[()]
            n_bins = handle["effective_quantile_bins"][()]
            values = handle["mean_log10_1p_CPM"][()]
            libraries = handle["n_used_libraries"][()]
            used_cells = handle["n_used_cells"][()]
        selected = [(segment, b) for segment in range(9) for b in range(int(n_bins[segment]))]
        matrix = np.stack([values[s, b] for s, b in selected], axis=1)
        library_n = np.asarray([libraries[s, b] for s, b in selected])
        assert matrix.shape[0] == len(names) == 9 and np.nanmin(matrix) >= 0
        # Set this storage path independently before running. candidate_group_summary.tsv.
        post = pd.read_csv(ROOT, sep="\t")
        post_order = [f"{stage} | Type {state}" for stage in ["P1", "P7", "P28"] for state in ["Ia", "Ib", "Ic", "II"]]
        post_matrix = post.pivot(index="gene_symbol", columns="group", values="mean_log10_1p_CPM_equal_library").reindex(index=names, columns=post_order)
        assert not post_matrix.isna().any().any()
        post_lib = post.drop_duplicates("group").set_index("group").reindex(post_order).n_measured_libraries.to_numpy()
        colors = ["#8A6E9C", "#5A89AD", "#7F7197", "#4F9695", "#339EB8", "#72AA8F", "#A18AB2", "#D66E73", "#A363A8"]
        fig = plt.figure(figsize=(13.5, 5.4), layout="constrained")
        grid = fig.add_gridspec(3, 2, width_ratios=[5, 1.6], height_ratios=[.42, 3.2, .32], wspace=.045, hspace=.035)
        top = fig.add_subplot(grid[0, 0])
        ax = fig.add_subplot(grid[1, 0])
        depth = fig.add_subplot(grid[2, 0], sharex=ax)
        post_top = fig.add_subplot(grid[0, 1])
        post_ax = fig.add_subplot(grid[1, 1], sharey=ax)
        post_depth = fig.add_subplot(grid[2, 1], sharex=post_ax)
        cmap = plt.colormaps["magma"].copy()
        cmap.set_bad("#d9d9d9")
        im = ax.imshow(matrix, cmap=cmap, vmin=0, vmax=3, interpolation="none", aspect="auto")
        post_ax.imshow(post_matrix.to_numpy(), cmap=cmap, vmin=0, vmax=3, interpolation="none", aspect="auto")
        ax.set_yticks(range(9), names, fontsize=10, fontstyle="italic")
        ax.tick_params(axis="y", length=0)
        ax.set_xticks([])
        post_ax.tick_params(axis="x", labelbottom=False, bottom=False)
        post_ax.tick_params(axis="y", labelleft=False, left=False)
        start = 0
        for s, name in enumerate(segments):
            stop = start + int(n_bins[s])
            top.axvspan(start - .5, stop - .5, color=colors[s], ec="white", lw=.5)
            top.text((start + stop - 1) / 2, .5, name, ha="center", va="center", fontsize=8, color="white", fontweight="bold", rotation=90 if s < 2 else 0)
            ax.axvline(stop - .5, color="white", lw=.6)
            depth.axvline(stop - .5, color="white", lw=.6)
            start = stop
        for start, label in zip([0, 4, 8], ["P1", "P7", "P28"]):
            post_top.axvspan(start - .5, start + 3.5, color="#777777", ec="white", lw=.5)
            post_top.text(start + 1.5, .5, label, ha="center", va="center", fontsize=9, color="white")
            post_ax.axvline(start - .5, color="white", lw=.7)
            post_depth.axvline(start - .5, color="white", lw=.7)
        for plot, width in [(top, matrix.shape[1]), (post_top, 12)]:
            plot.set_xlim(-.5, width - .5)
            plot.set_ylim(0, 1)
            plot.axis("off")
        max_libraries = max(library_n.max(), post_lib.max())
        for plot, data in [(depth, library_n), (post_depth, post_lib)]:
            lib_im = plot.imshow(np.asarray(data)[None, :], cmap="Greys", vmin=0, vmax=max_libraries, aspect="auto", interpolation="none")
            plot.set_yticks([])
            plot.set_xticks([])
        post_depth.set_xticks(range(12), ["Ia", "Ib", "Ic", "II"] * 3, fontsize=8, rotation=55, ha="right")
        post_depth.tick_params(length=0)
        depth.set_xlabel("Embryonic native pseudotime increases within each frozen segment →", fontsize=9, labelpad=8)
        post_depth.set_xlabel("Postnatal stages (discrete)", fontsize=9, labelpad=8)
        depth.set_ylabel("Libraries", fontsize=8, rotation=0, ha="right", va="center")
        for plot in [ax, post_ax, depth, post_depth]:
            for spine in plot.spines.values():
                spine.set_visible(False)
        cb = fig.colorbar(im, ax=[ax, post_ax], location="right", shrink=.78, pad=.015)
        cb.set_label("Mean log₁₀(1 + CPM), equal library weights", fontsize=9)
        lib_cb = fig.colorbar(lib_im, ax=[depth, post_depth], location="right", fraction=.025, aspect=5, pad=.015, ticks=[0, max_libraries])
        lib_cb.ax.tick_params(labelsize=7)
        fig.suptitle("Genetic candidates across the accepted developmental tree", x=.015, ha="left", fontweight="bold", fontsize=14)
        top.set_title("CVG → {VGN, T}; T → {AC → Ib/Ic, AB → Ia/II}", loc="left", fontsize=9, pad=8)
        post_top.set_title("Accepted postnatal annotations", loc="left", fontsize=9, pad=8)
        return fig

    def development_risk(mode="primary"):
        """Observed accepted stage/state groups; no missing cohorts are imputed."""
        # Set this storage path independently before running. score.h5.
        with h5py.File(ROOT, "r") as handle:
            group = handle[f"{mode}/group_association"]
            table = pd.DataFrame({key: group[key][()] for key in group if key != "index"}, index=group["index"].asstr()[()])
        stages = ["E9.5", "E10.5", "E11.5", "E13.5", "E14.5", "E15.5", "E16.5", "E17.5", "E18.5", "P1", "P7", "P28"]
        table["stage"] = [name.split(" | ", 1)[0] for name in table.index]
        table["state"] = [name.split(" | ", 1)[1] for name in table.index]
        with h5py.File(WHOLE, "r") as accepted:
            category_order = accepted["obs/source_stage_celltype/categories"].asstr()[()]
        states = list(dict.fromkeys(name.split(" | ", 1)[1] for name in category_order))
        value = table.pivot(index="state", columns="stage", values="assoc_mcz").reindex(index=states, columns=stages)
        fig, ax = plt.subplots(figsize=(11, 8), layout="constrained")
        cmap = plt.colormaps["RdBu_r"].copy()
        cmap.set_bad("#e5e5e5")
        im = ax.imshow(value, cmap=cmap, vmin=-3, vmax=3, aspect="auto", interpolation="none")
        for label, row in table.iterrows():
            x, y = stages.index(row.stage), states.index(row.state)
            ax.text(x, y, str(int(row.n_cell)), ha="center", va="center", fontsize=7, color="white" if abs(row.assoc_mcz) > 2 else "black")
            if row.assoc_fdr_bh < .05:
                ax.add_patch(plt.Rectangle((x-.47, y-.47), .94, .94, fill=False, ec="#255881", lw=1.3))
        ax.set_xticks(range(len(stages)), stages, rotation=45, ha="right", fontsize=9)
        ax.set_yticks(range(len(states)), states, fontsize=8)
        ax.tick_params(length=0)
        ax.axvline(8.5, c="white", lw=3)
        passed = int((table.assoc_fdr_bh < .05).sum())
        ax.set_title(f"Exploratory hearing-loss GWAS association across developmental states\n{passed}/36 groups pass BH q < 0.05 ({mode})", loc="left", fontweight="bold", fontsize=11, pad=15)
        ax.set_xlabel("Categorical developmental stage\nBlue outline: group association BH q < 0.05 (36 groups); numbers: cells; gray: no cohort", fontsize=9, labelpad=12)
        cb = fig.colorbar(im, ax=ax, fraction=.035, pad=.02)
        cb.set_label("Group association MC Z (95th-quantile test)", fontsize=9)
        for spine in ax.spines.values():
            spine.set_visible(False)
        return fig

    def developmental_set_enrichment():
        # Set this storage path independently before running. E3-primary.tsv.
        table = pd.read_csv(ROOT, sep="\t")
        order = ["CVG", "VGN", "T", "AC", "Ib", "Ic", "AB", "Ia", "II"]
        assert len(table) == 18 and table.VARIABLE.is_unique
        limit = max(.01, float(np.abs(table.BETA_STD).max()))
        fig, ax = plt.subplots(figsize=(6.7, 5.0), layout="constrained")
        for j, direction in enumerate(["up", "down"]):
            for i, segment in enumerate(order):
                row = table.loc[table.VARIABLE.eq(segment + "_" + direction)].iloc[0]
                ax.scatter(j, i, c=row.BETA_STD, cmap="RdBu_r", vmin=-limit, vmax=limit, s=120,
                           edgecolors="black" if row.Q_BH_18 < .05 else "#b0b0b0", linewidths=1.1 if row.Q_BH_18 < .05 else .6)
                ax.text(j+.12, i, f"{row.Q_BH_18:.2g}", va="center", fontsize=8)
        ax.set_xticks([0, 1], ["Increasing along segment", "Decreasing along segment"], fontsize=9)
        ax.set_yticks(range(9), order)
        ax.invert_yaxis()
        ax.set_xlim(-.4, 1.65)
        passed = int((table.Q_BH_18 < .05).sum())
        ax.set_title(f"Hearing-loss enrichment of frozen dynamic gene sets\n{passed}/18 sets pass BH q < 0.05", loc="left", fontweight="bold", fontsize=11)
        ax.set_xlabel("Numbers: BH q over 18 sets; outline: q < 0.05\nConditional on mean expression, global dynamic membership and MAGMA technical covariates", fontsize=8, labelpad=12)
        from matplotlib.cm import ScalarMappable
        cb = fig.colorbar(ScalarMappable(norm=plt.Normalize(-limit,limit),cmap="RdBu_r"), ax=ax, fraction=.04)
        cb.set_label("Standardized competitive coefficient", fontsize=8)
        for spine in ax.spines.values():
            spine.set_visible(False)
        return fig

    def segment_risk(mode="primary"):
        with h5py.File(TASK / f"rna/segment_score_adaptive_{mode}.h5", "r") as handle:
            segments = handle["segment_order"].asstr()[()]
            observed = handle["observed_equal_library_mean"][()]
            envelope = handle["null_envelope95"][()]
            n_lib = handle["n_used_libraries"][()]
        # Set this storage path independently before running. candidate9_segment_adaptive.h5.
        with h5py.File(ROOT, "r") as handle:
            t = handle["mean_t_equal_library"][()]
            assert np.array_equal(n_lib, handle["n_used_libraries"][()])
        colors = ["#8A6E9C", "#5A89AD", "#7F7197", "#4F9695", "#339EB8", "#72AA8F", "#A18AB2", "#D66E73", "#A363A8"]
        fig, axes = plt.subplots(3, 3, figsize=(10, 7), sharey=True, layout="constrained")
        finite = np.concatenate([observed[np.isfinite(observed)], envelope[np.isfinite(envelope)]])
        padding = max(.05, .07 * np.ptp(finite))
        for s, ax in enumerate(axes.flat):
            keep = n_lib[s] > 0
            x, y = t[s, keep], observed[s, keep]
            ax.fill_between(x, envelope[s, keep, 0], envelope[s, keep, 1], color="#dddddd", linewidth=0)
            ax.plot(x, y, "o-", c=colors[s], markersize=3, lw=1.2)
            ax.axhline(0, c=".6", ls="--", lw=.6)
            ax.set_title(segments[s], loc="left", color=colors[s], fontweight="bold", fontsize=11)
            ax.set_ylim(finite.min()-padding, finite.max()+padding)
            ax.tick_params(labelsize=8)
            ax.ticklabel_format(axis="x", useOffset=False)
            if s % 3 == 0:
                ax.set_ylabel("Equal-library mean scDRS", fontsize=9)
            if s >= 6:
                ax.set_xlabel("Frozen native pseudotime", fontsize=9)
        fig.suptitle(f"Hearing-loss gene-program scores along frozen tree segments ({mode})\nGray: 95% matched-control envelope; no per-bin significance test", x=.02, ha="left", fontsize=12, fontweight="bold")
        return fig

    def sgn_reference_association():
        """Display both SGN classes; keep the original full-cochlea BH results."""
        labels = ["Type I Neurons", "Type II Neurons"]
        fig, axes = plt.subplots(1, 2, sharey=True, figsize=(8.6, 3.5), layout="constrained")
        for ax, technology in zip(axes, ["snRNA", "scRNA"]):
            table = pd.read_csv(TASK / f"gear/E1-{technology}-primary-groups.tsv", sep="\t").set_index("group")
            assert len(table) == (21 if technology == "snRNA" else 25)
            selected = table.loc[labels]
            assert np.isfinite(selected[["assoc_mcz", "assoc_fdr_bh", "n_cell"]]).all().all()
            for row, (_, values) in enumerate(selected.iterrows()):
                ax.plot([0, values.assoc_mcz], [row, row], c="#87939f", lw=1.1)
                ax.scatter(values.assoc_mcz, row, s=55, c="#456e91", zorder=3)
                ax.annotate(f"n = {int(values.n_cell)}; q = {values.assoc_fdr_bh:.3f}",
                            (values.assoc_mcz, row), xytext=(0, 12), textcoords="offset points", ha="center", fontsize=9)
            ax.axvline(0, ls="--", c=".6", lw=.7)
            ax.set_xlim(-.4, .8)
            ax.set_ylim(1.5, -.5)
            ax.set_yticks([0, 1], ["Type I SGN", "Type II SGN"])
            ax.set_xlabel("Group association MC Z", fontsize=9)
            ax.set_title("P8 snRNA reference" if technology == "snRNA" else "Pooled scRNA reference", fontsize=11)
            ax.tick_params(labelsize=9)
        fig.suptitle("SGNs were tested: neither class passes BH q < 0.05", fontsize=12, fontweight="bold")
        fig.supxlabel("Source labels: Type I/II Neurons. BH correction retains all 21 snRNA / 25 scRNA types.", fontsize=8)
        return fig

    def preview_e1(result_stem, cell_type_order=None, figsize=None):
        """Return a Figure and axes for one technology/mode, preserving author labels.

        result_stem is e.g. gear/E1-snRNA-primary. The two panels show the
        official-formula association MC Z and descriptive observed-library means.
        Dots in the second panel are libraries; no animal-level inference is drawn.
        """
        #1. =========== Read matched result tables
        result_stem = Path(result_stem)
        groups = pd.read_csv(str(result_stem) + "-groups.tsv", sep="\t", index_col=0)
        libraries = pd.read_csv(str(result_stem) + "-sample-type.tsv", sep="\t")
        assert groups.index.is_unique
        assert groups.technology.nunique() == groups.analysis_mode.nunique() == 1
        assert libraries.technology.nunique() == 1
        assert groups.technology.iloc[0] == libraries.technology.iloc[0]
        assert set(libraries.cell_type_author) == set(groups.index)
        assert np.isfinite(groups[["assoc_mcz", "assoc_fdr_bh"]]).all().all()
        assert groups.assoc_fdr_bh.between(0, 1).all()
        library_n = libraries.groupby("cell_type_author", observed=True).n_cells.sum()
        assert np.array_equal(library_n.reindex(groups.index), groups.n_cell)
        order = list(groups.index) if cell_type_order is None else list(cell_type_order)
        assert len(order) == len(set(order)) and set(order) == set(groups.index)
        groups = groups.loc[order]

        #2. =========== Preview appearance only; no statistical filters
        if figsize is None:
            figsize = (8.0, max(4.0, 0.23 * len(order) + 1.2))
        significant_color = "#456E91"
        other_color = "#A7AFB7"
        jitter = 0.09
        rng = np.random.default_rng(42)
        with plt.rc_context({"font.size": 8, "axes.titlesize": 8, "axes.labelsize": 8}):
            fig, axes = plt.subplots(1, 2, sharey=True, figsize=figsize,
                                     gridspec_kw={"width_ratios": [1.0, 1.15]}, layout="constrained")
            y = np.arange(len(order))
            colors = np.where(groups.assoc_fdr_bh.to_numpy() < 0.05, significant_color, other_color)
            axes[0].hlines(y, 0, groups.assoc_mcz, color=colors, linewidth=0.8)
            axes[0].scatter(groups.assoc_mcz, y, s=28, color=colors, linewidths=0.4, edgecolors="white", zorder=3)
            axes[0].set_yticks(y, [f"{label} (n={int(groups.loc[label, 'n_cell'])})" for label in order], fontsize=7)
            axes[0].set_xlabel("Group association MC Z")
            axes[0].set_title("Association: blue indicates BH FDR < 0.05", loc="left")
            for position, label in enumerate(order):
                subset = libraries.loc[libraries.cell_type_author.eq(label)].sort_values("sample_id_author")
                offsets = rng.uniform(-jitter, jitter, len(subset))
                axes[1].scatter(subset.mean_norm_score, position + offsets, s=13, alpha=0.75,
                                color=other_color, linewidths=0, zorder=2)
                axes[1].scatter([subset.mean_norm_score.mean()], [position], marker="|", s=70,
                                color="#20252A", linewidths=1.2, zorder=3)
            axes[1].set_title("Observed-library means\nDots: libraries; |: equal-library mean", loc="left")
            axes[1].set_xlabel("Mean normalized scDRS")
            axes[1].tick_params(axis="y", left=False, labelleft=False)
            for ax in axes:
                ax.axvline(0, color="#767C82", linewidth=0.6, linestyle="--", zorder=0)
                ax.set_ylim(len(order) - 0.5, -0.5)
                ax.set_axisbelow(True)
                ax.yaxis.grid(True, color="#ECEFF1", linewidth=0.45)
                ax.spines[["top", "right"]].set_visible(False)
                ax.spines[["left", "bottom"]].set_linewidth(0.8)
                ax.tick_params(width=0.6, length=3, labelsize=7)
            assay_label = "P8 snRNA" if groups.technology.iloc[0] == "snRNA" else "Pooled scRNA"
            fig.suptitle(f"Adult hearing-loss GWAS (Trpchevska2022) | {assay_label} | {groups.analysis_mode.iloc[0]}", fontsize=10)
            fig.canvas.draw()
        return fig, axes
    return locals()


#9. =========== Published-candidate identity and native-source coverage audit
def audit_published_candidates():
    """Audit the 19 published candidates; do not fit models or make figures."""
    # Required references only; existing source matrices remain read-only.
    from concurrent.futures import ThreadPoolExecutor
    from datetime import datetime, timezone
    from xml.sax.saxutils import escape

    # Set this storage path independently before running.
    data_dir = ROOT
    data_dir.mkdir(parents=True, exist_ok=True)
    # Set this storage path independently before running.
    paper_path = ROOT
    # Set this storage path independently before running. Ensembl116_human_mouse.tsv.
    biomart_path = ROOT
    # Set this storage path independently before running. Ensembl116_candidate_details.json.
    details_path = ROOT
    # Set this storage path independently before running. HGNC_candidate_ids.tsv.
    hgnc_path = ROOT
    # Set this storage path independently before running. Candidate_data_sources.json.
    sources_path = ROOT
    biomart_base = "https://jun2026.archive.ensembl.org/biomart/martservice"
    # Set this storage path independently before running. hgnc_complete_set.txt.
    hgnc_full = ROOT
    hgnc_url = "https://storage.googleapis.com/public-download-files/hgnc/tsv/tsv/hgnc_complete_set.txt"
    attributes = [
        "ensembl_gene_id", "external_gene_name",
        "mmusculus_homolog_ensembl_gene", "mmusculus_homolog_associated_gene_name",
        "mmusculus_homolog_orthology_type", "mmusculus_homolog_orthology_confidence",
        "mmusculus_homolog_perc_id", "mmusculus_homolog_perc_id_r1",
    ]
    field_names = [
        "human_ensembl_id", "human_symbol", "mouse_ensembl_id", "mouse_symbol",
        "orthology_type", "orthology_confidence", "mouse_percent_identity", "human_percent_identity",
    ]

    def require(condition, message):
        if not condition:
            raise ValueError(message)

    def retain(path, content):
        if path.exists():
            require(path.read_bytes() == content, f"Existing reference differs: {path.name}")
        else:
            with path.open("xb") as stream:
                stream.write(content)

    def public_get(url):
        request = urllib.request.Request(url, headers={"User-Agent": "Fig4E-candidate-audit/1.0"})
        with urllib.request.urlopen(request, timeout=45) as response:
            require(response.status == 200, "Public reference request failed")
            return response.read()

    retain(paper_path, fetch_verified(COLOC_URL, COLOC_SHA256))
    coloc = read_reference_sheet(paper_path.read_bytes(), "Data 4", header=3)
    coloc = coloc.loc[coloc["Gene symbol"].notna()].copy()
    coloc["published_gene"] = coloc["Gene symbol"].astype(str).str.replace(r"\*+$", "", regex=True).str.strip()
    coloc["human_ensembl_id"] = coloc["Ensembl ID"].astype(str).str.replace(r"\.\d+$", "", regex=True)
    probability_columns = [f"PPH{i}" for i in range(5)]
    for column in probability_columns:
        coloc[column] = pd.to_numeric(coloc[column], errors="raise")
    require(len(coloc) == 55 and coloc.human_ensembl_id.nunique() == 19, "Published candidate set changed")
    require(coloc.PPH4.ge(0.5).all(), "Published H4 inclusion rule changed")
    require(not coloc.duplicated(["human_ensembl_id", "GTEx_tissue"]).any(), "Duplicate published gene-tissue pair")
    require(np.allclose(coloc[probability_columns].sum(axis=1), 1, atol=1e-10, rtol=0),
            "Published posterior probabilities do not sum to one")
    candidates = coloc[["published_gene", "human_ensembl_id"]].drop_duplicates()
    require(candidates.human_ensembl_id.is_unique, "Published gene identity is ambiguous")
    human_ids = candidates.human_ensembl_id.tolist()

    query = (
        '<Query virtualSchemaName="default" formatter="TSV" header="1" uniqueRows="1" '
        'count="" datasetConfigVersion="0.6"><Dataset name="hsapiens_gene_ensembl" interface="default">'
        f'<Filter name="ensembl_gene_id" value="{escape(",".join(human_ids))}"/>'
        + "".join(f'<Attribute name="{name}"/>' for name in attributes)
        + "</Dataset></Query>"
    )
    biomart_url = biomart_base + "?" + urllib.parse.urlencode({"query": query})
    if not biomart_path.exists():
        body = public_get(biomart_url)
        check = pd.read_csv(io.BytesIO(body), sep="\t", dtype=str, keep_default_na=False)
        require(check.shape == (19, 8), "Unexpected Ensembl116 BioMart response")
        require(set(check.iloc[:, 0]) == set(human_ids), "BioMart omitted a human candidate")
        retain(biomart_path, body)
    homology = pd.read_csv(biomart_path, sep="\t", dtype=str, keep_default_na=False)
    require(homology.shape[1] == len(field_names), "BioMart field layout changed")
    homology.columns = field_names
    require(homology.human_ensembl_id.is_unique and set(homology.human_ensembl_id) == set(human_ids),
            "Missing or repeated human BioMart rows")

    hgnc = pd.read_csv(hgnc_full, sep="\t", dtype=str, keep_default_na=False)
    hgnc_columns = ["hgnc_id", "symbol", "ensembl_gene_id", "entrez_id",
                    "prev_symbol", "alias_symbol", "status", "locus_group", "locus_type"]
    subset = hgnc.loc[hgnc.ensembl_gene_id.isin(human_ids), hgnc_columns].copy()
    require(subset.ensembl_gene_id.is_unique and subset.status.eq("Approved").all(),
            "HGNC candidate identities are not unique approved records")
    retain(hgnc_path, subset.to_csv(sep="\t", index=False, lineterminator="\n").encode("utf-8"))
    hgnc_candidates = pd.read_csv(hgnc_path, sep="\t", dtype=str, keep_default_na=False).set_index("ensembl_gene_id")

    unmapped_ids = homology.loc[homology.mouse_ensembl_id.eq(""), "human_ensembl_id"].tolist()
    rest_urls = {"release": "https://rest.ensembl.org/info/data?content-type=application/json"}
    for gene_id in unmapped_ids:
        rest_urls[f"lookup/{gene_id}"] = f"https://rest.ensembl.org/lookup/id/{gene_id}?content-type=application/json"
        rest_urls[f"archive/{gene_id}"] = f"https://rest.ensembl.org/archive/id/{gene_id}?content-type=application/json"
        rest_urls[f"homology/{gene_id}"] = (
            f"https://rest.ensembl.org/homology/id/human/{gene_id}"
            "?target_species=mus_musculus;type=orthologues;sequence=none;content-type=application/json"
        )
    if not details_path.exists():
        def fetch_detail(item):
            key, url = item
            return key, {"url": url, "response": json.loads(public_get(url))}
        with ThreadPoolExecutor(max_workers=3) as pool:
            details = dict(pool.map(fetch_detail, rest_urls.items()))
        require(details["release"]["response"]["releases"] == [116], "REST release is not the pinned BioMart release")
        retain(details_path, (json.dumps(details, indent=2, ensure_ascii=True) + "\n").encode("utf-8"))
    details = json.loads(details_path.read_text(encoding="utf-8"))
    require(details["release"]["response"]["releases"] == [116], "Stored REST release is not 116")
    for gene_id in unmapped_ids:
        lookup = details[f"lookup/{gene_id}"]["response"]
        archive = details[f"archive/{gene_id}"]["response"]
        result = details[f"homology/{gene_id}"]["response"]["data"]
        require(lookup["id"] == gene_id and lookup["biotype"] == "lncRNA", "Unmapped candidate annotation changed")
        require(str(archive["is_current"]) == "1" and not archive["possible_replacement"],
                "Unmapped candidate is retired or has an unresolved replacement")
        require(len(result) == 1 and result[0]["id"] == gene_id and not result[0]["homologies"],
                "REST and BioMart disagree on mouse orthology")

    references = {
        "schema": "Fig4E_candidate_reference_sources_v1",
        "created_utc": datetime.now(timezone.utc).isoformat(),
        "published_workbook": {"path": paper_path.name, "url": COLOC_URL, "sha256": digest(paper_path),
                               "sheet": "Data 4", "rows": 55, "genes": 19},
        "homology": {"path": biomart_path.name, "url": biomart_url, "sha256": digest(biomart_path),
                     "ensembl_release": 116, "attributes": attributes},
        "hgnc": {"path": hgnc_path.name, "source_url": hgnc_url, "sha256": digest(hgnc_path),
                  "full_source_local": str(hgnc_full.relative_to(ROOT)), "full_source_sha256": digest(hgnc_full),
                  "selection": "Exact published Ensembl gene IDs; approved records only"},
        "unmapped_details": {"path": details_path.name, "sha256": digest(details_path)},
        "full_eQTL_summary_statistics_downloaded": False,
    }
    if sources_path.exists():
        prior = json.loads(sources_path.read_text(encoding="utf-8"))
        references["created_utc"] = prior["created_utc"]
    retain(sources_path, (json.dumps(references, indent=2, ensure_ascii=True) + "\n").encode("utf-8"))

    # Compare frozen candidate counts against the supplied native source matrices.
    native = native_rna_tools()
    with h5py.File(WHOLE, "r") as handle:
        obs, whole_var = native["frame"](handle["obs"]), native["frame"](handle["var"])
    # Set this storage path independently before running. candidates.h5.
    candidate_cache = ROOT
    with h5py.File(candidate_cache, "r") as handle:
        cell_ids = handle["cell_ids"].asstr()[()]
        cached_gene_ids = pd.Index(handle["gene_ids"].asstr()[()])
        cached_symbols = handle["gene_symbols"].asstr()[()]
        counts = handle["counts"][()]
        measured = handle["measured"][()]
        cached_studies = handle["study_id"].asstr()[()]
    require(np.array_equal(cell_ids, obs.index.to_numpy()), "Candidate-cache cells differ from the accepted cohort")
    require(len(obs) == 8340 and np.array_equal(cached_studies, obs.study_id.astype(str).to_numpy()),
            "Accepted source assignment changed")
    require(np.array_equal(np.isfinite(counts), measured), "Missing counts were not preserved as NA")
    valid = homology.mouse_ensembl_id.ne("")
    require(homology.loc[valid, "orthology_type"].eq("ortholog_one2one").all()
            and homology.loc[valid, "orthology_confidence"].eq("1").all(),
            "One-to-many or low-confidence candidate requires separate review")
    require(homology.loc[valid, "mouse_ensembl_id"].is_unique, "Repeated mouse orthologue")
    require(set(homology.loc[valid, "mouse_ensembl_id"]) == set(cached_gene_ids),
            "New orthologue set differs from existing count-cache identities")
    symbol_check = homology.loc[valid].set_index("mouse_ensembl_id").mouse_symbol
    require(symbol_check.reindex(cached_gene_ids).tolist() == list(cached_symbols), "Mouse symbols and IDs disagree")
    # Set this storage path independently before running. native_mapping.tsv.gz.
    mapping = pd.read_csv(ROOT, sep="\t", keep_default_na=False)
    source_checks, source_stats = [], []
    studies = list(native["SOURCES"])
    for study in studies:
        selected = np.flatnonzero(obs.study_id.astype(str).eq(study).to_numpy())
        table = mapping.loc[mapping.study.eq(study)].sort_values("source_column").copy()
        require(np.array_equal(table.source_column.to_numpy(), np.arange(len(table))), "Source feature order changed")
        target_counts = table.loc[table.target_id.isin(cached_gene_ids), "target_id"].value_counts()
        require(target_counts.le(1).all(), f"Ambiguous native candidate feature: {study}")
        available = [gene_id for gene_id in cached_gene_ids if gene_id in target_counts.index]
        positions = cached_gene_ids.get_indexer(available)
        require(np.array_equal(measured[selected], np.tile(cached_gene_ids.isin(available), (len(selected), 1))),
                f"Supplied source feature coverage differs from the cache: {study}")
        table["common_column"] = pd.Index(available).get_indexer(table.target_id)
        checked_paths = []
        if study == "GSE178931":
            for library in obs.iloc[selected].library_id.astype(str).unique():
                rows = selected[obs.iloc[selected].library_id.astype(str).eq(library).to_numpy()]
                stage = str(obs.iloc[rows[0]].stage_standardized).split(".")[0]
                path = native["QC"] / f"{stage}_GSE178931__{library}.h5ad"
                with h5py.File(path, "r") as handle:
                    require(native["frame"](handle["var"]).index.tolist() == table.source_feature.tolist(),
                            "178931 library feature axis changed")
                actual = native["h5_rows"](path, obs.index[rows], table.common_column.to_numpy(), len(available)).toarray()
                require(np.array_equal(actual, counts[np.ix_(rows, positions)]), "178931 native candidate counts differ")
                checked_paths.append(str(path.relative_to(ROOT)))
        elif study in {"GSE165502", "Local_P7_SET7"}:
            path = native["SOURCES"][study]
            with h5py.File(path, "r") as handle:
                require(native["frame"](handle["var"]).index.tolist() == table.source_feature.tolist(),
                        "Native feature axis changed")
            actual = native["h5_rows"](path, obs.index[selected], table.common_column.to_numpy(), len(available)).toarray()
            require(np.array_equal(actual, counts[np.ix_(selected, positions)]), f"{study} native candidate counts differ")
            checked_paths.append(str(path.relative_to(ROOT)))
        elif study == "GSE195500":
            actual, path = native["csv195500"](obs, selected, table, len(available))
            require(np.array_equal(actual.toarray(), counts[np.ix_(selected, positions)]), "195500 original CSV counts differ")
            checked_paths.append(str(path.relative_to(ROOT)))
        elif study == "GSE231513":
            path = native["SOURCES"][study]
            with gzip.open(path, "rt", encoding="utf-8") as stream:
                header = next(stream).rstrip("\n").split("\t")
            require(header[0] == "cell_id_global" and header[1:] == table.source_feature.tolist(),
                    "E10 source feature axis changed")
            selected_table = table.loc[table.common_column.ge(0)].sort_values("common_column")
            raw = pd.read_csv(path, sep="\t", index_col=0,
                              usecols=["cell_id_global"] + selected_table.source_feature.tolist())
            actual = raw.loc[obs.index[selected], selected_table.source_feature].to_numpy()
            require(np.array_equal(actual, counts[np.ix_(selected, positions)]), "E10 native candidate counts differ")
            checked_paths.append(str(path.relative_to(ROOT)))
        else:
            for library in obs.iloc[selected].library_id.astype(str).unique():
                rows = selected[obs.iloc[selected].library_id.astype(str).eq(library).to_numpy()]
                actual, path = native["mtx_p28"](obs, rows, table, len(available), library)
                require(np.array_equal(actual.toarray(), counts[np.ix_(rows, positions)]), "P28 original MTX counts differ")
                checked_paths.append(str(path.relative_to(ROOT)))
        source_checks.append({
            "study": study, "cells": len(selected), "libraries": int(obs.iloc[selected].library_id.nunique()),
            "stages": list(dict.fromkeys(obs.iloc[selected].stage_standardized.astype(str))),
            "supplied_features": len(table), "candidate_features": len(available),
            "native_counts_elementwise_equal": True, "paths": checked_paths,
        })
        for j, gene_id in enumerate(cached_gene_ids):
            present = bool(measured[selected, j].all())
            n_detected = int((counts[selected, j] > 0).sum()) if present else None
            source_stats.append({
                "study": study, "mouse_ensembl_id": gene_id, "mouse_symbol": cached_symbols[j],
                "n_cells": len(selected), "feature_available": present, "detected_cells": n_detected,
                "status": ("detected" if n_detected else "measured_zero") if present else "source_feature_not_provided",
            })

    candidate_rows = []
    for candidate in candidates.itertuples(index=False):
        row = homology.set_index("human_ensembl_id").loc[candidate.human_ensembl_id]
        frame = coloc.loc[coloc.human_ensembl_id.eq(candidate.human_ensembl_id)]
        best = frame.loc[frame.PPH4.idxmax()]
        mouse_id = row.mouse_ensembl_id
        human_symbol = hgnc_candidates.loc[candidate.human_ensembl_id, "symbol"] if candidate.human_ensembl_id in hgnc_candidates.index else row.human_symbol
        available_sources = [s["study"] for s in source_stats if s["mouse_ensembl_id"] == mouse_id and s["feature_available"]]
        candidate_rows.append({
            "published_gene": candidate.published_gene, "human_gene": human_symbol or candidate.published_gene,
            "human_ensembl_id": candidate.human_ensembl_id,
            "mouse_gene": row.mouse_symbol or None, "mouse_ensembl_id": mouse_id or None,
            "orthology_type": row.orthology_type or None, "orthology_confidence": int(row.orthology_confidence) if mouse_id else None,
            "mapping_status": "high_confidence_one_to_one" if mouse_id else "no_mouse_orthologue_in_Ensembl116",
            "coverage_status": "all_sources" if len(available_sources) == len(studies) else ("partial_sources" if mouse_id else "not_assessable"),
            "available_sources": available_sources,
            "missing_sources": [study for study in studies if study not in available_sources] if mouse_id else None,
            "best_published_tissue": best.GTEx_tissue, "best_published_H3": float(best.PPH3),
            "best_published_H4": float(best.PPH4), "published_tissue_rows": len(frame),
        })
    result = {
        "stage": "published_candidate_identity_and_source_coverage",
        "candidate_rows": candidate_rows, "source_checks": source_checks, "source_stats": source_stats,
        "checks": {"published_rows": len(coloc), "published_genes": len(candidates),
                   "high_confidence_one_to_one": int(valid.sum()), "unmapped_lncrna": len(unmapped_ids),
                   "ambiguous_human_or_mouse_pairs": 0, "fixed_cells": len(obs),
                   "candidate_cache_sha256": digest(candidate_cache), "frozen_Whole_sha256": digest(WHOLE),
                   "reference_manifest": str(sources_path.relative_to(ROOT)), "figures_created": False},
    }
    print(json.dumps(result, ensure_ascii=True, indent=2))


#10. =========== Fig4E: source-faceted observed expression and published evidence
def make_candidate_figure_data():
    """Read the approved 19-gene audit; aggregate without changing the frozen tree."""
    # Every source has ONE peak per gene across its displayed embryo AND postnatal
    # values. This is descriptive within-source scaling, not batch correction.
    # Set this storage path independently before running.
    data_dir = ROOT
    segment_order = ["CVG", "T", "AC", "Ib", "Ic", "AB", "Ia", "II"]
    common = ["Acadvl", "Cldn7", "Ctdnep1", "Dlg4", "Mmab", "Nucks1", "Rab29", "Sptbn1", "Ythdf1"]
    partial = ["Ccdc68", "Crip3", "Cyp26c1", "Eaf2", "Lmo7", "Nol12", "Ptk2", "Tcf19"]
    unmapped = ["ENSG00000255176", "MMP2-AS1"]
    row_order = common + partial + unmapped
    embryo_sources = ["GSE178931", "GSE231513", "GSE195500", "GSE165502"]
    post_stages = ["P1", "P7", "P28"]
    post_sources = ["GSE195500", "Local_P7_SET7", "GSE202920"]
    post_types = ["Ia", "Ib", "Ic", "II"]
    all_sources = embryo_sources + post_sources[1:]
    min_library_bin_cells = 3

    # Set this storage path independently before running. candidates.h5.
    with h5py.File(ROOT, "r") as handle:
        values = handle["log10_1p_CPM"][()]
        counts = handle["counts"][()]
        measured = handle["measured"][()].astype(bool)
        gene_symbols = handle["gene_symbols"].asstr()[()].tolist()
        gene_ids = handle["gene_ids"].asstr()[()].tolist()
        cells = handle["cell_ids"].asstr()[()]
        studies = handle["study_id"].asstr()[()]
        libraries = handle["library_id"].asstr()[()]
        stages = handle["stage_standardized"].asstr()[()]
        states = handle["state"].asstr()[()]
    with h5py.File(HI, "r") as handle:
        native_cells = handle["cells/_index"].asstr()[()]
        seg = handle["cells/seg"].asstr()[()]
        native_t = handle["cells/t"][()]
        tree_meta = json.loads(handle["meta/input"][()])
    # Set this storage path independently before running. candidate9_segment_adaptive.h5.
    with h5py.File(ROOT, "r") as handle:
        saved_segments = handle["segment_order"].asstr()[()].tolist()
        edges = handle["native_t_quantile_edges"][()]
        effective = handle["effective_quantile_bins"][()]
        pooled_reference = handle["mean_log10_1p_CPM"][()]
        common_reference = handle["gene_symbols"].asstr()[()].tolist()
    assert len(cells) == 8340 and len(native_cells) == 7355
    assert np.array_equal(cells[:len(native_cells)], native_cells)
    assert np.isnan(values[~measured]).all() and np.isfinite(values[measured]).all()
    assert set(gene_symbols) == set(common + partial)
    expected_edges = {"CVG": [159, 154], "VGN": [154, 142], "T": [154, 35],
                      "AC": [35, 91], "Ib": [91, 106], "Ic": [91, 81],
                      "AB": [35, 87], "Ia": [87, 60], "II": [87, 38]}
    assert tree_meta["segment_edges"] == expected_edges
    gene_columns = [gene_symbols.index(g) for g in row_order[:17]]
    values = values[:, gene_columns]
    counts = counts[:, gene_columns]
    measured = measured[:, gene_columns]

    slot_order = [(s, b) for s in segment_order for b in range(int(effective[saved_segments.index(s)]))]
    matrices = {s: np.full((19, len(slot_order)), np.nan) for s in all_sources}
    library_depth = {s: np.zeros(len(slot_order), dtype=int) for s in all_sources}
    pooled_means = []
    used_cells = set()
    bin_cell_counts = {}
    for si, segment in enumerate(saved_segments):
        selected = np.flatnonzero(seg == segment)
        finite_edges = edges[si, np.isfinite(edges[si])]
        # pd.qcut uses right-closed intervals; no jitter, smoothing, or new bins.
        bin_ids = np.searchsorted(finite_edges[1:-1], native_t[selected], side="left")
        original_bins = pd.qcut(native_t[selected], q=int(min(20, max(3, len(selected) // 50))),
                                labels=False, duplicates="drop")
        assert np.array_equal(bin_ids, original_bins)
        for b in range(int(effective[si])):
            members = selected[bin_ids == b]
            library_means = []
            source_means = {}
            for library in sorted(set(libraries[members])):
                rows = members[libraries[members] == library]
                if len(rows) < min_library_bin_cells:
                    continue
                source = np.unique(studies[rows])
                assert len(source) == 1
                observed = measured[rows].all(axis=0)
                assert np.array_equal(measured[rows], np.broadcast_to(observed, (len(rows), 17)))
                average = np.full(17, np.nan)
                average[observed] = values[np.ix_(rows, np.flatnonzero(observed))].mean(axis=0)
                library_means.append(average)
                source_means.setdefault(source[0], []).append(average)
                if segment in segment_order:
                    used_cells.update(rows.tolist())
            common_columns = [row_order.index(g) for g in common_reference]
            check = np.asarray(library_means)[:, common_columns].mean(axis=0)
            assert np.allclose(check, pooled_reference[si, b], atol=1e-12, rtol=0)
            pooled_means.append(check)
            if segment not in segment_order:
                continue
            slot = slot_order.index((segment, b))
            bin_cell_counts[(segment, b)] = len(members)
            for source, source_values in source_means.items():
                stack = np.asarray(source_values)
                present = np.isfinite(stack).any(axis=0)
                matrices[source][np.flatnonzero(present), slot] = stack[:, present].mean(axis=0)
                library_depth[source][slot] = len(stack)

    # Postnatal classes are frozen labels, not extrapolated embryo pseudotime.
    post_values = np.full((19, 12), np.nan)
    post_cell_n, post_library_n = [], []
    for j, (stage, source, subtype) in enumerate((stage, source, subtype)
            for stage, source in zip(post_stages, post_sources) for subtype in post_types):
        selected = np.flatnonzero((stages == stage) & (states == "Type " + subtype))
        assert len(selected) > 0 and set(studies[selected]) == {source}
        present = measured[selected].all(axis=0)
        assert np.array_equal(measured[selected], np.broadcast_to(present, (len(selected), 17)))
        library_means = [values[np.ix_(selected[libraries[selected] == library], np.flatnonzero(present))].mean(axis=0)
                         for library in sorted(set(libraries[selected]))]
        post_values[np.flatnonzero(present), j] = np.mean(library_means, axis=0)
        post_cell_n.append(len(selected))
        post_library_n.append(len(library_means))
    # Independently compare all postnatal summaries to the pre-existing cache.
    # Set this storage path independently before running. candidate_group_summary.tsv.
    reference_post = pd.read_csv(ROOT, sep="\t")
    for j, (stage, subtype) in enumerate((stage, subtype) for stage in post_stages for subtype in post_types):
        ref = reference_post.set_index(["group", "gene_symbol"]).loc[
            [(f"{stage} | Type {subtype}", g) for g in row_order[:17]], "mean_log10_1p_CPM_equal_library"]
        assert np.allclose(post_values[:17, j], ref.to_numpy(), atol=1e-12, rtol=0, equal_nan=True)

    peaks = {}
    relative = {}
    relative_post = post_values.copy()
    for source in all_sources:
        source_post_columns = [j for j in range(12) if post_sources[j // 4] == source]
        combined = np.concatenate([matrices[source], post_values[:, source_post_columns]], axis=1)
        observed_rows = np.isfinite(combined).any(axis=1)
        peak = np.full(19, np.nan)
        peak[observed_rows] = np.nanmax(combined[observed_rows], axis=1)
        divisor = np.where(peak > 0, peak, 1)
        relative[source] = matrices[source] / divisor[:, None]
        relative_post[:, source_post_columns] = post_values[:, source_post_columns] / divisor[:, None]
        peaks[source] = peak
        assert np.array_equal(np.isnan(relative[source]), np.isnan(matrices[source]))
    assert np.isfinite(relative_post[:9]).all()
    assert np.isnan(relative_post[9:17, :4]).all()
    assert np.isnan(relative_post[17:]).all()
    for array in list(relative.values()) + [relative_post]:
        finite = array[np.isfinite(array)]
        if finite.size:
            assert finite.min() >= 0 and finite.max() <= 1 + 1e-12

    # Set this storage path independently before running. Ensembl116_human_mouse.tsv.
    homology = pd.read_csv(ROOT, sep="\t").fillna("")
    homology.columns = ["human_id", "human", "mouse_id", "mouse", "orthology", "confidence", "identity_mouse", "identity_human"]
    assert homology.human_id.is_unique and len(homology) == 19
    mapped = homology.loc[homology.mouse.ne("")]
    assert set(mapped.mouse) == set(row_order[:17])
    assert mapped.orthology.eq("ortholog_one2one").all() and mapped.confidence.eq(1).all()
    assert [mapped.set_index("mouse").loc[g, "mouse_id"] for g in row_order[:17]] == [gene_ids[gene_symbols.index(g)] for g in row_order[:17]]
    # Set this storage path independently before running.
    assert digest(ROOT) == COLOC_SHA256
    # Set this storage path independently before running.
    coloc = read_reference_sheet((ROOT).read_bytes(), "Data 4", header=3)
    coloc = coloc.loc[coloc["Gene symbol"].notna()].copy()
    coloc["human_id"] = coloc["Ensembl ID"].astype(str).str.replace(r"\.\d+$", "", regex=True)
    for key in ["PPH3", "PPH4"]:
        coloc[key] = pd.to_numeric(coloc[key], errors="raise")
    assert len(coloc) == 55 and coloc.human_id.nunique() == 19
    evidence = []
    for gene in row_order:
        match = homology.loc[homology.mouse.eq(gene) | homology.human.eq(gene) | homology.human_id.eq(gene)]
        assert len(match) == 1
        human_id = match.iloc[0].human_id
        source_rows = coloc.loc[coloc.human_id.eq(human_id)]
        best = source_rows.loc[source_rows.PPH4.idxmax()]
        evidence.append({"mouse": gene, "human": match.iloc[0].human or human_id,
                         "H4": float(best.PPH4), "H3": float(best.PPH3), "tissue": str(best.GTEx_tissue)})
    detected = np.r_[(counts > 0).sum(axis=0), [0, 0]]
    measurable = np.r_[measured.sum(axis=0), [0, 0]]
    assert int(detected[row_order.index("Cyp26c1")]) == 6
    assert int(measurable[row_order.index("Cyp26c1")]) == 3410
    checks = {"cells_frozen": len(cells), "embryo_frozen": len(native_cells),
              "embryo_SGN_plus_shared_CVG_used_in_bin_means": len(used_cells),
              "VGN_omitted_from_SGN_display": int((seg == "VGN").sum()),
              "native_bins_with_VGN": len(pooled_means), "displayed_bins_per_source": len(slot_order),
              "legacy_nine_gene_bin_means_agree": True, "all_postnatal_cache_means_agree": True,
              "source_facets": embryo_sources, "normalization": "one gene peak per study across displayed embryo bins and postnatal groups",
              "postnatal_cells": post_cell_n, "postnatal_libraries": post_library_n,
              "candidate_rows": len(row_order), "mapped_one_to_one": len(mapped), "no_mouse_orthologue": 2,
              "published_H4_ge_0_8_all19": sum(e["H4"] >= .8 for e in evidence),
              "published_H4_ge_0_8_mapped17": sum(e["H4"] >= .8 for e in evidence[:17]),
              "Cyp26c1_detected_measurable": [int(detected[row_order.index("Cyp26c1")]), int(measurable[row_order.index("Cyp26c1")])],
              "figure_files_saved": False}
    return locals()


def make_candidate_figure(panel="embryo"):
    """Fig. 6a style: lineage blocks, thin grid, RdBu_r; in-memory preview only."""
    # Shared layout and style settings; no gene clustering or fitted trajectory.
    import matplotlib
    matplotlib.use("Agg")
    import matplotlib.pyplot as plt
    from matplotlib.patches import Rectangle
    from matplotlib.colors import Normalize
    from matplotlib.cm import ScalarMappable
    plt.rcParams.update({"font.family": "Arial", "font.size": 8, "axes.linewidth": .6,
                         "pdf.fonttype": 42, "ps.fonttype": 42})
    d = make_candidate_figure_data()
    genes, segments, slots = d["row_order"], d["segment_order"], d["slot_order"]
    cmap = plt.colormaps["RdBu_r"].copy()
    cmap.set_bad("#d9d9d9")
    group_lines = [9, 17]
    labels = [g + (" †" if g == "Cyp26c1" else "") for g in genes]
    block_gap = 1.4

    def row_style(ax, show_labels=True):
        ax.set_ylim(19, 0)
        ax.set_yticks(np.arange(19) + .5, labels if show_labels else [""] * 19,
                      fontsize=8, fontstyle="italic")
        ax.tick_params(axis="both", length=0, pad=4)
        for y in group_lines:
            ax.axhline(y, color="black", lw=1.0, zorder=4)
        for spine in ax.spines.values():
            spine.set_visible(False)

    def draw_blocks(ax, matrix, groups, tick_labels=False):
        centers, ranges = {}, {}
        x = 0
        for name, columns in groups:
            n = len(columns)
            ax.pcolormesh(np.arange(n + 1) + x, np.arange(20), np.ma.masked_invalid(matrix[:, columns]),
                          cmap=cmap, vmin=0, vmax=1, edgecolors="black", linewidth=.18, shading="flat", rasterized=False)
            ax.add_patch(Rectangle((x, 0), n, 19, fill=False, ec="black", lw=.65))
            centers[name] = x + n / 2
            ranges[name] = (x, x + n)
            x += n + block_gap
        ax.set_xlim(0, x - block_gap)
        if tick_labels:
            ticks = [ranges[name][0] + k + .5 for name, columns in groups for k in range(len(columns))]
            ax.set_xticks(ticks, d["post_types"] * 3, rotation=0, fontsize=8)
        else:
            ax.set_xticks([])
        return centers, ranges

    def tree_top(ax, centers, xlim):
        # Parent-child edges use the accepted nine-segment topology. The VGN
        # sibling is not displayed; this schematic has no branch-length scale.
        children = {"CVG": ["T"], "T": ["AC", "AB"], "AC": ["Ib", "Ic"], "AB": ["Ia", "II"]}
        heights = {"CVG": .96, "T": .72, "AC": .43, "AB": .43, "Ib": .12, "Ic": .12, "Ia": .12, "II": .12}
        for parent, descendants in children.items():
            for child in descendants:
                ax.plot([centers[parent], centers[parent], centers[child]],
                        [heights[parent], heights[child], heights[child]], c="black", lw=.65)
        for name in segments:
            ax.plot([centers[name], centers[name]], [heights[name], -.12], c="black", lw=.55)
            ax.text(centers[name], -.18, name, ha="center", va="top", fontsize=8,
                    bbox={"fc": "white", "ec": "none", "pad": .3})
        ax.set(xlim=xlim, ylim=(-.4, 1.04))
        ax.axis("off")

    if panel == "embryo":
        fig = plt.figure(figsize=(16, 11), facecolor="white")
        fig.suptitle("Fig. 4E  |  Hearing-loss candidates in the accepted SGN developmental tree",
                     x=.065, y=.983, ha="left", fontsize=13, fontweight="bold")
        subtitles = {"GSE178931": "E9.5, E11.5, E13.5", "GSE231513": "E10.5",
                     "GSE195500": "E14.5, E16.5, E18.5", "GSE165502": "E14.5–E18.5"}
        groups = [(s, [j for j, pair in enumerate(slots) if pair[0] == s]) for s in segments]
        for k, source in enumerate(d["embryo_sources"]):
            left = .125 if k % 2 == 0 else .60
            bottom = .555 if k < 2 else .15
            width, height = .37, .275
            ax = fig.add_axes([left, bottom, width, height])
            centers, ranges = draw_blocks(ax, d["relative"][source], groups)
            row_style(ax)
            top = fig.add_axes([left, bottom + height + .008, width, .074])
            tree_top(top, centers, ax.get_xlim())
            fig.text(left, bottom + height + .102, f"{'ABCD'[k]}   {source}  |  {subtitles[source]}",
                     fontsize=10, fontweight="bold", ha="left")
            source_cells = int(((d["studies"][:7355] == source) & (d["seg"] != "VGN")).sum())
            contributing_libraries = len(set(d["libraries"][list(d["used_cells"])][d["studies"][list(d["used_cells"]) ] == source]))
            ax.set_xlabel(f"Native pseudotime increases within each segment →    |    {source_cells:,} cohort cells; {contributing_libraries} contributing libraries",
                          fontsize=7, labelpad=7)
        fig.text(.065, .092, "Rows: 9 genes present in every source  /  8 genes absent from the GSE195500 feature set  /  2 human lncRNAs with no mouse orthologue reported in Ensembl 116.", fontsize=8)
        fig.text(.065, .071, "† Cyp26c1: detected in 6 / 3,410 measurable whole-atlas cells (including VGN). CVG is a shared precursor; VGN is omitted from the heatmaps. Tree heights have no time scale.", fontsize=8)
        fig.text(.065, .05, "Observed bin means, equal weights per eligible library (≥3 cells per library/bin). No fitting or interpolation. Each gene is scaled to its own peak within each study.", fontsize=8)
        fig.text(.065, .029, "Gray: no eligible library/bin, absent source feature, or no Ensembl 116 mouse match. Block width has no time scale. Style reference: Petitpré et al., 2022, Fig. 6a.", fontsize=8)
        cax = fig.add_axes([.72, .945, .14, .011])
        cb = fig.colorbar(ScalarMappable(norm=Normalize(0, 1), cmap=cmap), cax=cax, orientation="horizontal", ticks=[0, .5, 1])
        cb.ax.tick_params(labelsize=7, length=2, pad=1)
        fig.text(.715, .966, "Relative expression (within gene and study)", fontsize=8)
        fig.text(.89, .946, "■ NA", fontsize=8, color="#8c8c8c")
    elif panel == "evidence":
        fig = plt.figure(figsize=(14.6, 7.4), facecolor="white")
        fig.suptitle("Fig. 4E  |  Postnatal expression and published human genetic evidence",
                     x=.10, y=.975, ha="left", fontsize=13, fontweight="bold")
        ax = fig.add_axes([.14, .20, .32, .60])
        groups = [(stage, list(range(i * 4, (i + 1) * 4))) for i, stage in enumerate(d["post_stages"])]
        centers, ranges = draw_blocks(ax, d["relative_post"], groups, tick_labels=True)
        row_style(ax)
        for i, stage in enumerate(d["post_stages"]):
            x = centers[stage]
            ax.text(x, -.8, stage, ha="center", va="bottom", fontsize=10, fontweight="bold")
            short_source = ["GSE195500", "Local P7", "GSE202920"][i]
            ax.text(x, -1.6, short_source, ha="center", va="bottom", fontsize=8)
            lo, hi = ranges[stage]
            for j in range(4):
                ax.text(lo + j + .5, 19.9, str(d["post_cell_n"][i * 4 + j]), ha="center", va="top", fontsize=7)
            n_libraries = d['post_library_n'][i * 4]
            ax.text(x, 21, f"{n_libraries} {'library' if n_libraries == 1 else 'libraries'}", ha="center", va="top", fontsize=7)
        fig.text(.14, .886, "E   Accepted SGN subtypes (discrete stages)", fontsize=10, fontweight="bold")
        ev = fig.add_axes([.49, .20, .49, .60])
        ev.set_xlim(0, 10)
        row_style(ev, show_labels=False)
        ev.set_xticks([])
        headers = [(0, "H4"), (.9, "H3"), (2, "GTEx v8 tissue"), (7.2, "Whole-atlas detection (n/N)")]
        for x, title in headers:
            ev.text(x, -.8, title, ha="left", va="bottom", fontsize=8, fontweight="bold")
        tissue_labels = {"Brain_Nucleus_accumbens_basal_ganglia": "Brain: nucleus accumbens",
                         "Brain_Substantia_nigra": "Brain: substantia nigra", "Brain_Hypothalamus": "Brain: hypothalamus",
                         "Brain_Hippocampus": "Brain: hippocampus", "Breast_Mammary_Tissue": "Breast: mammary tissue"}
        for i, evidence in enumerate(d["evidence"]):
            y = i + .5
            ev.text(0, y, f"{evidence['H4']:.3f}", va="center", fontsize=8, fontweight="bold" if evidence["H4"] >= .8 else "normal")
            ev.text(.9, y, f"{evidence['H3']:.3f}", va="center", fontsize=8)
            ev.text(2, y, tissue_labels.get(evidence["tissue"], evidence["tissue"].replace("_", ": ", 1).replace("_", " ")), va="center", fontsize=8)
            n, total = d["detected"][i], d["measurable"][i]
            ev.text(7.2, y, f"{n:,} / {total:,}" if total else "No Ensembl 116 match", va="center", fontsize=8)
        fig.text(.49, .886, "F   Published colocalization (Praveen et al., Data 4)", fontsize=10, fontweight="bold")
        fig.text(.10, .102, "Expression scale is shared with the embryo facet from the same study. Each gene has a separate study-specific peak; colors cannot compare absolute expression between studies.", fontsize=8)
        fig.text(.10, .078, "H4: shared-variant model; H3: different-variant model. The highest-H4 tissue per gene is shown; H3 comes from the same row. Bold H4 ≥ 0.8 is a display threshold.", fontsize=8)
        fig.text(.10, .054, "GTEx tissues are adult human, non-cochlear tissues; these results do not establish an SGN eQTL. Detection uses all measurable atlas cells, including VGN and cells outside eligible bins.", fontsize=8)
        cax = fig.add_axes([.14, .936, .17, .012])
        cb = fig.colorbar(ScalarMappable(norm=Normalize(0, 1), cmap=cmap), cax=cax, orientation="horizontal", ticks=[0, .5, 1])
        cb.ax.tick_params(labelsize=7, length=2, pad=1)
        fig.text(.345, .938, "Relative expression within gene/study; gray: NA; 0 = measured zero; †: sparse detection", fontsize=8)
    else:
        raise ValueError("Fig4E panel must be embryo or evidence")
    return fig, d["checks"]


#11. =========== Cortex input acquisition and eGene screening (no colocalization)
def prepare_cortex_egenes():
    """Extract only the official Cortex eGene member; retain all tested genes."""
    import tarfile
    from datetime import datetime, timezone
    # Set this storage path independently before running.
    data_dir = ROOT
    # Set this storage path independently before running. GTEx_Cortex_eGenes.tsv.gz.
    output_path = ROOT
    # Set this storage path independently before running. GTEx_Cortex_sources.json.
    manifest_path = ROOT
    archive_url = "https://storage.googleapis.com/adult-gtex/bulk-qtl/v8/single-tissue-cis-qtl/GTEx_Analysis_v8_eQTL.tar"
    member_name = "GTEx_Analysis_v8_eQTL/Brain_Cortex.v8.egenes.txt.gz"
    header_offset, member_size = 348336640, 2728234
    archive_size = 1562828800
    archive_etag = "d35b32152bdb21316b2509c46b0af998"
    q_threshold = 0.05
    network_timeout = 60

    def ranged(start, stop):
        request = urllib.request.Request(archive_url, headers={"Range": f"bytes={start}-{stop}", "If-Match": '"' + archive_etag + '"'})
        with urllib.request.urlopen(request, timeout=network_timeout) as response:
            assert response.status == 206
            assert response.headers["Content-Range"] == f"bytes {start}-{stop}/{archive_size}"
            assert response.headers["ETag"].strip('"') == archive_etag
            content = response.read()
        assert len(content) == stop - start + 1
        return content

    data_dir.mkdir(parents=True, exist_ok=True)
    if output_path.exists():
        content = output_path.read_bytes()
        assert len(content) == member_size
    else:
        header = tarfile.TarInfo.frombuf(ranged(header_offset, header_offset + 511), "utf-8", "strict")
        assert header.name == member_name and header.size == member_size
        content = ranged(header_offset + 512, header_offset + 511 + member_size)
        raw = gzip.decompress(content)  # Includes gzip CRC and length checks.
        assert raw.startswith(b"gene_id\tgene_name\t")
        with output_path.open("xb") as stream:
            stream.write(content)
    raw = gzip.decompress(content)
    table = pd.read_csv(io.BytesIO(raw), sep="\t")
    required = ["gene_id", "gene_name", "gene_chr", "gene_start", "gene_end", "strand", "num_var", "variant_id", "rs_id_dbSNP151_GRCh38p7", "slope", "slope_se", "qval"]
    assert set(required).issubset(table.columns) and table.gene_id.is_unique
    assert table.qval.between(0, 1).all() and table.slope_se.gt(0).all()
    selected = table.loc[table.qval.le(q_threshold)]
    assert 0 < len(selected) < len(table)
    entry = {"schema": "GTEx_Cortex_inputs_v1", "dataset": "gtex_v8", "tissue": "Brain_Cortex",
             "genome_build": "GRCh38", "annotation": "GENCODE26", "eqtl_samples": 205,
             "egenes": {"path": output_path.name, "archive_url": archive_url, "archive_etag": archive_etag,
                         "archive_size": archive_size, "member": member_name,
                         "member_offset": header_offset + 512, "bytes": len(content),
                         "sha256": hashlib.sha256(content).hexdigest(), "gzip_crc_verified": True,
                         "tested_genes": len(table), "q_threshold": q_threshold, "selected_genes": len(selected),
                         "tested_variant_gene_pairs_expected": int(table.num_var.sum()),
                         "selected_variant_gene_pairs_expected": int(selected.num_var.sum())},
             "full_associations": {"status": "not_downloaded"},
             "scope": "all tested Cortex genes; no restriction to 19 published genes; no colocalization",
             "retrieved_utc": datetime.now(timezone.utc).isoformat()}
    if manifest_path.exists():
        old = json.loads(manifest_path.read_text(encoding="utf-8"))
        assert old["egenes"]["sha256"] == entry["egenes"]["sha256"]
        entry["full_associations"] = old.get("full_associations", entry["full_associations"])
    manifest_path.write_text(json.dumps(entry, indent=2) + "\n", encoding="utf-8")
    summary = {"tested_genes": len(table), "eGenes_q_le_0_05": len(selected),
               "selected_percent": round(100 * len(selected) / len(table), 3),
               "q_threshold": q_threshold, "eqtl_samples": 205,
               "expected_all_tested_pairs": int(table.num_var.sum()),
               "expected_eGene_pairs": int(selected.num_var.sum()),
               "chromosomes": selected.gene_chr.value_counts().sort_index().to_dict(),
               "most_significant": selected.nsmallest(10, "qval")[["gene_id", "gene_name", "qval", "num_var"]].to_dict("records"),
               "saved_bytes": len(content), "file": str(output_path), "colocalization_run": False}
    print(json.dumps(summary, ensure_ascii=True, indent=2))


def download_cortex_pairs():
    """Download the EBI import of original GTEx v8 statistics, not a remapping."""
    import concurrent.futures
    import time
    # Set this storage path independently before running.
    data_dir = ROOT
    # Set this storage path independently before running. GTEx_Cortex_sources.json.
    manifest_path = ROOT
    # Set this storage path independently before running. GTEx_Cortex_pairs.tsv.gz.
    pair_path = ROOT
    source_url = "https://ftp.ebi.ac.uk/pub/databases/spot/eQTL/imported/GTEx_V8/ge/Brain_Cortex.tsv.gz"
    expected_size = 3658491712
    expected_md5 = "1a83c7cb898602b87a6fa2d94c8cc71c"
    index_md5 = "44d48196fea4d73bcf91a9bbbb34abce"
    chunk_size = 16 * 1024 * 1024
    workers = 8
    retries = 4
    timeout = 90
    manifest = json.loads(manifest_path.read_text(encoding="utf-8"))
    assert manifest["dataset"] == "gtex_v8" and manifest["tissue"] == "Brain_Cortex"
    with urllib.request.urlopen(source_url.rsplit("/", 1)[0] + "/md5sum.txt", timeout=timeout) as stream:
        checksums = stream.read().decode("utf-8")
    assert any(expected_md5 in line and line.rstrip().endswith("Brain_Cortex.tsv.gz") for line in checksums.splitlines())
    assert any(index_md5 in line and line.rstrip().endswith("Brain_Cortex.tsv.gz.tbi") for line in checksums.splitlines())

    def md5_file(path):
        h = hashlib.md5()
        with path.open("rb") as stream:
            for block in iter(lambda: stream.read(8 * 1024 * 1024), b""):
                h.update(block)
        return h.hexdigest()

    def persist():
        manifest_path.write_text(json.dumps(manifest, indent=2) + "\n", encoding="utf-8")

    if pair_path.exists():
        assert pair_path.stat().st_size == expected_size and md5_file(pair_path) == expected_md5
    else:
        partial = pair_path.with_name(pair_path.name + ".partial")
        with urllib.request.urlopen(urllib.request.Request(source_url, method="HEAD"), timeout=timeout) as response:
            assert int(response.headers["Content-Length"]) == expected_size
            etag = response.headers.get("ETag", "")
        previous = manifest.get("full_associations", {})
        completed = set(previous.get("completed_chunks", [])) if previous.get("url") == source_url else set()
        if not partial.exists():
            completed.clear()
            with partial.open("xb") as stream:
                stream.truncate(expected_size)
        assert partial.stat().st_size == expected_size
        manifest["full_associations"] = {"status": "downloading", "url": source_url,
            "provider": "EBI eQTL Catalogue imported GTEx_V8/ge; original statistics in harmonized format",
            "path": pair_path.name, "expected_bytes": expected_size, "provider_md5": expected_md5,
            "source_etag": etag, "chunk_bytes": chunk_size, "completed_chunks": sorted(completed),
            "completeness": "pending comparison against official per-gene num_var"}
        persist()
        n_chunks = (expected_size + chunk_size - 1) // chunk_size

        def one_chunk(chunk):
            start = chunk * chunk_size
            stop = min(expected_size, start + chunk_size) - 1
            for attempt in range(retries):
                try:
                    headers = {"Range": f"bytes={start}-{stop}"}
                    if etag:
                        headers["If-Match"] = etag
                    with urllib.request.urlopen(urllib.request.Request(source_url, headers=headers), timeout=timeout) as response:
                        assert response.status == 206
                        assert response.headers["Content-Range"] == f"bytes {start}-{stop}/{expected_size}"
                        content = response.read()
                    assert len(content) == stop - start + 1
                    with partial.open("r+b") as stream:
                        stream.seek(start)
                        assert stream.write(content) == len(content)
                    return chunk
                except Exception:
                    if attempt + 1 == retries:
                        raise
                    time.sleep(attempt + 1)
        started = time.monotonic()
        previous_print = started
        with concurrent.futures.ThreadPoolExecutor(max_workers=workers) as pool:
            pending = [pool.submit(one_chunk, i) for i in range(n_chunks) if i not in completed]
            for future in concurrent.futures.as_completed(pending):
                completed.add(future.result())
                manifest["full_associations"]["completed_chunks"] = sorted(completed)
                persist()
                now = time.monotonic()
                if now - previous_print >= 20 or len(completed) == n_chunks:
                    done = sum(min(chunk_size, expected_size - i * chunk_size) for i in completed)
                    print(f"Cortex associations: {done / 1e9:.3f}/{expected_size / 1e9:.3f} GB", flush=True)
                    previous_print = now
        assert md5_file(partial) == expected_md5
        partial.rename(pair_path)
    index_path = pair_path.with_name(pair_path.name + ".tbi")
    if not index_path.exists():
        with urllib.request.urlopen(source_url + ".tbi", timeout=timeout) as response:
            content = response.read()
        assert hashlib.md5(content).hexdigest() == index_md5
        with index_path.open("xb") as stream:
            stream.write(content)
    assert md5_file(index_path) == index_md5
    with gzip.open(pair_path, "rt", encoding="utf-8") as stream:
        header = stream.readline().rstrip("\n").split("\t")
        example = stream.readline().rstrip("\n").split("\t")
    assert {"variant", "pvalue", "molecular_trait_id", "gene_id", "beta", "se", "chromosome", "position", "ref", "alt"}.issubset(header)
    manifest["full_associations"].update({"status": "downloaded_checksum_verified", "bytes": pair_path.stat().st_size,
        "md5_verified": True, "index_bytes": index_path.stat().st_size, "index_md5_verified": True,
        "columns": header})
    manifest["full_associations"].pop("completed_chunks", None)
    persist()
    print(json.dumps({"file": str(pair_path), "bytes": pair_path.stat().st_size,
                      "md5_verified": True, "header": header, "first_row": example,
                      "completeness": "row-count comparison pending", "colocalization_run": False}, indent=2))


def check_cortex_gwas_coverage():
    """Audit all Cortex genes and variant overlap with GWAS; do not run coloc."""
    import time
    from pyliftover import LiftOver
    # Set this storage path independently before running.
    data_dir = ROOT
    # Set this storage path independently before running. GTEx_Cortex_eGenes.tsv.gz.
    gene_path = ROOT
    # Set this storage path independently before running. GTEx_Cortex_pairs.tsv.gz.
    pair_path = ROOT
    # Set this storage path independently before running. GTEx_Cortex_sources.json.
    manifest_path = ROOT
    # Set this storage path independently before running. Cortex_GWAS_coverage.tsv.
    result_path = ROOT
    # Set this storage path independently before running. hg19_hg38.chain.gz.
    chain_path = ROOT
    chain_url = "https://hgdownload.soe.ucsc.edu/goldenPath/hg19/liftOver/hg19ToHg38.over.chain.gz"
    # Set this storage path independently before running. hl_ma_nov18_v2_summarystats_09122021.txt.gz.
    gwas_path = ROOT
    # Set this storage path independently before running. Trpchevska2022.leads.tsv.
    lead_path = ROOT
    q_threshold = 0.05
    gwas_threshold = 5e-8
    gwas_chunk_size = 500000
    pairs_chunk_size = 750000
    wait_for_download_seconds = 7200
    chromosome_codes = {str(i): i for i in range(1, 23)} | {"X": 23, "23": 23}
    base_codes = {"A": 0, "C": 1, "G": 2, "T": 3}
    genes = pd.read_csv(gene_path, sep="\t")
    genes["stable_id"] = genes.gene_id.str.replace(r"\.\d+$", "", regex=True)
    assert genes.stable_id.is_unique
    gene_codes = dict(zip(genes.stable_id, range(len(genes))))
    is_egene = genes.qval.le(q_threshold).to_numpy()
    assert int(is_egene.sum()) == 9082
    if not chain_path.exists():
        with urllib.request.urlopen(chain_url, timeout=60) as response:
            chain_content = response.read()
        assert gzip.decompress(chain_content).startswith(b"chain ")
        with chain_path.open("xb") as stream:
            stream.write(chain_content)
    lifter = LiftOver(str(chain_path))
    leads = pd.read_csv(lead_path, sep="\t")
    assert len(leads) == 48 and leads.SNP.is_unique
    lead_ids = set(leads.SNP)
    lead_keys, lead_details = {}, {}
    key_parts, p_parts = [], []
    stats = {"raw_gwas_rows": 0, "raw_gwas_rows_by_chr": {}, "valid_biallelic_snvs": 0, "invalid_statistics_or_alleles": 0,
             "unmapped_or_multiple": 0, "reverse_strand_lifted": 0, "uniquely_lifted_rows": 0}
    fields = ["SNP", "CHR", "BP", "Allele1", "Allele2", "Freq1", "Effect", "StdErr", "P.value", "N"]
    dtypes = {"SNP": str, "CHR": str, "BP": np.int64, "Allele1": str, "Allele2": str,
              "Freq1": float, "Effect": float, "StdErr": float, "P.value": float, "N": float}
    # Keep the original GWAS, including its rare lead SNP; do not reuse MAGMA QC.
    for frame in pd.read_csv(gwas_path, sep=r"\s+", usecols=fields, dtype=dtypes, chunksize=gwas_chunk_size):
        stats["raw_gwas_rows"] += len(frame)
        for chromosome, count in frame.CHR.value_counts().items():
            stats["raw_gwas_rows_by_chr"][chromosome] = stats["raw_gwas_rows_by_chr"].get(chromosome, 0) + int(count)
        a1 = frame.Allele1.str.upper().map(base_codes)
        a2 = frame.Allele2.str.upper().map(base_codes)
        valid = (a1.notna() & a2.notna() & a1.ne(a2) & frame.CHR.isin(chromosome_codes)
                 & np.isfinite(frame.Effect) & np.isfinite(frame.StdErr) & frame.StdErr.gt(0)
                 & frame["P.value"].between(0, 1) & frame.Freq1.gt(0) & frame.Freq1.lt(1)
                 & np.isfinite(frame.N) & frame.N.gt(0))
        stats["invalid_statistics_or_alleles"] += int((~valid).sum())
        part = frame.loc[valid]
        left = a1.loc[valid].to_numpy(dtype=np.int8)
        right = a2.loc[valid].to_numpy(dtype=np.int8)
        stats["valid_biallelic_snvs"] += len(part)
        keys = np.zeros(len(part), dtype=np.uint64)
        for j, (rsid, chromosome, position, allele1, allele2) in enumerate(zip(
                part.SNP, part.CHR, part.BP, left, right)):
            source_chr = "chr" + ("X" if chromosome == "23" else chromosome)
            mapped = lifter.convert_coordinate(source_chr, int(position) - 1)
            unique = {(x[0], int(x[1]), x[2]) for x in mapped} if mapped else set()
            if len(unique) != 1:
                stats["unmapped_or_multiple"] += 1
                continue
            target_chr, target_zero, strand = next(iter(unique))
            target_code = chromosome_codes.get(target_chr.removeprefix("chr"))
            if target_code is None:
                stats["unmapped_or_multiple"] += 1
                continue
            if strand == "-":
                allele1, allele2 = 3 - allele1, 3 - allele2
                stats["reverse_strand_lifted"] += 1
            lo, hi = sorted([int(allele1), int(allele2)])
            key = (target_code << 36) + ((target_zero + 1) << 4) + lo * 4 + hi
            keys[j] = key
            if rsid in lead_ids:
                lead_keys[rsid] = key
                lead_details[rsid] = {"snp": rsid, "hg19_chr": chromosome, "hg19_position": int(position),
                                       "hg38_chr": target_chr, "hg38_position": target_zero + 1, "strand": strand}
        keep = keys > 0
        stats["uniquely_lifted_rows"] += int(keep.sum())
        key_parts.append(keys[keep])
        p_parts.append(part["P.value"].to_numpy()[keep])
        print(f"GWAS coordinate audit: {stats['raw_gwas_rows']:,} rows", flush=True)
    all_keys = np.concatenate(key_parts)
    all_p = np.concatenate(p_parts)
    del key_parts, p_parts
    order = np.argsort(all_keys, kind="stable")
    sorted_keys, sorted_p = all_keys[order], all_p[order]
    del all_keys, all_p, order
    unique_keys, first, multiplicity = np.unique(sorted_keys, return_index=True, return_counts=True)
    # Two different hg19 records can map to the same hg38 physical variant.
    # Exclude every such key, rather than choose the more significant record.
    ambiguous = multiplicity > 1
    stats["unique_gwas_variant_keys"] = len(unique_keys)
    stats["duplicated_gwas_key_rows"] = int((multiplicity - 1).sum())
    stats["excluded_ambiguous_keys"] = []
    for key, start, count in zip(unique_keys[ambiguous], first[ambiguous], multiplicity[ambiguous]):
        key = int(key)
        chromosome_code = key >> 36
        pair = key & 15
        stats["excluded_ambiguous_keys"].append({
            "chr38": "chrX" if chromosome_code == 23 else "chr" + str(chromosome_code),
            "position38": (key & ((1 << 36) - 1)) >> 4,
            "alleles": "ACGT"[pair // 4] + "ACGT"[pair % 4],
            "rows": int(count), "GWAS_P_values": sorted_p[start:start + count].tolist()})
    stats["excluded_ambiguous_rows"] = int(multiplicity[ambiguous].sum())
    unique_p = sorted_p[first[~ambiguous]]
    unique_keys = unique_keys[~ambiguous]
    stats["usable_unique_gwas_variant_keys"] = len(unique_keys)
    del sorted_keys, sorted_p, first, multiplicity
    assert stats["raw_gwas_rows"] == 8244937
    assert set(lead_keys) == lead_ids
    stats["all_48_published_leads_uniquely_lifted"] = True

    waited = 0
    while True:
        if pair_path.exists():
            try:
                ready_source = json.loads(manifest_path.read_text(encoding="utf-8"))
                if ready_source["full_associations"]["status"] == "downloaded_checksum_verified":
                    break
            except json.JSONDecodeError:
                pass
        if waited >= wait_for_download_seconds:
            raise RuntimeError("Complete Cortex associations are not available yet")
        time.sleep(10)
        waited += 10
    source = json.loads(manifest_path.read_text(encoding="utf-8"))
    assert source["full_associations"]["status"] == "downloaded_checksum_verified"
    # Confirm that harmonization preserved original GTEx effect statistics.
    official_leads = genes[["stable_id", "variant_id", "slope", "slope_se", "pval_nominal"]].copy()
    official_leads["variant"] = official_leads.variant_id.str.removesuffix("_b38")
    sample = pd.read_csv(pair_path, sep="\t", nrows=100000,
                         usecols=["gene_id", "variant", "beta", "se", "pvalue"])
    sample = sample.drop_duplicates(["gene_id", "variant"])
    matched_leads = sample.merge(official_leads, left_on=["gene_id", "variant"],
                                  right_on=["stable_id", "variant"], validate="many_to_one")
    lead_stat_checks = {left: bool(matched_leads[left].eq(matched_leads[right]).all())
                        for left, right in [("beta", "slope"), ("se", "slope_se"), ("pvalue", "pval_nominal")]}
    assert len(matched_leads) == 22 and all(lead_stat_checks.values())
    lead_stat_audit = {"sample_rows": 100000, "matched_official_leads": len(matched_leads),
                       "exact_checks": lead_stat_checks, "all_66_values_exact": True}
    del sample, matched_leads, official_leads
    n_genes = len(genes)
    counts = np.zeros(n_genes, dtype=np.int64)
    valid_snvs = counts.copy()
    shared = counts.copy()
    significant = counts.copy()
    minimum_p = np.full(n_genes, np.inf)
    lead_shared_genes = {snp: set() for snp in lead_ids}
    lead_key_to_id = {v: k for k, v in lead_keys.items()}
    lead_key_array = np.array(sorted(lead_key_to_id), dtype=np.uint64)
    raw_pairs, deduplicated_pairs = 0, 0
    carry = None
    usecols = ["variant", "pvalue", "gene_id", "beta", "se", "chromosome", "position", "ref", "alt"]
    pair_dtypes = {"variant": str, "pvalue": float, "gene_id": str, "beta": float, "se": float,
                   "chromosome": str, "position": np.int64, "ref": str, "alt": str}

    def consume(frame):
        nonlocal deduplicated_pairs
        frame = frame.drop_duplicates(["gene_id", "variant"])
        deduplicated_pairs += len(frame)
        codes = frame.gene_id.map(gene_codes)
        if codes.isna().any():
            raise ValueError("Imported associations contain genes absent from the official eGene table")
        codes = codes.to_numpy(dtype=np.int32)
        counts[:] += np.bincount(codes, minlength=n_genes)
        b1, b2 = frame.ref.map(base_codes), frame.alt.map(base_codes)
        chr_codes = frame.chromosome.str.removeprefix("chr").map(chromosome_codes)
        valid = (b1.notna() & b2.notna() & b1.ne(b2) & chr_codes.notna()
                 & np.isfinite(frame.beta) & np.isfinite(frame.se) & frame.se.gt(0)
                 & frame.pvalue.between(0, 1)).to_numpy()
        selected_codes = codes[valid]
        valid_snvs[:] += np.bincount(selected_codes, minlength=n_genes)
        c = chr_codes.to_numpy()[valid].astype(np.uint64)
        pos = frame.position.to_numpy()[valid].astype(np.uint64)
        a = b1.to_numpy()[valid].astype(np.uint64)
        b = b2.to_numpy()[valid].astype(np.uint64)
        keys = (c << np.uint64(36)) + (pos << np.uint64(4)) + np.minimum(a, b) * 4 + np.maximum(a, b)
        insertion = np.searchsorted(unique_keys, keys)
        within = insertion < len(unique_keys)
        match = np.zeros(len(keys), dtype=bool)
        match[within] = unique_keys[insertion[within]] == keys[within]
        matched_codes = selected_codes[match]
        shared[:] += np.bincount(matched_codes, minlength=n_genes)
        p = unique_p[insertion[match]]
        np.minimum.at(minimum_p, matched_codes, p)
        significant[:] += np.bincount(matched_codes[p < gwas_threshold], minlength=n_genes)
        lead_match = np.isin(keys, lead_key_array)
        for key, code in zip(keys[lead_match], selected_codes[lead_match]):
            lead_shared_genes[lead_key_to_id[int(key)]].add(int(code))

    for chunk in pd.read_csv(pair_path, sep="\t", usecols=usecols, dtype=pair_dtypes, chunksize=pairs_chunk_size):
        raw_pairs += len(chunk)
        if carry is not None:
            chunk = pd.concat([carry, chunk], ignore_index=True)
        # The indexed archive is position-sorted. Hold the last coordinate to
        # deduplicate alternative rsID annotations across chunk boundaries.
        last = chunk.iloc[-1]
        at_last = chunk.chromosome.eq(last.chromosome) & chunk.position.eq(last.position)
        carry = chunk.loc[at_last].copy()
        consume(chunk.loc[~at_last])
        if raw_pairs % (pairs_chunk_size * 12) == 0:
            print(f"Cortex coverage audit: {raw_pairs:,} association rows", flush=True)
    if carry is not None:
        consume(carry)
    expected = genes.num_var.to_numpy(dtype=np.int64)
    complete = counts == expected
    if not complete.all():
        raise ValueError("Full Cortex associations do not match official per-gene num_var")
    result = genes[["gene_id", "gene_name", "gene_chr", "gene_start", "gene_end", "strand", "qval"]].copy()
    result["is_eGene"] = is_egene
    result["GTEx_expected_variants"] = expected
    result["archive_unique_variants"] = counts
    result["archive_count_matches_GTEx"] = complete
    result["valid_eqtl_SNVS"] = valid_snvs
    result["GWAS_shared_SNVS"] = shared
    result["GWAS_fraction_of_valid_SNVS"] = np.divide(shared, valid_snvs, out=np.zeros(n_genes), where=valid_snvs > 0)
    result["minimum_shared_GWAS_P"] = np.where(np.isfinite(minimum_p), minimum_p, np.nan)
    result["GWAS_significant_SNVS"] = significant
    result["coverage_status"] = np.select([~complete, shared == 0], ["source_count_mismatch", "no_shared_valid_SNV"], default="covered")
    result.sort_values(["is_eGene", "qval"], ascending=[False, True]).to_csv(result_path, sep="\t", index=False)
    summary = {"tested_genes": n_genes, "eGenes_q_le_0_05": int(is_egene.sum()),
        "raw_archive_rows": raw_pairs, "deduplicated_gene_variant_pairs": deduplicated_pairs,
        "official_expected_pairs": int(expected.sum()), "genes_matching_official_num_var": int(complete.sum()),
        "genes_with_count_mismatch": int((~complete).sum()),
        "eGenes_with_shared_GWAS_SNVS": int(((shared > 0) & is_egene).sum()),
        "eGenes_with_shared_GWAS_P_lt_5e8": int(((significant > 0) & is_egene).sum()),
        "eGene_shared_SNVS_median": float(np.median(shared[is_egene])),
        "eGene_shared_fraction_median": float(np.median(result.loc[is_egene, "GWAS_fraction_of_valid_SNVS"])),
        "eGenes_without_GWAS_overlap_by_chr": result.loc[is_egene & (shared == 0), "gene_chr"].value_counts().to_dict(),
        "published_leads_in_Cortex": sum(bool(v) for v in lead_shared_genes.values()),
        "published_leads_in_eGenes": sum(any(is_egene[g] for g in v) for v in lead_shared_genes.values()),
        "GWAS": stats, "colocalization_run": False,
        "effect_scale_status": "GWAS Effect scale still requires author-model confirmation before coloc",
        "output": str(result_path)}
    for snp in sorted(lead_ids):
        lead_details[snp]["tested_genes_with_variant"] = len(lead_shared_genes[snp])
        lead_details[snp]["eGenes_with_variant"] = sum(bool(is_egene[g]) for g in lead_shared_genes[snp])
    source = json.loads(manifest_path.read_text(encoding="utf-8"))
    source["coordinate_chain"] = {"url": chain_url, "path": chain_path.name, "sha256": digest(chain_path)}
    source["GWAS_input"] = {"path": str(gwas_path), "bytes": gwas_path.stat().st_size,
        "sha256": digest(gwas_path), "source_url": "https://zenodo.org/records/5769707",
        "study": "Trpchevska et al. 2022 hearing-loss meta-analysis", "genome_build": "GRCh37/hg19",
        "downloaded_this_stage": False, "source_reused_without_MAGMA_filters": True}
    source["coverage"] = summary
    source["published_lead_coverage"] = list(lead_details.values())
    source["coverage_table"] = {"path": result_path.name, "sha256": digest(result_path)}
    source["full_associations"]["completeness"] = {"deduplicated_pairs": deduplicated_pairs,
        "expected_pairs": int(expected.sum()), "genes_matching_num_var": int(complete.sum()), "tested_genes": n_genes}
    source["full_associations"]["original_statistics_audit"] = lead_stat_audit
    manifest_path.write_text(json.dumps(source, indent=2) + "\n", encoding="utf-8")
    print(json.dumps(summary, indent=2))
    print(result.loc[is_egene & (significant > 0)].nsmallest(15, "minimum_shared_GWAS_P")[[
        "gene_name", "gene_chr", "qval", "GWAS_shared_SNVS", "minimum_shared_GWAS_P"]].to_string(index=False))


#12. =========== Model-independent preparation for Cortex colocalization
def prepare_cortex_coloc(write_pair_inputs=False):
    """Audit complete target-gene inputs without assuming the GWAS effect scale."""
    import struct
    import zlib
    from pyliftover import LiftOver

    # Set this storage path independently before running.
    data_dir = ROOT
    # Set this storage path independently before running. GTEx_Cortex_sources.json.
    manifest_path = ROOT
    # Set this storage path independently before running. Cortex_coloc_QC.tsv.
    result_path = ROOT
    # Set this storage path independently before running. GTEx_Cortex_pairs.tsv.gz.
    pair_path = ROOT
    # Set this storage path independently before running. GTEx_Cortex_eGenes.tsv.gz.
    gene_path = ROOT
    # Set this storage path independently before running. Cortex_GWAS_coverage.tsv.
    coverage_path = ROOT
    # Set this storage path independently before running. hl_ma_nov18_v2_summarystats_09122021.txt.gz.
    gwas_path = ROOT
    q_threshold, gwas_threshold = 0.05, 5e-8
    cis_distance = 1000000  # Original GTEx cis window; padding only locates BGZF blocks.
    gwas_chunk_size = 500000
    # Set this storage path independently before running.
    pair_input_dir = ROOT
    if write_pair_inputs:
        pair_input_dir.mkdir(parents=True, exist_ok=True)
    base_codes = {"A": 0, "C": 1, "G": 2, "T": 3}
    chromosome_codes = {str(i): i for i in range(1, 23)} | {"X": 23, "23": 23}
    header = ["variant", "r2", "pvalue", "molecular_trait_object_id", "molecular_trait_id",
              "maf", "gene_id", "median_tpm", "beta", "se", "an", "ac", "chromosome",
              "position", "ref", "alt", "type", "rsid"]
    source = json.loads(manifest_path.read_text(encoding="utf-8"))
    assert source["full_associations"]["status"] == "downloaded_checksum_verified"
    assert digest(coverage_path) == source["coverage_table"]["sha256"]
    assert digest(gwas_path) == source["GWAS_input"]["sha256"]
    genes = pd.read_csv(gene_path, sep="\t")
    coverage = pd.read_csv(coverage_path, sep="\t")
    candidates = coverage.loc[coverage.is_eGene & coverage.GWAS_significant_SNVS.gt(0)].copy()
    assert candidates.qval.le(q_threshold).all() and candidates.gene_id.is_unique
    assert len(candidates) == 530
    candidates = candidates.merge(genes[["gene_id", "num_var", "variant_id", "slope", "slope_se", "pval_nominal"]],
                                  on="gene_id", validate="one_to_one")
    candidates["stable_id"] = candidates.gene_id.str.replace(r"\.\d+$", "", regex=True)
    assert candidates.stable_id.is_unique

    # Read the existing TBI index. Its generic coordinates are 1-based inclusive.
    tbi = {}
    with gzip.open(str(pair_path) + ".tbi", "rb") as stream:
        assert stream.read(4) == b"TBI\x01"
        nref, fmt, cseq, cbeg, cend, meta, skip, names_length = struct.unpack("<8i", stream.read(32))
        assert (fmt, cseq, cbeg, cend, skip) == (0, 13, 14, 14, 1)
        names = stream.read(names_length).rstrip(b"\0").decode().split("\0")
        assert len(names) == nref
        for name in names:
            bins = {}
            for _ in range(struct.unpack("<i", stream.read(4))[0]):
                bid, count = struct.unpack("<Ii", stream.read(8))
                bins[bid] = [struct.unpack("<QQ", stream.read(16)) for _ in range(count)]
            count = struct.unpack("<i", stream.read(4))[0]
            linear = struct.unpack("<%dQ" % count, stream.read(8 * count))
            tbi[name] = bins, linear

    def query_chunks(chromosome, start, stop):
        assert 0 <= start < stop <= 1 << 29
        bins, linear = tbi[chromosome]
        if not linear or (start >> 14) >= len(linear):
            return []
        minimum = linear[start >> 14]
        end = stop - 1
        ids = [0]
        for shift, offset in [(26, 1), (23, 9), (20, 73), (17, 585), (14, 4681)]:
            ids.extend(range(offset + (start >> shift), offset + (end >> shift) + 1))
        return [(max(a, minimum), b) for bid in ids for a, b in bins.get(bid, ()) if b > minimum]

    def merge_intervals(intervals):
        merged = []
        for a, b in sorted(intervals):
            if not merged or a > merged[-1][1]:
                merged.append([a, b])
            else:
                merged[-1][1] = max(merged[-1][1], b)
        return merged

    intervals = {}
    for gene in candidates.itertuples(index=False):
        chromosome = gene.gene_chr.removeprefix("chr")
        intervals.setdefault(chromosome, []).append((max(0, gene.gene_start - cis_distance - 1),
                                                     gene.gene_end + cis_distance))
    regions = [(chromosome, a, b) for chromosome, locations in intervals.items()
               for a, b in merge_intervals(locations)]
    chunks = merge_intervals([chunk for chromosome, a, b in regions
                              for chunk in query_chunks(chromosome, a, b)])
    buffers = {gene.encode(): io.BytesIO() for gene in candidates.stable_id}
    extracted_rows, compressed_bytes = 0, 0

    def read_block(stream, offset):
        nonlocal compressed_bytes
        stream.seek(offset)
        head = stream.read(12)
        assert len(head) == 12 and head[:4] == b"\x1f\x8b\x08\x04"
        extra_length = struct.unpack_from("<H", head, 10)[0]
        extra = stream.read(extra_length)
        i, size = 0, None
        while i < len(extra):
            sub_length = struct.unpack_from("<H", extra, i + 2)[0]
            if extra[i:i + 2] == b"BC" and sub_length == 2:
                size = struct.unpack_from("<H", extra, i + 4)[0] + 1
            i += 4 + sub_length
        assert size is not None and 26 <= size <= 65536
        raw = head + extra + stream.read(size - 12 - extra_length)
        assert len(raw) == size
        compressed_bytes += size
        return zlib.decompress(raw, 31), offset + size  # Verifies CRC32 and ISIZE.

    with pair_path.open("rb") as stream:
        for first, last in chunks:
            offset, end_offset, carry = first >> 16, last >> 16, b""
            while offset <= end_offset:
                block, next_offset = read_block(stream, offset)
                left = (first & 65535) if offset == (first >> 16) else 0
                right = (last & 65535) if offset == end_offset else len(block)
                lines = (carry + block[left:right]).split(b"\n")
                carry = lines.pop()
                for line in lines:
                    if not line or line.startswith(b"variant\t") or line.startswith(b"#"):
                        continue
                    fields = line.split(b"\t")
                    assert len(fields) == len(header)
                    if fields[6] in buffers:
                        buffers[fields[6]].write(line + b"\n")
                        extracted_rows += 1
                if offset == end_offset:
                    break
                offset = next_offset
            assert not carry, "TBI chunk ends within a record"
    print(f"Cortex extraction: {extracted_rows:,} target rows; {compressed_bytes / 1e6:.1f} MB read", flush=True)

    # Retain original, unscaled GWAS statistics and study-presence masks in memory.
    dtype = np.dtype([("key", "u8"), ("effect_allele", "u1"), ("beta", "f8"), ("se", "f8"),
                      ("p", "f8"), ("freq1", "f8"), ("N", "u4"), ("cohort_mask", "u4")])
    fields = ["SNP", "CHR", "BP", "Allele1", "Allele2", "Freq1", "Effect", "StdErr", "P.value", "N", "Direction"]
    # Set this storage path independently before running. hg19_hg38.chain.gz.
    lifter = LiftOver(str(ROOT))
    parts = []
    raw_gwas_rows, unmapped_rows, reverse_rows = 0, 0, 0
    for frame in pd.read_csv(gwas_path, sep=r"\s+", usecols=fields, dtype={"CHR": str, "Direction": str}, chunksize=gwas_chunk_size):
        raw_gwas_rows += len(frame)
        allele1 = frame.Allele1.str.upper().map(base_codes)
        allele2 = frame.Allele2.str.upper().map(base_codes)
        valid = (allele1.notna() & allele2.notna() & allele1.ne(allele2)
                 & frame.CHR.isin(chromosome_codes) & np.isfinite(frame.Effect)
                 & np.isfinite(frame.StdErr) & frame.StdErr.gt(0) & frame["P.value"].between(0, 1)
                 & frame.Freq1.gt(0) & frame.Freq1.lt(1) & frame.N.gt(0) & np.isfinite(frame.N))
        assert valid.all(), "GWAS validity differs from the completed coverage audit"
        assert frame.N.eq(np.floor(frame.N)).all() and frame.N.le(np.iinfo(np.uint32).max).all()
        assert frame.Direction.str.len().eq(17).all()
        directions = frame.Direction.to_numpy(dtype="S17").view(np.uint8).reshape(-1, 17)
        # METAL also reports an observed zero effect as '0'; '?' denotes absence.
        assert np.isin(directions, [ord("+"), ord("-"), ord("0"), ord("?")]).all()
        masks = ((directions != ord("?")).astype(np.uint32) * (np.uint32(1) << np.arange(17, dtype=np.uint32))).sum(axis=1, dtype=np.uint32)
        records = np.zeros(len(frame), dtype=dtype)
        records["beta"], records["se"] = frame.Effect, frame.StdErr
        records["p"], records["freq1"], records["N"] = frame["P.value"], frame.Freq1, frame.N
        records["cohort_mask"] = masks
        for i, (chromosome, position, a1, a2) in enumerate(zip(frame.CHR, frame.BP, allele1, allele2)):
            chromosome = "chr" + ("X" if chromosome == "23" else chromosome)
            hits = lifter.convert_coordinate(chromosome, int(position) - 1)
            unique = {(x[0], int(x[1]), x[2]) for x in hits} if hits else set()
            if len(unique) != 1:
                unmapped_rows += 1
                continue
            chromosome38, zero38, strand = next(iter(unique))
            code = chromosome_codes.get(chromosome38.removeprefix("chr"))
            if code is None:
                unmapped_rows += 1
                continue
            a1, a2 = int(a1), int(a2)
            if strand == "-":
                a1, a2 = 3 - a1, 3 - a2
                reverse_rows += 1
            records["key"][i] = (code << 36) + ((zero38 + 1) << 4) + min(a1, a2) * 4 + max(a1, a2)
            records["effect_allele"][i] = a1
        parts.append(records[records["key"] > 0])
        if raw_gwas_rows % 2000000 == 0:
            print(f"GWAS input preparation: {raw_gwas_rows:,} rows", flush=True)
    gwas = np.concatenate(parts)
    del parts, records, frame
    order = np.argsort(gwas["key"], kind="stable")
    gwas = gwas[order]
    del order
    repeated = gwas["key"][1:] == gwas["key"][:-1]
    ambiguous = np.r_[False, repeated] | np.r_[repeated, False]
    excluded_ambiguous_rows = int(ambiguous.sum())
    gwas = gwas[~ambiguous]
    assert raw_gwas_rows == 8244937 and unmapped_rows == 4460
    assert excluded_ambiguous_rows == 2 and len(gwas) == 8240475
    mask_sizes = {int(mask): int(mask).bit_count() for mask in np.unique(gwas["cohort_mask"])}
    outcomes = []
    duplicate_fields = [name for name in header if name != "rsid"]
    for i, gene in enumerate(candidates.itertuples(index=False), 1):
        buffer = buffers.pop(gene.stable_id.encode())
        buffer.seek(0)
        frame = pd.read_csv(buffer, sep="\t", names=header, dtype={"chromosome": str, "gene_id": str, "variant": str})
        buffer.close()
        assert frame.gene_id.eq(gene.stable_id).all()
        raw_count = len(frame)
        duplicates = frame.variant.duplicated(keep=False)
        if duplicates.any():
            groups = frame.loc[duplicates].groupby("variant", sort=False)[duplicate_fields].nunique(dropna=False)
            assert not groups.gt(1).any().any(), "Duplicate GTEx associations disagree"
        frame = frame.drop_duplicates("variant")
        assert len(frame) == gene.num_var
        assert frame.chromosome.str.removeprefix("chr").eq(gene.gene_chr.removeprefix("chr")).all()
        identity = ("chr" + frame.chromosome.str.removeprefix("chr") + "_" + frame.position.astype(str)
                    + "_" + frame.ref + "_" + frame.alt)
        assert identity.eq(frame.variant).all(), "Variant identity columns disagree"
        lead_variant = gene.variant_id.removesuffix("_b38")
        lead = frame.loc[frame.variant.eq(lead_variant)]
        assert len(lead) == 1
        assert np.allclose(lead[["beta", "se", "pvalue"]].to_numpy(float)[0],
                           [gene.slope, gene.slope_se, gene.pval_nominal], rtol=1e-12, atol=0)
        ref, alt = frame.ref.map(base_codes), frame.alt.map(base_codes)
        good = (ref.notna() & alt.notna() & ref.ne(alt) & np.isfinite(frame.beta)
                & np.isfinite(frame.se) & frame.se.gt(0) & frame.pvalue.between(0, 1))
        snvs = frame.loc[good]
        code = chromosome_codes[gene.gene_chr.removeprefix("chr")]
        a, b = ref.loc[good].to_numpy(np.uint64), alt.loc[good].to_numpy(np.uint64)
        keys = (np.uint64(code) << np.uint64(36)) + (snvs.position.to_numpy(np.uint64) << np.uint64(4)) + np.minimum(a, b) * 4 + np.maximum(a, b)
        assert len(np.unique(keys)) == len(keys)
        insertion = np.searchsorted(gwas["key"], keys)
        in_bounds = insertion < len(gwas)
        shared = np.zeros(len(keys), dtype=bool)
        shared[in_bounds] = gwas["key"][insertion[in_bounds]] == keys[in_bounds]
        matched = gwas[insertion[shared]]
        assert len(snvs) == gene.valid_eqtl_SNVS and len(matched) == gene.GWAS_shared_SNVS
        assert int((matched["p"] < gwas_threshold).sum()) == gene.GWAS_significant_SNVS
        if write_pair_inputs:
            selected = snvs.loc[shared]
            r_input = pd.DataFrame({"snp": selected.variant.to_numpy(), "position": selected.position.to_numpy(),
                "eqtl_beta": selected.beta.to_numpy(), "eqtl_se": selected.se.to_numpy(),
                "GWAS_P": matched["p"], "GWAS_MAF": np.minimum(matched["freq1"], 1 - matched["freq1"]),
                "GWAS_N": matched["N"]})
            r_input.to_csv(pair_input_dir / (gene.stable_id + ".tsv.gz"), sep="\t", index=False, compression="gzip")
        tss = gene.gene_start if gene.strand == "+" else gene.gene_end
        start38, stop38 = max(1, tss - cis_distance), tss + cis_distance
        assert frame.position.between(start38, stop38).all()
        left = np.searchsorted(gwas["key"], np.uint64((code << 36) + (start38 << 4)))
        right = np.searchsorted(gwas["key"], np.uint64((code << 36) + ((stop38 + 1) << 4)))
        region = gwas[left:right]
        regional_lead = region[np.argmin(region["p"])]
        eqtl_lead_shared = bool(snvs.loc[shared, "variant"].eq(lead_variant).any())
        gwas_lead_shared = bool(np.isin(regional_lead["key"], keys))
        regional_key = int(regional_lead["key"])
        regional_pair = regional_key & 15
        regional_variant = f"{gene.gene_chr}:{(regional_key & ((1 << 36) - 1)) >> 4}:{'ACGT'[regional_pair // 4]}/{'ACGT'[regional_pair % 4]}"
        cohort_counts = np.array([mask_sizes[int(mask)] for mask in matched["cohort_mask"]])
        notes = []
        if not eqtl_lead_shared:
            notes.append("eQTL_lead_not_shared")
        if not gwas_lead_shared:
            notes.append("GWAS_regional_lead_not_shared")
        outcomes.append({"gene_id": gene.gene_id, "gene_name": gene.gene_name, "gene_chr": gene.gene_chr,
            "qval": gene.qval, "cis_start38": start38, "cis_end38": stop38,
            "GTEx_expected_variants": int(gene.num_var), "archive_raw_rows": raw_count,
            "archive_unique_variants": len(frame), "duplicate_statistics_agree": True,
            "official_lead_statistics_agree": True, "valid_eqtl_SNVS": len(snvs), "GWAS_shared_SNVS": len(matched),
            "GWAS_shared_fraction": len(matched) / len(snvs), "eQTL_lead_variant": lead_variant,
            "eQTL_lead_shared": eqtl_lead_shared, "GWAS_regional_lead_allele_key": regional_variant,
            "GWAS_regional_lead_P": float(regional_lead["p"]), "GWAS_regional_lead_shared": gwas_lead_shared,
            "GWAS_shared_N_min": int(matched["N"].min()), "GWAS_shared_N_median": float(np.median(matched["N"])),
            "GWAS_shared_N_max": int(matched["N"].max()), "GWAS_shared_N_unique": len(np.unique(matched["N"])),
            "GWAS_shared_cohort_masks": len(np.unique(matched["cohort_mask"])),
            "GWAS_shared_cohorts_min": int(cohort_counts.min()), "GWAS_shared_cohorts_max": int(cohort_counts.max()),
            "coverage_notes": ";".join(notes) or "both_leads_shared",
            "model_status": "GWAS_effect_scale_unconfirmed", "colocalization_run": False})
        if i % 100 == 0 or i == len(candidates):
            print(f"Cortex gene input audit: {i}/{len(candidates)}", flush=True)
    result = pd.DataFrame(outcomes)
    result.sort_values(["gene_chr", "cis_start38", "gene_id"]).to_csv(result_path, sep="\t", index=False)
    summary = {"target_genes": len(result), "expected_unique_pairs": int(candidates.num_var.sum()),
        "observed_unique_pairs": int(result.archive_unique_variants.sum()), "extracted_raw_rows": extracted_rows,
        "duplicate_rows": int(result.archive_raw_rows.sum() - result.archive_unique_variants.sum()),
        "all_duplicate_statistics_agree": True, "all_official_lead_statistics_agree": True,
        "both_leads_shared": int((result.eQTL_lead_shared & result.GWAS_regional_lead_shared).sum()),
        "eQTL_lead_not_shared": int((~result.eQTL_lead_shared).sum()),
        "GWAS_regional_lead_not_shared": int((~result.GWAS_regional_lead_shared).sum()),
        "genes_with_multiple_cohort_masks": int(result.GWAS_shared_cohort_masks.gt(1).sum()),
        "shared_N_min": int(result.GWAS_shared_N_min.min()), "shared_N_max": int(result.GWAS_shared_N_max.max()),
        "merged_read_regions": len(regions), "compressed_bytes_read": compressed_bytes,
        "unmapped_GWAS_rows": unmapped_rows, "reverse_strand_GWAS_rows": reverse_rows,
        "excluded_ambiguous_GWAS_rows": excluded_ambiguous_rows, "colocalization_run": False,
        "model_status": "Binary phenotype confirmed; Effect units and variant-specific case/control composition unconfirmed",
        "input_policy": "No fabricated logOR, fixed N, or per-SNP case fraction; original statistics used only for identity and coverage QC",
        "table": result_path.name, "sha256": digest(result_path)}
    source["coloc_preparation"] = summary
    manifest_path.write_text(json.dumps(source, indent=2) + "\n", encoding="utf-8")
    print(json.dumps(summary, indent=2))


#13. =========== Official coloc runtime and original GWAS approximation
def cortex_coloc_runtime():
    """Use pinned official Windows binaries in task scratch, not the system library."""
    import zipfile
    packages = {
        "reshape": ("0.8.10", "d7e6bcd39a838c08210fbaf552619724784ce8af25345efdf69c4d41d9772437"),
        "mixsqp": ("0.3-54", "56914a49e0473c6550b47aef6d9c7bd44feb92aaa2b41d800244ec43b0a9eb7a"),
        "susieR": ("0.14.2", "7dfa02420b7328a79305c38206784bc5da8f1c252d75e30c2d2cfd12306c71a2"),
        "coloc": ("5.2.3", "b5dbf24d27177886110102d755698deb90657d178d00163fa0079a0b19b02c5d")}
    # Set this storage path independently before running. Rscript.exe.
    rscript = ROOT
    # Set this storage path independently before running.
    scratch = ROOT
    # Set this storage path independently before running.
    library = ROOT
    library.mkdir(parents=True, exist_ok=True)
    provenance = {}
    for name, (version, expected_hash) in packages.items():
        url = f"https://cloud.r-project.org/bin/windows/contrib/4.5/{name}_{version}.zip"
        # Set this storage path independently before running.
        description = ROOT
        if not description.exists():
            with urllib.request.urlopen(url, timeout=120) as stream:
                content = stream.read()
            assert hashlib.sha256(content).hexdigest() == expected_hash
            with zipfile.ZipFile(io.BytesIO(content)) as archive:
                for entry in archive.infolist():
                    assert (library / entry.filename).resolve().is_relative_to(library.resolve())
                archive.extractall(library)
        assert f"Version: {version}\n" in description.read_text(encoding="utf-8")
        provenance[name] = {"version": version, "url": url, "sha256": expected_hash}
    assert rscript.is_file()
    return rscript, library, provenance


def execute_cortex_coloc(input_dir, qc_path, output_path, mode, case_fraction=None):
    """Run the official R package; keep per-SNP N for the labelled meta approximation."""
    import subprocess
    p1, p2, p12, p12_conservative = 1e-4, 1e-4, 1e-5, 1e-6
    eqtl_sdY, h4_threshold = 1.0, 0.8
    rscript, library, packages = cortex_coloc_runtime()
    # Set this storage path independently before running.
    scratch = ROOT
    # Set this storage path independently before running. coloc-run.R.
    runner = ROOT
    code = "\nargs <- commandArgs(trailingOnly=TRUE)\n.libPaths(c(args[1],.libPaths()))\nsuppressPackageStartupMessages(library(coloc))\nstopifnot(as.character(packageVersion('coloc'))=='5.2.3')\nqc <- read.delim(args[2],check.names=FALSE)\ninput_dir <- args[3]; output <- args[4]; mode <- args[5]\ns <- if(args[6]=='NA') NA_real_ else as.numeric(args[6]); p1 <- as.numeric(args[7]); p2 <- as.numeric(args[8])\np12 <- as.numeric(args[9]); p12_low <- as.numeric(args[10]); sdY <- as.numeric(args[11]); cutoff <- as.numeric(args[12])\n# Verify arithmetic with variable N against official package functions before real data.\nt1 <- list(pvalues=c(1e-8,.1,.01),MAF=c(.2,.3,.4),N=c(700000,650000,600000),s=.2047361613,type='cc',snp=c('a','b','c'))\nt2 <- list(beta=c(.3,.1,.01),varbeta=c(.01,.02,.01),type='quant',sdY=1,snp=c('a','b','c'))\ninvisible(capture.output(verify <- suppressWarnings(coloc.abf(t1,t2))))\ndirect <- coloc:::combine.abf(coloc:::approx.bf.p(t1$pvalues,t1$MAF,'cc',t1$N,t1$s)$lABF,\n    coloc:::approx.bf.estimates(t2$beta/sqrt(t2$varbeta),t2$varbeta,'quant',sdY=1)$lABF,1e-4,1e-4,1e-5,quiet=TRUE)\nstopifnot(max(abs(verify$summary[names(direct)]-direct))<1e-12)\nstopifnot(isTRUE(tools::checkMD5sums(package='coloc')))\nrows <- vector('list',nrow(qc))\nfor(i in seq_len(nrow(qc))) {\n    gid <- sub('\\\\.[0-9]+$','',qc$gene_id[i])\n    d <- read.delim(gzfile(file.path(input_dir,paste0(gid,'.tsv.gz'))),check.names=FALSE)\n    stopifnot(!anyDuplicated(d$snp),all(d$eqtl_se>0),nrow(d)>1)\n    if(mode=='p_approx') {\n        # Official 5.2.3 accepts N vectors and computes SNP-specific approximate variance.\n        # s is the published overall fraction, an explicit assumption for this meta-analysis.\n        gwas <- list(pvalues=d$GWAS_P,MAF=d$GWAS_MAF,N=d$GWAS_N,s=s,type='cc',snp=d$snp,position=d$position)\n    } else {\n        stopifnot(mode=='beta',all(d$GWAS_se>0))\n        gwas <- list(beta=d$GWAS_beta,varbeta=d$GWAS_se^2,type='cc',snp=d$snp,position=d$position)\n    }\n    eqtl <- list(beta=d$eqtl_beta,varbeta=d$eqtl_se^2,type='quant',sdY=sdY,snp=d$snp,position=d$position)\n    warnings <- character()\n    run <- function(prior) {\n        answer <- NULL\n        invisible(capture.output(answer <- withCallingHandlers(coloc.abf(gwas,eqtl,p1=p1,p2=p2,p12=prior),\n            warning=function(w) {warnings <<- unique(c(warnings,conditionMessage(w))); invokeRestart('muffleWarning')})))\n        answer\n    }\n    primary <- run(p12); conservative <- run(p12_low)\n    probabilities <- primary$summary[paste0('PP.H',0:4,'.abf')]\n    stopifnot(all(is.finite(probabilities)),abs(sum(probabilities)-1)<1e-10)\n    posterior <- primary$results[order(primary$results$SNP.PP.H4,decreasing=TRUE),]\n    size95 <- which(cumsum(posterior$SNP.PP.H4)>=.95)[1]\n    vars95 <- if(probabilities[5]>=cutoff) paste(posterior$snp[seq_len(size95)],collapse=';') else ''\n    rows[[i]] <- data.frame(gene_id=qc$gene_id[i],analysed_SNVS=unname(primary$summary['nsnps']),\n        H0=unname(probabilities[1]),H1=unname(probabilities[2]),H2=unname(probabilities[3]),\n        H3=unname(probabilities[4]),H4=unname(probabilities[5]),H4_p12_1e6=unname(conservative$summary['PP.H4.abf']),\n        effective_p1=unname(primary$priors['p1']),effective_p2=unname(primary$priors['p2']),\n        H4_top_SNP=posterior$snp[1],H4_top_SNP_PP_conditional=posterior$SNP.PP.H4[1],\n        H4_95set_size=size95,H4_95set_variants=vars95,\n        warnings=gsub('[\\r\\n]+',' ',paste(warnings,collapse=' | ')),check.names=FALSE)\n    if(i%%100==0 || i==nrow(qc))cat('Official coloc:',i,'/',nrow(qc),'\\n')\n}\nwrite.table(do.call(rbind,rows),output,sep='\\t',row.names=FALSE,quote=FALSE,na='NA')\ncat('Official arithmetic and package integrity checks passed\\n')\n"
    runner.write_text(code, encoding="utf-8")
    environment = os.environ.copy()
    environment.update(TMP=str(scratch), TEMP=str(scratch), LC_ALL="C")
    command = [str(rscript), "--vanilla", str(runner), str(library), str(qc_path), str(input_dir), str(output_path),
               mode, str(case_fraction) if case_fraction is not None else "NA", str(p1), str(p2), str(p12),
               str(p12_conservative), str(eqtl_sdY), str(h4_threshold)]
    completed = subprocess.run(command, cwd=ROOT, env=environment, text=True, encoding="utf-8", errors="replace",
                               stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    if completed.stdout:
        print(completed.stdout.rstrip(), flush=True)
    if completed.stderr:
        print(completed.stderr.rstrip(), file=sys.stderr, flush=True)
    if completed.returncode:
        raise RuntimeError("Official coloc R run failed")
    return {"R": "4.5.1", "packages": packages, "p1": p1, "p2": p2, "p12": p12,
            "p12_conservative": p12_conservative, "eqtl_sdY": eqtl_sdY, "H4_threshold": h4_threshold,
            "mode": mode, "case_fraction": case_fraction, "official_function_arithmetic_verified": True,
            "package_MD5_verified": True,
            "prior_adjustment": "Official p1/p2 adjustment when n*p>=1; effective values retained per gene"}


def analyze_original_cortex_coloc():
    """Exploratory P-value colocalization for the original heterogeneous GWAS."""
    # Set this storage path independently before running.
    data_dir = ROOT
    # Set this storage path independently before running. GTEx_Cortex_sources.json.
    manifest_path = ROOT
    # Set this storage path independently before running. Cortex_coloc_QC.tsv.
    qc_path = ROOT
    # Set this storage path independently before running. GWAS_coloc_results.tsv.
    result_path = ROOT
    # Set this storage path independently before running. GWAS_coloc_genes.tsv.
    genes_path = ROOT
    # Set this storage path independently before running. GWAS-native.tsv.
    native_path = ROOT
    case_fraction = 148152 / (148152 + 575472)  # Dataset-specific Zenodo counts.
    h4_threshold = 0.8
    mhc_chromosome, mhc_start, mhc_end = "chr6", 25000000, 34000000
    prepare_cortex_coloc(write_pair_inputs=True)
    # Set this storage path independently before running.
    runtime = execute_cortex_coloc(ROOT, qc_path, native_path, "p_approx", case_fraction)
    qc = pd.read_csv(qc_path, sep="\t")
    native = pd.read_csv(native_path, sep="\t")
    result = qc.merge(native, on="gene_id", validate="one_to_one")
    assert len(result) == 530 and result.analysed_SNVS.eq(result.GWAS_shared_SNVS).all()
    assert np.allclose(result[["H0", "H1", "H2", "H3", "H4"]].sum(axis=1), 1, atol=1e-10)
    source = json.loads(manifest_path.read_text(encoding="utf-8"))
    leads = pd.DataFrame(source["published_lead_coverage"])
    result["published_GWAS_leads_in_cis"] = [int((leads.hg38_chr.eq(gene.gene_chr)
        & leads.hg38_position.between(gene.cis_start38, gene.cis_end38)).sum()) for gene in result.itertuples(index=False)]
    result["broad_MHC_caution"] = (result.gene_chr.eq(mhc_chromosome) & result.cis_start38.le(mhc_end)
                                    & result.cis_end38.ge(mhc_start))
    result["known_multiple_GWAS_leads"] = result.published_GWAS_leads_in_cis.gt(1)
    result["H4_prior_robust"] = result.H4.ge(h4_threshold) & result.H4_p12_1e6.ge(h4_threshold)
    result["coverage_pass"] = result.eQTL_lead_shared & result.GWAS_regional_lead_shared
    result["main_candidate"] = (result.H4_prior_robust & result.coverage_pass
                                 & ~result.broad_MHC_caution & ~result.known_multiple_GWAS_leads)
    result["evidence_class"] = np.select([result.H4_prior_robust, result.H4.ge(h4_threshold), result.H3.ge(h4_threshold)],
        ["shared_support_prior_robust", "shared_support_prior_sensitive", "different_signal_support"], default="insufficient_support")
    result["GWAS_input_method"] = "P_value_MAF_per_SNP_N_overall_case_fraction_approximation"
    result["exploratory_only"] = True
    result["model_status"] = "Original_Effect_scale_unconfirmed;meta_case_fraction_assumed_constant"
    result["colocalization_run"] = True
    result = result.sort_values(["main_candidate", "H4"], ascending=[False, False])
    result.to_csv(result_path, sep="\t", index=False)
    selected = result.loc[result.main_candidate].copy()
    selected.to_csv(genes_path, sep="\t", index=False)
    summary = {"analysis": "Exploratory original-GWAS P-value approximation", "genes_tested": len(result),
        "H4_ge_0_8": int(result.H4.ge(h4_threshold).sum()), "H4_prior_robust": int(result.H4_prior_robust.sum()),
        "main_candidates_exploratory": len(selected), "class_counts": result.evidence_class.value_counts().to_dict(),
        "runtime": runtime, "genome_build": "GRCh38 after unique strand-aware liftOver",
        "approximation_limits": "Unconfirmed raw Effect units; overall case fraction cannot establish per-SNP case composition or imputation information",
        "input_definitions": {"MAF": "min(Freq1,1-Freq1), using the reported meta-analysis mean allele frequency",
            "N": "Original per-SNP total N, not effective N", "s": "Dataset-specific published overall case fraction; assumed constant",
            "GWAS_Effect": "Not used because its scale is unconfirmed"},
        "input_source": "https://zenodo.org/records/5769707", "case_counts": [148152, 575472],
        "scope": "All 530 prespecified overlap eGenes; complete shared SNVs; no significance/MAF pruning",
        "main_QC": "Both lead variants shared; H4>=0.8 at p12=1e-5 and 1e-6; no broad MHC caution or multiple published GWAS leads",
        "MHC_caution_window": [mhc_chromosome, mhc_start, mhc_end], "single_causal_variant_assumption": True,
        "colocalization_run": True, "result_file": result_path.name, "result_sha256": digest(result_path),
        "candidate_file": genes_path.name, "candidate_sha256": digest(genes_path)}
    source["original_GWAS_coloc"] = summary
    source["scope"] = "Cortex v8 input and coverage audits plus separately labelled original-GWAS exploratory colocalization; no restriction to 19 genes"
    summary["comparison_limit"] = "Original GWAS includes earlier FinnGen; comparison with R13 is not independent replication"
    manifest_path.write_text(json.dumps(source, indent=2) + "\n", encoding="utf-8")
    print(json.dumps(summary, indent=2))
    print(selected[["gene_name", "H3", "H4", "H4_p12_1e6", "analysed_SNVS"]].to_string(index=False))


#14. =========== FinnGen R13 discovery independent of the original GWAS shortlist
def finngen_bgzf_tools(include_chunks=False):
    """Read generic TBI/BGZF files without installing a platform-specific extension."""
    import struct
    import zlib

    def merge_intervals(intervals):
        merged = []
        for start, stop in sorted(intervals):
            if not merged or start > merged[-1][1]:
                merged.append([start, stop])
            else:
                merged[-1][1] = max(merged[-1][1], stop)
        return merged

    def read_index(path, sequence_column, position_column, header=None, index_path=None):
        if header is None:
            with gzip.open(path, "rt") as stream:
                header = stream.readline().rstrip("\r\n").split("\t")
        with gzip.open(index_path or (str(path) + ".tbi"), "rb") as stream:
            assert stream.read(4) == b"TBI\x01"
            nref, fmt, cseq, cbeg, cend, meta, skip, length = struct.unpack("<8i", stream.read(32))
            assert fmt == 0, "A one-based generic TBI index is required"
            assert header[cseq - 1] == sequence_column
            assert header[cbeg - 1] == position_column and cend == cbeg
            names = stream.read(length).rstrip(b"\0").decode().split("\0")
            assert len(names) == nref
            references = {}
            for name in names:
                bins = {}
                for _ in range(struct.unpack("<i", stream.read(4))[0]):
                    bid, count = struct.unpack("<Ii", stream.read(8))
                    bins[bid] = [struct.unpack("<QQ", stream.read(16)) for _ in range(count)]
                count = struct.unpack("<i", stream.read(4))[0]
                linear = struct.unpack("<%dQ" % count, stream.read(8 * count))
                references[name] = bins, linear
        return {"header": header, "references": references, "format": fmt,
                "sequence_column": cseq, "position_column": cbeg, "meta": meta, "skip": skip}

    def resolve_contig(chromosome, index):
        token = str(chromosome).removeprefix("chr")
        token = "X" if token == "23" else token
        options = [token, "chr" + token]
        if token == "X":
            options += ["23", "chr23"]
        found = [name for name in options if name in index["references"]]
        assert len(found) == 1, ("Missing or ambiguous index contig", chromosome, found)
        return found[0]

    def chunks_for(index, chromosome, start, stop):
        assert 0 <= start < stop <= 1 << 29
        bins, linear = index["references"][resolve_contig(chromosome, index)]
        if not linear or (start >> 14) >= len(linear):
            return []
        minimum, end = linear[start >> 14], stop - 1
        ids = [0]
        for shift, offset in [(26, 1), (23, 9), (20, 73), (17, 585), (14, 4681)]:
            ids.extend(range(offset + (start >> shift), offset + (end >> shift) + 1))
        return [(max(a, minimum), b) for bid in ids for a, b in bins.get(bid, ()) if b > minimum]

    def iter_rows(path, index, regions, statistics):
        chunks = merge_intervals([chunk for chromosome, start, stop in regions
                                  for chunk in chunks_for(index, chromosome, start, stop)])
        statistics.update(chunks=len(chunks), compressed_bytes=0, blocks=0)
        with path.open("rb") as stream:
            for first, last in chunks:
                offset, end_offset, carry = first >> 16, last >> 16, b""
                while offset <= end_offset:
                    stream.seek(offset)
                    head = stream.read(12)
                    assert len(head) == 12 and head[:4] == b"\x1f\x8b\x08\x04"
                    extra_length = struct.unpack_from("<H", head, 10)[0]
                    extra = stream.read(extra_length)
                    i, size = 0, None
                    while i < len(extra):
                        sub_length = struct.unpack_from("<H", extra, i + 2)[0]
                        if extra[i:i + 2] == b"BC" and sub_length == 2:
                            size = struct.unpack_from("<H", extra, i + 4)[0] + 1
                        i += 4 + sub_length
                    assert size is not None and 26 <= size <= 65536
                    raw = head + extra + stream.read(size - 12 - extra_length)
                    assert len(raw) == size
                    block = zlib.decompress(raw, 31)
                    statistics["compressed_bytes"] += size
                    statistics["blocks"] += 1
                    left = (first & 65535) if offset == (first >> 16) else 0
                    right = (last & 65535) if offset == end_offset else len(block)
                    lines = (carry + block[left:right]).split(b"\n")
                    carry = lines.pop()
                    for line in lines:
                        if not line or (index["meta"] and line[0] == index["meta"]):
                            continue
                        fields = line.split(b"\t")
                        if fields[0] == index["header"][0].encode():
                            continue
                        assert len(fields) == len(index["header"])
                        yield fields, line
                    if offset == end_offset:
                        break
                    offset += size
                assert not carry, "TBI chunk boundary splits a record"

    if include_chunks:
        return read_index, iter_rows, merge_intervals, chunks_for
    return read_index, iter_rows, merge_intervals


def download_finngen_cortex():
    """Download or verify only the full FinnGen endpoint and its original index."""
    import base64
    from datetime import datetime, timezone

    # Set this storage path independently before running.
    data_dir = ROOT
    # Set this storage path independently before running.
    scratch = ROOT
    endpoint = "H8_HL_SEN_NAS"
    base_url = "https://storage.googleapis.com/finngen-public-data-r13/summary_stats/finngen_R13_" + endpoint + ".gz"
    resources = [("FinnGen_Sensorineural_GWAS.gz", base_url, 809282989, "4d8e5660d2bb28f1c43c4399e8b2d6a1"),
                 ("FinnGen_Sensorineural_GWAS.gz.tbi", base_url + ".tbi", 1770687, "9f02bf822d617986e7f25a212480f7a7")]
    # Set this storage path independently before running. FinnGen_Cortex_sources.json.
    manifest_path = ROOT
    data_dir.mkdir(parents=True, exist_ok=True)
    scratch.mkdir(parents=True, exist_ok=True)
    source = json.loads(manifest_path.read_text(encoding="utf-8")) if manifest_path.exists() else {}
    provenance = {}
    for filename, url, expected_size, expected_md5 in resources:
        path = data_dir / filename
        headers, provider_md5 = {}, expected_md5
        if not path.exists():
            request = urllib.request.Request(url, method="HEAD")
            with urllib.request.urlopen(request, timeout=90) as response:
                headers = dict(response.headers.items())
                assert int(response.headers["Content-Length"]) == expected_size
                hashes = response.headers.get("x-goog-hash", "")
                provider_md5 = next((base64.b64decode(item.strip()[4:]).hex()
                                     for item in hashes.split(",") if item.strip().startswith("md5=")), None)
            if expected_md5 is not None:
                assert provider_md5 == expected_md5
            assert provider_md5 is not None
            partial = scratch / (filename + ".part")
            assert not partial.exists(), f"Inspect unfinished download before retrying: {partial}"
            downloaded = 0
            with urllib.request.urlopen(url, timeout=180) as response, partial.open("xb") as stream:
                assert int(response.headers["Content-Length"]) == expected_size
                for block in iter(lambda: response.read(8 * 1024 * 1024), b""):
                    stream.write(block)
                    downloaded += len(block)
            assert downloaded == expected_size
            verify_path = partial
        else:
            verify_path = path
        assert provider_md5 is not None, "Pinned provider MD5 is required to reuse an existing input offline"
        assert verify_path.stat().st_size == expected_size
        md5, sha256 = hashlib.md5(), hashlib.sha256()
        with verify_path.open("rb") as stream:
            for block in iter(lambda: stream.read(8 * 1024 * 1024), b""):
                md5.update(block)
                sha256.update(block)
        assert md5.hexdigest() == provider_md5, "FinnGen checksum mismatch"
        if verify_path != path:
            verify_path.replace(path)
        provenance[filename] = {"url": url, "path": filename, "bytes": expected_size,
            "provider_md5": provider_md5, "md5_verified": True, "sha256": sha256.hexdigest(),
            "HEAD": headers}
        print(f"FinnGen input verified: {filename}", flush=True)
    read_index, _, _ = finngen_bgzf_tools()
    index = read_index(data_dir / resources[0][0], "#chrom", "pos")
    required = {"#chrom", "pos", "ref", "alt", "pval", "beta", "sebeta", "af_alt"}
    assert required.issubset(index["header"])
    source.update(schema="FinnGen_Cortex_inputs_v1", release="R13", endpoint=endpoint,
        phenotype="Sensorineural hearing loss", genome_build="GRCh38",
        summary_statistics=provenance, header=index["header"],
        index_schema={key: value for key, value in index.items() if key not in ["header", "references"]},
        index_contigs=list(index["references"]),
        effect_scale="log_odds_ratio_for_ALT", effect_source="https://finngen.gitbook.io/documentation/data-description",
        endpoint_metadata_source="https://r13.finngen.fi/pheno/H8_HL_SEN_NAS",
        cases=48388, controls=432132, total_participants=480520, release_date="2026-06-02",
        manifest_url="https://storage.googleapis.com/finngen-public-data-r13/summary_stats/finngen_R13_manifest.tsv",
        manifest_endpoint_row={"phenocode": endpoint, "phenotype": "Sensorineural hearing loss",
            "category": "VIII Diseases of the ear and mastoid process (H8_)",
            "num_cases": 48388, "num_controls": 432132,
            "path_bucket": "gs://finngen-public-data-r13/summary_stats/finngen_R13_H8_HL_SEN_NAS.gz",
            "path_https": base_url},
        metadata_policy="Beta/SE coloc does not use fixed N, case fraction or allele frequency",
        retrieved_utc=datetime.now(timezone.utc).isoformat())
    manifest_path.write_text(json.dumps(source, indent=2) + "\n", encoding="utf-8")


def analyze_finngen_cortex(tissue="Cortex"):
    """Colocate complete tissue-specific eGene inputs with exact FinnGen REF/ALT overlaps."""
    from bisect import bisect_right

    tissue_names = {"Cortex": "Brain_Cortex", "Spinal": "Brain_Spinal_cord_cervical_c-1", "Tibial": "Nerve_Tibial"}
    assert tissue in tissue_names
    # Set this storage path independently before running.
    data_dir = ROOT
    manifest_path = data_dir / ("FinnGen_Cortex_sources.json" if tissue == "Cortex" else f"GTEx_{tissue}_sources.json")
    # Set this storage path independently before running. FinnGen_Sensorineural_GWAS.gz.
    finngen_path = ROOT
    eqtl_path = data_dir / f"GTEx_{tissue}_pairs.tsv.gz"
    egene_path = data_dir / f"GTEx_{tissue}_eGenes.tsv.gz"
    screen_path = data_dir / ("FinnGen_eGene_screen.tsv" if tissue == "Cortex" else f"FinnGen_{tissue}_screen.tsv")
    result_path = data_dir / ("FinnGen_coloc_results.tsv" if tissue == "Cortex" else f"FinnGen_{tissue}_coloc.tsv")
    # Set this storage path independently before running. FinnGen_coloc_genes.tsv.
    selected_path = ROOT if tissue == "Cortex" else None
    # Set this storage path independently before running.
    scratch = ROOT
    prefix = "FinnGen" if tissue == "Cortex" else f"FinnGen-{tissue}"
    input_dir = scratch / f"{prefix}-inputs"
    qc_path, native_path = scratch / f"{prefix}-QC.tsv", scratch / f"{prefix}-native.tsv"
    q_threshold, gwas_threshold, cis_distance = 0.05, 5e-8, 1000000
    scan_chunk_size, h4_threshold = 750000, 0.8
    mhc_chromosome, mhc_start, mhc_end = "chr6", 25000000, 34000000
    source = json.loads(manifest_path.read_text(encoding="utf-8"))
    eqtl_source = json.loads((data_dir / f"GTEx_{tissue}_sources.json").read_text(encoding="utf-8"))
    # Set this storage path independently before running. FinnGen_Cortex_sources.json.
    gwas_source = source if tissue == "Cortex" else json.loads((ROOT).read_text(encoding="utf-8"))
    assert gwas_source["release"] == "R13" and gwas_source["endpoint"] == "H8_HL_SEN_NAS"
    assert eqtl_source["dataset"] == "gtex_v8" and eqtl_source["tissue"] == tissue_names[tissue]
    for filename, record in gwas_source["summary_statistics"].items():
        path = data_dir / filename
        assert record["md5_verified"] and path.stat().st_size == record["bytes"]
        assert digest(path) == record["sha256"]
    assert digest(egene_path) == eqtl_source["egenes"]["sha256"]
    assert eqtl_path.stat().st_size == eqtl_source["full_associations"]["bytes"]
    if tissue == "Cortex":
        assert eqtl_source["full_associations"]["md5_verified"]
    else:
        assert eqtl_source["full_associations"]["subset_complete_for_selected_genes"] is True
        assert digest(eqtl_path) == eqtl_source["full_associations"]["sha256"]
    input_dir.mkdir(parents=True, exist_ok=True)
    read_index, iter_rows, merge_intervals = finngen_bgzf_tools()
    finngen_index = read_index(finngen_path, "#chrom", "pos")
    if tissue == "Cortex":
        eqtl_index = read_index(eqtl_path, "chromosome", "position")
        eqtl_header = eqtl_index["header"]
        genes = pd.read_csv(egene_path, sep="\t")
        genes = genes.loc[genes.qval.le(q_threshold)].copy()
        genes["stable_id"] = genes.gene_id.str.replace(r"\.\d+$", "", regex=True)
        assert len(genes) == 9082 and genes.stable_id.is_unique
        genes["tss38"] = np.where(genes.strand.eq("+"), genes.gene_start, genes.gene_end)
        genes["cis_start38"] = np.maximum(1, genes.tss38 - cis_distance)
        genes["cis_end38"] = genes.tss38 + cis_distance

        # This first pass stores only significant positions, not a truncated GWAS input.
        significant, raw_gwas_rows, invalid_p_rows = {}, 0, 0
        for frame in pd.read_csv(finngen_path, sep="\t", usecols=["#chrom", "pos", "pval"],
                                 dtype={"#chrom": str, "pos": np.int64}, chunksize=scan_chunk_size):
            raw_gwas_rows += len(frame)
            valid = np.isfinite(frame.pval) & frame.pval.between(0, 1)
            invalid_p_rows += int((~valid).sum())
            for chromosome, part in frame.loc[valid & frame.pval.lt(gwas_threshold)].groupby("#chrom", sort=False):
                chromosome = str(chromosome).removeprefix("chr")
                chromosome = "X" if chromosome == "23" else chromosome
                significant.setdefault(chromosome, []).append(part.pos.to_numpy(np.int64))
            if raw_gwas_rows % 3000000 == 0:
                print(f"FinnGen genome-wide screen: {raw_gwas_rows:,} variants", flush=True)
        significant = {chromosome: np.unique(np.concatenate(parts)) for chromosome, parts in significant.items()}
        screen_counts = []
        for gene in genes.itertuples(index=False):
            positions = significant.get(gene.gene_chr.removeprefix("chr"), np.empty(0, dtype=np.int64))
            screen_counts.append(int(np.searchsorted(positions, gene.cis_end38, side="right")
                                     - np.searchsorted(positions, gene.cis_start38)))
        genes["FinnGen_significant_positions_in_cis"] = screen_counts
        genes["selected_for_coloc"] = genes.FinnGen_significant_positions_in_cis.gt(0)
        genes[["gene_id", "gene_name", "gene_chr", "qval", "cis_start38", "cis_end38",
               "FinnGen_significant_positions_in_cis", "selected_for_coloc"]].to_csv(screen_path, sep="\t", index=False)
    else:
        tested = pd.read_csv(egene_path, sep="\t")
        screen = pd.read_csv(screen_path, sep="\t")
        required = set(tested.columns) | {"stable_id", "tss38", "cis_start38", "cis_end38",
            "FinnGen_significant_positions_in_cis", "selected_for_coloc"}
        assert required.issubset(screen.columns) and screen.gene_id.is_unique and tested.gene_id.is_unique
        assert set(screen.gene_id) == set(tested.gene_id), "Screen omits tested tissue genes"
        assert screen.stable_id.eq(screen.gene_id.str.replace(r"\.\d+$", "", regex=True)).all()
        assert screen.stable_id.is_unique and screen.selected_for_coloc.isin([True, False]).all()
        assert screen.tss38.eq(np.where(screen.strand.eq("+"), screen.gene_start, screen.gene_end)).all()
        assert screen.cis_start38.eq(np.maximum(1, screen.tss38 - cis_distance)).all()
        assert screen.cis_end38.eq(screen.tss38 + cis_distance).all()
        original = tested.set_index("gene_id").loc[screen.gene_id]
        for column in ["num_var", "gene_start", "gene_end", "qval", "slope", "slope_se", "pval_nominal"]:
            assert np.allclose(screen[column].to_numpy(float), original[column].to_numpy(float),
                               rtol=1e-12, atol=0, equal_nan=True), ("Screen metadata differs", column)
        for column in ["gene_chr", "strand", "variant_id"]:
            assert np.array_equal(screen[column].to_numpy(), original[column].to_numpy()), ("Screen metadata differs", column)
        genes = screen.loc[screen.qval.le(q_threshold)].copy()
        assert genes.selected_for_coloc.eq(genes.FinnGen_significant_positions_in_cis.gt(0)).all()
        raw_gwas_rows = gwas_source["Cortex_coloc"]["raw_FinnGen_rows"]
        invalid_p_rows = gwas_source["Cortex_coloc"]["invalid_P_rows"]
        with gzip.open(eqtl_path, "rt", encoding="utf-8") as stream:
            eqtl_header = stream.readline().rstrip("\r\n").split("\t")
        assert eqtl_header == eqtl_source["full_associations"]["columns"]
    candidates = genes.loc[genes.selected_for_coloc].copy()
    assert len(candidates) > 0, "No tissue eGenes overlap a significant FinnGen association"
    print(f"FinnGen discovery: {len(candidates)} of {len(genes)} {tissue} eGenes selected", flush=True)
    locations = {}
    for gene in candidates.itertuples(index=False):
        locations.setdefault(gene.gene_chr.removeprefix("chr"), []).append((gene.cis_start38 - 1, gene.cis_end38))
    locations = {chromosome: merge_intervals(intervals) for chromosome, intervals in locations.items()}
    regions = [(chromosome, start, stop) for chromosome, intervals in locations.items() for start, stop in intervals]
    buffers = {gene.encode(): io.BytesIO() for gene in candidates.stable_id}
    eqtl_io, gwas_io = {}, {}
    if tissue == "Cortex":
        for fields, line in iter_rows(eqtl_path, eqtl_index, regions, eqtl_io):
            if fields[6] in buffers:
                buffers[fields[6]].write(line + b"\n")
    else:
        eqtl_io.update(format="ordinary_gzip_selected_gene_subset", compressed_bytes=eqtl_path.stat().st_size, rows=0)
        with gzip.open(eqtl_path, "rb") as stream:
            assert stream.readline().rstrip(b"\r\n").split(b"\t") == [name.encode() for name in eqtl_header]
            for line in stream:
                line = line.rstrip(b"\r\n")
                fields = line.split(b"\t")
                assert len(fields) == len(eqtl_header) and fields[6] in buffers, "Unexpected subset gene or schema"
                buffers[fields[6]].write(line + b"\n")
                eqtl_io["rows"] += 1
    starts = {chromosome: [a for a, b in intervals] for chromosome, intervals in locations.items()}
    gwas_buffer = io.BytesIO()
    for fields, line in iter_rows(finngen_path, finngen_index, regions, gwas_io):
        chromosome = fields[0].decode().removeprefix("chr")
        chromosome = "X" if chromosome == "23" else chromosome
        if chromosome not in locations:
            continue
        position0 = int(fields[1]) - 1
        interval = bisect_right(starts[chromosome], position0) - 1
        if interval >= 0 and position0 < locations[chromosome][interval][1]:
            gwas_buffer.write(line + b"\n")
    gwas_buffer.seek(0)
    gwas = pd.read_csv(gwas_buffer, sep="\t", names=finngen_index["header"], dtype={"#chrom": str})
    gwas_buffer.close()
    gwas["chromosome"] = gwas["#chrom"].str.removeprefix("chr").replace({"23": "X"})
    gwas["key"] = ("chr" + gwas.chromosome + ":" + gwas.pos.astype(str) + ":" + gwas.ref + ":" + gwas.alt)
    duplicate_columns = [column for column in finngen_index["header"] if column not in ["rsids", "nearest_genes"]]
    duplicates = gwas.key.duplicated(keep=False)
    if duplicates.any():
        differences = gwas.loc[duplicates].groupby("key")[duplicate_columns].nunique(dropna=False)
        assert not differences.gt(1).any().any(), "Conflicting FinnGen duplicate variants"
    gwas_raw_regional_rows = len(gwas)
    gwas = gwas.drop_duplicates("key").sort_values(["chromosome", "pos", "key"])
    gwas["valid_alleles"] = gwas.ref.str.fullmatch("[ACGT]+") & gwas.alt.str.fullmatch("[ACGT]+") & gwas.ref.ne(gwas.alt)
    gwas["valid_statistics"] = (np.isfinite(gwas.beta) & np.isfinite(gwas.sebeta) & gwas.sebeta.gt(0)
                                 & np.isfinite(gwas.pval) & gwas.pval.between(0, 1))
    by_chromosome = {chromosome: part for chromosome, part in gwas.groupby("chromosome", sort=False)}
    outcomes, total_eqtl_raw, total_eqtl_unique = [], 0, 0
    for number, gene in enumerate(candidates.itertuples(index=False), 1):
        buffer = buffers.pop(gene.stable_id.encode())
        buffer.seek(0)
        frame = pd.read_csv(buffer, sep="\t", names=eqtl_header, dtype={"chromosome": str})
        buffer.close()
        assert frame.gene_id.eq(gene.stable_id).all()
        raw_count = len(frame)
        duplicate_columns = [column for column in eqtl_header if column != "rsid"]
        duplicates = frame.variant.duplicated(keep=False)
        if duplicates.any():
            differences = frame.loc[duplicates].groupby("variant")[duplicate_columns].nunique(dropna=False)
            assert not differences.gt(1).any().any(), "Conflicting GTEx duplicate associations"
        frame = frame.drop_duplicates("variant")
        assert len(frame) == gene.num_var, "GTEx complete-gene variant count mismatch"
        assert frame.chromosome.str.removeprefix("chr").eq(gene.gene_chr.removeprefix("chr")).all()
        assert frame.position.between(gene.cis_start38, gene.cis_end38).all()
        reconstructed = (gene.gene_chr + "_" + frame.position.astype(str) + "_" + frame.ref + "_" + frame.alt)
        assert frame.variant.eq(reconstructed).all()
        lead_variant = gene.variant_id.removesuffix("_b38")
        lead = frame.loc[frame.variant.eq(lead_variant)]
        assert len(lead) == 1 and np.allclose(lead[["beta", "se", "pvalue"]].to_numpy(float)[0],
            [gene.slope, gene.slope_se, gene.pval_nominal], rtol=1e-12, atol=0)
        frame["valid_alleles"] = frame.ref.str.fullmatch("[ACGT]+") & frame.alt.str.fullmatch("[ACGT]+") & frame.ref.ne(frame.alt)
        frame["valid_statistics"] = (np.isfinite(frame.beta) & np.isfinite(frame.se) & frame.se.gt(0)
                                      & np.isfinite(frame.pvalue) & frame.pvalue.between(0, 1))
        frame["key"] = (gene.gene_chr + ":" + frame.position.astype(str) + ":" + frame.ref + ":" + frame.alt)
        valid_eqtl = frame.loc[frame.valid_alleles & frame.valid_statistics]
        chromosome = by_chromosome[gene.gene_chr.removeprefix("chr")]
        positions = chromosome.pos.to_numpy()
        a = np.searchsorted(positions, gene.cis_start38, side="left")
        b = np.searchsorted(positions, gene.cis_end38, side="right")
        regional = chromosome.iloc[a:b]
        lead_pool = regional.loc[np.isfinite(regional.pval) & regional.pval.between(0, 1)]
        order_columns = ["pval"] + (["mlogp"] if "mlogp" in lead_pool else []) + ["key"]
        regional_lead = lead_pool.sort_values(order_columns,
            ascending=[True] + ([False] if "mlogp" in lead_pool else []) + [True]).iloc[0]
        valid_gwas = regional.loc[regional.valid_alleles & regional.valid_statistics]
        shared = valid_eqtl.merge(valid_gwas[["key", "beta", "sebeta", "pval"]], on="key",
                                  suffixes=("_eqtl", "_GWAS"), validate="one_to_one")
        eqtl_lead_key = (gene.gene_chr + ":" + str(lead.position.iloc[0])
                         + ":" + lead.ref.iloc[0] + ":" + lead.alt.iloc[0])
        eqtl_lead_shared = bool(shared.key.eq(eqtl_lead_key).any())
        gwas_lead_shared = bool(shared.key.eq(regional_lead.key).any())
        n_shared = len(shared)
        if n_shared > 1:
            pair_input = pd.DataFrame({"snp": shared.key, "position": shared.position,
                "eqtl_beta": shared.beta_eqtl, "eqtl_se": shared.se,
                "GWAS_beta": shared.beta_GWAS, "GWAS_se": shared.sebeta})
            assert pair_input.snp.is_unique and np.isfinite(pair_input.iloc[:, 2:].to_numpy(float)).all()
            pair_input.to_csv(input_dir / (gene.stable_id + ".tsv.gz"), sep="\t", index=False,
                              float_format="%.17g", compression="gzip")
        snvs = shared.ref.str.len().eq(1) & shared.alt.str.len().eq(1)
        indels = shared.ref.str.len().ne(shared.alt.str.len())
        total_eqtl_raw += raw_count
        total_eqtl_unique += len(frame)
        notes = []
        if not eqtl_lead_shared:
            notes.append("eQTL_lead_not_shared")
        if not gwas_lead_shared:
            notes.append("GWAS_regional_lead_not_shared")
        if n_shared < 2:
            notes.append("insufficient_shared_variants")
        outcomes.append({"gene_id": gene.gene_id, "gene_name": gene.gene_name, "gene_chr": gene.gene_chr,
            "qval": gene.qval, "cis_start38": gene.cis_start38, "cis_end38": gene.cis_end38,
            "GTEx_expected_variants": int(gene.num_var), "archive_raw_rows": raw_count,
            "archive_unique_variants": len(frame), "duplicate_statistics_agree": True,
            "official_lead_statistics_agree": True, "valid_eqtl_variants": len(valid_eqtl),
            "eqtl_invalid_alleles": int((~frame.valid_alleles).sum()),
            "eqtl_invalid_statistics": int((~frame.valid_statistics).sum()),
            "GWAS_regional_variants": len(regional), "GWAS_regional_valid_variants": len(valid_gwas),
            "GWAS_regional_invalid_alleles": int((~regional.valid_alleles).sum()),
            "GWAS_regional_invalid_statistics": int((~regional.valid_statistics).sum()),
            "GWAS_shared_variants": n_shared, "shared_SNVS": int(snvs.sum()), "shared_indels": int(indels.sum()),
            "shared_other_sequence_variants": int((~snvs & ~indels).sum()),
            "GWAS_shared_fraction": n_shared / len(valid_eqtl) if len(valid_eqtl) else np.nan,
            "eQTL_lead_variant": eqtl_lead_key, "eQTL_lead_shared": eqtl_lead_shared,
            "GWAS_regional_lead_variant": regional_lead.key, "GWAS_regional_lead_P": regional_lead.pval,
            "GWAS_regional_lead_shared": gwas_lead_shared, "coverage_notes": ";".join(notes) or "both_leads_shared",
            "broad_MHC_caution": bool(gene.gene_chr == mhc_chromosome and gene.cis_start38 <= mhc_end and gene.cis_end38 >= mhc_start),
            "multiple_signal_assessment": "not_performed;single_causal_variant_assumption",
            "analysis_eligible": n_shared > 1})
        if number % 100 == 0 or number == len(candidates):
            print(f"FinnGen complete-gene preparation: {number}/{len(candidates)}", flush=True)
    qc = pd.DataFrame(outcomes)
    eligible = qc.loc[qc.analysis_eligible]
    assert len(eligible) > 0
    eligible[["gene_id"]].to_csv(qc_path, sep="\t", index=False)
    runtime = execute_cortex_coloc(input_dir, qc_path, native_path, "beta")
    native = pd.read_csv(native_path, sep="\t").rename(columns={"analysed_SNVS": "analysed_variants"})
    result = qc.merge(native, on="gene_id", how="left", validate="one_to_one")
    analysed = result.loc[result.analysis_eligible]
    assert analysed.analysed_variants.eq(analysed.GWAS_shared_variants).all()
    assert np.allclose(analysed[["H0", "H1", "H2", "H3", "H4"]].sum(axis=1), 1, atol=1e-10)
    result["H4_prior_robust"] = result.H4.ge(h4_threshold) & result.H4_p12_1e6.ge(h4_threshold)
    result["coverage_pass"] = result.eQTL_lead_shared & result.GWAS_regional_lead_shared
    result["main_candidate"] = result.H4_prior_robust & result.coverage_pass & ~result.broad_MHC_caution
    result["evidence_class"] = np.select([result.H4_prior_robust, result.H4.ge(h4_threshold), result.H3.ge(h4_threshold)],
        ["shared_support_prior_robust", "shared_support_prior_sensitive", "different_signal_support"], default="insufficient_support")
    result["GWAS_input_method"] = "FinnGen_ALT_logOR_beta_and_SE"
    result["model_status"] = "Single_causal_variant_ABF;multiple_signals_not_assessed"
    result["colocalization_run"] = result.analysis_eligible
    if tissue != "Cortex":
        result["tissue"] = eqtl_source["tissue"]
    result = result.sort_values(["main_candidate", "H4"], ascending=[False, False])
    result.to_csv(result_path, sep="\t", index=False)
    selected = result.loc[result.main_candidate]
    if selected_path is not None:
        selected.to_csv(selected_path, sep="\t", index=False)
    summary_key = "Cortex_coloc" if tissue == "Cortex" else "FinnGen_coloc"
    source[summary_key] = {"screened_eGenes": len(genes), "selected_genes": len(candidates),
        "analysed_genes": len(eligible), "raw_FinnGen_rows": raw_gwas_rows, "invalid_P_rows": invalid_p_rows,
        "screen_thresholds": {"eGene_q": q_threshold, "GWAS_P": gwas_threshold, "cis_distance": cis_distance},
        "scope": f"New screen of all {len(genes)} {tissue} eGenes; no original-GWAS or published-gene restriction",
        "matching": "Exact GRCh38 chromosome, position, REF and ALT; SNVs, indels and other sequence variants retained",
        "allele_alignment": "Both beta values refer to the same ALT; no strand, REF/ALT or sign changes",
        "variant_filter": "All shared valid beta/SE associations; no significance or MAF pruning after screening",
        "GTEx_raw_rows": total_eqtl_raw, "GTEx_unique_pairs": total_eqtl_unique,
        "GTEx_duplicate_rows": total_eqtl_raw - total_eqtl_unique,
        "FinnGen_regional_raw_rows": gwas_raw_regional_rows, "FinnGen_regional_unique_variants": len(gwas),
        "all_duplicate_statistics_agree": True, "all_GTEx_counts_and_lead_statistics_agree": True,
        "merged_regions": len(regions), "GTEx_indexed_IO": eqtl_io, "FinnGen_indexed_IO": gwas_io,
        "eQTL_lead_not_shared": int((~result.eQTL_lead_shared).sum()),
        "GWAS_regional_lead_not_shared": int((~result.GWAS_regional_lead_shared).sum()),
        "H4_ge_0_8": int(result.H4.ge(h4_threshold).sum()), "H4_prior_robust": int(result.H4_prior_robust.sum()),
        "main_candidates": len(selected), "runtime": runtime,
        "main_QC": "Both leads shared; H4>=0.8 at p12=1e-5 and 1e-6; excludes broad chr6:25-34Mb overlap",
        "MHC_caution_window": [mhc_chromosome, mhc_start, mhc_end],
        "single_causal_variant_assumption": True, "multiple_signals_assessed": False,
        "comparison_limit": "Original GWAS includes earlier FinnGen; comparison with R13 is not independent replication",
        "limitations": ("Adult cerebral-cortex proxy" if tissue == "Cortex" else f"Adult {tissue_names[tissue]} proxy") + "; exact indel representation may limit overlap; no conditional analysis or multi-signal fine-mapping; candidate support is conditional on the single-causal-variant model",
        "Cortex_source": {"dataset": eqtl_source["dataset"], "tissue": eqtl_source["tissue"],
                          "eqtl_samples": eqtl_source["eqtl_samples"], "eGenes_sha256": digest(egene_path),
                          **({"full_pairs_md5": eqtl_source["full_associations"]["provider_md5"]} if tissue == "Cortex"
                             else {"selected_pairs_sha256": eqtl_source["full_associations"]["sha256"]})},
        "screen_file": screen_path.name, "screen_sha256": digest(screen_path),
        "result_file": result_path.name, "result_sha256": digest(result_path),
        "colocalization_run": True}
    if tissue == "Cortex":
        source[summary_key].update(candidate_file=selected_path.name, candidate_sha256=digest(selected_path))
    else:
        source[summary_key]["GTEx_source"] = source[summary_key].pop("Cortex_source")
        source[summary_key]["GTEx_subset_IO"] = source[summary_key].pop("GTEx_indexed_IO")
        source[summary_key].update(screened_tested_genes=len(screen), GWAS_global_scan_reused=True,
            selected_gene_subset_only=True, complete_target_genes_verified=len(candidates),
            expected_target_pairs=int(candidates.num_var.sum()), observed_target_pairs=total_eqtl_unique)
        source[summary_key]["FinnGen_source"] = {"manifest": "FinnGen_Cortex_sources.json",
            "release": gwas_source["release"], "endpoint": gwas_source["endpoint"],
            "file_sha256": {name: record["sha256"] for name, record in gwas_source["summary_statistics"].items()}}
    manifest_path.write_text(json.dumps(source, indent=2) + "\n", encoding="utf-8")
    print(json.dumps(source[summary_key], indent=2))
    print(selected[["gene_name", "H3", "H4", "H4_p12_1e6", "analysed_variants"]].to_string(index=False))


#15. =========== Fixed neural-tissue extension: acquisition and complete reporting
def prepare_neural_screen():
    """Screen all tested genes in each added tissue against the same FinnGen GWAS."""
    import tarfile
    from datetime import datetime, timezone

    # Set this storage path independently before running.
    data_dir = ROOT
    archive_url = "https://storage.googleapis.com/adult-gtex/bulk-qtl/v8/single-tissue-cis-qtl/GTEx_Analysis_v8_eQTL.tar"
    archive_size, archive_etag = 1562828800, "d35b32152bdb21316b2509c46b0af998"
    members = {"Spinal": (444287488, 2671977), "Tibial": (989021184, 2909515)}
    member_hashes = {"Spinal": "e810b55e48860a3d67981b375525b99b63cb7549c04f8f6b69e34a10438a28b8",
                     "Tibial": "8008633f736fbbc5b28e714212e39bbb3a6179405f3fa7cd0df88632a4e56f19"}
    metadata_url = "https://gtexportal.org/api/v2/dataset/tissueSiteDetail?datasetId=gtex_v8&itemsPerPage=100"
    q_threshold, gwas_threshold, cis_distance = 0.05, 5e-8, 1000000
    scan_chunk_size, network_timeout = 750000, 90
    # Set this storage path independently before running. FinnGen_Cortex_sources.json.
    fg_source = json.loads((ROOT).read_text(encoding="utf-8"))
    # Set this storage path independently before running. FinnGen_Sensorineural_GWAS.gz.
    fg_path = ROOT
    assert digest(fg_path) == fg_source["summary_statistics"][fg_path.name]["sha256"]
    with urllib.request.urlopen(metadata_url, timeout=network_timeout) as stream:
        metadata = {row["tissueSiteDetailId"]: row for row in json.load(stream)["data"]}

    def ranged(start, stop):
        request = urllib.request.Request(archive_url, headers={"Range": f"bytes={start}-{stop}",
            "If-Match": '"' + archive_etag + '"'})
        with urllib.request.urlopen(request, timeout=network_timeout) as response:
            assert response.status == 206
            assert response.headers["Content-Range"] == f"bytes {start}-{stop}/{archive_size}"
            assert response.headers["ETag"].strip('"') == archive_etag
            content = response.read()
        assert len(content) == stop - start + 1
        return content

    tables, sources = {}, {}
    for label, (tissue, n_samples) in NEURAL_TISSUES.items():
        path = data_dir / f"GTEx_{label}_eGenes.tsv.gz"
        source_path = data_dir / f"GTEx_{label}_sources.json"
        offset, size = members[label]
        member_name = f"GTEx_Analysis_v8_eQTL/{tissue}.v8.egenes.txt.gz"
        member = tarfile.TarInfo.frombuf(ranged(offset, offset + 511), "utf-8", "strict")
        assert member.name == member_name and member.size == size
        content = path.read_bytes() if path.exists() else ranged(offset + 512, offset + 511 + size)
        assert len(content) == size and hashlib.sha256(content).hexdigest() == member_hashes[label]
        table = pd.read_csv(io.BytesIO(gzip.decompress(content)), sep="\t")
        required = {"gene_id", "gene_name", "gene_chr", "gene_start", "gene_end", "strand", "num_var",
                    "variant_id", "slope", "slope_se", "pval_nominal", "qval"}
        assert required.issubset(table.columns) and table.gene_id.is_unique
        assert table.qval.between(0, 1).all() and table.slope_se.gt(0).all()
        assert table.strand.isin(["+", "-"]).all()
        assert metadata[tissue]["eqtlSampleSummary"]["totalCount"] == n_samples
        assert int(table.qval.le(q_threshold).sum()) == metadata[tissue]["eGeneCount"]
        if not path.exists():
            with path.open("xb") as stream:
                stream.write(content)
        source = json.loads(source_path.read_text(encoding="utf-8")) if source_path.exists() else {}
        if "egenes" in source:
            assert source["egenes"]["sha256"] == hashlib.sha256(content).hexdigest()
        source.update(schema="GTEx_neural_inputs_v1", dataset="gtex_v8", tissue=tissue, label=label,
            genome_build="GRCh38", annotation="GENCODE26", eqtl_samples=n_samples,
            tissue_metadata=metadata[tissue], metadata_url=metadata_url,
            egenes={"path": path.name, "archive_url": archive_url, "archive_etag": archive_etag,
                "archive_bytes": archive_size, "member": member_name, "member_offset": offset + 512,
                "bytes": len(content), "sha256": hashlib.sha256(content).hexdigest(), "gzip_crc_verified": True,
                "tested_genes": len(table), "q_threshold": q_threshold,
                "selected_genes": int(table.qval.le(q_threshold).sum())},
            fixed_tissue_panel=["Brain_Cortex"] + [value[0] for value in NEURAL_TISSUES.values()],
            retrieved_utc=datetime.now(timezone.utc).isoformat())
        table["stable_id"] = table.gene_id.str.replace(r"\.\d+$", "", regex=True)
        assert table.stable_id.is_unique
        table["tss38"] = np.where(table.strand.eq("+"), table.gene_start, table.gene_end)
        table["cis_start38"] = np.maximum(1, table.tss38 - cis_distance)
        table["cis_end38"] = table.tss38 + cis_distance
        tables[label], sources[label] = table, source

    significant, raw_rows, invalid_p_rows = {}, 0, 0
    for frame in pd.read_csv(fg_path, sep="\t", usecols=["#chrom", "pos", "pval"],
                             dtype={"#chrom": str, "pos": np.int64}, chunksize=scan_chunk_size):
        raw_rows += len(frame)
        valid = np.isfinite(frame.pval) & frame.pval.between(0, 1)
        invalid_p_rows += int((~valid).sum())
        for chromosome, part in frame.loc[valid & frame.pval.lt(gwas_threshold)].groupby("#chrom", sort=False):
            chromosome = str(chromosome).removeprefix("chr")
            chromosome = "X" if chromosome == "23" else chromosome
            significant.setdefault(chromosome, []).append(part.pos.to_numpy(np.int64))
        if raw_rows % 3000000 == 0:
            print(f"FinnGen shared neural screen: {raw_rows:,} variants", flush=True)
    significant = {chromosome: np.unique(np.concatenate(parts)) for chromosome, parts in significant.items()}
    assert raw_rows == fg_source["Cortex_coloc"]["raw_FinnGen_rows"] and invalid_p_rows == 0
    for label, table in tables.items():
        counts = []
        for gene in table.itertuples(index=False):
            positions = significant.get(gene.gene_chr.removeprefix("chr"), np.empty(0, dtype=np.int64))
            counts.append(int(np.searchsorted(positions, gene.cis_end38, side="right")
                              - np.searchsorted(positions, gene.cis_start38)))
        table["FinnGen_significant_positions_in_cis"] = counts
        table["selected_for_coloc"] = table.qval.le(q_threshold) & table.FinnGen_significant_positions_in_cis.gt(0)
        screen_path = data_dir / f"FinnGen_{label}_screen.tsv"
        table.to_csv(screen_path, sep="\t", index=False, float_format="%.17g")
        sources[label]["GWAS_screen"] = {"GWAS": "FinnGen R13 H8_HL_SEN_NAS",
            "GWAS_source_file": "FinnGen_Cortex_sources.json", "GWAS_sha256": digest(fg_path),
            "raw_FinnGen_rows": raw_rows, "invalid_P_rows": invalid_p_rows,
            "screen_thresholds": {"eGene_q": q_threshold, "GWAS_P": gwas_threshold, "cis_distance": cis_distance},
            "screen_file": screen_path.name, "screen_sha256": digest(screen_path),
            "tested_genes": len(table), "screened_eGenes": int(table.qval.le(q_threshold).sum()),
            "selected_genes": int(table.selected_for_coloc.sum()),
            "scope": "All tested genes in this tissue; no restriction to Cortex or previously reported candidates"}
        (data_dir / f"GTEx_{label}_sources.json").write_text(json.dumps(sources[label], indent=2) + "\n", encoding="utf-8")
        print(json.dumps({"tissue": label, **sources[label]["GWAS_screen"]}), flush=True)


def download_neural_pairs():
    """Use source TBI ranges to retain every tested variant for selected genes only."""
    import concurrent.futures
    import struct
    import time
    import zlib
    from collections import Counter
    from datetime import datetime, timezone

    # Set this storage path independently before running.
    data_dir = ROOT
    # Set this storage path independently before running.
    scratch = ROOT
    scratch.mkdir(parents=True, exist_ok=True)
    base_url = "https://ftp.ebi.ac.uk/pub/databases/spot/eQTL/imported/GTEx_V8/ge/"
    resources = {"Spinal": (3484587416, "f2dbf0353ab37ed55f1a6b8930836a3a", "578cb2e6eea3f1289d2f8b6b49048672"),
                 "Tibial": (4030051960, "7fb602befb048c63847913f87ec11741", "fbd24e1cc82918cec4f5f0d233edcd50")}
    workers, retries, timeout = 4, 4, 120
    transfer_part_size, transfer_workers = 8 * 1024 * 1024, 4
    read_index, _, merge_intervals, chunks_for = finngen_bgzf_tools(include_chunks=True)

    def request_bytes(url, start=None, stop=None, expected_size=None, etag=None):
        for attempt in range(retries):
            try:
                headers = {} if start is None else {"Range": f"bytes={start}-{stop}"}
                if etag:
                    headers["If-Match"] = etag
                with urllib.request.urlopen(urllib.request.Request(url, headers=headers), timeout=timeout) as response:
                    if start is not None:
                        assert response.status == 206
                        assert response.headers["Content-Range"] == f"bytes {start}-{stop}/{expected_size}"
                        assert response.headers.get("ETag") == etag
                    content = response.read()
                if start is not None:
                    assert len(content) == stop - start + 1
                return content
            except Exception:
                if attempt + 1 == retries:
                    raise
                time.sleep(attempt + 1)

    checksums = request_bytes(base_url + "md5sum.txt").decode()
    for label, (tissue, n_samples) in NEURAL_TISSUES.items():
        source_path = data_dir / f"GTEx_{label}_sources.json"
        source = json.loads(source_path.read_text(encoding="utf-8"))
        screen_path = data_dir / f"FinnGen_{label}_screen.tsv"
        assert digest(screen_path) == source["GWAS_screen"]["screen_sha256"]
        table = pd.read_csv(screen_path, sep="\t")
        candidates = table.loc[table.selected_for_coloc].copy()
        assert len(candidates) == source["GWAS_screen"]["selected_genes"] > 0
        pair_path = data_dir / f"GTEx_{label}_pairs.tsv.gz"
        previous = source.get("full_associations", {})
        if pair_path.exists():
            assert previous["subset_complete_for_selected_genes"] and digest(pair_path) == previous["sha256"]
            assert previous["screen_sha256"] == digest(screen_path)
            print(f"{label}: existing complete selected-gene subset verified", flush=True)
            continue
        url = base_url + tissue + ".tsv.gz"
        file_size, file_md5, index_md5 = resources[label]
        assert any(line.rstrip().endswith(tissue + ".tsv.gz") and file_md5 in line for line in checksums.splitlines())
        assert any(line.rstrip().endswith(tissue + ".tsv.gz.tbi") and index_md5 in line for line in checksums.splitlines())
        with urllib.request.urlopen(urllib.request.Request(url, method="HEAD"), timeout=timeout) as response:
            assert int(response.headers["Content-Length"]) == file_size
            etag = response.headers["ETag"]
        range_dir = scratch / f"{label}-ranges"
        range_dir.mkdir(parents=True, exist_ok=True)
        # Set this storage path independently before running. source.json.
        cache_source = ROOT
        cache_identity = {"url": url, "etag": etag, "bytes": file_size}
        if cache_source.exists():
            assert json.loads(cache_source.read_text(encoding="utf-8")) == cache_identity
        else:
            cache_source.write_text(json.dumps(cache_identity), encoding="utf-8")
        head = request_bytes(url, 0, 65535, file_size, etag)
        header = zlib.decompressobj(31).decompress(head).split(b"\n", 1)[0].decode().split("\t")
        assert header[0] == "variant" and header[6] == "gene_id"
        assert {"variant", "pvalue", "beta", "se", "chromosome", "position", "ref", "alt"}.issubset(header)
        index_path = data_dir / f"GTEx_{label}_remote.tbi"
        index_content = index_path.read_bytes() if index_path.exists() else request_bytes(url + ".tbi")
        assert hashlib.md5(index_content).hexdigest() == index_md5
        if not index_path.exists():
            with index_path.open("xb") as stream:
                stream.write(index_content)
        index = read_index(pair_path, "chromosome", "position", header=header, index_path=index_path)
        regions = {}
        for gene in candidates.itertuples(index=False):
            regions.setdefault(gene.gene_chr, []).append((gene.cis_start38 - 1, gene.cis_end38))
        regions = [(chromosome, start, stop) for chromosome, intervals in regions.items()
                   for start, stop in merge_intervals(intervals)]
        chunks = merge_intervals([chunk for chromosome, start, stop in regions
                                  for chunk in chunks_for(index, chromosome, start, stop)])
        expected = dict(zip(candidates.stable_id, candidates.num_var))
        counts, leads, duplicate_rows = {}, set(), 0
        lead_by_gene = dict(zip(candidates.stable_id, candidates.variant_id.str.removesuffix("_b38")))
        positions = {name: header.index(name) for name in ["gene_id", "variant", "an", "beta", "se", "pvalue"]}
        active_key, active_signature = None, None
        raw_rows, compressed_bytes, blocks, subset_rows = 0, 0, 0, 0
        an_counts, non_nominal_an_examples = Counter(), []
        partial = scratch / f"GTEx_{label}_pairs.tsv.gz"
        if partial.exists():
            _release_cleanup_target(partial).unlink()

        def get_chunk(chunk):
            first, last = chunk
            start, stop = first >> 16, min(file_size - 1, (last >> 16) + 65535)
            cache_path = range_dir / f"{start}-{stop}.bgzf"
            if cache_path.exists():
                content = cache_path.read_bytes()
                assert len(content) == stop - start + 1
                return content
            if stop - start + 1 > 2 * transfer_part_size:
                part_ranges = [(a, min(stop, a + transfer_part_size - 1))
                               for a in range(start, stop + 1, transfer_part_size)]

                def get_part(bounds):
                    a, b = bounds
                    part_path = range_dir / f"part-{a}-{b}.bgzf"
                    if part_path.exists():
                        content = part_path.read_bytes()
                        assert len(content) == b - a + 1
                        return content
                    content = request_bytes(url, a, b, file_size, etag)
                    temporary = part_path.with_suffix(".partial")
                    temporary.write_bytes(content)
                    temporary.replace(part_path)
                    return content

                with concurrent.futures.ThreadPoolExecutor(max_workers=transfer_workers) as transfers:
                    content = b"".join(transfers.map(get_part, part_ranges))
            else:
                part_ranges = []
                content = request_bytes(url, start, stop, file_size, etag)
            with cache_path.open("xb") as stream:
                stream.write(content)
            for a, b in part_ranges:
                (_release_cleanup_target(range_dir / f"part-{a}-{b}.bgzf")).unlink()
            return content

        with gzip.open(partial, "wb", compresslevel=6) as output:
            output.write(("\t".join(header) + "\n").encode())
            pending = bytearray()
            with concurrent.futures.ThreadPoolExecutor(max_workers=workers) as pool:
                for batch_start in range(0, len(chunks), workers):
                    batch = chunks[batch_start:batch_start + workers]
                    for (first, last), content in zip(batch, pool.map(get_chunk, batch)):
                        compressed_bytes += len(content)
                        offset, end_offset, carry = first >> 16, last >> 16, b""
                        base = offset
                        while offset <= end_offset:
                            cursor = offset - base
                            head = content[cursor:cursor + 12]
                            assert head[:4] == b"\x1f\x8b\x08\x04"
                            extra_length = struct.unpack_from("<H", head, 10)[0]
                            extra = content[cursor + 12:cursor + 12 + extra_length]
                            i, size = 0, None
                            while i < len(extra):
                                sub_length = struct.unpack_from("<H", extra, i + 2)[0]
                                if extra[i:i + 2] == b"BC" and sub_length == 2:
                                    size = struct.unpack_from("<H", extra, i + 4)[0] + 1
                                i += 4 + sub_length
                            assert size is not None and 26 <= size <= 65536
                            block = zlib.decompress(content[cursor:cursor + size], 31)
                            blocks += 1
                            left = (first & 65535) if offset == base else 0
                            right = (last & 65535) if offset == end_offset else len(block)
                            lines = (carry + block[left:right]).split(b"\n")
                            carry = lines.pop()
                            for line in lines:
                                fields = line.split(b"\t")
                                if len(fields) != len(header):
                                    assert not line
                                    continue
                                gid = fields[positions["gene_id"]].decode()
                                if gid not in expected:
                                    continue
                                key = (gid, fields[positions["variant"]])
                                # EBI rsID annotation duplicates are adjacent in this variant-sorted source.
                                signature = tuple(value for i, value in enumerate(fields) if header[i] != "rsid")
                                if key == active_key:
                                    assert signature == active_signature, "Conflicting adjacent source duplicate"
                                    duplicate_rows += 1
                                active_key, active_signature = key, signature
                                an_text = fields[positions["an"]].decode()
                                an_value = float(an_text)
                                an_counts[an_text] += 1
                                if (not np.isfinite(an_value) or an_value != 2 * n_samples) and len(non_nominal_an_examples) < 3:
                                    non_nominal_an_examples.append({"gene_id": gid,
                                        "variant": fields[positions["variant"]].decode(), "AN": an_text})
                                counts[gid] = counts.get(gid, 0) + 1
                                if fields[positions["variant"]].decode() == lead_by_gene[gid]:
                                    leads.add(gid)
                                pending.extend(line + b"\n")
                                raw_rows += 1
                                if len(pending) >= 1024 * 1024:
                                    output.write(pending)
                                    pending.clear()
                            if offset == end_offset:
                                break
                            offset += size
                        assert not carry, "Remote TBI chunk boundary splits a record"
                    print(f"{label} remote ranges: {min(batch_start + workers, len(chunks))}/{len(chunks)}, "
                          f"{compressed_bytes / 1e6:.1f} MB transferred", flush=True)
            if pending:
                output.write(pending)
        # Independent per-gene unique-count and lead-statistic validation after streaming extraction.
        seen = {gid: set() for gid in expected}
        lead_stats, duplicates = {}, 0
        for frame in pd.read_csv(partial, sep="\t", chunksize=500000, dtype={"chromosome": str}):
            subset_rows += len(frame)
            assert frame.gene_id.isin(expected).all()
            for gid, part in frame.groupby("gene_id", sort=False):
                previous_n = len(seen[gid])
                seen[gid].update(part.variant)
                duplicates += len(part) - (len(seen[gid]) - previous_n)
                lead = part.loc[part.variant.eq(lead_by_gene[gid])]
                if not lead.empty:
                    lead_stats.setdefault(gid, []).extend(lead[["beta", "se", "pvalue"]].to_numpy(float).tolist())
        assert subset_rows == raw_rows and set(counts) == set(expected) and leads == set(expected)
        assert all(len(seen[gid]) == int(expected[gid]) for gid in expected), "Incomplete remote target gene"
        for gene in candidates.itertuples(index=False):
            assert np.allclose(lead_stats[gene.stable_id], [gene.slope, gene.slope_se, gene.pval_nominal], rtol=2e-12, atol=0)
        partial.rename(pair_path)
        source["full_associations"] = {"url": url, "path": pair_path.name,
            "provider": "EBI import of original GTEx v8; original statistics in harmonized format",
            "source_bytes": file_size, "provider_md5": file_md5, "source_etag": etag,
            "whole_source_md5_verified": False, "bytes": pair_path.stat().st_size, "sha256": digest(pair_path),
            "columns": header, "subset_complete_for_selected_genes": True,
            "subset_policy": "Every original tested variant for every selected gene; all P values retained; not whole-tissue allpairs",
            "selected_genes": len(candidates), "raw_rows": raw_rows,
            "unique_pairs": sum(map(len, seen.values())), "duplicate_rows": duplicates,
            "all_target_num_var_and_lead_statistics_verified": True, "BGZF_block_CRC_verified": True,
            "AN_row_counts": dict(sorted(an_counts.items())),
            "non_nominal_AN_examples": non_nominal_an_examples,
            "AN_policy": "Observed source field retained unchanged; nominal tissue N from official metadata; beta/SE coloc does not use AN or infer per-variant N",
            "screen_sha256": digest(screen_path),
            "source_index": {"url": url + ".tbi", "path": index_path.name, "bytes": len(index_content),
                "provider_md5": index_md5, "md5_verified": True, "sha256": digest(index_path),
                "usage": "Index of remote complete source; cannot index the local selected-gene subset"},
            "remote_IO": {"merged_regions": len(regions), "chunks": len(chunks),
                          "transferred_bytes": compressed_bytes, "BGZF_blocks": blocks},
            "retrieved_utc": datetime.now(timezone.utc).isoformat()}
        source_path.write_text(json.dumps(source, indent=2) + "\n", encoding="utf-8")
        print(json.dumps({"tissue": label, **source["full_associations"]}), flush=True)


def analyze_neural_finngen():
    """Keep Cortex fixed; run each added tissue separately and report the entire panel."""
    # Set this storage path independently before running.
    data_dir = ROOT
    h4_threshold, q_threshold = 0.8, 0.05
    panel = {"Cortex": ("Brain_Cortex", 205), **NEURAL_TISSUES}
    # Set this storage path independently before running. FinnGen_Cortex_sources.json.
    fixed_source = json.loads((ROOT).read_text(encoding="utf-8"))
    # Set this storage path independently before running. FinnGen_coloc_results.tsv.
    cortex_path = ROOT
    cortex_hash = digest(cortex_path)
    assert cortex_hash == fixed_source["Cortex_coloc"]["result_sha256"]
    for label in NEURAL_TISSUES:
        source = json.loads((data_dir / f"GTEx_{label}_sources.json").read_text(encoding="utf-8"))
        previous = source.get("FinnGen_coloc", {})
        result_path = data_dir / f"FinnGen_{label}_coloc.tsv"
        if (result_path.exists() and previous.get("result_sha256") == digest(result_path)
                and previous.get("screen_sha256") == digest(data_dir / f"FinnGen_{label}_screen.tsv")
                and previous.get("GTEx_source", {}).get("selected_pairs_sha256") == source["full_associations"]["sha256"]):
            print(f"{label}: completed result and input hashes verified", flush=True)
        else:
            analyze_finngen_cortex(label)
    assert digest(cortex_path) == cortex_hash

    references, screens, results = {}, {}, {}
    for label in panel:
        ref = pd.read_csv(data_dir / f"GTEx_{label}_eGenes.tsv.gz", sep="\t")
        ref["stable_id"] = ref.gene_id.str.replace(r"\.\d+$", "", regex=True)
        assert ref.stable_id.is_unique
        references[label] = ref
        screens[label] = pd.read_csv(data_dir / ("FinnGen_eGene_screen.tsv" if label == "Cortex"
                                               else f"FinnGen_{label}_screen.tsv"), sep="\t")
        results[label] = pd.read_csv(cortex_path if label == "Cortex"
                                    else data_dir / f"FinnGen_{label}_coloc.tsv", sep="\t")
    identity_columns = ["stable_id", "gene_id", "gene_name", "gene_chr"]
    identity = pd.concat([table[identity_columns] for table in references.values()], ignore_index=True)
    assert not identity.groupby("stable_id")[["gene_id", "gene_name", "gene_chr"]].nunique().gt(1).any().any()
    identity = identity.drop_duplicates("stable_id").sort_values(["gene_name", "stable_id"])
    rows = []
    for label, (tissue, sample_n) in panel.items():
        ref, screen, result = references[label], screens[label], results[label]
        base = identity.merge(ref[["stable_id", "qval"]], on="stable_id", how="left", validate="one_to_one")
        screen = screen[["gene_id", "FinnGen_significant_positions_in_cis", "selected_for_coloc"]]
        base = base.merge(screen, on="gene_id", how="left", validate="one_to_one")
        result = result.drop(columns=["gene_name", "gene_chr", "qval"])
        base = base.merge(result, on="gene_id", how="left", validate="one_to_one")
        base["label"], base["tissue"], base["eqtl_samples"] = label, tissue, sample_n
        base["tested_in_tissue"] = base.qval.notna()
        base["is_eGene"] = base.qval.le(q_threshold)
        selected = base.selected_for_coloc.eq(True)
        evaluated = base.colocalization_run.eq(True)
        base["evaluation_status"] = np.select([~base.tested_in_tissue, ~base.is_eGene,
            ~selected, ~evaluated], ["not_tested_in_tissue", "not_eGene", "no_significant_GWAS_in_cis",
                                  "insufficient_shared_variants"], default="colocalization_evaluated")
        base["default_support_QC_pass"] = base.H4.ge(h4_threshold) & base.coverage_pass.eq(True) & base.broad_MHC_caution.eq(False)
        base["prior_robust_QC_pass"] = base.main_candidate.eq(True)
        assert base.loc[~evaluated, "H4"].isna().all()
        assert int(evaluated.sum()) == int(result.colocalization_run.sum())
        rows.append(base)
    matrix = pd.concat(rows, ignore_index=True)
    matrix = matrix.sort_values(["gene_name", "stable_id", "label"])
    assert len(matrix) == len(identity) * len(panel) and not matrix.duplicated(["stable_id", "tissue"]).any()
    # Set this storage path independently before running. FinnGen_Neural_matrix.tsv.
    matrix_path = ROOT
    matrix.to_csv(matrix_path, sep="\t", index=False)
    supported = matrix.loc[matrix.default_support_QC_pass, "stable_id"].unique()
    candidates = []
    for gid in supported:
        group = matrix.loc[matrix.stable_id.eq(gid)].set_index("label")
        record = group.iloc[0][["stable_id", "gene_id", "gene_name", "gene_chr"]].to_dict()
        record["default_support_tissues"] = ";".join(group.index[group.default_support_QC_pass])
        record["prior_robust_tissues"] = ";".join(group.index[group.prior_robust_QC_pass])
        record["evaluated_tissues"] = ";".join(group.index[group.evaluation_status.eq("colocalization_evaluated")])
        for label in panel:
            for column in ["H3", "H4", "H4_p12_1e6", "coverage_pass", "evaluation_status"]:
                record[label + "_" + column] = group.loc[label, column]
        candidates.append(record)
    candidate_table = pd.DataFrame(candidates).sort_values("gene_name")
    # Set this storage path independently before running. FinnGen_Neural_candidates.tsv.
    candidate_path = ROOT
    candidate_table.to_csv(candidate_path, sep="\t", index=False)
    summary = {label: {"tested_genes": len(references[label]),
        "eGenes": int(references[label].qval.le(q_threshold).sum()),
        "selected_genes": int(screens[label].selected_for_coloc.sum()),
        "analysed_genes": int(results[label].colocalization_run.sum()),
        "H4_ge_0_8": int(results[label].H4.ge(h4_threshold).sum()),
        "default_support_QC_pass": int(matrix.loc[matrix.label.eq(label), "default_support_QC_pass"].sum()),
        "prior_robust_QC_pass": int(results[label].main_candidate.sum())} for label in panel}
    summary.update(candidate_genes=len(candidate_table), union_tested_genes=len(identity),
        matrix_rows=len(matrix), matrix_sha256=digest(matrix_path), candidates_sha256=digest(candidate_path),
        cortex_result_unchanged=digest(cortex_path) == cortex_hash,
        interpretation="Descriptive union of tissue-wise support; not independent replication or a combined posterior",
        multiplicity="Fixed three-tissue exploratory panel; all comparisons reported; H4 is not a P value and is not BH adjusted")
    for label in NEURAL_TISSUES:
        source_path = data_dir / f"GTEx_{label}_sources.json"
        source = json.loads(source_path.read_text(encoding="utf-8"))
        source["neural_panel_summary"] = summary
        source_path.write_text(json.dumps(source, indent=2) + "\n", encoding="utf-8")
    print(json.dumps(summary, indent=2), flush=True)
    print(candidate_table.to_string(index=False), flush=True)


def prepare_sgn_orthologs():
    """Audit the fixed six human candidates against official Ensembl 116 records."""
    import datetime
    import time
    import xml.etree.ElementTree as ET
    from concurrent.futures import ThreadPoolExecutor
    from xml.sax.saxutils import escape
    import requests

    # Fixed candidate scope and release; this step does not extract RNA or change caches.
    # Set this storage path independently before running.
    data_dir = ROOT
    # Set this storage path independently before running. FinnGen_Neural_candidates.tsv.
    candidate_path = ROOT
    # Set this storage path independently before running. hgnc_complete_set.txt.
    hgnc_path = ROOT
    # Set this storage path independently before running. SGN_Candidate_orthologs.tsv.
    table_path = ROOT
    # Set this storage path independently before running. SGN_Ortholog_sources.json.
    sources_path = ROOT
    release = 116
    timeout = 45
    request_attempts = 3
    workers = 3
    expected_ids = {"ENSG00000238018", "ENSG00000087302", "ENSG00000095596",
                    "ENSG00000168350", "ENSG00000161682", "ENSG00000115306"}
    candidates = pd.read_csv(candidate_path, sep="\t", dtype=str, keep_default_na=False)
    if (len(candidates) != 6 or set(candidates.stable_id) != expected_ids
            or candidates.stable_id.duplicated().any()):
        raise ValueError("Expected the fixed six-candidate neural colocalization union")
    candidate_sha, hgnc_sha = digest(candidate_path), digest(hgnc_path)
    ids = candidates.stable_id.tolist()

    def validate_table(table):
        required = {"human_id", "human_gtex_gene_id", "human_gtex_symbol", "human_symbol",
                    "human_biotype", "mouse_id", "mouse_symbol", "orthology_type",
                    "orthology_confidence", "human_percent_identity", "mouse_percent_identity",
                    "mouse_verified_synonyms", "status"}
        if not required.issubset(table.columns) or table.human_id.tolist() != ids:
            raise ValueError("Orthologue table schema or candidate order changed")
        if table.human_gtex_gene_id.tolist() != candidates.gene_id.tolist():
            raise ValueError("Original GTEx gene versions were not preserved")
        if table.human_gtex_symbol.tolist() != candidates.gene_name.tolist():
            raise ValueError("Original GTEx gene names were not preserved")
        mapped = table.loc[table.mouse_id.ne("")]
        missing = table.loc[table.mouse_id.eq("")]
        if (len(mapped) != 5 or mapped.mouse_id.duplicated().any()
                or not mapped.orthology_type.eq("ortholog_one2one").all()
                or not mapped.orthology_confidence.eq("1").all()
                or not mapped.status.eq("high_confidence_one_to_one").all()):
            raise ValueError("Expected five distinct high-confidence one-to-one orthologues")
        if (missing.human_id.tolist() != ["ENSG00000238018"]
                or missing.human_biotype.tolist() != ["lncRNA"]
                or missing.status.tolist() != ["no_mouse_orthologue_in_Ensembl116"]):
            raise ValueError("Missing orthologue must remain an explicitly unassigned lncRNA")
        rtraf = table.set_index("human_id").loc["ENSG00000087302"]
        if (rtraf.human_symbol != "RTRAF" or rtraf.mouse_symbol != "Rtraf"
                or rtraf.mouse_id != "ENSMUSG00000021807"
                or "2700060E02Rik" not in rtraf.mouse_verified_synonyms.split(";")):
            raise ValueError("Official Rtraf historical-symbol binding was not established")

    if table_path.exists() or sources_path.exists():
        if not (table_path.is_file() and sources_path.is_file()):
            raise RuntimeError("Incomplete orthologue outputs; refusing an unverified overwrite")
        saved = json.loads(sources_path.read_text(encoding="utf-8"))
        if (saved.get("schema") != "Fig4E-six-candidate-orthologues-v1"
                or saved.get("ensembl_release") != release
                or saved.get("inputs", {}).get("candidates_sha256") != candidate_sha
                or saved.get("inputs", {}).get("hgnc_sha256") != hgnc_sha
                or saved.get("table_sha256") != digest(table_path)):
            raise ValueError("Existing orthologue provenance does not match the accepted inputs")
        for record in saved["official_responses"].values():
            body = record["raw_text"].encode("utf-8")
            if len(body) != record["bytes"] or hashlib.sha256(body).hexdigest() != record["sha256"]:
                raise ValueError("Cached official response failed its SHA256 check")
        table = pd.read_csv(table_path, sep="\t", dtype=str, keep_default_na=False)
        validate_table(table)
        print("Orthologues: verified existing six rows; five mapped, one unassigned lncRNA", flush=True)
        return table

    def get_record(url, params=None, accept=None):
        headers = {"User-Agent": "Fig4E-orthologue-audit/1.0"}
        if accept:
            headers["Accept"] = accept
        for attempt in range(1, request_attempts + 1):
            try:
                response = requests.get(url, params=params, headers=headers, timeout=timeout)
                response.raise_for_status()
                break
            except requests.RequestException as error:
                status = error.response.status_code if error.response is not None else None
                if attempt == request_attempts or (status is not None and status not in {429, 500, 502, 503, 504}):
                    raise
                time.sleep(attempt)
        body = response.content
        raw = body.decode("utf-8")
        return {"url": response.url, "http_status": response.status_code, "attempts": attempt,
                "content_type": response.headers.get("Content-Type", ""),
                "bytes": len(body), "sha256": hashlib.sha256(body).hexdigest(), "raw_text": raw}

    def get_many(urls, accept=None):
        with ThreadPoolExecutor(max_workers=workers) as pool:
            values = list(pool.map(lambda url: get_record(url, accept=accept), urls.values()))
        return dict(zip(urls, values))

    records = get_many({
        "ensembl_release": "https://rest.ensembl.org/info/data?content-type=application/json",
        "biomart_registry": "https://useast.ensembl.org/biomart/martservice?type=registry"})
    if json.loads(records["ensembl_release"]["raw_text"]).get("releases") != [release]:
        raise RuntimeError("Live Ensembl REST is no longer release 116; do not silently change mappings")
    registry = ET.fromstring(records["biomart_registry"]["raw_text"])
    marts = [node.attrib for node in registry.iter("MartURLLocation")
             if node.attrib.get("name") == "ENSEMBL_MART_ENSEMBL"]
    if len(marts) != 1 or marts[0].get("database") != "ensembl_mart_116":
        raise RuntimeError("Official useast BioMart registry does not identify Ensembl Genes 116")

    def biomart(key, attributes, names):
        query = ('<Query virtualSchemaName="default" formatter="TSV" header="1" '
                 'uniqueRows="1" count="" datasetConfigVersion="0.6">'
                 '<Dataset name="hsapiens_gene_ensembl" interface="default">'
                 '<Filter name="ensembl_gene_id" value="' + escape(",".join(ids), {'"': '&quot;'})
                 + '"/>' + ''.join('<Attribute name="' + name + '"/>' for name in attributes)
                 + '</Dataset></Query>')
        record = get_record("https://useast.ensembl.org/biomart/martservice", params={"query": query})
        record["query_xml"] = query
        record["attributes"] = attributes
        raw = record["raw_text"]
        if "Query ERROR" in raw or raw.lstrip().startswith("<"):
            raise RuntimeError("Official BioMart returned an error or HTML, not a mapping table")
        frame = pd.read_csv(io.StringIO(raw), sep="\t", dtype=str, keep_default_na=False)
        if len(frame.columns) != len(names):
            raise ValueError("Unexpected BioMart field count")
        record["returned_headers"] = frame.columns.tolist()
        frame.columns = names
        if len(frame) != 6 or set(frame.human_id) != expected_ids or frame.human_id.duplicated().any():
            raise ValueError("BioMart must return exactly one record for each fixed human stable ID")
        records[key] = record
        return frame.set_index("human_id")

    orthology = biomart("biomart_orthologues", [
        "ensembl_gene_id", "external_gene_name", "mmusculus_homolog_ensembl_gene",
        "mmusculus_homolog_associated_gene_name", "mmusculus_homolog_orthology_type",
        "mmusculus_homolog_orthology_confidence", "mmusculus_homolog_perc_id",
        "mmusculus_homolog_perc_id_r1"], ["human_id", "human_symbol", "mouse_id", "mouse_symbol",
        "orthology_type", "orthology_confidence", "perc_id", "perc_id_r1"])
    # Gene annotations and homology fields belong to different BioMart attribute pages.
    annotation = biomart("biomart_gene_annotations", [
        "ensembl_gene_id", "ensembl_gene_id_version", "external_gene_name", "gene_biotype", "hgnc_id"],
        ["human_id", "human_current_gene_id", "human_symbol", "human_biotype", "hgnc_id"])
    urls = {}
    for gene in ids:
        urls["human_lookup:" + gene] = "https://rest.ensembl.org/lookup/id/" + gene + "?content-type=application/json"
        urls["human_homology:" + gene] = ("https://rest.ensembl.org/homology/id/human/" + gene
            + "?target_species=mus_musculus;type=orthologues;sequence=none;content-type=application/json")
        urls["hgnc_current:" + gene] = "https://rest.genenames.org/fetch/ensembl_gene_id/" + gene
    records.update(get_many(urls, accept="application/json"))
    mouse_ids = orthology.loc[orthology.mouse_id.ne(""), "mouse_id"].tolist()
    records.update(get_many({"mouse_lookup:" + gene:
        "https://rest.ensembl.org/lookup/id/" + gene + "?content-type=application/json" for gene in mouse_ids}))
    for gene in orthology.index[orthology.mouse_id.eq("")]:
        records["human_archive:" + gene] = get_record(
            "https://rest.ensembl.org/archive/id/" + gene + "?content-type=application/json")
    records.update(get_many({
        "rtraf_official_xrefs": "https://rest.ensembl.org/xrefs/id/ENSMUSG00000021807?content-type=application/json",
        "rtraf_historical_symbol": "https://rest.ensembl.org/xrefs/symbol/mus_musculus/2700060E02Rik?object_type=gene;content-type=application/json"}))
    xrefs = json.loads(records["rtraf_official_xrefs"]["raw_text"])
    mgi = [row for row in xrefs if row.get("dbname") == "MGI" and row.get("primary_id") == "MGI:1915295"]
    entrez = [row for row in xrefs if row.get("dbname") == "EntrezGene"]
    aliases = json.loads(records["rtraf_historical_symbol"]["raw_text"])
    if (len(mgi) != 1 or mgi[0].get("display_id") != "Rtraf"
            or "2700060E02Rik" not in mgi[0].get("synonyms", [])
            or aliases != [{"type": "gene", "id": "ENSMUSG00000021807"}]
            or len(entrez) != 1 or entrez[0].get("primary_id") != "68045"):
        raise ValueError("Official MGI/Ensembl historical Rtraf synonym is not uniquely bound")

    hgnc = pd.read_csv(hgnc_path, sep="\t", dtype=str, keep_default_na=False,
                       usecols=["hgnc_id", "symbol", "status", "ensembl_gene_id", "prev_symbol", "alias_symbol"])
    local_records, rows = {}, []
    for candidate in candidates.itertuples(index=False):
        gene = candidate.stable_id
        ann, homolog = annotation.loc[gene], orthology.loc[gene]
        human = json.loads(records["human_lookup:" + gene]["raw_text"])
        hgnc_result = json.loads(records["hgnc_current:" + gene]["raw_text"])["response"]
        if hgnc_result["numFound"] != 1:
            raise ValueError("HGNC current stable-ID lookup must resolve uniquely: " + gene)
        current_hgnc = hgnc_result["docs"][0]
        if (human["id"] != gene or human["assembly_name"] != "GRCh38"
                or human["display_name"] != ann.human_symbol
                or human["biotype"] != ann.human_biotype
                or gene + "." + str(human["version"]) != ann.human_current_gene_id
                or homolog.human_symbol != ann.human_symbol
                or current_hgnc.get("symbol") != ann.human_symbol
                or current_hgnc.get("hgnc_id") != ann.hgnc_id
                or current_hgnc.get("status") != "Approved"):
            raise ValueError("Human BioMart/REST/HGNC annotations disagree: " + gene)
        local = hgnc.loc[hgnc.ensembl_gene_id.eq(gene)]
        local_records[gene] = local.to_dict("records")
        if not local.empty and not local.hgnc_id.eq(ann.hgnc_id).all():
            raise ValueError("HGNC stable-ID identity changed from the accepted snapshot: " + gene)
        hom_data = json.loads(records["human_homology:" + gene]["raw_text"])["data"]
        if len(hom_data) != 1 or hom_data[0]["id"] != gene:
            raise ValueError("Unexpected Ensembl homology response")
        relations = hom_data[0]["homologies"]
        row = {"human_id": gene, "human_gtex_gene_id": candidate.gene_id,
               "human_gtex_symbol": candidate.gene_name, "human_symbol": ann.human_symbol,
               "hgnc_id": ann.hgnc_id, "hgnc_status": current_hgnc["status"],
               "human_biotype": ann.human_biotype, "human_current_gene_id": ann.human_current_gene_id,
               "human_previous_symbols": ";".join(current_hgnc.get("prev_symbol", [])),
               "mouse_id": "", "mouse_symbol": "", "mouse_biotype": "", "mouse_current_gene_id": "",
               "orthology_type": "", "orthology_confidence": "", "human_percent_identity": "",
               "mouse_percent_identity": "", "mouse_verified_synonyms": "", "mouse_mgi_id": "",
               "mouse_entrez_id": "", "status": "no_mouse_orthologue_in_Ensembl116"}
        if homolog.mouse_id:
            if len(relations) != 1:
                raise ValueError("Mapped candidate does not have exactly one mouse orthologue")
            relation = relations[0]
            mouse = json.loads(records["mouse_lookup:" + homolog.mouse_id]["raw_text"])
            if (relation["source"]["id"] != gene or relation["target"]["id"] != homolog.mouse_id
                    or relation["type"] != homolog.orthology_type or mouse["id"] != homolog.mouse_id
                    or mouse["display_name"] != homolog.mouse_symbol or mouse["assembly_name"] != "GRCm39"
                    or not np.isclose(float(homolog.perc_id), float(relation["source"]["perc_id"]), atol=1e-4)
                    or not np.isclose(float(homolog.perc_id_r1), float(relation["target"]["perc_id"]), atol=1e-4)):
                raise ValueError("Mouse BioMart/REST identity or sequence identities disagree: " + gene)
            row.update(mouse_id=homolog.mouse_id, mouse_symbol=homolog.mouse_symbol,
                mouse_biotype=mouse["biotype"], mouse_current_gene_id=homolog.mouse_id + "." + str(mouse["version"]),
                orthology_type=homolog.orthology_type, orthology_confidence=homolog.orthology_confidence,
                human_percent_identity=str(relation["source"]["perc_id"]),
                mouse_percent_identity=str(relation["target"]["perc_id"]), status="high_confidence_one_to_one")
            mgi_ids = current_hgnc.get("mgd_id", [])
            if len(mgi_ids) == 1:
                row["mouse_mgi_id"] = mgi_ids[0]
            if gene == "ENSG00000087302":
                row.update(mouse_verified_synonyms="2700060E02Rik", mouse_mgi_id="MGI:1915295", mouse_entrez_id="68045")
        else:
            archive = json.loads(records["human_archive:" + gene]["raw_text"])
            if relations or archive.get("is_current") != "1" or str(archive.get("release")) != str(release):
                raise ValueError("No-orthologue assignment lacks a current stable-ID and empty homology result")
        rows.append(row)
    table = pd.DataFrame(rows)
    validate_table(table)
    table_bytes = table.to_csv(sep="\t", index=False, lineterminator="\n").encode("utf-8")
    sources = {"schema": "Fig4E-six-candidate-orthologues-v1",
        "created_utc": datetime.datetime.now(datetime.timezone.utc).isoformat(),
        "ensembl_release": release, "biomart_registry_entry": marts[0],
        "assemblies": {"human": "GRCh38", "mouse": "GRCm39"},
        "release_provenance": "Live official useast BioMart registry and Ensembl REST both verified release 116; not an archive snapshot",
        "inputs": {"candidates_path": str(candidate_path.relative_to(ROOT)), "candidates_sha256": candidate_sha,
                   "hgnc_path": str(hgnc_path.relative_to(ROOT)), "hgnc_sha256": hgnc_sha},
        "table_path": str(table_path.relative_to(ROOT)), "table_sha256": hashlib.sha256(table_bytes).hexdigest(),
        "official_responses": records, "local_hgnc_selected_rows": local_records,
        "identity_definition": "human_percent_identity=REST homology source.perc_id; mouse_percent_identity=target.perc_id; independently compared with BioMart attributes",
        "scope": "Six fixed candidates retained; five protein-coding high-confidence one-to-one orthologues; unassigned human lncRNA retained without mouse expression",
        "rna_note": "No RNA extraction or global feature mapping/cache changes. Rtraf historical synonym is candidate-specific evidence only; original raw feature versions are not overwritten.",
        "checks": {"candidate_rows": len(table), "mapped_rows": int(table.mouse_id.ne("").sum()),
                   "unique_mapped_mouse_ids": int(table.loc[table.mouse_id.ne(""), "mouse_id"].nunique()),
                   "unassigned_rows": int(table.mouse_id.eq("").sum()), "raw_official_responses": len(records)}}
    if digest(candidate_path) != candidate_sha or digest(hgnc_path) != hgnc_sha:
        raise RuntimeError("An accepted input changed while official annotations were being retrieved")
    sources_bytes = (json.dumps(sources, indent=2, ensure_ascii=False) + "\n").encode("utf-8")
    with table_path.open("xb") as handle:
        handle.write(table_bytes)
    with sources_path.open("xb") as handle:
        handle.write(sources_bytes)
    print("Orthologues: saved six rows; five high-confidence one-to-one, one unassigned lncRNA", flush=True)
    return table


#17. =========== Six fixed genetic candidates in source-aware SGN development
def prepare_sgn_development():
    """Recover native counts for the six candidates without changing cells or tree."""
    from scipy import sparse

    # Set this storage path independently before running.
    data_dir = ROOT
    # Set this storage path independently before running.
    rna_dir = ROOT
    min_bin_library_cells = 3
    gene_order = ["ENSG00000168350", "ENSG00000161682", "ENSG00000115306",
                  "ENSG00000087302", "ENSG00000095596", "ENSG00000238018"]
    n = native_rna_tools()
    audited_obs, _, fixed_audit = audit_fixed_inputs()
    with h5py.File(WHOLE, "r") as handle:
        frozen_obs = n["frame"](handle["obs"])
    assert frozen_obs.index.equals(audited_obs.index)
    # Set this storage path independently before running. SGN_Candidate_orthologs.tsv.
    ortholog_path = ROOT
    orthology = pd.read_csv(ortholog_path, sep="\t", dtype=str, keep_default_na=False).set_index("human_id").loc[gene_order].reset_index()
    assert len(orthology) == 6 and orthology.human_id.is_unique
    mapped = orthology.mouse_id.ne("")
    assert mapped.sum() == 5 and orthology.loc[mapped, "mouse_id"].is_unique
    assert orthology.loc[mapped, "orthology_type"].eq("ortholog_one2one").all()
    assert orthology.loc[mapped, "orthology_confidence"].eq("1").all()
    assert orthology.set_index("human_id").loc["ENSG00000087302", "mouse_verified_synonyms"].find("2700060E02Rik") >= 0
    # Set this storage path independently before running. candidates.h5.
    candidate_path = ROOT
    # Set this storage path independently before running. candidates.json.
    old_manifest = json.loads((ROOT).read_text(encoding="utf-8"))
    assert digest(candidate_path) == old_manifest["cache_sha256"]
    with h5py.File(candidate_path, "r") as handle:
        cells = handle["cell_ids"].asstr()[()]
        assert np.array_equal(cells, frozen_obs.index.to_numpy())
        metadata = {key: handle[key].asstr()[()] for key in ["stage_standardized", "state", "study_id", "library_id", "count_unit_verified"]}
        denominator = handle["denominator"][()]
        old_sptbn1 = handle["gene_symbols"].asstr()[()].tolist().index("Sptbn1")
        old_sptbn1_counts = handle["counts"][:, old_sptbn1]
        old_sptbn1_expression = handle["log10_1p_CPM"][:, old_sptbn1]
    assert len(cells) == 8340 and np.isfinite(denominator).all() and (denominator > 0).all()
    for key in ["stage_standardized", "study_id", "library_id"]:
        assert np.array_equal(metadata[key], frozen_obs[key].astype(str).to_numpy())
    with h5py.File(HI, "r") as handle:
        assert np.array_equal(cells[:7355], handle["cells/_index"].asstr()[()])
        assert np.array_equal(denominator[:7355], handle["cells/library_total"][()])
        assert np.array_equal(metadata["state"][:7355], handle["cells/state"].asstr()[()])
        segment = np.full(8340, "Postnatal", dtype=object)
        segment[:7355] = handle["cells/seg"].asstr()[()]
        native_t = np.full(8340, np.nan)
        native_t[:7355] = handle["cells/t"][()]
        tree_input = json.loads(handle["meta/input"].asstr()[()])
    assert tree_input["tree_bundle_sha256"] == EXPECTED_HASHES[TREE]
    # Set this storage path independently before running. candidate9_segment_adaptive.h5.
    with h5py.File(ROOT, "r") as handle:
        segment_order = handle["segment_order"].asstr()[()]
        bin_edges = handle["native_t_quantile_edges"][()]
        effective_bins = handle["effective_quantile_bins"][()]
    assert segment_order.tolist() == tree_input["segment_order"]
    scope = np.where(segment == "VGN", "VGN_reference", np.where(segment == "CVG", "shared_CVG_context",
                      np.where(segment == "Postnatal", "postnatal_SGN", "SGN_frozen_tree")))
    obs = pd.DataFrame({**metadata, "segment": segment, "native_t": native_t, "scope": scope}, index=cells)
    counts = np.full((8340, 6), np.nan)
    measured = np.zeros((8340, 6), dtype=bool)
    mapped_columns = np.flatnonzero(mapped)
    mouse_ids = pd.Index(orthology.loc[mapped, "mouse_id"])
    # Set this storage path independently before running. counts.json.
    common_manifest = json.loads((ROOT).read_text(encoding="utf-8"))
    # Set this storage path independently before running. common_counts.h5.
    common_path = ROOT
    assert digest(common_path) == common_manifest["cache_sha256"]
    with h5py.File(common_path, "r") as handle:
        assert np.array_equal(handle["cell_ids"].asstr()[()], cells)
        common_ids = pd.Index(handle["gene_ids"].asstr()[()])
        common = sparse.csr_matrix((handle["counts/data"][()], handle["counts/indices"][()],
                                    handle["counts/indptr"][()]), shape=tuple(handle["counts"].attrs["shape"]))
    common_positions = common_ids.get_indexer(mouse_ids)
    assert int((common_positions >= 0).sum()) == 3
    # Set this storage path independently before running. native_mapping.tsv.gz.
    source_mapping = pd.read_csv(ROOT, sep="\t", keep_default_na=False)
    assert digest(n["MGI"]) == n["MGI_SHA256"] and digest(n["GTF"]) == n["GTF_SHA256"]
    alias_evidence = []
    with gzip.open(n["MGI"], "rt", encoding="utf-8") as stream:
        header = next(stream).rstrip("\n").split("\t")
        for line_number, line in enumerate(stream, 2):
            fields = line.rstrip("\n").split("\t")
            record = dict(zip(header, fields))
            if (record.get("Marker Symbol") == "2700060E02Rik" and record.get("Status") == "W"
                    and record.get("Current MGI Accession ID (if withdrawn)") == "MGI:1915295"
                    and record.get("Current Marker Symbol (if withdrawn)") == "Rtraf"):
                alias_evidence.append({"line": line_number, "record": record, "type": "withdrawn_successor"})
            elif (record.get("MGI Accession ID") == "MGI:1915295" and record.get("Marker Symbol") == "Rtraf"
                    and "2700060E02Rik" in record.get("Marker Synonyms (pipe-separated)", "").split("|")):
                alias_evidence.append({"line": line_number, "record": record, "type": "official_synonym"})
    assert len(alias_evidence) == 2
    source_reports, overridden = [], []
    expected_hashes = {str(item["path"]).replace("\\", "/"): item["sha256"] for item in common_manifest["sources"]}
    verified_paths = {}

    def verify_source(path):
        relative = path.relative_to(ROOT).as_posix()
        if relative not in verified_paths:
            actual = digest(path)
            assert actual == expected_hashes[relative], ("Native source fingerprint changed", relative)
            verified_paths[relative] = actual

    def endogenous_axis(path):
        with h5py.File(path, "r") as handle:
            features = n["frame"](handle["var"])
        eligible = np.ones(len(features), dtype=bool)
        for key in ["technical_feature", "ERCC_feature", "reporter_or_vector_feature", "external_spike_feature"]:
            if key in features:
                eligible &= ~features[key].to_numpy(dtype=bool)
        return features, eligible

    for source, path in n["SOURCES"].items():
        rows = np.flatnonzero(obs.study_id.eq(source).to_numpy())
        table = source_mapping.loc[source_mapping.study.eq(source)].sort_values("source_column").copy()
        assert np.array_equal(table.source_column, np.arange(len(table)))
        if source in ["GSE195500", "GSE165502"]:
            legacy = table.source_feature.eq("2700060E02Rik")
            assert legacy.sum() == 1 and not table.target_id.eq("ENSMUSG00000021807").any()
            old_record = table.loc[legacy].iloc[0].to_dict()
            assert old_record["decision"] == "excluded_alias_only_candidate"
            table.loc[legacy, "target_id"] = "ENSMUSG00000021807"
            table.loc[legacy, "decision"] = "candidate_only_unique_MGI_withdrawn_successor_and_official_synonym"
            overridden.append({"source": source, "original_mapping": old_record,
                               "mouse_id": "ENSMUSG00000021807", "MGI": "MGI:1915295"})
        available = np.array([int(table.target_id.eq(gene).sum()) for gene in mouse_ids])
        assert (available <= 1).all(), "Multiple source columns for one candidate"
        present = available == 1
        measured[np.ix_(rows, mapped_columns)] = present[None, :]
        target_columns = mouse_ids.get_indexer(table.target_id)
        table["common_column"] = target_columns
        values = np.zeros((len(rows), len(mouse_ids)))
        if source == "GSE178931":
            for library in pd.unique(obs.iloc[rows].library_id):
                local = np.flatnonzero(obs.iloc[rows].library_id.eq(library).to_numpy())
                selected = rows[local]
                stage = obs.iloc[selected[0]].stage_standardized.split(".")[0]
                native_path = n["QC"] / f"{stage}_GSE178931__{library}.h5ad"
                verify_source(native_path)
                features, eligible = endogenous_axis(native_path)
                assert features.index.tolist() == table.source_feature.tolist()
                matrix, totals = n["h5_rows"](native_path, cells[selected], target_columns, len(mouse_ids),
                    endogenous_eligible=eligible, return_totals=True)
                assert np.array_equal(totals, denominator[selected])
                values[local] = matrix.toarray()
        elif source in ["GSE165502", "Local_P7_SET7"]:
            verify_source(path)
            features, eligible = endogenous_axis(path)
            assert features.index.tolist() == table.source_feature.tolist()
            matrix, totals = n["h5_rows"](path, cells[rows], target_columns, len(mouse_ids),
                endogenous_eligible=eligible, return_totals=True)
            assert np.array_equal(totals, denominator[rows])
            values = matrix.toarray()
        elif source == "GSE195500":
            present_ids = mouse_ids[present]
            subset = table.copy()
            subset["common_column"] = present_ids.get_indexer(subset.target_id)
            matrix, native_path = n["csv195500"](frozen_obs, rows, subset, len(present_ids))
            verify_source(native_path)
            values[:, present] = matrix.toarray()
        elif source == "GSE231513":
            verify_source(path)
            raw = pd.read_csv(path, sep="\t", index_col=0)
            assert raw.columns.tolist() == table.source_feature.tolist() and set(raw.index) == set(cells[rows])
            assert np.array_equal(raw.loc[cells[rows]].sum(axis=1), denominator[rows])
            for j in np.flatnonzero(present):
                source_column = int(table.loc[table.target_id.eq(mouse_ids[j]), "source_column"].iloc[0])
                values[:, j] = raw.loc[cells[rows]].iloc[:, source_column].to_numpy()
        else:
            for library in pd.unique(obs.iloc[rows].library_id):
                local = np.flatnonzero(obs.iloc[rows].library_id.eq(library).to_numpy())
                selected = rows[local]
                qc_path = n["QC"] / f"P28_GSE202920__{library}.h5ad"
                features, eligible = endogenous_axis(qc_path)
                assert np.array_equal(features.source_feature_id.astype(str), table.source_feature.astype(str))
                matrix, native_path, totals = n["mtx_p28"](frozen_obs, selected, table, len(mouse_ids), library,
                    endogenous_eligible=eligible, return_totals=True)
                verify_source(native_path)
                assert np.array_equal(totals, denominator[selected])
                values[local] = matrix.toarray()
        assert np.isfinite(values).all() and (values >= 0).all() and np.equal(values, np.rint(values)).all()
        counts[np.ix_(rows, mapped_columns[present])] = values[:, present]
        for j in np.flatnonzero(common_positions >= 0):
            assert np.array_equal(values[:, j], common[rows, common_positions[j]].toarray().ravel()), (source, mouse_ids[j])
        source_reports.append({"source": source, "cells": len(rows), "present_mouse_genes": mouse_ids[present].tolist(),
            "full_source_denominator_verified": True, "shared_candidate_raw_counts_exact": True})
        print(f"SGN native counts verified: {source}, {len(rows)} fixed cells", flush=True)
    assert np.array_equal(np.isfinite(counts), measured)
    assert np.array_equal(counts[:, gene_order.index("ENSG00000115306")], old_sptbn1_counts)
    cpm = counts / denominator[:, None] * 1e6
    expression = np.log10(1 + cpm)
    assert np.array_equal(expression[:, gene_order.index("ENSG00000115306")], old_sptbn1_expression)
    detection = np.where(measured, counts > 0, np.nan)
    identity_columns = ["human_id", "human_gtex_symbol", "human_symbol", "mouse_id", "mouse_symbol"]
    coverage_rows = []
    for source in ["All"] + list(n["SOURCES"]):
        rows = np.arange(8340) if source == "All" else np.flatnonzero(obs.study_id.eq(source).to_numpy())
        core = rows[obs.scope.iloc[rows].isin(["SGN_frozen_tree", "postnatal_SGN"]).to_numpy()]
        for j, gene in orthology.iterrows():
            valid, core_valid = rows[measured[rows, j]], core[measured[core, j]]
            coverage_rows.append({**gene[identity_columns].to_dict(), "source": source,
                "n_cells": len(rows), "n_measured_cells": len(valid),
                "n_detected_cells": int((counts[valid, j] > 0).sum()) if len(valid) else np.nan,
                "detection_fraction": float((counts[valid, j] > 0).mean()) if len(valid) else np.nan,
                "sgn_cells": len(core), "sgn_measured_cells": len(core_valid),
                "sgn_detected_cells": int((counts[core_valid, j] > 0).sum()) if len(core_valid) else np.nan,
                "n_measured_libraries": int(obs.iloc[valid].library_id.nunique()),
                "measured": bool(len(valid)), "missing_reason": ("" if len(valid) else
                    "no_mouse_orthologue" if not gene.mouse_id else "source_feature_absent")})
    coverage = pd.DataFrame(coverage_rows)
    # Set this storage path independently before running. SGN_Candidate_coverage.tsv.
    coverage_path = ROOT
    coverage.to_csv(coverage_path, sep="\t", index=False)
    library_rows, summary_rows = [], []

    def aggregate(context, group_metadata, rows, minimum=1):
        libraries = pd.unique(obs.iloc[rows].library_id)
        eligible = [rows[obs.iloc[rows].library_id.eq(library).to_numpy()] for library in libraries]
        eligible = [selected for selected in eligible if len(selected) >= minimum]
        for j, gene in orthology.iterrows():
            identity = gene[identity_columns].to_dict()
            local_records = []
            for selected in eligible:
                valid = selected[measured[selected, j]]
                assert len(valid) in [0, len(selected)]
                record = {**identity, "context": context, **group_metadata,
                    "library_id": obs.iloc[selected[0]].library_id, "n_cells": len(selected),
                    "n_measured_cells": len(valid), "n_detected_cells": int((counts[valid, j] > 0).sum()) if len(valid) else np.nan,
                    "mean_log10_1p_CPM": float(expression[valid, j].mean()) if len(valid) else np.nan,
                    "detection_fraction": float(detection[valid, j].mean()) if len(valid) else np.nan,
                    "mean_CPM": float(cpm[valid, j].mean()) if len(valid) else np.nan,
                    "gene_count_sum": float(counts[valid, j].sum()) if len(valid) else np.nan,
                    "denominator_sum": float(denominator[valid].sum()) if len(valid) else np.nan,
                    "t_mean": float(np.mean(native_t[selected])) if np.isfinite(native_t[selected]).all() else np.nan,
                    "t_min": float(np.min(native_t[selected])) if np.isfinite(native_t[selected]).all() else np.nan,
                    "t_max": float(np.max(native_t[selected])) if np.isfinite(native_t[selected]).all() else np.nan}
                record["missing_reason"] = ("no_mouse_orthologue" if not gene.mouse_id else
                    "source_feature_absent" if not len(valid) else
                    "measured_zero" if record["n_detected_cells"] == 0 else "observed")
                local_records.append(record)
                library_rows.append(record)
            valid_records = [record for record in local_records if record["n_measured_cells"] > 0]
            reason = ("no_mouse_orthologue" if not gene.mouse_id else
                      "no_eligible_library_bin" if not eligible else
                      "source_feature_absent" if not valid_records else
                      "measured_zero" if sum(record["n_detected_cells"] for record in valid_records) == 0 else "observed")
            row = {**identity, "context": context, **group_metadata, "n_cells": len(rows),
                "n_libraries": len(libraries), "n_eligible_libraries": len(eligible),
                "n_contributing_cells": sum(len(selected) for selected in eligible),
                "n_measured_libraries": len(valid_records),
                "n_measured_cells": sum(record["n_measured_cells"] for record in valid_records),
                "n_detected_cells": sum(record["n_detected_cells"] for record in valid_records) if valid_records else np.nan,
                "missing_reason": reason}
            for column in ["mean_log10_1p_CPM", "detection_fraction", "mean_CPM", "t_mean"]:
                row[column] = float(np.mean([record[column] for record in valid_records])) if valid_records else np.nan
            row["t_min"] = float(np.min([record["t_min"] for record in valid_records])) if valid_records else np.nan
            row["t_max"] = float(np.max([record["t_max"] for record in valid_records])) if valid_records else np.nan
            for column in ["gene_count_sum", "denominator_sum"]:
                row[column] = float(sum(record[column] for record in valid_records)) if valid_records else np.nan
            summary_rows.append(row)

    display_scope = obs.scope.ne("VGN_reference").to_numpy()
    for source in n["SOURCES"]:
        for stage in pd.unique(obs.loc[obs.study_id.eq(source) & display_scope, "stage_standardized"]):
            rows = np.flatnonzero(obs.study_id.eq(source).to_numpy() & obs.stage_standardized.eq(stage).to_numpy() & display_scope)
            aggregate("stage", {"source": source, "stage": stage, "state": "All retained states", "segment": "", "bin": -1}, rows)
            for state in pd.unique(obs.iloc[rows].state):
                local = rows[obs.iloc[rows].state.eq(state).to_numpy()]
                aggregate("stage_state", {"source": source, "stage": stage, "state": state, "segment": "", "bin": -1}, local)
    for si, name in enumerate(segment_order):
        members = np.flatnonzero(segment == name)
        edges = bin_edges[si, np.isfinite(bin_edges[si])]
        bins = np.searchsorted(edges[1:-1], native_t[members], side="left")
        expected_bins = pd.qcut(native_t[members], q=int(min(20, max(3, len(members) // 50))), labels=False, duplicates="drop")
        assert np.array_equal(bins, expected_bins)
        for b in range(int(effective_bins[si])):
            selected = members[bins == b]
            for source in ["GSE178931", "GSE231513", "GSE195500", "GSE165502"]:
                local = selected[obs.iloc[selected].study_id.eq(source).to_numpy()]
                aggregate("segment_bin", {"source": source, "stage": "Embryonic", "state": "Frozen tree",
                    "segment": name, "bin": b}, local, min_bin_library_cells)
    libraries, development = pd.DataFrame(library_rows), pd.DataFrame(summary_rows)
    # Set this storage path independently before running. SGN_Candidate_libraries.tsv.
    library_path = ROOT
    # Set this storage path independently before running. SGN_Candidate_development.tsv.
    development_path = ROOT
    libraries.to_csv(library_path, sep="\t", index=False)
    development.to_csv(development_path, sep="\t", index=False)
    report = {"schema": "SGN_six_candidate_development_v1", "candidate_ids": gene_order,
        "orthologs_sha256": digest(ortholog_path), "cells": 8340, "embryo_cells": 7355,
        "scope_counts": obs.scope.value_counts().to_dict(), "fixed_input_audit": fixed_audit,
        "original_cell_annotations_and_tree_unchanged": True, "frozen_native_t_and_quantile_edges_reused": True,
        "source_counts": source_reports, "source_input_hashes": verified_paths,
        "candidate_only_alias_overrides": overridden, "MGI_alias_evidence": alias_evidence,
        "MGI_sha256": n["MGI_SHA256"], "M25_sha256": n["GTF_SHA256"],
        "old_common_counts_sha256": digest(common_path), "old_candidate_cache_sha256": digest(candidate_path),
        "old_Sptbn1_counts_and_expression_exact": True, "three_common_candidate_counts_exact": True,
        "normalization": "log10(1 + native count / full available endogenous source denominator * 1e6)",
        "denominator": "Unchanged previously audited vector; all extraction totals rechecked except archived source-bound P1 CSV totals",
        "bin_library_min_cells": min_bin_library_cells,
        "aggregation": "Mean across cells within each true library; equal means across measured eligible libraries within source",
        "inference": "Descriptive observations; no cell-as-animal P values, cross-source maturation fit or imputation",
        "source_confounds": "P1/P7/P28 are different sources/platforms; read counts, UMI counts and assigned pairs are not interchangeable",
        "missingness": "No mouse orthologue and absent source feature remain NA; measured zero remains zero; bins without eligible libraries remain NA"}
    # Set this storage path independently before running. SGN_Candidate_expression.h5.
    expression_path = ROOT
    with h5py.File(expression_path, "w") as handle:
        text = h5py.string_dtype("utf-8")
        for key, value in [("counts", counts), ("measured", measured), ("detection", detection),
                           ("denominator", denominator), ("log10_1p_CPM", expression), ("native_t", native_t)]:
            handle.create_dataset(key, data=value, compression="gzip", shuffle=True)
        for key, value in [("cell_ids", cells), ("human_ids", orthology.human_id),
            ("human_symbols", orthology.human_symbol), ("human_gtex_symbols", orthology.human_gtex_symbol),
            ("mouse_ids", orthology.mouse_id), ("mouse_symbols", orthology.mouse_symbol),
            ("segment", segment), ("scope", scope), *metadata.items()]:
            handle.create_dataset(key, data=np.asarray(value, dtype=object), dtype=text)
        tree = handle.create_group("tree")
        tree.create_dataset("segment_order", data=np.asarray(segment_order, dtype=object), dtype=text)
        tree.create_dataset("native_t_quantile_edges", data=bin_edges)
        tree.create_dataset("effective_quantile_bins", data=effective_bins)
        tree.create_dataset("segment_edges", data=json.dumps(tree_input["segment_edges"]), dtype=text)
        handle.create_dataset("manifest_json", data=json.dumps(report), dtype=text)
    report["outputs"] = {p.name: {"bytes": p.stat().st_size, "sha256": digest(p)}
        for p in [expression_path, coverage_path, library_path, development_path]}
    # Set this storage path independently before running. SGN_Candidate_sources.json.
    (ROOT).write_text(json.dumps(report, indent=2) + "\n", encoding="utf-8")
    print(coverage.loc[coverage.source.eq("All"), ["human_symbol", "mouse_symbol", "n_measured_cells", "n_detected_cells",
        "sgn_measured_cells", "sgn_detected_cells"]].to_string(index=False), flush=True)
    print(json.dumps({"cell_order_and_annotations_fixed": True, "tree_and_bins_fixed": True,
        "library_records": len(libraries), "development_records": len(development), "source_input_hashes_checked": len(verified_paths)}), flush=True)


```

```python
#1. =========== Public rendering helpers (inlined; no external analysis script)
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib.transforms import Bbox
from matplotlib.colors import Normalize
from matplotlib.patches import Rectangle, FancyArrowPatch, FancyBboxPatch

def vector_colorbar(ax, cmap, norm, ticks, tick_size=7, decimals=None, n=256):
    for i in range(n):
        color=cmap((i+.5)/n)
        ax.add_patch(Rectangle((i/n,0),1/n,1,facecolor=color,edgecolor=color,linewidth=.12,clip_on=False))
    labels=[f"{v:.{decimals}f}" if decimals is not None else f"{v:g}" for v in ticks]
    ax.set(xlim=(0,1),ylim=(0,1));ax.set_xticks(np.asarray(norm(ticks),float),labels)
    ax.set_yticks([]);ax.tick_params(labelsize=tick_size,length=2,pad=2)
    for spine in ax.spines.values():spine.set_visible(False)

REFERENCES=[
 dict(authors="Hamel et al.",year=2024,journal="Nature Communications",panels="Fig. 1b, 1d",
      doi="10.1038/s41467-023-44380-y",url="https://www.nature.com/articles/s41467-023-44380-y/figures/1",
      JIF_year=2025,JIF=18.1,used="Parallel GWAS/eQTL inputs, colocalization, then atlas expression mapping"),
 dict(authors="Haglund et al.",year=2025,journal="Nature Genetics",panels="Fig. 1a",
      doi="10.1038/s41588-024-02050-9",url="https://www.nature.com/articles/s41588-024-02050-9/figures/1",
      JIF_year=2025,JIF=25.5,used="Modular analysis workflow and aligned regional association schematic"),
]
METRICS_URL="https://www.nature.com/nature-portfolio/about-journals/journal-metrics"
```

```python
"""Strict vector PDF export for the Fig4 B/C/D Matplotlib canvas.

No analysis is performed. Each native Matplotlib label becomes one live Arial
text object, including the TF italic run in the two supported mathtext titles.
All geometry is written directly, without XObjects, glyph outlines or images.
The caller declares at most four data axes; their clip rectangles are collected
into one sibling viewport group per axis. Other artists must fit their axes.
"""

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")
from pathlib import Path
import re
import sys
import math
import numpy as np
from matplotlib.backend_bases import RendererBase
from matplotlib.path import Path as MplPath

try:
    from reportlab.pdfgen import canvas
except ImportError:
    # Set this storage path independently before running.
    sys.path.append(str(ROOT))
    from reportlab.pdfgen import canvas
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont

_FONTS = {'normal':'FigArial','italic':'FigArialItalic',
          'bold':'FigArialBold','bolditalic':'FigArialBoldItalic'}

def _register():
    for key, file in [('normal','arial.ttf'),('italic','ariali.ttf'),
                      ('bold','arialbd.ttf'),('bolditalic','arialbi.ttf')]:
        if _FONTS[key] not in pdfmetrics.getRegisteredFontNames():
            # Set this storage path independently before running.
            pdfmetrics.registerFont(TTFont(_FONTS[key],str(ROOT/file)))

def _font(prop):
    bold=prop.get_weight() in ('bold','heavy','black','semibold','demibold') or isinstance(prop.get_weight(),(int,float)) and prop.get_weight()>=600
    italic=prop.get_style() in ('italic','oblique')
    return _FONTS['bolditalic' if bold and italic else 'bold' if bold else 'italic' if italic else 'normal']

def _runs(text,prop,ismath):
    if not ismath:
        if '$' in text: raise ValueError('Unexpected dollar in plain label: '+text)
        return [(text,_font(prop))]
    # Current Fig4 titles use only this explicitly audited mathtext form.
    found=re.fullmatch(r'\$\\it\{([^{}]+)\}([^$]*)\$(.*)',text)
    if found is None: raise ValueError('Unsupported math label: '+text)
    tf,inside,after=found.groups()
    return [(tf,_FONTS['italic']),(inside+after,_FONTS['normal'])]

class _Recorder(RendererBase):
    def __init__(self,fig,agg,viewports,halo_labels):
        super().__init__()
        self.dpi=fig.dpi;self.unit=72/fig.dpi;self.agg=agg
        self.width,self.height=fig.bbox.size;self.ops=[];self.viewports=viewports;self.halo_labels=halo_labels
    def points_to_pixels(self,p): return p/self.unit
    def flipy(self):return False
    def get_canvas_width_height(self):return self.width,self.height
    def get_text_width_height_descent(self,s,prop,ismath):
        runs=_runs(s,prop,ismath)
        width=sum(pdfmetrics.stringWidth(t,f,prop.get_size_in_points()) for t,f in runs)/self.unit
        # Match the source Arial baseline/ascent geometry; width uses the same
        # embedded font as the PDF, including the semantic mixed italic title.
        _,h,d=self.agg.get_text_width_height_descent(re.sub(r'\$|\\it\{|\}', '',s),prop,False)
        return width,h,d
    def _clip(self,gc):
        if gc.get_clip_path()[0] is not None:
            raise ValueError('Nonrectangular artist clip is unsupported')
        b=gc.get_clip_rectangle()
        if b is None:return None
        for i,box in enumerate(self.viewports):
            if np.allclose(b.bounds,box,rtol=0,atol=1e-5):return i
        return None
    def draw_path(self,gc,path,transform,rgbFace=None):
        if gc.get_hatch():raise ValueError('Hatching/Pattern is forbidden')
        p=transform.transform_path(path)
        self.ops.append(('path',self._clip(gc),p,gc.get_rgb(),rgbFace,
                         gc.get_linewidth(),gc.get_alpha(),gc.get_forced_alpha(),
                         gc.get_dashes(),gc.get_capstyle(),gc.get_joinstyle()))
    def draw_text(self,gc,x,y,s,prop,angle,ismath=False,mtext=None):
        self.ops.append(('text',None,float(x),float(y),str(s),prop.copy(),float(angle),
                         ismath,gc.get_rgb(),gc.get_alpha(),id(mtext) in self.halo_labels))
    def draw_image(self,*args,**kwargs):raise ValueError('Raster image is forbidden')
    def start_rasterizing(self):pass
    def stop_rasterizing(self):pass
    def start_filter(self):raise ValueError('Raster filter is forbidden')
    def stop_filter(self,*args):raise ValueError('Raster filter is forbidden')

def _pdf_path(pdf,path,unit):
    out=pdf.beginPath();current=None;start=None
    for vertices,code in path.iter_segments(remove_nans=True,simplify=False,curves=True):
        v=np.asarray(vertices).reshape(-1,2)*unit
        if code==MplPath.MOVETO:
            current=v[0];start=current;out.moveTo(*current)
        elif code==MplPath.LINETO:
            current=v[0];out.lineTo(*current)
        elif code==MplPath.CURVE3:
            control,end=v
            a=current+(control-current)*2/3;b=end+(control-end)*2/3
            out.curveTo(*a,*b,*end);current=end
        elif code==MplPath.CURVE4:
            out.curveTo(*v.ravel());current=v[-1]
        elif code==MplPath.CLOSEPOLY:
            out.close();current=start
        elif code==MplPath.STOP:break
        else:raise ValueError('Unknown path opcode '+str(code))
    return out

def strict_editable_pdf(fig,destination,*,data_axes=(),live_halo_width=.20,crop_inches=None):
    """Write a strict vector PDF. Return declared clip boxes in PDF points.

    The artist canvas remains unchanged after return. Rasterization flags and
    white text path effects are replaced by a narrow live text fill+stroke
    (render mode 2) in the same single label object; no outlined font paths are
    emitted. The width is narrower than Agg's halo to preserve the black stems.
    """
    _register()
    if len(data_axes)>4:raise ValueError('At most four sibling data viewports')
    fig.canvas.draw();agg=fig.canvas.get_renderer()
    viewports=[tuple(ax.bbox.bounds) for ax in data_axes]
    effects=[]
    for artist in fig.findobj():
        if isinstance(artist, __import__('matplotlib.text',fromlist=['Text']).Text) and hasattr(artist,'get_path_effects'):
            old=artist.get_path_effects()
            if old:effects.append((artist,old));artist.set_path_effects([])
    rec=_Recorder(fig,agg,viewports,{id(a) for a,e in effects})
    try:fig.draw(rec)
    finally:
        for a,e in effects:a.set_path_effects(e)
    dst=Path(destination);dst.parent.mkdir(parents=True,exist_ok=True)
    crop=np.asarray(crop_inches if crop_inches is not None else [0,0,*fig.get_size_inches()],float)
    unit=rec.unit
    # Discard figure/axes backgrounds and objects belonging to the other panels.
    kept=[]
    x0,y0,w,h=crop*fig.dpi
    for op in rec.ops:
        if op[0]=='text':
            xx,yy=op[2:4]
            inside=x0-1<=xx<=x0+w+1 and y0-1<=yy<=y0+h+1
        else:
            vv=op[2].vertices
            if not len(vv):continue
            box=Bbox.from_extents(*np.r_[vv.min(axis=0),vv.max(axis=0)])
            is_background=(box.width>=fig.bbox.width-1 and box.height>=fig.bbox.height-1)
            inside=(not is_background and box.x1>=x0 and box.x0<=x0+w and box.y1>=y0 and box.y0<=y0+h)
        if inside:kept.append(op)
    rec.ops=kept
    pdf=canvas.Canvas(str(dst),pagesize=tuple(crop[2:]*72),
                      pageCompression=1,invariant=1,initialFontName=_FONTS['normal'])
    pdf.translate(-crop[0]*72,-crop[1]*72)
    def emit(op):
        if op[0]=='text':
            _,_,x,y,s,prop,angle,ismath,color,alpha,halo=op
            if not s:return
            pdf.saveState();pdf.setFillColorRGB(*color[:3]);pdf.setFillAlpha(alpha)
            text=pdf.beginText();rad=math.radians(angle)
            if halo and live_halo_width>0:
                pdf.setStrokeColorRGB(1,1,1);pdf.setStrokeAlpha(alpha)
                pdf.setLineWidth(float(live_halo_width));text.setTextRenderMode(2)
            text.setTextTransform(math.cos(rad),math.sin(rad),-math.sin(rad),math.cos(rad),x*unit,y*unit)
            for value,font in _runs(s,prop,ismath):
                if value:text.setFont(font,prop.get_size_in_points());text.textOut(value)
            pdf.drawText(text);pdf.restoreState();return
        _,_,path,stroke,face,lw,alpha,forced,dashes,cap,join=op
        pdf.saveState();pdf.setLineWidth(lw)
        pdf.setLineCap({'butt':0,'round':1,'projecting':2}[cap])
        pdf.setLineJoin({'miter':0,'round':1,'bevel':2}[join])
        offset,pattern=dashes
        pdf.setDash([] if pattern is None else [float(v) for v in pattern],0.0 if offset is None else float(offset))
        if stroke is not None:
            pdf.setStrokeColorRGB(*stroke[:3]);pdf.setStrokeAlpha(alpha if forced else (stroke[3] if len(stroke)>3 else alpha))
        if face is not None:
            pdf.setFillColorRGB(*face[:3]);pdf.setFillAlpha(alpha if forced else (face[3] if len(face)>3 else alpha))
        pdf.drawPath(_pdf_path(pdf,path,unit),stroke=int(stroke is not None and lw>0),fill=int(face is not None),fillMode=1)
        pdf.restoreState()
    done=set()
    for op in rec.ops:
        key=op[1]
        if key is None:emit(op);continue
        if key in done:continue
        done.add(key);pdf.saveState()
        p=pdf.beginPath();p.rect(*(np.asarray(viewports[key])*unit))
        pdf.clipPath(p,stroke=0,fill=0,fillMode=1)
        for candidate in rec.ops:
            if candidate[1]==key:emit(candidate)
        pdf.restoreState()
    pdf.showPage();pdf.save()
    if done!=set(range(len(data_axes))):raise AssertionError('Declared clip has no artist')
    return {'expected_clips':len(done),'clip_boxes_pt':(np.asarray(viewports)*unit).tolist(),
            'editable_label_objects':sum(op[0]=='text' for op in rec.ops),
            'label_strings':[op[4] for op in rec.ops if op[0]=='text'],
            'vector_paths':sum(op[0]=='path' for op in rec.ops),
            'text_effects':'White live text fill+stroke, single object; stroke width '+str(live_halo_width)+' pt.',
            'halo_labels':len(effects)}
```

```python

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")
def build_BC_canvas():
    """Plot a short genetic workflow, five frozen whole-atlas features and a state DotPlot."""
    import matplotlib
    matplotlib.use("Agg")
    import matplotlib.pyplot as plt
    from matplotlib.colors import LinearSegmentedColormap, Normalize, FuncNorm
    from matplotlib.cm import ScalarMappable
    from matplotlib.patches import FancyArrowPatch, PathPatch, Ellipse, Circle, Rectangle
    from matplotlib.text import Text
    from matplotlib.path import Path as MplPath
    import matplotlib.patheffects as pe
    from scipy.interpolate import UnivariateSpline
    from scipy.spatial import cKDTree

    #1. =========== Figure-only parameters; the existing atlas and counts stay fixed
    # Set this storage path independently before running.
    data_dir = ROOT
    # Set this storage path independently before running. SGN_Candidate_expression.h5.
    expression_path = ROOT
    # Set this storage path independently before running. FinnGen_Neural_candidates.tsv.
    candidate_path = ROOT
    # Set this storage path independently before running. SGN_Candidate_orthologs.tsv.
    ortholog_path = ROOT
    # Set this storage path independently before running. SGN_Candidate_sources.json.
    source_path = ROOT
    # Set this storage path independently before running. Fig4E_Candidate_development.png.
    output_path = ROOT
    # Set this storage path independently before running. Development_preview.png.
    preview_path = ROOT
    human_order = ["DEGS2", "FAM171A2", "SPTBN1", "RTRAF", "CYP26A1", "SPTBN1-AS2"]
    figure_size, final_dpi, preview_dpi = (13.6, 8.0), 600, 140
    feature_area, feature_alpha = 1.85, .90
    feature_vmin, feature_vmax = 0.0, 4.0  # Common actual log10(1+CPM), not per-gene maxima.
    dot_max_area, column_gap_ratio = 144.0, .5  # Point AREA is linear in detection fraction.
    gene_font, state_font = 8.0, 7.4
    stage_font, stage_left = 7.2, 8.95
    curve_t_start_bin, curve_samples = 2, 200
    curve_smoothing_rms = .35  # UMAP units; display smoothing only, with fixed endpoints.
    curve_cloud_max_distance = .50  # In the unchanged UMAP coordinate units.
    curve_labels = {"AC": ("Ib/Ic", (4.4, 7.6)), "AB": ("Ia/II", (-7.5, 5.2))}
    ink, muted, zero_color, missing_color = "#20242A", "#6B7078", "#D9D9D9", "#BBBBBB"
    color_log_tau = 1.0  # Display normalization only; existing RNA values are not transformed again.
    feature_norm = FuncNorm((lambda x: np.log1p(x/color_log_tau),
                            lambda y: color_log_tau*np.expm1(y)),
                           vmin=feature_vmin, vmax=feature_vmax, clip=True)
    color_knots = [0, .5, 1, 1.5, 2, 2.5, 3, 3.5, 4]
    color_values = ["#F2F2F2", "#FFD8D3", "#FFC0B8", "#FF9A8A", "#FF6D63",
                    "#F54E4A", "#D92D37", "#BC1428", "#9C001E"]
    feature_cmap = LinearSegmentedColormap.from_list("RNA_nonlinear_red",
        list(zip(feature_norm(color_knots), color_values)), N=1024)
    dot_cmap = LinearSegmentedColormap.from_list("candidate_expression", ["#F1F1F1", "#E8C6DE", "#C67ABA", "#952E7D", "#69005F"])
    dot_norm = Normalize(-1, 1, clip=True)
    plt.rcParams.update({"font.family": "Arial", "font.size": 8, "text.color": ink,
        "axes.unicode_minus": False, "axes.linewidth": .55, "mathtext.default": "regular"})

    #2. =========== Exact cell/embedding correspondence and fixed candidate checks
    source = json.loads(source_path.read_text(encoding="utf-8"))
    input_paths = [expression_path, candidate_path, ortholog_path, WHOLE]
    input_hashes = {p.name: digest(p) for p in input_paths}
    assert input_hashes[expression_path.name] == source["outputs"][expression_path.name]["sha256"]
    assert input_hashes[WHOLE.name] == EXPECTED_HASHES[WHOLE]
    with h5py.File(expression_path, "r") as h:
        humans = h["human_symbols"].asstr()[()].tolist()
        human_ids = h["human_ids"].asstr()[()].tolist()
        mouse_ids = h["mouse_ids"].asstr()[()].tolist()
        mouse = h["mouse_symbols"].asstr()[()].tolist()[:5]
        cells = h["cell_ids"].asstr()[()]
        states = h["state"].asstr()[()]
        sources = h["study_id"].asstr()[()]
        libraries = h["library_id"].asstr()[()]
        stages = h["stage_standardized"].asstr()[()]
        cell_segments = h["segment"].asstr()[()]
        native_t = h["native_t"][()]
        saved_segment_order = h["tree/segment_order"].asstr()[()].tolist()
        native_edges = h["tree/native_t_quantile_edges"][()]
        counts = h["counts"][()][:, :5]
        measured = h["measured"][()].astype(bool)[:, :5]
        expression = h["log10_1p_CPM"][()][:, :5]
        assert not h["measured"][:, -1].any() and np.isnan(h["counts"][:, -1]).all()
    assert humans == human_order and counts.shape == measured.shape == expression.shape == (8340, 5)
    assert len(set(cells)) == 8340 and all(mouse_ids[:5]) and mouse_ids[-1] == ""
    assert np.array_equal(np.isfinite(expression), measured)
    assert np.equal(expression[measured] > 0, counts[measured] > 0).all()
    assert np.isfinite(expression[measured]).all() and expression[measured].min() >= feature_vmin
    assert expression[measured].max() <= feature_vmax
    with h5py.File(WHOLE, "r") as h:
        assert np.array_equal(index(h["obs"]).astype(str), cells)
        xy = h["obsm/X_umap"][()]
        accepted_hash = h["uns/css/umap_sha256"].asstr()[()]
        full_labels = np.asarray(vector(h["obs/source_stage_celltype"])).astype(str)
        original_row_order = h["uns/css_figure_contract/dot_row_order"].asstr()[()].tolist()
    embedding_digest = hashlib.sha256()
    embedding_digest.update(str(xy.shape).encode()); embedding_digest.update(str(xy.dtype).encode())
    embedding_digest.update(np.ascontiguousarray(xy).tobytes())
    assert embedding_digest.hexdigest() == accepted_hash == "4cd0a00aaccc010c8f8ed2f258462ab1524d2df93f9541396001c9bda7f2254b"
    assert np.array_equal(states, [x.split(" | ", 1)[1] for x in full_labels])
    assert np.array_equal(stages, [x.split(" | ", 1)[0] for x in full_labels])
    state_order = list(dict.fromkeys(x.split(" | ", 1)[1] for x in original_row_order))
    assert len(state_order) == 18 and set(state_order) == set(states)
    stage_sets = [sorted(set(stages[states == state]), key=lambda age: (age[0] == "P", float(age[1:])))
                  for state in state_order]
    stage_display_rules = {
        ("E11.5", "E13.5"): "E11.5 / E13.5",
        ("E14.5", "E15.5", "E16.5", "E18.5"): "E14.5–E16.5; E18.5",
        ("E14.5", "E15.5", "E16.5", "E17.5"): "E14.5–E17.5",
        ("P1", "P7", "P28"): "P1 / P7 / P28",
    }
    stage_labels = [values[0] if len(values) == 1 else stage_display_rules[tuple(values)]
                    for values in stage_sets]
    # Display guides project accepted native-time bins onto the fixed UMAP; no tree refit.
    guides, guide_controls = {}, {}
    for segment in ["T", "AC", "AB"]:
        members = np.flatnonzero(cell_segments == segment)
        edge = native_edges[saved_segment_order.index(segment)]
        edge = edge[np.isfinite(edge)]
        bin_ids = np.searchsorted(edge[1:-1], native_t[members], side="left")
        start_bin = curve_t_start_bin if segment == "T" else 0
        anchors, records = [], []
        for bin_id in range(start_bin, len(edge)-1):
            current = members[bin_ids == bin_id]
            neighborhood = members[(np.abs(bin_ids-bin_id) <= 1) & (bin_ids >= start_bin)]
            assert len(current) > 0
            robust_center = np.median(xy[neighborhood], axis=0)
            nearest = current[np.argmin(np.linalg.norm(xy[current]-robust_center, axis=1))]
            anchors.append(xy[nearest].astype(float))
            records.append({"segment": segment, "native_bin": bin_id, "n_bin_cells": len(current),
                "cell_id": str(cells[nearest]), "cell_native_t": float(native_t[nearest]),
                "bin_median_t": float(np.median(native_t[current])), "xy": xy[nearest].astype(float).tolist()})
        anchors = np.asarray(anchors)
        guide_controls[segment] = records
        controls = anchors
        times = np.array([record["bin_median_t"] for record in records])
        if segment != "T":
            controls = np.vstack([guides["T"]["xy"][-1], controls])
            times = np.r_[guides["T"]["t"][-1], times]
        assert np.all(np.diff(times) > 0)
        display_t = np.linspace(times[0], times[-1], curve_samples)
        boundary_weights = np.ones(len(controls))
        boundary_weights[[0, -1]] = 1000.0
        display_xy = np.column_stack([UnivariateSpline(times, controls[:, j], w=boundary_weights, k=3,
            s=len(controls)*curve_smoothing_rms**2)(display_t) for j in range(2)])
        progress = (display_t-display_t[0]) / (display_t[-1]-display_t[0])
        display_xy += ((1-progress[:, None])*(controls[0]-display_xy[0])
                       + progress[:, None]*(controls[-1]-display_xy[-1]))
        display_xy[[0, -1]] = controls[[0, -1]]
        related = np.isin(cell_segments, ["T", segment])
        cloud_distance = cKDTree(xy[related]).query(display_xy)[0]
        assert cloud_distance.max() < curve_cloud_max_distance, (segment, cloud_distance.max())
        guides[segment] = {"xy": display_xy, "t": display_t,
            "max_cloud_distance": float(cloud_distance.max()),
            "p95_cloud_distance": float(np.quantile(cloud_distance, .95)),
            "length": float(np.linalg.norm(np.diff(display_xy, axis=0), axis=1).sum())}
    assert np.array_equal(guides["T"]["xy"][-1], guides["AC"]["xy"][0])
    assert np.array_equal(guides["T"]["xy"][-1], guides["AB"]["xy"][0])
    for segment in ["AC", "AB"]:
        assert guides["T"]["length"] + guides[segment]["length"] > 10
        assert guides["T"]["xy"][0, 1] - guides[segment]["xy"][-1, 1] > 8
    # Extend the shared display stem from observed E9.5 CVG cells; show the UMAP gap as dashed.
    e9 = np.flatnonzero((stages == "E9.5") & (cell_segments == "CVG"))
    earliest = e9[np.isclose(native_t[e9], native_t[e9].min(), atol=1e-12)]
    start_index = earliest[np.argmin(np.linalg.norm(xy[earliest]-np.median(xy[earliest], axis=0), axis=1))]
    assert len(e9) == 11 and native_t[start_index] == 0

    def segment_bins(name):
        members = np.flatnonzero(cell_segments == name)
        edge = native_edges[saved_segment_order.index(name)]
        edge = edge[np.isfinite(edge)]
        return members, np.searchsorted(edge[1:-1], native_t[members], side="left")

    def median_cell(indices):
        center = np.median(xy[indices], axis=0)
        return indices[np.argmin(np.linalg.norm(xy[indices]-center, axis=1))]

    cvg_members, cvg_bins = segment_bins("CVG")
    t_members, t_bins = segment_bins("T")
    transition = t_members[t_bins == 1]
    x_order = transition[np.argsort(xy[transition, 0])]
    split = int(np.diff(xy[x_order, 0]).argmax())+1
    main_transition, island_transition = x_order[:split], x_order[split:]
    visual_split_x = float((xy[x_order[split-1], 0]+xy[x_order[split], 0])/2)
    assert len(main_transition) > 5 and len(island_transition) > 5
    pair_distance = np.linalg.norm(xy[island_transition, None]-xy[main_transition][None], axis=2)
    ordered = native_t[island_transition, None] < native_t[main_transition][None]
    pair_distance[~ordered] = np.inf
    pair = np.unravel_index(pair_distance.argmin(), pair_distance.shape)
    island_boundary, main_boundary = island_transition[pair[0]], main_transition[pair[1]]
    island_indices = np.array([start_index,
        median_cell(cvg_members[cvg_bins == 1]), median_cell(cvg_members[cvg_bins == 2]),
        median_cell(t_members[t_bins == 0]), median_cell(island_transition), island_boundary])
    stem_start_index = int(np.flatnonzero(cells == guide_controls["T"][0]["cell_id"])[0])
    lead_indices = np.array([main_boundary, median_cell(main_transition), stem_start_index])
    all_early_indices = np.r_[island_indices, lead_indices]
    assert np.all(np.diff(native_t[all_early_indices]) > 0)

    def early_curve(indices):
        times = native_t[indices]
        controls = xy[indices].astype(float)
        grid = np.linspace(times[0], times[-1], curve_samples)
        weights = np.ones(len(indices)); weights[[0, -1]] = 1000
        result = np.column_stack([UnivariateSpline(times, controls[:, j], w=weights,
            k=min(3, len(indices)-1), s=len(indices)*curve_smoothing_rms**2)(grid) for j in range(2)])
        progress = (grid-grid[0])/(grid[-1]-grid[0])
        result += ((1-progress[:, None])*(controls[0]-result[0])
                   + progress[:, None]*(controls[-1]-result[-1]))
        result[[0, -1]] = controls[[0, -1]]
        return result

    early_island = early_curve(island_indices)
    early_main = early_curve(lead_indices)
    gap_curve = np.linspace(xy[island_boundary], xy[main_boundary], 80)
    early_support = np.isin(cell_segments, ["CVG", "T"])
    early_distance_tree = cKDTree(xy[early_support])
    island_max_distance = float(early_distance_tree.query(early_island)[0].max())
    lead_max_distance = float(early_distance_tree.query(early_main)[0].max())
    gap_max_distance = float(early_distance_tree.query(gap_curve)[0].max())
    assert island_max_distance < curve_cloud_max_distance and lead_max_distance < curve_cloud_max_distance
    assert gap_max_distance > curve_cloud_max_distance, "The displayed gap must be distinguished from a supported solid curve"
    assert np.array_equal(early_main[-1], guides["T"]["xy"][0])
    early_records = [{"cell_id": str(cells[i]), "xy": xy[i].astype(float).tolist(),
                      "native_t": float(native_t[i]), "stage": str(stages[i]), "segment": str(cell_segments[i])}
                     for i in all_early_indices]

    candidates = pd.read_csv(candidate_path, sep="\t")
    orthologs = pd.read_csv(ortholog_path, sep="\t")
    assert len(candidates) == 6 and set(candidates.stable_id) == set(human_ids)
    assert orthologs.mouse_id.notna().sum() == 5
    support = np.nanmax(candidates[["Cortex_H4", "Spinal_H4", "Tibial_H4"]].to_numpy(float), axis=1)
    assert np.all(support >= .8)
    center = (xy.min(axis=0) + xy.max(axis=0)) / 2
    span = max(float(np.ptp(xy, axis=0).max()) * 1.12, 1.0)
    xlim, ylim = (center[0]-span/2, center[0]+span/2), (center[1]-span/2, center[1]+span/2)

    #3. =========== One descriptive state summary; balance libraries within source, then sources
    # Use every measured cell (no bin cutoff); missing source features are never filled with zero.
    n_states, n_genes = len(state_order), len(mouse)
    means, fractions = np.full((n_states, n_genes), np.nan), np.full((n_states, n_genes), np.nan)
    measurement_coverage = np.zeros((n_states, n_genes))
    detected_cells = np.zeros((n_states, n_genes), dtype=int)
    measured_cells = np.zeros((n_states, n_genes), dtype=int)
    source_counts, library_counts = np.zeros((n_states, n_genes), int), np.zeros((n_states, n_genes), int)
    total_cells = np.array([(states == s).sum() for s in state_order])
    for si, state in enumerate(state_order):
        state_mask = states == state
        for gi in range(n_genes):
            take = state_mask & measured[:, gi]
            measured_cells[si, gi] = take.sum()
            detected_cells[si, gi] = (counts[take, gi] > 0).sum()
            measurement_coverage[si, gi] = take.sum() / state_mask.sum()
            source_values, source_detection = [], []
            for src in pd.unique(sources[take]):
                local = take & (sources == src)
                library_values, library_detection = [], []
                for lib in pd.unique(libraries[local]):
                    valid = local & (libraries == lib)
                    assert valid.sum() >= 1
                    library_values.append(float(expression[valid, gi].mean()))
                    library_detection.append(float((counts[valid, gi] > 0).mean()))
                source_values.append(float(np.mean(library_values)))
                source_detection.append(float(np.mean(library_detection)))
                library_counts[si, gi] += len(library_values)
            source_counts[si, gi] = len(source_values)
            if len(source_values):
                means[si, gi], fractions[si, gi] = np.mean(source_values), np.mean(source_detection)
    zscores = np.full_like(means, np.nan)
    for gi in range(n_genes):
        valid = np.isfinite(means[:, gi])
        values = means[valid, gi]
        sd = float(values.std(ddof=0))
        zscores[valid, gi] = (values-values.mean()) / sd if sd > 0 else 0.0
    assert np.isfinite(means).all() and (fractions >= 0).all() and (fractions <= 1).all()
    assert np.isclose(zscores.mean(axis=0), 0).all() and np.isclose(zscores.std(axis=0), 1).all()
    assert (measurement_coverage[:, :4] == 1).all()
    assert measured[:, 4].sum() == 3410 and (counts[measured[:, 4], 4] > 0).sum() == 10

    #4. =========== Retained B/C canvas; the independent workflow is drawn separately
    fig = plt.figure(figsize=figure_size, dpi=preview_dpi, facecolor="white")
    width, height = figure_size
    def text(x, y, label, **kwargs):
        return fig.text(x/width, y/height, label, **kwargs)

    #5. =========== B: the user's gray/red template on the exact whole-period UMAP
    text(.23, 6.53, "B", fontsize=17, fontweight="bold")
    text(.67, 6.53, "Whole-period expression", fontsize=12, fontweight="bold")
    feature_positions = [(.40, 3.55), (3.18, 3.55), (5.96, 3.55), (.40, .60), (3.18, .60)]
    feature_axes, feature_records, arrow_artists = [], [], []
    feature_width = 2.58
    for gi, (gene, (left, bottom)) in enumerate(zip(mouse, feature_positions)):
        ax = fig.add_axes([left/width, bottom/height, feature_width/width, feature_width/height])
        feature_axes.append(ax)
        zero = measured[:, gi] & (counts[:, gi] == 0)
        missing = ~measured[:, gi]
        positive = np.flatnonzero(measured[:, gi] & (counts[:, gi] > 0))
        positive = positive[np.argsort(expression[positive, gi], kind="stable")]
        ax.scatter(xy[zero, 0], xy[zero, 1], s=feature_area, c=zero_color, alpha=.9, linewidths=0)
        if missing.any():
            ax.scatter(xy[missing, 0], xy[missing, 1], s=feature_area, facecolors="none",
                       edgecolors=missing_color, linewidths=.16, alpha=.60)
        dots = ax.scatter(xy[positive, 0], xy[positive, 1], s=feature_area,
            c=expression[positive, gi], cmap=feature_cmap, norm=feature_norm,
            alpha=feature_alpha, edgecolors="none", linewidths=0)
        assert np.array_equal(dots.get_array(), expression[positive, gi])
        assert np.array_equal(dots.get_offsets(), xy[positive])
        assert zero.sum() + missing.sum() + len(positive) == 8340
        ax.set(xlim=xlim, ylim=ylim)
        ax.set_box_aspect(1); ax.axis("off")
        ax.set_title(gene, fontsize=12, fontstyle="italic", pad=5)
        # One shared stem, then two long branch curves with one arrowhead per branch.
        halo = [pe.Stroke(linewidth=1.45, foreground="white", alpha=.9), pe.Normal()]
        shared = guides["T"]["xy"]
        for supported in [early_island, early_main, shared]:
            ax.plot(supported[:, 0], supported[:, 1], color="#363636", lw=.75, zorder=5, path_effects=halo)
        ax.plot(gap_curve[:, 0], gap_curve[:, 1], color="#555555", lw=.75,
                linestyle=(0, (2.4, 2.0)), zorder=5, path_effects=halo)
        ax.scatter([xy[start_index, 0]], [xy[start_index, 1]], s=10, facecolors="white",
                   edgecolors="#363636", linewidths=.65, zorder=7)
        ax.text(16.6, 8.4, "E9.5 CVG", ha="right", va="center", fontsize=7.0, color=ink, zorder=7)
        for segment, (label, label_xy) in curve_labels.items():
            vertices = guides[segment]["xy"]
            codes = np.r_[MplPath.MOVETO, np.full(len(vertices)-1, MplPath.LINETO)]
            arrow = FancyArrowPatch(path=MplPath(vertices, codes), arrowstyle="-|>", mutation_scale=8.5,
                                    lw=.75, color="#363636", zorder=5, path_effects=halo)
            ax.add_patch(arrow)
            arrow_artists.append(arrow)
            ax.text(*label_xy, label, fontsize=8, ha="left", va="center", color=ink,
                    bbox={"facecolor": "white", "edgecolor": "none", "alpha": .82, "pad": .4}, zorder=6)
        if gene == "Cyp26a1":
            ax.text(.5, .985, "10 / 3,410 measured cells detected", ha="center", va="top",
                    fontsize=7.2, color=muted, transform=ax.transAxes)
        feature_records.append({"gene": gene, "positive_cells": len(positive), "zero_cells": int(zero.sum()),
                                "unmeasured_cells": int(missing.sum()), "same_cell_coordinates": True})
    # Use the unused sixth tile for a single shared legend rather than another plot.
    text(6.18, 2.55, "Expression", fontsize=9.5)
    cax = fig.add_axes([6.18/width, 2.27/height, 1.90/width, .10/height])
    vector_colorbar(cax, feature_cmap, feature_norm, [0, 1, 2, 3, 4], tick_size=7)
    text(6.18, 1.92, "log10(1 + CPM)", fontsize=8, color=muted)
    legend_axis = fig.add_axes([0, 0, 1, 1], zorder=-1)
    legend_axis.set(xlim=(0, width), ylim=(0, height)); legend_axis.axis("off")
    legend_axis.scatter([6.24], [1.61], s=13, c=zero_color, linewidths=0)
    text(6.40, 1.57, "Measured zero", fontsize=8, color=muted)
    legend_axis.scatter([6.24], [1.31], s=13, facecolors="none", edgecolors=missing_color, linewidths=.55)
    text(6.40, 1.27, "Not measured", fontsize=8, color=muted)
    text(6.18, .82, "Shared nonlinear color scale\nfor all five genes", fontsize=8, color=muted, linespacing=1.45)

    #6. =========== C: compact candidate-by-state DotPlot; one global gene-wise scaling
    text(8.90, 6.53, "C", fontsize=17, fontweight="bold")
    text(9.25, 6.53, "Developmental states", fontsize=12, fontweight="bold")
    dot_diameter = np.sqrt(dot_max_area * float(fractions.max()))
    pitch = dot_diameter * (1+column_gap_ratio) / 72
    matrix_width, matrix_height = n_genes*pitch, n_states*pitch
    matrix_left, matrix_bottom = 11.80, 1.08
    ax = fig.add_axes([matrix_left/width, matrix_bottom/height, matrix_width/width, matrix_height/height])
    xx, yy = np.meshgrid(np.arange(n_genes), np.arange(n_states))
    valid = np.isfinite(fractions) & np.isfinite(zscores)
    visible = valid & (fractions > 0)
    dots = ax.scatter(xx[visible], yy[visible], s=dot_max_area*fractions[visible], c=zscores[visible],
                      cmap=dot_cmap, norm=dot_norm, edgecolors="none", linewidths=0)
    no_data = ~valid
    if no_data.any():
        ax.scatter(xx[no_data], yy[no_data], marker="x", s=dot_max_area*.2, c=missing_color, linewidths=.6)
    ax.set(xlim=(-.5, n_genes-.5), ylim=(n_states-.5, -.5))
    labels = [g + (" †" if gi == 4 else "") for gi, g in enumerate(mouse)]
    ax.set_xticks(range(n_genes), labels, rotation=55, ha="right", fontsize=gene_font, fontstyle="italic")
    ax.set_yticks(range(n_states), state_order, fontsize=state_font)
    for state_label_artist in ax.get_yticklabels():
        state_label_artist.set_verticalalignment("center")
    ax.tick_params(length=0, pad=5)
    stage_artists = []
    for si, stage_label in enumerate(stage_labels):
        row_y = matrix_bottom + matrix_height * (1 - (si + .5) / n_states)
        stage_artists.append(text(stage_left, row_y, stage_label, fontsize=stage_font,
                                  ha="left", va="center", color=muted))
    text(stage_left, 5.66, "Stage(s)", fontsize=7.5, color=muted)
    text(10.65, 5.66, "Cell state", fontsize=7.5, color=muted, ha="center")
    for spine in ax.spines.values(): spine.set_visible(False)
    ax.plot([-.42, n_genes-.58], [-.95, -.95], lw=1.6, color="#952E7D", clip_on=False)
    ax.text((n_genes-1)/2, -1.35, "Hearing-loss candidates", fontsize=7.5, color=muted,
            ha="center", va="bottom", clip_on=False)
    # Legends remain above the matrix and are not repeated for each gene.
    dca = fig.add_axes([matrix_left/width, 6.05/height, matrix_width/width, .075/height])
    vector_colorbar(dca, dot_cmap, dot_norm, [-1, -.5, 0, .5, 1], tick_size=6.2, decimals=1)
    dca.set_title("Avg. Exp. (scaled)", fontsize=7.5, loc="left", pad=6)
    text(9.05, 6.27, "% Exp.", fontsize=7.5)
    for x, fraction in zip(np.linspace(9.15, 10.65, 5), [0, .25, .5, .75, 1]):
        if fraction > 0:
            legend_axis.scatter([x], [6.07], s=dot_max_area*fraction, c="#555555", edgecolors="none", linewidths=0)
        text(x, 5.82, f"{int(fraction*100)}", fontsize=7, ha="center", color=muted)
    text(9.05, .41, "† Sparse; GSE195500 feature absent.", fontsize=7.2, color=muted)
    text(.40, .17, "Native RNA expression; shared nonlinear color scale. Arrows display the accepted embryonic tree; dashed line spans the UMAP gap.", fontsize=6.5, color=muted)
    text(8.95, .16, "Equal source/library weights; gene-wise z-scores; measured cells only.", fontsize=6.1, color=muted)

    #7. =========== Verify displayed points, physical spacing, text and frozen inputs
    def check_layout(dpi):
        fig.set_dpi(dpi); fig.canvas.draw()
        renderer, canvas = fig.canvas.get_renderer(), fig.bbox
        clipped = []
        for item in fig.findobj(match=lambda item: isinstance(item, Text)):
            if not item.get_visible() or not item.get_text(): continue
            box = item.get_window_extent(renderer)
            if box.width and box.height and (box.x0 < canvas.x0-1 or box.y0 < canvas.y0-1 or box.x1 > canvas.x1+1 or box.y1 > canvas.y1+1):
                clipped.append(item.get_text())
        assert not clipped, "Clipped text: " + repr(clipped)
        a_boxes = [(item.get_text(), item.get_window_extent(renderer)) for item in fig.texts
                   if item.get_visible() and item.get_position()[1]*height > 6.85]
        a_overlaps = [(name_a,name_b) for i,(name_a,box_a) in enumerate(a_boxes)
                      for name_b,box_b in a_boxes[i+1:] if box_a.overlaps(box_b)]
        assert not a_overlaps, "Overlapping workflow labels: " + repr(a_overlaps)
        for feature in feature_axes:
            extent = feature.get_window_extent(renderer)
            assert np.isclose(extent.width, extent.height), "Feature axes are not square"
            assert np.allclose(feature.get_xlim(), xlim) and np.allclose(feature.get_ylim(), ylim)
        assert np.allclose(dots.get_sizes(), dot_max_area*fractions[visible])
        assert np.allclose(dots.get_array(), zscores[visible])
        diameter_px = dots.get_paths()[0].get_extents().width * dots.get_transforms()[:, 0, 0].max()
        centers = ax.transData.transform([[0, 0], [1, 0], [0, 1]])
        gap_ratios = np.array([abs(centers[1, 0]-centers[0, 0]), abs(centers[2, 1]-centers[0, 1])])/diameter_px-1
        assert np.allclose(gap_ratios, column_gap_ratio, atol=.005), gap_ratios
        row_boxes = [t.get_window_extent(renderer) for t in ax.get_yticklabels()]
        assert not any(a.overlaps(b) for i, a in enumerate(row_boxes) for b in row_boxes[i+1:])
        for stage_artist, row_box in zip(stage_artists, row_boxes):
            stage_box = stage_artist.get_window_extent(renderer)
            assert stage_box.x1 + 3 < row_box.x0, "A stage overlaps its state label"
            assert np.isclose((stage_box.y0 + stage_box.y1)/2, (row_box.y0 + row_box.y1)/2, atol=2)
        for feature in feature_axes:
            owned = [item for item in arrow_artists if item.axes is feature]
            assert len(owned) == 2
            for item in owned:
                assert len(item.get_path().vertices) > 100, "A short callout remains in place of a long curve"
        # The state labels must remain outside the feature-map tiles.
        assert min(box.x0 for box in row_boxes) > feature_axes[2].get_window_extent(renderer).x1 + 2
        return {"dpi": dpi, "all_text_inside_canvas": True, "feature_axes_square": True,
                "dot_gap_ratio_xy": gap_ratios.tolist(), "state_labels_clear_of_features": True,
                "stage_state_row_alignment": True, "curved_branch_arrows_per_feature": 2,
                "canvas_pixels": [int(round(canvas.width)), int(round(canvas.height))]}
    layout_checks = [check_layout(preview_dpi)]
    assert input_hashes == {p.name: digest(p) for p in input_paths}
    assert np.allclose(means, source["figure"]["C"]["mean_log10_1p_CPM"], atol=1e-12)
    assert np.allclose(fractions, source["figure"]["C"]["mean_detection_fraction"], atol=1e-12)
    assert np.allclose(zscores, source["figure"]["C"]["gene_wise_zscore"], atol=1e-12)
    return dict(fig=fig, candidates=candidates, orthologs=orthologs, xy=xy, source=source,
        input_hashes=input_hashes, feature_axes=feature_axes, state_order=state_order, mouse=mouse,
        human_ids=human_ids, cells=cells, means=means, fractions=fractions, zscores=zscores,
        measurement_coverage=measurement_coverage, native_t=native_t, stages=stages,
        stage_sets=stage_sets, stage_labels=stage_labels, guides=guides, guide_controls=guide_controls,
        feature_records=feature_records, layout_checks=layout_checks,
        crops={"B":(.20,.05,8.50,6.75), "C":(8.78,.05,4.72,6.75)})
```

## Panel A design references
- [Hamel et al., Nature Communications 2024, Fig. 1b and 1d](https://www.nature.com/articles/s41467-023-44380-y/figures/1): genetic colocalization followed by single-cell expression mapping.
- [Haglund et al., Nature Genetics 2025, Fig. 1a](https://www.nature.com/articles/s41588-024-02050-9/figures/1): modular genomic workflow and aligned association tracks.
- [Official 2025 journal metrics](https://www.nature.com/nature-portfolio/about-journals/journal-metrics): JIF 18.1 and 25.5, respectively.

The workflow drawing is original; schematic variant/association marks are not claimed as actual locus results. MR, Hi-C and cell-type enrichment methods from those papers are not added to this project.

## Original standalone source: fig4_Integration_genes.py

This complete existing program retains its original command-line entry point and runtime assumptions. Use it in its original independent script context. Executing the cell in Jupyter would invoke the unchanged entry point with Jupyter arguments; this section is not a new notebook dispatcher.


```python
"""Pure M25/MGI identity functions used by the Figure 4 native RNA workflow.

The human-genetics notebook extracts read_reference, read_alias_evidence, and
map_features without executing other analysis workflows. Their original mapping
logic and fixed reference checks are retained. The two primitive validation/hash
functions are copied from Integration-inputs.py for standalone library use.
"""
from __future__ import annotations

import csv
import gzip
import hashlib
import json
import re
from collections import defaultdict
from functools import lru_cache
from pathlib import Path
import pandas as pd

# Set each reference-file placeholder independently before calling the functions.
ROOT = Path(".")
# Set gencode.vM25.primary_assembly.annotation.gtf.gz, the fixed GENCODE M25 primary-assembly reference.
GTF = ROOT
GTF_SHA256 = "b3ceeef170048fc86c5dead33ba7af61d71e50c9a19ca5b4b9784b830802f5cd"
# Set MRK_List1.rpt.gz, the fixed MGI gene-symbol and synonym report snapshot.
MGI = ROOT
MGI_SHA256 = "714e4390ae85807d4640802ab7303ebc4040d8b4e5a83f2b47e0be49c64b762f"


def require(condition, message):
    if not condition:
        raise ValueError(message)

@lru_cache(maxsize=32)
def _file_sha256(path, size, mtime_ns):
    digest = hashlib.sha256()
    with Path(path).open("rb") as stream:
        for chunk in iter(lambda: stream.read(1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()


def file_sha256(path):
    path = Path(path).resolve()
    stat = path.stat()
    return _file_sha256(str(path), stat.st_size, stat.st_mtime_ns)

def read_reference():
    require(file_sha256(GTF) == GTF_SHA256, "M25 reference fingerprint changed.")
    records = []
    with gzip.open(GTF, "rt", encoding="utf-8") as stream:
        for line in stream:
            if line.startswith("#"):
                continue
            fields = line.rstrip("\n").split("\t")
            if fields[2] != "gene":
                continue
            attrs = dict(re.findall(r'(\w+) "([^"]+)"', fields[8]))
            match = re.fullmatch(r"(ENSMUSG\d{11})\.(\d+)", attrs["gene_id"])
            require(match is not None, "Unexpected reference gene ID syntax.")
            records.append({"target_id": match[1], "target_version": int(match[2]), "target_versioned_id": attrs["gene_id"],
                "target_symbol": attrs["gene_name"], "target_biotype": attrs["gene_type"], "chromosome": fields[0],
                "start": int(fields[3]), "end": int(fields[4]), "strand": fields[6], "mgi_id": attrs.get("mgi_id", "")})
    ref = pd.DataFrame(records).set_index("target_id", drop=False)
    require(ref.index.is_unique and len(ref) == 55487, "Unexpected M25 gene identity axis.")
    ref["target_excluded"] = ref.chromosome.eq("chrM") | ref.target_symbol.str.casefold().str.startswith(("hba", "hbb"))
    require(int(ref.target_excluded.sum()) == 49, "Reference exclusion classes changed.")
    symbols = ref.groupby("target_symbol", sort=False).target_id.agg(lambda x: tuple(sorted(x))).to_dict()
    return ref, symbols

def read_alias_evidence(queries, ref):
    """Read only matching MGI rows; keep every candidate and redirect evidence."""
    require(file_sha256(MGI) == MGI_SHA256, "MGI snapshot changed.")
    bridges = ref.loc[ref.mgi_id.ne("")].groupby("mgi_id").target_id.agg(lambda x: tuple(sorted(x))).to_dict()
    evidence = defaultdict(list)
    statuses = defaultdict(int)
    with gzip.open(MGI, "rt", encoding="utf-8", newline="") as stream:
        reader = csv.reader(stream, delimiter="\t")
        header = next(reader)
        require(len(header) == 14 and "Synonyms" in header[11], "Unexpected MGI report format.")
        for row_number, row in enumerate(reader, 2):
            require(len(row) == 14, "Malformed MGI row.")
            statuses[row[7]] += 1
            terms = {row[6]: "approved_symbol" if row[7] != "W" else "withdrawn_symbol"}
            for synonym in row[11].split("|"):
                if synonym:
                    terms.setdefault(synonym, "synonym")
            matched = queries.intersection(terms)
            if not matched:
                continue
            mgi_text = row[12] if row[7] == "W" else row[0]
            ids = tuple(sorted(set(re.findall(r"MGI:\d+", mgi_text))))
            targets = tuple(sorted({target for mgi_id in ids for target in bridges.get(mgi_id, ())}))
            for term in matched:
                evidence[term].append({"report_row": row_number, "match_kind": terms[term], "recorded_symbol": row[6], "status": row[7],
                    "recorded_mgi_id": row[0], "current_mgi_ids": ids, "current_symbol": row[13] if row[7] == "W" else row[6], "m25_targets": targets})
    # Set reference_manifest.json for the same frozen MGI snapshot; its hash, bytes, columns, and row count are checked below.
    manifest = json.loads((ROOT).read_text(encoding="utf-8"))
    require(manifest["sha256"] == MGI_SHA256 and manifest["bytes"] == MGI.stat().st_size and manifest["columns"] == header and manifest["rows"] == sum(statuses.values()), "MGI snapshot metadata differs from parsed data.")
    return dict(evidence), dict(statuses)

def map_features(names, id_type, ref, symbols, aliases):
    rows = []
    for name in names:
        if id_type == "ensembl_gene_id_versioned":
            match = re.fullmatch(r"(ENSMUSG\d{11})\.(\d+)", name)
            require(match is not None, "Non-Ensembl source value passed as an Ensembl ID.")
            base, version = match[1], int(match[2])
            exact = (base,) if base in ref.index and ref.at[base, "target_version"] == version else ()
            status = "exact_id_version" if exact else ("stable_id_version_difference" if base in ref.index else "old_id_absent_from_target")
        else:
            exact = symbols.get(name, ())
            status = "exact_symbol_unique_target" if len(exact) == 1 else ("exact_symbol_ambiguous_target" if exact else "unmapped_exact")
        hits = aliases.get(name, []) if id_type == "gene_symbol" else []
        mgi_ids = tuple(sorted({m for hit in hits for m in hit["current_mgi_ids"]}))
        alias_targets = tuple(sorted({g for hit in hits for g in hit["m25_targets"]}))
        if not hits:
            alias_status = "no_alias_evidence"
        elif not mgi_ids:
            alias_status = "withdrawn_without_explicit_successor" if all(h["status"] == "W" for h in hits) else "mgi_record_without_valid_id"
        elif len(mgi_ids) > 1:
            alias_status = "alias_ambiguous_mgi"
        elif not alias_targets:
            alias_status = "alias_mgi_absent_from_M25_PRI"
        elif len(alias_targets) > 1:
            alias_status = "alias_multiple_target"
        else:
            alias_status = "withdrawn_redirect_proposed" if any(h["status"] == "W" for h in hits) else "alias_unique_proposed"
        strict = exact[0] if len(exact) == 1 else ""
        proposal = alias_targets[0] if not exact and len(mgi_ids) == 1 and len(alias_targets) == 1 else ""
        suffix = bool(id_type == "gene_symbol" and re.search(r"\.\d+$", name))
        rows.append({"source_feature": name, "mapping_status": status, "exact_candidates": exact, "n_exact_candidates": len(exact),
            "strict_target": strict, "alias_status": alias_status, "alias_mgi_ids": mgi_ids, "alias_candidates": alias_targets,
            "alias_proposal": proposal, "alias_disagrees_with_exact": bool(strict and alias_targets and strict not in alias_targets),
            "symbol_numeric_suffix_preserved": suffix, "alias_evidence": hits})
    table = pd.DataFrame(rows).set_index("source_feature", drop=False)
    table["unique_proposed_target"] = table.strict_target.where(table.strict_target.ne(""), table.alias_proposal)
    multiplicity = table.loc[table.unique_proposed_target.ne(""), "unique_proposed_target"].value_counts()
    table["many_source_to_one_target"] = table.unique_proposed_target.map(multiplicity).fillna(0).gt(1)
    table["strict_target_excluded"] = table.strict_target.map(ref.target_excluded).eq(True)
    return table
```

## Original standalone source: fig4_Integration_axis.py

```python
"""Exact mapping decisions used by the Figure 4 native RNA workflow.

resolve_mapping is copied from Integration-axis.py without changing its
parameters, identity rules, collision handling, or returned metadata fields.
The human-genetics notebook extracts this function into its existing namespace.
"""
from __future__ import annotations

import pandas as pd
# The companion released module supplies the original pure mapping and validation functions.
from fig4_Integration_genes import map_features, require

GENES = {"map_features": map_features}

def resolve_mapping(names, id_type, ref, symbols, aliases):
    table = GENES["map_features"](names, id_type, ref, symbols, aliases)
    records = []
    for row in table.itertuples():
        current = [h for h in row.alias_evidence if h["status"] == "O" and h["match_kind"] == "approved_symbol"]
        withdrawn = [h for h in row.alias_evidence if h["status"] == "W" and h["match_kind"] == "withdrawn_symbol"]
        current_ids = tuple(sorted({m for h in current for m in h["current_mgi_ids"]}))
        withdrawn_ids = tuple(sorted({m for h in withdrawn for m in h["current_mgi_ids"]}))
        expected_mgi = str(ref.at[row.strict_target, "mgi_id"]) if row.strict_target else ""
        accepted, reason = "", ""
        if row.n_exact_candidates > 1:
            reason = "excluded_multiple_exact_M25_models"
        elif not row.strict_target:
            reason = "excluded_alias_only_candidate" if row.alias_evidence else "excluded_no_exact_or_alias_evidence"
        elif row.strict_target_excluded:
            reason = "excluded_reference_feature_class"
        elif id_type == "ensembl_gene_id_versioned":
            accepted, reason = row.strict_target, "accepted_exact_Ensembl_version"
        elif current:
            if expected_mgi and current_ids == (expected_mgi,) and all(h["current_mgi_ids"] for h in current):
                accepted, reason = row.strict_target, "accepted_exact_current_official_agreement"
            else:
                reason = "excluded_current_official_identity_conflict"
        elif withdrawn:
            if withdrawn_ids and (not expected_mgi or withdrawn_ids != (expected_mgi,)):
                reason = "excluded_withdrawn_identity_conflict"
            elif not expected_mgi or any(not h["current_mgi_ids"] for h in withdrawn):
                reason = "excluded_withdrawn_successor_unanchored"
            else:
                accepted, reason = row.strict_target, "accepted_exact_withdrawn_agreement"
        else:
            accepted, reason = row.strict_target, "accepted_exact_no_official_contradiction"
        synonym_ids = {m for h in row.alias_evidence if h["match_kind"] == "synonym" for m in h["current_mgi_ids"]}
        records.append({"accepted_gene_id": accepted, "decision": reason, "M25_mgi_id": expected_mgi,
            "current_official_mgi_ids": current_ids, "withdrawn_redirect_mgi_ids": withdrawn_ids,
            "ordinary_synonym_overlap": bool(expected_mgi and synonym_ids - {expected_mgi}),
            "historical_withdrawn_overlap": bool(expected_mgi and set(withdrawn_ids) - {expected_mgi})})
    decision = pd.DataFrame(records, index=table.index)
    table = pd.concat([table, decision], axis=1)
    collision = table.accepted_gene_id.ne("") & table.accepted_gene_id.duplicated(keep=False)
    table.loc[collision, "accepted_gene_id"] = ""
    table.loc[collision, "decision"] = "excluded_accepted_target_collision"
    table["mapping_accepted"] = table.accepted_gene_id.ne("")
    table["canonical_gene_id"] = pd.Categorical(table.accepted_gene_id.replace("", None))
    table["source_reference_status"] = "verified_M25" if id_type == "ensembl_gene_id_versioned" else "unknown/not_verified"
    require(table.loc[table.mapping_accepted, "accepted_gene_id"].is_unique, "Accepted mappings are not one-to-one.")
    return table
```

```python
#!/usr/bin/env python3
"""Fail-closed validation for 600-DPI PNGs and editable figure PDFs.

The default PDF profile is intentionally strict. It is designed for figures that
must remain straightforward to edit after opening in Adobe Illustrator:

* no more than four explicit, nonzero-winding clipping paths per page;
* clipping paths are sibling, axis-aligned rectangles (never nested);
* every clipping path is explicitly scoped by q/Q and ends with W n;
* no raster images, soft/image masks, inline images, text clipping, shading,
  or uninspected recursive Form XObjects;
* live, embedded, Unicode-mapped Arial/SimSun text; and
* at least one painted vector path.

The program writes its report to stdout only. Poppler render files are created in
a task-local temporary directory and removed before exit unless the caller uses
``--render-dir``; that option always creates and retains a new unique child
directory, so existing files are never overwritten.
"""

from __future__ import annotations

import argparse
import math
import os
import re
import secrets
import shutil
import subprocess
import sys
import tempfile
from collections import Counter
from dataclasses import dataclass, field, replace
from pathlib import Path
from typing import Any, Iterable, Sequence

try:
    from PIL import Image
    from pypdf import PdfReader
    from pypdf.generic import ArrayObject, ContentStream, DictionaryObject, IndirectObject
except ImportError as exc:  # pragma: no cover - exercised only in a broken runtime
    print(f"[NOT CERTIFIED] Required Python dependency is unavailable: {exc}")
    raise SystemExit(1)


EPSILON = 1e-7
PATH_BUILDING = {b"m", b"l", b"c", b"v", b"y", b"h", b"re"}
PATH_ENDING = {b"S", b"s", b"f", b"F", b"f*", b"B", b"B*", b"b", b"b*", b"n"}
TEXT_SHOWING = {b"Tj", b"TJ", b"'", b'"'}
FILLING = {b"f", b"F", b"f*"}
STROKING = {b"S", b"s"}
FILL_STROKING = {b"B", b"B*", b"b", b"b*"}
INLINE_IMAGE_OPS = {b"BI", b"ID", b"EI", b"INLINE IMAGE"}
SUBSET_PREFIX = re.compile(r"^[A-Z]{6}\+")
CJK_RE = re.compile(
    "[\u3400-\u4dbf\u4e00-\u9fff\uf900-\ufaff\U00020000-\U0002fa1f]"
)
ASCII_ALNUM_RE = re.compile(r"[A-Za-z0-9]")


@dataclass
class Issue:
    code: str
    message: str
    outcome: str = "FAIL"


@dataclass
class ClipEvent:
    number: int
    scope: str
    depth: int
    bbox: tuple[float, float, float, float] | None
    kinds: Counter[str] = field(default_factory=Counter)


@dataclass
class FontUse:
    scope: str
    resource_name: str
    base_font: str
    subtype: str
    embedded: bool
    to_unicode: bool


@dataclass
class GraphicsState:
    ctm: tuple[float, float, float, float, float, float] = (1, 0, 0, 1, 0, 0)
    clip_depth: int = 0
    active_clips: tuple[int, ...] = ()
    font_resource: str | None = None
    font_object: Any = None
    text_render_mode: int = 0


@dataclass
class PdfPageStats:
    page_number: int
    crop_size_points: tuple[float, float] = (0.0, 0.0)
    media_size_points: tuple[float, float] = (0.0, 0.0)
    rotation: int = 0
    clip_events: list[ClipEvent] = field(default_factory=list)
    max_clip_depth: int = 0
    q_count: int = 0
    Q_count: int = 0
    form_calls: int = 0
    image_count: int = 0
    inline_image_count: int = 0
    text_show_count: int = 0
    text_object_count: int = 0
    vector_paint_count: int = 0
    vector_ops: Counter[str] = field(default_factory=Counter)
    font_uses: dict[tuple[str, str, str], FontUse] = field(default_factory=dict)


@dataclass
class ValidationResult:
    path: Path
    kind: str
    issues: list[Issue] = field(default_factory=list)
    details: list[str] = field(default_factory=list)

    @property
    def passed(self) -> bool:
        return not self.issues

    @property
    def status(self) -> str:
        if any(issue.outcome == "FAIL" for issue in self.issues):
            return "FAIL"
        if self.issues:
            return "NOT CERTIFIED"
        return "PASS"

    def fail(self, code: str, message: str) -> None:
        self.issues.append(Issue(code, message, "FAIL"))

    def not_certified(self, code: str, message: str) -> None:
        self.issues.append(Issue(code, message, "NOT CERTIFIED"))


def dereference(value: Any) -> Any:
    try:
        return value.get_object()
    except AttributeError:
        return value


def dictionary(value: Any) -> DictionaryObject | dict:
    value = dereference(value)
    return value if isinstance(value, (DictionaryObject, dict)) else {}


def numeric(value: Any) -> float:
    return float(dereference(value))


def normalize_font_name(value: Any) -> str:
    name = str(value or "").lstrip("/")
    return SUBSET_PREFIX.sub("", name)


def font_family(base_font: str) -> str | None:
    compact = re.sub(r"[^a-z0-9]", "", normalize_font_name(base_font).lower())
    if compact.startswith("arial"):
        return "Arial"
    if compact.startswith("simsun") or compact.startswith("nsimsun"):
        return "SimSun"
    return None


def font_descriptor(font_obj: Any) -> dict:
    font = dictionary(font_obj)
    subtype = str(font.get("/Subtype", ""))
    if subtype == "/Type0":
        descendants = dereference(font.get("/DescendantFonts", []))
        if descendants:
            descendant = dictionary(descendants[0])
            return dictionary(descendant.get("/FontDescriptor", {}))
    return dictionary(font.get("/FontDescriptor", {}))


def font_is_embedded(font_obj: Any) -> bool:
    descriptor = font_descriptor(font_obj)
    return any(key in descriptor for key in ("/FontFile", "/FontFile2", "/FontFile3"))


def object_identity(value: Any) -> tuple[Any, ...]:
    if isinstance(value, IndirectObject):
        return ("indirect", value.idnum, value.generation)
    obj = dereference(value)
    ref = getattr(obj, "indirect_reference", None)
    if isinstance(ref, IndirectObject):
        return ("indirect", ref.idnum, ref.generation)
    return ("direct", id(obj))


def matrix_multiply(
    first: tuple[float, float, float, float, float, float],
    second: tuple[float, float, float, float, float, float],
) -> tuple[float, float, float, float, float, float]:
    """Compose affine transforms so result(point) == first(second(point))."""
    a1, b1, c1, d1, e1, f1 = first
    a2, b2, c2, d2, e2, f2 = second
    return (
        a1 * a2 + c1 * b2,
        b1 * a2 + d1 * b2,
        a1 * c2 + c1 * d2,
        b1 * c2 + d1 * d2,
        a1 * e2 + c1 * f2 + e1,
        b1 * e2 + d1 * f2 + f1,
    )


def transform_point(
    matrix: tuple[float, float, float, float, float, float], x: float, y: float
) -> tuple[float, float]:
    a, b, c, d, e, f = matrix
    return (a * x + c * y + e, b * x + d * y + f)


def normalized_bbox(points: Sequence[tuple[float, float]]) -> tuple[float, float, float, float]:
    xs = [point[0] for point in points]
    ys = [point[1] for point in points]
    left, right = min(xs), max(xs)
    bottom, top = min(ys), max(ys)
    return (left, bottom, right - left, top - bottom)


def nearly_equal(left: float, right: float, tolerance: float) -> bool:
    return abs(left - right) <= tolerance


def boxes_match(
    observed: tuple[float, float, float, float],
    expected: tuple[float, float, float, float],
) -> bool:
    tolerance = max(0.5, max(abs(value) for value in expected[2:]) * 1e-4)
    return all(nearly_equal(a, b, tolerance) for a, b in zip(observed, expected))


def image_blank_reason(image: Image.Image) -> str | None:
    """Return a concrete blank-page reason without allocating a full RGBA copy."""
    extrema: Any = image.getextrema()
    if not isinstance(extrema, tuple):
        extrema = (extrema,)
    elif extrema and not isinstance(extrema[0], tuple):
        extrema = (extrema,)
    bands = image.getbands()
    if "A" in bands and extrema[bands.index("A")] == (0, 0):
        return "all pixels are fully transparent"
    if all(low == high for low, high in extrema):
        return "all pixels have one uniform color"
    return None


def parse_simple_rectangle(
    path_ops: Sequence[tuple[bytes, list[Any]]],
    ctm: tuple[float, float, float, float, float, float],
) -> tuple[tuple[float, float, float, float] | None, str | None]:
    if len(path_ops) == 1 and path_ops[0][0] == b"re":
        operands = path_ops[0][1]
        if len(operands) != 4:
            return None, "rectangle operator does not have four operands"
        x, y, width, height = (numeric(value) for value in operands)
        if abs(width) <= EPSILON or abs(height) <= EPSILON:
            return None, "rectangle has zero width or height"
        raw_points = [
            (x, y),
            (x + width, y),
            (x + width, y + height),
            (x, y + height),
        ]
    else:
        operators = [operator for operator, _ in path_ops]
        if not operators or operators[0] != b"m" or any(
            operator not in {b"m", b"l", b"h"} for operator in operators
        ):
            return None, "clip path is not one re rectangle or one m/l/h rectangle"
        if operators.count(b"m") != 1 or operators.count(b"h") > 1:
            return None, "clip path contains multiple subpaths"
        raw_points: list[tuple[float, float]] = []
        for operator, operands in path_ops:
            if operator in {b"m", b"l"}:
                if len(operands) != 2:
                    return None, "rectangle vertex does not have two coordinates"
                raw_points.append((numeric(operands[0]), numeric(operands[1])))
        if len(raw_points) == 5 and all(
            nearly_equal(a, b, 1e-6) for a, b in zip(raw_points[0], raw_points[-1])
        ):
            raw_points.pop()
        if len(raw_points) != 4 or (operators[-1] != b"h" and len(path_ops) != 5):
            return None, "m/l/h clip does not contain exactly four rectangle vertices"

    points = [transform_point(ctm, x, y) for x, y in raw_points]
    tolerance = max(1e-5, max(abs(value) for point in points for value in point) * 1e-7)
    for start, end in zip(points, points[1:] + points[:1]):
        dx, dy = abs(end[0] - start[0]), abs(end[1] - start[1])
        if not (dx <= tolerance < dy or dy <= tolerance < dx):
            return None, "clip rectangle is rotated, skewed, or self-degenerate"
    unique_x = []
    unique_y = []
    for x, y in points:
        if not any(nearly_equal(x, item, tolerance) for item in unique_x):
            unique_x.append(x)
        if not any(nearly_equal(y, item, tolerance) for item in unique_y):
            unique_y.append(y)
    if len(unique_x) != 2 or len(unique_y) != 2:
        return None, "clip path does not form one axis-aligned rectangle"
    return normalized_bbox(points), None


class PdfAuditor:
    def __init__(
        self,
        path: Path,
        result: ValidationResult,
        max_clips: int,
        expected_clips: int | None,
        declared_clip_boxes: list[tuple[float, float, float, float]],
        allow_raster_images: bool,
        allow_form_xobjects: bool,
        allow_no_text: bool,
        allow_no_vector: bool,
    ) -> None:
        self.path = path
        self.result = result
        self.max_clips = max_clips
        self.expected_clips = expected_clips
        self.declared_clip_boxes = declared_clip_boxes
        self.allow_raster_images = allow_raster_images
        self.allow_form_xobjects = allow_form_xobjects
        self.allow_no_text = allow_no_text
        self.allow_no_vector = allow_no_vector
        self.reader: PdfReader | None = None
        self._scanned_resources: set[tuple[Any, ...]] = set()
        self._reported_fonts: set[tuple[Any, ...]] = set()

    def audit(self) -> list[PdfPageStats]:
        try:
            self.reader = PdfReader(str(self.path), strict=True)
        except Exception as exc:
            self.result.not_certified("PDF_PARSE", f"PDF cannot be parsed strictly: {exc}")
            return []
        if self.reader.is_encrypted:
            self.result.not_certified("PDF_ENCRYPTED", "encrypted PDFs cannot be inspected")
            return []
        if not self.reader.pages:
            self.result.fail("PDF_EMPTY", "PDF contains no pages")
            return []

        page_stats: list[PdfPageStats] = []
        for page_number, page in enumerate(self.reader.pages, start=1):
            crop_box = page.cropbox
            media_box = page.mediabox
            user_unit = float(page.get("/UserUnit", 1) or 1)
            rotation = int(page.get("/Rotate", 0) or 0) % 360
            stats = PdfPageStats(
                page_number,
                crop_size_points=(
                    float(crop_box.width) * user_unit,
                    float(crop_box.height) * user_unit,
                ),
                media_size_points=(
                    float(media_box.width) * user_unit,
                    float(media_box.height) * user_unit,
                ),
                rotation=rotation,
            )
            page_stats.append(stats)
            resources = dictionary(page.get("/Resources", {}))
            page_boxes = [
                (
                    float(crop_box.left),
                    float(crop_box.bottom),
                    float(crop_box.width),
                    float(crop_box.height),
                ),
                (
                    float(media_box.left),
                    float(media_box.bottom),
                    float(media_box.width),
                    float(media_box.height),
                ),
            ]
            allowed_boxes = page_boxes + self.declared_clip_boxes
            contents = page.get_contents()
            if contents is None:
                self.result.fail("PDF_NO_CONTENT", f"page {page_number} has no content stream")
                continue
            try:
                content_stream = ContentStream(contents, self.reader)
            except Exception as exc:
                self.result.not_certified(
                    "CONTENT_PARSE", f"page {page_number} content stream cannot be parsed: {exc}"
                )
                continue
            self._scan_resources(resources, f"page {page_number}", stats)
            self._analyze_operations(
                content_stream.operations,
                resources,
                f"page {page_number}",
                stats,
                GraphicsState(),
                allowed_boxes,
                form_stack=set(),
                implicit_scope=False,
            )
            self._validate_page_text(page, stats)
            self._validate_page_summary(stats)
        return page_stats

    def _scan_resources(self, resources: dict, scope: str, stats: PdfPageStats) -> None:
        identity = object_identity(resources)
        if identity in self._scanned_resources:
            return
        self._scanned_resources.add(identity)

        ext_states = dictionary(resources.get("/ExtGState", {}))
        for name, reference in ext_states.items():
            ext_state = dictionary(reference)
            if "/SMask" in ext_state and str(dereference(ext_state["/SMask"])) != "/None":
                self.result.fail(
                    "SOFT_MASK", f"{scope} ExtGState {name} contains a non-None /SMask"
                )

        patterns = dictionary(resources.get("/Pattern", {}))
        if patterns:
            self.result.fail(
                "PATTERN_RESOURCE",
                f"{scope} contains Pattern resources; their Illustrator editability is not certified",
            )
        shadings = dictionary(resources.get("/Shading", {}))
        if shadings:
            self.result.fail(
                "SHADING_RESOURCE",
                f"{scope} contains Shading resources; their Illustrator editability is not certified",
            )

        xobjects = dictionary(resources.get("/XObject", {}))
        for name, reference in xobjects.items():
            xobject = dictionary(reference)
            subtype = str(xobject.get("/Subtype", ""))
            if "/SMask" in xobject:
                self.result.fail("SOFT_MASK", f"{scope} XObject {name} contains /SMask")
            if "/Mask" in xobject:
                self.result.fail("IMAGE_MASK", f"{scope} XObject {name} contains /Mask")
            if bool(xobject.get("/ImageMask", False)):
                self.result.fail("IMAGE_MASK", f"{scope} XObject {name} is an image mask")
            if subtype == "/Image":
                stats.image_count += 1
                if not self.allow_raster_images:
                    self.result.fail("RASTER_IMAGE", f"{scope} contains image XObject {name}")
            elif subtype == "/Form" and not self.allow_form_xobjects:
                self.result.fail(
                    "FORM_XOBJECT",
                    f"{scope} contains Form XObject {name}; use --allow-form-xobjects only for an intentionally reviewed Form workflow",
                )
            elif subtype not in {"/Form", ""}:
                self.result.fail(
                    "UNSUPPORTED_XOBJECT", f"{scope} contains unsupported XObject {name} ({subtype})"
                )

    def _record_clip_content(self, stats: PdfPageStats, state: GraphicsState, kind: str) -> None:
        for event_number in state.active_clips:
            stats.clip_events[event_number - 1].kinds[kind] += 1

    def _register_font(
        self, scope: str, resource_name: str, font_reference: Any, stats: PdfPageStats
    ) -> None:
        font = dictionary(font_reference)
        base_font = normalize_font_name(font.get("/BaseFont", ""))
        subtype = str(font.get("/Subtype", ""))
        embedded = font_is_embedded(font)
        to_unicode = "/ToUnicode" in font
        key = (scope, resource_name, base_font)
        stats.font_uses[key] = FontUse(
            scope, resource_name, base_font, subtype, embedded, to_unicode
        )
        identity = object_identity(font_reference)
        if identity in self._reported_fonts:
            return
        self._reported_fonts.add(identity)
        family = font_family(base_font)
        if family is None:
            self.result.fail(
                "FONT_FAMILY",
                f"{scope} font {resource_name} uses {base_font or '<unnamed>'}; only Arial and SimSun are allowed",
            )
        if not embedded:
            self.result.fail(
                "FONT_NOT_EMBEDDED",
                f"{scope} font {resource_name} ({base_font or '<unnamed>'}) is not embedded",
            )
        if not to_unicode:
            self.result.fail(
                "FONT_NO_TOUNICODE",
                f"{scope} font {resource_name} ({base_font or '<unnamed>'}) has no /ToUnicode map",
            )

    def _analyze_operations(
        self,
        operations: Sequence[tuple[list[Any], bytes]],
        resources: dict,
        scope: str,
        stats: PdfPageStats,
        initial_state: GraphicsState,
        allowed_boxes: list[tuple[float, float, float, float]],
        form_stack: set[tuple[Any, ...]],
        implicit_scope: bool,
    ) -> None:
        state = replace(initial_state)
        base_clip_depth = initial_state.clip_depth
        stack: list[GraphicsState] = []
        path_ops: list[tuple[bytes, list[Any]]] = []
        pending_clip = False
        in_text = False
        text_objects_at_start = stats.text_object_count

        for index, (operands, operator) in enumerate(operations):
            operands = list(operands)
            if operator == b"q":
                stats.q_count += 1
                stack.append(replace(state))
                continue
            if operator == b"Q":
                stats.Q_count += 1
                if not stack:
                    self.result.fail("Q_UNDERFLOW", f"{scope} has Q without a matching q at operation {index}")
                else:
                    state = stack.pop()
                path_ops = []
                pending_clip = False
                continue
            if operator == b"cm":
                if len(operands) != 6:
                    self.result.fail("BAD_CTM", f"{scope} has malformed cm at operation {index}")
                else:
                    matrix = tuple(numeric(value) for value in operands)
                    state.ctm = matrix_multiply(state.ctm, matrix)  # type: ignore[arg-type]
                continue

            if operator in PATH_BUILDING:
                path_ops.append((operator, operands))
                stats.vector_ops[operator.decode("ascii")] += 1
                continue
            if operator in {b"W", b"W*"}:
                if pending_clip:
                    self.result.fail(
                        "CLIP_PENDING", f"{scope} has consecutive clipping operators at operation {index}"
                    )
                if operator == b"W*":
                    self.result.fail(
                        "CLIP_EVEN_ODD",
                        f"{scope} uses W* at operation {index}; the strict profile requires nonzero-winding W",
                    )
                pending_clip = True
                continue
            if operator in PATH_ENDING:
                if pending_clip:
                    if operator != b"n":
                        self.result.fail(
                            "CLIP_PAINTED_PATH",
                            f"{scope} applies clipping with W/W* followed by {operator.decode('latin1')}; strict clipping must end with n",
                        )
                    bbox, reason = parse_simple_rectangle(path_ops, state.ctm)
                    event_number = len(stats.clip_events) + 1
                    if reason:
                        self.result.fail(
                            "CLIP_NOT_RECTANGLE",
                            f"{scope} clip {event_number} is invalid: {reason}",
                        )
                    elif bbox is not None and not any(
                        boxes_match(bbox, candidate) for candidate in allowed_boxes
                    ):
                        pretty = ", ".join(f"{value:.3f}" for value in bbox)
                        self.result.fail(
                            "CLIP_BOX_UNDECLARED",
                            f"{scope} clip {event_number} box ({pretty}) is neither the page box nor a declared --clip-box",
                        )
                    if not stack and not implicit_scope:
                        self.result.fail(
                            "CLIP_NOT_SCOPED",
                            f"{scope} clip {event_number} is not enclosed by an explicit q/Q scope",
                        )
                    state.clip_depth += 1
                    state.active_clips = state.active_clips + (event_number,)
                    stats.max_clip_depth = max(stats.max_clip_depth, state.clip_depth)
                    stats.clip_events.append(
                        ClipEvent(event_number, scope, state.clip_depth, bbox)
                    )
                    if state.clip_depth > 1:
                        self.result.fail(
                            "CLIP_NESTED",
                            f"{scope} clip {event_number} reaches effective clipping depth {state.clip_depth}",
                        )
                    pending_clip = False
                if operator in FILLING:
                    stats.vector_paint_count += 1
                    self._record_clip_content(stats, state, "vector-fill")
                elif operator in STROKING:
                    stats.vector_paint_count += 1
                    self._record_clip_content(stats, state, "vector-stroke")
                elif operator in FILL_STROKING:
                    stats.vector_paint_count += 1
                    self._record_clip_content(stats, state, "vector-fill+stroke")
                path_ops = []
                continue

            if operator == b"BT":
                if in_text:
                    self.result.fail("TEXT_NESTED", f"{scope} contains nested BT operators")
                in_text = True
                stats.text_object_count += 1
                continue
            if operator == b"ET":
                if not in_text:
                    self.result.fail("TEXT_ET_UNMATCHED", f"{scope} contains ET without BT")
                in_text = False
                continue
            if operator == b"Tr":
                try:
                    state.text_render_mode = int(numeric(operands[0]))
                except Exception:
                    self.result.fail("TEXT_RENDER_MODE", f"{scope} contains malformed Tr")
                    continue
                if state.text_render_mode in {4, 5, 6, 7}:
                    self.result.fail(
                        "TEXT_CLIPPING",
                        f"{scope} uses clipping text rendering mode {state.text_render_mode}",
                    )
                continue
            if operator == b"Tf":
                if len(operands) < 2:
                    self.result.fail("FONT_SELECT", f"{scope} contains malformed Tf")
                    continue
                resource_name = str(operands[0])
                fonts = dictionary(resources.get("/Font", {}))
                font_reference = fonts.get(resource_name)
                if font_reference is None:
                    self.result.fail(
                        "FONT_MISSING", f"{scope} selects missing font resource {resource_name}"
                    )
                    state.font_resource = None
                    state.font_object = None
                else:
                    state.font_resource = resource_name
                    state.font_object = font_reference
                continue
            if operator in TEXT_SHOWING:
                stats.text_show_count += 1
                if not in_text:
                    self.result.fail("TEXT_OUTSIDE_BT", f"{scope} shows text outside BT/ET")
                if state.text_render_mode in {3, 7}:
                    self.result.fail(
                        "INVISIBLE_TEXT", f"{scope} contains invisible text rendering mode {state.text_render_mode}"
                    )
                if state.text_render_mode in {4, 5, 6, 7}:
                    self.result.fail(
                        "TEXT_CLIPPING",
                        f"{scope} shows text in clipping rendering mode {state.text_render_mode}",
                    )
                if not state.font_resource or state.font_object is None:
                    self.result.fail("TEXT_NO_FONT", f"{scope} shows text without a resolved font")
                else:
                    self._register_font(scope, state.font_resource, state.font_object, stats)
                self._record_clip_content(stats, state, "text")
                continue

            if operator == b"Do":
                if not operands:
                    self.result.fail("XOBJECT_CALL", f"{scope} has malformed Do operator")
                    continue
                resource_name = str(operands[0])
                xobjects = dictionary(resources.get("/XObject", {}))
                reference = xobjects.get(resource_name)
                if reference is None:
                    self.result.fail(
                        "XOBJECT_MISSING", f"{scope} invokes missing XObject {resource_name}"
                    )
                    continue
                xobject = dictionary(reference)
                subtype = str(xobject.get("/Subtype", ""))
                if subtype == "/Image":
                    self._record_clip_content(stats, state, "image")
                    continue
                if subtype != "/Form":
                    self.result.fail(
                        "UNSUPPORTED_XOBJECT",
                        f"{scope} invokes unsupported XObject {resource_name} ({subtype})",
                    )
                    continue
                stats.form_calls += 1
                self._record_clip_content(stats, state, "form")
                identity = object_identity(reference)
                if identity in form_stack:
                    self.result.fail(
                        "FORM_RECURSION", f"{scope} has recursive Form XObject {resource_name}"
                    )
                    continue
                form_resources = dictionary(xobject.get("/Resources", resources))
                self._scan_resources(form_resources, f"{scope}/Form {resource_name}", stats)
                form_matrix_raw = dereference(xobject.get("/Matrix", [1, 0, 0, 1, 0, 0]))
                try:
                    form_matrix = tuple(numeric(value) for value in form_matrix_raw)
                    if len(form_matrix) != 6:
                        raise ValueError
                except Exception:
                    self.result.fail(
                        "FORM_MATRIX", f"{scope}/Form {resource_name} has malformed /Matrix"
                    )
                    form_matrix = (1, 0, 0, 1, 0, 0)
                form_state = replace(state)
                form_state.ctm = matrix_multiply(state.ctm, form_matrix)  # type: ignore[arg-type]
                try:
                    form_stream = ContentStream(dereference(reference), self.reader)
                except Exception as exc:
                    self.result.not_certified(
                        "FORM_PARSE", f"{scope}/Form {resource_name} cannot be parsed: {exc}"
                    )
                    continue
                self._analyze_operations(
                    form_stream.operations,
                    form_resources,
                    f"{scope}/Form {resource_name}",
                    stats,
                    form_state,
                    allowed_boxes,
                    form_stack | {identity},
                    implicit_scope=True,
                )
                continue

            if operator in INLINE_IMAGE_OPS:
                stats.inline_image_count += 1
                self.result.fail("INLINE_IMAGE", f"{scope} contains an inline image")
                self._record_clip_content(stats, state, "inline-image")
                continue
            if operator == b"sh":
                self.result.fail("SHADING", f"{scope} paints a shading object")
                self._record_clip_content(stats, state, "shading")
                continue

        if pending_clip:
            self.result.fail("CLIP_UNFINISHED", f"{scope} ends before its W/W* clip is applied")
        if path_ops:
            self.result.fail("PATH_UNFINISHED", f"{scope} ends with an unfinished path")
        if in_text:
            self.result.fail("TEXT_UNFINISHED", f"{scope} ends before ET")
        if stack:
            self.result.fail(
                "Q_UNBALANCED", f"{scope} has {len(stack)} q operator(s) without matching Q"
            )
        if state.clip_depth != base_clip_depth and not implicit_scope:
            self.result.fail(
                "CLIP_NOT_RESTORED",
                f"{scope} ends at clipping depth {state.clip_depth}, expected {base_clip_depth}",
            )
        if stats.text_object_count < text_objects_at_start:
            self.result.not_certified("INTERNAL_TEXT_COUNT", f"{scope} text-object accounting failed")

    def _validate_page_text(self, page: Any, stats: PdfPageStats) -> None:
        fragments: list[tuple[str, float, float, float, str]] = []
        parsed_font_names = {
            normalize_font_name(font.base_font) for font in stats.font_uses.values()
        }

        def visitor(text: str, cm: Sequence[float], tm: Sequence[float], font: Any, size: float) -> None:
            cleaned = text.strip()
            if not cleaned:
                return
            base_font = normalize_font_name(dictionary(font).get("/BaseFont", ""))
            try:
                x = float(cm[0]) * float(tm[4]) + float(cm[2]) * float(tm[5]) + float(cm[4])
                y = float(cm[1]) * float(tm[4]) + float(cm[3]) * float(tm[5]) + float(cm[5])
            except Exception:
                x = y = 0.0
            fragments.append((cleaned, x, y, float(size or 0), base_font))
            family = font_family(base_font)
            # pypdf can report a parent-page default font for text inside a Form
            # XObject. The content-stream pass above resolves the Form's actual
            # font. Apply script-specific checks only when the visitor font is a
            # font that the operation parser observed on a text-show operation.
            visitor_font_is_resolved = base_font in parsed_font_names
            if visitor_font_is_resolved and ASCII_ALNUM_RE.search(cleaned) and family != "Arial":
                self.result.fail(
                    "ASCII_FONT",
                    f"page {stats.page_number} text {cleaned!r} uses {base_font or '<unknown>'}, not Arial",
                )
            if visitor_font_is_resolved and CJK_RE.search(cleaned) and family != "SimSun":
                self.result.fail(
                    "CJK_FONT",
                    f"page {stats.page_number} Chinese text {cleaned!r} uses {base_font or '<unknown>'}, not SimSun",
                )

        try:
            page.extract_text(visitor_text=visitor)
        except Exception as exc:
            self.result.not_certified(
                "TEXT_EXTRACT", f"page {stats.page_number} live text cannot be decoded: {exc}"
            )
            return

        # Detect the common bad export in which a word or label becomes one text
        # object per character. Single tick labels are excluded by requiring a run.
        single_fragments = [
            item
            for item in fragments
            if len(item[0]) == 1 and (item[0].isalpha() or CJK_RE.search(item[0]))
        ]
        for index, current in enumerate(single_fragments):
            run = [current]
            for candidate in single_fragments[index + 1 :]:
                prior = run[-1]
                same_line = abs(candidate[2] - prior[2]) <= max(0.75, prior[3] * 0.35)
                same_font = candidate[4] == prior[4] and abs(candidate[3] - prior[3]) <= 0.25
                gap = candidate[1] - prior[1]
                nearby = -0.25 * max(prior[3], 1) <= gap <= 2.5 * max(prior[3], 1)
                if same_line and same_font and nearby:
                    run.append(candidate)
                    if len(run) >= 3:
                        word = "".join(item[0] for item in run)
                        self.result.fail(
                            "TEXT_FRAGMENTED",
                            f"page {stats.page_number} appears to split label {word!r} into individual character objects",
                        )
                        return
                elif candidate[2] < prior[2] - max(1.0, prior[3]):
                    break

    def _validate_page_summary(self, stats: PdfPageStats) -> None:
        clip_count = len(stats.clip_events)
        if clip_count > self.max_clips:
            self.result.fail(
                "TOO_MANY_CLIPS",
                f"page {stats.page_number} contains {clip_count} clipping paths; maximum is {self.max_clips}",
            )
        if self.expected_clips is not None and clip_count != self.expected_clips:
            self.result.fail(
                "CLIP_COUNT_EXPECTED",
                f"page {stats.page_number} contains {clip_count} clipping paths; expected exactly {self.expected_clips}",
            )
        if stats.max_clip_depth > 1:
            self.result.fail(
                "CLIP_DEPTH",
                f"page {stats.page_number} reaches clipping depth {stats.max_clip_depth}; maximum is 1",
            )
        for event in stats.clip_events:
            if not event.kinds:
                self.result.fail(
                    "EMPTY_CLIP",
                    f"page {stats.page_number} clip {event.number} contains no painted vector or live text content",
                )
        if stats.q_count != stats.Q_count:
            self.result.fail(
                "Q_COUNT",
                f"page {stats.page_number} recursively contains q={stats.q_count}, Q={stats.Q_count}",
            )
        if not self.allow_no_text and stats.text_show_count == 0:
            self.result.fail("NO_LIVE_TEXT", f"page {stats.page_number} contains no live text")
        if not self.allow_no_vector and stats.vector_paint_count == 0:
            self.result.fail("NO_VECTOR", f"page {stats.page_number} contains no painted vector path")


def validate_png(
    path: Path,
    target_dpi: float,
    dpi_tolerance: float,
    expected_size: tuple[float, float] | None,
) -> ValidationResult:
    result = ValidationResult(path, "PNG")
    if path.suffix.lower() != ".png":
        result.fail("PNG_EXTENSION", "filename does not end in .png")
    try:
        with Image.open(path) as image:
            if image.format != "PNG":
                result.fail("PNG_FORMAT", f"file content is {image.format}, not PNG")
            width, height = image.size
            dpi = image.info.get("dpi")
            image.load()
            blank_reason = image_blank_reason(image)
            if blank_reason:
                result.fail("PNG_BLANK", blank_reason)
    except Exception as exc:
        result.not_certified("PNG_DECODE", f"PNG cannot be decoded: {exc}")
        return result

    result.details.append(f"pixels={width}x{height}")
    if not dpi or len(dpi) < 2:
        result.fail("PNG_DPI_MISSING", "PNG has no two-axis physical-resolution metadata")
    else:
        x_dpi, y_dpi = float(dpi[0]), float(dpi[1])
        result.details.append(f"dpi={x_dpi:.3f}x{y_dpi:.3f}")
        if abs(x_dpi - target_dpi) > dpi_tolerance or abs(y_dpi - target_dpi) > dpi_tolerance:
            result.fail(
                "PNG_DPI",
                f"PNG resolution is {x_dpi:.3f}x{y_dpi:.3f} DPI; expected {target_dpi:g}+/-{dpi_tolerance:g}",
            )
        if expected_size:
            expected_width = expected_size[0] * target_dpi
            expected_height = expected_size[1] * target_dpi
            pixel_tolerance = max(1.0, target_dpi * 0.005)
            if abs(width - expected_width) > pixel_tolerance or abs(height - expected_height) > pixel_tolerance:
                result.fail(
                    "PNG_PIXEL_SIZE",
                    f"pixels {width}x{height} do not match {expected_size[0]:g}x{expected_size[1]:g} inches at {target_dpi:g} DPI",
                )
            result.details.append(
                f"expected-physical-size={expected_size[0]:g}x{expected_size[1]:g}in"
            )
        else:
            result.details.append(
                f"inferred-physical-size={width / x_dpi:.3f}x{height / y_dpi:.3f}in"
            )
    return result


def find_pdftoppm(explicit: str | None) -> str | None:
    if explicit:
        candidate = Path(explicit)
        return str(candidate) if candidate.is_file() else None
    return shutil.which("pdftoppm")


def _render_size_matches(
    observed: tuple[int, int], expected_points: tuple[float, float], dpi: float, rotation: int
) -> bool:
    width_points, height_points = expected_points
    if rotation in {90, 270}:
        width_points, height_points = height_points, width_points
    expected = (width_points * dpi / 72.0, height_points * dpi / 72.0)
    # Poppler normally rounds page boundaries outward. Accept at most one pixel
    # of normal raster-grid rounding in either dimension.
    return all(abs(actual - target) <= 1.05 for actual, target in zip(observed, expected))


def _prepare_render_directory(
    path: Path, result: ValidationResult, requested_base: Path | None
) -> tuple[Path | None, tempfile.TemporaryDirectory[str] | None]:
    if requested_base is not None:
        try:
            base = requested_base.expanduser().resolve(strict=False)
            if base.exists() and not base.is_dir():
                result.not_certified(
                    "RENDER_DIR_NOT_DIRECTORY", f"--render-dir is not a directory: {base}"
                )
                return None, None
            base.mkdir(parents=True, exist_ok=True)
            safe_stem = re.sub(r"[^A-Za-z0-9._-]+", "-", path.stem).strip("-.") or "figure"
            # Always create a normal child directory so Windows applies the
            # parent's inherited ACL. Path.mkdir(exist_ok=False) is atomic and
            # cannot overwrite an existing entry; the retry loop handles the
            # extremely unlikely random-name collision.
            output = None
            for _ in range(20):
                candidate = base / f"{safe_stem[:60]}-render-{secrets.token_hex(4)}"
                try:
                    candidate.mkdir(exist_ok=False)
                except FileExistsError:
                    continue
                output = candidate
                break
            if output is None:
                result.not_certified(
                    "RENDER_DIR_COLLISION",
                    "could not allocate a unique render child after 20 attempts",
                )
                return None, None
        except OSError as exc:
            result.not_certified(
                "RENDER_DIR_CREATE", f"cannot create a unique child under --render-dir: {exc}"
            )
            return None, None
        result.details.append(f"render-output-dir={output} (retained; unique child, no overwrite)")
        return output, None

    task_tmp_root = Path.cwd() / "tmp"
    try:
        task_tmp_root.mkdir(parents=True, exist_ok=True)
        temporary = tempfile.TemporaryDirectory(prefix="validate-figure-", dir=task_tmp_root)
    except OSError:
        try:
            temporary = tempfile.TemporaryDirectory(prefix="validate-figure-")
        except OSError as exc:
            result.not_certified("RENDER_DIR_CREATE", f"cannot create a temporary render directory: {exc}")
            return None, None
    return Path(temporary.name), temporary


def render_pdf(
    path: Path,
    result: ValidationResult,
    page_stats: Sequence[PdfPageStats],
    executable: str | None,
    render_dpi: float,
    render_dir: Path | None,
) -> None:
    pdftoppm = find_pdftoppm(executable)
    if not pdftoppm:
        result.not_certified("POPPLER_MISSING", "pdftoppm was not found; visual rendering is mandatory")
        return
    output_dir, temporary = _prepare_render_directory(path, result, render_dir)
    if output_dir is None:
        return
    try:
        prefix = output_dir / "page"
        command = [
            pdftoppm,
            "-png",
            "-r",
            f"{render_dpi:g}",
            "-f",
            "1",
            "-l",
            str(len(page_stats)),
            str(path),
            str(prefix),
        ]
        try:
            completed = subprocess.run(
                command,
                stdout=subprocess.PIPE,
                stderr=subprocess.PIPE,
                text=True,
                timeout=max(60, len(page_stats) * 30),
                check=False,
            )
        except Exception as exc:
            result.not_certified("POPPLER_RUN", f"Poppler rendering could not run: {exc}")
            return
        if completed.returncode != 0:
            detail = completed.stderr.strip() or f"exit code {completed.returncode}"
            result.not_certified("POPPLER_RENDER", f"Poppler failed to render the PDF: {detail}")
            return
        rendered: dict[int, Path] = {}
        for candidate in output_dir.glob("page-*.png"):
            match = re.fullmatch(r"page-(\d+)\.png", candidate.name, flags=re.IGNORECASE)
            if match:
                rendered[int(match.group(1))] = candidate
        if len(rendered) != len(page_stats) or set(rendered) != set(
            range(1, len(page_stats) + 1)
        ):
            result.not_certified(
                "POPPLER_PAGE_COUNT",
                f"Poppler produced page numbers {sorted(rendered)}, expected 1..{len(page_stats)}",
            )
            return
        for stats in page_stats:
            output = rendered[stats.page_number]
            try:
                with Image.open(output) as image:
                    image.verify()
                with Image.open(output) as image:
                    image.load()
                    observed_size = image.size
                    blank_reason = image_blank_reason(image)
            except Exception as exc:
                result.not_certified("POPPLER_OUTPUT", f"rendered page {output.name} is invalid: {exc}")
                continue
            if stats.rotation not in {0, 90, 180, 270}:
                result.not_certified(
                    "PAGE_ROTATION",
                    f"page {stats.page_number} has unsupported /Rotate={stats.rotation}",
                )
                matched_box = None
            elif _render_size_matches(
                observed_size, stats.crop_size_points, render_dpi, stats.rotation
            ):
                matched_box = "CropBox"
            elif _render_size_matches(
                observed_size, stats.media_size_points, render_dpi, stats.rotation
            ):
                matched_box = "MediaBox"
            else:
                matched_box = None
                result.fail(
                    "RENDER_PIXEL_SIZE",
                    f"page {stats.page_number} rendered to {observed_size[0]}x{observed_size[1]} px at {render_dpi:g} DPI; "
                    f"this matches neither CropBox {stats.crop_size_points[0]:.3f}x{stats.crop_size_points[1]:.3f} pt "
                    f"nor MediaBox {stats.media_size_points[0]:.3f}x{stats.media_size_points[1]:.3f} pt",
                )
            if blank_reason:
                result.fail(
                    "RENDER_BLANK",
                    f"page {stats.page_number} rendered blank: {blank_reason}",
                )
            match_text = matched_box or "no-box-match"
            content_text = "blank" if blank_reason else "nonblank"
            result.details.append(
                f"render-page={stats.page_number} pixels={observed_size[0]}x{observed_size[1]} "
                f"dpi={render_dpi:g} box={match_text} content={content_text}"
            )
        result.details.append(f"poppler-rendered-pages={len(rendered)}")
    finally:
        if temporary is not None:
            temporary.cleanup()


def validate_pdf(path: Path, args: argparse.Namespace) -> ValidationResult:
    result = ValidationResult(path, "PDF")
    if path.suffix.lower() != ".pdf":
        result.fail("PDF_EXTENSION", "filename does not end in .pdf")
    auditor = PdfAuditor(
        path,
        result,
        max_clips=args.max_clips,
        expected_clips=args.expected_clips,
        declared_clip_boxes=args.clip_box,
        allow_raster_images=args.allow_raster_images,
        allow_form_xobjects=args.allow_form_xobjects,
        allow_no_text=args.allow_no_text,
        allow_no_vector=args.allow_no_vector,
    )
    page_stats = auditor.audit()
    if page_stats:
        for stats in page_stats:
            fonts = sorted({font.base_font for font in stats.font_uses.values()})
            clip_summary = "; ".join(
                f"#{event.number} depth={event.depth} kinds={'+'.join(sorted(event.kinds)) or 'empty'}"
                for event in stats.clip_events
            )
            result.details.append(
                f"page={stats.page_number} clips={len(stats.clip_events)} max-depth={stats.max_clip_depth} "
                f"q/Q={stats.q_count}/{stats.Q_count} text-shows={stats.text_show_count} "
                f"vector-paints={stats.vector_paint_count} images={stats.image_count + stats.inline_image_count} "
                f"forms={stats.form_calls} fonts={','.join(fonts) or 'none'}"
            )
            if clip_summary:
                result.details.append(f"page={stats.page_number} clip-groups: {clip_summary}")
        render_pdf(
            path,
            result,
            page_stats,
            args.pdftoppm,
            args.render_dpi,
            args.render_dir,
        )
    return result


def parse_box(value: str) -> tuple[float, float, float, float]:
    try:
        fields = tuple(float(part.strip()) for part in value.split(","))
    except ValueError as exc:
        raise argparse.ArgumentTypeError("clip box must be x,y,width,height") from exc
    if len(fields) != 4 or fields[2] <= 0 or fields[3] <= 0:
        raise argparse.ArgumentTypeError("clip box must be x,y,width,height with positive size")
    return fields  # type: ignore[return-value]


def parse_size(value: str) -> tuple[float, float]:
    match = re.fullmatch(r"\s*([0-9]*\.?[0-9]+)\s*[xX×]\s*([0-9]*\.?[0-9]+)\s*", value)
    if not match:
        raise argparse.ArgumentTypeError("expected size must be WIDTHxHEIGHT in inches")
    return (float(match.group(1)), float(match.group(2)))


def positive_float(value: str) -> float:
    try:
        parsed = float(value)
    except ValueError as exc:
        raise argparse.ArgumentTypeError("value must be a positive number") from exc
    if not math.isfinite(parsed) or parsed <= 0:
        raise argparse.ArgumentTypeError("value must be a finite positive number")
    return parsed


def build_parser() -> argparse.ArgumentParser:
    parser = argparse.ArgumentParser(
        description="Validate 600-DPI PNGs or Illustrator-editable vector figure PDFs."
    )
    parser.add_argument("files", nargs="+", type=Path)
    parser.add_argument("--target-dpi", type=float, default=600.0)
    parser.add_argument("--dpi-tolerance", type=float, default=1.0)
    parser.add_argument(
        "--expected-size-in",
        type=parse_size,
        metavar="WIDTHxHEIGHT",
        help="also verify PNG pixel dimensions against a physical size in inches",
    )
    parser.add_argument("--max-clips", type=int, default=4, choices=range(0, 5))
    parser.add_argument(
        "--expected-clips",
        type=int,
        choices=range(0, 5),
        help="require exactly this many clipping paths on every PDF page (use 3 for the full-period regression profile)",
    )
    parser.add_argument(
        "--clip-box",
        type=parse_box,
        action="append",
        default=[],
        metavar="X,Y,W,H",
        help="declare an allowed non-page rectangular PDF clipping box (repeatable)",
    )
    parser.add_argument(
        "--allow-raster-images",
        action="store_true",
        help="allow intentional image XObjects; masks and inline images remain forbidden",
    )
    parser.add_argument(
        "--allow-form-xobjects",
        action="store_true",
        help="allow recursively inspected Form XObjects; the strict default rejects all Forms",
    )
    parser.add_argument("--allow-no-text", action="store_true")
    parser.add_argument("--allow-no-vector", action="store_true")
    parser.add_argument("--pdftoppm", help="explicit path to Poppler pdftoppm")
    parser.add_argument(
        "--render-dpi",
        type=positive_float,
        default=72.0,
        help="Poppler verification resolution (default: 72 DPI)",
    )
    parser.add_argument(
        "--render-dir",
        type=Path,
        help="retain rendered pages in a newly created unique child of DIR; existing contents are never overwritten",
    )
    return parser


def print_result(result: ValidationResult) -> None:
    print(f"[{result.status}] {result.path}")
    for detail in result.details:
        print(f"  INFO {detail}")
    for issue in result.issues:
        label = "ERROR" if issue.outcome == "FAIL" else "BLOCKER"
        print(f"  {label} [{issue.code}] {issue.message}")


def main(argv: Sequence[str] | None = None) -> int:
    parser = build_parser()
    args = parser.parse_args(argv)
    if args.expected_clips is not None and args.expected_clips > args.max_clips:
        parser.error("--expected-clips cannot exceed --max-clips")
    results: list[ValidationResult] = []
    for raw_path in args.files:
        path = raw_path.resolve()
        if not path.is_file():
            result = ValidationResult(path, "UNKNOWN")
            result.not_certified("FILE_MISSING", "file does not exist or is not a regular file")
        elif path.suffix.lower() == ".png":
            result = validate_png(path, args.target_dpi, args.dpi_tolerance, args.expected_size_in)
        elif path.suffix.lower() == ".pdf":
            result = validate_pdf(path, args)
        else:
            result = ValidationResult(path, "UNKNOWN")
            result.not_certified("FILE_TYPE", "only .png and .pdf are supported")
        results.append(result)
        print_result(result)
    statuses = Counter(result.status for result in results)
    passed = statuses["PASS"]
    print(
        f"SUMMARY passed={passed} failed={statuses['FAIL']} "
        f"not-certified={statuses['NOT CERTIFIED']} total={len(results)}"
    )
    return 0 if passed == len(results) else 1


if __name__ == "__main__":
    raise SystemExit(main())
```

# Figure 4F: hearing-loss candidate expression across developmental states

```python
#1. =========== Saved inputs and output locations
from pathlib import Path
import os
import sys
import re
import math
import json
import hashlib
import h5py
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib import font_manager
from matplotlib.transforms import Bbox
from matplotlib.patches import Rectangle

ROOT = Path(".")  # Set the working root and replace each individual path placeholder.
# Set SGN_Candidate_expression.h5, produced by the native-source RNA extraction in 05_fig4E.ipynb.
EXPRESSION_FILE = ROOT
# Set FinnGen_Neural_candidates.tsv, produced by the completed human colocalization screen.
CANDIDATE_FILE = ROOT
# Set SGN_Candidate_orthologs.tsv, produced by the candidate orthology-mapping step.
ORTHOLOG_FILE = ROOT
# Set SGN_Candidate_sources.json, produced by the audited RNA/candidate input-preparation step.
SOURCE_FILE = ROOT
# Set Whole_CSS.h5ad, produced by the whole-period integration used for Fig1.
WHOLE = ROOT
EXPECTED_HASHES = {WHOLE: "57217aabefc823bf64fa9f6db5e9cd5c2314d7164f3a758d7458650bd36c24dc"}
PNG_FILE = ROOT  # Set the PNG output file path for Fig4F.
PDF_FILE = ROOT  # Set the PDF output file path for Fig4F.
WORK = ROOT  # Set the temporary plotting-cache directory.
TASK = WORK
FONT_FILES = {
    "normal": ROOT,  # Set the Arial regular font file path.
    "italic": ROOT,  # Set the Arial italic font file path.
    "bold": ROOT,  # Set the Arial bold font file path.
    "bolditalic": ROOT,  # Set the Arial bold italic font file path.
}
WORK.mkdir(parents=True, exist_ok=True)
os.environ["MPLCONFIGDIR"] = str(WORK)
sys.dont_write_bytecode = True
for font_path in FONT_FILES.values():
    font_manager.fontManager.addfont(str(font_path))
```

```python
#2. =========== Read-only input and colorbar helpers
def digest(path):
    value = hashlib.sha256()
    with path.open("rb") as stream:
        for block in iter(lambda: stream.read(8 * 1024 * 1024), b""):
            value.update(block)
    return value.hexdigest()

def vector(node):
    if isinstance(node, h5py.Dataset):
        return node.asstr()[()] if node.dtype.kind in "OSU" else node[()]
    encoding = node.attrs.get("encoding-type", "")
    if isinstance(encoding, bytes):
        encoding = encoding.decode()
    if encoding == "categorical":
        return pd.Categorical.from_codes(node["codes"][()], vector(node["categories"]))
    if "values" in node and "mask" in node:
        values = np.asarray(vector(node["values"]), dtype=object)
        values[node["mask"][()]] = None
        return values
    raise ValueError(f"Unsupported vector encoding: {node.name}: {encoding}")

def index(group):
    key = group.attrs.get("_index", "_index")
    if isinstance(key, bytes):
        key = key.decode()
    return pd.Index(vector(group[key]).astype(str))

def vector_colorbar(ax, cmap, norm, ticks, tick_size=7, decimals=None, n=256):
    for i in range(n):
        color=cmap((i+.5)/n)
        ax.add_patch(Rectangle((i/n,0),1/n,1,facecolor=color,edgecolor=color,linewidth=.12,clip_on=False))
    labels=[f"{v:.{decimals}f}" if decimals is not None else f"{v:g}" for v in ticks]
    ax.set(xlim=(0,1),ylim=(0,1));ax.set_xticks(np.asarray(norm(ticks),float),labels)
    ax.set_yticks([]);ax.tick_params(labelsize=tick_size,length=2,pad=2)
    for spine in ax.spines.values():spine.set_visible(False)
```

```python
#3. =========== Editable PDF exporter
"""Strict vector PDF export for the selected Matplotlib canvas.

No analysis is performed. Each native Matplotlib label becomes one live Arial
text object, including the TF italic run in the two supported mathtext titles.
All geometry is written directly, without XObjects, glyph outlines or images.
The caller declares at most four data axes; their clip rectangles are collected
into one sibling viewport group per axis. Other artists must fit their axes.
"""
from pathlib import Path
import re
import sys
import math
import numpy as np
from matplotlib.backend_bases import RendererBase
from matplotlib.path import Path as MplPath

from reportlab.pdfgen import canvas
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont

_FONTS = {'normal':'FigArial','italic':'FigArialItalic',
          'bold':'FigArialBold','bolditalic':'FigArialBoldItalic'}

def _register():
    for key, font_path in FONT_FILES.items():
        if _FONTS[key] not in pdfmetrics.getRegisteredFontNames():
            pdfmetrics.registerFont(TTFont(_FONTS[key],str(font_path)))

def _font(prop):
    bold=prop.get_weight() in ('bold','heavy','black','semibold','demibold') or isinstance(prop.get_weight(),(int,float)) and prop.get_weight()>=600
    italic=prop.get_style() in ('italic','oblique')
    return _FONTS['bolditalic' if bold and italic else 'bold' if bold else 'italic' if italic else 'normal']

def _runs(text,prop,ismath):
    if not ismath:
        if '$' in text: raise ValueError('Unexpected dollar in plain label: '+text)
        return [(text,_font(prop))]
    # Current Fig4 titles use only this explicitly audited mathtext form.
    found=re.fullmatch(r'\$\\it\{([^{}]+)\}([^$]*)\$(.*)',text)
    if found is None: raise ValueError('Unsupported math label: '+text)
    tf,inside,after=found.groups()
    return [(tf,_FONTS['italic']),(inside+after,_FONTS['normal'])]

class _Recorder(RendererBase):
    def __init__(self,fig,agg,viewports,halo_labels):
        super().__init__()
        self.dpi=fig.dpi;self.unit=72/fig.dpi;self.agg=agg
        self.width,self.height=fig.bbox.size;self.ops=[];self.viewports=viewports;self.halo_labels=halo_labels
    def points_to_pixels(self,p): return p/self.unit
    def flipy(self):return False
    def get_canvas_width_height(self):return self.width,self.height
    def get_text_width_height_descent(self,s,prop,ismath):
        runs=_runs(s,prop,ismath)
        width=sum(pdfmetrics.stringWidth(t,f,prop.get_size_in_points()) for t,f in runs)/self.unit
        # Match the source Arial baseline/ascent geometry; width uses the same
        # embedded font as the PDF, including the semantic mixed italic title.
        _,h,d=self.agg.get_text_width_height_descent(re.sub(r'\$|\\it\{|\}', '',s),prop,False)
        return width,h,d
    def _clip(self,gc):
        if gc.get_clip_path()[0] is not None:
            raise ValueError('Nonrectangular artist clip is unsupported')
        b=gc.get_clip_rectangle()
        if b is None:return None
        for i,box in enumerate(self.viewports):
            if np.allclose(b.bounds,box,rtol=0,atol=1e-5):return i
        return None
    def draw_path(self,gc,path,transform,rgbFace=None):
        if gc.get_hatch():raise ValueError('Hatching/Pattern is forbidden')
        p=transform.transform_path(path)
        self.ops.append(('path',self._clip(gc),p,gc.get_rgb(),rgbFace,
                         gc.get_linewidth(),gc.get_alpha(),gc.get_forced_alpha(),
                         gc.get_dashes(),gc.get_capstyle(),gc.get_joinstyle()))
    def draw_text(self,gc,x,y,s,prop,angle,ismath=False,mtext=None):
        self.ops.append(('text',None,float(x),float(y),str(s),prop.copy(),float(angle),
                         ismath,gc.get_rgb(),gc.get_alpha(),id(mtext) in self.halo_labels))
    def draw_image(self,*args,**kwargs):raise ValueError('Raster image is forbidden')
    def start_rasterizing(self):pass
    def stop_rasterizing(self):pass
    def start_filter(self):raise ValueError('Raster filter is forbidden')
    def stop_filter(self,*args):raise ValueError('Raster filter is forbidden')

def _pdf_path(pdf,path,unit):
    out=pdf.beginPath();current=None;start=None
    for vertices,code in path.iter_segments(remove_nans=True,simplify=False,curves=True):
        v=np.asarray(vertices).reshape(-1,2)*unit
        if code==MplPath.MOVETO:
            current=v[0];start=current;out.moveTo(*current)
        elif code==MplPath.LINETO:
            current=v[0];out.lineTo(*current)
        elif code==MplPath.CURVE3:
            control,end=v
            a=current+(control-current)*2/3;b=end+(control-end)*2/3
            out.curveTo(*a,*b,*end);current=end
        elif code==MplPath.CURVE4:
            out.curveTo(*v.ravel());current=v[-1]
        elif code==MplPath.CLOSEPOLY:
            out.close();current=start
        elif code==MplPath.STOP:break
        else:raise ValueError('Unknown path opcode '+str(code))
    return out

def strict_editable_pdf(fig,destination,*,data_axes=(),live_halo_width=.20,crop_inches=None):
    """Write a strict vector PDF. Return declared clip boxes in PDF points.

    The artist canvas remains unchanged after return. Rasterization flags and
    white text path effects are replaced by a narrow live text fill+stroke
    (render mode 2) in the same single label object; no outlined font paths are
    emitted. The width is narrower than Agg's halo to preserve the black stems.
    """
    _register()
    if len(data_axes)>4:raise ValueError('At most four sibling data viewports')
    fig.canvas.draw();agg=fig.canvas.get_renderer()
    viewports=[tuple(ax.bbox.bounds) for ax in data_axes]
    effects=[]
    for artist in fig.findobj():
        if isinstance(artist, __import__('matplotlib.text',fromlist=['Text']).Text) and hasattr(artist,'get_path_effects'):
            old=artist.get_path_effects()
            if old:effects.append((artist,old));artist.set_path_effects([])
    rec=_Recorder(fig,agg,viewports,{id(a) for a,e in effects})
    try:fig.draw(rec)
    finally:
        for a,e in effects:a.set_path_effects(e)
    dst=Path(destination);dst.parent.mkdir(parents=True,exist_ok=True)
    crop=np.asarray(crop_inches if crop_inches is not None else [0,0,*fig.get_size_inches()],float)
    unit=rec.unit
    # Discard figure/axes backgrounds and objects belonging to the other panels.
    kept=[]
    x0,y0,w,h=crop*fig.dpi
    for op in rec.ops:
        if op[0]=='text':
            xx,yy=op[2:4]
            inside=x0-1<=xx<=x0+w+1 and y0-1<=yy<=y0+h+1
        else:
            vv=op[2].vertices
            if not len(vv):continue
            box=Bbox.from_extents(*np.r_[vv.min(axis=0),vv.max(axis=0)])
            is_background=(box.width>=fig.bbox.width-1 and box.height>=fig.bbox.height-1)
            inside=(not is_background and box.x1>=x0 and box.x0<=x0+w and box.y1>=y0 and box.y0<=y0+h)
        if inside:kept.append(op)
    rec.ops=kept
    pdf=canvas.Canvas(str(dst),pagesize=tuple(crop[2:]*72),
                      pageCompression=1,invariant=1,initialFontName=_FONTS['normal'])
    pdf.translate(-crop[0]*72,-crop[1]*72)
    def emit(op):
        if op[0]=='text':
            _,_,x,y,s,prop,angle,ismath,color,alpha,halo=op
            if not s:return
            pdf.saveState();pdf.setFillColorRGB(*color[:3]);pdf.setFillAlpha(alpha)
            text=pdf.beginText();rad=math.radians(angle)
            if halo and live_halo_width>0:
                pdf.setStrokeColorRGB(1,1,1);pdf.setStrokeAlpha(alpha)
                pdf.setLineWidth(float(live_halo_width));text.setTextRenderMode(2)
            text.setTextTransform(math.cos(rad),math.sin(rad),-math.sin(rad),math.cos(rad),x*unit,y*unit)
            for value,font in _runs(s,prop,ismath):
                if value:text.setFont(font,prop.get_size_in_points());text.textOut(value)
            pdf.drawText(text);pdf.restoreState();return
        _,_,path,stroke,face,lw,alpha,forced,dashes,cap,join=op
        pdf.saveState();pdf.setLineWidth(lw)
        pdf.setLineCap({'butt':0,'round':1,'projecting':2}[cap])
        pdf.setLineJoin({'miter':0,'round':1,'bevel':2}[join])
        offset,pattern=dashes
        pdf.setDash([] if pattern is None else [float(v) for v in pattern],0.0 if offset is None else float(offset))
        if stroke is not None:
            pdf.setStrokeColorRGB(*stroke[:3]);pdf.setStrokeAlpha(alpha if forced else (stroke[3] if len(stroke)>3 else alpha))
        if face is not None:
            pdf.setFillColorRGB(*face[:3]);pdf.setFillAlpha(alpha if forced else (face[3] if len(face)>3 else alpha))
        pdf.drawPath(_pdf_path(pdf,path,unit),stroke=int(stroke is not None and lw>0),fill=int(face is not None),fillMode=1)
        pdf.restoreState()
    done=set()
    for op in rec.ops:
        key=op[1]
        if key is None:emit(op);continue
        if key in done:continue
        done.add(key);pdf.saveState()
        p=pdf.beginPath();p.rect(*(np.asarray(viewports[key])*unit))
        pdf.clipPath(p,stroke=0,fill=0,fillMode=1)
        for candidate in rec.ops:
            if candidate[1]==key:emit(candidate)
        pdf.restoreState()
    pdf.showPage();pdf.save()
    if done!=set(range(len(data_axes))):raise AssertionError('Declared clip has no artist')
    return {'expected_clips':len(done),'clip_boxes_pt':(np.asarray(viewports)*unit).tolist(),
            'editable_label_objects':sum(op[0]=='text' for op in rec.ops),
            'label_strings':[op[4] for op in rec.ops if op[0]=='text'],
            'vector_paths':sum(op[0]=='path' for op in rec.ops),
            'text_effects':'White live text fill+stroke, single object; stroke width '+str(live_halo_width)+' pt.',
            'halo_labels':len(effects)}
```

```python
#4. =========== Original candidate-expression canvas
def build_BC_canvas():
    """Plot a short genetic workflow, five frozen whole-atlas features and a state DotPlot."""
    import matplotlib
    matplotlib.use("Agg")
    import matplotlib.pyplot as plt
    from matplotlib.colors import LinearSegmentedColormap, Normalize, FuncNorm
    from matplotlib.cm import ScalarMappable
    from matplotlib.patches import FancyArrowPatch, PathPatch, Ellipse, Circle, Rectangle
    from matplotlib.text import Text
    from matplotlib.path import Path as MplPath
    import matplotlib.patheffects as pe
    from scipy.interpolate import UnivariateSpline
    from scipy.spatial import cKDTree

    #1. =========== Figure-only parameters; the existing atlas and counts stay fixed
    data_dir = ROOT  # Set the saved-input directory if needed.
    expression_path = EXPRESSION_FILE
    candidate_path = CANDIDATE_FILE
    ortholog_path = ORTHOLOG_FILE
    source_path = SOURCE_FILE
    output_path = PNG_FILE
    preview_path = PNG_FILE
    human_order = ["DEGS2", "FAM171A2", "SPTBN1", "RTRAF", "CYP26A1", "SPTBN1-AS2"]
    figure_size, final_dpi, preview_dpi = (13.6, 8.0), 600, 140
    feature_area, feature_alpha = 1.85, .90
    feature_vmin, feature_vmax = 0.0, 4.0  # Common actual log10(1+CPM), not per-gene maxima.
    dot_max_area, column_gap_ratio = 144.0, .5  # Point AREA is linear in detection fraction.
    gene_font, state_font = 8.0, 7.4
    stage_font, stage_left = 7.2, 8.95
    curve_t_start_bin, curve_samples = 2, 200
    curve_smoothing_rms = .35  # UMAP units; display smoothing only, with fixed endpoints.
    curve_cloud_max_distance = .50  # In the unchanged UMAP coordinate units.
    curve_labels = {"AC": ("Ib/Ic", (4.4, 7.6)), "AB": ("Ia/II", (-7.5, 5.2))}
    ink, muted, zero_color, missing_color = "#20242A", "#6B7078", "#D9D9D9", "#BBBBBB"
    color_log_tau = 1.0  # Display normalization only; existing RNA values are not transformed again.
    feature_norm = FuncNorm((lambda x: np.log1p(x/color_log_tau),
                            lambda y: color_log_tau*np.expm1(y)),
                           vmin=feature_vmin, vmax=feature_vmax, clip=True)
    color_knots = [0, .5, 1, 1.5, 2, 2.5, 3, 3.5, 4]
    color_values = ["#F2F2F2", "#FFD8D3", "#FFC0B8", "#FF9A8A", "#FF6D63",
                    "#F54E4A", "#D92D37", "#BC1428", "#9C001E"]
    feature_cmap = LinearSegmentedColormap.from_list("RNA_nonlinear_red",
        list(zip(feature_norm(color_knots), color_values)), N=1024)
    dot_cmap = LinearSegmentedColormap.from_list("candidate_expression", ["#F1F1F1", "#E8C6DE", "#C67ABA", "#952E7D", "#69005F"])
    dot_norm = Normalize(-1, 1, clip=True)
    plt.rcParams.update({"font.family": "Arial", "font.size": 8, "text.color": ink,
        "axes.unicode_minus": False, "axes.linewidth": .55, "mathtext.default": "regular"})

    #2. =========== Exact cell/embedding correspondence and fixed candidate checks
    source = json.loads(source_path.read_text(encoding="utf-8"))
    input_paths = [expression_path, candidate_path, ortholog_path, WHOLE]
    input_hashes = {p.name: digest(p) for p in input_paths}
    assert input_hashes[expression_path.name] == source["outputs"][expression_path.name]["sha256"]
    assert input_hashes[WHOLE.name] == EXPECTED_HASHES[WHOLE]
    with h5py.File(expression_path, "r") as h:
        humans = h["human_symbols"].asstr()[()].tolist()
        human_ids = h["human_ids"].asstr()[()].tolist()
        mouse_ids = h["mouse_ids"].asstr()[()].tolist()
        mouse = h["mouse_symbols"].asstr()[()].tolist()[:5]
        cells = h["cell_ids"].asstr()[()]
        states = h["state"].asstr()[()]
        sources = h["study_id"].asstr()[()]
        libraries = h["library_id"].asstr()[()]
        stages = h["stage_standardized"].asstr()[()]
        cell_segments = h["segment"].asstr()[()]
        native_t = h["native_t"][()]
        saved_segment_order = h["tree/segment_order"].asstr()[()].tolist()
        native_edges = h["tree/native_t_quantile_edges"][()]
        counts = h["counts"][()][:, :5]
        measured = h["measured"][()].astype(bool)[:, :5]
        expression = h["log10_1p_CPM"][()][:, :5]
        assert not h["measured"][:, -1].any() and np.isnan(h["counts"][:, -1]).all()
    assert humans == human_order and counts.shape == measured.shape == expression.shape == (8340, 5)
    assert len(set(cells)) == 8340 and all(mouse_ids[:5]) and mouse_ids[-1] == ""
    assert np.array_equal(np.isfinite(expression), measured)
    assert np.equal(expression[measured] > 0, counts[measured] > 0).all()
    assert np.isfinite(expression[measured]).all() and expression[measured].min() >= feature_vmin
    assert expression[measured].max() <= feature_vmax
    with h5py.File(WHOLE, "r") as h:
        assert np.array_equal(index(h["obs"]).astype(str), cells)
        xy = h["obsm/X_umap"][()]
        accepted_hash = h["uns/css/umap_sha256"].asstr()[()]
        full_labels = np.asarray(vector(h["obs/source_stage_celltype"])).astype(str)
        original_row_order = h["uns/css_figure_contract/dot_row_order"].asstr()[()].tolist()
    embedding_digest = hashlib.sha256()
    embedding_digest.update(str(xy.shape).encode()); embedding_digest.update(str(xy.dtype).encode())
    embedding_digest.update(np.ascontiguousarray(xy).tobytes())
    assert embedding_digest.hexdigest() == accepted_hash == "4cd0a00aaccc010c8f8ed2f258462ab1524d2df93f9541396001c9bda7f2254b"
    assert np.array_equal(states, [x.split(" | ", 1)[1] for x in full_labels])
    assert np.array_equal(stages, [x.split(" | ", 1)[0] for x in full_labels])
    state_order = list(dict.fromkeys(x.split(" | ", 1)[1] for x in original_row_order))
    assert len(state_order) == 18 and set(state_order) == set(states)
    stage_sets = [sorted(set(stages[states == state]), key=lambda age: (age[0] == "P", float(age[1:])))
                  for state in state_order]
    stage_display_rules = {
        ("E11.5", "E13.5"): "E11.5 / E13.5",
        ("E14.5", "E15.5", "E16.5", "E18.5"): "E14.5–E16.5; E18.5",
        ("E14.5", "E15.5", "E16.5", "E17.5"): "E14.5–E17.5",
        ("P1", "P7", "P28"): "P1 / P7 / P28",
    }
    stage_labels = [values[0] if len(values) == 1 else stage_display_rules[tuple(values)]
                    for values in stage_sets]
    # Display guides project accepted native-time bins onto the fixed UMAP; no tree refit.
    guides, guide_controls = {}, {}
    for segment in ["T", "AC", "AB"]:
        members = np.flatnonzero(cell_segments == segment)
        edge = native_edges[saved_segment_order.index(segment)]
        edge = edge[np.isfinite(edge)]
        bin_ids = np.searchsorted(edge[1:-1], native_t[members], side="left")
        start_bin = curve_t_start_bin if segment == "T" else 0
        anchors, records = [], []
        for bin_id in range(start_bin, len(edge)-1):
            current = members[bin_ids == bin_id]
            neighborhood = members[(np.abs(bin_ids-bin_id) <= 1) & (bin_ids >= start_bin)]
            assert len(current) > 0
            robust_center = np.median(xy[neighborhood], axis=0)
            nearest = current[np.argmin(np.linalg.norm(xy[current]-robust_center, axis=1))]
            anchors.append(xy[nearest].astype(float))
            records.append({"segment": segment, "native_bin": bin_id, "n_bin_cells": len(current),
                "cell_id": str(cells[nearest]), "cell_native_t": float(native_t[nearest]),
                "bin_median_t": float(np.median(native_t[current])), "xy": xy[nearest].astype(float).tolist()})
        anchors = np.asarray(anchors)
        guide_controls[segment] = records
        controls = anchors
        times = np.array([record["bin_median_t"] for record in records])
        if segment != "T":
            controls = np.vstack([guides["T"]["xy"][-1], controls])
            times = np.r_[guides["T"]["t"][-1], times]
        assert np.all(np.diff(times) > 0)
        display_t = np.linspace(times[0], times[-1], curve_samples)
        boundary_weights = np.ones(len(controls))
        boundary_weights[[0, -1]] = 1000.0
        display_xy = np.column_stack([UnivariateSpline(times, controls[:, j], w=boundary_weights, k=3,
            s=len(controls)*curve_smoothing_rms**2)(display_t) for j in range(2)])
        progress = (display_t-display_t[0]) / (display_t[-1]-display_t[0])
        display_xy += ((1-progress[:, None])*(controls[0]-display_xy[0])
                       + progress[:, None]*(controls[-1]-display_xy[-1]))
        display_xy[[0, -1]] = controls[[0, -1]]
        related = np.isin(cell_segments, ["T", segment])
        cloud_distance = cKDTree(xy[related]).query(display_xy)[0]
        assert cloud_distance.max() < curve_cloud_max_distance, (segment, cloud_distance.max())
        guides[segment] = {"xy": display_xy, "t": display_t,
            "max_cloud_distance": float(cloud_distance.max()),
            "p95_cloud_distance": float(np.quantile(cloud_distance, .95)),
            "length": float(np.linalg.norm(np.diff(display_xy, axis=0), axis=1).sum())}
    assert np.array_equal(guides["T"]["xy"][-1], guides["AC"]["xy"][0])
    assert np.array_equal(guides["T"]["xy"][-1], guides["AB"]["xy"][0])
    for segment in ["AC", "AB"]:
        assert guides["T"]["length"] + guides[segment]["length"] > 10
        assert guides["T"]["xy"][0, 1] - guides[segment]["xy"][-1, 1] > 8
    # Extend the shared display stem from observed E9.5 CVG cells; show the UMAP gap as dashed.
    e9 = np.flatnonzero((stages == "E9.5") & (cell_segments == "CVG"))
    earliest = e9[np.isclose(native_t[e9], native_t[e9].min(), atol=1e-12)]
    start_index = earliest[np.argmin(np.linalg.norm(xy[earliest]-np.median(xy[earliest], axis=0), axis=1))]
    assert len(e9) == 11 and native_t[start_index] == 0

    def segment_bins(name):
        members = np.flatnonzero(cell_segments == name)
        edge = native_edges[saved_segment_order.index(name)]
        edge = edge[np.isfinite(edge)]
        return members, np.searchsorted(edge[1:-1], native_t[members], side="left")

    def median_cell(indices):
        center = np.median(xy[indices], axis=0)
        return indices[np.argmin(np.linalg.norm(xy[indices]-center, axis=1))]

    cvg_members, cvg_bins = segment_bins("CVG")
    t_members, t_bins = segment_bins("T")
    transition = t_members[t_bins == 1]
    x_order = transition[np.argsort(xy[transition, 0])]
    split = int(np.diff(xy[x_order, 0]).argmax())+1
    main_transition, island_transition = x_order[:split], x_order[split:]
    visual_split_x = float((xy[x_order[split-1], 0]+xy[x_order[split], 0])/2)
    assert len(main_transition) > 5 and len(island_transition) > 5
    pair_distance = np.linalg.norm(xy[island_transition, None]-xy[main_transition][None], axis=2)
    ordered = native_t[island_transition, None] < native_t[main_transition][None]
    pair_distance[~ordered] = np.inf
    pair = np.unravel_index(pair_distance.argmin(), pair_distance.shape)
    island_boundary, main_boundary = island_transition[pair[0]], main_transition[pair[1]]
    island_indices = np.array([start_index,
        median_cell(cvg_members[cvg_bins == 1]), median_cell(cvg_members[cvg_bins == 2]),
        median_cell(t_members[t_bins == 0]), median_cell(island_transition), island_boundary])
    stem_start_index = int(np.flatnonzero(cells == guide_controls["T"][0]["cell_id"])[0])
    lead_indices = np.array([main_boundary, median_cell(main_transition), stem_start_index])
    all_early_indices = np.r_[island_indices, lead_indices]
    assert np.all(np.diff(native_t[all_early_indices]) > 0)

    def early_curve(indices):
        times = native_t[indices]
        controls = xy[indices].astype(float)
        grid = np.linspace(times[0], times[-1], curve_samples)
        weights = np.ones(len(indices)); weights[[0, -1]] = 1000
        result = np.column_stack([UnivariateSpline(times, controls[:, j], w=weights,
            k=min(3, len(indices)-1), s=len(indices)*curve_smoothing_rms**2)(grid) for j in range(2)])
        progress = (grid-grid[0])/(grid[-1]-grid[0])
        result += ((1-progress[:, None])*(controls[0]-result[0])
                   + progress[:, None]*(controls[-1]-result[-1]))
        result[[0, -1]] = controls[[0, -1]]
        return result

    early_island = early_curve(island_indices)
    early_main = early_curve(lead_indices)
    gap_curve = np.linspace(xy[island_boundary], xy[main_boundary], 80)
    early_support = np.isin(cell_segments, ["CVG", "T"])
    early_distance_tree = cKDTree(xy[early_support])
    island_max_distance = float(early_distance_tree.query(early_island)[0].max())
    lead_max_distance = float(early_distance_tree.query(early_main)[0].max())
    gap_max_distance = float(early_distance_tree.query(gap_curve)[0].max())
    assert island_max_distance < curve_cloud_max_distance and lead_max_distance < curve_cloud_max_distance
    assert gap_max_distance > curve_cloud_max_distance, "The displayed gap must be distinguished from a supported solid curve"
    assert np.array_equal(early_main[-1], guides["T"]["xy"][0])
    early_records = [{"cell_id": str(cells[i]), "xy": xy[i].astype(float).tolist(),
                      "native_t": float(native_t[i]), "stage": str(stages[i]), "segment": str(cell_segments[i])}
                     for i in all_early_indices]

    candidates = pd.read_csv(candidate_path, sep="\t")
    orthologs = pd.read_csv(ortholog_path, sep="\t")
    assert len(candidates) == 6 and set(candidates.stable_id) == set(human_ids)
    assert orthologs.mouse_id.notna().sum() == 5
    support = np.nanmax(candidates[["Cortex_H4", "Spinal_H4", "Tibial_H4"]].to_numpy(float), axis=1)
    assert np.all(support >= .8)
    center = (xy.min(axis=0) + xy.max(axis=0)) / 2
    span = max(float(np.ptp(xy, axis=0).max()) * 1.12, 1.0)
    xlim, ylim = (center[0]-span/2, center[0]+span/2), (center[1]-span/2, center[1]+span/2)

    #3. =========== One descriptive state summary; balance libraries within source, then sources
    # Use every measured cell (no bin cutoff); missing source features are never filled with zero.
    n_states, n_genes = len(state_order), len(mouse)
    means, fractions = np.full((n_states, n_genes), np.nan), np.full((n_states, n_genes), np.nan)
    measurement_coverage = np.zeros((n_states, n_genes))
    detected_cells = np.zeros((n_states, n_genes), dtype=int)
    measured_cells = np.zeros((n_states, n_genes), dtype=int)
    source_counts, library_counts = np.zeros((n_states, n_genes), int), np.zeros((n_states, n_genes), int)
    total_cells = np.array([(states == s).sum() for s in state_order])
    for si, state in enumerate(state_order):
        state_mask = states == state
        for gi in range(n_genes):
            take = state_mask & measured[:, gi]
            measured_cells[si, gi] = take.sum()
            detected_cells[si, gi] = (counts[take, gi] > 0).sum()
            measurement_coverage[si, gi] = take.sum() / state_mask.sum()
            source_values, source_detection = [], []
            for src in pd.unique(sources[take]):
                local = take & (sources == src)
                library_values, library_detection = [], []
                for lib in pd.unique(libraries[local]):
                    valid = local & (libraries == lib)
                    assert valid.sum() >= 1
                    library_values.append(float(expression[valid, gi].mean()))
                    library_detection.append(float((counts[valid, gi] > 0).mean()))
                source_values.append(float(np.mean(library_values)))
                source_detection.append(float(np.mean(library_detection)))
                library_counts[si, gi] += len(library_values)
            source_counts[si, gi] = len(source_values)
            if len(source_values):
                means[si, gi], fractions[si, gi] = np.mean(source_values), np.mean(source_detection)
    zscores = np.full_like(means, np.nan)
    for gi in range(n_genes):
        valid = np.isfinite(means[:, gi])
        values = means[valid, gi]
        sd = float(values.std(ddof=0))
        zscores[valid, gi] = (values-values.mean()) / sd if sd > 0 else 0.0
    assert np.isfinite(means).all() and (fractions >= 0).all() and (fractions <= 1).all()
    assert np.isclose(zscores.mean(axis=0), 0).all() and np.isclose(zscores.std(axis=0), 1).all()
    assert (measurement_coverage[:, :4] == 1).all()
    assert measured[:, 4].sum() == 3410 and (counts[measured[:, 4], 4] > 0).sum() == 10

    #4. =========== Retained B/C canvas; the independent workflow is drawn separately
    fig = plt.figure(figsize=figure_size, dpi=preview_dpi, facecolor="white")
    width, height = figure_size
    def text(x, y, label, **kwargs):
        return fig.text(x/width, y/height, label, **kwargs)

    #5. =========== B: the user's gray/red template on the exact whole-period UMAP
    text(.23, 6.53, "B", fontsize=17, fontweight="bold")
    text(.67, 6.53, "Whole-period expression", fontsize=12, fontweight="bold")
    feature_positions = [(.40, 3.55), (3.18, 3.55), (5.96, 3.55), (.40, .60), (3.18, .60)]
    feature_axes, feature_records, arrow_artists = [], [], []
    feature_width = 2.58
    for gi, (gene, (left, bottom)) in enumerate(zip(mouse, feature_positions)):
        ax = fig.add_axes([left/width, bottom/height, feature_width/width, feature_width/height])
        feature_axes.append(ax)
        zero = measured[:, gi] & (counts[:, gi] == 0)
        missing = ~measured[:, gi]
        positive = np.flatnonzero(measured[:, gi] & (counts[:, gi] > 0))
        positive = positive[np.argsort(expression[positive, gi], kind="stable")]
        ax.scatter(xy[zero, 0], xy[zero, 1], s=feature_area, c=zero_color, alpha=.9, linewidths=0)
        if missing.any():
            ax.scatter(xy[missing, 0], xy[missing, 1], s=feature_area, facecolors="none",
                       edgecolors=missing_color, linewidths=.16, alpha=.60)
        dots = ax.scatter(xy[positive, 0], xy[positive, 1], s=feature_area,
            c=expression[positive, gi], cmap=feature_cmap, norm=feature_norm,
            alpha=feature_alpha, edgecolors="none", linewidths=0)
        assert np.array_equal(dots.get_array(), expression[positive, gi])
        assert np.array_equal(dots.get_offsets(), xy[positive])
        assert zero.sum() + missing.sum() + len(positive) == 8340
        ax.set(xlim=xlim, ylim=ylim)
        ax.set_box_aspect(1); ax.axis("off")
        ax.set_title(gene, fontsize=12, fontstyle="italic", pad=5)
        # One shared stem, then two long branch curves with one arrowhead per branch.
        halo = [pe.Stroke(linewidth=1.45, foreground="white", alpha=.9), pe.Normal()]
        shared = guides["T"]["xy"]
        for supported in [early_island, early_main, shared]:
            ax.plot(supported[:, 0], supported[:, 1], color="#363636", lw=.75, zorder=5, path_effects=halo)
        ax.plot(gap_curve[:, 0], gap_curve[:, 1], color="#555555", lw=.75,
                linestyle=(0, (2.4, 2.0)), zorder=5, path_effects=halo)
        ax.scatter([xy[start_index, 0]], [xy[start_index, 1]], s=10, facecolors="white",
                   edgecolors="#363636", linewidths=.65, zorder=7)
        ax.text(16.6, 8.4, "E9.5 CVG", ha="right", va="center", fontsize=7.0, color=ink, zorder=7)
        for segment, (label, label_xy) in curve_labels.items():
            vertices = guides[segment]["xy"]
            codes = np.r_[MplPath.MOVETO, np.full(len(vertices)-1, MplPath.LINETO)]
            arrow = FancyArrowPatch(path=MplPath(vertices, codes), arrowstyle="-|>", mutation_scale=8.5,
                                    lw=.75, color="#363636", zorder=5, path_effects=halo)
            ax.add_patch(arrow)
            arrow_artists.append(arrow)
            ax.text(*label_xy, label, fontsize=8, ha="left", va="center", color=ink,
                    bbox={"facecolor": "white", "edgecolor": "none", "alpha": .82, "pad": .4}, zorder=6)
        if gene == "Cyp26a1":
            ax.text(.5, .985, "10 / 3,410 measured cells detected", ha="center", va="top",
                    fontsize=7.2, color=muted, transform=ax.transAxes)
        feature_records.append({"gene": gene, "positive_cells": len(positive), "zero_cells": int(zero.sum()),
                                "unmeasured_cells": int(missing.sum()), "same_cell_coordinates": True})
    # Use the unused sixth tile for a single shared legend rather than another plot.
    text(6.18, 2.55, "Expression", fontsize=9.5)
    cax = fig.add_axes([6.18/width, 2.27/height, 1.90/width, .10/height])
    vector_colorbar(cax, feature_cmap, feature_norm, [0, 1, 2, 3, 4], tick_size=7)
    text(6.18, 1.92, "log10(1 + CPM)", fontsize=8, color=muted)
    legend_axis = fig.add_axes([0, 0, 1, 1], zorder=-1)
    legend_axis.set(xlim=(0, width), ylim=(0, height)); legend_axis.axis("off")
    legend_axis.scatter([6.24], [1.61], s=13, c=zero_color, linewidths=0)
    text(6.40, 1.57, "Measured zero", fontsize=8, color=muted)
    legend_axis.scatter([6.24], [1.31], s=13, facecolors="none", edgecolors=missing_color, linewidths=.55)
    text(6.40, 1.27, "Not measured", fontsize=8, color=muted)
    text(6.18, .82, "Shared nonlinear color scale\nfor all five genes", fontsize=8, color=muted, linespacing=1.45)

    #6. =========== C: compact candidate-by-state DotPlot; one global gene-wise scaling
    text(8.90, 6.53, "C", fontsize=17, fontweight="bold")
    text(9.25, 6.53, "Developmental states", fontsize=12, fontweight="bold")
    dot_diameter = np.sqrt(dot_max_area * float(fractions.max()))
    pitch = dot_diameter * (1+column_gap_ratio) / 72
    matrix_width, matrix_height = n_genes*pitch, n_states*pitch
    matrix_left, matrix_bottom = 11.80, 1.08
    ax = fig.add_axes([matrix_left/width, matrix_bottom/height, matrix_width/width, matrix_height/height])
    xx, yy = np.meshgrid(np.arange(n_genes), np.arange(n_states))
    valid = np.isfinite(fractions) & np.isfinite(zscores)
    visible = valid & (fractions > 0)
    dots = ax.scatter(xx[visible], yy[visible], s=dot_max_area*fractions[visible], c=zscores[visible],
                      cmap=dot_cmap, norm=dot_norm, edgecolors="none", linewidths=0)
    no_data = ~valid
    if no_data.any():
        ax.scatter(xx[no_data], yy[no_data], marker="x", s=dot_max_area*.2, c=missing_color, linewidths=.6)
    ax.set(xlim=(-.5, n_genes-.5), ylim=(n_states-.5, -.5))
    labels = [g + (" †" if gi == 4 else "") for gi, g in enumerate(mouse)]
    ax.set_xticks(range(n_genes), labels, rotation=55, ha="right", fontsize=gene_font, fontstyle="italic")
    ax.set_yticks(range(n_states), state_order, fontsize=state_font)
    for state_label_artist in ax.get_yticklabels():
        state_label_artist.set_verticalalignment("center")
    ax.tick_params(length=0, pad=5)
    stage_artists = []
    for si, stage_label in enumerate(stage_labels):
        row_y = matrix_bottom + matrix_height * (1 - (si + .5) / n_states)
        stage_artists.append(text(stage_left, row_y, stage_label, fontsize=stage_font,
                                  ha="left", va="center", color=muted))
    text(stage_left, 5.66, "Stage(s)", fontsize=7.5, color=muted)
    text(10.65, 5.66, "Cell state", fontsize=7.5, color=muted, ha="center")
    for spine in ax.spines.values(): spine.set_visible(False)
    ax.plot([-.42, n_genes-.58], [-.95, -.95], lw=1.6, color="#952E7D", clip_on=False)
    ax.text((n_genes-1)/2, -1.35, "Hearing-loss candidates", fontsize=7.5, color=muted,
            ha="center", va="bottom", clip_on=False)
    # Legends remain above the matrix and are not repeated for each gene.
    dca = fig.add_axes([matrix_left/width, 6.05/height, matrix_width/width, .075/height])
    vector_colorbar(dca, dot_cmap, dot_norm, [-1, -.5, 0, .5, 1], tick_size=6.2, decimals=1)
    dca.set_title("Avg. Exp. (scaled)", fontsize=7.5, loc="left", pad=6)
    text(9.05, 6.27, "% Exp.", fontsize=7.5)
    for x, fraction in zip(np.linspace(9.15, 10.65, 5), [0, .25, .5, .75, 1]):
        if fraction > 0:
            legend_axis.scatter([x], [6.07], s=dot_max_area*fraction, c="#555555", edgecolors="none", linewidths=0)
        text(x, 5.82, f"{int(fraction*100)}", fontsize=7, ha="center", color=muted)
    text(9.05, .41, "† Sparse; GSE195500 feature absent.", fontsize=7.2, color=muted)
    text(.40, .17, "Native RNA expression; shared nonlinear color scale. Arrows display the accepted embryonic tree; dashed line spans the UMAP gap.", fontsize=6.5, color=muted)
    text(8.95, .16, "Equal source/library weights; gene-wise z-scores; measured cells only.", fontsize=6.1, color=muted)

    #7. =========== Verify displayed points, physical spacing, text and frozen inputs
    def check_layout(dpi):
        fig.set_dpi(dpi); fig.canvas.draw()
        renderer, canvas = fig.canvas.get_renderer(), fig.bbox
        clipped = []
        for item in fig.findobj(match=lambda item: isinstance(item, Text)):
            if not item.get_visible() or not item.get_text(): continue
            box = item.get_window_extent(renderer)
            if box.width and box.height and (box.x0 < canvas.x0-1 or box.y0 < canvas.y0-1 or box.x1 > canvas.x1+1 or box.y1 > canvas.y1+1):
                clipped.append(item.get_text())
        assert not clipped, "Clipped text: " + repr(clipped)
        a_boxes = [(item.get_text(), item.get_window_extent(renderer)) for item in fig.texts
                   if item.get_visible() and item.get_position()[1]*height > 6.85]
        a_overlaps = [(name_a,name_b) for i,(name_a,box_a) in enumerate(a_boxes)
                      for name_b,box_b in a_boxes[i+1:] if box_a.overlaps(box_b)]
        assert not a_overlaps, "Overlapping workflow labels: " + repr(a_overlaps)
        for feature in feature_axes:
            extent = feature.get_window_extent(renderer)
            assert np.isclose(extent.width, extent.height), "Feature axes are not square"
            assert np.allclose(feature.get_xlim(), xlim) and np.allclose(feature.get_ylim(), ylim)
        assert np.allclose(dots.get_sizes(), dot_max_area*fractions[visible])
        assert np.allclose(dots.get_array(), zscores[visible])
        diameter_px = dots.get_paths()[0].get_extents().width * dots.get_transforms()[:, 0, 0].max()
        centers = ax.transData.transform([[0, 0], [1, 0], [0, 1]])
        gap_ratios = np.array([abs(centers[1, 0]-centers[0, 0]), abs(centers[2, 1]-centers[0, 1])])/diameter_px-1
        assert np.allclose(gap_ratios, column_gap_ratio, atol=.005), gap_ratios
        row_boxes = [t.get_window_extent(renderer) for t in ax.get_yticklabels()]
        assert not any(a.overlaps(b) for i, a in enumerate(row_boxes) for b in row_boxes[i+1:])
        for stage_artist, row_box in zip(stage_artists, row_boxes):
            stage_box = stage_artist.get_window_extent(renderer)
            assert stage_box.x1 + 3 < row_box.x0, "A stage overlaps its state label"
            assert np.isclose((stage_box.y0 + stage_box.y1)/2, (row_box.y0 + row_box.y1)/2, atol=2)
        for feature in feature_axes:
            owned = [item for item in arrow_artists if item.axes is feature]
            assert len(owned) == 2
            for item in owned:
                assert len(item.get_path().vertices) > 100, "A short callout remains in place of a long curve"
        # The state labels must remain outside the feature-map tiles.
        assert min(box.x0 for box in row_boxes) > feature_axes[2].get_window_extent(renderer).x1 + 2
        return {"dpi": dpi, "all_text_inside_canvas": True, "feature_axes_square": True,
                "dot_gap_ratio_xy": gap_ratios.tolist(), "state_labels_clear_of_features": True,
                "stage_state_row_alignment": True, "curved_branch_arrows_per_feature": 2,
                "canvas_pixels": [int(round(canvas.width)), int(round(canvas.height))]}
    layout_checks = [check_layout(preview_dpi)]
    assert input_hashes == {p.name: digest(p) for p in input_paths}
    assert np.allclose(means, source["figure"]["C"]["mean_log10_1p_CPM"], atol=1e-12)
    assert np.allclose(fractions, source["figure"]["C"]["mean_detection_fraction"], atol=1e-12)
    assert np.allclose(zscores, source["figure"]["C"]["gene_wise_zscore"], atol=1e-12)
    return dict(fig=fig, candidates=candidates, orthologs=orthologs, xy=xy, source=source,
        input_hashes=input_hashes, feature_axes=feature_axes, state_order=state_order, mouse=mouse,
        human_ids=human_ids, cells=cells, means=means, fractions=fractions, zscores=zscores,
        measurement_coverage=measurement_coverage, native_t=native_t, stages=stages,
        stage_sets=stage_sets, stage_labels=stage_labels, guides=guides, guide_controls=guide_controls,
        feature_records=feature_records, layout_checks=layout_checks,
        crops={"B":(.20,.05,8.50,6.75), "C":(8.78,.05,4.72,6.75)})
```
