# Fig4 B–D: selected early SGN candidate programs

This notebook reads the formal frozen upstream module and input archives. It retains all **19 programs** across Foxo6, Zfpm2, Neurod2, Nhlh1 and Neurod1, with their exact members, per-cell scores, all main C/D GAMs, association signs and available upstream validation. Changing `PROGRAM_SELECTION` selects saved programs; it does not reselect genes or redefine modules. Prox1 is excluded.

The selected local module IDs, in fixed top/bottom and left/right order, are Foxo6 0/1, Zfpm2 0/2, Neurod2 1/0, Nhlh1 0/2, and Neurod1 0/6. The user's Neurod2 “m1/m2” means both available programs; original archive local IDs remain 0/1. Display aliases keep the reference font and panel spacing; full program names remain in the result archive.

B uses the previously audited complete positive motif-supported SCENIC signatures and shared-background weighted AUCell. Missing activity remains missing: Zfpm2 and Neurod1 have no estimable B activity under the fixed rules. The approved B analysis is unchanged. C/D are frozen RNA-associated modules, except the original early motif-supported Nhlh1 modules; they are not asserted to be subsets of the independently selected B signature. Neurod1's left module is inverse RNA-associated, which is not evidence of direct repression.

GAM uses the same source/library balancing, cubic shrinkage spline, k≤10, gamma=1.5 and GCV.Cp as upstream. All 19 main programs are fitted; previously accepted main C/D predictions are checked for equivalence. Available prior held-out metrics are retained; no held-out fit is invented for the newly displayed modules. Nhlh1's eight-gene program is an exploratory selection with limited membership stability.

Layout is defined on the supplied **1458 × 313** reference coordinate system at 75 logical pixels/inch, rendered from the source at 600 DPI. Outputs are five PNGs and five live-text vector PDFs in `Figures/fig4B-D/SGN-1`, plus exactly two result files in `Processed Data/07_Fig4B-D/SGN-1`. The notebook is the only final code file.

```python
#2. =========== Result serialization and frozen C/D input
def write_result_value(group, name, value):
    if isinstance(value, pd.DataFrame):
        node = group.create_group(name)
        node.attrs['kind'] = 'dataframe'
        node.attrs['columns'] = json.dumps(value.columns.tolist())
        write_result_value(node, '_index', value.index.astype(str).to_numpy())
        for col in value:
            vals = value[col]
            if vals.dtype.kind in 'OUS':
                vals = vals.fillna('').astype(str).to_numpy()
            else:
                vals = vals.to_numpy()
            write_result_value(node, str(col), vals)
    elif isinstance(value, dict):
        node = group.create_group(name)
        node.attrs['kind'] = 'dict'
        for key, item in value.items():
            write_result_value(node, str(key), item)
    elif isinstance(value, (list, tuple)) and (not value or any(isinstance(x, (dict, list, tuple, pd.DataFrame, np.ndarray)) for x in value)):
        node = group.create_group(name)
        node.attrs['kind'] = 'list'
        for i, item in enumerate(value):
            write_result_value(node, str(i), item)
    elif value is None:
        node = group.create_dataset(name, data='', dtype=h5py.string_dtype())
        node.attrs['kind'] = 'none'
    elif isinstance(value, (str, Path)):
        group.create_dataset(name, data=str(value), dtype=h5py.string_dtype())
    else:
        a = np.asarray(value)
        if a.dtype.kind in 'OUS':
            group.create_dataset(name, data=a.astype(str).astype(object), dtype=h5py.string_dtype(),
                                 compression='gzip' if a.ndim else None)
        else:
            group.create_dataset(name, data=a, compression='gzip' if a.ndim else None,
                                 shuffle=bool(a.ndim))

def read_result_value(node):
    if isinstance(node, h5py.Group):
        kind = node.attrs.get('kind', 'dict')
        if kind == 'dataframe':
            columns = json.loads(node.attrs['columns'])
            table = pd.DataFrame({col: read_result_value(node[col]) for col in columns})
            table.index = read_result_value(node['_index'])
            return table
        if kind == 'list':
            return [read_result_value(node[key]) for key in sorted(node, key=int)]
        return {key: read_result_value(node[key]) for key in node}
    if node.attrs.get('kind') == 'none':
        return None
    return node.asstr()[()] if node.dtype.kind in 'OSU' else node[()]
```

```python

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")

# Cleanup must never interpret the project/current directory as a temporary target.
def _release_cleanup_target(path):
    candidate = Path(path).resolve()
    current = Path.cwd().resolve()
    if candidate == current or candidate in current.parents or candidate == Path(candidate.anchor):
        raise RuntimeError("Configure the owned temporary cleanup path before running this section.")
    return path

#3. =========== Frozen programs and complete main-program GAMs
from pathlib import Path
import os, json, hashlib, subprocess
import numpy as np
import pandas as pd
import h5py

# These choose existing modules only; memberships and normalization stay frozen.
PROGRAM_SELECTION = {'Foxo6': [0, 1], 'Zfpm2': [0, 2], 'Neurod2': [1, 0],
                     'Nhlh1': [0, 2], 'Neurod1': [0, 6]}
PROGRAM_TITLES = {'Foxo6': ['Transient developmental genes', 'Broad early-rising genes', 'Early transition-associated genes'], 'Zfpm2': ['Neuronal morphogenesis genes', 'Broad early-rising genes', 'Transient developmental genes', 'Early-high genes, inverse-associated'], 'Neurod2': ['Broad early-rising genes', 'Transient developmental genes'], 'Nhlh1': ['Transient developmental genes', 'Weakly varying genes', 'Early transition-associated genes'], 'Neurod1': ['Postsynaptic receptor regulation genes, inverse-associated', 'Mid-trajectory peaking genes, inverse-associated', 'Late-rising genes (Calb1/Kcna1), inverse-associated', 'Progressively rising genes, mostly inverse-associated', 'Early-high genes', 'Late-rising genes (Cnr1/Kcnq2), inverse-associated', 'Transient differentiation-associated genes']}
PROGRAM_PATH_SEGMENTS = ('CVG', 'T', 'AC', 'Ic')
PROGRAM_K_MAX, PROGRAM_GAM_GAMMA, PROGRAM_GRID_N = 10, 1.5, 250


def _program_read(node):
    if isinstance(node, h5py.Group):
        kind = node.attrs.get('kind', 'dict')
        if kind == 'dataframe':
            columns = json.loads(node.attrs['columns'])
            table = pd.DataFrame({c: _program_read(node[c]) for c in columns})
            table.index = _program_read(node['_index'])
            return table
        if kind == 'list':
            return [_program_read(node[k]) for k in sorted(node, key=int)]
        return {k: _program_read(v) for k, v in node.items()}
    if node.attrs.get('kind') == 'none':
        return None
    return node.asstr()[()] if node.dtype.kind in 'OSU' else node[()]


def _program_file_hash(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as stream:
        for block in iter(lambda: stream.read(4 * 1024 * 1024), b''):
            digest.update(block)
    return digest.hexdigest()


def _program_nearest_jaccard(genes, candidates):
    genes = set(genes)
    return max((len(genes & set(x)) / len(genes | set(x)) for x in candidates), default=0.)


def load_module_result(factor, upstream_modules=None, upstream_inputs=None, selection=None):
    """Load every frozen qualifying module, never the old two-module result.

    Formal upstream H5s are the only inputs. No association/module selection,
    minmax normalization, or RNA score is re-estimated by this loader.
    """
    upstream_modules = Path(upstream_modules or globals()['UPSTREAM_MODULES'])
    upstream_inputs = Path(upstream_inputs or globals()['UPSTREAM_INPUTS'])
    chosen = list(map(int, PROGRAM_SELECTION[factor] if selection is None else selection))
    fields = ('factor', 'obs', 'X_umap', 'modules', 'module_scores', 'primary_path', 'good',
              'fork_t', 'association_audit', 'methods', 'summary', 'sensitivity', 'folds',
              'analysis_pool', 'association_selection', 'association_libraries')
    with h5py.File(upstream_modules, 'r') as h:
        group = h['factors/' + factor]
        result = {key: _program_read(group[key]) for key in fields}
        original = _program_read(group['reviewed_annotations'])
        old_titles = original['reviewed_program_titles_json']
        old_titles = list(old_titles.values()) if isinstance(old_titles, dict) else old_titles
        original_validation = _program_read(group['validation'])
        old_selected = _program_read(group['selected_modules']).astype(int).tolist()
        old_grid = group['fit_grid'][:]
        # These exact old main fits are QA references only, never module>=2 fits.
        result['_upstream_main_reference'] = {'grid': old_grid,
            'fitted': group['fitted'][:, 1:3], 'grid_fitted': group['grid_fitted'][:, 1:3],
            'D_fitted': group['D_fitted'][:], 'D_jobs': group['D_jobs'][:],
            'module_ids': old_selected}
        np.testing.assert_array_equal(result['_upstream_main_reference']['D_jobs'], np.asarray(old_selected) + 1)
        folds = []
        for key in sorted(group['LOO'], key=int):
            fold = group['LOO'][key]
            folds.append({k: _program_read(fold[k]) for k in
                          ('fold', 'heldout_library', 'pool_genes', 'modules')})
    obs = result['obs']
    with h5py.File(upstream_inputs, 'r') as h:
        np.testing.assert_array_equal(obs.index.astype(str), h['cell_ids'].asstr()[:])
        np.testing.assert_array_equal(result['X_umap'], h['X_umap'][:])
    modules, scores = result['modules'], np.asarray(result['module_scores'])
    assert scores.shape == (len(obs), len(modules)) and np.isfinite(scores).all()
    assert [int(m['module_id']) for m in modules] == list(range(len(modules)))
    assert len(chosen) == 2 and len(set(chosen)) == 2 and set(chosen).issubset(range(len(modules)))
    audit = result['association_audit'].set_index('gene', drop=False)
    title_records, metadata, validation_rows = {}, [], []
    for mid, module in enumerate(modules):
        genes = list(map(str, module['genes']))
        assert len(genes) == int(module['n_genes']) and len(set(genes)) == len(genes)
        rho = audit.loc[genes, 'mean_library_rho'].to_numpy(float)
        signs = ['positive' if x > 0 else 'negative' if x < 0 else 'zero' if x == 0 else 'not_evaluable' for x in rho]
        gene_sha = hashlib.sha256('\n'.join(sorted(genes)).encode()).hexdigest()
        previous_title = next((x['title'] for x in old_titles if x['TF'] == factor and x['gene_sha256'] == gene_sha), None)
        module.update(original_display=bool(module['display']), display=mid in chosen, job=mid + 1,
            gene_sha256=gene_sha, association_sign=dict(zip(genes, signs)),
            mean_library_RNA_rho=dict(zip(genes, map(float, rho))),
            upstream_title=previous_title, title=PROGRAM_TITLES[factor][mid])
        title_records[str(mid)] = {'TF': factor, 'module_id': mid, 'gene_sha256': gene_sha,
            'title': module['title'], 'upstream_title': previous_title,
            'name_evidence': 'Exploratory paired-background GO' if (factor, mid) in (('Zfpm2', 0), ('Neurod1', 0))
                             else 'Descriptive observed expression pattern; no new functional enrichment claim'}
        metadata.append({'TF': factor, 'module_id': mid, 'job': mid + 1, 'n_genes': len(genes),
            'selected_for_display': mid in chosen, 'upstream_display': mid in old_selected,
            'title': module['title'], 'gene_sha256': gene_sha,
            'RNA_positive_genes': signs.count('positive'), 'RNA_negative_genes': signs.count('negative'),
            'mean_library_RNA_rho': float(np.nanmean(rho)),
            'graph_sensitivity_min_jaccard': float(module['graph_sensitivity_min_jaccard']),
            'has_upstream_heldout_GAM': bool(original_validation.module_id.eq(mid).any())})
        for fold in folds:
            jac = _program_nearest_jaccard(genes, fold['modules'])
            match = original_validation.loc[original_validation.module_id.eq(mid) & original_validation.fold.eq(fold['fold'])]
            if len(match):
                assert len(match) == 1
                np.testing.assert_allclose(jac, match.iloc[0].membership_jaccard, rtol=0, atol=1e-15)
            validation_rows.append({'TF': factor, 'module_id': mid, 'fold': int(fold['fold']),
                'heldout_library': fold['heldout_library'], 'membership_jaccard': jac,
                'heldout_prediction_spearman': float(match.iloc[0].heldout_prediction_spearman) if len(match) else np.nan,
                'heldout_GAM_available': bool(len(match)),
                'scope': 'Closest membership in frozen training-only partitions; no new LOO fitting'})
    genes_in_modules = set(g for m in modules for g in m['genes'])
    result['association_audit'] = result['association_audit'].loc[result['association_audit'].gene.isin(genes_in_modules)].copy()
    result['selected_modules'] = chosen
    result['reviewed_annotations'] = dict(original, reviewed_program_titles_json=title_records)
    result['program_metadata'] = pd.DataFrame(metadata)
    result['validation'] = pd.DataFrame(validation_rows)
    result['upstream_validation'] = original_validation.rename(columns={'job': 'upstream_job'})
    result['LOO_membership_metadata'] = folds
    result['jobs'] = pd.DataFrame([{'job': 0, 'mode': 'B_activity_placeholder', 'TF': factor, 'module_id': -1}] +
        [{'job': m + 1, 'mode': 'C_main', 'TF': factor, 'module_id': m} for m in range(len(modules))])
    result['summary'].update(upstream_display_module_ids=old_selected,
        display_module_ids=chosen, all_qualifying_modules_retained=True)
    result['methods'] = {'C_D_frozen_upstream': True, 'C_D_input_file': str(upstream_modules),
        'C_D_original_methods': result['methods'], 'all_main_programs_fitted': True,
        'job_contract': 'fitted/grid column0 is separate B activity; module_id m uses column m+1; D_jobs uses those same main-job IDs',
        'upstream_validation_indices': 'upstream_job refers only to the frozen upstream fit, not this main-fit matrix',
        'B_C_D_target_identity': 'B is an independently selected full SCENIC signature; C/D retain frozen expression modules and are not asserted to be subsets of B'}
    return result


PROGRAM_GAM_SOURCE = r'''
suppressPackageStartupMessages({library(mgcv);library(rhdf5)})
a<-commandArgs(trailingOnly=TRUE)
meta<-read.delim(a[1],check.names=FALSE,stringsAsFactors=FALSE)
for(k in c('good','primary','on_path')) meta[[k]]<-tolower(as.character(meta[[k]])) %in% c('true','1')
X<-h5read(a[2],'program_scores'); if(ncol(X)==nrow(meta)) X<-t(X)
stopifnot(nrow(X)==nrow(meta),all(is.finite(X)))
grid<-seq(min(meta$t[meta$primary]),max(meta$t[meta$primary]),length.out=as.integer(a[6]))
P<-matrix(NA_real_,nrow(meta),ncol(X));G<-matrix(NA_real_,length(grid),ncol(X));D<-P
kmax<-as.integer(a[4]);gamma_value<-as.numeric(a[5])
balanced<-function(i){
 d<-meta[i,];sources<-unique(d$study_id);w<-numeric(length(i))
 for(s in sources){m<-which(d$study_id==s);libs<-unique(d$library_id[m])
  for(l in libs){z<-m[d$library_id[m]==l];w[z]<-1/length(sources)/length(libs)/length(z)}}
 w*length(i)/sum(w)
}
segment_predict<-function(train,rows,j){
 k<-min(kmax,length(unique(meta$t[train]))-1)
 if(length(train)<12 || k<3 || sd(X[train,j])<1e-10) return(rep(mean(X[train,j]),length(rows)))
 d<-data.frame(value=X[train,j],t=meta$t[train],w=balanced(train))
 model<-gam(value~s(t,bs='ts',k=k),data=d,weights=w,gamma=gamma_value,method='GCV.Cp')
 as.numeric(predict(model,newdata=data.frame(t=meta$t[rows])))
}
for(j in seq_len(ncol(X))){
 train<-which(meta$primary & is.finite(X[,j]))
 k<-min(kmax,length(unique(meta$t[train]))-1)
 stopifnot(length(train)>=12,k>=3)
 d<-data.frame(value=X[train,j],t=meta$t[train],w=balanced(train))
 model<-gam(value~s(t,bs='ts',k=k),data=d,weights=w,gamma=gamma_value,method='GCV.Cp')
 P[,j]<-as.numeric(predict(model,newdata=data.frame(t=meta$t)))
 G[,j]<-as.numeric(predict(model,newdata=data.frame(t=grid)))
 D[meta$on_path,j]<-P[meta$on_path,j]
 for(segment in unique(meta$seg[!meta$on_path])){
  rows<-which(meta$seg==segment);train<-rows[meta$good[rows] & is.finite(X[rows,j])]
  if(length(train)>=12 && length(unique(meta$t[train]))>=4) D[rows,j]<-segment_predict(train,rows,j)
 }
}
h5createFile(a[3]);h5write(P,a[3],'fitted');h5write(G,a[3],'grid_fitted')
h5write(grid,a[3],'grid');h5write(D,a[3],'D_fitted')
h5write(paste(R.version.string,as.character(packageVersion('mgcv')),as.character(packageVersion('rhdf5')),sep=' | '),a[3],'R_versions')
h5closeAll()
'''


def fit_all_programs(module_results, work_dir, rscript):
    """Fit all frozen programs in one R call, preserving separate B column0.

    No old LOO job is repurposed as a main program. Existing main C/D values
    for modules0/1 are checked against the recomputed values, then preserved
    byte-for-byte so prior approved panels remain exactly reproducible.
    """
    work_dir = Path(work_dir)
    work_dir.mkdir(parents=True, exist_ok=True)
    factors = list(module_results)
    shared = module_results[factors[0]]
    obs = shared['obs'].copy()
    obs['good'] = shared['good']; obs['primary'] = shared['primary_path']
    obs['on_path'] = obs.seg.isin(PROGRAM_PATH_SEGMENTS)
    blocks, offsets, cursor = [], {}, 0
    for factor, result in module_results.items():
        pd.testing.assert_frame_equal(result['obs'], shared['obs'])
        for key in ('good', 'primary_path', 'X_umap'):
            np.testing.assert_array_equal(result[key], shared[key])
        count = len(result['modules'])
        assert result['module_scores'].shape == (len(obs), count)
        offsets[factor] = np.arange(cursor, cursor + count)
        blocks.append(result['module_scores']); cursor += count
    values = np.column_stack(blocks)
    assert values.shape[1] == 19 and np.isfinite(values).all()
    files = {k: work_dir / v for k, v in {'meta': 'program-cells.tsv', 'input': 'program-input.h5',
        'source': 'program-fit.R', 'output': 'program-output.h5'}.items()}
    try:
        obs[['t','seg','study_id','library_id','good','primary','on_path']].to_csv(files['meta'],sep='\t',index=False)
        with h5py.File(files['input'],'w') as h:h.create_dataset('program_scores',data=values)
        files['source'].write_text(PROGRAM_GAM_SOURCE,encoding='utf-8')
        if files['output'].exists(): _release_cleanup_target(files['output']).unlink()
        env=os.environ.copy();env.update(TMPDIR=str(work_dir),TMP=str(work_dir),TEMP=str(work_dir),LC_ALL='C',LANG='C')
        call=subprocess.run([str(rscript),'--vanilla',str(files['source']),str(files['meta']),str(files['input']),
            str(files['output']),str(PROGRAM_K_MAX),str(PROGRAM_GAM_GAMMA),str(PROGRAM_GRID_N)],
            env=env,capture_output=True,text=True,errors='replace')
        if call.returncode: raise RuntimeError(call.stdout+call.stderr)
        with h5py.File(files['output'],'r') as h:
            fitted=h['fitted'][:].T;grid_fitted=h['grid_fitted'][:].T
            grid=h['grid'][:];d_fitted=h['D_fitted'][:].T
            versions=h['R_versions'].asstr()[()].tolist()
        assert fitted.shape==values.shape and np.isfinite(fitted).all()
        assert grid_fitted.shape==(PROGRAM_GRID_N,19) and np.isfinite(grid_fitted).all()
        assert d_fitted.shape==values.shape
        on_path=obs.on_path.to_numpy();np.testing.assert_array_equal(d_fitted[on_path],fitted[on_path])
        for factor,result in module_results.items():
            columns=offsets[factor];count=len(columns);old=result.pop('_upstream_main_reference')
            p=fitted[:,columns].copy();g=grid_fitted[:,columns].copy();d=d_fitted[:,columns].copy()
            np.testing.assert_allclose(grid,old['grid'],rtol=0,atol=1e-12)
            differences={}
            for c,mid in enumerate(old['module_ids']):
                for label,new,previous in [('C',p[:,mid],old['fitted'][:,c]),
                    ('grid',g[:,mid],old['grid_fitted'][:,c]),('D',d[:,mid],old['D_fitted'][:,c])]:
                    np.testing.assert_allclose(new,previous,rtol=0,atol=1e-12,equal_nan=True)
                    finite=np.isfinite(new)&np.isfinite(previous)
                    differences[f'm{mid}_{label}']=float(np.max(np.abs(new[finite]-previous[finite]))) if finite.any() else 0.
                # Preserve accepted main values exactly after verifying same-method equivalence.
                p[:,mid]=old['fitted'][:,c];g[:,mid]=old['grid_fitted'][:,c];d[:,mid]=old['D_fitted'][:,c]
            result.update(fitted=np.column_stack([np.full(len(obs),np.nan),p]),
                grid_fitted=np.column_stack([np.full(len(grid),np.nan),g]),fit_grid=grid.copy(),
                D_fitted=d,D_jobs=np.arange(1,count+1,dtype=int),
                fit_R_source=PROGRAM_GAM_SOURCE,
                fit_methods={'formula':f"value~s(t,bs='ts',k<={PROGRAM_K_MAX})",'gamma':PROGRAM_GAM_GAMMA,'method':'GCV.Cp',
                    'weights':'Equal source, equal libraries per source, equal cells per library; rescaled to n',
                    'C':'All frozen modules on the exact good CVG/T/AC/Ic primary path',
                    'D':'Exact C on primary segments; other native segments separately fitted; unsupported segments NaN',
                    'normalization':'Frozen upstream all-atlas per-gene minmax then unweighted module mean; unchanged',
                    'prediction_clipping':False,'R_versions':versions,'main_program_count':count,
                    'existing_main_fit_max_abs_difference':differences,
                    'existing_main_fits_exactly_preserved':True,
                    'LOO':'Only upstream existing heldout metrics retained; no additional heldout GAMs were fitted'})
            np.testing.assert_array_equal(result['D_fitted'][on_path],result['fitted'][on_path][:,result['D_jobs']])
        return module_results
    finally:
        for path in files.values():
            if path.exists():_release_cleanup_target(path).unlink()


def build_all_program_results(upstream_modules=None, upstream_inputs=None, work_dir=None, rscript=None):
    """Rebuild every main C/D fit from the formal frozen module archive.

    The final notebook can call this without any temporary precomputed cache.
    The returned input/code/parameter fingerprints also support guarded reuse
    of a saved cache; existence alone is never evidence of equivalence.
    """
    upstream_modules = Path(upstream_modules or globals()['UPSTREAM_MODULES'])
    upstream_inputs = Path(upstream_inputs or globals()['UPSTREAM_INPUTS'])
    # Set this storage path independently before running.
    work_dir = Path(work_dir or (ROOT))
    rscript = Path(rscript or globals()['RSCRIPT'])
    fingerprints = {'upstream_modules_file': str(upstream_modules),
        'upstream_modules_sha256': _program_file_hash(upstream_modules),
        'upstream_inputs_file': str(upstream_inputs),
        'upstream_inputs_sha256': _program_file_hash(upstream_inputs),
        'GAM_source_sha256': hashlib.sha256(PROGRAM_GAM_SOURCE.encode()).hexdigest(),
        'GAM_k_max': PROGRAM_K_MAX, 'GAM_gamma': PROGRAM_GAM_GAMMA,
        'GAM_grid_n': PROGRAM_GRID_N, 'GAM_method': 'GCV.Cp',
        'module_memberships': 'Frozen; no reclustering or gene reselection',
        'score_normalization': 'Frozen full-endogenous logCPM then all-atlas gene-minmax/unweighted mean'}
    results = {factor: load_module_result(factor, upstream_modules, upstream_inputs)
               for factor in PROGRAM_SELECTION}
    assert sum(len(result['modules']) for result in results.values()) == 19
    fit_all_programs(results, work_dir, rscript)
    for result in results.values():
        result['program_provenance'] = dict(fingerprints)
    return results


def all_program_results():
    # Reuse only a schema- and input/method-matched final archive.
    expected = {'upstream_modules_sha256': file_sha256(UPSTREAM_MODULES),
                'upstream_inputs_sha256': file_sha256(UPSTREAM_INPUTS),
                'GAM_source_sha256': hashlib.sha256(PROGRAM_GAM_SOURCE.encode()).hexdigest(),
                'GAM_k_max': PROGRAM_K_MAX, 'GAM_gamma': PROGRAM_GAM_GAMMA,
                'GAM_grid_n': PROGRAM_GRID_N, 'GAM_method': 'GCV.Cp'}
    results = None
    if RESULT_FILE.exists() and not RECOMPUTE_PROGRAMS:
        with h5py.File(RESULT_FILE, 'r') as h:
            if h.attrs.get('schema') == 'Fig4BD_all_programs_v2':
                saved = {tf: read_result_value(h['factors/' + tf]) for tf in FACTORS}
                if all(all(r['program_provenance'].get(k) == v for k,v in expected.items()) for r in saved.values()):
                    common = read_result_value(h['shared'])
                    for result in saved.values(): result.update(common)
                    results = saved
    if results is None:
        results = build_all_program_results()
    for tf, result in results.items():
        chosen = list(PROGRAM_SELECTION[tf])
        result['selected_modules'] = chosen
        result['summary']['display_module_ids'] = chosen
        result['fitted'][:, 0] = np.nan
        result['grid_fitted'][:, 0] = np.nan
        for mid, module in enumerate(result['modules']):
            title = PROGRAM_TITLES[tf][mid]
            module.update(display=mid in chosen, title=title)
            result['reviewed_annotations']['reviewed_program_titles_json'][str(mid)]['title'] = title
            rows = result['program_metadata']['module_id'].eq(mid)
            result['program_metadata'].loc[rows,'title'] = title
            result['program_metadata'].loc[rows,'selected_for_display'] = mid in chosen
        assert result['module_scores'].shape[1] == len(result['modules'])
        assert result['fitted'].shape[1] == len(result['modules']) + 1
    assert sum(len(r['modules']) for r in results.values()) == 19
    return results
```

```python

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")

# Cleanup must never interpret the project/current directory as a temporary target.
def _release_cleanup_target(path):
    candidate = Path(path).resolve()
    current = Path.cwd().resolve()
    if candidate == current or candidate in current.parents or candidate == Path(candidate.anchor):
        raise RuntimeError("Configure the owned temporary cleanup path before running this section.")
    return path

#4. =========== Reproducible SCENIC audit and fixed-background AUCell
ACTIVITY_WORKER_SOURCE = '"""Complete existing-GRN motif pruning and shared-axis canonical AUCell worker.\n\nRun with the existing pySCENIC interpreter. The functions can be embedded as\nworker source in the final notebook; no fitted SCENIC result is overwritten.\n"""\n\n# Configure each file/directory placeholder independently; no storage addresses are shipped.\nfrom pathlib import Path\nROOT = Path(".")\n\nfrom pathlib import Path\nimport hashlib\nimport importlib.metadata\nimport json\nimport os\nimport sys\nimport tempfile\nimport time\nimport warnings\n\nsys.dont_write_bytecode = True\nfor _key in (\'OMP_NUM_THREADS\', \'MKL_NUM_THREADS\', \'OPENBLAS_NUM_THREADS\', \'NUMBA_NUM_THREADS\'):\n    os.environ[_key] = \'1\'\n\nTFS = (\'Foxo6\', \'Neurod2\', \'Nhlh1\', \'Neurod1\')\nRANK_THRESHOLDS = (5000, 1500)  # Author CLI primary; prior project sensitivity.\nMIN_MODULE_GENES, NES_THRESHOLD = 20, 3.0\nRHO_THRESHOLD, MOTIF_AUC_THRESHOLD = .03, .05\nPRUNE_WORKERS = 2\nSOURCE_PRIORITY = (\'whole_early\', \'CVG_VGN\', \'late_SS2\', \'late_UMI\')\nDATABASES = (\'mm10_10kbp_up_10kbp_down_full_tx_v10_clust.genes_vs_motifs.rankings.feather\',\n             \'mm10_500bp_up_100bp_down_full_tx_v10_clust.genes_vs_motifs.rankings.feather\')\nANNOTATIONS = \'motifs-v10nr_clust-nr.mgi-m0.001-o0.0.tbl\'\n\n\ndef sha256(path):\n    value = hashlib.sha256()\n    with Path(path).open(\'rb\') as stream:\n        for block in iter(lambda: stream.read(8 * 1024 * 1024), b\'\'):\n            value.update(block)\n    return value.hexdigest()\n\n\ndef plain(value):\n    import numpy as np\n    if isinstance(value, dict):\n        return {str(k): plain(v) for k, v in value.items()}\n    if isinstance(value, (set, frozenset)):\n        return sorted(plain(v) for v in value)\n    if isinstance(value, (tuple, list, np.ndarray)):\n        return [plain(v) for v in value]\n    if isinstance(value, np.generic):\n        return value.item()\n    return value\n\n\ndef put_json(group, name, value):\n    import h5py\n    import numpy as np\n    group.create_dataset(name, data=np.frombuffer(json.dumps(plain(value)).encode(\'utf-8\'), dtype=np.uint8),\n                         compression=\'gzip\', shuffle=True)\n    group[name].attrs[\'kind\'] = \'json\'\n\n\ndef annotation_class(text, motif_id):\n    text = str(text)\n    lower = text.lower()\n    orthology = \'ortholog\' in lower\n    similarity = \'similar motif\' in lower or \'motif similar\' in lower\n    direct = lower.startswith(\'gene is directly annotated\')\n    label = (\'direct\' if direct else \'orthology+similarity\' if orthology and similarity\n             else \'orthology\' if orthology else \'similarity\' if similarity else \'other\')\n    return dict(annotation_class=label, exact_direct_annotation=direct,\n        orthology_supported=orthology, motif_similarity_supported=similarity,\n        complex_motif=(\'tf_pairs__\' in str(motif_id) or \'tfdimers__\' in str(motif_id)))\n\n\ndef load_existing_grn(root, source):\n    import h5py\n    import numpy as np\n    import pandas as pd\n    from scipy import sparse\n    root = Path(root)\n    if source == \'whole_early\':\n        # Set this storage path independently before running.\n        folder = ROOT\n        # Set this storage path independently before running. GSE178931_axes.npz.\n        paths = [ROOT, ROOT,\n                 # Set this storage path independently before running. adjacencies.tsv.\n                 ROOT]\n        with np.load(paths[0]) as axes:\n            keep = axes[\'inference_keep\']\n            genes, cells = axes[\'genes\'][keep], axes[\'cells\']\n        values = sparse.load_npz(paths[1])[:, keep].toarray()\n        adjacency = pd.read_csv(paths[2], sep=\'\\t\', usecols=[\'TF\', \'target\', \'importance\'])\n        metadata = dict(cohort=\'GSE178931 accepted CVG/SGN atlas cells\',\n            stages=[\'E9.5\', \'E11.5\', \'E13.5\'], cells=1161,\n            expression_unit=\'Existing GRN input: log1p(CP10k), full eligible source axis, float32\',\n            no_GRNBoost_refit=True)\n        assert values.shape == (1161, 14901) and len(adjacency) == 1479271\n    elif source in (\'late_SS2\', \'late_UMI\'):\n        study = \'GSE165502\' if source == \'late_SS2\' else \'GSE195500\'\n        # Set this storage path independently before running.\n        folder = ROOT\n        # Set this storage path independently before running.\n        paths = [ROOT / (study + \'_axes.npz\'),\n                 # Set this storage path independently before running.\n                 ROOT / (study + \'_expression.npz\'),\n                 # Set this storage path independently before running. adjacencies.tsv.\n                 ROOT,\n                 # Set this storage path independently before running. Inputs.h5. Candidate input contract; supply the immutable input container (standalone producer not located).\n                 ROOT]\n        with np.load(paths[0]) as axes:\n            keep, all_cells = axes[\'inference_keep\'], axes[\'cells\']\n            genes = axes[\'genes\'][keep]\n        with h5py.File(paths[3], \'r\') as handle:\n            index = pd.Index(handle[\'cell_ids\'].asstr()[:]).get_indexer(all_cells)\n            assert (index >= 0).all()\n            permitted = handle[\'obs/source_state_evaluation_allowed\'][:][index]\n            mixture = handle[\'obs/source_rna_mixture_review_flag\'][:][index]\n        training = permitted & ~mixture\n        cells = all_cells[training]\n        values = sparse.load_npz(paths[1])[training][:, keep].toarray()\n        adjacency = pd.read_csv(paths[2], sep=\'\\t\', usecols=[\'TF\', \'target\', \'importance\'])\n        expected_cells = 1112 if study == \'GSE165502\' else 4616\n        assert len(cells) == expected_cells\n        metadata = dict(cohort=study + \' accepted native-source good training cells\',\n            cells=expected_cells, original_source_cells=len(all_cells),\n            expression_unit=\'Existing GRN input: log1p(CP10k), full eligible source axis, float32\',\n            training_rule=\'source_state_evaluation_allowed and not source_rna_mixture_review_flag\',\n            no_GRNBoost_refit=True)\n    elif source == \'CVG_VGN\':\n        # Set this storage path independently before running. Network.h5. Produced by fig4_VGN_network_analysis.py.\n        paths = [ROOT]\n        with h5py.File(paths[0], \'r\') as handle:\n            genes, cells = handle[\'genes\'].asstr()[:], handle[\'cell_ids\'].asstr()[:]\n            values = handle[\'inference_expression\'][:]\n            tfs = handle[\'TFs\'].asstr()[:]\n            adjacency = pd.DataFrame({\'TF\': tfs[handle[\'grn/TF_index\'][:]],\n                \'target\': genes[handle[\'grn/target_index\'][:]],\n                \'importance\': handle[\'grn/importance\'][:]})\n        metadata = dict(cohort=\'Actual curated CVG490 plus VGN520; SGN excluded\',\n            stages=[\'E9.5\', \'E11.5\', \'E13.5\'], cells=1010,\n            expression_unit=\'Existing GRN input: log10(1+native count/full endogenous total*1e6), float64\',\n            no_GRNBoost_refit=True)\n        assert values.shape == (1010, 15214) and len(adjacency) == 1765865\n    else:\n        raise ValueError(\'Unknown fixed GRN source: \' + source)\n    assert not adjacency.duplicated([\'TF\', \'target\']).any()\n    assert not adjacency.TF.eq(adjacency.target).any()\n    assert set(adjacency.TF).issubset(genes) and set(adjacency.target).issubset(genes)\n    assert np.isfinite(values).all() and np.isfinite(adjacency.importance).all()\n    metadata[\'input_sha256\'] = {str(p.relative_to(root)): sha256(p) for p in paths}\n    metadata[\'all_TF_competition_edges\'] = len(adjacency)\n    metadata[\'input_TFs\'] = int(adjacency.TF.nunique())\n    return pd.DataFrame(values, index=cells, columns=genes), adjacency, metadata\n\n\ndef encode_regulon(regulon, genes, native_axes):\n    full = {str(g): float(w) for g, w in regulon.gene2weight.items()}\n    tf = regulon.transcription_factor\n    targets = {g: w for g, w in full.items() if g != tf}\n    return dict(name=regulon.name, transcription_factor=tf, gene2weight=full,\n        gene2weight_no_self=targets, n_full=len(full), n_no_self=len(targets),\n        self_in_full=tf in full, context=sorted(regulon.context),\n        score=float(regulon.score), nes=float(regulon.nes), annotation=str(regulon.annotation),\n        orthologous_identity=float(regulon.orthologous_identity),\n        similarity_qvalue=float(regulon.similarity_qvalue),\n        inference_coverage=len(set(targets).intersection(genes)) / len(targets) if targets else None,\n        native_measured_coverage={source: dict(matched=len(set(targets).intersection(axis)),\n            full=len(targets), coverage=len(set(targets).intersection(axis)) / len(targets) if targets else None,\n            missing=sorted(set(targets).difference(axis))) for source, axis in native_axes.items()},\n        direct_binding_or_causal_claim_allowed=False)\n\n\ndef run_motif_audit(root, output, sources=SOURCE_PRIORITY):\n    #1. =========== Runtime, resource integrity and native gene coverage axes\n    import h5py\n    import numpy as np\n    import pandas as pd\n    import dask\n    dask.config.set({\'dataframe.query-planning\': False})\n    from ctxcore.rnkdb import FeatherRankingDatabase\n    from pyscenic.utils import modules_from_adjacencies\n    from pyscenic.prune import prune2df, df2regulons\n    warnings.filterwarnings(\'ignore\', message=\'Not prepending group keys\', category=FutureWarning)\n    root, output = Path(root).resolve(), Path(output).resolve()\n    # Set this storage path independently before running.\n    resources = ROOT\n    resource_hashes = {name: sha256(resources / name) for name in (*DATABASES, ANNOTATIONS, \'allTFs_mm.txt\')}\n    native_axes = {}\n    for source in (\'GSE178931\', \'GSE165502\', \'GSE195500\', \'GSE231513\'):\n        # Set this storage path independently before running.\n        with np.load(ROOT / (source + \'_axes.npz\')) as axes:\n            native_axes[source] = set(axes[\'genes\'].tolist())\n    databases = [FeatherRankingDatabase(str(resources / name), name=name) for name in DATABASES]\n    # Set this storage path independently before running. allTFs_mm.txt.\n    tf_catalog = set((ROOT).read_text().splitlines())\n    assert set(TFS).issubset(tf_catalog) and \'Zfpm2\' not in tf_catalog\n    resource_annotations = {tf: [] for tf in (*TFS, \'Zfpm2\')}\n    for chunk in pd.read_csv(resources / ANNOTATIONS, sep=\'\\t\', chunksize=100000):\n        for tf in resource_annotations:\n            hit = chunk.loc[chunk.gene_name.eq(tf)]\n            resource_annotations[tf].extend(hit.to_dict(\'records\'))\n    metadata = dict(TFs=TFS, excluded_candidate=\'Zfpm2\',\n        excluded_reason=\'Absent from official TF pool and motif gene-name annotation; no substitute TF or motif added\',\n        primary_rank_threshold=5000, sensitivity_rank_threshold=1500,\n        source_priority=SOURCE_PRIORITY,\n        scoring_policy=dict(primary_axis=\'Three native source measured intersection, 11736 genes\',\n            E10_excluded_from_scoring=True, full_canonical_signature=True,\n            self_targets_preserved=True, min_full_signature_coverage=.8,\n            post_prune_min_targets=None, first_valid_source_only=True),\n        min_module_genes=MIN_MODULE_GENES, NES_threshold=NES_THRESHOLD,\n        rho_threshold=RHO_THRESHOLD, motif_auc_threshold=MOTIF_AUC_THRESHOLD,\n        thresholds=[.75, .9], top_n_targets=[50], top_n_regulators=[5, 10, 50],\n        activating_only=True, dropout_masking=False, no_final_20_target_gate=True,\n        resources=resource_hashes, workers=PRUNE_WORKERS,\n        runtime_versions={name: importlib.metadata.version(name) for name in\n            (\'pyscenic\', \'arboreto\', \'ctxcore\', \'numpy\', \'pandas\', \'dask\')},\n        limits=[\'Existing GRNs are reused, not refitted under a changed expression unit\',\n            \'TF(+) means positive coexpression context, not direct-only motif annotation\',\n            \'No independent GRN validation or causal TF evidence is supplied by pruning\',\n            \'Native gene axes differ between sources; raw AUC cannot be blindly pooled\'])\n    output.parent.mkdir(parents=True, exist_ok=True)\n    with h5py.File(output, \'a\') as handle:\n        handle.attrs[\'schema\'] = \'candidate_positive_SCENIC_targeted_pruning_audit_v1\'\n        handle.attrs[\'stage\'] = \'running\'\n        if \'methods_json\' in handle:\n            del handle[\'methods_json\']\n        put_json(handle, \'methods_json\', metadata)\n        if \'resource_annotation_evidence_json\' not in handle:\n            put_json(handle, \'resource_annotation_evidence_json\', resource_annotations)\n        if \'executed_source\' in handle:\n            if \'initial_two_source_executed_source\' not in handle:\n                handle.create_dataset(\'initial_two_source_executed_source\',\n                    data=handle[\'executed_source\'].asstr()[()], dtype=h5py.string_dtype())\n            del handle[\'executed_source\']\n        handle.create_dataset(\'executed_source\', data=Path(__file__).read_text(encoding=\'utf-8\'), dtype=h5py.string_dtype())\n    #2. =========== Generate full-competition modules, then prune only four TFs\n    for source in sources:\n        with h5py.File(output, \'r\') as handle:\n            complete = all(source + \'/rank\' + str(rank) + \'/regulons_json\' in handle\n                           for rank in RANK_THRESHOLDS)\n        if complete:\n            print(source, \'complete audit reused; no rerun\', flush=True)\n            continue\n        started = time.perf_counter()\n        frame, adjacency, source_metadata = load_existing_grn(root, source)\n        modules = modules_from_adjacencies(adjacency, frame, thresholds=(.75, .9),\n            top_n_targets=(50,), top_n_regulators=(5, 10, 50), min_genes=MIN_MODULE_GENES,\n            rho_dichotomize=True, keep_only_activating=True,\n            rho_threshold=RHO_THRESHOLD, rho_mask_dropouts=False)\n        selected = [module for module in modules if module.transcription_factor in TFS]\n        source_metadata.update(all_modules=len(modules), tested_modules=len(selected),\n            module_counts={tf: sum(module.transcription_factor == tf for module in selected) for tf in TFS},\n            module_creation_seconds=time.perf_counter() - started)\n        print(source, \'complete modules\', len(modules), \'four-TF modules\', len(selected), flush=True)\n        module_records = [dict(TF=m.transcription_factor, context=sorted(m.context),\n            gene2weight=dict(m.gene2weight), genes=len(m)) for m in selected]\n        candidate_edges = adjacency.loc[adjacency.TF.isin(TFS)]\n        with h5py.File(output, \'a\') as handle:\n            group = handle.require_group(source)\n            put_json(group, \'source_metadata_json\', source_metadata)\n            put_json(group, \'candidate_modules_json\', module_records)\n            put_json(group, \'candidate_adjacency_json\', candidate_edges.to_dict(\'records\'))\n            group.create_dataset(\'genes\', data=frame.columns.to_numpy(dtype=object), dtype=h5py.string_dtype())\n            group.create_dataset(\'cells\', data=frame.index.to_numpy(dtype=object), dtype=h5py.string_dtype())\n        for threshold in RANK_THRESHOLDS:\n            start = time.perf_counter()\n            motif = prune2df(databases, selected, str(resources / ANNOTATIONS),\n                rank_threshold=threshold, auc_threshold=MOTIF_AUC_THRESHOLD,\n                nes_threshold=NES_THRESHOLD, client_or_address=\'dask_multiprocessing\',\n                num_workers=PRUNE_WORKERS)\n            regs = (df2regulons(motif, save_columns=[\'NES\', \'Annotation\',\n                    \'MotifSimilarityQvalue\', \'OrthologousIdentity\']) if not motif.empty else [])\n            rows = []\n            for index, row in motif.iterrows():\n                tf, motif_id = index\n                record = dict(TF=tf, MotifID=motif_id)\n                record.update({\'|\'.join(key) if isinstance(key, tuple) else str(key): plain(value)\n                               for key, value in row.items()})\n                annotation = record.get(\'Enrichment|Annotation\', record.get(\'Annotation\', \'\'))\n                record.update(annotation_class(annotation, motif_id))\n                rows.append(record)\n            encoded = [encode_regulon(reg, set(frame.columns), native_axes) for reg in regs]\n            summary = []\n            for tf in TFS:\n                reg = next((r for r in encoded if r[\'transcription_factor\'] == tf), None)\n                evidence = [r for r in rows if r[\'TF\'] == tf]\n                summary.append(dict(TF=tf, status=\'positive_regulon_available\' if reg else \'no_motif_supported_positive_regulon\',\n                    targets_full=reg[\'n_full\'] if reg else 0, targets_no_self=reg[\'n_no_self\'] if reg else 0,\n                    motif_rows=len(evidence), direct_rows=sum(r[\'exact_direct_annotation\'] for r in evidence),\n                    orthology_rows=sum(r[\'orthology_supported\'] for r in evidence),\n                    similarity_rows=sum(r[\'motif_similarity_supported\'] for r in evidence),\n                    complex_motif_rows=sum(r[\'complex_motif\'] for r in evidence)))\n            calibration = {}\n            if source == \'whole_early\' and threshold == 1500:\n                # Set this storage path independently before running. SCENIC.h5. Produced by fig4_SCENIC_analysis.ipynb.\n                with h5py.File(ROOT, \'r\') as original:\n                    original_regs = json.loads(original[\'state/regulons/GSE178931\'][:].tobytes().decode())\n                frozen = next(r for r in original_regs if r[\'transcription_factor\'] == \'Nhlh1\')\n                actual = next((r for r in encoded if r[\'transcription_factor\'] == \'Nhlh1\'), None)\n                calibration = dict(frozen_targets=len(frozen[\'gene2weight\']),\n                    exact_membership=actual is not None and set(actual[\'gene2weight\']) == set(frozen[\'gene2weight\']),\n                    exact_weights=actual is not None and actual[\'gene2weight\'] == frozen[\'gene2weight\'])\n                assert calibration[\'exact_membership\'] and calibration[\'exact_weights\']\n            with h5py.File(output, \'a\') as handle:\n                group = handle[source].create_group(\'rank\' + str(threshold))\n                put_json(group, \'regulons_json\', encoded)\n                put_json(group, \'motif_evidence_json\', rows)\n                put_json(group, \'summary_json\', summary)\n                put_json(group, \'frozen_Nhlh1_calibration_json\', calibration)\n                group.attrs[\'elapsed_seconds\'] = time.perf_counter() - start\n            print(source, \'rank\', threshold, [(r[\'TF\'], r[\'targets_no_self\']) for r in summary], flush=True)\n        del frame, adjacency, modules, selected\n    #3. =========== Close with provenance and no changes to original sources\n    with h5py.File(output, \'a\') as handle:\n        handle.attrs[\'stage\'] = \'complete\'\n    return output\n\n\n\n\n#4. =========== Unified full-signature AUCell methods\nimport argparse\n\nimport gc\n\nAU_TFS = (\'Foxo6\', \'Zfpm2\', \'Neurod2\', \'Nhlh1\', \'Neurod1\')\n\nAU_STUDIES = (\'GSE178931\', \'GSE165502\', \'GSE195500\', \'GSE231513\')\n\nAU_SCORE_STUDIES = AU_STUDIES[:3]\n\nAU_SOURCE_PRIORITY = (\'whole_early\', \'CVG_VGN\', \'late_SS2\', \'late_UMI\')\n\ndef score_file_sha256(path):\n    value = hashlib.sha256()\n    with Path(path).open(\'rb\') as stream:\n        for block in iter(lambda: stream.read(8 * 1024 * 1024), b\'\'):\n            value.update(block)\n    return value.hexdigest()\n\ndef get_json(dataset):\n    import numpy as np\n    value = dataset[()]\n    if isinstance(value, np.ndarray):\n        value = value.tobytes().decode(\'utf-8\')\n    elif isinstance(value, bytes):\n        value = value.decode(\'utf-8\')\n    return json.loads(value)\n\ndef score_put_json(group, name, value):\n    import h5py\n    group.create_dataset(name, data=json.dumps(value, ensure_ascii=False), dtype=h5py.string_dtype())\n\ndef load_axes(root):\n    import numpy as np\n    axes = {}\n    for study in AU_STUDIES:\n        # Set this storage path independently before running.\n        with np.load(ROOT / (study + \'_axes.npz\')) as z:\n            cells, genes = z[\'cells\'].astype(str), z[\'genes\'].astype(str)\n        assert len(set(cells)) == len(cells) and len(set(genes)) == len(genes)\n        axes[study] = dict(cells=cells, genes=genes)\n    common = set(axes[AU_SCORE_STUDIES[0]][\'genes\'])\n    for study in AU_SCORE_STUDIES[1:]:\n        common.intersection_update(axes[study][\'genes\'])\n    # The source\'s original measured axis determines tie/shuffle input order.\n    primary = np.asarray([g for g in axes[AU_SCORE_STUDIES[0]][\'genes\'] if g in common])\n    assert len(primary) == 11736, len(primary)\n    return axes, primary\n\ndef _csr_rows(group, indices):\n    """Read arbitrary formal CSR rows without loading the entire CSR matrix."""\n    import numpy as np\n    from scipy import sparse\n    pointers = group[\'indptr\'][:]\n    data, cols, lengths = [], [], []\n    for row in indices:\n        a, b = int(pointers[row]), int(pointers[row + 1])\n        data.append(group[\'data\'][a:b]); cols.append(group[\'indices\'][a:b]); lengths.append(b - a)\n    ptr = np.r_[0, np.cumsum(lengths)]\n    return sparse.csr_matrix((np.concatenate(data), np.concatenate(cols), ptr),\n                             shape=(len(indices), int(group.attrs[\'shape\'][1])))\n\ndef validate_inputs(root):\n    """Exhaustively compare native common counts to formal counts, chunked.\n\n    Returns the small extra-gene archive in memory; does not write an H5.\n    Measured true zero remains zero; E10.5 unmeasured extra genes remain NaN.\n    """\n    import numpy as np\n    import h5py\n    import pandas as pd\n    from scipy import sparse\n    root = Path(root).resolve()\n    axes, primary = load_axes(root)\n    # Set this storage path independently before running. Inputs.h5. Candidate input contract; supply the immutable input container (standalone producer not located).\n    inputs = ROOT\n    # Set this storage path independently before running. CSS_3000_Embryonic_Inte_adata.h5ad.\n    formal = ROOT\n    with h5py.File(inputs, \'r\') as f:\n        cells = f[\'cell_ids\'].asstr()[:]\n        source = f[\'obs/study_id\'].asstr()[:]\n        formal_genes = f[\'genes\'].asstr()[:]\n        native_tfs = f[\'native_TFs\'].asstr()[:]\n        native_tf_counts = f[\'native_counts\'][:]\n        native_tf_measured = f[\'native_measured\'][:].astype(bool)\n    assert len(cells) == 7355 and len(set(cells)) == len(cells)\n    assert len(formal_genes) == 11467 and set(formal_genes).issubset(primary)\n    all_native_cells = np.concatenate([axes[s][\'cells\'] for s in AU_STUDIES])\n    assert len(set(all_native_cells)) == 7355 and set(all_native_cells) == set(cells)\n    formal_set = set(formal_genes)\n    extra_genes = np.asarray([g for g in primary if g not in formal_set])\n    assert len(extra_genes) == 269\n    extra_counts = np.full((7355, len(extra_genes)), np.nan)\n    extra_measured = np.zeros(extra_counts.shape, dtype=bool)\n    primary_positive = np.full(7355, np.nan)\n    result = {\'primary_genes\': len(primary), \'formal_genes\': len(formal_genes),\n              \'extra_genes\': len(extra_genes), \'sources\': [], \'all_common_counts_exact\': True,\n              \'scope\': \'Three native source measured intersection; E10.5 excluded from AUCell\',\n              \'source_files\': {}}\n    with h5py.File(formal, \'r\') as f:\n        f_cells = f[\'obs/_index\'].asstr()[:]\n        symbol_key = \'gene_symbol\' if \'gene_symbol\' in f[\'var\'] else \'_index\'\n        gene_node = f[\'var/\' + symbol_key]\n        if isinstance(gene_node, h5py.Dataset):\n            f_genes = gene_node.asstr()[:]\n        elif \'values\' in gene_node:  # AnnData nullable-string-array\n            assert not gene_node[\'mask\'][:].any()\n            f_genes = gene_node[\'values\'].asstr()[:]\n        else:\n            assert (gene_node[\'codes\'][:] >= 0).all()\n            f_genes = gene_node[\'categories\'].asstr()[:][gene_node[\'codes\'][:]]\n        assert np.array_equal(f_cells, cells)\n        assert np.array_equal(f_genes, formal_genes), \'Formal counts gene order mismatch\'\n        for study in AU_STUDIES:\n            # Set this storage path independently before running.\n            paths = [ROOT / (study + suffix) for suffix in (\'_axes.npz\', \'_counts.npz\')]\n            for path in paths:\n                result[\'source_files\'][str(path.relative_to(root))] = score_file_sha256(path)\n            matrix = sparse.load_npz(paths[1]).tocsr()\n            matrix.sum_duplicates(); matrix.sort_indices()\n            assert matrix.shape == (len(axes[study][\'cells\']), len(axes[study][\'genes\']))\n            assert np.issubdtype(matrix.data.dtype, np.integer), (study, matrix.data.dtype)\n            assert (matrix.data >= 0).all()\n            at = pd.Index(cells).get_indexer(axes[study][\'cells\'])\n            assert (at >= 0).all() and np.all(source[at] == study)\n            gene_index = pd.Index(axes[study][\'genes\'])\n            common_at = gene_index.get_indexer(formal_genes)\n            assert (common_at >= 0).all()\n            checked = 0\n            for start in range(0, len(at), 128):\n                stop = min(start + 128, len(at))\n                frozen = _csr_rows(f[\'layers/counts\'], at[start:stop])\n                native = matrix[start:stop, common_at]\n                delta = frozen - native\n                assert delta.nnz == 0 or not np.any(delta.data), \'Native/formal counts differ in \' + study\n                checked += native.shape[0] * native.shape[1]\n            ex = gene_index.get_indexer(extra_genes)\n            available = ex >= 0\n            extra_counts[np.ix_(at, np.flatnonzero(available))] = matrix[:, ex[available]].toarray()\n            extra_measured[np.ix_(at, np.flatnonzero(available))] = True\n            tf_at = gene_index.get_indexer(native_tfs)\n            measured = np.broadcast_to(tf_at >= 0, (len(at), len(native_tfs)))\n            assert np.array_equal(native_tf_measured[at], measured), study\n            if measured.any():\n                observed = matrix[:, tf_at[tf_at >= 0]].toarray()\n                expected = native_tf_counts[at][:, tf_at >= 0]\n                assert np.array_equal(observed, expected), \'Native TF counts mismatch: \' + study\n            if study in AU_SCORE_STUDIES:\n                ix = gene_index.get_indexer(primary)\n                assert (ix >= 0).all()\n                primary_positive[at] = np.asarray((matrix[:, ix] > 0).sum(axis=1)).ravel()\n            result[\'sources\'].append(dict(study=study, cells=len(at), native_genes=matrix.shape[1],\n                raw_dtype=str(matrix.data.dtype), max_count=int(matrix.data.max()),\n                common_values_compared=checked, common_counts_exact=True,\n                native_TF_counts_exact=True, extra_measured_genes=int(available.sum())))\n            del matrix\n            gc.collect()\n    result[\'source_files\'][str(inputs.relative_to(root))] = score_file_sha256(inputs)\n    result[\'source_files\'][str(formal.relative_to(root))] = score_file_sha256(formal)\n    result[\'scoring_cells\'] = int(np.isfinite(primary_positive).sum())\n    result[\'excluded_cells\'] = int(np.isnan(primary_positive).sum())\n    assert result[\'scoring_cells\'] == 7282 and result[\'excluded_cells\'] == 73\n    e10 = source == \'GSE231513\'\n    assert int(extra_measured[e10][0].sum()) == 4 and np.isnan(extra_counts[e10]).sum() == 73 * 265\n    return dict(summary=result, cells=cells, source=source, primary_genes=primary,\n                extra_genes=extra_genes, extra_counts=extra_counts, extra_measured=extra_measured,\n                positive_count=primary_positive, axes=axes)\n\ndef select_signatures(h5_path, primary_genes):\n    """First full canonical positive regulon with >=80% axis coverage.\n\n    No post-pruning target-count threshold; self targets remain if present.\n    """\n    import h5py\n    import numpy as np\n    candidates, selected = [], {}\n    measured = set(primary_genes)\n    with h5py.File(h5_path, \'r\') as f:\n        assert f.attrs.get(\'stage\') == \'complete\', \'All source pruning must be closed and complete\'\n        for source in AU_SOURCE_PRIORITY:\n            path = source + \'/rank5000/regulons_json\'\n            assert path in f, \'Pruning incomplete: \' + path\n            regs = get_json(f[path])\n            for tf in AU_TFS:\n                hits = [r for r in regs if r[\'transcription_factor\'] == tf]\n                assert len(hits) <= 1, (source, tf, len(hits))\n                reg = hits[0] if hits else None\n                weights = ({str(g): float(w) for g, w in reg[\'gene2weight\'].items()} if reg else {})\n                assert all(np.isfinite(w) and w >= 0 for w in weights.values())\n                if reg:\n                    assert reg[\'name\'].endswith(\'(+)\'), \'Only canonical positive regulons are allowed\'\n                matched = sorted(measured.intersection(weights))\n                missing = sorted(set(weights).difference(measured))\n                coverage = len(matched) / len(weights) if weights else None\n                valid = bool(weights) and coverage >= AU_MIN_COVERAGE\n                take = valid and tf not in selected\n                record = dict(TF=tf, source=source, regulon_name=reg[\'name\'] if reg else None,\n                    targets_full=len(weights), self_in_signature=tf in weights,\n                    matched=len(matched), coverage=coverage, missing=missing,\n                    valid=valid, selected=take, post_prune_min_targets=None)\n                candidates.append(record)\n                if take:\n                    selected[tf] = dict(record, gene2weight=weights)\n    return candidates, selected\n\ndef rank_content_sha256(rankings):\n    import numpy as np\n    value = hashlib.sha256()\n    for start in range(0, len(rankings), 128):\n        value.update(np.asarray(rankings.iloc[start:start + 128].values, dtype=\'<u4\').tobytes(order=\'C\'))\n    return value.hexdigest()\n\ndef score(root, h5_path):\n    """Explicit scoring step; appends only activity after success.\n\n    Run only when all four rank5000 pruning blocks are complete. It reuses one\n    common-axis ranking and passes the FULL signature to ctxcore\'s coverage QC.\n    """\n    import h5py\n    import numpy as np\n    import pandas as pd\n    from scipy import sparse\n    from ctxcore.genesig import GeneSignature\n    from ctxcore.recovery import derive_rank_cutoff\n    from pyscenic.aucell import create_rankings, aucell4r\n    root, h5_path = Path(root).resolve(), Path(h5_path).resolve()\n    data = validate_inputs(root)\n    genes, cells, source = data[\'primary_genes\'], data[\'cells\'], data[\'source\']\n    candidates, selected = select_signatures(h5_path, genes)\n    scope = np.isin(source, AU_SCORE_STUDIES)\n    cutoff = derive_rank_cutoff(AU_AUC_THRESHOLD, len(genes))\n    eligible = scope & (data[\'positive_count\'] > cutoff)\n    assert cutoff == 586\n    # The pandas ranking seed shuffles columns once; dropping ineligible rows\n    # first does not change any eligible cell\'s rank or the gene permutation.\n    score_cells = cells[eligible]\n    values = np.empty((len(score_cells), len(genes)), dtype=np.int32)\n    axes = data[\'axes\']\n    for study in AU_SCORE_STUDIES:\n        at = pd.Index(score_cells).get_indexer(axes[study][\'cells\'])\n        keep = at >= 0\n        ix = pd.Index(axes[study][\'genes\']).get_indexer(genes)\n        # Set this storage path independently before running.\n        matrix = sparse.load_npz(ROOT / (study + \'_counts.npz\')).tocsr()\n        values[at[keep]] = matrix[keep][:, ix].toarray()\n        del matrix\n    rankings = create_rankings(pd.DataFrame(values, index=score_cells, columns=genes), seed=AU_SEED)\n    del values\n    gc.collect()\n    assert rankings.shape == (int(eligible.sum()), 11736)\n    rank_hash = rank_content_sha256(rankings)\n    raw_auc = np.full((len(cells), len(AU_TFS)), np.nan)\n    signatures = [GeneSignature(name=tf, gene2weight=selected[tf][\'gene2weight\']) for tf in AU_TFS if tf in selected]\n    if signatures:\n        scores = aucell4r(rankings, signatures, auc_threshold=AU_AUC_THRESHOLD,\n                         noweights=False, normalize=False, num_workers=1)\n        assert scores.index.is_unique\n        scores = scores.reindex(score_cells)\n        assert np.isfinite(scores.values).all() and (scores.values >= 0).all()\n        for j, tf in enumerate(AU_TFS):\n            if tf in selected:\n                raw_auc[eligible, j] = scores[tf].to_numpy()\n    methods = dict(TFs=list(AU_TFS), source_priority=list(AU_SOURCE_PRIORITY), primary_rank_threshold=5000,\n        axis_rule=\'Measured gene intersection of GSE178931/GSE165502/GSE195500, original GSE178931 order\',\n        all_measured_features_retained=True, exclude_all_zero_genes=False, missing_gene_imputation=False,\n        scoring_source_scope=list(AU_SCORE_STUDIES), excluded_source=\'GSE231513\',\n        excluded_reason=\'E10.5 source lacks 265 primary-axis genes; all 73 cells remain NaN\',\n        full_canonical_signature=True, self_targets_preserved=True,\n        min_full_signature_coverage=AU_MIN_COVERAGE, post_prune_min_targets=None,\n        weighted=True, noweights=False, normalize=False, auc_threshold=AU_AUC_THRESHOLD, seed=AU_SEED,\n        rank_cutoff=int(cutoff), cell_gate=\'positive_genes > derive_rank_cutoff(.05, 11736)\',\n        cutoff_zero_based=True, num_workers=1,\n        parameter_attribution=\'Explicit project scoring policy; author executed weight flag has not been independently established\',\n        rank_values_sha256=rank_hash, rank_hash_layout=\'eligible cells, shuffled rank columns, row-major little-endian uint32\',\n        raw_zero_is_valid=True, unsupported_and_ineligible_are_NaN=True,\n        input_validation=data[\'summary\'], prune_file_sha256_before_append=score_file_sha256(h5_path),\n        runtime=sys.executable, python=sys.version,\n        versions={name: importlib.metadata.version(name) for name in\n                  (\'pyscenic\', \'ctxcore\', \'numpy\', \'pandas\', \'scipy\', \'h5py\')})\n    reason = {tf: (\'Selected first full canonical positive regulon with >=80% measured coverage\' if tf in selected\n        else \'No rank5000 source supplied a nonempty canonical positive regulon with >=80% coverage\') for tf in AU_TFS}\n    if \'Zfpm2\' not in selected:\n        reason[\'Zfpm2\'] = \'Absent from TF-pool/motif annotation; strict activity unsupported; no RNA substitution\'\n    measured_sets = {study: set(axes[study][\'genes\']) for study in AU_STUDIES}\n    # No ranking matrix or duplicated full count matrix is persisted.\n    with h5py.File(h5_path, \'a\') as f:\n        assert \'activity\' not in f, \'Existing activity is not overwritten\'\n        g = f.create_group(\'activity\')\n        g.attrs[\'schema\'] = \'common_native11736_fullcanonical_weighted_AUCell_v1\'\n        for name, array in [(\'cell_ids\', cells), (\'source\', source), (\'genes\', genes),\n                            (\'TFs\', np.asarray(AU_TFS)), (\'rank_cell_ids\', score_cells),\n                            (\'rank_gene_order\', rankings.columns.to_numpy()), (\'extra_genes\', data[\'extra_genes\']),\n                            (\'extra_cell_ids\', cells[scope]), (\'measured_source_ids\', np.asarray(AU_STUDIES))]:\n            g.create_dataset(name, data=np.asarray(array, dtype=object), dtype=h5py.string_dtype())\n        for name, array in [(\'raw_auc\', raw_auc), (\'positive_genes\', np.where(scope, data[\'positive_count\'], -1).astype(np.int32)),\n                            (\'source_scope\', scope), (\'rank_gate\', eligible), (\'score_available\', np.isfinite(raw_auc)),\n                            (\'extra_counts\', data[\'extra_counts\'][scope]),\n                            (\'gene_measured_by_source\', np.asarray([[g in measured_sets[study] for g in genes] for study in AU_STUDIES], dtype=bool))]:\n            g.create_dataset(name, data=array, compression=\'gzip\', shuffle=True)\n        g.create_dataset(\'rank_cutoff\', data=int(cutoff))\n        score_put_json(g, \'candidate_sources_json\', candidates)\n        score_put_json(g, \'selected_regulons_json\', {tf: dict(selected.get(tf, {}),\n            status=\'available\' if tf in selected else \'not_estimable\', reason=reason[tf]) for tf in AU_TFS})\n        score_put_json(g, \'status_reason_json\', reason)\n        score_put_json(g, \'metadata_json\', methods)\n        g.create_dataset(\'executed_source\', data=Path(__file__).read_text(encoding=\'utf-8\'), dtype=h5py.string_dtype())\n        g.attrs[\'complete\'] = True\n    return dict(selected={tf:dict(source=record[\'source\'], full=record[\'targets_full\'], matched=record[\'matched\'],\n                                 coverage=record[\'coverage\']) for tf, record in selected.items()},\n                rank_shape=list(rankings.shape), rank_hash=rank_hash,\n                eligible_cells=int(eligible.sum()), in_scope_cells=int(scope.sum()),\n                unsupported=[tf for tf in AU_TFS if tf not in selected])\n\n#4. =========== Complete reproducible command-line workflow\nif __name__ == \'__main__\':\n    parser = argparse.ArgumentParser()\n    parser.add_argument(\'--root\', type=Path, required=True)\n    parser.add_argument(\'--output\', type=Path, required=True)\n    parser.add_argument(\'--mode\', choices=(\'all\', \'prune\', \'score\', \'validate\'), default=\'all\')\n    args = parser.parse_args()\n    args.output.parent.mkdir(parents=True, exist_ok=True)\n    with tempfile.TemporaryDirectory(prefix=\'SCENIC-work-\', dir=args.output.parent) as work:\n        for name in (\'TMP\', \'TEMP\', \'NUMBA_CACHE_DIR\'):\n            os.environ[name] = work\n        if args.mode in (\'all\', \'prune\'):\n            run_motif_audit(args.root, args.output)\n        if args.mode in (\'all\', \'score\'):\n            summary = score(args.root, args.output)\n            print(json.dumps(summary, ensure_ascii=False, indent=2), flush=True)\n        elif args.mode == \'validate\':\n            print(json.dumps(validate_inputs(args.root)[\'summary\'], ensure_ascii=False, indent=2), flush=True)\n'

def ensure_activity_inputs():
    if not ACTIVITY_FILE.exists() and not RECOMPUTE_ACTIVITY and PRIOR_ACTIVITY_FILE.exists():
        assert file_sha256(PRIOR_ACTIVITY_FILE) == PRIOR_ACTIVITY_SHA256, 'Review the prior activity cache before reuse'
        shutil.copyfile(PRIOR_ACTIVITY_FILE, ACTIVITY_FILE)
        assert file_sha256(ACTIVITY_FILE) == PRIOR_ACTIVITY_SHA256
    if ACTIVITY_FILE.exists() and not RECOMPUTE_ACTIVITY:
        with h5py.File(ACTIVITY_FILE,'r') as h:
            assert h.attrs.get('stage') == 'complete'
            assert h['activity'].attrs.get('complete')
            assert h['activity/metadata_json'].shape == ()
        return
    assert SCENIC_PYTHON.is_file(), 'The existing pySCENIC environment is required to rebuild activity'
    # Set this storage path independently before running. SCENIC-worker.py. Use fig4_SGN_activity_worker_source.py, or the identical embedded worker.
    worker = ROOT
    worker.write_text(ACTIVITY_WORKER_SOURCE,encoding='utf-8')
    # Set this storage path independently before running. Activity.pending.h5.
    temporary = ROOT
    if temporary.exists(): _release_cleanup_target(temporary).unlink()
    env = os.environ.copy()
    # Set this storage path independently before running.
    env.update(TMP=str(WORK_DIR),TEMP=str(WORK_DIR),NUMBA_CACHE_DIR=str(ROOT))
    subprocess.run([str(SCENIC_PYTHON),'-B',str(worker),'--root',str(PROJECT_ROOT),
                    '--output',str(temporary)],env=env,check=True)
    with h5py.File(temporary,'r') as h:
        assert h.attrs.get('stage') == 'complete' and h['activity'].attrs.get('complete')
    temporary.replace(ACTIVITY_FILE)


def read_activity_inputs(expected_cell_ids):
    ensure_activity_inputs()
    with h5py.File(ACTIVITY_FILE,'r') as h:
        g=h['activity']
        np.testing.assert_array_equal(g['cell_ids'].asstr()[:],np.asarray(expected_cell_ids).astype(str))
        np.testing.assert_array_equal(g['TFs'].asstr()[:],np.asarray(FACTORS))
        raw=g['raw_auc'][:]
        mask=g['score_available'][:].astype(bool)
        gate=g['rank_gate'][:].astype(bool)
        scope=g['source_scope'][:].astype(bool)
        sources=g['source'].asstr()[:]
        genes=g['genes'].asstr()[:]
        selected=json.loads(g['selected_regulons_json'].asstr()[()])
        audit=json.loads(g['candidate_sources_json'].asstr()[()])
        metadata=json.loads(g['metadata_json'].asstr()[()])
    assert raw.shape==(len(expected_cell_ids),len(FACTORS))
    np.testing.assert_array_equal(mask,np.isfinite(raw))
    assert len(genes)==11736 and int(scope.sum())==7282
    assert np.isnan(raw[~scope]).all() and np.isnan(raw[~gate]).all()
    assert (sources[~scope]=='GSE231513').all()
    assert metadata['full_canonical_signature'] and metadata['self_targets_preserved']
    assert metadata['weighted'] and not metadata['normalize']
    assert metadata['seed']==SEED and metadata['auc_threshold']==.05
    assert metadata['post_prune_min_targets'] is None and metadata['min_full_signature_coverage']==.8
    for j,tf in enumerate(FACTORS):
        if selected[tf]['status']=='available':
            assert selected[tf]['coverage']>=.8
            assert len(selected[tf]['gene2weight'])==selected[tf]['targets_full']
            assert np.isfinite(raw[scope & gate,j]).all() and (raw[scope & gate,j]>=0).all()
        else:
            assert np.isnan(raw[:,j]).all()
    return raw, selected, audit, metadata
```

```python

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")

# Cleanup must never interpret the project/current directory as a temporary target.
def _release_cleanup_target(path):
    candidate = Path(path).resolve()
    current = Path.cwd().resolve()
    if candidate == current or candidate in current.parents or candidate == Path(candidate.anchor):
        raise RuntimeError("Configure the owned temporary cleanup path before running this section.")
    return path

#5. =========== Branch-wise AUCell GAM and fixed display reference
B_GAM_SOURCE = r"""
suppressPackageStartupMessages({library(mgcv); library(rhdf5)})
a<-commandArgs(trailingOnly=TRUE)
meta<-read.delim(a[1],check.names=FALSE,stringsAsFactors=FALSE)
meta$good<-tolower(as.character(meta$good)) %in% c('true','1')
X<-h5read(a[2],'raw_auc'); if(ncol(X)==nrow(meta)) X<-t(X)
stopifnot(nrow(X)==nrow(meta)); P<-matrix(NA_real_,nrow(X),ncol(X))
balanced<-function(i){
 d<-meta[i,]; sources<-unique(d$study_id); w<-numeric(length(i))
 for(s in sources){m<-which(d$study_id==s); libs<-unique(d$library_id[m])
  for(l in libs){z<-m[d$library_id[m]==l]; w[z]<-1/length(sources)/length(libs)/length(z)}}
 w*length(i)/sum(w)
}
for(j in seq_len(ncol(X))){for(s in unique(meta$seg)){
 rows<-which(meta$seg==s & is.finite(X[,j])); train<-rows[meta$good[rows]]
 if(!length(train)) next
 k<-min(10,length(unique(meta$t[train]))-1)
 if(length(train)<12 || k<3 || sd(X[train,j])<1e-10){
  P[rows,j]<-weighted.mean(X[train,j],balanced(train))
 }else{
  d<-data.frame(value=X[train,j],t=meta$t[train],w=balanced(train))
  model<-gam(value~s(t,bs='ts',k=k),data=d,weights=w,gamma=1.5,method='GCV.Cp')
  P[rows,j]<-as.numeric(predict(model,newdata=data.frame(t=meta$t[rows])))
 }
}}
h5createFile(a[3]);h5write(P,a[3],'fitted')
h5write(paste(R.version.string,as.character(packageVersion('mgcv')),sep=' | '),a[3],'R_version')
h5closeAll()
"""

def fit_activity(raw_auc, obs, good):
    # Set this storage path independently before running.
    folder = ROOT
    folder.mkdir(parents=True, exist_ok=True)
    meta = obs[['t','seg','study_id','library_id']].copy()
    meta['good'] = good
    # Set this storage path independently before running. cells.tsv.
    meta.to_csv(ROOT, sep='\t', index=False)
    # Set this storage path independently before running. input.h5.
    with h5py.File(ROOT,'w') as h: h.create_dataset('raw_auc',data=raw_auc)
    # Set this storage path independently before running. fit.R.
    (ROOT).write_text(B_GAM_SOURCE,encoding='utf-8')
    # Set this storage path independently before running. output.h5.
    output = ROOT
    if output.exists(): _release_cleanup_target(output).unlink()
    env = os.environ.copy()
    env.update(TMPDIR=str(folder), TMP=str(folder), TEMP=str(folder), LC_ALL='C', LANG='C')
    # Set this storage path independently before running. fit.R.
    call = subprocess.run([str(RSCRIPT),'--vanilla',str(ROOT),
        # Set this storage path independently before running. cells.tsv.
        str(ROOT),str(ROOT),str(output)],
        env=env,capture_output=True,text=True,errors='replace')
    if call.returncode: raise RuntimeError(call.stdout + call.stderr)
    with h5py.File(output,'r') as h:
        fitted = h['fitted'][:].T
        versions = h['R_version'].asstr()[()].tolist()
    assert fitted.shape == raw_auc.shape
    assert not np.isfinite(fitted[~np.isfinite(raw_auc)]).any()
    return fitted, {'formula':"value~s(t,bs='ts',k<=10)", 'gamma':1.5,
        'method':'GCV.Cp', 'weights':'Equal source, equal libraries, equal cells within library',
        'branches':'Each frozen native segment is fitted independently',
        'unsupported_cells':'Remain NaN; never imputed as zero', 'R_version':versions}

def activity_for_factor(factor, j, raw_auc, fitted, obs, good, selected, candidate_audit, fit_methods):
    values = fitted[:,j]
    mask = np.isfinite(raw_auc[:,j]) & np.isfinite(values)
    status = 'available' if mask.any() else 'not_estimable'
    record = selected.get(factor, {})
    if record is None: record = {}
    # This is an explicit project display reference. The author's regime label
    # is not assumed to mean this exact age/segment rule.
    late = (obs['age'].to_numpy(float)>=14.5) & ~obs['seg'].isin(('CVG','VGN')).to_numpy() & good & mask
    vmax = None
    reference_used = 'Not applicable'
    if status == 'available':
        maximum = float(np.max(values[late])) if late.any() else np.nan
        if np.isfinite(maximum) and maximum > 0:
            vmax, reference_used = maximum, 'Good scored SGN-segment cells at age >= E14.5'
        else:
            vmax = max(float(np.max(values[mask])),1e-6)
            reference_used = 'All scored cells; late reference had no positive fitted maximum'
    reason = '' if status == 'available' else ('No motif annotation or inferred positive regulon' if factor=='Zfpm2'
        else 'No complete positive regulon passed the fixed >=80% mapping rule on the shared measured gene axis')
    signature = record.get('regulon',record.get('signature',record)) if isinstance(record,dict) else {}
    weights = signature.get('gene2weight',{}) if isinstance(signature,dict) else {}
    return {'status':status,'reason':reason,'regulon_name':factor+'(+)' if status=='available' else '',
        'raw':raw_auc[:,j], 'fitted':values, 'display_mask':mask,'display_vmax':vmax,
        'display_reference':reference_used,'display_saturated_cells':int((values[mask]>vmax).sum()) if vmax else 0,
        'display_numeric_clipping':False,'fit_methods':fit_methods,
        'selected_network':record,'candidate_network_audit':candidate_audit,
        'signature_gene_count':len(weights),'signature_self_gene_retained':factor in weights,
        'metric':'Weighted AUCell; top 5%; seed 42; normalize=False',
        'scope':'Three original SCENIC studies on one 11736-gene measured axis. E10 auxiliary cells are unscored and gray.'}
```

```python
"""Published Fig.3d vector ribbons with current module gene names substituted."""

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")

import numpy as np
import matplotlib.patheffects as pe
from matplotlib.path import Path as MplPath
from matplotlib.patches import PathPatch
from matplotlib.font_manager import FontProperties
from matplotlib.textpath import TextPath


# Literature-discussed representatives, not significance or stronger-edge gates.

NETWORK_HIGHLIGHT_EVIDENCE = {'Foxo6': {'Nhlh1': {'reason': 'Highlighted in the reference paper as a neurogenesis-related member.', 'source': 'https://www.nature.com/articles/s41467-022-31580-1'}, 'Sox11': {'reason': 'Representative neuronal differentiation factor; general CNS evidence, not demonstrated SGN fate specification.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC1698453/'}}, 'Zfpm2': {'Isl1': {'reason': 'Representative with direct cochleovestibular neuron differentiation, migration and axon evidence; does not validate a Zfpm2-to-Isl1 edge.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC10883659/'}, 'Sox11': {'reason': 'Neuronal differentiation representative; general CNS evidence.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC1698453/'}}, 'Neurod2': {'Nhlh1': {'reason': 'Reference-paper neurogenesis representative already present in this module.', 'source': 'https://www.nature.com/articles/s41467-022-31580-1'}, 'Neurod6': {'reason': 'Reference-paper neurogenesis representative already present in this module.', 'source': 'https://www.nature.com/articles/s41467-022-31580-1'}, 'Isl1': {'reason': 'Direct cochleovestibular developmental evidence; does not establish the current association as direct regulation.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC10883659/'}}, 'Nhlh1': {'Nhlh2': {'reason': 'Developmental sensory-neuron expression representative; expression evidence is not functional necessity and Nhlh1 knockout effects are not transferred to Nhlh2.', 'source': 'https://pubmed.ncbi.nlm.nih.gov/17004922/'}}, 'Neurod1': {'Myt1l': {'reason': 'Representative neuronal identity factor; not evidence of direct Neurod1 regulation.', 'source': 'https://www.nature.com/articles/nature21722'}, 'Sema3a': {'reason': 'Representative with auditory-neuron connectivity/excitability evidence; not early SGN fate or direct Neurod1-target validation.', 'source': 'https://pubmed.ncbi.nlm.nih.gov/36284453/'}}, 'Prox1': {}}
NETWORK_HIGHLIGHTS = {tf: tuple(genes) for tf, genes in NETWORK_HIGHLIGHT_EVIDENCE.items()}

# Set this storage path independently before running. Single-cell RNA-sequencing analysis of the developing mouse inner ear identifies molecular logic of auditory neuron diversification.pdf.
PAPER_NETWORK_TEMPLATE = {'source': 'Petitpre et al. 2022, Nature Communications, Fig.3d', 'path': 'ROOT', 'sha256': '8dda1b31fe727e1ee88d73932b6b0dd485ab070855ec39bd4d5587e1b2826bd4', 'page_1based': 6, 'bbox': [302.0, 208.0, 466.0, 286.0], 'center': [384.737473256552, 248.26325348336002], 'base_font_size': 4.34579991999999, 'red_font_size': 4.1339998199999854, 'red_rgb': [0.72941, 0.0, 0.1098], 'guides': [{'path': [['m', [383.54000854, 219.98101855999994]], ['l', [383.54000854, 221.68801856999994]]], 'pts': [[383.54000854, 219.98101855999994], [383.54000854, 221.68801856999994]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 219.98101855999994, 'bottom': 221.68801856999994, 'linewidth': 0.329, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 222.28698780000002]], ['l', [383.54000854, 242.96197558999995]]], 'pts': [[383.54000854, 222.28698780000002], [383.54000854, 242.96197558999995]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 222.28698780000002, 'bottom': 242.96197558999995, 'linewidth': 0.329, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[3.5955, 0.59925997], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 254.99102832999995]], ['l', [383.54000854, 256.5500283399999]]], 'pts': [[383.54000854, 254.99102832999995], [383.54000854, 256.5500283399999]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 254.99102832999995, 'bottom': 256.5500283399999, 'linewidth': 0.30000001, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 257.04101612]], ['l', [383.54000854, 274.01001025999994]]], 'pts': [[383.54000854, 257.04101612], [383.54000854, 274.01001025999994]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 257.04101612, 'bottom': 274.01001025999994, 'linewidth': 0.30000001, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[2.95099997, 0.49184], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 274.25601245999997]], ['l', [383.54000854, 275.81401242]]], 'pts': [[383.54000854, 274.25601245999997], [383.54000854, 275.81401242]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 274.25601245999997, 'bottom': 275.81401242, 'linewidth': 0.30000001, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}], 'slots': {'left': [{'visual_slot': 'St18', 'point': [346.20866888772, 262.10511139336006], 'bbox': [341.73901367, 259.93221143336007, 350.67832410544, 264.27801135336006], 'path': [['m', [384.55599975, 247.91198780000002]], ['c', [382.62599981, 252.00998784], [377.81299972, 255.82098771000005], [372.05499935, 258.58798744]], ['c', [366.29899979000004, 261.35598804000006], [359.60300064, 263.08798744], [353.93199921, 263.086988]], ['c', [351.02399825000003, 263.086988], [348.38800049, 262.63198806], [346.2879982, 261.63998748000006]], ['l', [346.11899948, 261.99698782999997]], ['c', [348.29299927, 263.02298786000006], [350.98299789000004, 263.48098804000006], [353.93199921, 263.48198750000006]], ['c', [359.68199921, 263.48098804000006], [366.42000008, 261.7349877500001], [372.22699928000003, 258.94398738000007]], ['c', [378.03099965, 256.14898826], [382.9089998, 252.31698800000004], [384.91299975000004, 248.0809878]], ['l', [384.55599975, 247.91198780000002]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Nhlh1', 'point': [356.088734364884, 244.5250502825599], 'bbox': [350.6889038, 242.4580503725599, 361.48856492976796, 246.5920501925599], 'path': [['m', [384.61499023, 247.83898975]], ['c', [382.29399013, 249.61298979000003], [378.49899006, 250.44698978999998], [374.39698982, 250.44498968999994]], ['c', [367.40699005, 250.45298982999998], [359.5329895, 248.02498975000003], [356.52198982, 244.05298974000004]], ['l', [356.20698928999997, 244.29298974999995]], ['c', [359.38298987999997, 248.43698973000005], [367.31798935, 250.83198977999996], [374.39698982, 250.83998967000002]], ['c', [378.55099011, 250.83798981999996], [382.41299033, 250.01098967999997], [384.85399022999997, 248.15398975000005]], ['l', [384.61499023, 247.83898975]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Rd3', 'point': [322.756078766624, 226.05335357336003], 'bbox': [318.77011108, 223.88045361336003, 326.742046453248, 228.22625353336002], 'path': [['m', [384.6499939, 247.81799365999996]], ['c', [381.24699379000003, 249.42599369999994], [376.98399402, 250.16099358999998], [372.28799344000004, 250.16099358999998]], ['c', [363.14399339000005, 250.16199375999997], [352.3639946, 247.37599365999995], [343.0929947, 242.91899349999994]], ['c', [333.81999207, 238.46499397999992], [326.06099320000004, 232.33299401], [322.9339943, 225.7029938899999]], ['l', [322.57699204000005, 225.87199451999993]], ['c', [325.77799226, 232.63999320999994], [333.60199357000005, 238.79299404999995], [342.92199326, 243.27499344]], ['c', [352.24299240000005, 247.75399366], [363.06499482000004, 250.5549936399999], [372.28799344000004, 250.55599356999994]], ['c', [377.0249939, 250.55599356999994], [381.34199382, 249.81699360999994], [384.8189939, 248.17499365999993]], ['l', [384.6499939, 247.81799365999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'St6galnac5', 'point': [345.172961909844, 238.51819732336003], 'bbox': [334.2999878, 236.34529736336003, 356.045936019688, 240.69109728336002], 'path': [['m', [384.63101196, 247.82800341999996]], ['c', [381.88601208, 249.52700341999991], [377.98401213, 250.3030033199999], [373.67301178, 250.30200338999998]], ['c', [368.28301239, 250.3030033199999], [362.25501251000003, 249.09300339999993], [357.04801178, 247.01500339999996]], ['c', [351.83901215000003, 244.93900349], [347.45501328, 241.98800326999992], [345.33501053000003, 238.5520034]], ['l', [344.999012, 238.76000359]], ['c', [347.19801331, 242.31200361999993], [351.64701081000004, 245.28300334999994], [356.90201186, 247.38200341999993]], ['c', [362.15801239, 249.47900343999993], [368.22401238000003, 250.69700336999995], [373.67301178, 250.69700336999995]], ['c', [378.03101206, 250.69600344999992], [381.99501205, 249.92000341999994], [384.83901196000005, 248.16500342999996]], ['l', [384.63101196, 247.82800341999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ntrk1', 'point': [325.02116664444, 241.59320952335997], 'bbox': [319.83010864, 239.42030956335998, 330.21222464887995, 243.76610948335997], 'path': [['m', [384.61099243, 247.84198046999995]], ['c', [379.48399258, 251.94298029999993], [370.88299275, 253.86098052], [361.68199158, 253.85898066999994]], ['c', [347.34299088, 253.86598062999997], [331.55499268, 249.20198047999997], [325.17399216, 241.19598056999996]], ['l', [324.8659935, 241.44298030999994]], ['c', [331.40899277, 249.61398041999996], [347.2599907, 254.24698066999997], [361.68199158, 254.25398064]], ['c', [370.93699265000004, 254.25098038], [379.60499239, 252.34198046999995], [384.85799243, 248.1509804499999]], ['l', [384.61099243, 247.84198046999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Egf', 'point': [335.78853088996004, 244.15320709336004], 'bbox': [332.52700805, 241.98030713336004, 339.05005372992, 246.32610705336003], 'path': [['m', [384.60598754, 247.84698535999996]], ['c', [380.19398737, 251.64298536], [372.34598732, 253.4589853399999], [364.09498787, 253.45598555999993]], ['c', [352.93998718, 253.46198513000002], [341.05598831, 250.13998533999995], [335.93298721, 244.16198541999995]], ['l', [335.63398742, 244.4199853199999]], ['c', [340.91498564, 250.54998541999998], [352.86398697, 253.8449855], [364.09498787, 253.85098554]], ['c', [372.40198707, 253.84798528], [380.31898737, 252.04398537999998], [384.86298754, 248.14598535999994]], ['l', [384.60598754, 247.84698535999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Clvs1', 'point': [336.701124429984, 255.54914215335998], 'bbox': [331.26800537, 253.37624219335999, 342.134243489968, 257.72204211336], 'path': [['m', [384.57598876, 247.87902880000001]], ['c', [379.39998864999995, 254.92302894], [365.57698822, 259.28602838000006], [353.48098945, 259.27702855999996]], ['c', [346.81998825, 259.27902841], [340.68598937999997, 257.95902872], [336.81698988999995, 255.10202884]], ['l', [336.58198927999996, 255.42002867999997]], ['c', [340.56798934, 258.35502863], [346.76898955999997, 259.67102861], [353.48098945, 259.67202901999997]], ['c', [365.66898918, 259.66402863999997], [379.55198859999996, 255.33602857000005], [384.89398874, 248.11402880000003]], ['l', [384.57598876, 247.87902880000001]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Raver2', 'point': [351.98725317106005, 222.76331451336], 'bbox': [344.98291015, 220.59041455336, 358.99159619212, 224.93621447336], 'path': [['m', [384.70999145, 247.79998829]], ['c', [384.20099145, 247.86398828999995], [383.67499149, 247.89498829000001], [383.13399148, 247.89498829000001]], ['c', [377.3329916, 247.89898829000003], [369.91099167000004, 244.30598834], [363.79399109, 239.31598809000002]], ['c', [357.67199135000004, 234.33098842999993], [352.85699081, 227.94598818999998], [352.18299103000004, 222.47198918000004]], ['l', [351.79099274000004, 222.52098896999996]], ['c', [352.50099182, 228.17798855], [357.37999153000004, 234.59198811], [363.54499054, 239.62198783999997]], ['c', [369.71499157, 244.64698818], [377.17499161, 248.2849883], [383.13399148, 248.28998829]], ['c', [383.68999147, 248.28998829], [384.23199147, 248.25798828999996], [384.75899145, 248.19198827000002]], ['l', [384.70999145, 247.79998829]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Foxo6', 'point': [331.56035424882396, 219.52234771335998], 'bbox': [325.5209961, 217.34944775335998, 337.599712397648, 221.69524767335997], 'path': [['m', [384.67800903, 247.80700732999992]], ['c', [382.77300906, 248.37600732999988], [380.65000916, 248.64300734999995], [378.37800883, 248.64300734999995]], ['c', [369.85500908, 248.64500732999988], [359.24900818, 244.87600732999988], [350.2530098, 239.30200721999995]], ['c', [341.2550087, 233.73200752999992], [333.87100983, 226.34800769999993], [331.74401093, 219.2010064299999]], ['l', [331.36600876, 219.31400729999996]], ['c', [333.55200958, 226.62300731999994], [341.00000763, 234.03100730999995], [350.04500961, 239.63800765999997]], ['c', [359.09300995, 245.2410073499999], [369.74100876, 249.03600729999994], [378.37800883, 249.03800726999987]], ['c', [380.68100906, 249.03800726999987], [382.84200907, 248.76700729999993], [384.79100903, 248.18600732999994]], ['l', [384.67800903, 247.80700732999992]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Shh', 'point': [347.06160342884004, 250.46924712336], 'bbox': [343.1960144, 248.29634716336, 350.92719245768, 252.64214708336], 'path': [['m', [384.58700561, 247.86499072999993]], ['c', [380.7070055, 252.23799086999998], [371.95900536, 254.62499094999998], [363.6190052, 254.6199908399999]], ['c', [357.00400543, 254.6229906299999], [350.64800644, 253.12199068999996], [347.18900681, 250.03599072999998]], ['l', [346.92700577, 250.3309908099999]], ['c', [350.52200699, 253.52399063999997], [356.94600487, 255.01099062999992], [363.6190052, 255.01499082999987]], ['c', [372.03200531, 255.00899077999998], [380.84500551, 252.6469907899999], [384.88200559, 248.12799073999997]], ['l', [384.58700561, 247.86499072999993]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Cecr2', 'point': [341.931003113304, 228.58631256336002], 'bbox': [336.13500976, 226.41341260336003, 347.726996466608, 230.75921252336002], 'path': [['m', [384.6659851, 247.8109746]], ['c', [382.79998505, 248.50097460000006], [380.62498522, 248.82197463], [378.2609849, 248.82197463]], ['c', [371.64598465, 248.82397461000005], [363.56498527, 246.30497456], [356.6639843, 242.46397448000005]], ['c', [349.76098633, 238.62597418999997], [344.042984, 233.4569745], [342.11598587000003, 228.22997523000004]], ['l', [341.74598694, 228.36697435999997]], ['c', [343.73898697, 233.74597502000006], [349.5219841, 238.93797445999996], [356.47198486, 242.80897475000006]], ['c', [363.42398453, 246.67697466000004], [371.54798507, 249.21597456999996], [378.2609849, 249.21697460999997]], ['c', [380.65998506, 249.21697460999997], [382.87998509, 248.89197457], [384.8029851, 248.1819746]], ['l', [384.6659851, 247.8109746]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Neurod6', 'point': [355.255089154796, 233.17708885255996], 'bbox': [347.3269043, 231.11008894255997, 363.183274009592, 235.24408876255995], 'path': [['m', [384.6749878, 247.80798388999995]], ['c', [383.58998777, 248.14998388999993], [382.36598779, 248.3109839099999], [381.04898787999997, 248.3109839099999]], ['c', [376.45298768, 248.31298388999994], [370.73598768, 246.34498383999994], [365.88198854, 243.41198398999995]], ['c', [361.02498819, 240.48198367999998], [357.03598787, 236.5799842099999], [355.84898759, 232.79198408999991]], ['l', [355.47198869, 232.91098358]], ['c', [356.71998788, 236.85798408999995], [360.77398874, 240.78498413], [365.67798806, 243.7499838099999]], ['c', [370.5839882, 246.71098388999997], [376.34298803, 248.70398390999992], [381.04898787999997, 248.70598388999997]], ['c', [382.39798785, 248.70598388999997], [383.66198779, 248.54198391], [384.7939878, 248.1849838899999]], ['l', [384.6749878, 247.80798388999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ppp1r17', 'point': [338.6215513488, 268.92810942336007], 'bbox': [330.4079895, 266.7552094633601, 346.8351131976, 271.10100938336], 'path': [['m', [384.5499878, 247.92700244999992]], ['c', [382.46198775, 253.54200221999997], [376.30998803, 259.0780024699999], [368.88598824999997, 263.18700266999997]], ['c', [361.46598817, 267.2990021899999], [352.7789879, 269.9910016199999], [345.6579895, 269.9900021699999]], ['c', [343.08498765, 269.9900021699999], [340.71698762, 269.63900234999994], [338.68798829, 268.8820023699999]], ['l', [338.5499878, 269.2520032099999]], ['c', [340.63598634, 270.0300030899999], [343.04898835, 270.3850026299999], [345.6579895, 270.3850026299999]], ['c', [352.87598802, 270.3830018199999], [361.6049881, 267.6700024799999], [369.07798768, 263.53300238999987]], ['c', [376.54798795, 259.3910021899999], [382.76498784, 253.83200264999994], [384.9199878, 248.06500244999995]], ['l', [384.5499878, 247.92700244999992]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Klc3', 'point': [313.35505509218797, 245.62421539336003], 'bbox': [309.1281128, 243.45131543336004, 317.581997384376, 247.79711535336003], 'path': [['m', [384.6000061, 247.85199023999996]], ['c', [377.81400633, 254.16299011], [364.97900582, 257.2509904], [351.80200577, 257.24699068999996]], ['c', [336.5300064, 257.25198985], [320.80100631999994, 253.1039900899999], [313.49700928, 245.22599031999994]], ['l', [313.20700836, 245.49499013000002]], ['c', [320.66500472999996, 253.51199007999992], [336.46000671999997, 257.6369900799999], [351.80200577, 257.6419902]], ['c', [365.03900527999997, 257.63799049], [377.94200609999996, 254.56699039], [384.8690061, 248.14099023999995]], ['l', [384.6000061, 247.85199023999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ppp1r14a', 'point': [324.71179432656004, 252.19410064336], 'bbox': [315.2901001, 250.02120068336, 334.13348855312, 254.36700060336], 'path': [['m', [384.58499145, 247.86700488999998]], ['c', [378.29899168000003, 255.14800502000003], [363.83999062000004, 259.16400481999995], [350.22599030000003, 259.15800525]], ['c', [340.01898955, 259.1610045599999], [330.28799057000003, 256.90500498999995], [324.83999252, 252.18400480000003]], ['l', [324.58098984000003, 252.48300505999998]], ['c', [330.16299057000003, 257.3060050099999], [339.96199035, 259.5500045], [350.22599030000003, 259.55300476]], ['c', [363.91599083, 259.54700518999994], [378.43899155, 255.55800487], [384.88399145, 248.12600489]], ['l', [384.58499145, 247.86700488999998]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Klhl25', 'point': [331.8726626188, 232.49829986336], 'bbox': [325.83200073, 230.32539990336, 337.9133245076, 234.67119982335998], 'path': [['m', [384.64099121, 247.82299853999996]], ['c', [381.36599111, 249.59399855999993], [377.02899123, 250.39999844], [372.22899150999996, 250.39999844]], ['c', [364.69799041, 250.39999844], [356.03499030999996, 248.41699852], [348.56399154999997, 245.13799859999995]], ['c', [341.09199143, 241.86099863999993], [334.81698989999995, 237.27999832], [332.04199219, 232.13899852999998]], ['l', [331.69499207, 232.32699825999998]], ['c', [334.54698944, 237.5959983099999], [340.88599015, 242.19699861999993], [348.40499115, 245.4999985899999]], ['c', [355.92499161, 248.79999853999993], [364.62899208, 250.79399848999992], [372.22899150999996, 250.79499841999996]], ['c', [377.07199145, 250.79399848999992], [381.46899127999995, 249.98499847999994], [384.82899120999997, 248.16999853999994]], ['l', [384.64099121, 247.82299853999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Necab2', 'point': [318.41408894212, 267.16614654336], 'bbox': [310.92410278, 264.99324658336, 325.90407510424, 269.33904650336], 'path': [['m', [384.56201171, 247.90002489999995]], ['c', [381.04601168, 254.22502469999995], [373.26701163999996, 259.80202532], [364.01501083, 263.7800250099999]], ['c', [354.76401138, 267.76002551], [344.04501340999997, 270.1470246399999], [334.66701126, 270.1470246399999]], ['c', [328.40001297, 270.1470246399999], [322.73401259999997, 269.0790257499999], [318.50901031, 266.7280240099999]], ['l', [318.31700896999996, 267.07302523]], ['c', [322.63001251, 269.47102404], [328.35601042999997, 270.54102563999993], [334.66701126, 270.54202509999993]], ['c', [344.11101149999996, 270.54102563999993], [354.87101172999996, 268.1430249299999], [364.17101097, 264.1430249299999]], ['c', [373.47001171, 260.14002467], [381.31401180999995, 254.54302501999996], [384.90701171, 248.09302489999993]], ['l', [384.56201171, 247.90002489999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Dscaml1', 'point': [357.05660059441595, 256.80414702336], 'bbox': [348.60488891, 254.63124706335998, 365.50831227883197, 258.97704698335997], 'path': [['m', [384.55999755, 247.90399218999994]], ['c', [383.13399755, 250.61099219999994], [379.83199738999997, 253.06199218999996], [375.88999748, 254.81699228999992]], ['c', [371.94899749, 256.57499170999995], [367.37299728, 257.64499235999995], [363.42199707, 257.64399196]], ['c', [360.99599838, 257.64399196], [358.80799675, 257.24099208999996], [357.1479969, 256.36299181999993]], ['l', [356.96399687999997, 256.71199179999996]], ['c', [358.70599745, 257.63199187], [360.95299719999997, 258.0379920099999], [363.42199707, 258.03899241]], ['c', [367.44299698, 258.0379920099999], [372.05999755, 256.95699263999995], [376.04999733, 255.17799234999995]], ['c', [380.03899765, 253.3969922199999], [383.40699754999997, 250.92499208999993], [384.90899755, 248.08899218999989]], ['l', [384.55999755, 247.90399218999994]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Cacna1f', 'point': [326.950596540992, 275.43103912335994], 'bbox': [318.85610961, 273.25813916335994, 335.045083471984, 277.60393908335993], 'path': [['m', [384.54901123, 247.92901659999995]], ['c', [381.96901131000004, 255.09601640999995], [374.19801140000004, 262.2010169099999], [364.81201172000004, 267.49001741999996]], ['c', [355.42801095000004, 272.78301669999996], [344.43400956000005, 276.26701593999996], [335.4480095, 276.2660165]], ['c', [332.34601212, 276.2660165], [329.48501206000003, 275.85101747999994], [327.01301193, 274.95901727999996]], ['l', [326.87901307000004, 275.33101701999993]], ['c', [329.40600968, 276.24201631999995], [332.31101227, 276.6600175], [335.4480095, 276.66101695]], ['c', [344.53401185, 276.65901613999995], [355.57001113, 273.15401697999994], [365.00601197000003, 267.83401728]], ['c', [374.43801117000004, 262.51201677999995], [382.27501131, 255.38401651999993], [384.92001123, 248.06401659999995]], ['l', [384.54901123, 247.93001659999993]], ['l', [384.54901123, 247.92901659999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Kcne1l', 'point': [328.69355004772, 263.50513580335996], 'bbox': [322.05099487, 261.33223584335997, 335.33610522544, 265.67803576335996], 'path': [['m', [384.56298828, 247.8980107499999]], ['c', [381.58098817, 253.1150107499999], [375.09998798000004, 257.68301058999987], [367.39598847, 260.93001032999996]], ['c', [359.69298745000003, 264.17801142999986], [350.76998902, 266.11201144999995], [342.91898728, 266.1110100899999]], ['c', [337.41698838, 266.11201144999995], [332.44298935, 265.16000987999996], [328.78899002, 263.0680108199999]], ['l', [328.59298706000004, 263.4110102799999]], ['c', [332.33698655, 265.55301142999986], [337.37198638, 266.5060105499999], [342.91898728, 266.5060105499999]], ['c', [350.83398818, 266.5060105499999], [359.79698754000003, 264.5620102999999], [367.54898835, 261.2940106499999]], ['c', [375.29998875, 258.0230107499999], [381.84498835, 253.43501091999997], [384.90598828000003, 248.09501074999991]], ['l', [384.56298828, 247.8980107499999]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Mfap4', 'point': [348.57326316882404, 277.13901642336], 'bbox': [342.53390502, 274.96611646336, 354.612621317648, 279.31191638335997], 'path': [['m', [384.53799438, 247.97601367999994]], ['c', [383.88999437999996, 254.18201352999995], [378.54999447, 261.50101328999995], [371.71799468, 267.23701335]], ['c', [364.89199448, 272.97901392999995], [356.57699393999997, 277.14101458999994], [350.06999587999996, 277.13601351999995]], ['c', [349.56599425999997, 277.13601351999995], [349.07199477999995, 277.11101389999993], [348.59099577999996, 277.06001330999993]], ['l', [348.54999541999996, 277.4530129599999]], ['c', [349.04499434999997, 277.50501299999996], [349.55299377, 277.53101397999995], [350.06999587999996, 277.53101397999995]], ['c', [356.73999405, 277.52601290999996], [365.09299468, 273.3170132799999], [371.97199439999997, 267.5400128499999]], ['c', [378.84599447, 261.75701381], [384.2499944, 254.40801383999997], [384.93099437999996, 248.01701368]], ['l', [384.53799438, 247.97601367999994]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Kcnh6', 'point': [315.9832609866, 234.44025542336], 'bbox': [309.8230896, 232.26735546336, 322.14343237319997, 236.61315538335998], 'path': [['m', [384.625, 247.83197069999994]], ['c', [379.5079999, 251.23897075999992], [371.94699955, 252.78997087999994], [363.64100075, 252.78797054999995]], ['c', [345.40999985, 252.7979707799999], [323.6080017, 245.3139708199999], [316.1439972, 234.0659708999999]], ['l', [315.81500245, 234.28497076999997]], ['c', [323.44699860000003, 245.72697068999992], [345.3069992, 253.1729707799999], [363.64100075, 253.18297052999992]], ['c', [371.99499989, 253.1819706], [379.62200022, 251.63297080999996], [384.844, 248.1609706999999]], ['l', [384.625, 247.83197069999994]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Btbd17', 'point': [355.26389850322, 268.97010162336005], 'bbox': [348.37689208, 266.79720166336006, 362.15090492644, 271.14300158336], 'path': [['m', [384.54000854, 247.96197558999995]], ['c', [383.74100852, 252.56897545999993], [379.46600866, 257.77397585999995], [374.10800838, 261.7969756199999]], ['c', [368.75400829, 265.8239751], [362.32200813, 268.67597629], [357.28400803, 268.67197466999994]], ['c', [356.59200859, 268.67197466999994], [355.92700768, 268.61897517], [355.29400826, 268.50797510999996]], ['l', [355.22600937, 268.8969750499999]], ['c', [355.88400841, 269.01197482], [356.57200813, 269.06697512999995], [357.28400803, 269.06697512999995]], ['c', [362.46600914, 269.06297541999993], [368.93900871, 266.17297603], [374.34500884, 262.11197519999996]], ['c', [379.74700831999996, 258.0469756199999], [384.08700854, 252.81497574999992], [384.92900854, 248.02997558999994]], ['l', [384.54000854, 247.96197558999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ebf2', 'point': [342.18483272487, 217.1121474525599], 'bbox': [337.93301391, 215.0451475425599, 346.43665153973996, 219.1791473625599], 'path': [['m', [384.70498657, 247.80102588]], ['c', [383.90998656, 247.92102588], [383.08198655, 247.97902588], [382.22698665, 247.97902588]], ['c', [374.84298611, 247.98302588], [365.42998696, 243.62802603], [357.62598610000003, 237.53002598]], ['c', [349.81798554, 231.43702556000005], [343.62198639, 223.59702541000001], [342.59298706000004, 216.73302507999995]], ['l', [342.20198822000003, 216.79202511000005]], ['c', [343.27098847, 223.83602572999996], [349.53098679000004, 231.70402576000004], [357.38298607, 237.84102585000005]], ['c', [365.23998642000004, 243.97302581999998], [374.69198704, 248.37002587999996], [382.22698665, 248.37402588999998]], ['c', [383.09998656, 248.37402588999998], [383.94798655, 248.31502588], [384.76398657000004, 248.19202587999996]], ['l', [384.70498657, 247.80102588]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Khdrbs3', 'point': [314.19896563104, 256.75110748336], 'bbox': [306.10708618, 254.57820752336, 322.29084508208, 258.92400744336], 'path': [['m', [384.57901, 247.87402393000002]], ['c', [377.0040102, 257.54302359999997], [357.98900985, 263.23302413], [340.85300827, 263.22602416000007]], ['c', [330.21900939, 263.22802403000003], [320.31101226, 261.03802348], [314.32000732, 256.32902385]], ['l', [314.07601165, 256.64002373000005]], ['c', [320.19001006999997, 261.43702365], [330.1660118, 263.61802436000005], [340.85300827, 263.62102367]], ['c', [358.07401083999997, 263.61302424999997], [377.15101003, 257.95502425000006], [384.89000998, 248.11802393000005]], ['l', [384.57901, 247.87402393000002]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Mfng', 'point': [337.58536149108, 279.76703766336], 'bbox': [332.75500488, 277.59413770336, 342.41571810216, 281.93993762336], 'path': [['m', [384.5409851, 247.95898487]], ['c', [383.1479851, 255.21098471999994], [376.32698536, 263.25498439], [367.8289852, 269.45598461]], ['c', [359.33598518, 275.66198398], [349.16998672, 280.03198672999997], [341.15998459, 280.02798511]], ['c', [339.92298507, 280.02798511], [338.73898697, 279.92398502], [337.61998367, 279.70898487]], ['l', [337.54498672, 280.09698535]], ['c', [338.69198608, 280.31798602], [339.90098572, 280.42298557], [341.15998459, 280.42298557]], ['c', [349.30898666, 280.4199834], [359.51598549, 276.01398517], [368.061985, 269.77498485]], ['c', [376.6039853, 263.53198482000005], [383.4899851, 255.46098471999994], [384.92898511, 248.03398487000004]], ['l', [384.5409851, 247.95898487]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}], 'right': [{'visual_slot': 'Cnih3', 'point': [431.303848375488, 281.52503326335994], 'bbox': [425.62649536, 279.35213330335995, 436.98120139097597, 283.69793322335994], 'path': [['m', [384.7669983, 248.1909794999999]], ['c', [385.73099828, 248.02997949999997], [386.74099827000003, 247.95297949999986], [387.7879982, 247.95297949999986]], ['c', [395.82999802, 247.94897949999995], [406.06399918, 252.55297947999986], [414.5739975, 259.03397990999986]], ['c', [423.0889969, 265.50997973999995], [429.87899781, 273.8679795399999], [431.10899735, 281.2719807799999]], ['l', [431.49899674, 281.2069783399999]], ['c', [430.22599793, 273.6239800599999], [423.3729973, 265.23897982999995], [414.81399918, 258.7199797799999]], ['c', [406.2499981, 252.20497942999987], [395.97699833, 247.56097949999992], [387.7879982, 247.5569794999999]], ['c', [386.7209983, 247.5569794999999], [385.68999827, 247.63697947999992], [384.7019983, 247.8019794999999]], ['l', [384.7669983, 248.1909794999999]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Lypd1', 'point': [452.10173249480397, 232.17214500255994], 'bbox': [446.47039794, 230.10514509255995, 457.733067049608, 234.23914491255994], 'path': [['m', [384.90200805, 248.10101367999994]], ['c', [388.55800795, 242.22601367999994], [396.04500769000003, 237.22001411999997], [404.92800711, 233.6990132599999]], ['c', [413.80900763, 230.17501307999999], [424.08000945000003, 228.13001300999997], [433.29100798, 228.13001300999997]], ['c', [440.75800705, 228.12901355999998], [447.52600859, 229.47601367999994], [452.28601073, 232.44301367999992]], ['l', [452.49501036000004, 232.10801364999998]], ['c', [447.6360092, 229.08301403999997], [440.80500792, 227.73601390999988], [433.29100798, 227.7350144599999]], ['c', [424.02200698, 227.73601390999988], [413.71100806, 229.79001285999993], [404.78200721, 233.33101322999994]], ['c', [395.85300826, 236.87501384999996], [388.30200815, 241.90101387999994], [384.56700805, 247.89201368]], ['l', [384.90200805, 248.10101367999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'H3f3a', 'point': [415.66477391326396, 235.01331451336], 'bbox': [409.86660766, 232.84041455336, 421.46294016652797, 237.18621447336], 'path': [['m', [384.91799926, 248.07000782]], ['c', [386.35499928999997, 244.46000791999995], [390.40399931999997, 240.93900778], [395.28099917, 238.34400799000002]], ['c', [400.15499972, 235.74600747], [405.85299871999996, 234.06600810999998], [410.55399893, 234.06700757]], ['c', [412.43099974, 234.06700757], [414.1479988, 234.33500817000004], [415.58799933, 234.91100742000003]], ['l', [415.73499869, 234.54400780000003]], ['c', [414.23299979, 233.94400742000005], [412.46799849, 233.67300751000005], [410.55399893, 233.67200806000005]], ['c', [405.76099967, 233.67400792], [400.02099893999997, 235.37300828000002], [395.0949993, 237.99600747]], ['c', [390.17099903999997, 240.62200786999995], [386.05699931, 244.16500785000005], [384.55099924999996, 247.92300781999995]], ['l', [384.91799926, 248.07000782]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Vav3', 'point': [440.62342571883994, 260.32312896336], 'bbox': [435.67138671, 258.15022900336, 445.57546472767996, 262.49602892336], 'path': [['m', [384.8420105, 248.16302539999992]], ['c', [388.88401031, 245.55302549999988], [394.7550106, 244.36402537999993], [401.22200966, 244.36602546999995]], ['c', [408.72200966, 244.36502529999996], [417.02201080000003, 245.9600253399999], [424.2000122, 248.7360254099999]], ['c', [431.37900924, 251.5110254399999], [437.43201065, 255.47402526999997], [440.45401001, 260.16602564999994]], ['l', [440.78601074, 259.9510254999999]], ['c', [437.68500900000004, 255.1470255999999], [431.56700897, 251.16402531999995], [424.3420105, 248.36802539999996]], ['c', [417.11600876, 245.57302548999996], [408.77801132, 243.97102548999987], [401.22200966, 243.97102548999987]], ['c', [394.70801067, 243.9720254199999], [388.77201056, 245.15902545999995], [384.6280105, 247.83002539999995]], ['l', [384.8420105, 248.16302539999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Bhlhe22', 'point': [434.266184673216, 227.52430084336004], 'bbox': [426.29251098, 225.35140088336004, 442.239858366432, 229.69720080336003], 'path': [['m', [384.91699218, 248.0709843799999]], ['c', [387.26799225, 242.3239841699999], [393.7309923, 236.78998424999986], [401.49799155000005, 232.71598483999992]], ['c', [409.26199149, 228.63898517999996], [418.32599258000005, 226.01298381999993], [425.84599303000005, 226.01398517999996]], ['c', [428.96099089, 226.01398517999996], [431.80899048000003, 226.4649834799999], [434.18899153, 227.43998386999988]], ['l', [434.33899306, 227.07398463999994]], ['c', [431.89599228000003, 226.07398463999994], [428.99799346000003, 225.61998416999995], [425.84599303000005, 225.61898471999996]], ['c', [418.23599242, 225.62098361999995], [409.12999152000003, 228.26398517999996], [401.31399153, 232.36598445999994]], ['c', [393.49999236, 236.47098399999993], [386.97199224, 242.02698423999993], [384.55199218, 247.9219843799999]], ['l', [384.91699218, 248.0709843799999]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Runx1t1', 'point': [409.813794761016, 258.22223785336007], 'bbox': [401.72061157, 256.0493378933601, 417.90697795203204, 260.39513781336007], 'path': [['m', [384.81298828, 248.17797899999994]], ['c', [386.05998825, 247.63897902999997], [387.58398818, 247.38897902999997], [389.25998830000003, 247.38997901999994]], ['c', [393.00698852, 247.38797897999996], [397.50498866000004, 248.63897899999995], [401.36098861, 250.60197901999993]], ['c', [405.21898841, 252.56197881999992], [408.4289875, 255.24197911999988], [409.62998771, 258.03297853999993]], ['l', [409.99198913000004, 257.87597895]], ['c', [408.72098922000004, 254.94097900999998], [405.44498824000004, 252.2399787999999], [401.53998756, 250.24997901999996]], ['c', [397.63398838, 248.26297899999997], [393.09298801, 246.9959789599999], [389.25998830000003, 246.99397898999996]], ['c', [387.54498839, 246.99497903999998], [385.96998822, 247.24897897999995], [384.65598828000003, 247.81497899999988]], ['l', [384.81298828, 248.17797899999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Sema3a', 'point': [409.52446611104006, 275.58503082335994], 'bbox': [401.43258666, 273.41213086335995, 417.61634556208, 277.75793078335994], 'path': [['m', [384.7250061, 248.19397021999998]], ['c', [389.75300596999995, 248.42897022], [395.93800639999995, 252.48497009999994], [400.83200645, 257.79397059999997]], ['c', [405.73300551999995, 263.09697009], [409.34500692999995, 269.65097095], [409.33800696, 274.77796985]], ['c', [409.33800696, 274.94997072999996], [409.33400535, 275.12097026], [409.32600592999995, 275.28896953]], ['l', [409.72100637999995, 275.30797054]], ['c', [409.72900581, 275.13296938999997], [409.73400687, 274.95596935], [409.73400687, 274.77796985]], ['c', [409.7270069, 269.47097064], [406.06100654, 262.88196994], [401.12300681, 257.52596999]], ['c', [396.17800616, 252.17597032999993], [389.96200608, 248.05797022000002], [384.7440061, 247.79897021]], ['l', [384.7250061, 248.19397021999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Hgf', 'point': [435.41528772774404, 215.20630767336002], 'bbox': [432.03338623, 213.03340771336002, 438.79718922548807, 217.37920763336], 'path': [['m', [384.92800903, 248.03802539999992]], ['c', [386.55800903, 240.47002556999996], [393.83500861, 232.21202518999996], [402.8480091, 225.87502529999995]], ['c', [411.85600853, 219.5330262399999], [422.59600829, 215.10602619999997], [431.08900833, 215.10902454999996]], ['c', [432.59400939, 215.10902454999996], [434.02901076, 215.24702503999993], [435.37100983, 215.53802539999992]], ['l', [435.45401, 215.15202381999995]], ['c', [434.08100890000003, 214.85502674999987], [432.61800765, 214.71402408999995], [431.08900833, 214.71402408999995]], ['c', [422.46300887, 214.7170262599999], [411.68200874, 219.17802478999988], [402.62000846, 225.5520253399999]], ['c', [393.56300925, 231.9300255999999], [386.22200905, 240.21402551999995], [384.54200903, 247.95502539999995]], ['l', [384.92800903, 248.03802539999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Gpr12', 'point': [423.68097555884003, 212.90430572336004], 'bbox': [417.64248657, 210.73140576336004, 429.71946454768, 215.07720568336003], 'path': [['m', [384.93200683, 248.00701954]], ['c', [385.30900683, 240.84601977], [391.10500668, 232.11301948999994], [398.66500663, 225.20201924000003]], ['c', [406.22000692999995, 218.28601887000002], [415.53700637, 213.19101764000004], [422.85400770999996, 213.19701816999998]], ['c', [423.12900543, 213.19701816999998], [423.40000533, 213.20401814000002], [423.66900633999995, 213.21902134000004]], ['l', [423.6910057, 212.82402089000004]], ['c', [423.41400526999996, 212.80901768], [423.13500594, 212.80201770999997], [422.85400770999996, 212.80201770999997]], ['c', [415.35900687, 212.80801823000002], [406.00600622999997, 217.95601894000004], [398.39900682999996, 224.91101887000002]], ['c', [390.7970066, 231.87101986999994], [384.93900683, 240.63601973000004], [384.53700682, 247.98601954000003]], ['l', [384.93200683, 248.00701954]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Prr15', 'point': [444.7351342022, 249.10224272336006], 'bbox': [439.4223938, 246.92934276336007, 450.0478746044, 251.27514268336006], 'path': [['m', [384.87200927, 248.13800098000002]], ['c', [390.71300935, 242.44900086000007], [402.22200964999996, 239.60300114000006], [413.85200881, 239.60800125000003]], ['c', [426.15200804, 239.60300114000006], [438.58501051999997, 242.78600074999997], [444.59201049, 248.97500099]], ['l', [444.87500761999996, 248.70000095]], ['c', [438.71900939, 242.38000108000006], [426.21900939, 239.21700146], [413.85200881, 239.21300078000002]], ['c', [402.15900991999996, 239.21700146], [390.5820093, 242.04400111999996], [384.59700927, 247.85500098]], ['l', [384.87200927, 248.13800098000002]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Galntl6', 'point': [454.981455833256, 256.27924469336006], 'bbox': [448.09640502, 254.10634473336006, 461.866506646512, 258.45214465336005], 'path': [['m', [384.8580017, 248.15100146999998]], ['c', [390.88200186, 243.35300162], [400.94300078, 241.11600163000003], [411.7140026, 241.11900141]], ['c', [428.63400267, 241.11100150999994], [447.29700087, 246.63800145999994], [454.82199858, 256.11600162]], ['l', [455.13100432, 255.87000133000004]], ['c', [447.44300078, 246.22600151999995], [428.71700285, 240.73100139999997], [411.7140026, 240.7230015]], ['c', [400.88900183, 240.72600126999998], [390.76000165, 242.95500137], [384.6120017, 247.84200149000003]], ['l', [384.8580017, 248.15100146999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Acsl1', 'point': [450.3621483522, 267.26709869336], 'bbox': [445.04940795, 265.09419873336003, 455.6748887544, 269.43999865336], 'path': [['m', [384.82998657, 248.16900683999995]], ['c', [388.97398662, 245.88600683999994], [394.50498677, 244.84800673999996], [400.62298679, 244.84900690999996]], ['c', [409.92998695, 244.84800673999996], [420.58998489000004, 247.24500681999996], [429.7869873, 251.23000693999995]], ['c', [438.98598479000003, 255.2120070599999], [446.71798705000003, 260.78900672], [450.18898772, 267.0950074399999]], ['l', [450.53498840000003, 266.9050069]], ['c', [446.98698807, 260.47200728999996], [439.18998718, 254.87500668999996], [429.94398498000004, 250.86700679]], ['c', [420.69698714000003, 246.8620067999999], [409.99698638, 244.45400693], [400.62298679, 244.45400693]], ['c', [394.46098613000004, 244.45400693], [388.86998653, 245.49400688999992], [384.63898657000004, 247.82300685999996]], ['l', [384.82998657, 248.16900683999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Rnf122', 'point': [405.27320903102, 240.44025542336], 'bbox': [398.26669311, 238.26735546336, 412.27972495204, 242.61315538335998], 'path': [['m', [384.91500854, 248.07702685999993]], ['c', [385.90400855999997, 245.84802675999993], [388.50200844, 243.7160267999999], [391.62200830999996, 242.16302682999992]], ['c', [394.74100875, 240.60702705999995], [398.37800883999995, 239.62302638999995], [401.41900825, 239.6240267999999]], ['c', [402.84800909999996, 239.6240267999999], [404.14400864, 239.8410267999999], [405.19000815, 240.3080268099999]], ['l', [405.35100936, 239.94702674999996]], ['c', [404.23500823999996, 239.45002700999999], [402.88800811, 239.22902728999998], [401.41900825, 239.22902728999998]], ['c', [398.29500865, 239.23002673999997], [394.61500834, 240.23002673999997], [391.44600866999997, 241.81002665999995]], ['c', [388.27900861999996, 243.39202691999992], [385.61600851, 245.54502679999996], [384.55400854, 247.91602686]], ['l', [384.91500854, 248.07702685999993]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Gm2516', 'point': [427.71574959658005, 258.04615142336], 'bbox': [419.3815918, 255.87325146336002, 436.04990739316, 260.21905138336], 'path': [['m', [384.83999633, 248.16400195999995]], ['c', [387.87399624, 246.25100197999996], [392.22999619, 245.37700201999996], [397.03699587, 245.37800195]], ['c', [402.85299681000004, 245.37800195], [409.32699584, 246.65200199999992], [414.92299651, 248.85500193999997]], ['c', [420.51999663000004, 251.05600190999996], [425.23399733, 254.19100190999995], [427.54499816000003, 257.8650021699999]], ['l', [427.87899779, 257.65500212999996]], ['c', [425.48899459, 253.8650021699999], [420.71099471, 250.71000196], [415.06799697, 248.4880019599999]], ['c', [409.42399596, 246.26600195999993], [402.90999603, 244.9830019599999], [397.03699587, 244.9830019599999]], ['c', [392.18299626000004, 244.98400189999995], [387.76399635, 245.85700204], [384.62899633, 247.8290019599999]], ['l', [384.83999633, 248.16400195999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Syn3', 'point': [419.74662518883997, 281.57102325335995], 'bbox': [414.79458618, 279.39812329335996, 424.69866419768, 283.74392321335995], 'path': [['m', [384.7380066, 248.19397021999998]], ['l', [384.9970066, 248.19197022000003]], ['c', [391.77700663, 248.18497021999997], [400.38200665, 253.17097045000003], [407.29200745, 259.87496996999994]], ['c', [414.20900727000003, 266.57197047], [419.42800522, 274.98597003], [419.5470047, 281.71897174]], ['l', [419.94200516, 281.71197177]], ['c', [419.80500794, 274.78596927], [414.5240059, 266.33997012], [407.56700707, 259.59097052999994]], ['c', [400.60500622, 252.84797001999993], [391.96500683, 247.80297022000002], [384.9970066, 247.79597021999996]], ['l', [384.7310066, 247.79897021]], ['l', [384.7380066, 248.19397021999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Myt1l', 'point': [441.36555751443996, 237.00721100336], 'bbox': [436.17449951, 234.83431104336, 446.55661551887994, 239.18011096336], 'path': [['m', [384.89801025, 248.10699511999997]], ['c', [391.0340104, 239.04499484999997], [408.66101073000004, 233.09199477999994], [423.53001021, 233.10199501]], ['c', [430.56201171000004, 233.09999515999993], [436.97301101000005, 234.42899466999995], [441.25201035000003, 237.33699466999997]], ['l', [441.47401046000004, 237.009995]], ['c', [437.08800887, 234.0339951699999], [430.6110115, 232.70799495999995], [423.53001021, 232.70699550999996]], ['c', [408.5600109, 232.71599531999993], [390.87401008, 238.63099529], [384.57101025000003, 247.88499511999999]], ['l', [384.89801025, 248.10699511999997]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Sptb', 'point': [405.52768864773196, 251.24720122335998], 'bbox': [401.0586853, 249.07430126335998, 409.99669199546395, 253.42010118335998], 'path': [['m', [384.8500061, 248.15698292000002]], ['c', [386.46300613, 246.98998287000006], [389.01300619999995, 246.44098296000004], [391.79800604999997, 246.44298293999998]], ['c', [397.06900595999997, 246.43398288000003], [403.16800689999997, 248.40998292000006], [405.36400603, 251.47998287000007]], ['l', [405.68500517999996, 251.24898292]], ['c', [403.32200622, 247.99698292000005], [397.16300581999997, 246.05598284999996], [391.79800604999997, 246.04798296]], ['c', [388.96300600999996, 246.04898288000004], [386.34600615, 246.59398296000006], [384.6190061, 247.83598292]], ['l', [384.8500061, 248.15698292000002]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Chrm3', 'point': [432.674079539984, 247.93421295335997], 'bbox': [426.1545105, 245.76131299335998, 439.193648579968, 250.10711291335997], 'path': [['m', [384.875, 248.13501025999994]], ['c', [389.62799978, 243.3010101499999], [399.375, 240.82301045999998], [409.06900024, 240.8270101799999]], ['c', [418.48500061, 240.82301045999998], [427.84799957, 243.16101028999992], [432.53400039999997, 247.78801025999996]], ['l', [432.81200027, 247.50701024]], ['c', [427.97999954, 242.75501014999998], [418.54899978000003, 240.4370103099999], [409.06900024, 240.43201019999992]], ['c', [399.30900001, 240.4370103099999], [389.49599981, 242.89401008999994], [384.59400002, 247.8580102499999]], ['l', [384.875, 248.13501025999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Pou6f2', 'point': [418.55852739324, 246.85724762335997], 'bbox': [411.67260742, 244.68434766335997, 425.44444736648, 249.03014758335996], 'path': [['m', [384.88000488, 248.13000537999994]], ['c', [388.31400489, 244.38000537999994], [395.85000514, 242.3670053699999], [403.13300513, 242.37200547999998]], ['c', [409.28800582, 242.36900522999997], [415.26000404, 243.80500557999994], [418.41900634, 246.71700539999995]], ['l', [418.68700408, 246.42600538]], ['c', [415.3880043, 243.40200521999998], [409.34800529, 241.98100521999993], [403.13300513, 241.9770055099999]], ['c', [395.77900504, 241.98300554999992], [388.17700481, 243.97200537999993], [384.58900488, 247.8630053999999]], ['l', [384.88000488, 248.13000537999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Fam65b', 'point': [427.05886894326, 238.62024810336004], 'bbox': [419.08758544, 236.44734814336005, 435.03015244652, 240.79314806336004], 'path': [['m', [384.90200805, 248.10199023999996]], ['c', [387.17600797, 244.47999026999992], [391.82100819, 241.38799001999996], [397.33700846, 239.2169900199999]], ['c', [402.85200881000003, 237.04399062999994], [409.23300742000004, 235.78599025999995], [414.96200751000003, 235.78599025999995]], ['c', [419.68700789, 235.78498986], [423.96700667, 236.64299060999997], [426.95100783, 238.52199032]], ['l', [427.16100692, 238.18799069]], ['c', [424.07800673, 236.24999000999992], [419.73400878, 235.39199021000002], [414.96200751000003, 235.39098979999994]], ['c', [409.17500876, 235.39199021000002], [402.75600815, 236.65798999999993], [397.19200801, 238.84898998999995]], ['c', [391.63100815, 241.04299023], [386.9210081, 244.1539902899999], [384.56700805, 247.89099023999995]], ['l', [384.90200805, 248.10199023999996]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'D130009I18Rik', 'point': [402.146551924224, 230.45234039336003], 'bbox': [387.16571044, 228.27944043336004, 417.12739340844803, 232.62524035336003], 'path': [['m', [384.93200683, 247.99401905000002]], ['l', [384.93200683, 247.91501905000007]], ['c', [384.92500683, 244.48801901000002], [387.49200677, 240.09401895000008], [390.94700668999997, 236.57901908999997]], ['c', [394.39500712999995, 233.05701878000002], [398.73000715999996, 230.41701939000006], [402.14200590999997, 230.3850198]], ['l', [402.13800619999995, 229.99001934]], ['c', [398.53300665999996, 230.03901913000004], [394.16500663, 232.7410188], [390.66500663, 236.30201867000005]], ['c', [387.17100690999996, 239.87001944999997], [384.54400682, 244.29801895000003], [384.53700682, 247.91501905000007]], ['l', [384.53700682, 247.99901905000002]], ['l', [384.93200683, 247.99401905000002]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Pou4f1', 'point': [442.052851574745, 276.09286034256], 'bbox': [435.50149536, 274.02586043255997, 448.60420778949, 278.15986025256], 'path': [['m', [384.7999878, 248.18298388999995]], ['c', [387.19898772, 247.34298391999994], [389.95698785999997, 246.95198395], [392.93998813999997, 246.95198395]], ['c', [401.94998741, 246.94998384999997], [413.00698853, 250.5229837999999], [422.43498612, 255.92498399]], ['c', [431.8659897, 261.32298423], [439.66398621, 268.55798388999995], [442.1979866, 275.80698442999994]], ['l', [442.56998825, 275.67598391999996]], ['c', [439.97198868, 268.27198459], [432.1089859, 261.01398421], [422.63198853, 255.58198404999996]], ['c', [413.15098763, 250.15298391], [402.0519886, 246.55798388999995], [392.93998813999997, 246.55698384999994]], ['c', [389.92198801, 246.55698384999994], [387.12198782, 246.95098388999997], [384.6689878, 247.8099838899999]], ['l', [384.7999878, 248.18298388999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Tnfrsf19', 'point': [420.20380911328, 269.61512115335995], 'bbox': [412.23361206, 267.44222119335996, 428.17400616656, 271.78802111335995], 'path': [['m', [384.78201293, 248.18798876999995]], ['c', [385.84701298, 247.92298878999998], [387.00501297, 247.79698876999998], [388.23101281, 247.79698876999998]], ['c', [394.04501246, 247.79398876999994], [401.37701224, 250.6419887699999], [407.55301283, 254.77498866999997]], ['c', [413.73301314, 258.90298892], [418.75401305, 264.32298899], [420.00601195, 269.37698793999994]], ['l', [420.38901137, 269.2809891799999]], ['c', [419.08401106, 264.0589890499999], [413.99801253, 258.61398840999993], [407.77301215, 254.44698858999993]], ['c', [401.54401206, 250.28298879], [394.17101286, 247.40498876999993], [388.23101281, 247.40198876]], ['c', [386.97801303, 247.40198876], [385.78901289, 247.5309887699999], [384.68701293, 247.80498878999992]], ['l', [384.78201293, 248.18798876999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Dcbld2', 'point': [405.62584035548, 264.29914215336], 'bbox': [398.86160278, 262.12624219336, 412.39007793096, 266.47204211336], 'path': [['m', [384.76000976, 248.19201708999992]], ['c', [385.09900974, 248.14801708999994], [385.45100974, 248.1260170899999], [385.8120098, 248.1260170899999]], ['c', [389.46600961, 248.1210170899999], [394.16700934, 250.36601710999992], [398.0450096, 253.4950170599999]], ['c', [401.92700957, 256.61901713], [404.98200988, 260.6300168099999], [405.42700957, 264.05701686]], ['l', [405.81801032, 264.0050168099999]], ['c', [405.33700942, 260.3960175599999], [402.21800994, 256.3560166399999], [398.29300976, 253.18701696999995]], ['c', [394.36200999, 250.02401709999992], [389.62200974, 247.73501710999994], [385.8120098, 247.7310170899999]], ['c', [385.43500977, 247.7310170899999], [385.06600976, 247.75401708999993], [384.70900976, 247.80101708999996]], ['l', [384.76000976, 248.19201708999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Runx1', 'point': [436.936950439864, 269.23595725256], 'bbox': [430.9624939, 267.16895734256, 442.911406979728, 271.30295716256], 'path': [['m', [384.81298828, 248.17797899999994]], ['c', [387.45498823, 247.0409790399999], [390.66298818, 246.51697898999998], [394.18198871000004, 246.5179790399999]], ['c', [402.07898902, 246.51697898999998], [411.53098869, 249.15297901999998], [419.63998794, 253.28597879999995]], ['c', [427.750988, 257.4169793199999], [434.51298904000004, 263.05297899999994], [437.06198882, 268.9849781999999]], ['l', [437.42498779, 268.82897805999994]], ['c', [434.80498885000003, 262.75297879999994], [427.97698974, 257.09497880999993], [419.81898879, 252.93397903999994]], ['c', [411.65898895000004, 248.77697896999996], [402.16498756, 246.12397909999993], [394.18198871000004, 246.12297893999994]], ['c', [390.62398815, 246.12297893999994], [387.36498832, 246.64997899999992], [384.65698828, 247.81497899999988]], ['l', [384.81298828, 248.17797899999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Prkce', 'point': [455.57032022776, 243.81525542336], 'bbox': [450.01638793, 241.64235546336, 461.12425252552, 245.98815538335998], 'path': [['m', [384.8829956, 248.12701464999998]], ['c', [392.26999568, 239.73701430999995], [408.93899535, 235.17601444000002], [424.76999663000004, 235.18201494999994]], ['c', [437.09799575, 235.17901469999993], [448.91399382000003, 237.94301464], [455.43799590000003, 243.70301486999995]], ['l', [455.6989975, 243.40701485]], ['c', [449.03999328000003, 237.54101420999996], [437.1559944, 234.79001474999995], [424.76999663000004, 234.78701449999994]], ['c', [408.86499595000004, 234.79301501999998], [392.12999581, 239.32701444999998], [384.5869956, 247.86601465]], ['l', [384.8829956, 248.12701464999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Tshz1', 'point': [422.11932123108, 225.07129791336], 'bbox': [416.20251464, 222.89839795336002, 428.03612782216, 227.24419787336], 'path': [['m', [384.92700195, 248.04101612]], ['c', [386.1940019, 242.60901595999997], [391.51800203, 236.73001624999995], [398.09100149999995, 232.23801566999998]], ['c', [404.66000175, 227.74201633999996], [412.47300147, 224.62601521], [418.65600205, 224.62901547]], ['c', [419.86500166999997, 224.62901547], [421.01100157999997, 224.74701549999997], [422.07300184999997, 224.99601604999998]], ['l', [422.16300200999996, 224.61101582000003]], ['c', [421.06700133, 224.35401585], [419.89100264999996, 224.23301556], [418.65600205, 224.23301556]], ['c', [412.34500120999996, 224.23601582000003], [404.49000166999997, 227.38501598999994], [397.86800193, 231.91201640999998]], ['c', [391.2500019, 236.44401600000003], [385.86100195999995, 242.34801629000003], [384.54200195, 247.95101611999996]], ['l', [384.92700195, 248.04101612]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Sgms1', 'point': [412.09867355552, 217.09345367336005], 'bbox': [405.33660888, 214.92055371336005, 418.86073823104005, 219.26635363336004], 'path': [['m', [384.93200683, 247.98498583999992]], ['c', [384.91800682999997, 247.75398583999993], [384.91200683, 247.5199858399999], [384.91200683, 247.28398586999992]], ['c', [384.90600682999997, 241.5469856399999], [388.87700676, 234.21798563999994], [394.2800064, 228.27098513999988]], ['c', [399.67700671, 222.31798602999993], [406.50600622999997, 217.7479863299999], [412.10800743, 217.42798663999997]], ['l', [412.08400724999996, 217.03298617999997]], ['c', [406.29400633999995, 217.3789849399999], [399.43300723, 222.01098490999993], [393.98700713, 228.00498629999993]], ['c', [388.54700683, 234.00498629999993], [384.52200682999995, 241.36998606999987], [384.51700682999996, 247.28398586999992]], ['c', [384.51700682999996, 247.52798585999994], [384.52300683, 247.7689858399999], [384.53700682, 248.00798583999995]], ['l', [384.93200683, 247.98498583999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Crtac1', 'point': [445.5786032222, 222.91431549336005], 'bbox': [439.17941284, 220.74141553336005, 451.9777936044, 225.08721545336005], 'path': [['m', [384.91699218, 248.07202197999993]], ['c', [387.81799220000005, 241.01102210999989], [395.75599192000004, 234.23402167999996], [405.29099272, 229.24402286999998]], ['c', [414.82299231, 224.2520222799999], [425.94699095000004, 221.03902103999997], [435.18699263, 221.04102184999988]], ['c', [439.03699111000003, 221.04002239999988], [442.55899048000003, 221.59802104999994], [445.50099181, 222.80902149999997]], ['l', [445.65199279, 222.44302226999991]], ['c', [442.64599228000003, 221.20702220999988], [439.07499312000004, 220.64602137999998], [435.18699263, 220.64502193999988]], ['c', [425.8569908, 220.6470227499999], [414.69099234000004, 223.8770222799999], [405.10699271000004, 228.89402248]], ['c', [395.52599238000005, 233.91402197999992], [387.52299212, 240.71402217999992], [384.55199218, 247.92102197999998]], ['l', [384.91699218, 248.07202197999993]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}]}, 'current_label_offsets': {'Foxo6': {'Cplx2': [0, 1.15], 'Mfap4': [0, -1.15], 'Pcdh10': [0, 1.35], 'Nhlh1': [0, -1.35]}, 'Zfpm2': {'Shox2': [1.5, 0], 'Nhsl1': [-1.5, 0], 'Ank2': [1.35, 0], 'Soga3': [-1.35, 0], 'Elavl3': [0, 0.8], 'Epha5': [0, -0.8], 'Rab3a': [0, 1.15], 'Map1b': [0, -1.15], 'Mapk8ip1': [0, 1.55], 'Tubb3': [0, -1.55], '3110039M20Rik': [0, 1.9], 'Afap1': [0, -1.9]}, 'Neurod2': {'Neurod6': [0, 1.35], 'Myo6': [0, -1.35], 'Kcnmb2': [1.55, 0], 'Mreg': [-1.55, 0]}, 'Nhlh1': {'1700037H04Rik': [0, 1.55], 'Ypel3': [0, -1.55], 'Nhlh2': [1.5, 0], 'Trp53i11': [-1.5, 0]}, 'Neurod1': {'Jun': [0, 0.241], 'Ntm': [-3.244, 0], 'Vcan': [0.973, 0], 'Edil3': [-0.762, 0.728], 'Cd81': [1.065, 3.971], 'Lrrn3': [-5.092, 0], 'Uchl1': [-1.284, 0], 'Acsl1': [-3.441, -1.537], 'Olfm1': [1.6, 0], 'Aplp1': [-1.6, 0], 'Chst2': [0.762, 0], 'Elavl2': [0, -3.499], 'Pcsk2': [4.538, 0.475], 'Ckmt1': [0, 1.106], 'Scn3a': [-0.973, 0.431], 'Unc5c': [1.284, -0.479], 'Snx18': [0, -0.475], 'Ddah1': [0, 0.804], 'Gnao1': [0, 0.516], 'Ywhah': [0, -0.037], 'Tmeff2': [-1.065, 0], 'Rbms1': [11.147, 0], 'Pou6f2': [0, -0.804], 'Ctnna2': [5.092, 0], 'Spock1': [0, -0.35], 'Hsp90aa1': [0, -1.09], 'C1ql3': [0, 0.874], 'Olfm2': [-1.814, 0], 'Rbms3': [0, -0.874], 'Cacng5': [1.814, 0], 'Sncg': [0, 2.5], 'Mllt11': [0, -2.5]}}}

def draw_module_network(ax, factor, left_genes, right_genes, kind='associated',
                        left_color='#2D7184', right_color='#5AAD7F', font_size=11.4,
                        highlight_genes=tuple()):
    """Substitute current module genes into the published Fig.3d vector layout.

    Original slot names are visual provenance only. Every current input member
    retains exactly one ribbon; template geometry carries no biological weight.
    Side-color parameters apply to ribbons only; outer panels are untouched.
    """
    #1. =========== Original vector template and appearance
    padding_fraction = .025
    halo_width_pt = .85
    highlight_color = '#BA001C'
    template = PAPER_NETWORK_TEMPLATE
    groups = [list(map(str, left_genes)), list(map(str, right_genes))]
    highlights = set(map(str, highlight_genes))
    if any(len(set(g)) != len(g) for g in groups):
        raise ValueError('A displayed module contains duplicate genes.')
    if font_size <= 0:
        raise ValueError('font_size must be positive.')
    ax.set_axis_off(); ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.set_facecolor('white')
    fig = ax.figure
    fig.canvas.draw()
    renderer = fig.canvas.get_renderer()
    area = ax.get_window_extent(renderer)
    point_scale = 72.0 / fig.dpi
    width, height = area.width * point_scale, area.height * point_scale
    bbox = np.asarray(template['bbox'], dtype=float)
    scale = min(width * (1 - 2 * padding_fraction) / (bbox[2] - bbox[0]),
                height * (1 - 2 * padding_fraction) / (bbox[3] - bbox[1]))
    used = (bbox[2:] - bbox[:2]) * scale
    offset = (np.array([width, height]) - used) / 2
    def xy(points):
        a = np.asarray(points, dtype=float)
        result = (a - bbox[:2]) * scale + offset
        result[..., 1] = height - result[..., 1]
        return result / np.array([width, height])
    def angle(slot, side):
        delta = np.asarray(slot['point']) - np.asarray(template['center'])
        return np.arctan2(delta[1], -delta[0] if side == 'left' else delta[0])
    def topology(slot):
        return tuple(step[0] for step in slot['path'])
    def blend(a, b):
        path = [[sa[0]] + [((np.asarray(pa) + np.asarray(pb)) / 2).tolist()
                for pa, pb in zip(sa[1:], sb[1:])]
                for sa, sb in zip(a['path'], b['path'])]
        return {'visual_slot': [a['visual_slot'], b['visual_slot']],
            'point': ((np.asarray(a['point']) + np.asarray(b['point'])) / 2).tolist(),
            'bbox': ((np.asarray(a['bbox']) + np.asarray(b['bbox'])) / 2).tolist(),
            'path': path, 'rgb': a['rgb'], 'interpolated': True}
    def choose_slots(side, count):
        slots = list(template['slots'][side])
        if count <= len(slots):
            if count == 0: return []
            points = np.asarray([slot['point'] for slot in slots])
            # Preserve the actual top, bottom and outer boundary before filling
            # the remaining slots from the original two-dimensional layout.
            outer = np.argmin(points[:, 0]) if side == 'left' else np.argmax(points[:, 0])
            chosen = list(dict.fromkeys([int(np.argmin(points[:, 1])),
                                         int(np.argmax(points[:, 1])), int(outer)]))[:count]
            while len(chosen) < count:
                distances = np.linalg.norm((points[:, None] - points[chosen][None])
                                           * np.array([.8, 1.0]), axis=2).min(axis=1)
                distances[chosen] = -1
                chosen.append(int(np.argmax(distances)))
            return [slots[i] for i in chosen]
        # Additional members use local interpolation of neighboring original vector
        # ribbons with identical command topology; no ellipse is generated.
        while len(slots) < count:
            slots.sort(key=lambda s: angle(s, side))
            points = np.asarray([slot['point'] for slot in slots])
            candidates = []
            boxes = np.asarray([slot['bbox'] for slot in slots])
            for i, a in enumerate(slots):
                for j in range(i + 1, len(slots)):
                    b = slots[j]
                    if topology(a) != topology(b) or abs(angle(a, side) - angle(b, side)) > .65:
                        continue
                    box = (np.asarray(a['bbox']) + np.asarray(b['bbox'])) / 2
                    dx = np.maximum(0, np.maximum(box[0] - boxes[:, 2], boxes[:, 0] - box[2]))
                    dy = np.maximum(0, np.maximum(box[1] - boxes[:, 3], boxes[:, 1] - box[3]))
                    ox = np.maximum(0, np.minimum(box[2], boxes[:, 2]) - np.maximum(box[0], boxes[:, 0]))
                    oy = np.maximum(0, np.minimum(box[3], boxes[:, 3]) - np.maximum(box[1], boxes[:, 1]))
                    # Insert into genuine local whitespace between neighboring
                    # source ribbons, rather than halfway through existing words.
                    score = np.hypot(dx, dy).min() - 100 * (ox * oy).sum()
                    candidates.append((float(score), i, j))
            if not candidates:
                raise ValueError('No compatible original ribbon pair for interpolation.')
            _, i, j = max(candidates)
            slots.append(blend(slots[i], slots[j]))
        return slots
    def mpl_path(commands):
        vertices, codes = [], []
        for step in commands:
            code = {'m': MplPath.MOVETO, 'l': MplPath.LINETO, 'c': MplPath.CURVE4}[step[0]]
            vertices.extend(step[1:]); codes.extend([code] * len(step[1:]))
        vertices.append(vertices[0]); codes.append(MplPath.CLOSEPOLY)
        return MplPath(xy(vertices), codes)
    stroke = [pe.withStroke(linewidth=halo_width_pt, foreground='white')]
    #2. =========== Original gray guide, ribbons and replacement gene labels
    for guide in template['guides']:
        points = xy(guide['pts'])
        line, = ax.plot(points[:, 0], points[:, 1], transform=ax.transAxes,
            color=guide['stroking_color'], linewidth=guide['linewidth'] * scale,
            solid_capstyle='butt', zorder=.2, clip_on=True)
        dash, phase = guide['dash']
        if dash:
            line.set_linestyle((phase / guide['linewidth'],
                                tuple(float(v) / guide['linewidth'] for v in dash)))
    factor_point = xy(template['center'])
    factor_text = ax.text(*factor_point, str(factor), transform=ax.transAxes,
        ha='center', va='center', fontfamily='Arial', fontstyle='italic',
        fontweight='normal', fontsize=font_size, color='black', path_effects=stroke,
        zorder=6, clip_on=False)
    records, patches, edge_records = [], [], []
    for side, genes in zip(['left', 'right'], groups):
        slots = choose_slots(side, len(genes))
        texts = [ax.text(.5, .5, gene, transform=ax.transAxes, ha='center', va='center',
            fontfamily='Arial', fontstyle='italic',
            fontsize=font_size * (template['red_font_size'] / template['base_font_size'] if gene in highlights else 1),
            color=highlight_color if gene in highlights else 'black', path_effects=stroke,
            zorder=5, clip_on=False) for gene in genes]
        fig.canvas.draw(); renderer = fig.canvas.get_renderer()
        # Match long replacement names to the genuinely wider original text
        # slots; this preserves the published geometry rather than repacking it.
        gene_widths = [TextPath((0, 0), gene, prop=FontProperties(family='Arial', style='italic', size=text.get_fontsize())).get_extents().width for gene, text in zip(genes, texts)]
        gene_order = sorted(range(len(genes)), key=lambda i: (gene_widths[i], genes[i]))
        slot_order = sorted(range(len(slots)), key=lambda i: (slots[i]['bbox'][2] - slots[i]['bbox'][0], i))
        for gi, si in zip(gene_order, slot_order):
            gene, text, slot = genes[gi], texts[gi], slots[si]
            shift = template.get('current_label_offsets', {}).get(str(factor), {}).get(gene, (0, 0))
            point = xy(slot['point']) + np.asarray(shift) / np.array([width, height])
            text.set_position(point)
            patch = PathPatch(mpl_path(slot['path']), transform=ax.transAxes,
                facecolor=left_color if side == 'left' else right_color, edgecolor='none', linewidth=0, alpha=1,
                zorder=1, clip_on=True)
            ax.add_patch(patch); patches.append(patch)
            records.append({'side': side, 'gene': gene, 'text': text, 'point': point})
            edge_records.append({'side': side, 'gene': gene,
                'visual_slot': slot['visual_slot'], 'template_interpolated': slot['interpolated'],
                'label_offset_points': list(shift)})
    #3. =========== Report actual layout and the fixed visual provenance
    fig.canvas.draw(); renderer = fig.canvas.get_renderer()
    named = [(str(factor), factor_text)] + [(r['side'] + ':' + r['gene'], r['text']) for r in records]
    boxes = [(name, text.get_window_extent(renderer)) for name, text in named]
    overlaps, clipped = [], []
    for i, (name, box) in enumerate(boxes):
        if box.x0 < area.x0 or box.y0 < area.y0 or box.x1 > area.x1 or box.y1 > area.y1:
            clipped.append(name)
        for other, b in boxes[i + 1:]:
            if min(box.x1, b.x1) > max(box.x0, b.x0) and min(box.y1, b.y1) > max(box.y0, b.y0):
                overlaps.append((name, other))
    positions = {'factor': tuple(map(float, factor_point)), 'left': {}, 'right': {}}
    for record in records:
        positions[record['side']][record['gene']] = tuple(map(float, record['point']))
    all_genes = set(groups[0]) | set(groups[1])
    reference = {'source': template['source'], 'path': template['path'],
        'sha256': template['sha256'], 'page_1based': template['page_1based'],
        'coordinate_system': 'Original PDF points, top-left origin; uniform affine transform',
        'original_slots': {'left': len(template['slots']['left']), 'right': len(template['slots']['right'])},
        'slot_mapping': edge_records,
        'slot_names_are_biological_targets': False,
        'runtime_pdf_or_Tmp_dependency': False,
        'local_text_offsets_points': template.get('current_label_offsets', {}).get(str(factor), {})}
    motif = str(kind).lower() in {'motif', 'motif_supported', 'early_motif_supported', 'regulon'}
    qa = {'ok': not (overlaps or clipped), 'factor': str(factor), 'kind': str(kind),
        'left_genes': len(groups[0]), 'right_genes': len(groups[1]),
        'gene_labels': len(records), 'edges': len(patches),
        'all_genes_connected': len(patches) == sum(map(len, groups)),
        'label_overlaps': overlaps, 'clipped_labels': clipped,
        'axes_width_inches': width / 72, 'axes_height_inches': height / 72,
        'font_size_points': float(font_size), 'factor_font_size_points': float(font_size),
        'highlighted_genes': sorted(highlights & all_genes),
        'requested_highlights_not_in_modules': sorted(highlights - all_genes),
        'highlight_semantics': 'Literature-discussed representatives; not statistical significance or stronger edges',
        'curve_geometry': 'Original filled Bezier ribbons; interpolation only for additional members',
        'paper_palette': {'left': '#2D7184', 'right': '#5AAD7F', 'red': highlight_color},
        'template_scale_points': float(scale),
        'reference_template': reference,
        'edge_semantics': 'Predicted motif-supported TF-target link' if motif else 'RNA module association'}
    return {'labeltexts': [text for _, text in named], 'edgepatches': patches,
        'positions': positions, 'factor_text': factor_text, 'edge_records': edge_records,
        'reference_template': reference, 'QA': qa}


# SGN-1 display only; append after the existing network helper cell.
# Red marks approved literature representatives, independent of RNA association sign.
NETWORK_HIGHLIGHTS = {
    "Foxo6": ("Nhlh1", "Sox11"),
    "Zfpm2": ("Isl1", "Nhlh1"),
    "Neurod2": ("Nhlh1", "Neurod6", "Isl1"),
    "Nhlh1": ("Nhlh2",),
    "Neurod1": ("Myt1l", "Nhlh2"),
}

# Reuse approved gene-level evidence; neither copy validates a new TF-target edge.
NETWORK_HIGHLIGHT_EVIDENCE["Zfpm2"]["Nhlh1"] = dict(
    NETWORK_HIGHLIGHT_EVIDENCE["Foxo6"]["Nhlh1"]
)
NETWORK_HIGHLIGHT_EVIDENCE["Neurod1"]["Nhlh2"] = dict(
    NETWORK_HIGHLIGHT_EVIDENCE["Nhlh1"]["Nhlh2"]
)
NETWORK_HIGHLIGHT_EVIDENCE = {
    factor: {gene: dict(NETWORK_HIGHLIGHT_EVIDENCE[factor][gene]) for gene in genes}
    for factor, genes in NETWORK_HIGHLIGHTS.items()
}

# SGN-1 local text-only offsets after changing displayed modules.
# Points: positive x is right; positive y is up. Paths, slots and fonts are unchanged.
PAPER_NETWORK_TEMPLATE["current_label_offsets"]['Nhlh1'].update({'Trp53i11': [-1.5, 3.1]})
PAPER_NETWORK_TEMPLATE["current_label_offsets"]['Neurod1'].update({'Cfl2': [0.001, 3.74], 'Btg1': [0.0, 0.338], 'Rtn1': [0.383, -3.064], 'Dner': [-5.286, 0.0], 'Sncg': [-0.556, 2.5], 'Mapt': [0.729, 0.0], 'Myt1l': [-0.972, 3.867], 'Chgb': [2.286, -4.683], 'Nav3': [-3.032, 0.0], 'Nsg2': [-0.568, -0.122], 'Trim2': [-3.122, -1.942], 'Fmn1': [1.562, 0.0], 'Scrn1': [-1.562, 0.0], 'Mllt11': [0.972, -2.5], 'C1ql3': [0.0, -1.359], 'Olfm2': [1.901, 0.677], 'Vdac1': [0.0, 1.106], 'Cxxc4': [-0.729, 0.836], 'Clec2l': [0.568, -1.087], 'Scn3b': [0.0, -0.677], 'Rims3': [0.0, 1.006], 'Eps15': [0.0, 0.718], 'Lancl1': [0.0, -0.239], 'Trim67': [0.379, 0.0], 'Vstm2l': [4.693, 0.0], 'Hpcal1': [0.0, -1.006], 'Ncam1': [3.032, 0.0], 'Cacng5': [1.814, 0.122], 'Akap12': [0.0, 0.608], 'Slc17a6': [-2.493, 1.454], 'S100a10': [0.0, 0.582], 'St18': [1.171, 0.0], 'Actc1': [0.113, 0.0], 'Insm1': [-0.113, 0.0], 'Fam110a': [-1.171, 0.0]})
```

```python
"""Strict vector PDF export for the Fig4 B/C/D Matplotlib canvas.

No analysis is performed. Each native Matplotlib label becomes one live Arial
text object, including the TF italic run in the two supported mathtext titles.
All geometry is written directly, without XObjects, glyph outlines or images.
The caller declares at most four data axes; their clip rectangles are collected
into one sibling viewport group per axis. Other artists must fit their axes.
"""

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")
from pathlib import Path
import re
import sys
import math
import numpy as np
from matplotlib.backend_bases import RendererBase
from matplotlib.path import Path as MplPath

try:
    from reportlab.pdfgen import canvas
except ImportError:
    # Set this storage path independently before running.
    sys.path.append(str(ROOT))
    from reportlab.pdfgen import canvas
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont

_FONTS = {'normal':'FigArial','italic':'FigArialItalic',
          'bold':'FigArialBold','bolditalic':'FigArialBoldItalic'}

def _register():
    for key, file in [('normal','arial.ttf'),('italic','ariali.ttf'),
                      ('bold','arialbd.ttf'),('bolditalic','arialbi.ttf')]:
        if _FONTS[key] not in pdfmetrics.getRegisteredFontNames():
            # Set this storage path independently before running.
            pdfmetrics.registerFont(TTFont(_FONTS[key],str(ROOT/file)))

def _font(prop):
    bold=prop.get_weight() in ('bold','heavy','black','semibold','demibold') or isinstance(prop.get_weight(),(int,float)) and prop.get_weight()>=600
    italic=prop.get_style() in ('italic','oblique')
    return _FONTS['bolditalic' if bold and italic else 'bold' if bold else 'italic' if italic else 'normal']

def _runs(text,prop,ismath):
    if not ismath:
        if '$' in text: raise ValueError('Unexpected dollar in plain label: '+text)
        return [(text,_font(prop))]
    # Current Fig4 titles use only this explicitly audited mathtext form.
    found=re.fullmatch(r'\$\\it\{([^{}]+)\}([^$]*)\$(.*)',text)
    if found is None: raise ValueError('Unsupported math label: '+text)
    tf,inside,after=found.groups()
    return [(tf,_FONTS['italic']),(inside+after,_FONTS['normal'])]

class _Recorder(RendererBase):
    def __init__(self,fig,agg,viewports,halo_labels):
        super().__init__()
        self.dpi=fig.dpi;self.unit=72/fig.dpi;self.agg=agg
        self.width,self.height=fig.bbox.size;self.ops=[];self.viewports=viewports;self.halo_labels=halo_labels
    def points_to_pixels(self,p): return p/self.unit
    def flipy(self):return False
    def get_canvas_width_height(self):return self.width,self.height
    def get_text_width_height_descent(self,s,prop,ismath):
        runs=_runs(s,prop,ismath)
        width=sum(pdfmetrics.stringWidth(t,f,prop.get_size_in_points()) for t,f in runs)/self.unit
        # Match the source Arial baseline/ascent geometry; width uses the same
        # embedded font as the PDF, including the semantic mixed italic title.
        _,h,d=self.agg.get_text_width_height_descent(re.sub(r'\$|\\it\{|\}', '',s),prop,False)
        return width,h,d
    def _clip(self,gc):
        if gc.get_clip_path()[0] is not None:
            raise ValueError('Nonrectangular artist clip is unsupported')
        b=gc.get_clip_rectangle()
        if b is None:return None
        for i,box in enumerate(self.viewports):
            if np.allclose(b.bounds,box,rtol=0,atol=1e-5):return i
        return None
    def draw_path(self,gc,path,transform,rgbFace=None):
        if gc.get_hatch():raise ValueError('Hatching/Pattern is forbidden')
        p=transform.transform_path(path)
        self.ops.append(('path',self._clip(gc),p,gc.get_rgb(),rgbFace,
                         gc.get_linewidth(),gc.get_alpha(),gc.get_forced_alpha(),
                         gc.get_dashes(),gc.get_capstyle(),gc.get_joinstyle()))
    def draw_text(self,gc,x,y,s,prop,angle,ismath=False,mtext=None):
        self.ops.append(('text',None,float(x),float(y),str(s),prop.copy(),float(angle),
                         ismath,gc.get_rgb(),gc.get_alpha(),id(mtext) in self.halo_labels))
    def draw_image(self,*args,**kwargs):raise ValueError('Raster image is forbidden')
    def start_rasterizing(self):pass
    def stop_rasterizing(self):pass
    def start_filter(self):raise ValueError('Raster filter is forbidden')
    def stop_filter(self,*args):raise ValueError('Raster filter is forbidden')

def _pdf_path(pdf,path,unit):
    out=pdf.beginPath();current=None;start=None
    for vertices,code in path.iter_segments(remove_nans=True,simplify=False,curves=True):
        v=np.asarray(vertices).reshape(-1,2)*unit
        if code==MplPath.MOVETO:
            current=v[0];start=current;out.moveTo(*current)
        elif code==MplPath.LINETO:
            current=v[0];out.lineTo(*current)
        elif code==MplPath.CURVE3:
            control,end=v
            a=current+(control-current)*2/3;b=end+(control-end)*2/3
            out.curveTo(*a,*b,*end);current=end
        elif code==MplPath.CURVE4:
            out.curveTo(*v.ravel());current=v[-1]
        elif code==MplPath.CLOSEPOLY:
            out.close();current=start
        elif code==MplPath.STOP:break
        else:raise ValueError('Unknown path opcode '+str(code))
    return out

def strict_editable_pdf(fig,destination,*,data_axes=(),live_halo_width=.20):
    """Write a strict vector PDF. Return declared clip boxes in PDF points.

    The artist canvas remains unchanged after return. Rasterization flags and
    white text path effects are replaced by a narrow live text fill+stroke
    (render mode 2) in the same single label object; no outlined font paths are
    emitted. The width is narrower than Agg's halo to preserve the black stems.
    """
    _register()
    if len(data_axes)>4:raise ValueError('At most four sibling data viewports')
    fig.canvas.draw();agg=fig.canvas.get_renderer()
    viewports=[tuple(ax.bbox.bounds) for ax in data_axes]
    effects=[]
    for artist in fig.findobj():
        if hasattr(artist,'get_path_effects'):
            old=artist.get_path_effects()
            if old:effects.append((artist,old));artist.set_path_effects([])
    rec=_Recorder(fig,agg,viewports,{id(a) for a,e in effects})
    try:fig.draw(rec)
    finally:
        for a,e in effects:a.set_path_effects(e)
    dst=Path(destination);dst.parent.mkdir(parents=True,exist_ok=True)
    pdf=canvas.Canvas(str(dst),pagesize=tuple(fig.get_size_inches()*72),
                      pageCompression=1,invariant=1,initialFontName=_FONTS['normal'])
    unit=rec.unit
    def emit(op):
        if op[0]=='text':
            _,_,x,y,s,prop,angle,ismath,color,alpha,halo=op
            if not s:return
            pdf.saveState();pdf.setFillColorRGB(*color[:3]);pdf.setFillAlpha(alpha)
            text=pdf.beginText();rad=math.radians(angle)
            if halo and live_halo_width>0:
                pdf.setStrokeColorRGB(1,1,1);pdf.setStrokeAlpha(alpha)
                pdf.setLineWidth(float(live_halo_width));text.setTextRenderMode(2)
            text.setTextTransform(math.cos(rad),math.sin(rad),-math.sin(rad),math.cos(rad),x*unit,y*unit)
            for value,font in _runs(s,prop,ismath):
                if value:text.setFont(font,prop.get_size_in_points());text.textOut(value)
            pdf.drawText(text);pdf.restoreState();return
        _,_,path,stroke,face,lw,alpha,forced,dashes,cap,join=op
        pdf.saveState();pdf.setLineWidth(lw)
        pdf.setLineCap({'butt':0,'round':1,'projecting':2}[cap])
        pdf.setLineJoin({'miter':0,'round':1,'bevel':2}[join])
        offset,pattern=dashes
        pdf.setDash([] if pattern is None else [float(v) for v in pattern],0.0 if offset is None else float(offset))
        if stroke is not None:
            pdf.setStrokeColorRGB(*stroke[:3]);pdf.setStrokeAlpha(alpha if forced else (stroke[3] if len(stroke)>3 else alpha))
        if face is not None:
            pdf.setFillColorRGB(*face[:3]);pdf.setFillAlpha(alpha if forced else (face[3] if len(face)>3 else alpha))
        pdf.drawPath(_pdf_path(pdf,path,unit),stroke=int(stroke is not None and lw>0),fill=int(face is not None),fillMode=1)
        pdf.restoreState()
    done=set()
    for op in rec.ops:
        key=op[1]
        if key is None:emit(op);continue
        if key in done:continue
        done.add(key);pdf.saveState()
        p=pdf.beginPath();p.rect(*(np.asarray(viewports[key])*unit))
        pdf.clipPath(p,stroke=0,fill=0,fillMode=1)
        for candidate in rec.ops:
            if candidate[1]==key:emit(candidate)
        pdf.restoreState()
    pdf.showPage();pdf.save()
    if done!=set(range(len(data_axes))):raise AssertionError('Declared clip has no artist')
    return {'expected_clips':len(done),'clip_boxes_pt':(np.asarray(viewports)*unit).tolist(),
            'editable_label_objects':sum(op[0]=='text' for op in rec.ops),
            'label_strings':[op[4] for op in rec.ops if op[0]=='text'],
            'vector_paths':sum(op[0]=='path' for op in rec.ops),
            'text_effects':'White live text fill+stroke, single object; stroke width '+str(live_halo_width)+' pt.',
            'halo_labels':len(effects)}
```

```python
#8. =========== Reference-proportional rendering
# Positions are left/top/width/height on the supplied 1458 x 313 reference.
REFERENCE_W, REFERENCE_H, LOGICAL_PPI = 1458, 313, 75
FIGSIZE = (REFERENCE_W / LOGICAL_PPI, np.nextafter(REFERENCE_H / LOGICAL_PPI, np.inf))
DPI = 600
REFERENCE_SHA256 = '72566b7a48fdf5247f9e6e9a6b364c70b018ecbb39b5ef43b2fd80641d304526'
C_DISPLAY_TITLES = {
    'Foxo6': {0:'Transient developmental', 1:'Broad early-rising', 2:'Early transition-associated'},
    'Zfpm2': {0:'Neuronal morphogenesis', 1:'Broad early-rising', 2:'Transient developmental', 3:'Early-high (inverse)'},
    'Neurod2': {0:'Broad early-rising', 1:'Transient developmental'},
    'Nhlh1': {0:'Transient developmental', 1:'Weakly varying', 2:'Early transition-associated'},
    'Neurod1': {0:'Postsynaptic (inverse)', 1:'Mid-trajectory peak (inverse)', 2:'Late-rising Calb1/Kcna1',
                3:'Progressively rising', 4:'Early-high', 5:'Late-rising Cnr1/Kcnq2', 6:'Transient differentiation'}}

def reference_box(x, y, width, height):
    return [x / REFERENCE_W, 1 - (y + height) / REFERENCE_H,
            width / REFERENCE_W, height / REFERENCE_H]

def reference_text(fig, x, baseline_y, text, **kwargs):
    return fig.text(x / REFERENCE_W, 1 - baseline_y / REFERENCE_H,
                    text, va='baseline', **kwargs)

import math
import hashlib
from pathlib import Path
import numpy as np
import pandas as pd
def render_factor(result, factor, output_dir, seed=42):
    import matplotlib
    matplotlib.use('Agg')
    import matplotlib.pyplot as plt
    from matplotlib.colors import LinearSegmentedColormap, Normalize
    from matplotlib.lines import Line2D
    B_POINT_AREA = 1.5
    C_POINT_AREA = 3.1
    BACKGROUND_ALPHA = .20
    FOREGROUND_ALPHA = .90
    NETWORK_FONT = 11.4
    PALETTE = {'CVG': '#1A6D79', 'T': '#1A6D79', 'VGN': '#00008B',
        'AB': '#7A4E94', 'AC': '#3585BC', 'Ia': '#F37271',
        'Ib': '#44B9DE', 'Ic': '#97C8A8', 'II': '#E377C2'}
    LEFT_COLOR, RIGHT_COLOR = '#1A6D79', '#97C8A8'
    plt.rcParams.update({'font.family': 'Arial', 'font.size': 8, 'axes.linewidth': .65,
                         'xtick.direction': 'out', 'ytick.direction': 'out'})
    assert factor in ('Foxo6', 'Zfpm2', 'Neurod2', 'Nhlh1', 'Neurod1'), factor
    obs, xy = result['obs'], result['X_umap']
    modules = result['modules']
    chosen = result['selected_modules']
    titles = result['reviewed_annotations']['reviewed_program_titles_json']
    reviewed = list(titles.values()) if isinstance(titles, dict) else titles
    on_path = obs.seg.isin(('CVG', 'T', 'AC', 'Ic')).to_numpy()
    grid, grid_fit = result['fit_grid'], result['grid_fitted']
    chosen = list(map(int, chosen))  # Fixed user order, never peak-sorted.
    assert len(chosen) == 2
    names = {}
    for m in chosen:
        gene_hash = hashlib.sha256('\n'.join(sorted(modules[m]['genes'])).encode()).hexdigest()
        found = [d for d in reviewed if d['TF'] == factor and d['gene_sha256'] == gene_hash]
        assert len(found) == 1, 'Review descriptive titles after changing gene sets'
        names[m] = found[0]['title']
    fig = plt.figure(figsize=FIGSIZE)
    text_items, curve_axes = [], []
    cm_b = LinearSegmentedColormap.from_list('activity', ['#D4D4D4', '#FFFFFF', '#F3151D'])
    activity = result['activity']
    status = str(activity['status'])
    if status not in ('available', 'not_estimable'):
        raise ValueError('Unknown activity status: ' + status)
    bscore = np.asarray(activity['fitted'], dtype=float)
    bmask = np.asarray(activity['display_mask'])
    assert bscore.shape == (len(obs),) and bmask.shape == (len(obs),)
    assert bmask.dtype.kind == 'b', 'Activity display_mask must be boolean'
    reason = str(activity['reason'])
    btitle = r'$\it{' + factor + '}$ (+) activity' if status == 'available' else r'$\it{' + factor + '}$ activity'
    if status == 'available':
        bmax = float(activity['display_vmax'])
        assert np.isfinite(bmax) and bmax > 0
        assert bmask.any() and np.isfinite(bscore[bmask]).all()
        metric = 'Regulon activity (AUC)'
    else:
        assert activity['display_vmax'] is None, 'Unavailable activity has no AUC color scale'
        assert not bmask.any(), 'Unavailable activity cannot display scored cells'
        assert np.isnan(bscore).all(), 'Unavailable activity must stay NaN, not zero'
        bmax, metric = None, 'Activity not estimable'
    lim = (xy.min(axis=0) - 1, xy.max(axis=0) + 1)
    def embedding(ax, values, cmap, vmax, mask=None):
        ax.scatter(xy[:, 0], xy[:, 1], s=B_POINT_AREA, c='#D4D4D4', alpha=.55, linewidths=0)
        sc = None
        if values is not None:
            mask = np.isfinite(values) if mask is None else mask & np.isfinite(values)
            order = np.flatnonzero(mask)[np.argsort(values[mask], kind='stable')]
            sc = ax.scatter(xy[order, 0], xy[order, 1], s=B_POINT_AREA, c=values[order],
                cmap=cmap, norm=Normalize(0, vmax, clip=True), alpha=.88, linewidths=0)
        ax.set_xlim(lim[0][0], lim[1][0]); ax.set_ylim(lim[0][1], lim[1][1])
        ax.set_aspect('equal'); ax.axis('off')
        return sc
    bax = fig.add_axes(reference_box(10, 52, 262, 206))
    sc = embedding(bax, bscore if status == 'available' else None, cm_b, bmax, mask=bmask if status == 'available' else None)
    text_items.append(reference_text(fig,137,40,btitle,fontsize=15.36,ha='center'))
    if status == 'available':
        cax = fig.add_axes(reference_box(78,278,118,8))
        cb = fig.colorbar(sc, cax=cax, orientation='horizontal')
        cb.set_ticks([0, bmax]); cb.ax.tick_params(labelsize=11.52, length=2, pad=1)
        cb.outline.set_linewidth(.4)
        cb.solids.set_edgecolor('face'); cb.solids.set_linewidth(.12)
        text_items.extend(cax.get_xticklabels())
    text_items.append(reference_text(fig,137,271,metric,fontsize=11.52,ha='center'))
    rng = np.random.default_rng(seed)
    background = rng.permutation(len(obs)); foreground = rng.permutation(np.flatnonzero(result['primary_path']))
    colors = np.array([PALETTE[s] for s in obs.seg])
    if not chosen:
        for i in range(2):
            ax = fig.add_axes(reference_box(318,51 if i == 0 else 179,220,103))
            curve_axes.append(ax)
            ax.set_xlim(0, 6.12); ax.set_ylim(0, .90)
            ax.set_xticks([0, 2, 4, 6]); ax.set_yticks([0, .45, .90])
            ax.tick_params(labelsize=7.68, length=2.5, width=.6, pad=2)
            text_items.append(ax.set_title('Program unavailable', fontsize=8.0, pad=4))
            text_items.append(ax.text(.5, .5, '\u2014', transform=ax.transAxes,
                                      ha='center', va='center', color='#999999', fontsize=16))
            if i == 0:
                ax.tick_params(labelbottom=False)
            else:
                text_items.append(ax.set_xlabel('Pseudotime', labelpad=2, fontsize=11.52))
    for i, m in enumerate(chosen):
        ax = fig.add_axes(reference_box(318,51 if i == 0 else 179,220,103))
        curve_axes.append(ax)
        values = result['module_scores'][:, m]
        ax.scatter(obs.t.to_numpy()[background], values[background], s=C_POINT_AREA,
                   c=colors[background], alpha=BACKGROUND_ALPHA, linewidths=0)
        ax.scatter(obs.t.to_numpy()[foreground], values[foreground], s=C_POINT_AREA,
                   c=colors[foreground], alpha=FOREGROUND_ALPHA, linewidths=0)
        ax.plot(grid, grid_fit[:, modules[m]['job']], lw=1.05, color='black', zorder=5)
        if i == 0:
            ax.axvline(result['fork_t'], lw=.65, ls=(0, (4, 3)), color='black')
            ax.tick_params(labelbottom=False)
        else:
            text_items.append(ax.set_xlabel('Pseudotime', labelpad=2, fontsize=11.52))
        ax.set_xlim(0, 6.12); ax.set_ylim(0, .90)
        ax.set_xticks([0, 2, 4, 6]); ax.set_yticks([0, .45, .90])
        ax.tick_params(labelsize=7.68, length=2.5, width=.6, pad=2)
        title = f'{C_DISPLAY_TITLES[factor][m]}  (n={modules[m]["n_genes"]})'
        text_items.append(reference_text(fig,428,46 if i == 0 else 174,title,fontsize=12.48,ha='center'))
    # The reference omits a vertical RNA-axis title; units remain in the archive.
    network = None
    spatial_limits = []
    if len(chosen) == 2:
        # The two spatial maps show precisely the same modules as the two C panels.
        for i, (m, color, x) in enumerate(zip(chosen, (LEFT_COLOR, RIGHT_COLOR), (586,1258))):
            dcol = int(np.flatnonzero(result['D_jobs'] == modules[m]['job'])[0])
            dscore = result['D_fitted'][:, dcol]
            dmax = math.ceil(np.nanmax(dscore[result['good']]) * 20) / 20
            spatial_limits.append(dmax)
            np.testing.assert_array_equal(dscore[on_path], result['fitted'][on_path, modules[m]['job']])
            dax = fig.add_axes(reference_box(x,91,198,155.3))
            cmap = LinearSegmentedColormap.from_list('program' + str(i), ['#D4D4D4', '#FFFFFF', color])
            scd = embedding(dax, dscore, cmap, dmax, mask=result['good'])
            barax = fig.add_axes(reference_box(636 if i == 0 else 1308,272,118,7))
            cbar = fig.colorbar(scd, cax=barax, orientation='horizontal')
            cbar.set_ticks([0, dmax]); cbar.ax.tick_params(labelsize=11.52, length=2, pad=1); cbar.outline.set_linewidth(.4)
            cbar.solids.set_edgecolor('face'); cbar.solids.set_linewidth(.12)
            text_items.extend(barax.get_xticklabels())
        axn = fig.add_axes(reference_box(780,62,480,224))
        network = draw_module_network(axn, factor, modules[chosen[0]]['genes'], modules[chosen[1]]['genes'],
            kind='motif' if factor == 'Nhlh1' else 'associated', left_color='#2D7184',
            right_color='#5AAD7F', font_size=NETWORK_FONT,
            highlight_genes=NETWORK_HIGHLIGHTS.get(factor, ()))
        text_items.extend(network['labeltexts'])
        left_title = 'Postsynaptic receptor regulation (inverse)' if factor == 'Neurod1' and chosen[0] == 0 else names[chosen[0]]
        text_items.append(reference_text(fig,783,40,left_title,fontsize=15.36,color='#00585D',ha='center'))
        text_items.append(reference_text(fig,1228,40,names[chosen[1]],fontsize=15.36,color='#006600',ha='center'))
        # Edge evidence is recorded in the notebook/archive, as in the supplied layout.
    else:
        text_items.append(fig.text(.67, .52, 'No supported early program', ha='center', fontsize=12, color='#777777'))
    for x, letter in ((16, 'B'), (284, 'C'), (548, 'D')):
        text_items.append(reference_text(fig,x,34,letter,fontsize=34.56,fontweight='normal'))
    legend_labels = {'CVG':'CVG','VGN':'VGN*','T':'SGN trunk','AC':'Ib/Ic shared','AB':'Ia/II shared',
                     'Ia':'Ia','Ib':'Ib','Ic':'Ic','II':'II'}
    legend_records = []
    for segment, y in zip(('CVG','VGN','T','AC','AB','Ia','Ib','Ic','II'),(54,67,80,94,107,121,134,147.5,161)):
        label = legend_labels[segment]
        text = fig.text(291/REFERENCE_W,1-y/REFERENCE_H,label,ha='right',va='center',fontsize=12.48)
        text_items.append(text)
        fig.canvas.draw()
        logical_width = text.get_window_extent(fig.canvas.get_renderer()).width * LOGICAL_PPI / fig.dpi
        marker_x = 291 - logical_width - 8
        fig.add_artist(Line2D([marker_x/REFERENCE_W],[1-y/REFERENCE_H],transform=fig.transFigure,
            marker='o',linestyle='None',color=PALETTE[segment],markersize=7.68,markeredgewidth=0))
        legend_records.append({'segment':segment,'label':label,'marker_xy':[marker_x,y],'text_right_x':291})
    for ax in curve_axes:
        text_items.extend(ax.get_xticklabels()); text_items.extend(ax.get_yticklabels())
    fig.canvas.draw()
    renderer = fig.canvas.get_renderer()
    clipped = [t.get_text() for t in text_items if t.get_visible() and t.get_text() and
        (not fig.bbox.contains(*t.get_window_extent(renderer).p0) or not fig.bbox.contains(*t.get_window_extent(renderer).p1))]
    assert not clipped, clipped
    assert network is None or network['QA']['ok'], network['QA']
    output_dir = Path(output_dir); output_dir.mkdir(parents=True, exist_ok=True)
    path = output_dir / f'{factor}.png'
    fig.savefig(path, dpi=DPI)
    pdf_path = output_dir / f'{factor}.pdf'
    pdf_info = strict_editable_pdf(fig, pdf_path, data_axes=curve_axes)
    plt.close(fig)
    result['figure'] = {'filename': path.name, 'filename_png': path.name, 'filename_pdf': pdf_path.name,
        'pdf_export': pdf_info, 'dpi': DPI, 'figsize_in': FIGSIZE,
        'B_activity_status': status, 'B_activity_reason': reason,
        'B_data': ('Motif-supported regulon AUCell GAM; no RNA or associated-program substitution' if status == 'available' else 'Unavailable activity; gray embedding only; no activity values plotted'),
        'B_display_mask_cells': int(bmask.sum()), 'B_display_vmax': bmax,
        'C_D_module_ids': chosen, 'C_D_comparison_applicable': len(chosen) == 2,
        'C_D_gene_sets_equal': True if len(chosen) == 2 else None,
        'C_D_primary_fitted_equal': True if len(chosen) == 2 else None,
        'title_records': {str(m): names[m] for m in chosen}, 'clipped_texts': clipped,
        'network': None if network is None else {k: v for k, v in network.items() if k not in ('labeltexts', 'edgepatches', 'factor_text')},
        'palette': PALETTE, 'C_background_alpha': BACKGROUND_ALPHA, 'C_foreground_alpha': FOREGROUND_ALPHA,
        'VGN_asterisk': 'Candidate VGN tree arm, distinct from independently curated VGN reference',
        'D_colorbar_maxima': spatial_limits,
        'C_display_titles': {str(m):C_DISPLAY_TITLES[factor][m] for m in chosen},
        'reference_layout': {'width':REFERENCE_W,'height':REFERENCE_H,'logical_ppi':LOGICAL_PPI,'sha256':REFERENCE_SHA256,
            'C_axes_xywh':[[318,51,220,103],[318,179,220,103]],'network_xywh':[780,62,480,224],
            'legend':legend_records,'font_sizes_pt':{'panel':34.56,'B_D_title':15.36,'C_title':12.48,'C_tick':7.68,'network':NETWORK_FONT}},
        'highlight_evidence': NETWORK_HIGHLIGHT_EVIDENCE.get(factor, {}),
        'highlight_semantics': 'Literature-discussed representatives; red is not significance, edge strength, direct binding, or fate determination.',
        'spatial_values': 'Module fitted RNA, not TF RNA or AUCell. Each colorbar maximum is the actual fitted maximum rounded up to0.05; stored fitted values are not clipped. Compare spatial patterns within a module, not saturation across modules.'}
    audit = result['association_audit'].set_index('gene')
    rows = []
    for side, m in enumerate(chosen):
        for gene in modules[m]['genes']:
            rho = float(audit.loc[gene, 'mean_library_rho'])
            rows.append({'factor': factor, 'gene': gene, 'module_id': m, 'side': 'left' if side == 0 else 'right',
                'program_title': names[m], 'evidence': 'motif-supported prediction' if factor == 'Nhlh1' else 'RNA association',
                'mean_library_RNA_rho': rho,
                'RNA_association_sign': 'not encoded for motif targets' if factor == 'Nhlh1' else ('positive' if rho > 0 else 'negative'),
                'arrowhead': False})
    result['network_edges'] = pd.DataFrame(rows)
    assert len(rows) == sum(modules[m]['n_genes'] for m in chosen)
    return path
```

```python

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")
#9. =========== Render five figures, validate both formats, and save necessary results
from concurrent.futures import ThreadPoolExecutor
VALIDATION_WORKERS = 2  # Independent PDF checks overlap the next figure export.
validation_pool = ThreadPoolExecutor(max_workers=VALIDATION_WORKERS)
validation_futures = {}
source_cells = [c.source for c in nbformat.read(NOTEBOOK_FILE,as_version=4).cells if c.cell_type=='code']
code_sha = hashlib.sha256('\n\n'.join(source_cells).encode()).hexdigest()
module_results = all_program_results()
shared = module_results[FACTORS[0]]
obs = shared['obs']
for result in module_results.values():
    pd.testing.assert_frame_equal(result['obs'],obs)
    for key in ('X_umap','good','primary_path'):
        np.testing.assert_array_equal(result[key],shared[key])
raw_auc, selected_regulons, activity_audit, activity_metadata = read_activity_inputs(obs.index)
fitted_activity, fit_methods = fit_activity(raw_auc,obs,shared['good'])
manifest, all_results = [], {}
for j,factor in enumerate(FACTORS):
    result = module_results[factor]
    result['activity'] = activity_for_factor(factor,j,raw_auc,fitted_activity,obs,
        result['good'],selected_regulons,activity_audit,fit_methods)
    result['B_raw'] = raw_auc[:,j]
    result['fitted'][:,0] = fitted_activity[:,j]
    np.testing.assert_array_equal(result['D_fitted'][obs.seg.isin(('CVG','T','AC','Ic'))],
        result['fitted'][obs.seg.isin(('CVG','T','AC','Ic'))][:,result['D_jobs']])
    png = render_factor(result,factor,FIGURE_DIR,seed=SEED)
    pdf = png.with_suffix('.pdf')
    pdf_info = result['figure']['pdf_export']
    validate_args = [str(VALIDATOR_PYTHON),str(VALIDATOR),str(png),str(pdf),
        '--expected-size-in',f'{FIGSIZE[0]}x{FIGSIZE[1]}','--expected-clips',str(pdf_info['expected_clips']),
        # Set this storage path independently before running.
        '--pdftoppm',str(PDFTOPPM),'--render-dpi','150','--render-dir',str(ROOT)]
    for box in pdf_info['clip_boxes_pt']:
        validate_args += ['--clip-box',','.join(format(float(x),'.8f') for x in box)]
    validation_futures[factor] = validation_pool.submit(subprocess.run,validate_args,capture_output=True,text=True,errors='replace')
    result['render_provenance'] = {'code_sha256':code_sha,'executed_code_cells':source_cells,
        'activity_file_sha256':file_sha256(ACTIVITY_FILE),
        'PNG_sha256':file_sha256(png),'PDF_sha256':file_sha256(pdf),
        'python':sys.version,'platform':platform.platform(),
        'B_methods':activity_metadata,'C_D_reselected':True,'all_programs_saved':True,'Prox1_rendered':False}
    manifest.append({'TF':factor,'B_status':result['activity']['status'],
        'C_D_genes':'+'.join(str(result['modules'][m]['n_genes']) for m in result['selected_modules']),
        'All_programs':len(result['modules']),
        'PNG':png.name,'PDF':pdf.name})
    all_results[factor] = result
    print(factor + ': exported; validation queued',flush=True)
    display(Image(filename=str(png),width=1458))
try:
    for factor,result in all_results.items():
        checked = validation_futures[factor].result()
        if checked.returncode: raise RuntimeError(checked.stdout + checked.stderr)
        result['figure']['output_validation'] = {'status':'PASS','validator_stdout':checked.stdout}
        print(factor + ': PNG/PDF PASS',flush=True)
finally:
    validation_pool.shutdown(wait=True)

# Set this storage path independently before running. Fig4B-D.pending.h5.
temporary = ROOT
with h5py.File(temporary,'w') as h:
    h.attrs['schema']='Fig4BD_all_programs_v2'
    common = h.create_group('shared')
    shared_names = ('obs','X_umap','good','primary_path','fork_t')
    for key in shared_names: write_result_value(common,key,shared[key])
    for factor,result in all_results.items():
        target=h.create_group('factors/'+factor)
        for key,value in result.items():
            if key not in shared_names: write_result_value(target,key,value)
    write_result_value(h,'manifest',pd.DataFrame(manifest))
    write_result_value(h,'provenance',{'code_sha256':code_sha,'activity_file_sha256':file_sha256(ACTIVITY_FILE),
        'upstream_module_file':str(UPSTREAM_MODULES.relative_to(PROJECT_ROOT)),
        'upstream_module_sha256':file_sha256(UPSTREAM_MODULES),
        'methods':'SCENIC motif-pruned signatures -> fixed-background AUCell -> branch-wise GAM; all 19 frozen C/D programs fitted and retained; fixed user-selected display',
        'scope':'Five named TFs only. Unsupported activity is missing, not zero.'})
temporary.replace(RESULT_FILE)
assert sum(len(r['modules']) for r in all_results.values()) == 19
assert sorted(p.name for p in RESULT_DIR.iterdir() if p.is_file() and p.name != 'SGN-lines.h5') == ['Activity.h5','Fig4B-D.h5']
display(pd.DataFrame(manifest))
```

## SGN selected-gene age curves

Seven requested genes, using the exact VGN-lines aggregation and layout. The legend contains only CVG, VGN and SGN; colored individual-library points remain visible. The original B–D results are retained. Frozen source files are recorded in the new SGN-lines.h5; that file also supports independent replotting.

```python

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")

# Cleanup must never interpret the project/current directory as a temporary target.
def _release_cleanup_target(path):
    candidate = Path(path).resolve()
    current = Path.cwd().resolve()
    if candidate == current or candidate in current.parents or candidate == Path(candidate.anchor):
        raise RuntimeError("Configure the owned temporary cleanup path before running this section.")
    return path

#11. =========== SGN selected-gene age curves
# Append-only analysis; the prior B/C/D figures and two H5 files are protected.
from pathlib import Path
import os, sys, json, hashlib, platform, subprocess, shutil
# Set this storage path independently before running.
SGN_LINE_ROOT = ROOT
SGN_LINE_GENES = ('Foxo6','Zfpm2','Neurod2','Nhlh1','Neurod1','Prox1','Insm1')
# Set this storage path independently before running.
SGN_LINE_FIGURE_DIR = ROOT
# Set this storage path independently before running. SGN-lines.h5.
SGN_LINE_RESULT = ROOT
# Set this storage path independently before running. 05_fig4B-D_SGN.ipynb. Use the supplied fig4_SGN_program_analysis.ipynb source.
SGN_LINE_NOTEBOOK = ROOT
# Set this storage path independently before running.
SGN_LINE_WORK = ROOT
SGN_LINE_FIGSIZE, SGN_LINE_DPI = (15.2,6.4),600
REBUILD_SGN_LINES = False  # True after changing data preparation; appearance changes only redraw.
for p in (SGN_LINE_WORK,SGN_LINE_FIGURE_DIR,SGN_LINE_RESULT.parent): p.mkdir(parents=True,exist_ok=True)
# Set this storage path independently before running.
os.environ['MPLCONFIGDIR']=str(ROOT)
sys.dont_write_bytecode=True
import h5py
import numpy as np
import pandas as pd
import nbformat
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.ticker import FixedLocator
from IPython.display import display,Image
from urllib.parse import quote,unquote
SGN_LINE_DATA_SOURCE_SHA256 = 'b712bd1fc4c64b302907823fc0df88428256e84c447e4a1d2a56a18f8dd2f416'


SGN_LINE_INPUTS = {
    # Set this storage path independently before running. RNA-screen.h5.
    'RNA': 'ROOT',
    # Set this storage path independently before running. activity-screen.h5.
    'identity': 'ROOT',
    # Set this storage path independently before running. VGN520-common-counts.npz.
    'VGN_membership': 'ROOT',
}
SGN_EXPECTED_AGES = {
    'CVG': [9.5, 10.5, 11.5, 13.5],
    'VGN': [11.5, 13.5],
    'SGN': [11.5, 13.5, 14.5, 15.5, 16.5, 17.5, 18.5],
}
sys.dont_write_bytecode = True


def _sgn_line_digest(path):
    value = hashlib.sha256()
    with Path(path).open('rb') as stream:
        for block in iter(lambda: stream.read(4 * 1024 * 1024), b''):
            value.update(block)
    return value.hexdigest()


def _sgn_line_array(node):
    if isinstance(node, h5py.Dataset):
        return node.asstr()[()] if node.dtype.kind in 'OUS' else node[()]
    categories = _sgn_line_array(node['categories'])
    codes = node['codes'][()]
    result = np.full(len(codes), '', dtype=object)
    valid = codes >= 0
    result[valid] = categories[codes[valid]]
    return result


def _sgn_line_frame(group):
    return pd.DataFrame({key: _sgn_line_array(group[key]) for key in group})


def _sgn_line_bool(values):
    values = np.asarray(values)
    return values if values.dtype.kind == 'b' else np.isin(values.astype(str), ['True', 'true', '1'])


#2. =========== Native library pseudobulk, then equal-library/equal-source means
def prepare_sgn_lines(root, genes=SGN_LINE_GENES):
    """Return native RNA tables; no disk writes, inference, normalization refit or plotting.

    The flat ``line_units``, ``line_source`` and ``line_points`` tables support
    the existing VGN ``render_lines(data)`` directly; each table is stored once.
    """
    root, genes = Path(root).resolve(), tuple(map(str, genes))
    assert genes and len(set(genes)) == len(genes), 'Gene names must be nonempty and unique'
    paths = {key: root / value for key, value in SGN_LINE_INPUTS.items()}
    hashes = {str(path.relative_to(root)): _sgn_line_digest(path) for path in paths.values()}
    with h5py.File(paths['RNA'], 'r') as handle:
        obs = _sgn_line_frame(handle['cells'])
        tfs = handle['TFs'].asstr()[()]
        positions = pd.Index(tfs).get_indexer(genes)
        if np.any(positions < 0):
            absent = np.asarray(genes)[positions < 0].tolist()
            raise ValueError(f'Requested genes absent from the frozen native-TF axis: {absent}')
        # HDF5 fancy indexing requires monotonically increasing source columns.
        order = np.argsort(positions)
        counts = np.full((len(obs), len(genes)), -1, dtype=np.int64)
        counts[:, order] = handle['counts'][:, positions[order]]
        native_measured = np.zeros(counts.shape, dtype=bool)
        native_measured[:, order] = handle['cell_TF_measured'][:, positions[order]]
        vgn = _sgn_line_frame(handle['VGN_native_library_pseudobulk'])
    with h5py.File(paths['identity'], 'r') as handle:
        aobs = _sgn_line_frame(handle['obs'])
        np.testing.assert_array_equal(obs.cell_id, handle['cells'].asstr()[()])
        np.testing.assert_array_equal(obs.cell_id, aobs['_index'])
    assert len(obs) == 7355 and obs.cell_id.is_unique
    np.testing.assert_array_equal(native_measured, counts >= 0)
    assert np.all(obs.library_total > 0)
    good = (_sgn_line_bool(obs.source_state_evaluation_allowed)
            & ~_sgn_line_bool(obs.source_rna_mixture_review_flag))
    obs['identity'] = np.where(obs.study_id.eq('GSE178931'),
        np.where(aobs.cell_type_plot.eq('Differentiating SGN'), 'SGN', 'CVG'),
        np.where(obs.study_id.eq('GSE231513'), 'CVG-like', 'SGN'))

    with np.load(paths['VGN_membership'], allow_pickle=False) as handle:
        reference = pd.DataFrame({'cell': handle['cells'], 'stage': handle['stage'],
            'library': handle['library_id'], 'total': handle['full_endogenous_total']})
    assert len(reference) == 520 and reference.cell.is_unique
    assert not set(reference.cell).intersection(obs.cell_id)
    assert (reference.total > 0).all()
    vgn_expected = reference.groupby(['stage', 'library']).agg(
        cells=('cell', 'size'), full_endogenous_total_sum=('total', 'sum'))

    records, missing_units = [], []
    for (study, stage, identity, library), part in obs.loc[good].groupby(
            ['study_id', 'stage_standardized', 'identity', 'library_id']):
        ii = part.index.to_numpy()
        total = float(part.library_total.sum())
        values = counts[ii]
        measured = (values >= 0).all(axis=0)
        np.testing.assert_array_equal(measured, (values >= 0).any(axis=0))
        for j, gene in enumerate(genes):
            if not measured[j]:
                missing_units.append({'gene': gene, 'study': study, 'stage': stage,
                    'identity': identity, 'library': library, 'cells': len(ii),
                    'reason': 'Native source feature was not measured; omitted, never zero-filled'})
                continue
            cpm = float(values[:, j].sum() / total * 1e6)
            records.append({'gene': gene, 'study': study, 'stage': stage,
                'identity': identity, 'library': library, 'cells': len(ii),
                'expr': float(np.log10(1 + cpm))})

    vgn_selected = vgn.loc[vgn.TF.isin(genes)].copy()
    assert not vgn_selected.duplicated(['TF', 'study', 'stage', 'library_id']).any()
    for gene in genes:
        assert len(vgn_selected.loc[vgn_selected.TF.eq(gene)]) == len(vgn_expected)
    for _, row in vgn_selected.iterrows():
        expected = vgn_expected.loc[(row.stage, row.library_id)]
        assert int(row.cells) == int(expected.cells)
        assert float(row.full_endogenous_total_sum) == float(expected.full_endogenous_total_sum)
        np.testing.assert_allclose(row.log10_1p_full_endogenous_pseudobulk_CPM,
            np.log10(1 + row.full_endogenous_pseudobulk_CPM), atol=1e-12, rtol=0)
        records.append({'gene': row.TF, 'study': row.study, 'stage': row.stage,
            'identity': 'VGN', 'library': row.library_id, 'cells': int(row.cells),
            'expr': float(row.log10_1p_full_endogenous_pseudobulk_CPM)})

    units = pd.DataFrame(records)
    assert not units.duplicated(['gene', 'study', 'stage', 'identity', 'library']).any()
    assert np.isfinite(units.expr).all()
    units['age'] = units.stage.str[1:].astype(float)
    units['display_type'] = units.identity.replace({'CVG-like': 'CVG'})
    # These two means deliberately follow _vgn_line_data, including log before averaging.
    source = units.groupby(['gene', 'study', 'stage', 'age', 'display_type'],
                           as_index=False).expr.mean()
    points = source.groupby(['gene', 'stage', 'age', 'display_type'],
                            as_index=False).expr.mean()
    for gene in genes:
        for identity, ages in SGN_EXPECTED_AGES.items():
            selected = points.loc[points.gene.eq(gene)
                & points.display_type.eq(identity)].sort_values('age')
            assert selected.age.tolist() == ages
    assert points.expr.between(0, 4.1).all()
    missing = {gene: int((counts[:, j] < 0).sum()) for j, gene in enumerate(genes)}
    missing_table = pd.DataFrame(missing_units, columns=[
        'gene', 'study', 'stage', 'identity', 'library', 'cells', 'reason'])
    for relative, expected in hashes.items():
        assert _sgn_line_digest(root / relative) == expected, relative
    if genes == SGN_LINE_GENES:
        assert (len(units), len(source), len(points)) == (338, 109, 91)
        assert missing == {g: (4616 if g == 'Zfpm2' else 0) for g in genes}
        assert len(missing_table) == 12 and missing_table.study.eq('GSE195500').all()

    lines = {
        'genes': np.asarray(genes), 'line_units': units, 'line_source': source,
        'line_points': points, 'line_missing_source_features': missing,
        'missing_library_genes': missing_table,
        'source_hashes': hashes,
        'methods': {
            # Set this storage path independently before running. 05_fig4B-D_VGN.ipynb. Use the supplied fig4_VGN_program_analysis.ipynb source.
            'adapted_from': 'Code_def/05_fig4B-D_VGN.ipynb::_vgn_line_data',
            'unit': 'log10(1+sum raw counts/sum full endogenous library totals*1e6)',
            'averaging': 'Equal library weights within source/stage/type, then equal source weights',
            'E10.5': 'CVG-like source identity retained; ordinary CVG display type and marker',
            'VGN': 'Independent curated VGN520; no candidate tree-arm substitution',
            'missing': 'Native source feature count=-1; absent library-gene rows omitted, never replaced by zero',
            'scope': 'Original frozen identities and quality flags; no new marker selection, clustering, fitting or expression imputation',
            'limits': ['Source averaging does not remove platform effects.',
                       'Age curves are cross-sectional descriptive summaries, not lineage tracking.',
                       'GSM/library identity is not assumed to be an independently documented animal replicate.',
                       'E9.5 is 12 frozen atlas cells from one library.'],
        },
        'diagnostics': {'native_atlas_cells': 7355, 'curated_VGN_cells': 520,
            'library_gene_rows': len(units), 'source_gene_rows': len(source),
            'curve_points': len(points), 'source_inputs_unchanged': True,
            'all_seven_genes_in_requested_order': genes == SGN_LINE_GENES},
    }
    return lines


# Serialization for the single SGN-lines result
def _sgn_line_plain(value):
    if isinstance(value, np.generic): return value.item()
    if isinstance(value, np.ndarray): return value.tolist()
    if isinstance(value, Path): return str(value)
    raise TypeError(type(value).__name__)


def write_sgn_line_value(parent, key, value):
    if isinstance(value, pd.DataFrame):
        group = parent.create_group(key)
        group.attrs['kind'], group.attrs['columns'] = 'dataframe', json.dumps(value.columns.tolist())
        write_sgn_line_value(group, '_index', value.index.to_numpy())
        for column in value: write_sgn_line_value(group, column, value[column].to_numpy())
    elif isinstance(value, dict):
        group = parent.create_group(key); group.attrs['kind'] = 'dict'
        for name, item in value.items(): write_sgn_line_value(group, quote(str(name), safe=''), item)
    elif isinstance(value, np.ndarray):
        node = (parent.create_dataset(key, data=value.astype(str).astype(object), dtype=h5py.string_dtype())
                if value.dtype.kind in 'OUS' else parent.create_dataset(
                    key, data=value, compression='gzip' if value.ndim else None))
        node.attrs['kind'] = 'ndarray'
    else:
        node = parent.create_dataset(key, data=json.dumps(value, default=_sgn_line_plain), dtype=h5py.string_dtype())
        node.attrs['kind'] = 'json'


def read_sgn_line_value(node):
    kind = node.attrs['kind']
    if kind == 'dataframe':
        columns = json.loads(node.attrs['columns'])
        out = pd.DataFrame({key: read_sgn_line_value(node[key]) for key in columns})
        out.index = read_sgn_line_value(node['_index']); return out
    if kind == 'dict': return {unquote(key): read_sgn_line_value(node[key]) for key in node}
    if kind == 'json': return json.loads(node.asstr()[()])
    return node.asstr()[()] if node.dtype.kind in 'OUS' else node[()]




"""Strict vector PDF export for the Fig4 B/C/D Matplotlib canvas.

No analysis is performed. Each native Matplotlib label becomes one live Arial
text object, including the TF italic run in the two supported mathtext titles.
All geometry is written directly, without XObjects, glyph outlines or images.
The caller declares at most four data axes; their clip rectangles are collected
into one sibling viewport group per axis. Other artists must fit their axes.
"""
from pathlib import Path
import re
import sys
import math
import numpy as np
from matplotlib.backend_bases import RendererBase
from matplotlib.path import Path as MplPath
try:
    from reportlab.pdfgen import canvas
except ImportError:
    # Set this storage path independently before running.
    sys.path.append(str(ROOT))
    from reportlab.pdfgen import canvas
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont
_SGN_LINE_FONTS = {'normal': 'FigArial', 'italic': 'FigArialItalic', 'bold': 'FigArialBold', 'bolditalic': 'FigArialBoldItalic'}

def _sgn_line_register_fonts():
    for key, file in [('normal', 'arial.ttf'), ('italic', 'ariali.ttf'), ('bold', 'arialbd.ttf'), ('bolditalic', 'arialbi.ttf')]:
        if _SGN_LINE_FONTS[key] not in pdfmetrics.getRegisteredFontNames():
            # Set this storage path independently before running.
            pdfmetrics.registerFont(TTFont(_SGN_LINE_FONTS[key], str(ROOT / file)))

def _sgn_line_font(prop):
    bold = prop.get_weight() in ('bold', 'heavy', 'black', 'semibold', 'demibold') or (isinstance(prop.get_weight(), (int, float)) and prop.get_weight() >= 600)
    italic = prop.get_style() in ('italic', 'oblique')
    return _SGN_LINE_FONTS['bolditalic' if bold and italic else 'bold' if bold else 'italic' if italic else 'normal']

def _sgn_line_runs(text, prop, ismath):
    if not ismath:
        if '$' in text:
            raise ValueError('Unexpected dollar in plain label: ' + text)
        return [(text, _sgn_line_font(prop))]
    found = re.fullmatch('\\$\\\\it\\{([^{}]+)\\}([^$]*)\\$(.*)', text)
    if found is None:
        raise ValueError('Unsupported math label: ' + text)
    tf, inside, after = found.groups()
    return [(tf, _SGN_LINE_FONTS['italic']), (inside + after, _SGN_LINE_FONTS['normal'])]

class _SGNLineRecorder(RendererBase):

    def __init__(self, fig, agg, viewports, halo_labels):
        super().__init__()
        self.dpi = fig.dpi
        self.unit = 72 / fig.dpi
        self.agg = agg
        self.width, self.height = fig.bbox.size
        self.ops = []
        self.viewports = viewports
        self.halo_labels = halo_labels

    def points_to_pixels(self, p):
        return p / self.unit

    def flipy(self):
        return False

    def get_canvas_width_height(self):
        return (self.width, self.height)

    def get_text_width_height_descent(self, s, prop, ismath):
        runs = _sgn_line_runs(s, prop, ismath)
        width = sum((pdfmetrics.stringWidth(t, f, prop.get_size_in_points()) for t, f in runs)) / self.unit
        _, h, d = self.agg.get_text_width_height_descent(re.sub('\\$|\\\\it\\{|\\}', '', s), prop, False)
        return (width, h, d)

    def _clip(self, gc):
        if gc.get_clip_path()[0] is not None:
            raise ValueError('Nonrectangular artist clip is unsupported')
        b = gc.get_clip_rectangle()
        if b is None:
            return None
        for i, box in enumerate(self.viewports):
            if np.allclose(b.bounds, box, rtol=0, atol=1e-05):
                return i
        return None

    def draw_path(self, gc, path, transform, rgbFace=None):
        if gc.get_hatch():
            raise ValueError('Hatching/Pattern is forbidden')
        p = transform.transform_path(path)
        self.ops.append(('path', self._clip(gc), p, gc.get_rgb(), rgbFace, gc.get_linewidth(), gc.get_alpha(), gc.get_forced_alpha(), gc.get_dashes(), gc.get_capstyle(), gc.get_joinstyle()))

    def draw_text(self, gc, x, y, s, prop, angle, ismath=False, mtext=None):
        self.ops.append(('text', None, float(x), float(y), str(s), prop.copy(), float(angle), ismath, gc.get_rgb(), gc.get_alpha(), id(mtext) in self.halo_labels))

    def draw_image(self, *args, **kwargs):
        raise ValueError('Raster image is forbidden')

    def start_rasterizing(self):
        pass

    def stop_rasterizing(self):
        pass

    def start_filter(self):
        raise ValueError('Raster filter is forbidden')

    def stop_filter(self, *args):
        raise ValueError('Raster filter is forbidden')

def _sgn_line_pdf_path(pdf, path, unit):
    out = pdf.beginPath()
    current = None
    start = None
    for vertices, code in path.iter_segments(remove_nans=True, simplify=False, curves=True):
        v = np.asarray(vertices).reshape(-1, 2) * unit
        if code == MplPath.MOVETO:
            current = v[0]
            start = current
            out.moveTo(*current)
        elif code == MplPath.LINETO:
            current = v[0]
            out.lineTo(*current)
        elif code == MplPath.CURVE3:
            control, end = v
            a = current + (control - current) * 2 / 3
            b = end + (control - end) * 2 / 3
            out.curveTo(*a, *b, *end)
            current = end
        elif code == MplPath.CURVE4:
            out.curveTo(*v.ravel())
            current = v[-1]
        elif code == MplPath.CLOSEPOLY:
            out.close()
            current = start
        elif code == MplPath.STOP:
            break
        else:
            raise ValueError('Unknown path opcode ' + str(code))
    return out

def sgn_line_editable_pdf(fig, destination, *, data_axes=(), live_halo_width=0.2):
    """Write a strict vector PDF. Return declared clip boxes in PDF points.

    The artist canvas remains unchanged after return. Rasterization flags and
    white text path effects are replaced by a narrow live text fill+stroke
    (render mode 2) in the same single label object; no outlined font paths are
    emitted. The width is narrower than Agg's halo to preserve the black stems.
    """
    _sgn_line_register_fonts()
    if len(data_axes) > 4:
        raise ValueError('At most four sibling data viewports')
    fig.canvas.draw()
    agg = fig.canvas.get_renderer()
    viewports = [tuple(ax.bbox.bounds) for ax in data_axes]
    effects = []
    for artist in fig.findobj():
        if hasattr(artist, 'get_path_effects'):
            old = artist.get_path_effects()
            if old:
                effects.append((artist, old))
                artist.set_path_effects([])
    rec = _SGNLineRecorder(fig, agg, viewports, {id(a) for a, e in effects})
    try:
        fig.draw(rec)
    finally:
        for a, e in effects:
            a.set_path_effects(e)
    dst = Path(destination)
    dst.parent.mkdir(parents=True, exist_ok=True)
    pdf = canvas.Canvas(str(dst), pagesize=tuple(fig.get_size_inches() * 72), pageCompression=1, invariant=1, initialFontName=_SGN_LINE_FONTS['normal'])
    unit = rec.unit

    def emit(op):
        if op[0] == 'text':
            _, _, x, y, s, prop, angle, ismath, color, alpha, halo = op
            if not s:
                return
            pdf.saveState()
            pdf.setFillColorRGB(*color[:3])
            pdf.setFillAlpha(alpha)
            text = pdf.beginText()
            rad = math.radians(angle)
            if halo and live_halo_width > 0:
                pdf.setStrokeColorRGB(1, 1, 1)
                pdf.setStrokeAlpha(alpha)
                pdf.setLineWidth(float(live_halo_width))
                text.setTextRenderMode(2)
            text.setTextTransform(math.cos(rad), math.sin(rad), -math.sin(rad), math.cos(rad), x * unit, y * unit)
            for value, font in _sgn_line_runs(s, prop, ismath):
                if value:
                    text.setFont(font, prop.get_size_in_points())
                    text.textOut(value)
            pdf.drawText(text)
            pdf.restoreState()
            return
        _, _, path, stroke, face, lw, alpha, forced, dashes, cap, join = op
        pdf.saveState()
        pdf.setLineWidth(lw)
        pdf.setLineCap({'butt': 0, 'round': 1, 'projecting': 2}[cap])
        pdf.setLineJoin({'miter': 0, 'round': 1, 'bevel': 2}[join])
        offset, pattern = dashes
        pdf.setDash([] if pattern is None else [float(v) for v in pattern], 0.0 if offset is None else float(offset))
        if stroke is not None:
            pdf.setStrokeColorRGB(*stroke[:3])
            pdf.setStrokeAlpha(alpha if forced else stroke[3] if len(stroke) > 3 else alpha)
        if face is not None:
            pdf.setFillColorRGB(*face[:3])
            pdf.setFillAlpha(alpha if forced else face[3] if len(face) > 3 else alpha)
        pdf.drawPath(_sgn_line_pdf_path(pdf, path, unit), stroke=int(stroke is not None and lw > 0), fill=int(face is not None), fillMode=1)
        pdf.restoreState()
    done = set()
    for op in rec.ops:
        key = op[1]
        if key is None:
            emit(op)
            continue
        if key in done:
            continue
        done.add(key)
        pdf.saveState()
        p = pdf.beginPath()
        p.rect(*np.asarray(viewports[key]) * unit)
        pdf.clipPath(p, stroke=0, fill=0, fillMode=1)
        for candidate in rec.ops:
            if candidate[1] == key:
                emit(candidate)
        pdf.restoreState()
    pdf.showPage()
    pdf.save()
    if done != set(range(len(data_axes))):
        raise AssertionError('Declared clip has no artist')
    return {'expected_clips': len(done), 'clip_boxes_pt': (np.asarray(viewports) * unit).tolist(), 'editable_label_objects': sum((op[0] == 'text' for op in rec.ops)), 'label_strings': [op[4] for op in rec.ops if op[0] == 'text'], 'vector_paths': sum((op[0] == 'path' for op in rec.ops)), 'text_effects': 'White live text fill+stroke, single object; stroke width ' + str(live_halo_width) + ' pt.', 'halo_labels': len(effects)}

from PIL import Image as PILImage


def save_sgn_line_figure(fig, axes):
    fig.canvas.draw();renderer=fig.canvas.get_renderer()
    clipped=[]
    for text in fig.findobj(matplotlib.text.Text):
        if text.get_visible() and text.get_text():
            box=text.get_window_extent(renderer)
            if not fig.bbox.contains(*box.p0) or not fig.bbox.contains(*box.p1):clipped.append(text.get_text())
    assert not clipped,clipped
    # No data clip is needed: both centers and marker radii fit the declared axes.
    for ax in axes:
        for artist in ax.collections:
            xy=ax.transData.transform(np.asarray(artist.get_offsets(),float))
            radius=np.sqrt(np.max(artist.get_sizes()))/2*fig.dpi/72
            assert (xy[:,0]-radius >= ax.bbox.x0).all() and (xy[:,0]+radius <= ax.bbox.x1).all()
            assert (xy[:,1]-radius >= ax.bbox.y0).all() and (xy[:,1]+radius <= ax.bbox.y1).all()
        for artist in ax.lines:
            xy=ax.transData.transform(np.column_stack((artist.get_xdata(),artist.get_ydata())))
            radius=artist.get_markersize()/2*fig.dpi/72
            assert (xy[:,0]-radius >= ax.bbox.x0).all() and (xy[:,0]+radius <= ax.bbox.x1).all()
            assert (xy[:,1]-radius >= ax.bbox.y0).all() and (xy[:,1]+radius <= ax.bbox.y1).all()
    # Set this storage path independently before running. SGN-lines.png.
    png=ROOT;pdf=png.with_suffix('.pdf')
    fig.savefig(png,dpi=SGN_LINE_DPI,facecolor='white')
    qa=sgn_line_editable_pdf(fig,pdf,data_axes=())
    with PILImage.open(png) as im:
        assert im.size == (9120,3840)
        assert all(abs(x-SGN_LINE_DPI)<.1 for x in im.info['dpi'])
        qa.update(pixels=list(im.size),dpi=list(im.info['dpi']))
    qa.update(clipped_labels=clipped,legend_labels=['CVG','VGN','SGN'],
        colored_library_points_retained=True,figsize_in=list(SGN_LINE_FIGSIZE),
        library_alpha=.25,library_area=10.,
        palette={'CVG':'#8D77A1','VGN':'#5B85A7','SGN':'#69A18B'})
    plt.close(fig)
    return qa


def render_sgn_lines(data):
    FIGSIZE = SGN_LINE_FIGSIZE
    X_LIMITS, Y_LIMITS = (9.1, 18.9), (-.10, 4.10)
    X_TICKS, Y_TICKS = (9.5, 11.5, 13.5, 16.5, 18.5), (0, 1, 2, 3, 4)
    MEAN_LINE_WIDTH, MEAN_MARKER_SIZE = 1.65, 3.8
    LIBRARY_AREA, LIBRARY_ALPHA = 10.0, .25
    ORDER = ('CVG', 'VGN', 'SGN')
    COLORS = {'CVG': '#8D77A1', 'VGN': '#5B85A7', 'SGN': '#69A18B'}
    units, points = data['line_units'], data['line_points']
    plt.rcParams.update({'font.family': 'Arial', 'font.size': 9,
        'text.color': '#30343A', 'axes.labelcolor': '#30343A',
        'xtick.color': '#50555B', 'ytick.color': '#50555B', 'savefig.facecolor': 'white'})
    fig = plt.figure(figsize=FIGSIZE)
    grid = fig.add_gridspec(2, 4, left=.058, right=.985, bottom=.13, top=.94,
                           hspace=.53, wspace=.25)
    axes = []
    for j, gene in enumerate(SGN_LINE_GENES):
        row, col = (0, j) if j < 3 else (1, j - 3)
        ax = fig.add_subplot(grid[row, col])
        axes.append(ax)
        for identity in ORDER:
            u = units.loc[units.gene.eq(gene) & units.display_type.eq(identity)]
            p = points.loc[points.gene.eq(gene) & points.display_type.eq(identity)].sort_values('age')
            ax.scatter(u.age, u.expr, s=LIBRARY_AREA, color=COLORS[identity],
                       alpha=LIBRARY_ALPHA, linewidths=0, zorder=2)
            line, = ax.plot(p.age, p.expr, color=COLORS[identity], lw=MEAN_LINE_WIDTH,
                           marker='o', ms=MEAN_MARKER_SIZE, markeredgewidth=0, zorder=3)
            np.testing.assert_array_equal(line.get_ydata(), p.expr)
        ax.set_xlim(X_LIMITS); ax.set_ylim(Y_LIMITS)
        ax.set_xticks(X_TICKS, [f'E{x:g}' for x in X_TICKS])
        ax.xaxis.set_minor_locator(FixedLocator(np.arange(9.5, 19)))
        ax.set_yticks(Y_TICKS)
        if col: ax.tick_params(labelleft=False)
        ax.tick_params(which='major', length=2.6, width=.6, labelsize=8.5, pad=3)
        ax.tick_params(axis='x', which='minor', length=1.5, width=.45, color='#A5AAB0')
        ax.grid(axis='y', color='#EBECEE', linewidth=.55, zorder=0)
        for side in ('top', 'right'): ax.spines[side].set_visible(False)
        for side in ('left', 'bottom'):
            ax.spines[side].set_color('#A3A7AD'); ax.spines[side].set_linewidth(.65)
        ax.set_title(gene, fontstyle='italic', fontsize=12, pad=8)
    legend_ax = fig.add_subplot(grid[0, 3]); legend_ax.axis('off')
    handles = [Line2D([0], [0], color=COLORS[t], lw=MEAN_LINE_WIDTH, marker='o',
                      ms=4, markeredgewidth=0, label=t) for t in ORDER]
    line_legend = legend_ax.legend(handles=handles, loc='center', frameon=False, fontsize=9.5,
                     handlelength=2.2, labelspacing=1.0)
    fig.supylabel('RNA: mean log10(1 + pseudobulk CPM)', x=.012, y=.55, fontsize=10)
    fig.supxlabel('Embryonic age', x=.53, y=.035, fontsize=10)
    fig.canvas.draw(); renderer = fig.canvas.get_renderer()
    for ax in axes:
        boxes = [t.get_window_extent(renderer) for t in ax.get_xticklabels() if t.get_visible()]
        assert not any(a.overlaps(b) for i, a in enumerate(boxes) for b in boxes[i + 1:])
    assert [t.get_text() for t in line_legend.get_texts()] == list(ORDER)
    return save_sgn_line_figure(fig,axes)


# Frozen inputs and prior outputs are never overwritten by this appended block.
sgn_line_protected=[SGN_LINE_RESULT.parent/name for name in ('Activity.h5','Fig4B-D.h5')]
sgn_line_protected += [SGN_LINE_FIGURE_DIR/f'{tf}.{ext}' for tf in ('Foxo6','Zfpm2','Neurod2','Nhlh1','Neurod1') for ext in ('png','pdf')]
# Set this storage path independently before running.
sgn_line_protected += [ROOT/f'VGN-lines.{ext}' for ext in ('png','pdf')]
sgn_line_before={str(p.relative_to(SGN_LINE_ROOT)):_sgn_line_digest(p) for p in sgn_line_protected}
sgn_line_data=None
if SGN_LINE_RESULT.exists() and not REBUILD_SGN_LINES:
    with h5py.File(SGN_LINE_RESULT,'r') as h:
        if h.attrs.get('schema')=='SGN-seven-gene-lines-v1' and h['provenance/data_source_sha256'].asstr()[()]==SGN_LINE_DATA_SOURCE_SHA256:
            candidate=read_sgn_line_value(h['data'])
            if tuple(candidate['genes'])==SGN_LINE_GENES:
                present_hashes={p:_sgn_line_digest(SGN_LINE_ROOT/p) for p in candidate['source_hashes'] if (SGN_LINE_ROOT/p).exists()}
                if all(candidate['source_hashes'][p]==v for p,v in present_hashes.items()):sgn_line_data=candidate
if sgn_line_data is None:sgn_line_data=prepare_sgn_lines(SGN_LINE_ROOT,SGN_LINE_GENES)
np.testing.assert_array_equal(sgn_line_data['genes'],SGN_LINE_GENES)
with plt.rc_context():
    sgn_line_qa=render_sgn_lines(sgn_line_data)
# Set this storage path independently before running. validate_figure_output.py. Use the supplied fig4_validate_figure_output.py.
sgn_line_validator=ROOT
# Set this storage path independently before running.
sgn_line_dependencies=ROOT
# Set this storage path independently before running. python.exe.
sgn_line_check=subprocess.run([str(ROOT),str(sgn_line_validator),
    # Set this storage path independently before running. SGN-lines.png.
    str(ROOT),str(ROOT),
    '--expected-size-in','15.2x6.4','--expected-clips','0',
    # Set this storage path independently before running. pdftoppm.exe.
    '--pdftoppm',str(ROOT),
    # Set this storage path independently before running.
    '--render-dir',str(ROOT),'--render-dpi','150'],capture_output=True,text=True,errors='replace')
if sgn_line_check.returncode:raise RuntimeError(sgn_line_check.stdout+sgn_line_check.stderr)
sgn_line_qa.update(status='PASS',validator_stdout=sgn_line_check.stdout)
for p,old_hash in sgn_line_before.items():assert _sgn_line_digest(SGN_LINE_ROOT/p)==old_hash,p
sgn_line_notebook=nbformat.read(SGN_LINE_NOTEBOOK,as_version=4)
sgn_line_source=next(c.source for c in sgn_line_notebook.cells if c.cell_type=='code' and c.source.startswith('#11. =========== SGN selected-gene age curves'))
sgn_line_code_sha=hashlib.sha256(sgn_line_source.encode()).hexdigest()
# Set this storage path independently before running. SGN-lines.pending.h5.
sgn_line_pending=ROOT
with h5py.File(sgn_line_pending,'w') as h:
    h.attrs['schema']='SGN-seven-gene-lines-v1'
    write_sgn_line_value(h,'data',sgn_line_data)
    write_sgn_line_value(h,'figure',sgn_line_qa)
    p=h.create_group('provenance')
    p.create_dataset('data_source_sha256',data=SGN_LINE_DATA_SOURCE_SHA256,dtype=h5py.string_dtype())
    p.create_dataset('append_cell_source_sha256',data=sgn_line_code_sha,dtype=h5py.string_dtype())
    p.create_dataset('executed_append_cell_source',data=sgn_line_source,dtype=h5py.string_dtype())
    p.create_dataset('notebook_source_sha256',data=hashlib.sha256('\n\n'.join(c.source for c in sgn_line_notebook.cells if c.cell_type=='code').encode()).hexdigest(),dtype=h5py.string_dtype())
    write_sgn_line_value(p,'prior_outputs_sha256_unchanged',sgn_line_before)
    write_sgn_line_value(p,'output_sha256',{ext:_sgn_line_digest(SGN_LINE_FIGURE_DIR/f'SGN-lines.{ext}') for ext in ('png','pdf')})
    write_sgn_line_value(p,'runtime',{'python':sys.version,'platform':platform.platform(),'notebook_scope':'Only this new appended cell executed; prior B/C/D cells and artifacts retained'})
sgn_line_pending.replace(SGN_LINE_RESULT)
assert sorted(p.name for p in _release_cleanup_target(SGN_LINE_RESULT.parent).iterdir() if p.is_file())==['Activity.h5','Fig4B-D.h5','SGN-lines.h5']
print('SGN lines: 7 genes, 338 library values, 91 curve points; PNG/PDF PASS; prior outputs unchanged.')
# Set this storage path independently before running. SGN-lines.png.
display(Image(filename=str(ROOT),width=1400))
sgn_line_removed=0
for p in sorted(_release_cleanup_target(SGN_LINE_WORK).rglob('*'),key=lambda p:len(p.parts),reverse=True):
    assert p.resolve().is_relative_to(SGN_LINE_WORK.resolve())
    if p.is_file():sgn_line_removed+=p.stat().st_size;_release_cleanup_target(p).unlink()
    elif p.is_dir() and not any(_release_cleanup_target(p).iterdir()):_release_cleanup_target(p).rmdir()
if SGN_LINE_WORK.exists() and not any(_release_cleanup_target(SGN_LINE_WORK).iterdir()):_release_cleanup_target(SGN_LINE_WORK).rmdir()
print('Consumed runtime bytes removed:',sgn_line_removed)
```

### SCENIC B-only retry

This appended block preserves the existing B/C/D and age-line analyses. It reports all prespecified seed/database attempts and either selects the prespecified early seed42 canonical signature when positive-regulon presence repeats in at least two seeds (membership overlaps are reported separately) or explicitly uses the original CVG/VGN signature as a transfer score. AUCell is scored on the measured early/SS2 20871-gene intersection; UMI and E10.5 remain unscored. Missing activity never substitutes TF RNA. The complete retry is stored only under `Activity.h5/B_retry`. Default redraw verifies source, input and parameter hashes; review changes before enabling `RECOMPUTE_B_RETRY=True`.


```python

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")

# Cleanup must never interpret the project/current directory as a temporary target.
def _release_cleanup_target(path):
    candidate = Path(path).resolve()
    current = Path.cwd().resolve()
    if candidate == current or candidate in current.parents or candidate == Path(candidate.anchor):
        raise RuntimeError("Configure the owned temporary cleanup path before running this section.")
    return path

#12. =========== Repeated SCENIC evidence and two-source B-only activity
B_RETRY_SOURCES = {'resources.py': '"""Fetch only official mouse v9 gene-ranking resources for this SCENIC retry."""\n\n# Configure each file/directory placeholder independently; no storage addresses are shipped.\nfrom pathlib import Path\nROOT = Path(".")\nfrom pathlib import Path\nfrom html.parser import HTMLParser\nfrom urllib.parse import urljoin, urlparse\nimport argparse\nimport hashlib\nimport json\nimport time\nimport urllib.request\nfrom concurrent.futures import ThreadPoolExecutor, as_completed\nimport re\nimport sys\n\n# Set this storage path independently before running.\nROOT = ROOT\n# Set this storage path independently before running.\nOUT = ROOT\nBASE = \'https://resources.aertslab.org/cistarget/databases/mus_musculus/mm10/refseq_r80/mc9nr/\'\n\nclass Links(HTMLParser):\n    def __init__(self):\n        super().__init__()\n        self.links = []\n    def handle_starttag(self, tag, attrs):\n        if tag == \'a\':\n            href = dict(attrs).get(\'href\', \'\')\n            if href and not href.startswith((\'?\', \'#\')):\n                self.links.append(href)\n\ndef fetch(url):\n    with urllib.request.urlopen(url, timeout=60) as stream:\n        return stream.read()\n\ndef inventory():\n    pending, seen, files = [(BASE, 0)], set(), []\n    while pending:\n        url, depth = pending.pop(0)\n        if url in seen:\n            continue\n        seen.add(url)\n        parser = Links()\n        parser.feed(fetch(url).decode())\n        for href in parser.links:\n            target = urljoin(url, href)\n            if not target.startswith(BASE):\n                continue\n            path = urlparse(target).path\n            if path.endswith(\'/\') and depth < 3:\n                pending.append((target, depth + 1))\n            elif path.endswith((\'.feather\', \'.sha256sum\', \'.sha1sum\')):\n                files.append(target)\n    return sorted(set(files))\n\ndef download(url):\n    OUT.mkdir(parents=True, exist_ok=True)\n    path = OUT / urlparse(url).path.rsplit(\'/\', 1)[-1]\n    request = urllib.request.Request(url, headers={\'Range\':\'bytes=0-0\'})\n    with urllib.request.urlopen(request, timeout=90) as stream:\n        matched = re.fullmatch(r\'bytes 0-0/(\\d+)\', stream.headers.get(\'Content-Range\',\'\'))\n        if stream.status != 206 or matched is None:\n            raise RuntimeError(\'Official server did not honor byte-range request\')\n        total = int(matched.group(1))\n        assert len(stream.read()) == 1\n    start = path.stat().st_size if path.exists() else 0\n    assert start <= total\n    print(\'DOWNLOAD\', path.name, \'bytes\', total, \'existing\', start, flush=True)\n    parts = OUT / (path.name + \'.parts\')\n    parts.mkdir(exist_ok=True)\n    chunk_size = 16 * 1024 * 1024\n    spans = [(a,min(a+chunk_size,total)-1) for a in range(start,total,chunk_size)]\n    def part(span):\n        a,b=span\n        dst=parts / (str(a)+\'.part\')\n        for attempt in range(10):\n            have=dst.stat().st_size if dst.exists() else 0\n            if have==b-a+1:return dst\n            assert have<b-a+1\n            current=a+have\n            try:\n                req=urllib.request.Request(url,headers={\'Range\':f\'bytes={current}-{b}\'})\n                with urllib.request.urlopen(req,timeout=90) as stream:\n                    assert stream.status==206 and stream.headers.get(\'Content-Range\')==f\'bytes {current}-{b}/{total}\', repr(dict(stream.headers))\n                    with dst.open(\'ab\') as out:\n                        for block in iter(lambda:stream.read(65536),b\'\'):\n                            out.write(block)\n                assert dst.stat().st_size==b-a+1\n                return dst\n            except Exception as error:\n                print(\'RETRY\',a,attempt+1,type(error).__name__,str(error)[:200],flush=True)\n                if attempt==9:raise\n    done,last=start,time.monotonic()\n    with ThreadPoolExecutor(max_workers=32) as pool:\n        futures={pool.submit(part,span):span for span in spans}\n        for future in as_completed(futures):\n            dst=future.result();done+=dst.stat().st_size\n            if time.monotonic()-last>=30:\n                print(\'PROGRESS\',path.name,done,\'/\',total,flush=True);last=time.monotonic()\n    with path.open(\'ab\') as out:\n        for a,b in spans:\n            dst=parts/(str(a)+\'.part\')\n            with dst.open(\'rb\') as stream:\n                for block in iter(lambda:stream.read(4*1024*1024),b\'\'):out.write(block)\n    assert path.stat().st_size==total\n    sha = hashlib.sha256()\n    with path.open(\'rb\') as stream:\n        for block in iter(lambda: stream.read(8 * 1024 * 1024), b\'\'):\n            sha.update(block)\n    return dict(url=url, name=path.name, bytes=path.stat().st_size, sha256=sha.hexdigest())\n\nif __name__ == \'__main__\':\n    parser = argparse.ArgumentParser()\n    parser.add_argument(\'--download\', action=\'store_true\')\n    parser.add_argument(\'--annotation-only\', action=\'store_true\')\n    args = parser.parse_args()\n    if args.annotation_only:\n        print(download(\'https://resources.aertslab.org/cistarget/motif2tf/motifs-v9-nr.mgi-m0.001-o0.0.tbl\'),flush=True)\n        sys.exit(0)\n    entries = inventory()\n    print(json.dumps(entries, indent=2), flush=True)\n    if args.download:\n        candidates = [x for x in entries if \'genes_vs_motifs.rankings.feather\' in x]\n        if not candidates:\n            candidates = [x for x in entries if x.endswith(\'.feather\')]\n        selected = []\n        for token in (\'10kb\',):\n            match = [x for x in candidates if token in x]\n            if match:\n                selected.append(sorted(match)[0])\n        if not selected:\n            raise RuntimeError(\'No official v9 gene-ranking resource identified\')\n        records = [download(x) for x in selected]\n        annotation = \'https://resources.aertslab.org/cistarget/motif2tf/motifs-v9-nr.mgi-m0.001-o0.0.tbl\'\n        records.append(download(annotation))\n        # Set this storage path independently before running. resources.json.\n        (ROOT).write_text(json.dumps(records, indent=2), encoding=\'utf-8\')\n        print(\'COMPLETE\', json.dumps(records), flush=True)\n', 'grn/GRN.py': '"""Frozen whole-early GRNBoost2 repeats, seed42 calibration then seeds43/44."""\n\n# Configure each file/directory placeholder independently; no storage addresses are shipped.\nfrom pathlib import Path\nROOT = Path(".")\n#1. =========== Shared configuration\nfrom pathlib import Path\nimport argparse, hashlib, importlib.metadata, inspect, json, os, sys, tempfile, time\nfor key in (\'OMP_NUM_THREADS\',\'OPENBLAS_NUM_THREADS\',\'MKL_NUM_THREADS\',\'NUMBA_NUM_THREADS\'):\n    os.environ[key]=\'1\'\nsys.dont_write_bytecode=True\nimport h5py\nimport numpy as np\nimport pandas as pd\nfrom scipy import sparse\nfrom joblib import Parallel, delayed\nfrom arboreto.core import EARLY_STOP_WINDOW_LENGTH, SGBM_KWARGS, infer_partial_network\nSEEDS=(43,44)\nJOBS=8\nCALIBRATION=(\'Neurod1\',\'Zfpm2\',\'Nhlh1\',\'Tubb3\',\'Sox2\',\'Gata3\')\nRESOURCE_NAMES=(\'allTFs_mm.txt\',\'mm10_10kbp_up_10kbp_down_full_tx_v10_clust.genes_vs_motifs.rankings.feather\',\'mm10_500bp_up_100bp_down_full_tx_v10_clust.genes_vs_motifs.rankings.feather\',\'motifs-v10nr_clust-nr.mgi-m0.001-o0.0.tbl\')\ndef digest(path):\n    v=hashlib.sha256()\n    with Path(path).open(\'rb\') as f:\n        for b in iter(lambda:f.read(8*1024*1024),b\'\'):v.update(b)\n    return v.hexdigest()\ndef putj(h,key,value):\n    if key in h:del h[key]\n    h.create_dataset(key,data=json.dumps(value,ensure_ascii=False),dtype=h5py.string_dtype())\ndef infer_target(index, expression, genes, tf_expression, tf_genes, seed):\n    return infer_partial_network(regressor_type="GBM", regressor_kwargs=SGBM_KWARGS,\n        tf_matrix=tf_expression, tf_matrix_gene_names=tf_genes,\n        target_gene_name=genes[index], target_gene_expression=expression[:, index],\n        include_meta=False, early_stop_window_length=EARLY_STOP_WINDOW_LENGTH, seed=seed)\n#2. =========== Frozen inputs and exact seed42 calibration\n\ndef run(root,out):\n    root,out=Path(root).resolve(),Path(out).resolve()\n    # Set this storage path independently before running.\n    assert out==ROOT\n    out.mkdir(parents=True,exist_ok=True)\n    # Set this storage path independently before running. GSE178931_axes.npz.\n    paths=[ROOT,ROOT,ROOT,ROOT]\n    with np.load(paths[0]) as z:\n        genes=z[\'genes\'][z[\'inference_keep\']].astype(str).tolist();cells=z[\'cells\'].astype(str)\n        keep=z[\'inference_keep\']\n    matrix=sparse.load_npz(paths[1])\n    np.testing.assert_array_equal(keep,np.asarray((matrix>0).sum(axis=0)).ravel()>=3)\n    x=np.ascontiguousarray(matrix[:,keep].toarray(),dtype=np.float32)\n    # Set this storage path independently before running. allTFs_mm.txt.\n    pool=set((ROOT).read_text().splitlines())\n    ti=[i for i,g in enumerate(genes) if g in pool];tfs=[genes[i] for i in ti]\n    tx=np.ascontiguousarray(x[:,ti])\n    assert x.shape==(1161,14901) and tx.shape==(1161,1401) and np.isfinite(x).all()\n    # Set this storage path independently before running.\n    meta=dict(cohort=\'GSE178931 accepted whole-early atlas\',stages=[\'E9.5\',\'E11.5\',\'E13.5\'],cells=1161,genes=14901,TFs=1401,expression=\'Original log1p(CP10k), native eligible denominator, float32\',minimum_detected_cells=3,all_TFs_compete_for_every_target=True,workers=JOBS,seeds=list(SEEDS),SGBM_KWARGS=SGBM_KWARGS,early_stop_window_length=EARLY_STOP_WINDOW_LENGTH,input_sha256={str(p.relative_to(root)):digest(p) for p in paths},resource_sha256={n:digest(ROOT/n) for n in RESOURCE_NAMES},versions={n:importlib.metadata.version(n) for n in [\'pyscenic\',\'arboreto\',\'ctxcore\',\'numpy\',\'pandas\',\'scikit-learn\',\'joblib\',\'h5py\']},python=sys.version,runtime=sys.executable)\n    meta[\'configuration_sha256\']=hashlib.sha256(json.dumps(meta,sort_keys=True).encode()).hexdigest()\n    meta[\'worker_source_sha256\']=digest(__file__)\n    with tempfile.TemporaryDirectory(prefix=\'runtime-\',dir=out) as work:\n        for key in (\'TMP\',\'TEMP\',\'NUMBA_CACHE_DIR\',\'MPLCONFIGDIR\',\'JOBLIB_TEMP_FOLDER\'):os.environ[key]=work\n        old=pd.read_csv(paths[2],sep=\'\\t\',float_precision=\'round_trip\')\n        rows=[];started=time.monotonic()\n        frames=Parallel(n_jobs=JOBS,max_nbytes=\'1M\',temp_folder=work,batch_size=1)(delayed(infer_target)(genes.index(g),x,genes,tx,tfs,42) for g in CALIBRATION)\n        for g,a in zip(CALIBRATION,frames):\n            b=old[old.target.eq(g)].sort_values(\'TF\').reset_index(drop=True);a=a.sort_values(\'TF\').reset_index(drop=True)\n            pd.testing.assert_frame_equal(a[[\'TF\',\'target\']],b[[\'TF\',\'target\']],check_dtype=False)\n            d=float(np.max(np.abs(a.importance.to_numpy()-b.importance.to_numpy())))\n            exact=np.array_equal(a.importance.to_numpy(),b.importance.to_numpy())\n            rows.append(dict(target=g,edges=len(a),TF_target_exact=True,weights_exact=bool(exact),max_abs_difference=d))\n            assert exact,(g,d)\n        calibration=dict(seed=42,passed=True,targets=rows,elapsed_seconds=time.monotonic()-started)\n        print(\'CALIBRATION \'+json.dumps(calibration),flush=True)\n        del old,frames\n        for seed in SEEDS:\n            output=out/f\'seed{seed}.h5\'\n            assert not output.exists(),\'Existing seed result is not overwritten\'\n            started=time.monotonic();pending=[];nedge=0;last=started\n            with h5py.File(output,\'w\') as h:\n                h.attrs.update(schema=\'whole_early_GRNBoost2_repeat_v1\',stage=\'running\',seed=seed,completed_targets=0)\n                for k,v in [(\'genes\',genes),(\'TFs\',tfs),(\'cell_ids\',cells)]:h.create_dataset(k,data=np.asarray(v,dtype=object),dtype=h5py.string_dtype())\n                for k,dtype in [(\'TF_index\',np.int32),(\'target_index\',np.int32),(\'importance\',np.float64)]:h.create_dataset(\'grn/\'+k,shape=(0,),maxshape=(None,),chunks=(65536,),dtype=dtype,compression=\'gzip\',shuffle=True)\n                putj(h,\'methods_json\',dict(meta,seed=seed,seed_configuration_sha256=hashlib.sha256((meta[\'configuration_sha256\']+str(seed)).encode()).hexdigest()))\n                putj(h,\'calibration_json\',calibration)\n                h.create_dataset(\'executed_source\',data=Path(__file__).read_text(encoding=\'utf-8-sig\'),dtype=h5py.string_dtype())\n            print(\'START \'+json.dumps(dict(seed=seed,targets=len(genes),workers=JOBS)),flush=True)\n            with Parallel(n_jobs=JOBS,max_nbytes=\'1M\',temp_folder=work,batch_size=1,return_as=\'generator\') as executor:\n                generated=executor(delayed(infer_target)(i,x,genes,tx,tfs,seed) for i in range(len(genes)))\n                for i,a in enumerate(generated):\n                    assert a.target.eq(genes[i]).all() and not a.TF.eq(a.target).any() and not a.TF.duplicated().any()\n                    assert np.isfinite(a.importance).all() and a.importance.gt(0).all()\n                    idx=pd.Index(tfs).get_indexer(a.TF);assert (idx>=0).all()\n                    pending.append((idx.astype(np.int32),np.full(len(a),i,np.int32),a.importance.to_numpy(np.float64)))\n                    if len(pending)==128 or i+1==len(genes):\n                        arr=[np.concatenate([p[k] for p in pending]) for k in range(3)]\n                        with h5py.File(output,\'a\') as h:\n                            for k,v in zip([\'TF_index\',\'target_index\',\'importance\'],arr):\n                                d=h[\'grn/\'+k];d.resize((nedge+len(v),));d[nedge:]=v\n                            nedge+=len(arr[0]);h.attrs[\'completed_targets\']=i+1;h.attrs[\'elapsed_seconds\']=time.monotonic()-started\n                        pending.clear()\n                        if time.monotonic()-last>=60 or i+1==len(genes):\n                            print(\'PROGRESS \'+json.dumps(dict(seed=seed,completed=i+1,total=len(genes),edges=nedge,seconds=round(time.monotonic()-started,2))),flush=True);last=time.monotonic()\n            with h5py.File(output,\'a\') as h:\n                arrays=[h[\'grn/\'+k][:] for k in [\'TF_index\',\'target_index\',\'importance\']]\n                order=pd.Series(arrays[2]).sort_values(ascending=False).index.to_numpy()\n                pair=arrays[0].astype(np.int64)*len(genes)+arrays[1]\n                assert len(np.unique(pair))==len(pair)\n                for k,a in zip([\'TF_index\',\'target_index\',\'importance\'],arrays):h[\'grn/\'+k][:]=a[order]\n                result=dict(seed=seed,targets_fitted=len(genes),edges=nedge,elapsed_seconds=time.monotonic()-started,workers=JOBS,duplicate_edges=0,self_edges=0,sort_order=\'importance descending, pandas quicksort\')\n                putj(h,\'result_json\',result);h.attrs[\'stage\']=\'complete\'\n            print(\'COMPLETE \'+json.dumps(result),flush=True)\n        for p in paths:assert digest(p)==meta[\'input_sha256\'][str(p.relative_to(root))]\n        # Set this storage path independently before running.\n        for n in RESOURCE_NAMES:assert digest(ROOT/n)==meta[\'resource_sha256\'][n]\n    print(\'ALL_COMPLETE\',flush=True)\nif __name__==\'__main__\':\n    parser=argparse.ArgumentParser();parser.add_argument(\'--root\',type=Path,required=True);parser.add_argument(\'--output-dir\',type=Path,required=True)\n    a=parser.parse_args();run(a.root,a.output_dir)\n', 'prune.py': '"""Seed and cisTarget-version sensitivity for two requested SCENIC factors."""\n\n# Configure each file/directory placeholder independently; no storage addresses are shipped.\nfrom pathlib import Path\nROOT = Path(".")\nfrom pathlib import Path\nimport argparse\nimport hashlib\nimport importlib.metadata\nimport json\nimport os\nimport sys\nimport time\nimport warnings\n\nsys.dont_write_bytecode = True\nfor key in (\'OMP_NUM_THREADS\', \'MKL_NUM_THREADS\', \'OPENBLAS_NUM_THREADS\', \'NUMBA_NUM_THREADS\'):\n    os.environ[key] = \'1\'\n\nimport h5py\nimport numpy as np\nimport pandas as pd\nfrom scipy import sparse\nimport dask\ndask.config.set({\'dataframe.query-planning\': False})\nfrom ctxcore.genesig import Regulon\nfrom ctxcore.rnkdb import FeatherRankingDatabase\nfrom pyscenic.utils import modules_from_adjacencies\nfrom pyscenic.prune import prune2df, df2regulons\n\n# Set this storage path independently before running.\nROOT = ROOT\n# Set this storage path independently before running.\nWORK = ROOT\n# Set this storage path independently before running. Activity.h5. Produced by the SCENIC workers in fig4_SGN_program_analysis.ipynb.\nACTIVITY = ROOT\n# Set this storage path independently before running. pruning.h5.\nOUTPUT = ROOT\nPARAMETERS = dict(rank_threshold=5000, auc_threshold=.05, nes_threshold=3.,\n                  min_genes=20, rho_threshold=.03, thresholds=[.75,.9],\n                  top_n_targets=[50], top_n_regulators=[5,10,50],\n                  keep_only_activating=True, rho_mask_dropouts=False)\n\ndef sha(path):\n    h = hashlib.sha256()\n    with Path(path).open(\'rb\') as stream:\n        for block in iter(lambda: stream.read(8 * 1024 * 1024), b\'\'):\n            h.update(block)\n    return h.hexdigest()\n\ndef plain(v):\n    if isinstance(v,dict): return {str(k):plain(x) for k,x in v.items()}\n    if isinstance(v,(list,tuple,set,frozenset,np.ndarray)): return [plain(x) for x in v]\n    if isinstance(v,np.generic): return v.item()\n    return v\n\ndef put(group, name, value):\n    group.create_dataset(name, data=np.frombuffer(json.dumps(plain(value)).encode(),dtype=np.uint8),\n                         compression=\'gzip\', shuffle=True)\n\ndef get(d): return json.loads(d[:].tobytes().decode())\n\ndef resources(version):\n    if version in (\'v10\',\'v10_10kb\'):\n        # Set this storage path independently before running.\n        folder = ROOT\n        files = sorted(folder.glob(\'*v10_clust.genes_vs_motifs.rankings.feather\'))\n        # Set this storage path independently before running.\n        annot = ROOT\n    elif version == \'v9\':\n        # Set this storage path independently before running.\n        folder = ROOT\n        files = sorted(folder.glob(\'*mc9nr.genes_vs_motifs.rankings.feather\'))\n        # Set this storage path independently before running.\n        annot = ROOT\n    else: raise ValueError(version)\n    if version in (\'v9\',\'v10_10kb\'): files=[p for p in files if \'10kb\' in p.name]\n    assert len(files) == (2 if version==\'v10\' else 1) and annot.is_file(), \'Matched annotation and declared cis windows are required\'\n    return files, annot\n\ndef load_modules(source, seed):\n    if seed == 42:\n        with h5py.File(ACTIVITY,\'r\') as f:\n            rows = get(f[source + \'/candidate_modules_json\'])\n            rows = [r for r in rows if r[\'TF\'] in (\'Neurod1\',\'Zfpm2\')]\n        modules = [Regulon(name=r[\'TF\'],gene2weight=r[\'gene2weight\'],gene2occurrence={},\n                          transcription_factor=r[\'TF\'],context=frozenset(r[\'context\'])) for r in rows]\n        return modules, dict(source=\'Frozen full-competition modules, original seed 42\',\n                             candidate_modules=len(modules), original_grn_reused=True,\n                             input_sha256=sha(ACTIVITY))\n    assert source == \'whole_early\'\n    # Set this storage path independently before running.\n    grnfile = ROOT / (\'seed\' + str(seed) + \'.h5\')\n    with h5py.File(grnfile,\'r\') as f:\n        assert f.attrs.get(\'stage\')==\'complete\', \'GRN training is incomplete\'\n        tfs, genes = f[\'TFs\'].asstr()[:], f[\'genes\'].asstr()[:]\n        adj = pd.DataFrame(dict(TF=tfs[f[\'grn/TF_index\'][:]], target=genes[f[\'grn/target_index\'][:]],\n                                importance=f[\'grn/importance\'][:]))\n    # Set this storage path independently before running. GSE178931_axes.npz.\n    axesfile=ROOT\n    with np.load(axesfile) as a:\n        use=a[\'inference_keep\']; actual=a[\'genes\'][use];cells=a[\'cells\']\n    assert np.array_equal(actual,genes)\n    # Set this storage path independently before running. GSE178931_expression.npz.\n    values=sparse.load_npz(ROOT)[:,use].toarray()\n    frame=pd.DataFrame(values,index=cells,columns=genes)\n    begun=time.monotonic()\n    all_modules=modules_from_adjacencies(adj,frame,thresholds=(.75,.9),top_n_targets=(50,),\n        top_n_regulators=(5,10,50),min_genes=20,rho_dichotomize=True,\n        keep_only_activating=True,rho_threshold=.03,rho_mask_dropouts=False)\n    modules=[m for m in all_modules if m.transcription_factor in (\'Neurod1\',\'Zfpm2\')]\n    return modules, dict(source=\'New complete GRNBoost2 competition\', seed=seed,\n        total_modules=len(all_modules), candidate_modules=len(modules), grn_edges=len(adj),\n        module_seconds=time.monotonic()-begun, grn_sha256=sha(grnfile))\n\ndef run(source,seed,version):\n    key=f\'{source}/seed{seed}/{version}\'\n    files,annot=resources(version)\n    hashes={p.name:sha(p) for p in [*files,annot]}\n    if seed == 42:\n        with h5py.File(ACTIVITY,\'r\') as f:\n            module_hash=hashlib.sha256(f[source+\'/candidate_modules_json\'][:].tobytes()).hexdigest()\n    else:\n        # Set this storage path independently before running.\n        module_hash=sha(ROOT/(\'seed\'+str(seed)+\'.h5\'))\n    guard=dict(source=source,seed=seed,version=version,parameters=PARAMETERS,\n               resources=hashes,full_competition_input_sha256=module_hash)\n    token=hashlib.sha256(json.dumps(guard,sort_keys=True).encode()).hexdigest()\n    if OUTPUT.exists():\n        with h5py.File(OUTPUT,\'r\') as f:\n            if key in f and f[key].attrs.get(\'complete\',False):\n                assert f[key].attrs[\'guard\']==token, \'Cached resource or parameter mismatch\'\n                print(\'REUSE\',key,flush=True);return\n    modules,meta=load_modules(source,seed)\n    annotations={tf:[] for tf in [\'Neurod1\',\'Zfpm2\',\'Fog2\']}\n    for chunk in pd.read_csv(annot,sep=\'\\t\',chunksize=100000,low_memory=False):\n        for tf in annotations:\n            annotations[tf].extend(chunk.loc[chunk.gene_name.eq(tf)].to_dict(\'records\'))\n    databases=[FeatherRankingDatabase(str(p),name=p.name) for p in files]\n    print(\'PRUNE\',key,\'modules\',len(modules),\'annotations\',\n          {k:len(v) for k,v in annotations.items()},flush=True)\n    begun=time.monotonic()\n    with warnings.catch_warnings():\n        warnings.filterwarnings(\'ignore\',message=\'Not prepending group keys\',category=FutureWarning)\n        motifs=prune2df(databases,modules,str(annot),rank_threshold=5000,auc_threshold=.05,\n            nes_threshold=3.,client_or_address=\'dask_multiprocessing\',num_workers=2)\n    regs=df2regulons(motifs,save_columns=[\'NES\',\'Annotation\',\'MotifSimilarityQvalue\',\'OrthologousIdentity\']) if not motifs.empty else []\n    evidence=[]\n    for (tf,motifid),row in motifs.iterrows():\n        r=dict(TF=tf,MotifID=motifid)\n        r.update({\'|\'.join(k) if isinstance(k,tuple) else str(k):plain(v) for k,v in row.items()})\n        evidence.append(r)\n    records=[]\n    # Set this storage path independently before running. GSE178931_axes.npz.\n    with np.load(ROOT) as z:early=set(z[\'genes\'])\n    # Set this storage path independently before running. GSE165502_axes.npz.\n    with np.load(ROOT) as z:ss2=set(z[\'genes\'])\n    # Set this storage path independently before running. GSE195500_axes.npz.\n    with np.load(ROOT) as z:umi=set(z[\'genes\'])\n    for r in regs:\n        weights=dict(r.gene2weight)\n        assert r.name.endswith(\'(+)\')\n        records.append(dict(name=r.name,TF=r.transcription_factor,gene2weight=weights,\n            targets=len(weights),context=sorted(r.context),nes=float(r.nes),annotation=str(r.annotation),\n            early_ss2_matched=len(set(weights)&early&ss2),\n            three_source_matched=len(set(weights)&early&ss2&umi),\n            early_ss2_coverage=len(set(weights)&early&ss2)/len(weights),\n            three_source_coverage=len(set(weights)&early&ss2&umi)/len(weights)))\n    summary=[dict(TF=tf,status=\'available\' if any(r[\'TF\']==tf for r in records) else \'no_motif_supported_positive_regulon\',\n                  annotation_rows=len(annotations[tf]),\n                  targets=next((r[\'targets\'] for r in records if r[\'TF\']==tf),0)) for tf in (\'Neurod1\',\'Zfpm2\')]\n    with h5py.File(OUTPUT,\'a\') as f:\n        if key in f:del f[key]\n        g=f.create_group(key);g.attrs[\'guard\']=token\n        put(g,\'methods\',guard);put(g,\'module_metadata\',meta)\n        put(g,\'modules\',[dict(TF=m.transcription_factor,context=sorted(m.context),gene2weight=dict(m.gene2weight)) for m in modules])\n        put(g,\'annotation_evidence\',annotations);put(g,\'motif_evidence\',evidence)\n        put(g,\'regulons\',records);put(g,\'summary\',summary)\n        g.attrs[\'elapsed_seconds\']=time.monotonic()-begun\n        g.attrs[\'complete\']=True\n    print(\'COMPLETE\',key,summary,\'seconds\',round(time.monotonic()-begun,1),flush=True)\n\nif __name__==\'__main__\':\n    parser=argparse.ArgumentParser()\n    parser.add_argument(\'--source\',choices=[\'whole_early\',\'CVG_VGN\'],default=\'whole_early\')\n    parser.add_argument(\'--seed\',type=int,default=42)\n    parser.add_argument(\'--version\',choices=[\'v9\',\'v10\',\'v10_10kb\'],required=True)\n    args=parser.parse_args()\n    run(args.source,args.seed,args.version)\n', 'render/B-only.py': '\n# Configure each file/directory placeholder independently; no storage addresses are shipped.\nfrom pathlib import Path\nROOT = Path(".")\n\n# Cleanup must never interpret the project/current directory as a temporary target.\ndef _release_cleanup_target(path):\n    candidate = Path(path).resolve()\n    current = Path.cwd().resolve()\n    if candidate == current or candidate in current.parents or candidate == Path(candidate.anchor):\n        raise RuntimeError("Configure the owned temporary cleanup path before running this section.")\n    return path\n\nB_WORKER_SOURCE = \'\\n# Configure each file/directory placeholder independently; no storage addresses are shipped.\\nfrom pathlib import Path\\nROOT = Path(".")\\nfrom pathlib import Path\\nimport os,sys,json,hashlib\\nfor k in (\\\'OMP_NUM_THREADS\\\',\\\'MKL_NUM_THREADS\\\',\\\'OPENBLAS_NUM_THREADS\\\',\\\'NUMBA_NUM_THREADS\\\'):os.environ[k]=\\\'1\\\'\\nimport h5py,numpy as np,pandas as pd\\nfrom scipy import sparse\\nfrom ctxcore.genesig import GeneSignature\\nfrom ctxcore.recovery import derive_rank_cutoff\\nfrom pyscenic.aucell import create_rankings,aucell4r\\n\\ndef score(config_path):\\n # Set this storage path independently before running.\\n c=json.loads(Path(config_path).read_text(encoding=\\\'utf-8\\\')); root=Path(c[\\\'root\\\']); native=ROOT\\n with h5py.File(c[\\\'shared\\\'],\\\'r\\\') as h:\\n  cells=h[\\\'shared/obs/_index\\\'].asstr()[:];source=h[\\\'shared/obs/study_id\\\'].asstr()[:]\\n # Set this storage path independently before running. Inputs.h5. Candidate input contract; supply the immutable input container (standalone producer not located).\\n with h5py.File(ROOT,\\\'r\\\') as h:\\n  np.testing.assert_array_equal(cells,h[\\\'cell_ids\\\'].asstr()[:])\\n studies=(\\\'GSE178931\\\',\\\'GSE165502\\\'); axes={}\\n for s in studies:\\n  with np.load(native/(s+\\\'_axes.npz\\\')) as z:axes[s]=(z[\\\'cells\\\'].astype(str),z[\\\'genes\\\'].astype(str))\\n  assert len(set(axes[s][0]))==len(axes[s][0]) and len(set(axes[s][1]))==len(axes[s][1])\\n  np.testing.assert_array_equal(np.sort(cells[source==s]),np.sort(axes[s][0]))\\n common=set(axes[studies[0]][1])&set(axes[studies[1]][1]);genes=np.asarray([g for g in axes[studies[0]][1] if g in common])\\n assert len(genes)==20871\\n scope=np.isin(source,studies);assert scope.sum()==2666\\n score_cells=cells[scope]; values=np.empty((scope.sum(),len(genes)),dtype=np.int32)\\n for s,(cs,gs) in axes.items():\\n  rows=pd.Index(score_cells).get_indexer(cs);cols=pd.Index(gs).get_indexer(genes)\\n  m=sparse.load_npz(native/(s+\\\'_counts.npz\\\')).tocsr();assert np.issubdtype(m.dtype,np.integer) and m.data.min()>=0\\n  values[rows]=m[:,cols].toarray();del m\\n positive=np.full(len(cells),-1,dtype=np.int32);positive[scope]=(values>0).sum(axis=1)\\n cutoff=derive_rank_cutoff(.05,len(genes));gate=scope&(positive>cutoff)\\n audit={};sigs=[];tfs=list(c[\\\'signatures\\\'])\\n for tf,w in c[\\\'signatures\\\'].items():\\n  w={} if w is None else {str(g):float(v) for g,v in w.items()}\\n  assert all(np.isfinite(v) and v>=0 for v in w.values())\\n  matched=sorted(set(w)&common);cov=len(matched)/len(w) if w else None\\n  valid=bool(w) and sum(w.values())>0 and cov>=.8\\n  audit[tf]=dict(status=\\\'available\\\' if valid else \\\'not_estimable\\\',full_gene2weight=w,full_count=len(w),matched=len(matched),coverage=cov,missing=sorted(set(w)-common),self_in_signature=tf in w)\\n  if valid:sigs.append(GeneSignature(name=tf,gene2weight=w))\\n raw=np.full((len(cells),len(tfs)),np.nan);rank_sha=None\\n if sigs:\\n  ranked=create_rankings(pd.DataFrame(values[gate[scope]],index=cells[gate],columns=genes),seed=42);del values\\n  digest=hashlib.sha256()\\n  for i in range(0,len(ranked),128):digest.update(np.asarray(ranked.iloc[i:i+128].values,dtype=\\\'<u4\\\').tobytes(order=\\\'C\\\'))\\n  rank_sha=digest.hexdigest();auc=aucell4r(ranked,sigs,auc_threshold=.05,noweights=False,normalize=False,num_workers=1).reindex(cells[gate])\\n  assert np.isfinite(auc.values).all() and (auc.values>=0).all()\\n  for j,tf in enumerate(tfs):\\n   if audit[tf][\\\'status\\\']==\\\'available\\\':raw[gate,j]=auc[tf].to_numpy()\\n assert np.isnan(raw[~scope]).all() and np.isnan(raw[~gate]).all()\\n dt=h5py.string_dtype(\\\'utf-8\\\')\\n with h5py.File(c[\\\'output\\\'],\\\'w\\\') as h:\\n  h.attrs[\\\'schema\\\']=\\\'two_source_native20871_B_v1\\\';h.attrs[\\\'cache_key\\\']=c[\\\'cache_key\\\']\\n  for k,v in dict(cell_ids=cells,source=source,genes=genes,TFs=np.asarray(tfs)).items():h.create_dataset(k,data=np.asarray(v,dtype=object),dtype=dt)\\n  for k,v in dict(raw_auc=raw,source_scope=scope,rank_gate=gate,positive_genes=positive).items():h.create_dataset(k,data=v,compression=\\\'gzip\\\')\\n  h.create_dataset(\\\'rank_cutoff\\\',data=int(cutoff))\\n  for k,v in dict(audit=audit,metadata=c.get(\\\'metadata\\\',{}),methods=dict(seed=42,auc_threshold=.05,weighted=True,normalize=False,axis_genes=len(genes),ranking_sha256=rank_sha,axis_rule=\\\'Native GSE178931/GSE165502 measured intersection, original GSE178931 order\\\',missing_gene_imputation=False,full_canonical_signature=True,self_targets_preserved=True,min_coverage=.8,comparison_limit=\\\'Different axis from the previous three-source 11736-gene score; absolute AUC not directly comparable\\\')).items():h.create_dataset(k+\\\'_json\\\',data=json.dumps(v),dtype=dt)\\n print(json.dumps(dict(axis=len(genes),scope=int(scope.sum()),eligible=int(gate.sum()),cutoff=int(cutoff),audit={tf:{k:v for k,v in a.items() if k!=\\\'full_gene2weight\\\'} for tf,a in audit.items()})),flush=True)\\nif __name__==\\\'__main__\\\':score(sys.argv[1])\\n\'\nB_GAM_SOURCE = "\\nsuppressPackageStartupMessages({library(mgcv); library(rhdf5)})\\na<-commandArgs(trailingOnly=TRUE)\\nmeta<-read.delim(a[1],check.names=FALSE,stringsAsFactors=FALSE)\\nmeta$good<-tolower(as.character(meta$good)) %in% c(\'true\',\'1\')\\nX<-h5read(a[2],\'raw_auc\'); if(ncol(X)==nrow(meta)) X<-t(X)\\nstopifnot(nrow(X)==nrow(meta)); P<-matrix(NA_real_,nrow(X),ncol(X))\\nbalanced<-function(i){\\n d<-meta[i,]; sources<-unique(d$study_id); w<-numeric(length(i))\\n for(s in sources){m<-which(d$study_id==s); libs<-unique(d$library_id[m])\\n  for(l in libs){z<-m[d$library_id[m]==l]; w[z]<-1/length(sources)/length(libs)/length(z)}}\\n w*length(i)/sum(w)\\n}\\nfor(j in seq_len(ncol(X))){for(s in unique(meta$seg)){\\n rows<-which(meta$seg==s & is.finite(X[,j])); train<-rows[meta$good[rows]]\\n if(!length(train)) next\\n k<-min(10,length(unique(meta$t[train]))-1)\\n if(length(train)<12 || k<3 || sd(X[train,j])<1e-10){\\n  P[rows,j]<-weighted.mean(X[train,j],balanced(train))\\n }else{\\n  d<-data.frame(value=X[train,j],t=meta$t[train],w=balanced(train))\\n  model<-gam(value~s(t,bs=\'ts\',k=k),data=d,weights=w,gamma=1.5,method=\'GCV.Cp\')\\n  P[rows,j]<-as.numeric(predict(model,newdata=data.frame(t=meta$t[rows])))\\n }\\n}}\\nh5createFile(a[3]);h5write(P,a[3],\'fitted\')\\nh5write(paste(R.version.string,as.character(packageVersion(\'mgcv\')),sep=\' | \'),a[3],\'R_version\')\\nh5closeAll()\\n"\nSTRICT_PDF_SOURCE = \'"""Strict vector PDF export for the Fig4 B/C/D Matplotlib canvas.\\n\\nNo analysis is performed. Each native Matplotlib label becomes one live Arial\\ntext object, including the TF italic run in the two supported mathtext titles.\\nAll geometry is written directly, without XObjects, glyph outlines or images.\\nThe caller declares at most four data axes; their clip rectangles are collected\\ninto one sibling viewport group per axis. Other artists must fit their axes.\\n"""\\n\\n# Configure each file/directory placeholder independently; no storage addresses are shipped.\\nfrom pathlib import Path\\nROOT = Path(".")\\nfrom pathlib import Path\\nimport re\\nimport sys\\nimport math\\nimport numpy as np\\nfrom matplotlib.backend_bases import RendererBase\\nfrom matplotlib.path import Path as MplPath\\n\\ntry:\\n    from reportlab.pdfgen import canvas\\nexcept ImportError:\\n    # Set this storage path independently before running.\\n    sys.path.append(str(ROOT))\\n    from reportlab.pdfgen import canvas\\nfrom reportlab.pdfbase import pdfmetrics\\nfrom reportlab.pdfbase.ttfonts import TTFont\\n\\n_FONTS = {\\\'normal\\\':\\\'FigArial\\\',\\\'italic\\\':\\\'FigArialItalic\\\',\\n          \\\'bold\\\':\\\'FigArialBold\\\',\\\'bolditalic\\\':\\\'FigArialBoldItalic\\\'}\\n\\ndef _register():\\n    for key, file in [(\\\'normal\\\',\\\'arial.ttf\\\'),(\\\'italic\\\',\\\'ariali.ttf\\\'),\\n                      (\\\'bold\\\',\\\'arialbd.ttf\\\'),(\\\'bolditalic\\\',\\\'arialbi.ttf\\\')]:\\n        if _FONTS[key] not in pdfmetrics.getRegisteredFontNames():\\n            # Set this storage path independently before running.\\n            pdfmetrics.registerFont(TTFont(_FONTS[key],str(ROOT/file)))\\n\\ndef _font(prop):\\n    bold=prop.get_weight() in (\\\'bold\\\',\\\'heavy\\\',\\\'black\\\',\\\'semibold\\\',\\\'demibold\\\') or isinstance(prop.get_weight(),(int,float)) and prop.get_weight()>=600\\n    italic=prop.get_style() in (\\\'italic\\\',\\\'oblique\\\')\\n    return _FONTS[\\\'bolditalic\\\' if bold and italic else \\\'bold\\\' if bold else \\\'italic\\\' if italic else \\\'normal\\\']\\n\\ndef _runs(text,prop,ismath):\\n    if not ismath:\\n        if \\\'$\\\' in text: raise ValueError(\\\'Unexpected dollar in plain label: \\\'+text)\\n        return [(text,_font(prop))]\\n    # Current Fig4 titles use only this explicitly audited mathtext form.\\n    found=re.fullmatch(r\\\'\\\\$\\\\\\\\it\\\\{([^{}]+)\\\\}([^$]*)\\\\$(.*)\\\',text)\\n    if found is None: raise ValueError(\\\'Unsupported math label: \\\'+text)\\n    tf,inside,after=found.groups()\\n    return [(tf,_FONTS[\\\'italic\\\']),(inside+after,_FONTS[\\\'normal\\\'])]\\n\\nclass _Recorder(RendererBase):\\n    def __init__(self,fig,agg,viewports,halo_labels):\\n        super().__init__()\\n        self.dpi=fig.dpi;self.unit=72/fig.dpi;self.agg=agg\\n        self.width,self.height=fig.bbox.size;self.ops=[];self.viewports=viewports;self.halo_labels=halo_labels\\n    def points_to_pixels(self,p): return p/self.unit\\n    def flipy(self):return False\\n    def get_canvas_width_height(self):return self.width,self.height\\n    def get_text_width_height_descent(self,s,prop,ismath):\\n        runs=_runs(s,prop,ismath)\\n        width=sum(pdfmetrics.stringWidth(t,f,prop.get_size_in_points()) for t,f in runs)/self.unit\\n        # Match the source Arial baseline/ascent geometry; width uses the same\\n        # embedded font as the PDF, including the semantic mixed italic title.\\n        _,h,d=self.agg.get_text_width_height_descent(re.sub(r\\\'\\\\$|\\\\\\\\it\\\\{|\\\\}\\\', \\\'\\\',s),prop,False)\\n        return width,h,d\\n    def _clip(self,gc):\\n        if gc.get_clip_path()[0] is not None:\\n            raise ValueError(\\\'Nonrectangular artist clip is unsupported\\\')\\n        b=gc.get_clip_rectangle()\\n        if b is None:return None\\n        for i,box in enumerate(self.viewports):\\n            if np.allclose(b.bounds,box,rtol=0,atol=1e-5):return i\\n        return None\\n    def draw_path(self,gc,path,transform,rgbFace=None):\\n        if gc.get_hatch():raise ValueError(\\\'Hatching/Pattern is forbidden\\\')\\n        p=transform.transform_path(path)\\n        self.ops.append((\\\'path\\\',self._clip(gc),p,gc.get_rgb(),rgbFace,\\n                         gc.get_linewidth(),gc.get_alpha(),gc.get_forced_alpha(),\\n                         gc.get_dashes(),gc.get_capstyle(),gc.get_joinstyle()))\\n    def draw_text(self,gc,x,y,s,prop,angle,ismath=False,mtext=None):\\n        self.ops.append((\\\'text\\\',None,float(x),float(y),str(s),prop.copy(),float(angle),\\n                         ismath,gc.get_rgb(),gc.get_alpha(),id(mtext) in self.halo_labels))\\n    def draw_image(self,*args,**kwargs):raise ValueError(\\\'Raster image is forbidden\\\')\\n    def start_rasterizing(self):pass\\n    def stop_rasterizing(self):pass\\n    def start_filter(self):raise ValueError(\\\'Raster filter is forbidden\\\')\\n    def stop_filter(self,*args):raise ValueError(\\\'Raster filter is forbidden\\\')\\n\\ndef _pdf_path(pdf,path,unit):\\n    out=pdf.beginPath();current=None;start=None\\n    for vertices,code in path.iter_segments(remove_nans=True,simplify=False,curves=True):\\n        v=np.asarray(vertices).reshape(-1,2)*unit\\n        if code==MplPath.MOVETO:\\n            current=v[0];start=current;out.moveTo(*current)\\n        elif code==MplPath.LINETO:\\n            current=v[0];out.lineTo(*current)\\n        elif code==MplPath.CURVE3:\\n            control,end=v\\n            a=current+(control-current)*2/3;b=end+(control-end)*2/3\\n            out.curveTo(*a,*b,*end);current=end\\n        elif code==MplPath.CURVE4:\\n            out.curveTo(*v.ravel());current=v[-1]\\n        elif code==MplPath.CLOSEPOLY:\\n            out.close();current=start\\n        elif code==MplPath.STOP:break\\n        else:raise ValueError(\\\'Unknown path opcode \\\'+str(code))\\n    return out\\n\\ndef strict_editable_pdf(fig,destination,*,data_axes=(),live_halo_width=.20):\\n    """Write a strict vector PDF. Return declared clip boxes in PDF points.\\n\\n    The artist canvas remains unchanged after return. Rasterization flags and\\n    white text path effects are replaced by a narrow live text fill+stroke\\n    (render mode 2) in the same single label object; no outlined font paths are\\n    emitted. The width is narrower than Agg\\\'s halo to preserve the black stems.\\n    """\\n    _register()\\n    if len(data_axes)>4:raise ValueError(\\\'At most four sibling data viewports\\\')\\n    fig.canvas.draw();agg=fig.canvas.get_renderer()\\n    viewports=[tuple(ax.bbox.bounds) for ax in data_axes]\\n    effects=[]\\n    for artist in fig.findobj():\\n        if hasattr(artist,\\\'get_path_effects\\\'):\\n            old=artist.get_path_effects()\\n            if old:effects.append((artist,old));artist.set_path_effects([])\\n    rec=_Recorder(fig,agg,viewports,{id(a) for a,e in effects})\\n    try:fig.draw(rec)\\n    finally:\\n        for a,e in effects:a.set_path_effects(e)\\n    dst=Path(destination);dst.parent.mkdir(parents=True,exist_ok=True)\\n    pdf=canvas.Canvas(str(dst),pagesize=tuple(fig.get_size_inches()*72),\\n                      pageCompression=1,invariant=1,initialFontName=_FONTS[\\\'normal\\\'])\\n    unit=rec.unit\\n    def emit(op):\\n        if op[0]==\\\'text\\\':\\n            _,_,x,y,s,prop,angle,ismath,color,alpha,halo=op\\n            if not s:return\\n            pdf.saveState();pdf.setFillColorRGB(*color[:3]);pdf.setFillAlpha(alpha)\\n            text=pdf.beginText();rad=math.radians(angle)\\n            if halo and live_halo_width>0:\\n                pdf.setStrokeColorRGB(1,1,1);pdf.setStrokeAlpha(alpha)\\n                pdf.setLineWidth(float(live_halo_width));text.setTextRenderMode(2)\\n            text.setTextTransform(math.cos(rad),math.sin(rad),-math.sin(rad),math.cos(rad),x*unit,y*unit)\\n            for value,font in _runs(s,prop,ismath):\\n                if value:text.setFont(font,prop.get_size_in_points());text.textOut(value)\\n            pdf.drawText(text);pdf.restoreState();return\\n        _,_,path,stroke,face,lw,alpha,forced,dashes,cap,join=op\\n        pdf.saveState();pdf.setLineWidth(lw)\\n        pdf.setLineCap({\\\'butt\\\':0,\\\'round\\\':1,\\\'projecting\\\':2}[cap])\\n        pdf.setLineJoin({\\\'miter\\\':0,\\\'round\\\':1,\\\'bevel\\\':2}[join])\\n        offset,pattern=dashes\\n        pdf.setDash([] if pattern is None else [float(v) for v in pattern],0.0 if offset is None else float(offset))\\n        if stroke is not None:\\n            pdf.setStrokeColorRGB(*stroke[:3]);pdf.setStrokeAlpha(alpha if forced else (stroke[3] if len(stroke)>3 else alpha))\\n        if face is not None:\\n            pdf.setFillColorRGB(*face[:3]);pdf.setFillAlpha(alpha if forced else (face[3] if len(face)>3 else alpha))\\n        pdf.drawPath(_pdf_path(pdf,path,unit),stroke=int(stroke is not None and lw>0),fill=int(face is not None),fillMode=1)\\n        pdf.restoreState()\\n    done=set()\\n    for op in rec.ops:\\n        key=op[1]\\n        if key is None:emit(op);continue\\n        if key in done:continue\\n        done.add(key);pdf.saveState()\\n        p=pdf.beginPath();p.rect(*(np.asarray(viewports[key])*unit))\\n        pdf.clipPath(p,stroke=0,fill=0,fillMode=1)\\n        for candidate in rec.ops:\\n            if candidate[1]==key:emit(candidate)\\n        pdf.restoreState()\\n    pdf.showPage();pdf.save()\\n    if done!=set(range(len(data_axes))):raise AssertionError(\\\'Declared clip has no artist\\\')\\n    return {\\\'expected_clips\\\':len(done),\\\'clip_boxes_pt\\\':(np.asarray(viewports)*unit).tolist(),\\n            \\\'editable_label_objects\\\':sum(op[0]==\\\'text\\\' for op in rec.ops),\\n            \\\'label_strings\\\':[op[4] for op in rec.ops if op[0]==\\\'text\\\'],\\n            \\\'vector_paths\\\':sum(op[0]==\\\'path\\\' for op in rec.ops),\\n            \\\'text_effects\\\':\\\'White live text fill+stroke, single object; stroke width \\\'+str(live_halo_width)+\\\' pt.\\\',\\n            \\\'halo_labels\\\':len(effects)}\\n\'\n\nfrom pathlib import Path\nimport os,sys,json,hashlib,subprocess\nimport numpy as np,pandas as pd,h5py\n\ndef _sha(path):\n d=hashlib.sha256()\n with Path(path).open(\'rb\') as h:\n  for b in iter(lambda:h.read(1048576),b\'\'):d.update(b)\n return d.hexdigest()\n\ndef _plain(v):\n if isinstance(v,dict):return {str(k):_plain(x) for k,x in v.items()}\n if isinstance(v,(tuple,list)):return [_plain(x) for x in v]\n if isinstance(v,np.ndarray):return v.tolist()\n if isinstance(v,np.generic):return v.item()\n if isinstance(v,Path):return str(v)\n return v\n\ndef _shared(root,path):\n with h5py.File(path,\'r\') as h:\n  g=h[\'shared/obs\'];cols=json.loads(g.attrs[\'columns\']);cells=g[\'_index\'].asstr()[:]\n  obs=pd.DataFrame({k:(g[k].asstr()[:] if g[k].dtype.kind in \'OSU\' else g[k][:]) for k in cols},index=cells)\n  xy=h[\'shared/X_umap\'][:];good=h[\'shared/good\'][:].astype(bool)\n # Set this storage path independently before running. Inputs.h5. Candidate input contract; supply the immutable input container (standalone producer not located).\n with h5py.File(ROOT,\'r\') as h:\n  np.testing.assert_array_equal(cells,h[\'cell_ids\'].asstr()[:]);np.testing.assert_array_equal(xy,h[\'X_umap\'][:])\n  np.testing.assert_array_equal(obs.study_id.to_numpy(),h[\'obs/study_id\'].asstr()[:])\n assert len(cells)==7355 and len(set(cells))==7355 and xy.shape==(7355,2)\n return cells,obs,xy,good\n\ndef _fit(raw_path,obs,good,folder,rscript,key):\n # Set this storage path independently before running. fit.h5.\n output=ROOT\n if output.exists():\n  with h5py.File(output,\'r\') as h:\n   if h.attrs.get(\'cache_key\')==key:return h[\'fitted\'][:].T,h[\'R_version\'].asstr()[()].tolist()\n  _release_cleanup_target(output).unlink()\n meta=obs[[\'t\',\'seg\',\'study_id\',\'library_id\']].copy();meta[\'good\']=good\n # Set this storage path independently before running. cells.tsv.\n meta.to_csv(ROOT,sep=\'\\t\',index=False);(ROOT).write_text(B_GAM_SOURCE,encoding=\'utf-8\')\n env=os.environ.copy();env.update(TMPDIR=str(folder),TMP=str(folder),TEMP=str(folder),LC_ALL=\'C\',LANG=\'C\')\n # Set this storage path independently before running. fit.R.\n cmd=[str(rscript),\'--vanilla\',str(ROOT),str(ROOT),str(raw_path),str(output)]\n c=subprocess.run(cmd,env=env,capture_output=True,text=True,errors=\'replace\')\n if c.returncode:raise RuntimeError(c.stdout+c.stderr)\n with h5py.File(output,\'r+\') as h:\n  h.attrs[\'cache_key\']=key;return h[\'fitted\'][:].T,h[\'R_version\'].asstr()[()].tolist()\n\ndef render_b(factor,activity,xy,output_dir):\n import matplotlib\n matplotlib.use(\'Agg\')\n import matplotlib.pyplot as plt\n from matplotlib.colors import LinearSegmentedColormap,Normalize\n w,h,ppi=300,343,75\n def box(x,y,ww,hh):return [x/w,1-(y+hh)/h,ww/w,hh/h]\n def text(fig,x,y,label,**kw):return fig.text(x/w,1-y/h,label,va=\'baseline\',**kw)\n with plt.rc_context({\'font.family\':\'Arial\',\'font.size\':8,\'axes.linewidth\':.65}):\n  fig=plt.figure(figsize=(w/ppi,h/ppi));ax=fig.add_axes(box(19,52,262,206))\n  ax.scatter(xy[:,0],xy[:,1],s=1.5,c=\'#D4D4D4\',alpha=.55,linewidths=0)\n  avail=activity[\'status\']==\'available\';mask=activity[\'display_mask\'];v=activity[\'fitted\'];sc=None\n  if avail:\n   ii=np.flatnonzero(mask)[np.argsort(v[mask],kind=\'stable\')]\n   cmap=LinearSegmentedColormap.from_list(\'activity\',[\'#D4D4D4\',\'#FFFFFF\',\'#F3151D\'])\n   sc=ax.scatter(xy[ii,0],xy[ii,1],s=1.5,c=v[ii],cmap=cmap,norm=Normalize(0,activity[\'display_vmax\'],clip=True),alpha=.88,linewidths=0)\n  lo=xy.min(0)-1;hi=xy.max(0)+1;ax.set_xlim(lo[0],hi[0]);ax.set_ylim(lo[1],hi[1]);ax.set_aspect(\'equal\');ax.axis(\'off\')\n  title=r\'$\\it{\'+factor+\'}$ (+) activity\' if avail else r\'$\\it{\'+factor+\'}$ activity\'\n  labels=[text(fig,150,40,title,fontsize=15.36,ha=\'center\')]\n  if avail:\n   cax=fig.add_axes(box(91,278,118,8));cb=fig.colorbar(sc,cax=cax,orientation=\'horizontal\')\n   cb.set_ticks([0,activity[\'display_vmax\']]);cb.ax.tick_params(labelsize=11.52,length=2,pad=1);cb.outline.set_linewidth(.4)\n   cb.solids.set_edgecolor(\'face\');cb.solids.set_linewidth(.12);labels.extend(cax.get_xticklabels())\n   labels.append(text(fig,150,271,\'Regulon activity (AUC)\',fontsize=11.52,ha=\'center\'))\n   origin=activity.get(\'network_origin\')\n   context=(\'CVG/VGN-derived transfer; early + SS2 scored\' if origin==\'CVG_VGN\' else \'Early-derived regulon (\'+str(activity.get(\'network_version\'))+\'); early + SS2 scored\') if origin else \'Early + SS2 scored; UMI / E10.5 / low-depth cells unscored\'\n   axis_note=\'UMI / E10.5 / low-depth unscored; 20871-gene AUC axis\' if origin else \'20871-gene axis; separate from the former three-source score\'\n   labels.append(text(fig,150,318,context,fontsize=7.2,ha=\'center\'))\n   labels.append(text(fig,150,332,axis_note,fontsize=6.9,ha=\'center\'))\n  else:\n   unavailable_label=\'Regulon coverage <80%; activity not estimable\' if activity[\'signature\'][\'full_count\'] else \'No motif-supported regulon (v9/v10)\'\n   labels.append(text(fig,150,278,unavailable_label,fontsize=10 if not activity[\'signature\'][\'full_count\'] else 9,ha=\'center\'))\n   labels.append(text(fig,150,311,\'Activity not estimable; gray does not mean absent RNA\',fontsize=7.6,ha=\'center\'))\n  fig.canvas.draw();renderer=fig.canvas.get_renderer()\n  clipped=[t.get_text() for t in labels if not(fig.bbox.contains(*t.get_window_extent(renderer).p0) and fig.bbox.contains(*t.get_window_extent(renderer).p1))]\n  assert not clipped,clipped\n  out=Path(output_dir);out.mkdir(parents=True,exist_ok=True);png=out/(factor+\'-B.png\');pdf=out/(factor+\'-B.pdf\')\n  fig.savefig(png,dpi=600)\n  if \'strict_editable_pdf\' not in globals():exec(STRICT_PDF_SOURCE,globals())\n  pdfinfo=strict_editable_pdf(fig,pdf,data_axes=());plt.close(fig)\n return dict(png=str(png),pdf=str(pdf),pdf_export=pdfinfo,dpi=600,pixels=[2400,2744],\n  fontsize_title=15.36,fontsize_metric=11.52,point_area=1.5,png_sha256=_sha(png),pdf_sha256=_sha(pdf),clipped_labels=clipped)\n\ndef validate_figures(figures,folder):\n # Set this storage path independently before running.\n dep=ROOT\n # Set this storage path independently before running. python.exe.\n py=ROOT;pop=ROOT\n # Set this storage path independently before running. validate_figure_output.py. Use the supplied fig4_validate_figure_output.py.\n validator=ROOT\n for tf,info in figures.items():\n  reports=[]\n  for cmd in [[str(py),str(validator),info[\'png\']],\n   # Set this storage path independently before running.\n   [str(py),str(validator),info[\'pdf\'],\'--expected-clips\',\'0\',\'--pdftoppm\',str(pop),\'--render-dpi\',\'150\',\'--render-dir\',str(ROOT/tf)]]:\n   c=subprocess.run(cmd,capture_output=True,text=True,errors=\'replace\')\n   if c.returncode:raise RuntimeError(c.stdout+c.stderr)\n   reports.append(c.stdout)\n  info[\'validator_reports\']=reports;info[\'validation_status\']=\'PASS\'\n\ndef score_and_render(signatures,metadata=None,*,project_root=None,work_dir=None,output_dir=None,shared_file=None,scenic_python=None,rscript=None,validate=True):\n """Complete signatures -> common 20871-gene weighted AUC -> native GAM -> B.\n\n All inputs are read-only. Defaults write only this task\'s render subdirectory.\n Unsupported signatures and unscored UMI/E10.5 remain NaN, never zero-filled.\n Metadata (resources, GRN/source/run evidence, predeclared choice policy) forms\n part of the cache key. The caller may merge returned arrays into its result.\n """\n # Set this storage path independently before running.\n root=Path(project_root or \'ROOT\').resolve()\n # Set this storage path independently before running.\n folder=Path(work_dir or ROOT).resolve();folder.mkdir(parents=True,exist_ok=True)\n # Set this storage path independently before running.\n out=Path(output_dir or ROOT);shared=Path(shared_file or ROOT)\n # Set this storage path independently before running. python.exe.\n py=Path(scenic_python or ROOT);r=Path(rscript or \'ROOT\')\n signatures={tf:(None if weights is None else {str(g):float(v) for g,v in weights.items()}) for tf,weights in signatures.items()};metadata=_plain(metadata or {})\n cells,obs,xy,good=_shared(root,shared)\n # Set this storage path independently before running. Inputs.h5. Candidate input contract; supply the immutable input container (standalone producer not located).\n files=[shared,ROOT]\n for study in [\'GSE178931\',\'GSE165502\']:files.extend([root/f\'Tmp/scenic/inputs/{study}_axes.npz\',root/f\'Tmp/scenic/inputs/{study}_counts.npz\'])\n hashes={str(p):_sha(p) for p in files}\n payload=dict(signatures=signatures,metadata=metadata,input_sha256=hashes,worker_sha256=hashlib.sha256(B_WORKER_SOURCE.encode()).hexdigest(),GAM_sha256=hashlib.sha256(B_GAM_SOURCE.encode()).hexdigest())\n # Set this storage path independently before running. raw.h5.\n key=hashlib.sha256(json.dumps(payload,sort_keys=True,separators=(\',\',\':\')).encode()).hexdigest();raw_path=ROOT;hit=False\n if raw_path.exists():\n  with h5py.File(raw_path,\'r\') as h:hit=h.attrs.get(\'cache_key\')==key\n if not hit:\n  config=dict(root=str(root),shared=str(shared),output=str(raw_path),cache_key=key,signatures=signatures,metadata=metadata)\n  # Set this storage path independently before running. score-config.json.\n  (ROOT).write_text(json.dumps(config),encoding=\'utf-8\');(ROOT).write_text(B_WORKER_SOURCE,encoding=\'utf-8\')\n  # Set this storage path independently before running.\n  env=os.environ.copy();env.update(TMP=str(folder),TEMP=str(folder),NUMBA_CACHE_DIR=str(ROOT))\n  # Set this storage path independently before running. score-worker.py.\n  c=subprocess.run([str(py),\'-B\',str(ROOT),str(ROOT)],env=env,capture_output=True,text=True,errors=\'replace\')\n  if c.returncode:raise RuntimeError(c.stdout+c.stderr)\n with h5py.File(raw_path,\'r\') as h:\n  np.testing.assert_array_equal(h[\'cell_ids\'].asstr()[:],cells);np.testing.assert_array_equal(h[\'TFs\'].asstr()[:],list(signatures))\n  raw=h[\'raw_auc\'][:];genes=h[\'genes\'].asstr()[:];scope=h[\'source_scope\'][:];gate=h[\'rank_gate\'][:];positive=h[\'positive_genes\'][:];cutoff=int(h[\'rank_cutoff\'][()])\n  audit=json.loads(h[\'audit_json\'].asstr()[()]);methods=json.loads(h[\'methods_json\'].asstr()[()])\n fitted,versions=_fit(raw_path,obs,good,folder,r,key)\n assert raw.shape==fitted.shape==(7355,len(signatures));assert np.isnan(fitted[~np.isfinite(raw)]).all()\n activities={};figures={}\n for j,tf in enumerate(signatures):\n  v=fitted[:,j];mask=np.isfinite(v)&np.isfinite(raw[:,j]);status=\'available\' if mask.any() else \'not_estimable\'\n  late=(obs.age.to_numpy(float)>=14.5)&~obs.seg.isin((\'CVG\',\'VGN\')).to_numpy()&good&mask;vmax=None;reference=\'Not applicable\'\n  if status==\'available\':\n   mx=float(v[late].max()) if late.any() else np.nan\n   if np.isfinite(mx) and mx>0:vmax=mx;reference=\'Good scored SGN-segment cells at age >= E14.5\'\n   else:vmax=max(float(v[mask].max()),1e-6);reference=\'All scored cells; late reference had no positive fitted maximum\'\n  activity=dict(status=status,raw=raw[:,j],fitted=v,display_mask=mask,display_vmax=vmax,display_reference=reference,\n   display_numeric_clipping=False,display_saturated_cells=int((v[mask]>vmax).sum()) if vmax else 0,signature=audit[tf],\n   reason=\'\' if status==\'available\' else (\'No motif-supported regulon (v9/v10)\' if not signatures[tf] else \'Signature fails the fixed 80% measured-coverage rule\'),\n   scope=\'GSE178931+GSE165502 only on native measured 20871-gene axis; UMI/E10.5 unscored\',metric=\'Weighted AUCell; top5%; seed42; normalize=False\')\n  choice=metadata.get(\'selection\',{}).get(tf,{})\n  activity[\'network_origin\']=choice.get(\'origin\');activity[\'network_version\']=choice.get(\'version\');activity[\'transfer_score\']=choice.get(\'origin\')==\'CVG_VGN\'\n  activities[tf]=activity;figures[tf]=render_b(tf,activity,xy,out)\n if validate:validate_figures(figures,folder)\n return dict(cell_ids=cells,genes=genes,TFs=list(signatures),raw_auc=raw,fitted=fitted,source=obs.study_id.to_numpy(),source_scope=scope,rank_gate=gate,positive_genes=positive,rank_cutoff=cutoff,\n  activities=activities,figures=figures,methods=methods,fit_methods=dict(formula="value~s(t,bs=\'ts\',k<=10)",gamma=1.5,method=\'GCV.Cp\',weights=\'Equal source, equal libraries, equal cells within library\',branches=\'Each native segment fitted independently\',R_version=versions),metadata=metadata,input_sha256=hashes,cache_key=key,score_cache_hit=hit,worker_source_sha256=hashlib.sha256(B_WORKER_SOURCE.encode()).hexdigest(),GAM_source_sha256=hashlib.sha256(B_GAM_SOURCE.encode()).hexdigest())\n'}
B_RETRY_POLICY = {'primary_cohort': 'whole_early', 'seeds': [42, 43, 44], 'database_order': ['v10', 'v9'], 'selection_rule': 'Prefer seed42 full canonical positive regulon from early cohort, requiring positive regulon support in at least two of three seeds for that same database version; report all target overlaps. Version v10 is considered first. If neither version meets this rule, retain the original CVG_VGN seed42 v10 full signature as an explicitly labelled transfer score.', 'no_visual_selection': True, 'no_target_truncation': True, 'min_full_signature_coverage': 0.8, 'score_scope': ['GSE178931', 'GSE165502'], 'score_axis': 'Original measured intersection, 20871 genes; UMI and E10.5 unscored', 'matched_version_comparison': 'v9 10kb versus v10 10kb; original two-window v10 retained as primary baseline', 'zfpm2_policy': 'Only use a named motif-supported positive regulon. If unavailable in paired v9/v10 annotations, retain missing activity with an explicit status label; no GATA or RNA substitution.'}
B_RETRY_BASELINE = {'files': {'Figures\\fig4B-D\\SGN-1\\Foxo6.png': {'sha256': '411b6a745535a974e63dc5144a347baa6c93b15af333e4e296d9ebe422e96924', 'bytes': 4302112}, 'Figures\\fig4B-D\\SGN-1\\Neurod1.png': {'sha256': '2d5429ddd5d2a5ce323355fb9d2c0b9dbb0f513f9b04167bca77f5c7df202541', 'bytes': 4614552}, 'Figures\\fig4B-D\\SGN-1\\Neurod2.png': {'sha256': '6c078ceae126d1fee6f6ebfabdec1bdc3f8b69e2459106439410a27844c2c0f8', 'bytes': 4292066}, 'Figures\\fig4B-D\\SGN-1\\Nhlh1.png': {'sha256': 'fa58b6fc665180ee78c383c2d997417fd7a7f8ff9bb3ecfce55d604e46745824', 'bytes': 4206148}, 'Figures\\fig4B-D\\SGN-1\\SGN-lines.png': {'sha256': '7808a33596c69f1385cdaaec0b3df31c6c17603a7e6d4014082b862275e03e2e', 'bytes': 790485}, 'Figures\\fig4B-D\\SGN-1\\Zfpm2.png': {'sha256': '006c844e32a63c622a423c253337c4462f5b08937982cb28229ce220d4a53d50', 'bytes': 4419892}, 'Figures\\fig4B-D\\SGN-1\\Foxo6.pdf': {'sha256': '4d6f0a5a24c8d66858b7bba8ae88292fafab75466af0dbcd597ea13c98235c9e', 'bytes': 11228554}, 'Figures\\fig4B-D\\SGN-1\\Neurod1.pdf': {'sha256': '6ab4d33105025f542295163ae7cdc3ad9d3a2b7e73bad4a11a7fe786688d2323', 'bytes': 9948982}, 'Figures\\fig4B-D\\SGN-1\\Neurod2.pdf': {'sha256': '3683871df3e2943629e9bf9cd8af1e752f467f1097804b1cd35fc54a7a274a1f', 'bytes': 11225733}, 'Figures\\fig4B-D\\SGN-1\\Nhlh1.pdf': {'sha256': 'bff1e885137d755511310f49d4a20bc0786b5e53e9554a23a503932ab9fe090e', 'bytes': 11210886}, 'Figures\\fig4B-D\\SGN-1\\SGN-lines.pdf': {'sha256': 'e692259c0778b80bfe6cf8f5af4ada5b33f1b0e0e0782de91ced9f5fe9828372', 'bytes': 140215}, 'Figures\\fig4B-D\\SGN-1\\Zfpm2.pdf': {'sha256': '84a410bf7bfd49291e276bf9f8fb8a7fece97f9c3236c7db758f25b945a08acd', 'bytes': 9947072}, 'Processed Data\\07_Fig4B-D\\SGN-1\\Activity.h5': {'sha256': 'f452ef42f22af6b007591af4a6815770d63465f038d6b5536bd77d9d77d35e15', 'bytes': 8918771}, 'Processed Data\\07_Fig4B-D\\SGN-1\\Fig4B-D.h5': {'sha256': '7d1c9030ae2052453521e474f77577f0ad2b5293f4717e37b3b4125b99ef2ff9', 'bytes': 25097189}, 'Processed Data\\07_Fig4B-D\\SGN-1\\SGN-lines.h5': {'sha256': '698c19bfd2ded659cb52ba95e8f0c8673dc1f9b967e65599f066affea31f2a86', 'bytes': 244880}}, 'Activity_existing_logical_hashes': {'/': '5bac991dfc076210b8e891a7327f2a8d8da6e688aec9207246252269e41cc5f8', 'CVG_VGN': '44136fa355b3678a1146ad16f7e8649e94fb4fc21fe77e8310c060f61caaff8a', 'CVG_VGN/candidate_adjacency_json': 'd27e07dd5c07f9c4d740f276eae71910672610524de43e873316254fd2c17ad4', 'CVG_VGN/candidate_modules_json': '898149c0faa830af443719bc4e6bcf400304a49887b417d15b7475ef25d3494e', 'CVG_VGN/cells': '0d14e9c5a6f5bba999d738506fea2f2f8b5588ae03aebb2e962dc99a8be8afa5', 'CVG_VGN/genes': '000de4866452740f022947c1b9638b53f58aa329ecf3bf4e409968a2683c79f6', 'CVG_VGN/rank1500': 'cea2721040b3ff58b00133992dd5a381b36b2ad022434e58021741a87bbcd15c', 'CVG_VGN/rank1500/frozen_Nhlh1_calibration_json': '5eb024478f142d55852bfe2e571ba753799f0478d5cdcc91ffef84af8d412629', 'CVG_VGN/rank1500/motif_evidence_json': 'efa3396c0a46f7b9cc0da49ba5046041797a58286b35f9b0e8ab8116333196bd', 'CVG_VGN/rank1500/regulons_json': '40e6cb9ed2a7af543d3e32031a41d0675b61118f7c1dd58a98e8255e36aeac41', 'CVG_VGN/rank1500/summary_json': '9e1b88c6df07fe586058a2de5986ff9e1d198da3f2d99917300d0c3fc61c293e', 'CVG_VGN/rank5000': '478470f456c2c8af949e0d2c2622d7034968f4e3ecfd68d10342de6c237d638e', 'CVG_VGN/rank5000/frozen_Nhlh1_calibration_json': '5eb024478f142d55852bfe2e571ba753799f0478d5cdcc91ffef84af8d412629', 'CVG_VGN/rank5000/motif_evidence_json': '3ffedee282c4bfd330feb5a1c2621a1d31c31935155b632541700e3485a4d3a1', 'CVG_VGN/rank5000/regulons_json': '5916d3142946b3b63277de97555cb33560343e4152b55aaaa8698355d6999bda', 'CVG_VGN/rank5000/summary_json': '83c010f27deecfdc27840df3f2f31c37fbc41a0ba9015a994c8fbeef90d7aa2f', 'CVG_VGN/source_metadata_json': '57e403f53294190278416e1f0dfc719d4b3140bd3476240cb8c4f9652f39c7a8', 'activity': 'bc485b5b25e82055383cdb2ebc9ce79285d910e95886f3aa1261a59df3e0d796', 'activity/TFs': '50953ebab25dd36f626622e0b7a3a2422664ea4be31a4411ea35a8d1ddfbee1b', 'activity/candidate_sources_json': '2c356c348e6ce54d01a1bcd27629f30220a8559289202864a75211f36ae540f7', 'activity/cell_ids': 'c03521724787d7fef7b047659f256b0706906c7fe53a57bda52fd225e8e5ed59', 'activity/executed_source': 'a20c6fb9c7aa7f1534108bcb0460e247d2a5c7246b3028d43c98ad2eec09e077', 'activity/extra_cell_ids': 'b0cc780dbe97ad35e43ee9b42578b2c9ed7d52a5b7da3654cc258ca59dc32487', 'activity/extra_counts': 'd4e36278b9eaa2600094f14de636b796020bfd312a27489b226a92703f7aa274', 'activity/extra_genes': '2bace67bc6ec242442a52f02eb7265f4738047f853f74aee59240f2a3b4c12d9', 'activity/gene_measured_by_source': '6bb1499f6d242bedeb98ae8038efd6b2de9ea03806da8c71c8cb2c4f2022bd4f', 'activity/genes': 'dbfad20bdcd9ebdc205a1779177e9a62ac95f2456b13e10eac4da6cf6fec720c', 'activity/measured_source_ids': '0edd80ec4618cb67aefe368585795fe6ad27144a330e7a600b4eeaf274c6e59f', 'activity/metadata_json': '03cefa4c2df4848d6b02739dae15d19b29abaeb02cc0aea87fe8b5673977f245', 'activity/positive_genes': '695de8d26395455d81c3b03c6cfb95db150b3581db4ab2bb8241bbeb7c574cb3', 'activity/rank_cell_ids': 'b0cc780dbe97ad35e43ee9b42578b2c9ed7d52a5b7da3654cc258ca59dc32487', 'activity/rank_cutoff': 'a2a466d81336c4e4205844c52fcbb12ce5df784327f36799d8296ab2dea072b8', 'activity/rank_gate': 'b391bb21b5f443eedf63ad6883de689f100e50a060aeb0ba6230930aff51e300', 'activity/rank_gene_order': '2492780a6cba08228a6bb59d517720e49b86d7b0783c0a3219693084f1f5564a', 'activity/raw_auc': '050bf92bc88a0ee212b2522c36831b8f783dea5546aa5ae823fdfc5bd38abb1d', 'activity/score_available': '2d54102c0dfbaa842ce83e0602471544c5812ea1de1fa64bc1f191c7658897f2', 'activity/selected_regulons_json': 'a1f10dcb18cc056287eb55f73776bc146c95df8c5dbc2b0dd0920d77d79f9043', 'activity/source': '2ead83a51cafbf3910d537dd6100fe090445ff12695338cb4aa35e2fc8fd264d', 'activity/source_scope': 'b391bb21b5f443eedf63ad6883de689f100e50a060aeb0ba6230930aff51e300', 'activity/status_reason_json': '7f8718ee2f6d3a6c1d6f7452f32bfb1e8bb9503558fd02b9555b3c64c16beec8', 'executed_source': '76e2747ab1402cedf8d9ee5c419c290c9d580a2d84598e2f86da7d775cab89e0', 'initial_two_source_executed_source': '18943171dac32ead0626086ad1c740d74703a3df870449e91f11e3a6346f716a', 'late_SS2': '44136fa355b3678a1146ad16f7e8649e94fb4fc21fe77e8310c060f61caaff8a', 'late_SS2/candidate_adjacency_json': 'd246ef9ae0a2f3b43cdbd790a4ae1b2aaba43b40261903af677fb16917b5bc47', 'late_SS2/candidate_modules_json': '0151d887c91251c2fa250ea56b1dd400e7955385e6c9be0220d1af1d721fe8f8', 'late_SS2/cells': 'a85df768d8d07750988826644fe3395af621bf2acbacc7aabe112746bad54d8f', 'late_SS2/genes': '9742e4a1b0b052d350fcb8ad8b24e24cb5ae6004ed2ebc96903ddccb2548e455', 'late_SS2/rank1500': 'c63f52f8e55547fdbaad088fa54fdb4e452234130b40158d7de2c8e23edcb1bc', 'late_SS2/rank1500/frozen_Nhlh1_calibration_json': '5eb024478f142d55852bfe2e571ba753799f0478d5cdcc91ffef84af8d412629', 'late_SS2/rank1500/motif_evidence_json': '7f2eeb4b9feb992f850b953fe65bd0ccbcd58800e6ff60a96ba60bf9b77d1eb7', 'late_SS2/rank1500/regulons_json': '9996ac71281dea9898dabc136fb5ce713ea3f0a552f0d34a73d351c5cda63aa1', 'late_SS2/rank1500/summary_json': 'b81316473de9f96f4d92cf7a37b9771f97343d464056c80aa458734302765944', 'late_SS2/rank5000': '2ca8fb544f0c09f97e7371d8c04f286be63629ca89216507d49bc1a1fdde0f04', 'late_SS2/rank5000/frozen_Nhlh1_calibration_json': '5eb024478f142d55852bfe2e571ba753799f0478d5cdcc91ffef84af8d412629', 'late_SS2/rank5000/motif_evidence_json': 'd3352afa83c4f50a5efe4c83f1b4d5b0cc903d6c43de67389453059917abac55', 'late_SS2/rank5000/regulons_json': '885ccc5d83786973dd9e14998f88406117468430b28ebdbc747b8b821e40d551', 'late_SS2/rank5000/summary_json': '0ee16ad3933ed69e20c5ffc086e01f3ac2b06f6f9953882f9c8c691b7737ced7', 'late_SS2/source_metadata_json': '90e25913e81253780f7341ab2d4afdcf504875281d2f934611641f6ab368cf79', 'late_UMI': '44136fa355b3678a1146ad16f7e8649e94fb4fc21fe77e8310c060f61caaff8a', 'late_UMI/candidate_adjacency_json': '31ada4e670591decec64f9e4526a8a0b6c5bbda77c802e17ba5ee55853dda036', 'late_UMI/candidate_modules_json': 'ffee84ba545e7595e1e7f1dc5fe5e1ea38490cdfeab2aba504251e13f8eb177e', 'late_UMI/cells': 'a2fc4e6af8900d28983ba09475fda7bc14885f4054e64ecce7eaaf4439b7f000', 'late_UMI/genes': '57d7664899bf3efe7b47365002f07d8edf10e4c51e87f37472c42b4a6ca4874b', 'late_UMI/rank1500': '43632d06ce009fa8576403f0d4eaa708a720c071b99bf64e338c9e7d35536c81', 'late_UMI/rank1500/frozen_Nhlh1_calibration_json': '5eb024478f142d55852bfe2e571ba753799f0478d5cdcc91ffef84af8d412629', 'late_UMI/rank1500/motif_evidence_json': '15895d590f9420e928ace0fa07275ed262c73e4f90e7c262a6624e850a4e610c', 'late_UMI/rank1500/regulons_json': 'ff5eb51a07da33615f3c2ac4be4abbaca8cfd1139698b119f00e65d41c082777', 'late_UMI/rank1500/summary_json': '484b6effbfa151fd46e8938568e873916b2677b45535c859f6fe4c699073e5bd', 'late_UMI/rank5000': 'f3a90054ae2d913f1f31ff315efa77d91c3469736260b7aa9afc56d88bbb4840', 'late_UMI/rank5000/frozen_Nhlh1_calibration_json': '5eb024478f142d55852bfe2e571ba753799f0478d5cdcc91ffef84af8d412629', 'late_UMI/rank5000/motif_evidence_json': '46311031266572406521b3e1800ba71008deeb33478ea597a6a34133a2e38c30', 'late_UMI/rank5000/regulons_json': '8301a2931ba40682fe537c050690604936290634af46199a9aa91ba7fb183a09', 'late_UMI/rank5000/summary_json': 'e8bade1ad40fd6a04a25f18c95591f2770da585bb28ff5f1c9573012085753ae', 'late_UMI/source_metadata_json': '5ef994b25516037bdc8af6d2ccf22a5df60a9c19f1f919d1558f4f1741c085cd', 'methods_json': '65b5946ec9768e84cd35edbe55eb6ad8103142f0cf1b5af4a85fa8037d4567c4', 'planned_scoring_policy_history_json': '7f9fd9fc74b00a1d27aef4d39f58174142e9c3ee3f43dbb804007e620b84c717', 'reproducible_worker_sha256': 'de5c74c7a8a0e9ac9ecd36338a574792abc529433ee1727730de43b3b806afe1', 'reproducible_worker_source': '2c1d82d3f4a9f34a8ba03c160e8af7d9431106fb51e91bdb17923a0c187b6b22', 'resource_annotation_evidence_json': 'b657bdb0de1738676b91f917156ff196f6e29f5e28ef23b7092f8d2a070cbb8d', 'whole_early': '44136fa355b3678a1146ad16f7e8649e94fb4fc21fe77e8310c060f61caaff8a', 'whole_early/candidate_adjacency_json': 'd3accb710920943361e0b2b8f9f783da2530406eb843e68f1247c30b56bf8ddf', 'whole_early/candidate_modules_json': 'd283deb1cbb27a171443950fc600e38e7ffb3b07da430c60ef400c8458c988e6', 'whole_early/cells': 'cb3313ec6611f1d0ea5b08fbf9d5b7b0b4f931455a5b19f9583db6cecf8e8c3b', 'whole_early/genes': '50d905fa803d69a588b4466acf3da6e13cd44bb500fc28a7d219833775bba1de', 'whole_early/rank1500': '82f5b2a80e4693a4d791f964deff29851f7da8bf9f8acee3f990228c118cf9cf', 'whole_early/rank1500/frozen_Nhlh1_calibration_json': '1f04d3ac25f7049a152d183f2ca9e373725cbf7ae8f5511aa3b4b8156d4be0c2', 'whole_early/rank1500/motif_evidence_json': '8af84559c0545417d28b7b5646ab677887b265290f2006c74f5814bad9d61bf6', 'whole_early/rank1500/regulons_json': '87068ae384bc62b85c7b5e2e0f15bae1147e24a654471f9dee29eeab06f6c894', 'whole_early/rank1500/summary_json': 'b179f55b95bcfcf3523369d98b7d57b52eaad755fd46cb27875fbceaca8878b6', 'whole_early/rank5000': '8b6b84063dd8ece51545258f39f2f8c8febc6bb2aeccb0d4963e92a9bc152125', 'whole_early/rank5000/frozen_Nhlh1_calibration_json': '5eb024478f142d55852bfe2e571ba753799f0478d5cdcc91ffef84af8d412629', 'whole_early/rank5000/motif_evidence_json': '8af84559c0545417d28b7b5646ab677887b265290f2006c74f5814bad9d61bf6', 'whole_early/rank5000/regulons_json': '87068ae384bc62b85c7b5e2e0f15bae1147e24a654471f9dee29eeab06f6c894', 'whole_early/rank5000/summary_json': 'b179f55b95bcfcf3523369d98b7d57b52eaad755fd46cb27875fbceaca8878b6', 'whole_early/source_metadata_json': '4c34d42b36339a7db244b60aa9a0470267118ddcf5abf0127cfb7e27d686acfd'}, 'old_codecells_sha256': ['a41d75cf12b13cf233847243eceb5e5f6c6e91c1a50d775644af0c70f027077d', '6337d1b0ffe772a774f470ce64fe4fac2fa955b69333e1c159170cabcb2cd17f', 'b822652991b5982d46c0b23b8db2388a5d9d3d3d3049847fd4975a7129ac7828', 'da649785c4aaa47c4d1f64304317c192d524506891aacdb65cb1081dd0fd582c', 'a5d835b4f152fd229bf25e9b1c27861ebabf280cefa170ac67f430430697930b', '4426bd340c722476ceb78d9dbff831832ab777f97cbca761b3867ea1eb0ade4c', 'cee8781224269a55152ca3c2b4f518e2fca606ed20663886249cbe97b3f42571', 'ccc7dbf3005f65b9ca33b7bf1c1385f5c744a82745586ec2e371b7f7305e1241', '4cde20a95962097e23848e466a06115b7b2888e62aec068176482d6c8f6cb077', 'a8e3468457c759d4f78616bdd1d05ae52a7b1ffaf7e544d5affc3f76925a0fb3', '55d81a242b2ce207e2051b6023444f11af8cba80121c466f2b558ca55a0ea51a']}
RECOMPUTE_B_RETRY = False

from pathlib import Path
from urllib.parse import quote, unquote
import ast, hashlib, json, os, shutil, subprocess, sys
import h5py, numpy as np

# Set this storage path independently before running.
_RETRY_ROOT = ROOT
# Set this storage path independently before running.
_RETRY_WORK = ROOT
# Set this storage path independently before running. Activity.h5. Produced by the SCENIC workers in fig4_SGN_program_analysis.ipynb.
_RETRY_ACTIVITY = ROOT
# Set this storage path independently before running. Fig4B-D.h5.
_RETRY_SHARED = ROOT
# Set this storage path independently before running.
_RETRY_FIGURES = ROOT
# Set this storage path independently before running. 05_fig4B-D_SGN.ipynb. Use the supplied fig4_SGN_program_analysis.ipynb source.
_RETRY_CODE = ROOT
# Set this storage path independently before running. python.exe.
_RETRY_PYTHON = ROOT
# Set this storage path independently before running. Rscript.exe.
_RETRY_R = ROOT
_RETRY_VERSIONS = ('v10', 'v10_10kb', 'v9')
_RETRY_JOBS = [('whole_early',seed,version) for version in _RETRY_VERSIONS for seed in (42,43,44)] + [('CVG_VGN',42,version) for version in _RETRY_VERSIONS]


def _retry_plain(v):
    if isinstance(v,bytes): return {'bytes_hex':v.hex()}
    if isinstance(v,np.ndarray): return [_retry_plain(x) for x in v.tolist()]
    if isinstance(v,np.generic): return _retry_plain(v.item())
    if isinstance(v,dict): return {str(k):_retry_plain(x) for k,x in v.items()}
    if isinstance(v,(tuple,list)): return [_retry_plain(x) for x in v]
    if isinstance(v,Path): return str(v)
    return v


def _retry_json(v):
    return json.dumps(_retry_plain(v),sort_keys=True,ensure_ascii=False,allow_nan=True)


def _retry_sha(path):
    d=hashlib.sha256()
    with Path(path).open('rb') as h:
        for block in iter(lambda:h.read(8*1024*1024),b''): d.update(block)
    return d.hexdigest()


def _retry_nodes(path,prefix=''):
    nodes={}
    with h5py.File(path,'r') as f:
        start=f[prefix] if prefix else f
        def visit(name,o):
            if (not prefix and (name=='B_retry' or name.startswith('B_retry/') or name=='B_paper' or name.startswith('B_paper/'))) or (prefix and name=='integrity_json'):return
            d=hashlib.sha256();d.update(_retry_json(dict(o.attrs)).encode())
            if isinstance(o,h5py.Dataset):
                d.update(str(o.dtype).encode());d.update(str(o.shape).encode())
                if o.dtype.kind in 'OSU':
                    value=o.asstr()[()];d.update(json.dumps(value.tolist() if o.shape else value,ensure_ascii=False).encode())
                elif o.shape:
                    rowbytes=max(1,int(np.prod(o.shape[1:]))*o.dtype.itemsize)
                    step=max(1,1048576//rowbytes)
                    for i in range(0,o.shape[0],step):d.update(np.ascontiguousarray(o[i:i+step]).tobytes())
                else:d.update(np.asarray(o[()]).tobytes())
            nodes[name]=d.hexdigest()
        visit('/',start);start.visititems(visit)
    return nodes


def _retry_runtime_snapshot():
    # Historical paths identify protected outputs; their current bytes, not
    # historical hashes, define the boundary of this particular #12 execution.
    files={}
    for rel in B_RETRY_BASELINE['files']:
        if Path(rel).name=='Activity.h5':continue
        path=_RETRY_ROOT/rel
        files[rel]=dict(sha256=_retry_sha(path),bytes=path.stat().st_size)
    notebook=json.loads(_RETRY_CODE.read_text(encoding='utf-8'))
    codecells=[hashlib.sha256(''.join(c['source']).encode()).hexdigest() for c in notebook['cells'] if c['cell_type']=='code' and not ''.join(c['source']).startswith('#12.')]
    return dict(files=files,Activity_existing_logical_hashes=_retry_nodes(_RETRY_ACTIVITY),preexisting_codecells_sha256=codecells)


def _retry_assert_protected(snapshot):
    for rel,record in snapshot['files'].items():
        assert _retry_sha(_RETRY_ROOT/rel)==record['sha256'],'This #12 run changed a protected output: '+rel
    assert _retry_nodes(_RETRY_ACTIVITY)==snapshot['Activity_existing_logical_hashes'],'This #12 run changed original Activity nodes'



def _retry_cell_source():
    notebook=json.loads(_RETRY_CODE.read_text(encoding='utf-8'))
    cells=[''.join(c['source']) for c in notebook['cells'] if c['cell_type']=='code' and ''.join(c['source']).startswith('#12.')]
    assert len(cells)==1,'Append exactly one numbered #12 cell before running this block'
    return cells[0]


def _retry_guard():
    source=_retry_cell_source()
    # Switching the recompute flag alone must not invalidate a reproducible cache.
    tree=ast.parse(source)
    for node in tree.body:
        if isinstance(node,ast.Assign) and any(isinstance(t,ast.Name) and t.id=='RECOMPUTE_B_RETRY' for t in node.targets):node.value=ast.Constant(False)
        if isinstance(node,ast.Assign) and any(isinstance(t,ast.Name) and t.id=='B_RETRY_BASELINE' for t in node.targets):node.value=ast.Constant('Historical audit payload excluded from scientific cache key')
    code_sha=hashlib.sha256(ast.dump(tree,include_attributes=False).encode()).hexdigest()
    # Set this storage path independently before running. Inputs.h5. Candidate input contract; supply the immutable input container (standalone producer not located).
    files=[ROOT]
    for study in ('GSE178931','GSE165502'):
        files.extend(_RETRY_ROOT/f'Tmp/scenic/inputs/{study}_{suffix}.npz' for suffix in ('axes','counts'))
    # Set this storage path independently before running. GSE178931_expression.npz.
    files.extend([ROOT,ROOT,ROOT,ROOT])
    shared_nodes=_retry_nodes(_RETRY_SHARED,'shared')
    guard=dict(schema='SCENIC_B_retry_guard_v2',source_ast_sha256=code_sha,policy=B_RETRY_POLICY,
        shared_nodes=shared_nodes,shared_nodes_sha256=hashlib.sha256(_retry_json(shared_nodes).encode()).hexdigest(),
        input_sha256={str(p.relative_to(_RETRY_ROOT)):_retry_sha(p) for p in files},
        base_Activity_nodes_sha256=hashlib.sha256(_retry_json(_retry_nodes(_RETRY_ACTIVITY)).encode()).hexdigest(),
        scoring=dict(studies=['GSE178931','GSE165502'],axis=20871,seed=42,weighted=True,normalize=False,top_fraction=.05,coverage=.8),
        source_sha256={k:hashlib.sha256(v.encode()).hexdigest() for k,v in B_RETRY_SOURCES.items()})
    return guard,hashlib.sha256(_retry_json(guard).encode()).hexdigest()


def _retry_write(parent,key,value):
    key=quote(str(key),safe='')
    if key in parent:del parent[key]
    if isinstance(value,dict):
        g=parent.create_group(key);g.attrs['kind']='dict'
        for k,v in value.items():_retry_write(g,k,v)
    elif isinstance(value,(list,tuple)):
        g=parent.create_group(key);g.attrs['kind']='list'
        for i,v in enumerate(value):_retry_write(g,str(i),v)
    elif value is None:
        g=parent.create_group(key);g.attrs['kind']='none'
    elif isinstance(value,np.ndarray):
        if value.dtype.kind in 'OSU':d=parent.create_dataset(key,data=np.asarray(value,dtype=object),dtype=h5py.string_dtype('utf-8'))
        else:d=parent.create_dataset(key,data=value,compression='gzip' if value.ndim else None)
        d.attrs['kind']='ndarray'
    elif isinstance(value,str):
        d=parent.create_dataset(key,data=value,dtype=h5py.string_dtype('utf-8'));d.attrs['kind']='str'
    else:
        d=parent.create_dataset(key,data=json.dumps(_retry_plain(value)),dtype=h5py.string_dtype('utf-8'));d.attrs['kind']='json'


def _retry_read(node):
    kind=node.attrs.get('kind','')
    if isinstance(node,h5py.Group):
        if kind=='none':return None
        if kind=='list':return [_retry_read(node[str(i)]) for i in range(len(node))]
        return {unquote(k):_retry_read(v) for k,v in node.items()}
    value=node.asstr()[()] if node.dtype.kind in 'OSU' else node[()]
    return json.loads(value) if kind=='json' else value


def _retry_get(node):
    return json.loads(node[:].tobytes().decode()) if node.shape else json.loads(node.asstr()[()])


def _retry_run_command(command):
    # Set this storage path independently before running.
    env=os.environ.copy();env.update(TMP=str(_RETRY_WORK),TEMP=str(_RETRY_WORK),NUMBA_CACHE_DIR=str(ROOT))
    subprocess.run([str(x) for x in command],cwd=_RETRY_ROOT,env=env,check=True)


def _retry_materialize():
    _RETRY_WORK.mkdir(parents=True,exist_ok=True)
    for name,text in B_RETRY_SOURCES.items():
        path=_RETRY_WORK/name;path.parent.mkdir(parents=True,exist_ok=True);path.write_text(text,encoding='utf-8')
    # Set this storage path independently before running. selection-plan.json.
    (ROOT).write_text(_retry_json(B_RETRY_POLICY),encoding='utf-8')


def _retry_grn_valid(path):
    if not path.exists():return False
    with h5py.File(path,'r') as h:
        if h.attrs.get('stage')!='complete':return False
        def normalized(text):
            tree=ast.parse(text)
            for n in tree.body:
                if isinstance(n,ast.Assign) and any(isinstance(t,ast.Name) and t.id=='SEEDS' for t in n.targets):n.value=ast.Tuple(elts=[ast.Constant(43),ast.Constant(44)],ctx=ast.Load())
            return ast.dump(tree,include_attributes=False)
        if normalized(h['executed_source'].asstr()[()])!=normalized(B_RETRY_SOURCES['grn/GRN.py']):return False
        methods=json.loads(h['methods_json'].asstr()[()])
        for rel,digest in methods['input_sha256'].items():
            if _retry_sha(_RETRY_ROOT/rel)!=digest:return False
        # Set this storage path independently before running. allTFs_mm.txt.
        if _retry_sha(ROOT)!=methods['resource_sha256']['allTFs_mm.txt']:return False
        assert h['genes'].shape==(14901,) and h['TFs'].shape==(1401,) and h['cell_ids'].shape==(1161,)
        assert int(h.attrs['completed_targets'])==14901
        calibration=json.loads(h['calibration_json'].asstr()[()]);assert calibration['passed']
        assert all(r['weights_exact'] and r['TF_target_exact'] for r in calibration['targets'])
    return True


def _retry_compute_attempts():
    _retry_materialize()
    # Set this storage path independently before running. resources.json.
    manifest=ROOT;ready=False
    if manifest.exists() and not RECOMPUTE_B_RETRY:
        records=json.loads(manifest.read_text(encoding='utf-8'))
        # Set this storage path independently before running.
        ready=bool(records) and all((ROOT/r['name']).is_file() and (ROOT/r['name']).stat().st_size==r['bytes'] and _retry_sha(ROOT/r['name'])==r['sha256'] for r in records)
    # Set this storage path independently before running. resources.py.
    if not ready:_retry_run_command([sys.executable,'-X','utf8','-B',ROOT,'--download'])
    # Set this storage path independently before running. pruning.h5.
    if RECOMPUTE_B_RETRY and (ROOT).exists():
        # Set this storage path independently before running. pruning.h5.
        (_release_cleanup_target(ROOT)).unlink()
    pending=[seed for seed in (43,44) if not _retry_grn_valid(_RETRY_WORK/f'grn/seed{seed}.h5')]
    if pending:
        for seed in pending:
            stale=_RETRY_WORK/f'grn/seed{seed}.h5'
            # Set this storage path independently before running.
            assert stale.parent.resolve()==ROOT
            if stale.exists():_release_cleanup_target(stale).unlink()
        # The full original implementation remains embedded; only pending seeds run.
        text=B_RETRY_SOURCES['grn/GRN.py'].replace('SEEDS=(43,44)','SEEDS='+repr(tuple(pending)))
        # Set this storage path independently before running. GRN.pending.py.
        path=ROOT;path.write_text(text,encoding='utf-8')
        # Set this storage path independently before running.
        _retry_run_command([_RETRY_PYTHON,'-X','utf8','-B',path,'--root',_RETRY_ROOT,'--output-dir',ROOT])
    for source,seed,version in _RETRY_JOBS:
        print('SCENIC retry:',source,seed,version,flush=True)
        # Set this storage path independently before running. prune.py.
        command=[_RETRY_PYTHON,'-X','utf8','-B',ROOT,'--source',source,'--seed',seed,'--version',version]
        try:_retry_run_command(command)
        except subprocess.CalledProcessError:
            # Failed/partial or stale retry cache is never treated as evidence.
            # Invalidation affects only this task's temporary pruning group.
            # Set this storage path independently before running. pruning.h5.
            path=ROOT;key=f'{source}/seed{seed}/{version}'
            if not path.exists():raise
            with h5py.File(path,'r+') as h:
                if key not in h:raise
                del h[key]
            _retry_run_command(command)


def _retry_select():
    attempts={};overlap=[];selection={};resource_records={}
    # Set this storage path independently before running. pruning.h5.
    with h5py.File(ROOT,'r') as h:
        for source,seed,version in _RETRY_JOBS:
            path=f'{source}/seed{seed}/{version}';g=h[path];assert g.attrs.get('complete')
            attempts[path]=dict(regulons=_retry_get(g['regulons']),summary=_retry_get(g['summary']),methods=_retry_get(g['methods']))
            for name,sha in attempts[path]['methods']['resources'].items():
                # Set this storage path independently before running.
                folder=ROOT if version=='v9' else ROOT
                file=folder/name;assert _retry_sha(file)==sha
                resource_records[str(file.relative_to(_RETRY_ROOT))]=dict(sha256=sha,bytes=file.stat().st_size)
    for tf in ('Neurod1','Zfpm2'):
        choice=None;checks=[]
        for version in B_RETRY_POLICY['database_order']:
            regs={}
            for seed in (42,43,44):
                hits=[r for r in attempts[f'whole_early/seed{seed}/{version}']['regulons'] if r['TF']==tf]
                assert len(hits)<=1;regs[seed]=hits[0] if hits else None
            supported=sum(r is not None for r in regs.values());base=regs[42]
            valid=base is not None and supported>=2 and base['early_ss2_coverage']>=.8
            checks.append(dict(version=version,positive_seeds=[s for s,r in regs.items() if r],seed42_targets=base['targets'] if base else 0,seed42_coverage=base['early_ss2_coverage'] if base else None,eligible=valid))
            for a,b in ((42,43),(42,44),(43,44)):
                aa=set(regs[a]['gene2weight']) if regs[a] else set();bb=set(regs[b]['gene2weight']) if regs[b] else set()
                overlap.append(dict(TF=tf,version=version,seeds=[a,b],sizes=[len(aa),len(bb)],intersection=len(aa&bb),jaccard=len(aa&bb)/len(aa|bb) if aa|bb else None,signature_scope='Full canonical gene sets, self retained when present'))
            if choice is None and valid:choice=dict(origin='whole_early',seed=42,version=version,record=base,reason='Predeclared seed42 canonical signature; same-version positive support in at least 2/3 seeds and >=80% measured coverage')
        if choice is None:
            hits=[r for r in attempts['CVG_VGN/seed42/v10']['regulons'] if r['TF']==tf]
            if hits and hits[0]['early_ss2_coverage']>=.8:
                assert len(hits)==1
                if tf=='Neurod1':
                    with h5py.File(_RETRY_ACTIVITY,'r') as h:
                        old=[r for r in _retry_get(h['CVG_VGN/rank5000/regulons_json']) if r['transcription_factor']==tf]
                    assert len(old)==1 and old[0]['gene2weight']==hits[0]['gene2weight'],'Fallback must exactly reproduce the frozen original canonical signature'
                choice=dict(origin='CVG_VGN',seed=42,version='v10',record=hits[0],reason='Neither early version met the fixed replication rule; original CVG_VGN seed42 v10 canonical signature retained as transfer score')
        if choice is None:choice=dict(origin=None,seed=None,version=None,record=None,reason='No named motif-supported positive regulon in the paired v9/v10 resources; no GATA or RNA substitution')
        selection[tf]=dict(choice,early_version_checks=checks)
    return selection,dict(policy=B_RETRY_POLICY,selection=selection,pairwise_target_overlaps=overlap,attempt_summary={k:v['summary'] for k,v in attempts.items()}),resource_records


def _retry_check_score(result):
    with h5py.File(_RETRY_SHARED,'r') as h:
        np.testing.assert_array_equal(result['cell_ids'],h['shared/obs/_index'].asstr()[:])
        source=h['shared/obs/study_id'].asstr()[:]
    np.testing.assert_array_equal(result['source_scope'],np.isin(source,['GSE178931','GSE165502']))
    assert len(result['genes'])==20871 and result['raw_auc'].shape==(7355,2)
    assert list(result['TFs'])==['Neurod1','Zfpm2']
    assert np.isnan(result['raw_auc'][~result['source_scope']]).all()
    assert np.isnan(result['fitted'][~np.isfinite(result['raw_auc'])]).all()
    assert result['rank_cutoff']==1043
    for tf,a in result['activities'].items():
        if a['status']=='available':assert a['signature']['coverage']>=.8
        else:assert np.isnan(a['raw']).all() and np.isnan(a['fitted']).all() and not a['display_mask'].any()


def _retry_present_resources(records):
    # Cached redraw does not require downloaded feathers to be retained. Any
    # resource that IS present must still match its recorded byte hash.
    for rel,record in records.items():
        p=_RETRY_ROOT/rel
        if p.exists():assert _retry_sha(p)==record['sha256'],'A cached SCENIC resource changed; set RECOMPUTE_B_RETRY=True after review'


def _retry_commit(result,guard,token,selection,resources,new,runtime_snapshot):
    # Set this storage path independently before running. Activity.with-B-retry.pending.h5.
    pending=ROOT;shutil.copyfile(_RETRY_ACTIVITY,pending)
    with h5py.File(pending,'r+') as h:
        if new:
            if 'B_retry' in h:del h['B_retry']
            g=h.create_group('B_retry');g.attrs.update(schema='two_source_SCENIC_B_retry_v1',complete=True,guard_sha256=token)
            _retry_write(g,'guard',guard);_retry_write(g,'score',result);_retry_write(g,'selection',selection);_retry_write(g,'resource_records',resources)
            _retry_write(g,'sources',dict(B_RETRY_SOURCES,cell_source=_retry_cell_source()))
            _retry_write(g,'historical_append_audit',B_RETRY_BASELINE)
            archive=g.create_group('pruning')
            # Set this storage path independently before running. pruning.h5.
            with h5py.File(ROOT,'r') as s:
                for key in s:s.copy(key,archive,name=key)
            archive=g.create_group('grn');grn_sha={}
            for seed in (43,44):
                path=_RETRY_WORK/f'grn/seed{seed}.h5';dest=archive.create_group('seed'+str(seed));grn_sha[str(seed)]=_retry_sha(path)
                with h5py.File(path,'r') as s:
                    for k,v in s.attrs.items():dest.attrs[k]=v
                    for key in s:s.copy(key,dest,name=key)
            _retry_write(g,'GRN_file_sha256',grn_sha)
        else:
            g=h['B_retry'];_retry_write(g['score'],'figures',result['figures'])
        _retry_write(g,'render_provenance',dict(executed_cell_sha256=hashlib.sha256(_retry_cell_source().encode()).hexdigest(),source_sha256=guard['source_sha256'],outputs={tf:{k:v[k] for k in ('png_sha256','pdf_sha256')} for tf,v in result['figures'].items()},protected_outputs_unchanged=runtime_snapshot['files'],preexisting_codecells_sha256_at_start=runtime_snapshot['preexisting_codecells_sha256'],original_Activity_nodes_sha256=guard['base_Activity_nodes_sha256'],current_shared_nodes_sha256=guard['shared_nodes_sha256']))
        if 'integrity_json' in g:del g['integrity_json']
    assert _retry_nodes(pending)==runtime_snapshot['Activity_existing_logical_hashes'],'Append changed original Activity nodes'
    integrity=_retry_nodes(pending,'B_retry')
    with h5py.File(pending,'r+') as h:h['B_retry'].create_dataset('integrity_json',data=_retry_json(integrity),dtype=h5py.string_dtype('utf-8'))
    pending.replace(_RETRY_ACTIVITY)


def _run_b_retry():
    _RETRY_WORK.mkdir(parents=True,exist_ok=True)
    # Set this storage path independently before running.
    os.environ['MPLCONFIGDIR']=str(ROOT)
    os.environ['PYTHONDONTWRITEBYTECODE']='1';sys.dont_write_bytecode=True
    runtime_snapshot=_retry_runtime_snapshot();_retry_assert_protected(runtime_snapshot)
    guard,token=_retry_guard();cached=False
    with h5py.File(_RETRY_ACTIVITY,'r') as h:
        if 'B_retry' in h and not RECOMPUTE_B_RETRY:
            g=h['B_retry'];assert g.attrs.get('complete')
            assert g.attrs.get('guard_sha256')==token,'Retry inputs/source/parameters changed. Review, then set RECOMPUTE_B_RETRY=True.'
            expected=json.loads(g['integrity_json'].asstr()[()]);result=_retry_read(g['score']);selection=_retry_read(g['selection']);resources=_retry_read(g['resource_records']);cached=True
    if cached:
        assert _retry_nodes(_RETRY_ACTIVITY,'B_retry')==expected,'Stored B_retry payload failed its logical hash check'
        _retry_present_resources(resources)
        ns={'__name__':'scenic_b_retry_render'};exec(B_RETRY_SOURCES['render/B-only.py'],ns)
        with h5py.File(_RETRY_SHARED,'r') as h:xy=h['shared/X_umap'][:]
        result['figures']={tf:ns['render_b'](tf,result['activities'][tf],xy,_RETRY_FIGURES) for tf in ('Neurod1','Zfpm2')}
        # Set this storage path independently before running.
        ns['validate_figures'](result['figures'],ROOT)
    else:
        _retry_compute_attempts();selection,metadata,resources=_retry_select()
        signatures={tf:selection[tf]['record']['gene2weight'] if selection[tf]['record'] else None for tf in ('Neurod1','Zfpm2')}
        ns={'__name__':'scenic_b_retry_render'};exec(B_RETRY_SOURCES['render/B-only.py'],ns)
        # Set this storage path independently before running.
        result=ns['score_and_render'](signatures,metadata,project_root=_RETRY_ROOT,work_dir=ROOT,output_dir=_RETRY_FIGURES,shared_file=_RETRY_SHARED,scenic_python=_RETRY_PYTHON,rscript=_RETRY_R,validate=True)
        guard,token=_retry_guard()
    _retry_check_score(result);_retry_assert_protected(runtime_snapshot);_retry_commit(result,guard,token,selection,resources,not cached,runtime_snapshot);_retry_assert_protected(runtime_snapshot)
    print('B retry:',{tf:a['status'] for tf,a in result['activities'].items()},'| scored',int(result['rank_gate'].sum()),'cells; saved only /B_retry and two PNG/PDF pairs')
    try:
        from IPython.display import display,Image
        for tf in ('Neurod1','Zfpm2'):display(Image(filename=result['figures'][tf]['png']))
    except ImportError:pass
    return result

B_RETRY_RESULT = _run_b_retry()
```

## 2. Frozen inputs and normalization

Read integer `layers/counts` from the formal common-axis integration file, with the 7,355 cells, 11,467 genes, embedding, native segments and source flags frozen. Calculate float64 log10(1 + count / full endogenous total x 10^6). Inputs.h5 preserves the six native TF columns, missing-feature masks, prior motif/AUC data and reviewed descriptive title evidence.

E10.5 integer counts can be reconstructed from `Processed Data/03_stagewise_selected/E10.5_GSE231513__plate_1-plate_6__O2_O3_for_integration.h5ad`; its 73x6 native factor columns were independently identical to the archive. The transformed QC X and the similarly named continuous-integration object are not raw counts. Unmeasured source features, including Zfpm2 in the UMI source, remain NaN rather than zero.

```python
#1. =========== Shared imports and fixed analysis decisions
from pathlib import Path
import json
import re

import h5py
import networkx as nx
import numpy as np
import pandas as pd
from scipy import sparse
from scipy.stats import rankdata, spearmanr


EARLY_SOURCE = "GSE178931"
PATH_SEGMENTS = ("CVG", "T", "AC", "Ic")
MT_PATTERN = r"^mt[-.]"
RIBO_PATTERN = r"^(?:Rpl.*|Rps(?!6k).*)$"
TECH_PATTERN = r"^(?:ERCC[-_].*|Egfp|eGFP|GFP|Gfp|mCherry|tdTomato|Cre|CreERT2|Luc2|luciferase)$"
ASSOCIATION_RULES = {
    "min_library_cells": 30,
    "min_TF_positive_cells": 3,
    "min_gene_detection": 0.10,
    "min_global_abs_rho": 0.20,
    "min_mean_library_abs_rho": 0.15,
    "min_direction_fraction": 0.70,
    "min_valid_libraries_full": 4,
    "min_valid_libraries_training": 3,
    "max_genes": 200,
}


def _readframe(node):
    columns = json.loads(node.attrs["columns"]) if "columns" in node.attrs else list(node)
    table = pd.DataFrame({
        k: node[k].asstr()[:] if node[k].dtype.kind in "OSU" else node[k][:]
        for k in columns if k != "_index"
    })
    if "_index" in node:
        table.index = node["_index"].asstr()[:]
    return table


def _strings(node):
    if isinstance(node, h5py.Dataset):
        return node.asstr()[:]
    encoding = node.attrs.get("encoding-type", "")
    if encoding == "nullable-string-array":
        result = node["values"].asstr()[:]
        if np.asarray(node["mask"][:], bool).any():
            raise ValueError("Missing gene symbols are not permitted on the fixed gene axis")
        return result
    if encoding == "categorical":
        codes = node["codes"][:]
        if (codes < 0).any():
            raise ValueError("Missing categorical gene symbols")
        return node["categories"].asstr()[:][codes]
    raise ValueError(f"Unsupported string encoding: {encoding}")


def _flags(values):
    a = np.asarray(values)
    if a.dtype.kind == "b":
        return a
    if a.dtype.kind in "iu":
        if not np.isin(a, [0, 1]).all():
            raise ValueError("Flag columns must contain booleans or 0/1")
        return a.astype(bool)
    text = pd.Series(a).astype(str).str.lower()
    if not text.isin(["true", "false", "1", "0"]).all():
        raise ValueError("Unrecognized boolean flag value")
    return text.isin(["true", "1"]).to_numpy()


def read_common_expression(inputs_path, rows=None, columns=None):
    """Read the formal CSR counts and return native-denominator log10(1+CPM).

    Rows and columns refer to the fixed axes stored in Inputs.h5. Neither the
    common-gene count sum nor a selected-gene sum replaces library_total.
    """
    inputs_path = Path(inputs_path).resolve()
    with h5py.File(inputs_path, "r") as h:
        genes = h["genes"].asstr()[:]
        cells = h["cell_ids"].asstr()[:]
        totals = h["library_total"][:]
        source = Path(h.attrs["common_counts_file"])
    if not source.is_absolute():
        source = inputs_path.parents[2] / source
    row_ids = np.arange(len(cells)) if rows is None else np.arange(len(cells))[rows]
    col_ids = np.arange(len(genes)) if columns is None else np.arange(len(genes))[columns]
    row_ids, col_ids = np.atleast_1d(row_ids), np.atleast_1d(col_ids)
    with h5py.File(source, "r") as h:
        obs_key = h["obs"].attrs.get("_index", "_index")
        np.testing.assert_array_equal(_strings(h["obs"][obs_key]), cells)
        np.testing.assert_array_equal(_strings(h["var/gene_symbol"]), genes)
        node = h["layers/counts"]
        if node.attrs.get("encoding-type") != "csr_matrix":
            raise ValueError("Formal layers/counts must be CSR")
        shape = tuple(int(v) for v in node.attrs["shape"])
        if shape != (len(cells), len(genes)):
            raise ValueError("Formal count shape differs from the frozen input axes")
        counts = sparse.csr_matrix(
            (node["data"][:], node["indices"][:], node["indptr"][:]), shape=shape
        )
        values = counts[row_ids][:, col_ids].toarray().astype(np.float64)
    if not np.isfinite(values).all() or (values < 0).any() or (totals[row_ids] <= 0).any():
        raise ValueError("Invalid counts or native library totals")
    return np.log10(1 + values / totals[row_ids, None] * 1e6)
```

## 3. Signed association screening

Discovery uses 1,161 GSE178931 cells at E9.5/E11.5/E13.5. Require |global Spearman rho|>=.20, |mean library rho|>=.15, >=70% direction consistency, >=10% detection, and >=4 evaluable libraries. Retain <=200 genes; exclude self, mitochondrial, structural ribosomal, and technical genes. Every holdout repeats training-only selection.

```python
#2. =========== Exact signed Spearman association selection
def _normalized_ranks(values):
    ranks = rankdata(values, axis=0, method="average", nan_policy="omit")
    finite_count = np.isfinite(ranks).sum(axis=0)
    means = np.divide(np.nansum(ranks, axis=0), finite_count,
                      out=np.zeros(ranks.shape[1]), where=finite_count > 0)
    ranks -= means
    norms = np.sqrt(np.nansum(ranks * ranks, axis=0))
    result = np.full_like(ranks, np.nan)
    np.divide(ranks, norms, out=result, where=norms > 0)
    return result


def _correlations(gene_values, tf_values):
    return _normalized_ranks(tf_values).T @ _normalized_ranks(gene_values)


def _exclusions(genes, factor):
    return {
        "excluded_mt": np.array([bool(re.search(MT_PATTERN, g, re.I)) for g in genes]),
        "excluded_ribosomal": np.array([bool(re.fullmatch(RIBO_PATTERN, g)) for g in genes]),
        "excluded_technical": np.array([bool(re.fullmatch(TECH_PATTERN, g, re.I)) for g in genes]),
        "self_gene": genes == factor,
    }


def _choose(genes, global_rho, per_library, detection, excluded, minimum_libraries):
    rules = ASSOCIATION_RULES
    valid = np.isfinite(per_library)
    n_valid = valid.sum(axis=0)
    mean = np.divide(np.nansum(per_library, axis=0), n_valid,
                     out=np.full_like(global_rho, np.nan), where=n_valid > 0)
    agreement = np.divide((valid & (per_library * mean[None] > 0)).sum(axis=0), n_valid,
                          out=np.full_like(global_rho, np.nan), where=n_valid > 0)
    tests = {
        "passes_detection": detection >= rules["min_gene_detection"],
        "passes_global_abs_rho": np.abs(global_rho) >= rules["min_global_abs_rho"],
        "passes_mean_library_abs_rho": np.abs(mean) >= rules["min_mean_library_abs_rho"],
        "passes_global_mean_direction": global_rho * mean > 0,
        "passes_direction_fraction": agreement >= rules["min_direction_fraction"],
        "passes_valid_libraries": n_valid >= minimum_libraries,
    }
    passed = ~np.logical_or.reduce(list(excluded.values()))
    for value in tests.values():
        passed &= value
    candidates = np.flatnonzero(passed)
    order = np.lexsort((genes[candidates], -np.abs(global_rho[candidates]), -np.abs(mean[candidates])))
    chosen = candidates[order[:rules["max_genes"]]]
    selected = np.zeros(len(genes), dtype=bool)
    selected[chosen] = True
    rank = np.zeros(len(genes), dtype=np.int16)
    rank[chosen] = np.arange(1, len(chosen) + 1)
    return dict(global_rho=global_rho, mean_rho=mean, valid_libraries=n_valid,
                direction_fraction=agreement, detection=detection,
                passed=passed, selected=selected, selection_rank=rank, **tests)


def _audit(genes, factor, result, excluded, analysis_pool, fold=-1):
    table = pd.DataFrame({"TF": factor, "fold": fold, "gene": genes,
        "global_rho": result["global_rho"], "mean_library_rho": result["mean_rho"],
        "valid_libraries": result["valid_libraries"],
        "direction_fraction": result["direction_fraction"],
        "early_detection": result["detection"], "passed_pool": result["passed"],
        "association_selected": result["selected"], "selection_rank": result["selection_rank"],
        "selected_for_analysis": np.isin(genes, analysis_pool),
        "RNA_association_sign": np.sign(np.nan_to_num(result["mean_rho"], nan=0.0)).astype(np.int8),
        **excluded, **{k: v for k, v in result.items() if k.startswith("passes_")}})
    return table
```

## 4. Natural dynamic modules and seven early holdouts

Use source/native-segment/pseudotime-third bins, equal-library means, source-wise centering/L2 scaling, a correlation/Jaccard graph and Louvain seed42. Keep every module with >=8 genes, display the largest two, and retain training-only genes in every holdout. Nhlh1 uses its frozen 46-gene common-axis motif-supported set and prior AUC, rather than its separate RNA-associated pool. Prox1 failed full-data selection; occasional 1-2-gene training-fold sets are retained as audit evidence and do not become a C program.

```python
#3. =========== Frozen-tree, source- and library-balanced dynamic profiles
def _profile_design(obs, good):
    n_bins, minimum_cells = 3, 3
    specs = []
    for source in sorted(obs.study_id.unique()):
        for seg in PATH_SEGMENTS:
            rows = np.flatnonzero(good & obs.study_id.eq(source).to_numpy()
                                 & obs.seg.eq(seg).to_numpy())
            if len(rows) < minimum_cells:
                continue
            low, high = obs.t.to_numpy()[rows].min(), obs.t.to_numpy()[rows].max()
            bins = np.clip(((obs.t.to_numpy()[rows] - low)
                / max(high - low, 1e-12) * n_bins).astype(int), 0, n_bins - 1)
            for b in range(n_bins):
                selected = rows[bins == b]
                if len(selected) >= minimum_cells:
                    specs.append({"source": source, "seg": seg, "bin": b, "rows": selected})
    return specs


def _profiles(raw, obs, specs, omit=None):
    minimum_cells = 3
    blocks = []
    libraries = obs.library_id.to_numpy()
    for source in sorted({s["source"] for s in specs}):
        columns = []
        for spec in specs:
            if spec["source"] != source:
                continue
            rows = spec["rows"]
            if omit is not None:
                rows = rows[libraries[rows] != omit]
            if len(rows) < minimum_cells:
                continue
            means = [raw[r].mean(axis=0) for lib in sorted(np.unique(libraries[rows]))
                     if len(r := rows[libraries[rows] == lib]) >= minimum_cells]
            if means:
                columns.append(np.mean(means, axis=0))
        if len(columns) < 2:
            continue
        block = np.asarray(columns).T
        block -= block.mean(axis=1, keepdims=True)
        block /= np.maximum(np.linalg.norm(block, axis=1, keepdims=True), 1e-12)
        blocks.append(block)
    if not blocks:
        raise ValueError("No evaluable dynamic profile blocks")
    return np.hstack(blocks) / np.sqrt(len(blocks))


def _cluster_profiles(patterns, seed=42, resolution=1.0):
    k_neighbors, minimum_module = 15, 8
    z = np.asarray(patterns, float).copy()
    z -= z.mean(axis=1, keepdims=True)
    lengths = np.linalg.norm(z, axis=1)
    valid = np.flatnonzero(lengths > 1e-10)
    if len(valid) < minimum_module:
        return []
    z = z[valid] / lengths[valid, None]
    similarity = z @ z.T
    np.fill_diagonal(similarity, -2)
    k = min(k_neighbors, len(z) - 1)
    nearest = np.argsort(-similarity, axis=1, kind="stable")[:, :k]
    neighbors = [set(row) for row in nearest]
    graph = nx.Graph()
    graph.add_nodes_from(range(len(z)))
    for i in range(len(z)):
        for j in nearest[i]:
            j = int(j)
            if graph.has_edge(i, j) or similarity[i, j] <= 0:
                continue
            shared = len(neighbors[i] & neighbors[j])
            if shared:
                graph.add_edge(i, j, weight=shared / (2 * k - shared) * float(similarity[i, j]))
    groups = nx.community.louvain_communities(
        graph, weight="weight", resolution=resolution, seed=seed)
    return sorted([valid[sorted(g)] for g in groups if len(g) >= minimum_module],
                  key=lambda g: (-len(g), int(g.min())))


def _closest_set(target, options):
    a = set(target)
    if not options:
        return 0.0, []
    scored = [(len(a & set(b)) / len(a | set(b)), list(b)) for b in options]
    return max(scored, key=lambda value: (value[0], len(value[1])))
```

```python
#4. =========== Independently analyze one candidate from the immutable inputs
def analyze_factor(inputs_path, factor, seed=42):
    """Return arrays, DataFrames and module definitions without saving any results.

    Association discovery is repeated from native early RNA, including all seven
    training folds. Nhlh1 instead uses the frozen motif-supported gene membership;
    its fold checks address module partitioning, not independent motif discovery.
    No trajectory, embedding, GRN, enrichment analysis or GAM is refitted here.
    """
    inputs_path = Path(inputs_path).resolve()
    with h5py.File(inputs_path, "r") as h:
        genes = h["genes"].asstr()[:]
        cell_ids = h["cell_ids"].asstr()[:]
        obs = _readframe(h["obs"])
        native_tfs = h["native_TFs"].asstr()[:]
        matches = np.flatnonzero(native_tfs == factor)
        if len(matches) != 1:
            raise ValueError(f"Native TF expression is unavailable or ambiguous: {factor}")
        j = int(matches[0])
        native_counts = h["native_counts"][:, j]
        native_rna_matrix = h["native_RNA"][:]
        native_rna = native_rna_matrix[:, j]
        native_measured = h["native_measured"][:, j].astype(bool)
        totals = h["library_total"][:]
        motif_sets = json.loads(h["early_motif_sets_json"].asstr()[()])
        nhlh1_auc = h["nhlh1_auc"][:]
        xy, fork_t = h["X_umap"][:], float(h["fork_t"][()])
        reviewed = {name: json.loads(h[name].asstr()[()]) for name in
                    ("reviewed_program_titles_json", "reviewed_title_evidence_json") if name in h}
    np.testing.assert_array_equal(obs.index.astype(str), cell_ids)
    if "cell_id" in obs:
        np.testing.assert_array_equal(obs.cell_id, cell_ids)
    if len(cell_ids) != 7355 or len(genes) != 11467 or not pd.Index(genes).is_unique:
        raise ValueError("The fixed 7,355-cell / 11,467-symbol axes changed")
    np.testing.assert_array_equal(native_counts >= 0, native_measured)
    if not np.isnan(native_rna[~native_measured]).all():
        raise ValueError("Unmeasured TF RNA must remain NaN")
    expected = np.log10(1 + native_counts[native_measured] / totals[native_measured] * 1e6)
    np.testing.assert_allclose(native_rna[native_measured], expected, rtol=0, atol=1e-12)
    if not np.isfinite(obs.t).all() or xy.shape != (len(obs), 2):
        raise ValueError("Invalid frozen trajectory or embedding")
    good = (_flags(obs.source_state_evaluation_allowed)
            & ~_flags(obs.source_rna_mixture_review_flag))
    primary = good & obs.seg.isin(PATH_SEGMENTS).to_numpy()
    early_rows = np.flatnonzero(obs.study_id.eq(EARLY_SOURCE))
    early = obs.iloc[early_rows].reset_index(drop=True)
    if len(early) != 1161 or set(early.stage_standardized) != {"E9.5", "E11.5", "E13.5"}:
        raise ValueError("Early discovery scope changed")
    x_early = read_common_expression(inputs_path, rows=early_rows)
    # Preserve the original six-TF matrix multiplication. A single-row BLAS
    # kernel can change the sign of numerically zero rho for tied expressions.
    tf_early = native_rna_matrix[early_rows]
    count_early, measured_early = native_counts[early_rows], native_measured[early_rows]
    if not (measured_early.all() or not measured_early.any()):
        raise ValueError("Mixed measured/unmeasured early TF observations are unsupported")
    library_rows, library_records, library_rhos = [], [], []
    for lib, positions in early[early.stage_standardized.isin(["E11.5", "E13.5"])].groupby("library_id", sort=True).groups.items():
        rows = np.asarray(positions, int)
        if len(rows) < ASSOCIATION_RULES["min_library_cells"]:
            continue
        positives = int((count_early[rows] > 0).sum())
        eligible = bool(measured_early[rows].all()
                        and positives >= ASSOCIATION_RULES["min_TF_positive_cells"])
        library_rows.append(rows)
        library_records.append({"library_id": lib, "stage": early.stage_standardized.iloc[rows[0]],
            "cells": len(rows), "TF_positive_cells": positives, "TF_evaluable": eligible})
        library_rhos.append(_correlations(x_early[rows], tf_early[rows])[j]
                            if eligible else np.full(len(genes), np.nan))
    if len(library_rows) != 7:
        raise ValueError("Expected seven eligible-size early libraries")
    library_rho = np.stack(library_rhos)
    global_rho = _correlations(x_early, tf_early)[j]
    detection = (x_early > 0).mean(axis=0)
    excluded = _exclusions(genes, factor)
    full = _choose(genes, global_rho, library_rho, detection, excluded,
                   ASSOCIATION_RULES["min_valid_libraries_full"])
    finite = np.flatnonzero(np.isfinite(global_rho))
    scipy_checks = sorted(set([0, len(genes) // 2, len(genes) - 1]
        + ([int(finite[np.argmax(np.abs(global_rho[finite]))])] if len(finite) else [])))
    if len(finite):
        for k in scipy_checks:
            reference = spearmanr(tf_early[:, j], x_early[:, k]).statistic
            np.testing.assert_allclose(global_rho[k], reference, rtol=0, atol=1e-12, equal_nan=True)

    frozen = factor == "Nhlh1"
    missing_frozen = []
    if frozen:
        weights = motif_sets[factor]["activating_motif"]["gene2weight"]
        missing_frozen = [g for g in weights if g not in set(genes)]
        pool = [g for g in weights if g in set(genes) and g != factor]
        positive = list(pool)
        kind, metric = "early_motif_supported", "Regulon AUC"
    else:
        chosen = np.flatnonzero(full["selected"])
        chosen = chosen[np.argsort(full["selection_rank"][chosen])]
        pool = genes[chosen].tolist()
        positive = genes[chosen[full["mean_rho"][chosen] > 0]].tolist()
        kind = "early_associated" if pool else "unavailable"
        metric = "Associated expression score" if positive else ("TF RNA expression" if factor == "Prox1" else "Unavailable")
    full_audit = _audit(genes, factor, full, excluded, pool)
    folds, fold_results, loo_gene_rows = [], [], []
    for fold_id, (heldout, record) in enumerate(zip(library_rows, library_records)):
        train = np.ones(len(early), bool)
        train[heldout] = False
        result = _choose(genes, _correlations(x_early[train], tf_early[train])[j],
            library_rho[np.arange(7) != fold_id], (x_early[train] > 0).mean(axis=0),
            excluded, ASSOCIATION_RULES["min_valid_libraries_training"])
        # Original BC uses the native gene-axis order for a reselected training pool.
        train_genes = list(pool) if frozen else genes[result["selected"]].tolist()
        folds.append({"fold": fold_id, **record, "training_cells": int(train.sum()),
                      "training_E11_E13_libraries": 6, "E9_single_library_retained": True})
        fold_results.append({"fold": fold_id, "heldout_library": record["library_id"],
            "selection": result, "pool_genes": train_genes,
            "gene_audit": _audit(genes, factor, result, excluded, train_genes, fold_id)})
        for k in np.flatnonzero(result["selected"]):
            heldout_rho = library_rho[fold_id, k]
            loo_gene_rows.append({"TF": factor, "fold": fold_id, "gene": genes[k],
                "heldout_library": record["library_id"],
                "train_rank": int(result["selection_rank"][k]),
                "train_global_rho": result["global_rho"][k],
                "train_mean_library_rho": result["mean_rho"][k],
                "heldout_rho": heldout_rho, "heldout_evaluable": bool(np.isfinite(heldout_rho)),
                "heldout_same_direction": bool(np.isfinite(heldout_rho)
                    and heldout_rho * result["mean_rho"][k] > 0)})
    full_audit["LOO_association_selected_folds"] = np.stack(
        [f["selection"]["selected"] for f in fold_results]).sum(axis=0)
    del x_early

    union = sorted(set(pool).union(*(set(f["pool_genes"]) for f in fold_results)))
    union_genes = np.asarray(union, dtype=str)
    gene_lookup = pd.Index(union_genes)
    positions = pd.Index(genes).get_indexer(union_genes)
    if (positions < 0).any():
        raise ValueError("A selected training gene is absent from the input gene axis")
    raw = read_common_expression(inputs_path, columns=positions)
    # Match BC's advanced column indexing, preserving the sorted union axis.
    raw = raw[:, np.arange(len(union_genes))]
    minimum = raw.min(axis=0)
    span = raw.max(axis=0) - minimum
    if (span <= 0).any():
        raise ValueError("A selected gene is constant across the complete atlas")
    scaled = (raw - minimum) / span
    specs = _profile_design(obs, good)
    features = _profiles(raw, obs, specs)
    cols = gene_lookup.get_indexer(pool)
    groups = _cluster_profiles(features[cols], seed=seed) if len(cols) else []
    variants = [(seed + 1, 1.0), (seed + 2, 1.0), (seed, 0.8), (seed, 1.2)]
    variant_groups = [_cluster_profiles(features[cols], seed=s, resolution=r)
                      if len(cols) else [] for s, r in variants]
    modules, module_score_columns, sensitivity = [], [], []
    pool_module = {g: -1 for g in pool}
    for module_id, local in enumerate(groups):
        indices = cols[local]
        members = union_genes[indices].tolist()
        stability = min(_closest_set(local, variant)[0] for variant in variant_groups)
        modules.append({"module_id": module_id, "TF": factor, "community": module_id,
            "genes": members, "n_genes": len(members), "kind": kind,
            "display": module_id < 2, "graph_sensitivity_min_jaccard": stability})
        module_score_columns.append(scaled[:, indices].mean(axis=1))
        pool_module.update({g: module_id for g in members})
        for variant_id, ((s, r), variant) in enumerate(zip(variants, variant_groups)):
            jac, matched = _closest_set(local, variant)
            sensitivity.append({"TF": factor, "module_id": module_id, "variant": variant_id,
                "seed": s, "resolution": r, "matched_jaccard": jac,
                "matched_genes": union_genes[cols[np.asarray(matched, int)]].tolist()})
    module_scores = np.column_stack(module_score_columns) if modules else np.empty((len(obs), 0))
    displayed = [m["module_id"] for m in modules if m["display"]]
    if frozen:
        b_raw = np.asarray(nhlh1_auc, float).reshape(-1)
        if b_raw.shape != (len(obs),) or not np.isfinite(b_raw).all():
            raise ValueError("Frozen Nhlh1 AUC is not aligned/finite")
    elif positive:
        b_raw = scaled[:, gene_lookup.get_indexer(positive)].mean(axis=1)
    elif factor == "Prox1":
        # RNA reference only: failed full-data program selection remains failed.
        b_raw = native_rna.copy()
    else:
        b_raw = np.full(len(obs), np.nan)
    jobs, job_values = [], []
    if np.isfinite(b_raw).all():
        jobs.append({"job": 0, "mode": "B", "TF": factor, "module_id": -1, "omit_library": ""})
        job_values.append(b_raw)
    for module in modules:
        if module["display"]:
            module["job"] = len(jobs)
            jobs.append({"job": len(jobs), "mode": "C", "TF": factor,
                         "module_id": module["module_id"], "omit_library": ""})
            job_values.append(module_scores[:, module["module_id"]])

    validations = []
    for fold in fold_results:
        lib, fold_id = fold["heldout_library"], fold["fold"]
        train_cells = good & ~obs.library_id.eq(lib).to_numpy()
        train_min = raw[train_cells].min(axis=0)
        train_span = raw[train_cells].max(axis=0) - train_min
        fold_features = _profiles(raw, obs, specs, omit=lib)
        train_cols = gene_lookup.get_indexer(fold["pool_genes"])
        local = _cluster_profiles(fold_features[train_cols], seed=seed) if len(train_cols) else []
        named = [union_genes[train_cols[g]].tolist() for g in local]
        fold.update({"modules": named, "train_gene_min": train_min, "train_gene_span": train_span,
                     "profile_features": fold_features[train_cols]})
        for module_id in displayed:
            jac, matched = _closest_set(modules[module_id]["genes"], named)
            row = {"TF": factor, "module_id": module_id, "fold": fold_id,
                "heldout_library": lib, "training_pool_genes": len(fold["pool_genes"]),
                "requested_training_pool_genes": len(fold["pool_genes"]),
                "training_module_genes": len(matched), "membership_jaccard": jac,
                "target_scope": ("Fixed full-data motif targets; validation concerns module memberships only"
                    if frozen else "Gene associations reselected without heldout library"),
                "job": -1, "training_module_gene_list": json.dumps(matched)}
            if matched:
                c = gene_lookup.get_indexer(matched)
                c = c[train_span[c] > 0]
                row["scored_training_module_genes"] = len(c)
                if len(c) >= 8:
                    value = ((raw[:, c] - train_min[c]) / train_span[c]).mean(axis=1)
                    row["job"] = len(jobs)
                    jobs.append({"job": len(jobs), "mode": "C", "TF": factor,
                                 "module_id": module_id, "omit_library": lib})
                    job_values.append(value)
            else:
                row["scored_training_module_genes"] = 0
            validations.append(row)
    jobs = pd.DataFrame(jobs, columns=["job", "mode", "TF", "module_id", "omit_library"])
    job_values = np.column_stack(job_values) if job_values else np.empty((len(obs), 0))
    full_audit["dynamic_module_id"] = full_audit.gene.map(pool_module).fillna(-1).astype(int)
    full_audit["in_qualifying_dynamic_module"] = full_audit.dynamic_module_id >= 0
    association_selected = np.flatnonzero(full["selected"])
    summary = {"TF": factor, "kind": kind, "association_pool_genes": int(full["passed"].sum()),
        "association_selected_genes": len(association_selected), "analysis_pool_genes": len(pool),
        "positive_genes_for_B": len(positive), "module_sizes": [len(g) for g in groups],
        "display_module_ids": displayed, "training_union_genes": len(union),
        "all_discovery_libraries": len(library_rows), "early_cells": len(early), "folds": len(folds)}
    print(f"{factor}: pool={len(pool)}, modules={summary['module_sizes']}, folds={len(folds)}", flush=True)
    return {
        "factor": factor, "seed": seed, "summary": summary,
        "genes": union_genes, "common_genes": genes, "cell_ids": cell_ids, "obs": obs,
        "RNA": raw, "gene_min": minimum, "gene_span": span, "good": good,
        "primary_path": primary, "X_umap": xy, "fork_t": fork_t,
        "native_RNA": native_rna, "native_counts": native_counts, "native_measured": native_measured,
        "analysis_pool": pool, "positive_genes": positive,
        "association_audit": full_audit, "association_selection": full,
        "association_library_rho": library_rho, "association_libraries": pd.DataFrame(library_records),
        "association_LOO_evidence": pd.DataFrame(loo_gene_rows),
        "modules": modules, "module_scores": module_scores, "selected_modules": displayed,
        "profile_features": features, "profile_design": specs,
        "sensitivity": sensitivity,
        "sensitivity_modules": [{"seed": s, "resolution": r,
            "modules": [union_genes[cols[g]].tolist() for g in variant]}
            for (s, r), variant in zip(variants, variant_groups)],
        "B_raw": b_raw, "jobs": jobs, "job_values": job_values,
        "validation": pd.DataFrame(validations), "folds": pd.DataFrame(folds), "LOO": fold_results,
        "reviewed_annotations": reviewed,
        "methods": {"source": EARLY_SOURCE, "association_rules": dict(ASSOCIATION_RULES),
            "expression": "log10(1+formal counts/full native endogenous total*1e6)",
            "gene_exclusions": {"mt": MT_PATTERN, "ribosomal_structural": RIBO_PATTERN, "technical": TECH_PATTERN},
            "analysis_kind": kind, "B_metric": metric, "frozen_motif_gene_set": frozen,
            "frozen_targets_absent_from_common_axis": missing_frozen,
            "profile_path": PATH_SEGMENTS, "bins_per_segment": 3, "min_cells_per_library_bin": 3,
            "profile_scaling": "Center/L2 within each source, concatenate with equal source block weight",
            "graph": "Correlation kNN15, shared-neighbor Jaccard times positive correlation, Louvain min8",
            "display_selection": "Two largest qualifying modules; all qualifying modules retained",
            "normalization_full": "Gene-wise minmax over all atlas cells, then unweighted mean",
            "normalization_LOO": "Good training cells only; heldout ranges are not clipped",
            "limits": ["Associated genes are not validated direct targets or TF activation",
                "The tree, profile bins and TF preselection are fixed",
                "Module matching to the exploratory full-data modules is not independent discovery",
                "Nhlh1 motif membership and AUC are frozen; motif discovery is not revalidated"]},
        "checks": {"native_TF_formula_tolerance": 1e-12, "Spearman_scipy_tolerance": 1e-12,
            "common_axes_exact": True, "training_only_genes_preserved": True,
            "no_GRN_refit": True, "no_tree_refit": True, "no_result_files_written": True},
    }
```

## 5. Independent late Prox1 check

Apply the unchanged source-specific correlation, detection and library-consistency gates to 5,728 good cells from GSE165502/GSE195500, with 22 libraries and eight source-stage holdouts. This is a separate late-stage sensitivity audit, not an early regulon or a substitute program.

```python
"""Recompute the fixed late Prox1 association audit from formal inputs.

This cell uses the notebook's read_common_expression and _correlations helpers.
It performs no file writes, plotting, trajectory fitting, or threshold tuning.
"""

import hashlib
import json
import re
from pathlib import Path as FilePath

import h5py
import numpy as np
import pandas as pd


def validate_late_prox1(inputs_path):
    """Return complete full-data and source-stage library-holdout evidence."""
    #1. =========== Fixed scope and association criteria
    sources = ('GSE165502', 'GSE195500')
    tf = 'Prox1'
    min_library_cells, min_tf_positive_cells = 30, 3
    min_detection, min_source_rho, min_library_rho = .10, .20, .15
    min_direction_fraction, min_valid_libraries, max_selected = .70, 4, 200
    exclusion_patterns = {
        'mt': r'^mt[-.]',
        'ribosomal_structural': r'^(?:Rpl.*|Rps(?!6k).*)$',
        'technical': r'^(?:ERCC[-_].*|Egfp|eGFP|GFP|Gfp|mCherry|tdTomato|Cre|CreERT2|Luc2|luciferase)$',
    }
    inputs_path = FilePath(inputs_path).resolve()

    def file_sha256(path):
        value = hashlib.sha256()
        with FilePath(path).open('rb') as stream:
            for block in iter(lambda: stream.read(8 * 1024 * 1024), b''):
                value.update(block)
        return value.hexdigest()

    #2. =========== Load native TF data and the common expression matrix once
    input_hash = file_sha256(inputs_path)
    with h5py.File(inputs_path, 'r') as handle:
        genes = handle['genes'].asstr()[:]
        cell_ids = handle['cell_ids'].asstr()[:]
        obs = pd.DataFrame({key: (node.asstr()[:] if node.dtype.kind in 'OSU'
                                  else node[:])
                            for key, node in handle['obs'].items() if key != '_index'})
        obs['cell_id'] = cell_ids
        np.testing.assert_array_equal(handle['obs/_index'].asstr()[:], cell_ids)
        allowed = obs.source_state_evaluation_allowed.astype(str).str.lower().isin(['true', '1'])
        mixture = obs.source_rna_mixture_review_flag.astype(str).str.lower().isin(['true', '1'])
        rows = np.flatnonzero(obs.study_id.isin(sources) & allowed & ~mixture)
        late = obs.iloc[rows].reset_index(drop=True)
        tf_index = list(handle['native_TFs'].asstr()[:]).index(tf)
        counts = handle['native_counts'][rows, tf_index]
        measured = handle['native_measured'][rows, tf_index]
        tf_values = handle['native_RNA'][rows, tf_index][:, None]
        totals = handle['library_total'][rows]
        formal_counts = FilePath(handle.attrs['common_counts_file'])
        provenance = json.loads(handle['provenance_json'].asstr()[()])
    if not formal_counts.is_absolute():
        formal_counts = inputs_path.parents[2] / formal_counts
    counts_hash = file_sha256(formal_counts)
    if counts_hash != provenance['source_sha256']['counts']:
        raise ValueError('Formal common-count file differs from Inputs provenance.')
    expression = read_common_expression(inputs_path, rows=rows)
    np.testing.assert_array_equal(totals, late.library_total)
    np.testing.assert_array_equal(counts >= 0, measured)
    np.testing.assert_allclose(tf_values[:, 0][measured],
        np.log10(1 + counts[measured] / totals[measured] * 1e6), rtol=0, atol=1e-12)
    if not np.isnan(tf_values[:, 0][~measured]).all():
        raise ValueError('Unmeasured native TF values must remain NaN.')
    if expression.shape != (5728, 11467) or not np.isfinite(expression).all():
        raise ValueError('The fixed late expression scope must be 5728 x 11467.')
    source_rows = [np.flatnonzero(late.study_id.eq(source)) for source in sources]
    if list(map(len, source_rows)) != [1112, 4616]:
        raise ValueError('The fixed late source cell counts differ.')
    excluded = np.array([any(
        re.search(pattern, gene, re.I) if key == 'mt' else
        re.fullmatch(pattern, gene, 0 if key == 'ribosomal_structural' else re.I)
        for key, pattern in exclusion_patterns.items()) for gene in genes], dtype=bool)
    permitted = ~excluded & (genes != tf)

    #3. =========== Source-specific global and independent within-library ranks
    source_rho = np.stack([_correlations(expression[r], tf_values[r])[0]
                           for r in source_rows])
    source_detection = np.stack([(expression[r] > 0).mean(axis=0) for r in source_rows])
    library_rows, library_records, library_rho = [], [], []
    grouping = late.groupby(['study_id', 'stage_standardized', 'library_id'], sort=True)
    for (source, stage, library), positions in grouping.groups.items():
        r = np.asarray(positions, dtype=int)
        if len(r) < min_library_cells:
            continue
        evaluable = bool(measured[r].all() and (counts[r] > 0).sum() >= min_tf_positive_cells)
        rho = (_correlations(expression[r], tf_values[r])[0] if evaluable
               else np.full(len(genes), np.nan))
        library_rows.append(r)
        library_records.append({'study': source, 'stage': stage, 'library_id': library,
            'cells': len(r), 'TF_positive_cells': int((counts[r] > 0).sum()),
            'TF_evaluable': evaluable})
        library_rho.append(rho)
    libraries = pd.DataFrame(library_records)
    library_rho = np.stack(library_rho)
    if len(libraries) != 22:
        raise ValueError('The fixed late audit must contain 22 eligible-size libraries.')

    def choose(sr, per_library, detection):
        valid = np.isfinite(per_library).sum(axis=0)
        mean = np.divide(np.nansum(per_library, axis=0), valid,
            out=np.full(len(genes), np.nan), where=valid > 0)
        agreement = np.divide(
            (np.isfinite(per_library) & (per_library * mean > 0)).sum(axis=0), valid,
            out=np.full(len(genes), np.nan), where=valid > 0)
        conservative = np.sign(sr[0]) * np.min(np.abs(sr), axis=0)
        passed = (permitted & (detection.min(axis=0) >= min_detection)
            & (np.abs(sr) >= min_source_rho).all(axis=0) & (sr[0] * sr[1] > 0)
            & (conservative * mean > 0) & (np.abs(mean) >= min_library_rho)
            & (agreement >= min_direction_fraction) & (valid >= min_valid_libraries))
        candidates = np.flatnonzero(passed)
        order = np.lexsort((genes[candidates], -np.abs(conservative[candidates]),
                            -np.abs(mean[candidates])))
        chosen = candidates[order[:max_selected]]
        selected = np.zeros(len(genes), dtype=bool)
        selected[chosen] = True
        rank = np.zeros(len(genes), dtype=np.int16)
        rank[chosen] = np.arange(1, len(chosen) + 1)
        return {'global_rho': conservative, 'mean_rho': mean,
                'valid_libraries': valid, 'direction_fraction': agreement,
                'passed': passed, 'selected': selected, 'rank': rank}

    full = choose(source_rho, library_rho, source_detection)

    #4. =========== Reselect after omitting one whole library per source-stage
    fold_ids = libraries.groupby(['study', 'stage'], sort=True).head(1).index.to_numpy()
    if len(fold_ids) != 8:
        raise ValueError('The fixed audit must contain eight source-stage holdouts.')
    folds, train_values, validation = [], [], []
    for fold, library_index in enumerate(fold_ids):
        record = library_records[library_index]
        heldout = library_rows[library_index]
        train_mask = np.ones(len(late), dtype=bool)
        train_mask[heldout] = False
        train_source_rho, train_detection = source_rho.copy(), source_detection.copy()
        source_index = sources.index(record['study'])
        r = source_rows[source_index]
        r = r[train_mask[r]]
        train_source_rho[source_index] = _correlations(expression[r], tf_values[r])[0]
        train_detection[source_index] = (expression[r] > 0).mean(axis=0)
        keep_libraries = np.arange(len(libraries)) != library_index
        result = choose(train_source_rho, library_rho[keep_libraries], train_detection)
        folds.append({'fold': fold, **record, 'training_cells': int(train_mask.sum()),
                      'training_libraries': int(keep_libraries.sum())})
        train_values.append({'source_rho': train_source_rho,
                             'source_detection': train_detection, **result})
        chosen = np.flatnonzero(result['selected'])
        chosen = chosen[np.argsort(result['rank'][chosen])]
        for k in chosen:
            rho = library_rho[library_index, k]
            validation.append({'fold': fold, 'TF': tf, 'gene': genes[k],
                'heldout_source': record['study'], 'heldout_stage': record['stage'],
                'heldout_library': record['library_id'], 'train_rank': int(result['rank'][k]),
                'train_global_rho_min_abs': result['global_rho'][k],
                'train_mean_library_rho': result['mean_rho'][k],
                'train_direction_fraction': result['direction_fraction'][k],
                'train_valid_libraries': int(result['valid_libraries'][k]),
                'heldout_rho': rho, 'heldout_evaluable': bool(np.isfinite(rho)),
                'heldout_same_direction': bool(np.isfinite(rho) and rho * result['mean_rho'][k] > 0)})

    #5. =========== Return full evidence and fold arrays without persistence
    frequency = np.stack([result['selected'] for result in train_values]).sum(axis=0)
    evidence = pd.DataFrame({'TF': tf, 'gene': genes,
        'global_rho_min_abs': full['global_rho'],
        'source_global_GSE165502': source_rho[0], 'source_global_GSE195500': source_rho[1],
        'mean_library_rho': full['mean_rho'], 'valid_libraries': full['valid_libraries'],
        'direction_fraction': full['direction_fraction'],
        'source_min_detection': source_detection.min(axis=0),
        'excluded_mt_ribo_technical_or_self': ~permitted,
        'passed_pool': full['passed'], 'selected': full['selected'],
        'selection_rank': full['rank'], 'LOO_selected_folds': frequency})
    selected = evidence[evidence.selected].sort_values('selection_rank').reset_index(drop=True)
    indices = np.flatnonzero(full['selected'])
    indices = indices[np.argsort(full['rank'][indices])]
    gene_sets = {tf: {'genes': genes[indices].tolist(),
        'positive': genes[indices[full['mean_rho'][indices] > 0]].tolist(),
        'negative': genes[indices[full['mean_rho'][indices] < 0]].tolist(),
        'gene2association': {str(genes[k]): float(full['mean_rho'][k]) for k in indices},
        'scope': 'late_GSE165502_and_GSE195500', 'label': 'Late-associated expression program',
        'motif_supported': False, 'independent_GRN_validation': False}}
    summary = pd.DataFrame([{'TF': tf, 'scope': 'Late-associated expression program',
        'pool_genes': int(full['passed'].sum()), 'selected_genes': len(selected),
        'selected_positive': len(gene_sets[tf]['positive']),
        'selected_negative': len(gene_sets[tf]['negative']),
        'libraries': len(libraries), 'cells': len(late), 'folds': len(folds),
        'status': 'available' if len(selected) else 'Unavailable'}])
    loo = {'folds': pd.DataFrame(folds), 'selected_evidence': pd.DataFrame(validation)}
    for key in ['source_rho', 'source_detection', 'global_rho', 'mean_rho',
                'valid_libraries', 'direction_fraction', 'passed', 'selected', 'rank']:
        loo['train_' + key] = np.stack([result[key] for result in train_values])
    if file_sha256(inputs_path) != input_hash:
        raise ValueError('Inputs.h5 changed during the late audit.')
    methods = {'label': 'Late-associated expression program', 'TF': tf, 'sources': sources,
        'good_cells': 'source_state_evaluation_allowed and not source_rna_mixture_review_flag',
        'stages': sorted(late.stage_standardized.unique()),
        'RNA_unit': 'log10(1+native count/full endogenous total*1e6)',
        'correlation': 'Source-specific global Spearman; equal-library mean of within-library Spearman',
        'rules': {'min_detection_each_source': min_detection,
            'min_abs_global_rho_each_source': min_source_rho,
            'source_global_directions_agree': True,
            'min_abs_equal_library_rho': min_library_rho,
            'global_and_library_mean_directions_agree': True,
            'min_library_direction_fraction': min_direction_fraction,
            'min_valid_libraries_full_and_training': min_valid_libraries,
            'min_library_cells': min_library_cells,
            'min_TF_positive_cells': min_tf_positive_cells,
            'max_selected_genes': max_selected},
        'ranking': 'abs(mean library rho), then min(abs(source global rho)), then gene symbol',
        'LOO': 'First lexicographic library per source-stage, entire library removed; affected-source global ranks/detection and selection recomputed; unchanged source summaries reused exactly',
        'limits': ['Late scope is not an early target set or a TF regulon',
                   'Source-specific gates do not establish causality',
                   'Preselected TF and accepted trajectory remain fixed'],
        'exclusion_patterns': exclusion_patterns}
    return {'genes': genes, 'sources': np.asarray(sources), 'cells': late,
        'TF_counts': counts, 'TF_expression': tf_values[:, 0], 'TF_measured': measured,
        'source_global_rho': source_rho, 'source_detection': source_detection,
        'library_rho': library_rho, 'libraries': libraries, 'full': full,
        'evidence': evidence, 'selected': selected, 'summary': summary,
        'selected_gene_sets': gene_sets, 'LOO': loo, 'methods': methods,
        'source_sha256': {'Inputs.h5': input_hash, str(formal_counts): counts_hash},
        'checks': {'input_hash_unchanged': True, 'no_pooled_platform_global_correlation': True,
                   'unchanged_source_reused_exactly': True, 'folds': len(folds)}}
```

## 6. Primary-path and branch fits

C uses source/library-balanced shrinkage GAMs on CVG-T-AC-Ic. D reuses exactly those primary predictions and independently fits the other native segments; unsupported segments stay grey. Prox1 displays measured RNA directly and fits no nonexistent C/D program.

```python

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")

# Cleanup must never interpret the project/current directory as a temporary target.
def _release_cleanup_target(path):
    candidate = Path(path).resolve()
    current = Path.cwd().resolve()
    if candidate == current or candidate in current.parents or candidate == Path(candidate.anchor):
        raise RuntimeError("Configure the owned temporary cleanup path before running this section.")
    return path

#5. =========== Source-balanced GAMs and branch-specific spatial predictions
GAM_SOURCE = r'''
suppressPackageStartupMessages({library(mgcv); library(rhdf5)})
a <- commandArgs(trailingOnly=TRUE)
meta <- read.delim(a[1],check.names=FALSE,stringsAsFactors=FALSE)
meta$good <- tolower(as.character(meta$good)) %in% c('true','1')
meta$primary <- tolower(as.character(meta$primary)) %in% c('true','1')
meta$on_path <- tolower(as.character(meta$on_path)) %in% c('true','1')
jobs <- read.delim(a[2],check.names=FALSE,stringsAsFactors=FALSE,na.strings='NA')
X <- h5read(a[3],'job_values'); if(ncol(X)==nrow(meta)) X <- t(X)
stopifnot(nrow(X)==nrow(meta),ncol(X)==nrow(jobs))
grid <- seq(min(meta$t[meta$primary]),max(meta$t[meta$primary]),length.out=250)
P <- matrix(NA_real_,nrow(meta),ncol(X)); G <- matrix(NA_real_,length(grid),ncol(X))
main <- which(jobs$mode=='C' & (is.na(jobs$omit_library) | jobs$omit_library==''))
D <- matrix(NA_real_,nrow(meta),length(main))
balanced <- function(i){
 d<-meta[i,]; sources<-unique(d$study_id); w<-numeric(length(i))
 for(s in sources){
  m<-which(d$study_id==s); libs<-unique(d$library_id[m])
  for(l in libs){ z<-m[d$library_id[m]==l]; w[z]<-1/length(sources)/length(libs)/length(z) }
 }; w*length(i)/sum(w)
}
segment_predict <- function(train,rows,j){
 k<-min(10,length(unique(meta$t[train]))-1)
 if(length(train)<12 || k<3 || sd(X[train,j])<1e-10) return(rep(mean(X[train,j]),length(rows)))
 d<-data.frame(value=X[train,j],t=meta$t[train],w=balanced(train))
 m<-gam(value~s(t,bs='ts',k=k),data=d,weights=w,gamma=1.5,method='GCV.Cp')
 as.numeric(predict(m,newdata=data.frame(t=meta$t[rows])))
}
for(j in seq_len(ncol(X))){
 if(jobs$mode[j]=='B'){
  for(s in unique(meta$seg)){
   rows<-which(meta$seg==s); train<-rows[meta$good[rows] & is.finite(X[rows,j])]
   if(length(train)>0) P[rows,j]<-segment_predict(train,rows,j)
  }
 }else{
  train<-which(meta$primary & is.finite(X[,j]))
  l<-jobs$omit_library[j]
  if(!is.na(l) && nchar(l)>0) train<-train[meta$library_id[train]!=l]
  k<-min(10,length(unique(meta$t[train]))-1)
  d<-data.frame(value=X[train,j],t=meta$t[train],w=balanced(train))
  m<-gam(value~s(t,bs='ts',k=k),data=d,weights=w,gamma=1.5,method='GCV.Cp')
  P[,j]<-as.numeric(predict(m,newdata=data.frame(t=meta$t)))
  G[,j]<-as.numeric(predict(m,newdata=data.frame(t=grid)))
 }
}
for(z in seq_along(main)){
 j<-main[z]
 D[meta$on_path,z]<-P[meta$on_path,j]
 for(s in unique(meta$seg[!meta$on_path])){
  rows<-which(meta$seg==s); train<-rows[meta$good[rows] & is.finite(X[rows,j])]
  if(length(train)>=12 && length(unique(meta$t[train]))>=4) D[rows,z]<-segment_predict(train,rows,j)
 }
}
h5createFile(a[4]);h5write(P,a[4],'fitted');h5write(G,a[4],'grid_fitted')
h5write(grid,a[4],'grid');h5write(D,a[4],'D_fitted');h5write(main-1,a[4],'D_jobs')
h5write(paste(R.version.string,as.character(packageVersion('mgcv')),as.character(packageVersion('rhdf5')),sep=' | '),a[4],'R_versions')
h5closeAll()
'''


def fit_factor(result, work_dir, rscript):
    if not result['selected_modules']:
        assert result['factor'] == 'Prox1' and result['methods']['analysis_kind'] == 'unavailable'
        assert np.isfinite(result['B_raw']).all()
        grid = np.linspace(result['obs'].t[result['primary_path']].min(),
                           result['obs'].t[result['primary_path']].max(), 250)
        result.update(fitted=result['job_values'].copy(),
            grid_fitted=np.full((len(grid), len(result['jobs'])), np.nan), fit_grid=grid,
            D_fitted=np.empty((len(result['obs']), 0)), D_jobs=np.array([], dtype=int),
            fit_methods={'B': 'Measured native Prox1 RNA; no smoothing',
                'C_D': 'Unavailable: no full-data early program passed the fixed criteria',
                'prediction_clipping': False}, fit_R_source='')
        return result
    work_dir = Path(work_dir)
    work_dir.mkdir(parents=True, exist_ok=True)
    obs = result['obs'].copy()
    obs['good'] = result['good']
    obs['primary'] = result['primary_path']
    obs['on_path'] = obs.seg.isin(('CVG', 'T', 'AC', 'Ic'))
    jobs = result['jobs']
    files = {key: work_dir / name for key, name in {
        'meta': 'fit_cells.tsv', 'jobs': 'fit_jobs.tsv', 'input': 'fit_input.h5',
        'source': 'fit.R', 'output': 'fit_output.h5'}.items()}
    try:
        obs[['t', 'seg', 'study_id', 'library_id', 'good', 'primary', 'on_path']].to_csv(files['meta'], sep='\t', index=False)
        jobs.fillna('').to_csv(files['jobs'], sep='\t', index=False)
        with h5py.File(files['input'], 'w') as h:
            h.create_dataset('job_values', data=result['job_values'])
        files['source'].write_text(GAM_SOURCE, encoding='utf-8')
        if files['output'].exists():
            _release_cleanup_target(files['output']).unlink()
        env = os.environ.copy()
        env.update(TMPDIR=str(work_dir), TMP=str(work_dir), TEMP=str(work_dir), LC_ALL='C', LANG='C')
        completed = subprocess.run([str(rscript), '--vanilla', str(files['source']), str(files['meta']),
            str(files['jobs']), str(files['input']), str(files['output'])], env=env,
            capture_output=True, text=True, errors='replace')
        if completed.returncode:
            raise RuntimeError(completed.stderr + completed.stdout)
        if completed.stderr.strip():
            print(completed.stderr.strip())
        with h5py.File(files['output'], 'r') as h:
            fitted = h['fitted'][:].T
            grid_fit = h['grid_fitted'][:].T
            grid = h['grid'][:]
            d_fit = h['D_fitted'][:].T
            d_jobs = np.asarray(h['D_jobs'][:]).ravel().astype(int)
            r_versions = h['R_versions'].asstr()[()].tolist()
        assert fitted.shape == result['job_values'].shape
        assert np.isfinite(fitted).all()
        assert d_fit.shape == (len(obs), len(d_jobs))
        on_path = obs.on_path.to_numpy()
        np.testing.assert_array_equal(d_fit[on_path], fitted[on_path][:, d_jobs])
        validation = result['validation'].copy()
        for i, row in validation.iterrows():
            hold = result['primary_path'] & obs.library_id.eq(row.heldout_library).to_numpy()
            j = int(row.job)
            validation.loc[i, 'heldout_path_cells'] = int(hold.sum())
            if j >= 0 and hold.sum() >= 5:
                validation.loc[i, 'heldout_prediction_spearman'] = spearmanr(result['job_values'][hold, j], fitted[hold, j]).statistic
                validation.loc[i, 'heldout_score_min'] = float(result['job_values'][hold, j].min())
                validation.loc[i, 'heldout_score_max'] = float(result['job_values'][hold, j].max())
        result.update(fitted=fitted, grid_fitted=grid_fit, fit_grid=grid,
            D_fitted=d_fit, D_jobs=d_jobs, validation=validation,
            fit_methods={'formula': "value~s(t,bs='ts',k<=10)", 'gamma': 1.5,
                'method': 'GCV.Cp', 'weights': 'Equal source, equal libraries per source, equal cells per library; rescaled to n',
                'C': 'Good CVG/T/AC/Ic primary path only',
                'D': 'Exact C prediction on the primary segments; other native segments fitted separately; unsupported segments remain NaN',
                'prediction_clipping': False, 'R_versions': r_versions}, fit_R_source=GAM_SOURCE)
        return result
    finally:
        for p in files.values():
            if p.exists():
                result.setdefault('temporary_cleanup', []).append({'name': p.name, 'bytes': p.stat().st_size})
                _release_cleanup_target(p).unlink()


```

## 7. Curved gene display and joint panels

All genes from each displayed C module appear on the corresponding D side. Labels occupy nested arcs, with fine curves and no arrowheads. Associated genes are not validated direct targets. Neurod1 C/D programs are inverse-associated with early TF RNA; this does not establish direct repression. Descriptive functional titles do not imply GO significance. D colorbar maxima use actual fitted maxima rounded up to .05, so compare spatial patterns within each module.

The D network reuses the published Fig.3d filled Bezier ribbon paths, node slots, palette, typography proportions and gray guide. Current module genes replace the original visual-slot names. Smaller modules retain source slots covering the full outline; additional members use interpolation between compatible source ribbons. The original slot names are visual provenance only. Red marks the existing literature-discussed representatives and does not encode significance, edge strength, direct binding or proven fate regulation. The embedded template records the source PDF path, SHA256 and page, with no runtime PDF or temporary-file dependency.

```python
"""Published Fig.3d vector ribbons with current module gene names substituted."""

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")

import numpy as np
import matplotlib.patheffects as pe
from matplotlib.path import Path as MplPath
from matplotlib.patches import PathPatch
from matplotlib.font_manager import FontProperties
from matplotlib.textpath import TextPath


# Literature-discussed representatives, not significance or stronger-edge gates.

NETWORK_HIGHLIGHT_EVIDENCE = {'Foxo6': {'Nhlh1': {'reason': 'Highlighted in the reference paper as a neurogenesis-related member.', 'source': 'https://www.nature.com/articles/s41467-022-31580-1'}, 'Sox11': {'reason': 'Representative neuronal differentiation factor; general CNS evidence, not demonstrated SGN fate specification.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC1698453/'}}, 'Zfpm2': {'Isl1': {'reason': 'Representative with direct cochleovestibular neuron differentiation, migration and axon evidence; does not validate a Zfpm2-to-Isl1 edge.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC10883659/'}, 'Sox11': {'reason': 'Neuronal differentiation representative; general CNS evidence.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC1698453/'}}, 'Neurod2': {'Nhlh1': {'reason': 'Reference-paper neurogenesis representative already present in this module.', 'source': 'https://www.nature.com/articles/s41467-022-31580-1'}, 'Neurod6': {'reason': 'Reference-paper neurogenesis representative already present in this module.', 'source': 'https://www.nature.com/articles/s41467-022-31580-1'}, 'Isl1': {'reason': 'Direct cochleovestibular developmental evidence; does not establish the current association as direct regulation.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC10883659/'}}, 'Nhlh1': {'Nhlh2': {'reason': 'Developmental sensory-neuron expression representative; expression evidence is not functional necessity and Nhlh1 knockout effects are not transferred to Nhlh2.', 'source': 'https://pubmed.ncbi.nlm.nih.gov/17004922/'}}, 'Neurod1': {'Myt1l': {'reason': 'Representative neuronal identity factor; not evidence of direct Neurod1 regulation.', 'source': 'https://www.nature.com/articles/nature21722'}, 'Sema3a': {'reason': 'Representative with auditory-neuron connectivity/excitability evidence; not early SGN fate or direct Neurod1-target validation.', 'source': 'https://pubmed.ncbi.nlm.nih.gov/36284453/'}}, 'Prox1': {}}
NETWORK_HIGHLIGHTS = {tf: tuple(genes) for tf, genes in NETWORK_HIGHLIGHT_EVIDENCE.items()}

# Set this storage path independently before running. Single-cell RNA-sequencing analysis of the developing mouse inner ear identifies molecular logic of auditory neuron diversification.pdf.
PAPER_NETWORK_TEMPLATE = {'source': 'Petitpre et al. 2022, Nature Communications, Fig.3d', 'path': 'ROOT', 'sha256': '8dda1b31fe727e1ee88d73932b6b0dd485ab070855ec39bd4d5587e1b2826bd4', 'page_1based': 6, 'bbox': [302.0, 208.0, 466.0, 286.0], 'center': [384.737473256552, 248.26325348336002], 'base_font_size': 4.34579991999999, 'red_font_size': 4.1339998199999854, 'red_rgb': [0.72941, 0.0, 0.1098], 'guides': [{'path': [['m', [383.54000854, 219.98101855999994]], ['l', [383.54000854, 221.68801856999994]]], 'pts': [[383.54000854, 219.98101855999994], [383.54000854, 221.68801856999994]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 219.98101855999994, 'bottom': 221.68801856999994, 'linewidth': 0.329, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 222.28698780000002]], ['l', [383.54000854, 242.96197558999995]]], 'pts': [[383.54000854, 222.28698780000002], [383.54000854, 242.96197558999995]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 222.28698780000002, 'bottom': 242.96197558999995, 'linewidth': 0.329, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[3.5955, 0.59925997], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 254.99102832999995]], ['l', [383.54000854, 256.5500283399999]]], 'pts': [[383.54000854, 254.99102832999995], [383.54000854, 256.5500283399999]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 254.99102832999995, 'bottom': 256.5500283399999, 'linewidth': 0.30000001, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 257.04101612]], ['l', [383.54000854, 274.01001025999994]]], 'pts': [[383.54000854, 257.04101612], [383.54000854, 274.01001025999994]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 257.04101612, 'bottom': 274.01001025999994, 'linewidth': 0.30000001, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[2.95099997, 0.49184], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 274.25601245999997]], ['l', [383.54000854, 275.81401242]]], 'pts': [[383.54000854, 274.25601245999997], [383.54000854, 275.81401242]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 274.25601245999997, 'bottom': 275.81401242, 'linewidth': 0.30000001, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}], 'slots': {'left': [{'visual_slot': 'St18', 'point': [346.20866888772, 262.10511139336006], 'bbox': [341.73901367, 259.93221143336007, 350.67832410544, 264.27801135336006], 'path': [['m', [384.55599975, 247.91198780000002]], ['c', [382.62599981, 252.00998784], [377.81299972, 255.82098771000005], [372.05499935, 258.58798744]], ['c', [366.29899979000004, 261.35598804000006], [359.60300064, 263.08798744], [353.93199921, 263.086988]], ['c', [351.02399825000003, 263.086988], [348.38800049, 262.63198806], [346.2879982, 261.63998748000006]], ['l', [346.11899948, 261.99698782999997]], ['c', [348.29299927, 263.02298786000006], [350.98299789000004, 263.48098804000006], [353.93199921, 263.48198750000006]], ['c', [359.68199921, 263.48098804000006], [366.42000008, 261.7349877500001], [372.22699928000003, 258.94398738000007]], ['c', [378.03099965, 256.14898826], [382.9089998, 252.31698800000004], [384.91299975000004, 248.0809878]], ['l', [384.55599975, 247.91198780000002]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Nhlh1', 'point': [356.088734364884, 244.5250502825599], 'bbox': [350.6889038, 242.4580503725599, 361.48856492976796, 246.5920501925599], 'path': [['m', [384.61499023, 247.83898975]], ['c', [382.29399013, 249.61298979000003], [378.49899006, 250.44698978999998], [374.39698982, 250.44498968999994]], ['c', [367.40699005, 250.45298982999998], [359.5329895, 248.02498975000003], [356.52198982, 244.05298974000004]], ['l', [356.20698928999997, 244.29298974999995]], ['c', [359.38298987999997, 248.43698973000005], [367.31798935, 250.83198977999996], [374.39698982, 250.83998967000002]], ['c', [378.55099011, 250.83798981999996], [382.41299033, 250.01098967999997], [384.85399022999997, 248.15398975000005]], ['l', [384.61499023, 247.83898975]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Rd3', 'point': [322.756078766624, 226.05335357336003], 'bbox': [318.77011108, 223.88045361336003, 326.742046453248, 228.22625353336002], 'path': [['m', [384.6499939, 247.81799365999996]], ['c', [381.24699379000003, 249.42599369999994], [376.98399402, 250.16099358999998], [372.28799344000004, 250.16099358999998]], ['c', [363.14399339000005, 250.16199375999997], [352.3639946, 247.37599365999995], [343.0929947, 242.91899349999994]], ['c', [333.81999207, 238.46499397999992], [326.06099320000004, 232.33299401], [322.9339943, 225.7029938899999]], ['l', [322.57699204000005, 225.87199451999993]], ['c', [325.77799226, 232.63999320999994], [333.60199357000005, 238.79299404999995], [342.92199326, 243.27499344]], ['c', [352.24299240000005, 247.75399366], [363.06499482000004, 250.5549936399999], [372.28799344000004, 250.55599356999994]], ['c', [377.0249939, 250.55599356999994], [381.34199382, 249.81699360999994], [384.8189939, 248.17499365999993]], ['l', [384.6499939, 247.81799365999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'St6galnac5', 'point': [345.172961909844, 238.51819732336003], 'bbox': [334.2999878, 236.34529736336003, 356.045936019688, 240.69109728336002], 'path': [['m', [384.63101196, 247.82800341999996]], ['c', [381.88601208, 249.52700341999991], [377.98401213, 250.3030033199999], [373.67301178, 250.30200338999998]], ['c', [368.28301239, 250.3030033199999], [362.25501251000003, 249.09300339999993], [357.04801178, 247.01500339999996]], ['c', [351.83901215000003, 244.93900349], [347.45501328, 241.98800326999992], [345.33501053000003, 238.5520034]], ['l', [344.999012, 238.76000359]], ['c', [347.19801331, 242.31200361999993], [351.64701081000004, 245.28300334999994], [356.90201186, 247.38200341999993]], ['c', [362.15801239, 249.47900343999993], [368.22401238000003, 250.69700336999995], [373.67301178, 250.69700336999995]], ['c', [378.03101206, 250.69600344999992], [381.99501205, 249.92000341999994], [384.83901196000005, 248.16500342999996]], ['l', [384.63101196, 247.82800341999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ntrk1', 'point': [325.02116664444, 241.59320952335997], 'bbox': [319.83010864, 239.42030956335998, 330.21222464887995, 243.76610948335997], 'path': [['m', [384.61099243, 247.84198046999995]], ['c', [379.48399258, 251.94298029999993], [370.88299275, 253.86098052], [361.68199158, 253.85898066999994]], ['c', [347.34299088, 253.86598062999997], [331.55499268, 249.20198047999997], [325.17399216, 241.19598056999996]], ['l', [324.8659935, 241.44298030999994]], ['c', [331.40899277, 249.61398041999996], [347.2599907, 254.24698066999997], [361.68199158, 254.25398064]], ['c', [370.93699265000004, 254.25098038], [379.60499239, 252.34198046999995], [384.85799243, 248.1509804499999]], ['l', [384.61099243, 247.84198046999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Egf', 'point': [335.78853088996004, 244.15320709336004], 'bbox': [332.52700805, 241.98030713336004, 339.05005372992, 246.32610705336003], 'path': [['m', [384.60598754, 247.84698535999996]], ['c', [380.19398737, 251.64298536], [372.34598732, 253.4589853399999], [364.09498787, 253.45598555999993]], ['c', [352.93998718, 253.46198513000002], [341.05598831, 250.13998533999995], [335.93298721, 244.16198541999995]], ['l', [335.63398742, 244.4199853199999]], ['c', [340.91498564, 250.54998541999998], [352.86398697, 253.8449855], [364.09498787, 253.85098554]], ['c', [372.40198707, 253.84798528], [380.31898737, 252.04398537999998], [384.86298754, 248.14598535999994]], ['l', [384.60598754, 247.84698535999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Clvs1', 'point': [336.701124429984, 255.54914215335998], 'bbox': [331.26800537, 253.37624219335999, 342.134243489968, 257.72204211336], 'path': [['m', [384.57598876, 247.87902880000001]], ['c', [379.39998864999995, 254.92302894], [365.57698822, 259.28602838000006], [353.48098945, 259.27702855999996]], ['c', [346.81998825, 259.27902841], [340.68598937999997, 257.95902872], [336.81698988999995, 255.10202884]], ['l', [336.58198927999996, 255.42002867999997]], ['c', [340.56798934, 258.35502863], [346.76898955999997, 259.67102861], [353.48098945, 259.67202901999997]], ['c', [365.66898918, 259.66402863999997], [379.55198859999996, 255.33602857000005], [384.89398874, 248.11402880000003]], ['l', [384.57598876, 247.87902880000001]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Raver2', 'point': [351.98725317106005, 222.76331451336], 'bbox': [344.98291015, 220.59041455336, 358.99159619212, 224.93621447336], 'path': [['m', [384.70999145, 247.79998829]], ['c', [384.20099145, 247.86398828999995], [383.67499149, 247.89498829000001], [383.13399148, 247.89498829000001]], ['c', [377.3329916, 247.89898829000003], [369.91099167000004, 244.30598834], [363.79399109, 239.31598809000002]], ['c', [357.67199135000004, 234.33098842999993], [352.85699081, 227.94598818999998], [352.18299103000004, 222.47198918000004]], ['l', [351.79099274000004, 222.52098896999996]], ['c', [352.50099182, 228.17798855], [357.37999153000004, 234.59198811], [363.54499054, 239.62198783999997]], ['c', [369.71499157, 244.64698818], [377.17499161, 248.2849883], [383.13399148, 248.28998829]], ['c', [383.68999147, 248.28998829], [384.23199147, 248.25798828999996], [384.75899145, 248.19198827000002]], ['l', [384.70999145, 247.79998829]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Foxo6', 'point': [331.56035424882396, 219.52234771335998], 'bbox': [325.5209961, 217.34944775335998, 337.599712397648, 221.69524767335997], 'path': [['m', [384.67800903, 247.80700732999992]], ['c', [382.77300906, 248.37600732999988], [380.65000916, 248.64300734999995], [378.37800883, 248.64300734999995]], ['c', [369.85500908, 248.64500732999988], [359.24900818, 244.87600732999988], [350.2530098, 239.30200721999995]], ['c', [341.2550087, 233.73200752999992], [333.87100983, 226.34800769999993], [331.74401093, 219.2010064299999]], ['l', [331.36600876, 219.31400729999996]], ['c', [333.55200958, 226.62300731999994], [341.00000763, 234.03100730999995], [350.04500961, 239.63800765999997]], ['c', [359.09300995, 245.2410073499999], [369.74100876, 249.03600729999994], [378.37800883, 249.03800726999987]], ['c', [380.68100906, 249.03800726999987], [382.84200907, 248.76700729999993], [384.79100903, 248.18600732999994]], ['l', [384.67800903, 247.80700732999992]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Shh', 'point': [347.06160342884004, 250.46924712336], 'bbox': [343.1960144, 248.29634716336, 350.92719245768, 252.64214708336], 'path': [['m', [384.58700561, 247.86499072999993]], ['c', [380.7070055, 252.23799086999998], [371.95900536, 254.62499094999998], [363.6190052, 254.6199908399999]], ['c', [357.00400543, 254.6229906299999], [350.64800644, 253.12199068999996], [347.18900681, 250.03599072999998]], ['l', [346.92700577, 250.3309908099999]], ['c', [350.52200699, 253.52399063999997], [356.94600487, 255.01099062999992], [363.6190052, 255.01499082999987]], ['c', [372.03200531, 255.00899077999998], [380.84500551, 252.6469907899999], [384.88200559, 248.12799073999997]], ['l', [384.58700561, 247.86499072999993]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Cecr2', 'point': [341.931003113304, 228.58631256336002], 'bbox': [336.13500976, 226.41341260336003, 347.726996466608, 230.75921252336002], 'path': [['m', [384.6659851, 247.8109746]], ['c', [382.79998505, 248.50097460000006], [380.62498522, 248.82197463], [378.2609849, 248.82197463]], ['c', [371.64598465, 248.82397461000005], [363.56498527, 246.30497456], [356.6639843, 242.46397448000005]], ['c', [349.76098633, 238.62597418999997], [344.042984, 233.4569745], [342.11598587000003, 228.22997523000004]], ['l', [341.74598694, 228.36697435999997]], ['c', [343.73898697, 233.74597502000006], [349.5219841, 238.93797445999996], [356.47198486, 242.80897475000006]], ['c', [363.42398453, 246.67697466000004], [371.54798507, 249.21597456999996], [378.2609849, 249.21697460999997]], ['c', [380.65998506, 249.21697460999997], [382.87998509, 248.89197457], [384.8029851, 248.1819746]], ['l', [384.6659851, 247.8109746]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Neurod6', 'point': [355.255089154796, 233.17708885255996], 'bbox': [347.3269043, 231.11008894255997, 363.183274009592, 235.24408876255995], 'path': [['m', [384.6749878, 247.80798388999995]], ['c', [383.58998777, 248.14998388999993], [382.36598779, 248.3109839099999], [381.04898787999997, 248.3109839099999]], ['c', [376.45298768, 248.31298388999994], [370.73598768, 246.34498383999994], [365.88198854, 243.41198398999995]], ['c', [361.02498819, 240.48198367999998], [357.03598787, 236.5799842099999], [355.84898759, 232.79198408999991]], ['l', [355.47198869, 232.91098358]], ['c', [356.71998788, 236.85798408999995], [360.77398874, 240.78498413], [365.67798806, 243.7499838099999]], ['c', [370.5839882, 246.71098388999997], [376.34298803, 248.70398390999992], [381.04898787999997, 248.70598388999997]], ['c', [382.39798785, 248.70598388999997], [383.66198779, 248.54198391], [384.7939878, 248.1849838899999]], ['l', [384.6749878, 247.80798388999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ppp1r17', 'point': [338.6215513488, 268.92810942336007], 'bbox': [330.4079895, 266.7552094633601, 346.8351131976, 271.10100938336], 'path': [['m', [384.5499878, 247.92700244999992]], ['c', [382.46198775, 253.54200221999997], [376.30998803, 259.0780024699999], [368.88598824999997, 263.18700266999997]], ['c', [361.46598817, 267.2990021899999], [352.7789879, 269.9910016199999], [345.6579895, 269.9900021699999]], ['c', [343.08498765, 269.9900021699999], [340.71698762, 269.63900234999994], [338.68798829, 268.8820023699999]], ['l', [338.5499878, 269.2520032099999]], ['c', [340.63598634, 270.0300030899999], [343.04898835, 270.3850026299999], [345.6579895, 270.3850026299999]], ['c', [352.87598802, 270.3830018199999], [361.6049881, 267.6700024799999], [369.07798768, 263.53300238999987]], ['c', [376.54798795, 259.3910021899999], [382.76498784, 253.83200264999994], [384.9199878, 248.06500244999995]], ['l', [384.5499878, 247.92700244999992]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Klc3', 'point': [313.35505509218797, 245.62421539336003], 'bbox': [309.1281128, 243.45131543336004, 317.581997384376, 247.79711535336003], 'path': [['m', [384.6000061, 247.85199023999996]], ['c', [377.81400633, 254.16299011], [364.97900582, 257.2509904], [351.80200577, 257.24699068999996]], ['c', [336.5300064, 257.25198985], [320.80100631999994, 253.1039900899999], [313.49700928, 245.22599031999994]], ['l', [313.20700836, 245.49499013000002]], ['c', [320.66500472999996, 253.51199007999992], [336.46000671999997, 257.6369900799999], [351.80200577, 257.6419902]], ['c', [365.03900527999997, 257.63799049], [377.94200609999996, 254.56699039], [384.8690061, 248.14099023999995]], ['l', [384.6000061, 247.85199023999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ppp1r14a', 'point': [324.71179432656004, 252.19410064336], 'bbox': [315.2901001, 250.02120068336, 334.13348855312, 254.36700060336], 'path': [['m', [384.58499145, 247.86700488999998]], ['c', [378.29899168000003, 255.14800502000003], [363.83999062000004, 259.16400481999995], [350.22599030000003, 259.15800525]], ['c', [340.01898955, 259.1610045599999], [330.28799057000003, 256.90500498999995], [324.83999252, 252.18400480000003]], ['l', [324.58098984000003, 252.48300505999998]], ['c', [330.16299057000003, 257.3060050099999], [339.96199035, 259.5500045], [350.22599030000003, 259.55300476]], ['c', [363.91599083, 259.54700518999994], [378.43899155, 255.55800487], [384.88399145, 248.12600489]], ['l', [384.58499145, 247.86700488999998]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Klhl25', 'point': [331.8726626188, 232.49829986336], 'bbox': [325.83200073, 230.32539990336, 337.9133245076, 234.67119982335998], 'path': [['m', [384.64099121, 247.82299853999996]], ['c', [381.36599111, 249.59399855999993], [377.02899123, 250.39999844], [372.22899150999996, 250.39999844]], ['c', [364.69799041, 250.39999844], [356.03499030999996, 248.41699852], [348.56399154999997, 245.13799859999995]], ['c', [341.09199143, 241.86099863999993], [334.81698989999995, 237.27999832], [332.04199219, 232.13899852999998]], ['l', [331.69499207, 232.32699825999998]], ['c', [334.54698944, 237.5959983099999], [340.88599015, 242.19699861999993], [348.40499115, 245.4999985899999]], ['c', [355.92499161, 248.79999853999993], [364.62899208, 250.79399848999992], [372.22899150999996, 250.79499841999996]], ['c', [377.07199145, 250.79399848999992], [381.46899127999995, 249.98499847999994], [384.82899120999997, 248.16999853999994]], ['l', [384.64099121, 247.82299853999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Necab2', 'point': [318.41408894212, 267.16614654336], 'bbox': [310.92410278, 264.99324658336, 325.90407510424, 269.33904650336], 'path': [['m', [384.56201171, 247.90002489999995]], ['c', [381.04601168, 254.22502469999995], [373.26701163999996, 259.80202532], [364.01501083, 263.7800250099999]], ['c', [354.76401138, 267.76002551], [344.04501340999997, 270.1470246399999], [334.66701126, 270.1470246399999]], ['c', [328.40001297, 270.1470246399999], [322.73401259999997, 269.0790257499999], [318.50901031, 266.7280240099999]], ['l', [318.31700896999996, 267.07302523]], ['c', [322.63001251, 269.47102404], [328.35601042999997, 270.54102563999993], [334.66701126, 270.54202509999993]], ['c', [344.11101149999996, 270.54102563999993], [354.87101172999996, 268.1430249299999], [364.17101097, 264.1430249299999]], ['c', [373.47001171, 260.14002467], [381.31401180999995, 254.54302501999996], [384.90701171, 248.09302489999993]], ['l', [384.56201171, 247.90002489999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Dscaml1', 'point': [357.05660059441595, 256.80414702336], 'bbox': [348.60488891, 254.63124706335998, 365.50831227883197, 258.97704698335997], 'path': [['m', [384.55999755, 247.90399218999994]], ['c', [383.13399755, 250.61099219999994], [379.83199738999997, 253.06199218999996], [375.88999748, 254.81699228999992]], ['c', [371.94899749, 256.57499170999995], [367.37299728, 257.64499235999995], [363.42199707, 257.64399196]], ['c', [360.99599838, 257.64399196], [358.80799675, 257.24099208999996], [357.1479969, 256.36299181999993]], ['l', [356.96399687999997, 256.71199179999996]], ['c', [358.70599745, 257.63199187], [360.95299719999997, 258.0379920099999], [363.42199707, 258.03899241]], ['c', [367.44299698, 258.0379920099999], [372.05999755, 256.95699263999995], [376.04999733, 255.17799234999995]], ['c', [380.03899765, 253.3969922199999], [383.40699754999997, 250.92499208999993], [384.90899755, 248.08899218999989]], ['l', [384.55999755, 247.90399218999994]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Cacna1f', 'point': [326.950596540992, 275.43103912335994], 'bbox': [318.85610961, 273.25813916335994, 335.045083471984, 277.60393908335993], 'path': [['m', [384.54901123, 247.92901659999995]], ['c', [381.96901131000004, 255.09601640999995], [374.19801140000004, 262.2010169099999], [364.81201172000004, 267.49001741999996]], ['c', [355.42801095000004, 272.78301669999996], [344.43400956000005, 276.26701593999996], [335.4480095, 276.2660165]], ['c', [332.34601212, 276.2660165], [329.48501206000003, 275.85101747999994], [327.01301193, 274.95901727999996]], ['l', [326.87901307000004, 275.33101701999993]], ['c', [329.40600968, 276.24201631999995], [332.31101227, 276.6600175], [335.4480095, 276.66101695]], ['c', [344.53401185, 276.65901613999995], [355.57001113, 273.15401697999994], [365.00601197000003, 267.83401728]], ['c', [374.43801117000004, 262.51201677999995], [382.27501131, 255.38401651999993], [384.92001123, 248.06401659999995]], ['l', [384.54901123, 247.93001659999993]], ['l', [384.54901123, 247.92901659999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Kcne1l', 'point': [328.69355004772, 263.50513580335996], 'bbox': [322.05099487, 261.33223584335997, 335.33610522544, 265.67803576335996], 'path': [['m', [384.56298828, 247.8980107499999]], ['c', [381.58098817, 253.1150107499999], [375.09998798000004, 257.68301058999987], [367.39598847, 260.93001032999996]], ['c', [359.69298745000003, 264.17801142999986], [350.76998902, 266.11201144999995], [342.91898728, 266.1110100899999]], ['c', [337.41698838, 266.11201144999995], [332.44298935, 265.16000987999996], [328.78899002, 263.0680108199999]], ['l', [328.59298706000004, 263.4110102799999]], ['c', [332.33698655, 265.55301142999986], [337.37198638, 266.5060105499999], [342.91898728, 266.5060105499999]], ['c', [350.83398818, 266.5060105499999], [359.79698754000003, 264.5620102999999], [367.54898835, 261.2940106499999]], ['c', [375.29998875, 258.0230107499999], [381.84498835, 253.43501091999997], [384.90598828000003, 248.09501074999991]], ['l', [384.56298828, 247.8980107499999]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Mfap4', 'point': [348.57326316882404, 277.13901642336], 'bbox': [342.53390502, 274.96611646336, 354.612621317648, 279.31191638335997], 'path': [['m', [384.53799438, 247.97601367999994]], ['c', [383.88999437999996, 254.18201352999995], [378.54999447, 261.50101328999995], [371.71799468, 267.23701335]], ['c', [364.89199448, 272.97901392999995], [356.57699393999997, 277.14101458999994], [350.06999587999996, 277.13601351999995]], ['c', [349.56599425999997, 277.13601351999995], [349.07199477999995, 277.11101389999993], [348.59099577999996, 277.06001330999993]], ['l', [348.54999541999996, 277.4530129599999]], ['c', [349.04499434999997, 277.50501299999996], [349.55299377, 277.53101397999995], [350.06999587999996, 277.53101397999995]], ['c', [356.73999405, 277.52601290999996], [365.09299468, 273.3170132799999], [371.97199439999997, 267.5400128499999]], ['c', [378.84599447, 261.75701381], [384.2499944, 254.40801383999997], [384.93099437999996, 248.01701368]], ['l', [384.53799438, 247.97601367999994]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Kcnh6', 'point': [315.9832609866, 234.44025542336], 'bbox': [309.8230896, 232.26735546336, 322.14343237319997, 236.61315538335998], 'path': [['m', [384.625, 247.83197069999994]], ['c', [379.5079999, 251.23897075999992], [371.94699955, 252.78997087999994], [363.64100075, 252.78797054999995]], ['c', [345.40999985, 252.7979707799999], [323.6080017, 245.3139708199999], [316.1439972, 234.0659708999999]], ['l', [315.81500245, 234.28497076999997]], ['c', [323.44699860000003, 245.72697068999992], [345.3069992, 253.1729707799999], [363.64100075, 253.18297052999992]], ['c', [371.99499989, 253.1819706], [379.62200022, 251.63297080999996], [384.844, 248.1609706999999]], ['l', [384.625, 247.83197069999994]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Btbd17', 'point': [355.26389850322, 268.97010162336005], 'bbox': [348.37689208, 266.79720166336006, 362.15090492644, 271.14300158336], 'path': [['m', [384.54000854, 247.96197558999995]], ['c', [383.74100852, 252.56897545999993], [379.46600866, 257.77397585999995], [374.10800838, 261.7969756199999]], ['c', [368.75400829, 265.8239751], [362.32200813, 268.67597629], [357.28400803, 268.67197466999994]], ['c', [356.59200859, 268.67197466999994], [355.92700768, 268.61897517], [355.29400826, 268.50797510999996]], ['l', [355.22600937, 268.8969750499999]], ['c', [355.88400841, 269.01197482], [356.57200813, 269.06697512999995], [357.28400803, 269.06697512999995]], ['c', [362.46600914, 269.06297541999993], [368.93900871, 266.17297603], [374.34500884, 262.11197519999996]], ['c', [379.74700831999996, 258.0469756199999], [384.08700854, 252.81497574999992], [384.92900854, 248.02997558999994]], ['l', [384.54000854, 247.96197558999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ebf2', 'point': [342.18483272487, 217.1121474525599], 'bbox': [337.93301391, 215.0451475425599, 346.43665153973996, 219.1791473625599], 'path': [['m', [384.70498657, 247.80102588]], ['c', [383.90998656, 247.92102588], [383.08198655, 247.97902588], [382.22698665, 247.97902588]], ['c', [374.84298611, 247.98302588], [365.42998696, 243.62802603], [357.62598610000003, 237.53002598]], ['c', [349.81798554, 231.43702556000005], [343.62198639, 223.59702541000001], [342.59298706000004, 216.73302507999995]], ['l', [342.20198822000003, 216.79202511000005]], ['c', [343.27098847, 223.83602572999996], [349.53098679000004, 231.70402576000004], [357.38298607, 237.84102585000005]], ['c', [365.23998642000004, 243.97302581999998], [374.69198704, 248.37002587999996], [382.22698665, 248.37402588999998]], ['c', [383.09998656, 248.37402588999998], [383.94798655, 248.31502588], [384.76398657000004, 248.19202587999996]], ['l', [384.70498657, 247.80102588]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Khdrbs3', 'point': [314.19896563104, 256.75110748336], 'bbox': [306.10708618, 254.57820752336, 322.29084508208, 258.92400744336], 'path': [['m', [384.57901, 247.87402393000002]], ['c', [377.0040102, 257.54302359999997], [357.98900985, 263.23302413], [340.85300827, 263.22602416000007]], ['c', [330.21900939, 263.22802403000003], [320.31101226, 261.03802348], [314.32000732, 256.32902385]], ['l', [314.07601165, 256.64002373000005]], ['c', [320.19001006999997, 261.43702365], [330.1660118, 263.61802436000005], [340.85300827, 263.62102367]], ['c', [358.07401083999997, 263.61302424999997], [377.15101003, 257.95502425000006], [384.89000998, 248.11802393000005]], ['l', [384.57901, 247.87402393000002]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Mfng', 'point': [337.58536149108, 279.76703766336], 'bbox': [332.75500488, 277.59413770336, 342.41571810216, 281.93993762336], 'path': [['m', [384.5409851, 247.95898487]], ['c', [383.1479851, 255.21098471999994], [376.32698536, 263.25498439], [367.8289852, 269.45598461]], ['c', [359.33598518, 275.66198398], [349.16998672, 280.03198672999997], [341.15998459, 280.02798511]], ['c', [339.92298507, 280.02798511], [338.73898697, 279.92398502], [337.61998367, 279.70898487]], ['l', [337.54498672, 280.09698535]], ['c', [338.69198608, 280.31798602], [339.90098572, 280.42298557], [341.15998459, 280.42298557]], ['c', [349.30898666, 280.4199834], [359.51598549, 276.01398517], [368.061985, 269.77498485]], ['c', [376.6039853, 263.53198482000005], [383.4899851, 255.46098471999994], [384.92898511, 248.03398487000004]], ['l', [384.5409851, 247.95898487]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}], 'right': [{'visual_slot': 'Cnih3', 'point': [431.303848375488, 281.52503326335994], 'bbox': [425.62649536, 279.35213330335995, 436.98120139097597, 283.69793322335994], 'path': [['m', [384.7669983, 248.1909794999999]], ['c', [385.73099828, 248.02997949999997], [386.74099827000003, 247.95297949999986], [387.7879982, 247.95297949999986]], ['c', [395.82999802, 247.94897949999995], [406.06399918, 252.55297947999986], [414.5739975, 259.03397990999986]], ['c', [423.0889969, 265.50997973999995], [429.87899781, 273.8679795399999], [431.10899735, 281.2719807799999]], ['l', [431.49899674, 281.2069783399999]], ['c', [430.22599793, 273.6239800599999], [423.3729973, 265.23897982999995], [414.81399918, 258.7199797799999]], ['c', [406.2499981, 252.20497942999987], [395.97699833, 247.56097949999992], [387.7879982, 247.5569794999999]], ['c', [386.7209983, 247.5569794999999], [385.68999827, 247.63697947999992], [384.7019983, 247.8019794999999]], ['l', [384.7669983, 248.1909794999999]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Lypd1', 'point': [452.10173249480397, 232.17214500255994], 'bbox': [446.47039794, 230.10514509255995, 457.733067049608, 234.23914491255994], 'path': [['m', [384.90200805, 248.10101367999994]], ['c', [388.55800795, 242.22601367999994], [396.04500769000003, 237.22001411999997], [404.92800711, 233.6990132599999]], ['c', [413.80900763, 230.17501307999999], [424.08000945000003, 228.13001300999997], [433.29100798, 228.13001300999997]], ['c', [440.75800705, 228.12901355999998], [447.52600859, 229.47601367999994], [452.28601073, 232.44301367999992]], ['l', [452.49501036000004, 232.10801364999998]], ['c', [447.6360092, 229.08301403999997], [440.80500792, 227.73601390999988], [433.29100798, 227.7350144599999]], ['c', [424.02200698, 227.73601390999988], [413.71100806, 229.79001285999993], [404.78200721, 233.33101322999994]], ['c', [395.85300826, 236.87501384999996], [388.30200815, 241.90101387999994], [384.56700805, 247.89201368]], ['l', [384.90200805, 248.10101367999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'H3f3a', 'point': [415.66477391326396, 235.01331451336], 'bbox': [409.86660766, 232.84041455336, 421.46294016652797, 237.18621447336], 'path': [['m', [384.91799926, 248.07000782]], ['c', [386.35499928999997, 244.46000791999995], [390.40399931999997, 240.93900778], [395.28099917, 238.34400799000002]], ['c', [400.15499972, 235.74600747], [405.85299871999996, 234.06600810999998], [410.55399893, 234.06700757]], ['c', [412.43099974, 234.06700757], [414.1479988, 234.33500817000004], [415.58799933, 234.91100742000003]], ['l', [415.73499869, 234.54400780000003]], ['c', [414.23299979, 233.94400742000005], [412.46799849, 233.67300751000005], [410.55399893, 233.67200806000005]], ['c', [405.76099967, 233.67400792], [400.02099893999997, 235.37300828000002], [395.0949993, 237.99600747]], ['c', [390.17099903999997, 240.62200786999995], [386.05699931, 244.16500785000005], [384.55099924999996, 247.92300781999995]], ['l', [384.91799926, 248.07000782]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Vav3', 'point': [440.62342571883994, 260.32312896336], 'bbox': [435.67138671, 258.15022900336, 445.57546472767996, 262.49602892336], 'path': [['m', [384.8420105, 248.16302539999992]], ['c', [388.88401031, 245.55302549999988], [394.7550106, 244.36402537999993], [401.22200966, 244.36602546999995]], ['c', [408.72200966, 244.36502529999996], [417.02201080000003, 245.9600253399999], [424.2000122, 248.7360254099999]], ['c', [431.37900924, 251.5110254399999], [437.43201065, 255.47402526999997], [440.45401001, 260.16602564999994]], ['l', [440.78601074, 259.9510254999999]], ['c', [437.68500900000004, 255.1470255999999], [431.56700897, 251.16402531999995], [424.3420105, 248.36802539999996]], ['c', [417.11600876, 245.57302548999996], [408.77801132, 243.97102548999987], [401.22200966, 243.97102548999987]], ['c', [394.70801067, 243.9720254199999], [388.77201056, 245.15902545999995], [384.6280105, 247.83002539999995]], ['l', [384.8420105, 248.16302539999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Bhlhe22', 'point': [434.266184673216, 227.52430084336004], 'bbox': [426.29251098, 225.35140088336004, 442.239858366432, 229.69720080336003], 'path': [['m', [384.91699218, 248.0709843799999]], ['c', [387.26799225, 242.3239841699999], [393.7309923, 236.78998424999986], [401.49799155000005, 232.71598483999992]], ['c', [409.26199149, 228.63898517999996], [418.32599258000005, 226.01298381999993], [425.84599303000005, 226.01398517999996]], ['c', [428.96099089, 226.01398517999996], [431.80899048000003, 226.4649834799999], [434.18899153, 227.43998386999988]], ['l', [434.33899306, 227.07398463999994]], ['c', [431.89599228000003, 226.07398463999994], [428.99799346000003, 225.61998416999995], [425.84599303000005, 225.61898471999996]], ['c', [418.23599242, 225.62098361999995], [409.12999152000003, 228.26398517999996], [401.31399153, 232.36598445999994]], ['c', [393.49999236, 236.47098399999993], [386.97199224, 242.02698423999993], [384.55199218, 247.9219843799999]], ['l', [384.91699218, 248.0709843799999]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Runx1t1', 'point': [409.813794761016, 258.22223785336007], 'bbox': [401.72061157, 256.0493378933601, 417.90697795203204, 260.39513781336007], 'path': [['m', [384.81298828, 248.17797899999994]], ['c', [386.05998825, 247.63897902999997], [387.58398818, 247.38897902999997], [389.25998830000003, 247.38997901999994]], ['c', [393.00698852, 247.38797897999996], [397.50498866000004, 248.63897899999995], [401.36098861, 250.60197901999993]], ['c', [405.21898841, 252.56197881999992], [408.4289875, 255.24197911999988], [409.62998771, 258.03297853999993]], ['l', [409.99198913000004, 257.87597895]], ['c', [408.72098922000004, 254.94097900999998], [405.44498824000004, 252.2399787999999], [401.53998756, 250.24997901999996]], ['c', [397.63398838, 248.26297899999997], [393.09298801, 246.9959789599999], [389.25998830000003, 246.99397898999996]], ['c', [387.54498839, 246.99497903999998], [385.96998822, 247.24897897999995], [384.65598828000003, 247.81497899999988]], ['l', [384.81298828, 248.17797899999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Sema3a', 'point': [409.52446611104006, 275.58503082335994], 'bbox': [401.43258666, 273.41213086335995, 417.61634556208, 277.75793078335994], 'path': [['m', [384.7250061, 248.19397021999998]], ['c', [389.75300596999995, 248.42897022], [395.93800639999995, 252.48497009999994], [400.83200645, 257.79397059999997]], ['c', [405.73300551999995, 263.09697009], [409.34500692999995, 269.65097095], [409.33800696, 274.77796985]], ['c', [409.33800696, 274.94997072999996], [409.33400535, 275.12097026], [409.32600592999995, 275.28896953]], ['l', [409.72100637999995, 275.30797054]], ['c', [409.72900581, 275.13296938999997], [409.73400687, 274.95596935], [409.73400687, 274.77796985]], ['c', [409.7270069, 269.47097064], [406.06100654, 262.88196994], [401.12300681, 257.52596999]], ['c', [396.17800616, 252.17597032999993], [389.96200608, 248.05797022000002], [384.7440061, 247.79897021]], ['l', [384.7250061, 248.19397021999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Hgf', 'point': [435.41528772774404, 215.20630767336002], 'bbox': [432.03338623, 213.03340771336002, 438.79718922548807, 217.37920763336], 'path': [['m', [384.92800903, 248.03802539999992]], ['c', [386.55800903, 240.47002556999996], [393.83500861, 232.21202518999996], [402.8480091, 225.87502529999995]], ['c', [411.85600853, 219.5330262399999], [422.59600829, 215.10602619999997], [431.08900833, 215.10902454999996]], ['c', [432.59400939, 215.10902454999996], [434.02901076, 215.24702503999993], [435.37100983, 215.53802539999992]], ['l', [435.45401, 215.15202381999995]], ['c', [434.08100890000003, 214.85502674999987], [432.61800765, 214.71402408999995], [431.08900833, 214.71402408999995]], ['c', [422.46300887, 214.7170262599999], [411.68200874, 219.17802478999988], [402.62000846, 225.5520253399999]], ['c', [393.56300925, 231.9300255999999], [386.22200905, 240.21402551999995], [384.54200903, 247.95502539999995]], ['l', [384.92800903, 248.03802539999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Gpr12', 'point': [423.68097555884003, 212.90430572336004], 'bbox': [417.64248657, 210.73140576336004, 429.71946454768, 215.07720568336003], 'path': [['m', [384.93200683, 248.00701954]], ['c', [385.30900683, 240.84601977], [391.10500668, 232.11301948999994], [398.66500663, 225.20201924000003]], ['c', [406.22000692999995, 218.28601887000002], [415.53700637, 213.19101764000004], [422.85400770999996, 213.19701816999998]], ['c', [423.12900543, 213.19701816999998], [423.40000533, 213.20401814000002], [423.66900633999995, 213.21902134000004]], ['l', [423.6910057, 212.82402089000004]], ['c', [423.41400526999996, 212.80901768], [423.13500594, 212.80201770999997], [422.85400770999996, 212.80201770999997]], ['c', [415.35900687, 212.80801823000002], [406.00600622999997, 217.95601894000004], [398.39900682999996, 224.91101887000002]], ['c', [390.7970066, 231.87101986999994], [384.93900683, 240.63601973000004], [384.53700682, 247.98601954000003]], ['l', [384.93200683, 248.00701954]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Prr15', 'point': [444.7351342022, 249.10224272336006], 'bbox': [439.4223938, 246.92934276336007, 450.0478746044, 251.27514268336006], 'path': [['m', [384.87200927, 248.13800098000002]], ['c', [390.71300935, 242.44900086000007], [402.22200964999996, 239.60300114000006], [413.85200881, 239.60800125000003]], ['c', [426.15200804, 239.60300114000006], [438.58501051999997, 242.78600074999997], [444.59201049, 248.97500099]], ['l', [444.87500761999996, 248.70000095]], ['c', [438.71900939, 242.38000108000006], [426.21900939, 239.21700146], [413.85200881, 239.21300078000002]], ['c', [402.15900991999996, 239.21700146], [390.5820093, 242.04400111999996], [384.59700927, 247.85500098]], ['l', [384.87200927, 248.13800098000002]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Galntl6', 'point': [454.981455833256, 256.27924469336006], 'bbox': [448.09640502, 254.10634473336006, 461.866506646512, 258.45214465336005], 'path': [['m', [384.8580017, 248.15100146999998]], ['c', [390.88200186, 243.35300162], [400.94300078, 241.11600163000003], [411.7140026, 241.11900141]], ['c', [428.63400267, 241.11100150999994], [447.29700087, 246.63800145999994], [454.82199858, 256.11600162]], ['l', [455.13100432, 255.87000133000004]], ['c', [447.44300078, 246.22600151999995], [428.71700285, 240.73100139999997], [411.7140026, 240.7230015]], ['c', [400.88900183, 240.72600126999998], [390.76000165, 242.95500137], [384.6120017, 247.84200149000003]], ['l', [384.8580017, 248.15100146999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Acsl1', 'point': [450.3621483522, 267.26709869336], 'bbox': [445.04940795, 265.09419873336003, 455.6748887544, 269.43999865336], 'path': [['m', [384.82998657, 248.16900683999995]], ['c', [388.97398662, 245.88600683999994], [394.50498677, 244.84800673999996], [400.62298679, 244.84900690999996]], ['c', [409.92998695, 244.84800673999996], [420.58998489000004, 247.24500681999996], [429.7869873, 251.23000693999995]], ['c', [438.98598479000003, 255.2120070599999], [446.71798705000003, 260.78900672], [450.18898772, 267.0950074399999]], ['l', [450.53498840000003, 266.9050069]], ['c', [446.98698807, 260.47200728999996], [439.18998718, 254.87500668999996], [429.94398498000004, 250.86700679]], ['c', [420.69698714000003, 246.8620067999999], [409.99698638, 244.45400693], [400.62298679, 244.45400693]], ['c', [394.46098613000004, 244.45400693], [388.86998653, 245.49400688999992], [384.63898657000004, 247.82300685999996]], ['l', [384.82998657, 248.16900683999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Rnf122', 'point': [405.27320903102, 240.44025542336], 'bbox': [398.26669311, 238.26735546336, 412.27972495204, 242.61315538335998], 'path': [['m', [384.91500854, 248.07702685999993]], ['c', [385.90400855999997, 245.84802675999993], [388.50200844, 243.7160267999999], [391.62200830999996, 242.16302682999992]], ['c', [394.74100875, 240.60702705999995], [398.37800883999995, 239.62302638999995], [401.41900825, 239.6240267999999]], ['c', [402.84800909999996, 239.6240267999999], [404.14400864, 239.8410267999999], [405.19000815, 240.3080268099999]], ['l', [405.35100936, 239.94702674999996]], ['c', [404.23500823999996, 239.45002700999999], [402.88800811, 239.22902728999998], [401.41900825, 239.22902728999998]], ['c', [398.29500865, 239.23002673999997], [394.61500834, 240.23002673999997], [391.44600866999997, 241.81002665999995]], ['c', [388.27900861999996, 243.39202691999992], [385.61600851, 245.54502679999996], [384.55400854, 247.91602686]], ['l', [384.91500854, 248.07702685999993]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Gm2516', 'point': [427.71574959658005, 258.04615142336], 'bbox': [419.3815918, 255.87325146336002, 436.04990739316, 260.21905138336], 'path': [['m', [384.83999633, 248.16400195999995]], ['c', [387.87399624, 246.25100197999996], [392.22999619, 245.37700201999996], [397.03699587, 245.37800195]], ['c', [402.85299681000004, 245.37800195], [409.32699584, 246.65200199999992], [414.92299651, 248.85500193999997]], ['c', [420.51999663000004, 251.05600190999996], [425.23399733, 254.19100190999995], [427.54499816000003, 257.8650021699999]], ['l', [427.87899779, 257.65500212999996]], ['c', [425.48899459, 253.8650021699999], [420.71099471, 250.71000196], [415.06799697, 248.4880019599999]], ['c', [409.42399596, 246.26600195999993], [402.90999603, 244.9830019599999], [397.03699587, 244.9830019599999]], ['c', [392.18299626000004, 244.98400189999995], [387.76399635, 245.85700204], [384.62899633, 247.8290019599999]], ['l', [384.83999633, 248.16400195999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Syn3', 'point': [419.74662518883997, 281.57102325335995], 'bbox': [414.79458618, 279.39812329335996, 424.69866419768, 283.74392321335995], 'path': [['m', [384.7380066, 248.19397021999998]], ['l', [384.9970066, 248.19197022000003]], ['c', [391.77700663, 248.18497021999997], [400.38200665, 253.17097045000003], [407.29200745, 259.87496996999994]], ['c', [414.20900727000003, 266.57197047], [419.42800522, 274.98597003], [419.5470047, 281.71897174]], ['l', [419.94200516, 281.71197177]], ['c', [419.80500794, 274.78596927], [414.5240059, 266.33997012], [407.56700707, 259.59097052999994]], ['c', [400.60500622, 252.84797001999993], [391.96500683, 247.80297022000002], [384.9970066, 247.79597021999996]], ['l', [384.7310066, 247.79897021]], ['l', [384.7380066, 248.19397021999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Myt1l', 'point': [441.36555751443996, 237.00721100336], 'bbox': [436.17449951, 234.83431104336, 446.55661551887994, 239.18011096336], 'path': [['m', [384.89801025, 248.10699511999997]], ['c', [391.0340104, 239.04499484999997], [408.66101073000004, 233.09199477999994], [423.53001021, 233.10199501]], ['c', [430.56201171000004, 233.09999515999993], [436.97301101000005, 234.42899466999995], [441.25201035000003, 237.33699466999997]], ['l', [441.47401046000004, 237.009995]], ['c', [437.08800887, 234.0339951699999], [430.6110115, 232.70799495999995], [423.53001021, 232.70699550999996]], ['c', [408.5600109, 232.71599531999993], [390.87401008, 238.63099529], [384.57101025000003, 247.88499511999999]], ['l', [384.89801025, 248.10699511999997]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Sptb', 'point': [405.52768864773196, 251.24720122335998], 'bbox': [401.0586853, 249.07430126335998, 409.99669199546395, 253.42010118335998], 'path': [['m', [384.8500061, 248.15698292000002]], ['c', [386.46300613, 246.98998287000006], [389.01300619999995, 246.44098296000004], [391.79800604999997, 246.44298293999998]], ['c', [397.06900595999997, 246.43398288000003], [403.16800689999997, 248.40998292000006], [405.36400603, 251.47998287000007]], ['l', [405.68500517999996, 251.24898292]], ['c', [403.32200622, 247.99698292000005], [397.16300581999997, 246.05598284999996], [391.79800604999997, 246.04798296]], ['c', [388.96300600999996, 246.04898288000004], [386.34600615, 246.59398296000006], [384.6190061, 247.83598292]], ['l', [384.8500061, 248.15698292000002]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Chrm3', 'point': [432.674079539984, 247.93421295335997], 'bbox': [426.1545105, 245.76131299335998, 439.193648579968, 250.10711291335997], 'path': [['m', [384.875, 248.13501025999994]], ['c', [389.62799978, 243.3010101499999], [399.375, 240.82301045999998], [409.06900024, 240.8270101799999]], ['c', [418.48500061, 240.82301045999998], [427.84799957, 243.16101028999992], [432.53400039999997, 247.78801025999996]], ['l', [432.81200027, 247.50701024]], ['c', [427.97999954, 242.75501014999998], [418.54899978000003, 240.4370103099999], [409.06900024, 240.43201019999992]], ['c', [399.30900001, 240.4370103099999], [389.49599981, 242.89401008999994], [384.59400002, 247.8580102499999]], ['l', [384.875, 248.13501025999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Pou6f2', 'point': [418.55852739324, 246.85724762335997], 'bbox': [411.67260742, 244.68434766335997, 425.44444736648, 249.03014758335996], 'path': [['m', [384.88000488, 248.13000537999994]], ['c', [388.31400489, 244.38000537999994], [395.85000514, 242.3670053699999], [403.13300513, 242.37200547999998]], ['c', [409.28800582, 242.36900522999997], [415.26000404, 243.80500557999994], [418.41900634, 246.71700539999995]], ['l', [418.68700408, 246.42600538]], ['c', [415.3880043, 243.40200521999998], [409.34800529, 241.98100521999993], [403.13300513, 241.9770055099999]], ['c', [395.77900504, 241.98300554999992], [388.17700481, 243.97200537999993], [384.58900488, 247.8630053999999]], ['l', [384.88000488, 248.13000537999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Fam65b', 'point': [427.05886894326, 238.62024810336004], 'bbox': [419.08758544, 236.44734814336005, 435.03015244652, 240.79314806336004], 'path': [['m', [384.90200805, 248.10199023999996]], ['c', [387.17600797, 244.47999026999992], [391.82100819, 241.38799001999996], [397.33700846, 239.2169900199999]], ['c', [402.85200881000003, 237.04399062999994], [409.23300742000004, 235.78599025999995], [414.96200751000003, 235.78599025999995]], ['c', [419.68700789, 235.78498986], [423.96700667, 236.64299060999997], [426.95100783, 238.52199032]], ['l', [427.16100692, 238.18799069]], ['c', [424.07800673, 236.24999000999992], [419.73400878, 235.39199021000002], [414.96200751000003, 235.39098979999994]], ['c', [409.17500876, 235.39199021000002], [402.75600815, 236.65798999999993], [397.19200801, 238.84898998999995]], ['c', [391.63100815, 241.04299023], [386.9210081, 244.1539902899999], [384.56700805, 247.89099023999995]], ['l', [384.90200805, 248.10199023999996]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'D130009I18Rik', 'point': [402.146551924224, 230.45234039336003], 'bbox': [387.16571044, 228.27944043336004, 417.12739340844803, 232.62524035336003], 'path': [['m', [384.93200683, 247.99401905000002]], ['l', [384.93200683, 247.91501905000007]], ['c', [384.92500683, 244.48801901000002], [387.49200677, 240.09401895000008], [390.94700668999997, 236.57901908999997]], ['c', [394.39500712999995, 233.05701878000002], [398.73000715999996, 230.41701939000006], [402.14200590999997, 230.3850198]], ['l', [402.13800619999995, 229.99001934]], ['c', [398.53300665999996, 230.03901913000004], [394.16500663, 232.7410188], [390.66500663, 236.30201867000005]], ['c', [387.17100690999996, 239.87001944999997], [384.54400682, 244.29801895000003], [384.53700682, 247.91501905000007]], ['l', [384.53700682, 247.99901905000002]], ['l', [384.93200683, 247.99401905000002]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Pou4f1', 'point': [442.052851574745, 276.09286034256], 'bbox': [435.50149536, 274.02586043255997, 448.60420778949, 278.15986025256], 'path': [['m', [384.7999878, 248.18298388999995]], ['c', [387.19898772, 247.34298391999994], [389.95698785999997, 246.95198395], [392.93998813999997, 246.95198395]], ['c', [401.94998741, 246.94998384999997], [413.00698853, 250.5229837999999], [422.43498612, 255.92498399]], ['c', [431.8659897, 261.32298423], [439.66398621, 268.55798388999995], [442.1979866, 275.80698442999994]], ['l', [442.56998825, 275.67598391999996]], ['c', [439.97198868, 268.27198459], [432.1089859, 261.01398421], [422.63198853, 255.58198404999996]], ['c', [413.15098763, 250.15298391], [402.0519886, 246.55798388999995], [392.93998813999997, 246.55698384999994]], ['c', [389.92198801, 246.55698384999994], [387.12198782, 246.95098388999997], [384.6689878, 247.8099838899999]], ['l', [384.7999878, 248.18298388999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Tnfrsf19', 'point': [420.20380911328, 269.61512115335995], 'bbox': [412.23361206, 267.44222119335996, 428.17400616656, 271.78802111335995], 'path': [['m', [384.78201293, 248.18798876999995]], ['c', [385.84701298, 247.92298878999998], [387.00501297, 247.79698876999998], [388.23101281, 247.79698876999998]], ['c', [394.04501246, 247.79398876999994], [401.37701224, 250.6419887699999], [407.55301283, 254.77498866999997]], ['c', [413.73301314, 258.90298892], [418.75401305, 264.32298899], [420.00601195, 269.37698793999994]], ['l', [420.38901137, 269.2809891799999]], ['c', [419.08401106, 264.0589890499999], [413.99801253, 258.61398840999993], [407.77301215, 254.44698858999993]], ['c', [401.54401206, 250.28298879], [394.17101286, 247.40498876999993], [388.23101281, 247.40198876]], ['c', [386.97801303, 247.40198876], [385.78901289, 247.5309887699999], [384.68701293, 247.80498878999992]], ['l', [384.78201293, 248.18798876999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Dcbld2', 'point': [405.62584035548, 264.29914215336], 'bbox': [398.86160278, 262.12624219336, 412.39007793096, 266.47204211336], 'path': [['m', [384.76000976, 248.19201708999992]], ['c', [385.09900974, 248.14801708999994], [385.45100974, 248.1260170899999], [385.8120098, 248.1260170899999]], ['c', [389.46600961, 248.1210170899999], [394.16700934, 250.36601710999992], [398.0450096, 253.4950170599999]], ['c', [401.92700957, 256.61901713], [404.98200988, 260.6300168099999], [405.42700957, 264.05701686]], ['l', [405.81801032, 264.0050168099999]], ['c', [405.33700942, 260.3960175599999], [402.21800994, 256.3560166399999], [398.29300976, 253.18701696999995]], ['c', [394.36200999, 250.02401709999992], [389.62200974, 247.73501710999994], [385.8120098, 247.7310170899999]], ['c', [385.43500977, 247.7310170899999], [385.06600976, 247.75401708999993], [384.70900976, 247.80101708999996]], ['l', [384.76000976, 248.19201708999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Runx1', 'point': [436.936950439864, 269.23595725256], 'bbox': [430.9624939, 267.16895734256, 442.911406979728, 271.30295716256], 'path': [['m', [384.81298828, 248.17797899999994]], ['c', [387.45498823, 247.0409790399999], [390.66298818, 246.51697898999998], [394.18198871000004, 246.5179790399999]], ['c', [402.07898902, 246.51697898999998], [411.53098869, 249.15297901999998], [419.63998794, 253.28597879999995]], ['c', [427.750988, 257.4169793199999], [434.51298904000004, 263.05297899999994], [437.06198882, 268.9849781999999]], ['l', [437.42498779, 268.82897805999994]], ['c', [434.80498885000003, 262.75297879999994], [427.97698974, 257.09497880999993], [419.81898879, 252.93397903999994]], ['c', [411.65898895000004, 248.77697896999996], [402.16498756, 246.12397909999993], [394.18198871000004, 246.12297893999994]], ['c', [390.62398815, 246.12297893999994], [387.36498832, 246.64997899999992], [384.65698828, 247.81497899999988]], ['l', [384.81298828, 248.17797899999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Prkce', 'point': [455.57032022776, 243.81525542336], 'bbox': [450.01638793, 241.64235546336, 461.12425252552, 245.98815538335998], 'path': [['m', [384.8829956, 248.12701464999998]], ['c', [392.26999568, 239.73701430999995], [408.93899535, 235.17601444000002], [424.76999663000004, 235.18201494999994]], ['c', [437.09799575, 235.17901469999993], [448.91399382000003, 237.94301464], [455.43799590000003, 243.70301486999995]], ['l', [455.6989975, 243.40701485]], ['c', [449.03999328000003, 237.54101420999996], [437.1559944, 234.79001474999995], [424.76999663000004, 234.78701449999994]], ['c', [408.86499595000004, 234.79301501999998], [392.12999581, 239.32701444999998], [384.5869956, 247.86601465]], ['l', [384.8829956, 248.12701464999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Tshz1', 'point': [422.11932123108, 225.07129791336], 'bbox': [416.20251464, 222.89839795336002, 428.03612782216, 227.24419787336], 'path': [['m', [384.92700195, 248.04101612]], ['c', [386.1940019, 242.60901595999997], [391.51800203, 236.73001624999995], [398.09100149999995, 232.23801566999998]], ['c', [404.66000175, 227.74201633999996], [412.47300147, 224.62601521], [418.65600205, 224.62901547]], ['c', [419.86500166999997, 224.62901547], [421.01100157999997, 224.74701549999997], [422.07300184999997, 224.99601604999998]], ['l', [422.16300200999996, 224.61101582000003]], ['c', [421.06700133, 224.35401585], [419.89100264999996, 224.23301556], [418.65600205, 224.23301556]], ['c', [412.34500120999996, 224.23601582000003], [404.49000166999997, 227.38501598999994], [397.86800193, 231.91201640999998]], ['c', [391.2500019, 236.44401600000003], [385.86100195999995, 242.34801629000003], [384.54200195, 247.95101611999996]], ['l', [384.92700195, 248.04101612]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Sgms1', 'point': [412.09867355552, 217.09345367336005], 'bbox': [405.33660888, 214.92055371336005, 418.86073823104005, 219.26635363336004], 'path': [['m', [384.93200683, 247.98498583999992]], ['c', [384.91800682999997, 247.75398583999993], [384.91200683, 247.5199858399999], [384.91200683, 247.28398586999992]], ['c', [384.90600682999997, 241.5469856399999], [388.87700676, 234.21798563999994], [394.2800064, 228.27098513999988]], ['c', [399.67700671, 222.31798602999993], [406.50600622999997, 217.7479863299999], [412.10800743, 217.42798663999997]], ['l', [412.08400724999996, 217.03298617999997]], ['c', [406.29400633999995, 217.3789849399999], [399.43300723, 222.01098490999993], [393.98700713, 228.00498629999993]], ['c', [388.54700683, 234.00498629999993], [384.52200682999995, 241.36998606999987], [384.51700682999996, 247.28398586999992]], ['c', [384.51700682999996, 247.52798585999994], [384.52300683, 247.7689858399999], [384.53700682, 248.00798583999995]], ['l', [384.93200683, 247.98498583999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Crtac1', 'point': [445.5786032222, 222.91431549336005], 'bbox': [439.17941284, 220.74141553336005, 451.9777936044, 225.08721545336005], 'path': [['m', [384.91699218, 248.07202197999993]], ['c', [387.81799220000005, 241.01102210999989], [395.75599192000004, 234.23402167999996], [405.29099272, 229.24402286999998]], ['c', [414.82299231, 224.2520222799999], [425.94699095000004, 221.03902103999997], [435.18699263, 221.04102184999988]], ['c', [439.03699111000003, 221.04002239999988], [442.55899048000003, 221.59802104999994], [445.50099181, 222.80902149999997]], ['l', [445.65199279, 222.44302226999991]], ['c', [442.64599228000003, 221.20702220999988], [439.07499312000004, 220.64602137999998], [435.18699263, 220.64502193999988]], ['c', [425.8569908, 220.6470227499999], [414.69099234000004, 223.8770222799999], [405.10699271000004, 228.89402248]], ['c', [395.52599238000005, 233.91402197999992], [387.52299212, 240.71402217999992], [384.55199218, 247.92102197999998]], ['l', [384.91699218, 248.07202197999993]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}]}, 'current_label_offsets': {'Foxo6': {'Cplx2': [0, 1.15], 'Mfap4': [0, -1.15], 'Pcdh10': [0, 1.35], 'Nhlh1': [0, -1.35]}, 'Zfpm2': {'Shox2': [1.5, 0], 'Nhsl1': [-1.5, 0], 'Ank2': [1.35, 0], 'Soga3': [-1.35, 0], 'Elavl3': [0, 0.8], 'Epha5': [0, -0.8], 'Rab3a': [0, 1.15], 'Map1b': [0, -1.15], 'Mapk8ip1': [0, 1.55], 'Tubb3': [0, -1.55], '3110039M20Rik': [0, 1.9], 'Afap1': [0, -1.9]}, 'Neurod2': {'Neurod6': [0, 1.35], 'Myo6': [0, -1.35], 'Kcnmb2': [1.55, 0], 'Mreg': [-1.55, 0]}, 'Nhlh1': {'1700037H04Rik': [0, 1.55], 'Ypel3': [0, -1.55], 'Nhlh2': [1.5, 0], 'Trp53i11': [-1.5, 0]}, 'Neurod1': {'Jun': [0, 0.241], 'Ntm': [-3.244, 0], 'Vcan': [0.973, 0], 'Edil3': [-0.762, 0.728], 'Cd81': [1.065, 3.971], 'Lrrn3': [-5.092, 0], 'Uchl1': [-1.284, 0], 'Acsl1': [-3.441, -1.537], 'Olfm1': [1.6, 0], 'Aplp1': [-1.6, 0], 'Chst2': [0.762, 0], 'Elavl2': [0, -3.499], 'Pcsk2': [4.538, 0.475], 'Ckmt1': [0, 1.106], 'Scn3a': [-0.973, 0.431], 'Unc5c': [1.284, -0.479], 'Snx18': [0, -0.475], 'Ddah1': [0, 0.804], 'Gnao1': [0, 0.516], 'Ywhah': [0, -0.037], 'Tmeff2': [-1.065, 0], 'Rbms1': [11.147, 0], 'Pou6f2': [0, -0.804], 'Ctnna2': [5.092, 0], 'Spock1': [0, -0.35], 'Hsp90aa1': [0, -1.09], 'C1ql3': [0, 0.874], 'Olfm2': [-1.814, 0], 'Rbms3': [0, -0.874], 'Cacng5': [1.814, 0], 'Sncg': [0, 2.5], 'Mllt11': [0, -2.5]}}}

def draw_module_network(ax, factor, left_genes, right_genes, kind='associated',
                        left_color='#2D7184', right_color='#5AAD7F', font_size=11.4,
                        highlight_genes=tuple()):
    """Substitute current module genes into the published Fig.3d vector layout.

    Original slot names are visual provenance only. Every current input member
    retains exactly one ribbon; template geometry carries no biological weight.
    Side-color parameters apply to ribbons only; outer panels are untouched.
    """
    #1. =========== Original vector template and appearance
    padding_fraction = .025
    halo_width_pt = .85
    highlight_color = '#BA001C'
    template = PAPER_NETWORK_TEMPLATE
    groups = [list(map(str, left_genes)), list(map(str, right_genes))]
    highlights = set(map(str, highlight_genes))
    if any(len(set(g)) != len(g) for g in groups):
        raise ValueError('A displayed module contains duplicate genes.')
    if font_size <= 0:
        raise ValueError('font_size must be positive.')
    ax.set_axis_off(); ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.set_facecolor('white')
    fig = ax.figure
    fig.canvas.draw()
    renderer = fig.canvas.get_renderer()
    area = ax.get_window_extent(renderer)
    point_scale = 72.0 / fig.dpi
    width, height = area.width * point_scale, area.height * point_scale
    bbox = np.asarray(template['bbox'], dtype=float)
    scale = min(width * (1 - 2 * padding_fraction) / (bbox[2] - bbox[0]),
                height * (1 - 2 * padding_fraction) / (bbox[3] - bbox[1]))
    used = (bbox[2:] - bbox[:2]) * scale
    offset = (np.array([width, height]) - used) / 2
    def xy(points):
        a = np.asarray(points, dtype=float)
        result = (a - bbox[:2]) * scale + offset
        result[..., 1] = height - result[..., 1]
        return result / np.array([width, height])
    def angle(slot, side):
        delta = np.asarray(slot['point']) - np.asarray(template['center'])
        return np.arctan2(delta[1], -delta[0] if side == 'left' else delta[0])
    def topology(slot):
        return tuple(step[0] for step in slot['path'])
    def blend(a, b):
        path = [[sa[0]] + [((np.asarray(pa) + np.asarray(pb)) / 2).tolist()
                for pa, pb in zip(sa[1:], sb[1:])]
                for sa, sb in zip(a['path'], b['path'])]
        return {'visual_slot': [a['visual_slot'], b['visual_slot']],
            'point': ((np.asarray(a['point']) + np.asarray(b['point'])) / 2).tolist(),
            'bbox': ((np.asarray(a['bbox']) + np.asarray(b['bbox'])) / 2).tolist(),
            'path': path, 'rgb': a['rgb'], 'interpolated': True}
    def choose_slots(side, count):
        slots = list(template['slots'][side])
        if count <= len(slots):
            if count == 0: return []
            points = np.asarray([slot['point'] for slot in slots])
            # Preserve the actual top, bottom and outer boundary before filling
            # the remaining slots from the original two-dimensional layout.
            outer = np.argmin(points[:, 0]) if side == 'left' else np.argmax(points[:, 0])
            chosen = list(dict.fromkeys([int(np.argmin(points[:, 1])),
                                         int(np.argmax(points[:, 1])), int(outer)]))[:count]
            while len(chosen) < count:
                distances = np.linalg.norm((points[:, None] - points[chosen][None])
                                           * np.array([.8, 1.0]), axis=2).min(axis=1)
                distances[chosen] = -1
                chosen.append(int(np.argmax(distances)))
            return [slots[i] for i in chosen]
        # Additional members use local interpolation of neighboring original vector
        # ribbons with identical command topology; no ellipse is generated.
        while len(slots) < count:
            slots.sort(key=lambda s: angle(s, side))
            points = np.asarray([slot['point'] for slot in slots])
            candidates = []
            boxes = np.asarray([slot['bbox'] for slot in slots])
            for i, a in enumerate(slots):
                for j in range(i + 1, len(slots)):
                    b = slots[j]
                    if topology(a) != topology(b) or abs(angle(a, side) - angle(b, side)) > .65:
                        continue
                    box = (np.asarray(a['bbox']) + np.asarray(b['bbox'])) / 2
                    dx = np.maximum(0, np.maximum(box[0] - boxes[:, 2], boxes[:, 0] - box[2]))
                    dy = np.maximum(0, np.maximum(box[1] - boxes[:, 3], boxes[:, 1] - box[3]))
                    ox = np.maximum(0, np.minimum(box[2], boxes[:, 2]) - np.maximum(box[0], boxes[:, 0]))
                    oy = np.maximum(0, np.minimum(box[3], boxes[:, 3]) - np.maximum(box[1], boxes[:, 1]))
                    # Insert into genuine local whitespace between neighboring
                    # source ribbons, rather than halfway through existing words.
                    score = np.hypot(dx, dy).min() - 100 * (ox * oy).sum()
                    candidates.append((float(score), i, j))
            if not candidates:
                raise ValueError('No compatible original ribbon pair for interpolation.')
            _, i, j = max(candidates)
            slots.append(blend(slots[i], slots[j]))
        return slots
    def mpl_path(commands):
        vertices, codes = [], []
        for step in commands:
            code = {'m': MplPath.MOVETO, 'l': MplPath.LINETO, 'c': MplPath.CURVE4}[step[0]]
            vertices.extend(step[1:]); codes.extend([code] * len(step[1:]))
        vertices.append(vertices[0]); codes.append(MplPath.CLOSEPOLY)
        return MplPath(xy(vertices), codes)
    stroke = [pe.withStroke(linewidth=halo_width_pt, foreground='white')]
    #2. =========== Original gray guide, ribbons and replacement gene labels
    for guide in template['guides']:
        points = xy(guide['pts'])
        line, = ax.plot(points[:, 0], points[:, 1], transform=ax.transAxes,
            color=guide['stroking_color'], linewidth=guide['linewidth'] * scale,
            solid_capstyle='butt', zorder=.2, clip_on=True)
        dash, phase = guide['dash']
        if dash:
            line.set_linestyle((phase / guide['linewidth'],
                                tuple(float(v) / guide['linewidth'] for v in dash)))
    factor_point = xy(template['center'])
    factor_text = ax.text(*factor_point, str(factor), transform=ax.transAxes,
        ha='center', va='center', fontfamily='Arial', fontstyle='italic',
        fontweight='normal', fontsize=font_size, color='black', path_effects=stroke,
        zorder=6, clip_on=False)
    records, patches, edge_records = [], [], []
    for side, genes in zip(['left', 'right'], groups):
        slots = choose_slots(side, len(genes))
        texts = [ax.text(.5, .5, gene, transform=ax.transAxes, ha='center', va='center',
            fontfamily='Arial', fontstyle='italic',
            fontsize=font_size * (template['red_font_size'] / template['base_font_size'] if gene in highlights else 1),
            color=highlight_color if gene in highlights else 'black', path_effects=stroke,
            zorder=5, clip_on=False) for gene in genes]
        fig.canvas.draw(); renderer = fig.canvas.get_renderer()
        # Match long replacement names to the genuinely wider original text
        # slots; this preserves the published geometry rather than repacking it.
        gene_widths = [TextPath((0, 0), gene, prop=FontProperties(family='Arial', style='italic', size=text.get_fontsize())).get_extents().width for gene, text in zip(genes, texts)]
        gene_order = sorted(range(len(genes)), key=lambda i: (gene_widths[i], genes[i]))
        slot_order = sorted(range(len(slots)), key=lambda i: (slots[i]['bbox'][2] - slots[i]['bbox'][0], i))
        for gi, si in zip(gene_order, slot_order):
            gene, text, slot = genes[gi], texts[gi], slots[si]
            shift = template.get('current_label_offsets', {}).get(str(factor), {}).get(gene, (0, 0))
            point = xy(slot['point']) + np.asarray(shift) / np.array([width, height])
            text.set_position(point)
            patch = PathPatch(mpl_path(slot['path']), transform=ax.transAxes,
                facecolor=left_color if side == 'left' else right_color, edgecolor='none', linewidth=0, alpha=1,
                zorder=1, clip_on=True)
            ax.add_patch(patch); patches.append(patch)
            records.append({'side': side, 'gene': gene, 'text': text, 'point': point})
            edge_records.append({'side': side, 'gene': gene,
                'visual_slot': slot['visual_slot'], 'template_interpolated': slot['interpolated'],
                'label_offset_points': list(shift)})
    #3. =========== Report actual layout and the fixed visual provenance
    fig.canvas.draw(); renderer = fig.canvas.get_renderer()
    named = [(str(factor), factor_text)] + [(r['side'] + ':' + r['gene'], r['text']) for r in records]
    boxes = [(name, text.get_window_extent(renderer)) for name, text in named]
    overlaps, clipped = [], []
    for i, (name, box) in enumerate(boxes):
        if box.x0 < area.x0 or box.y0 < area.y0 or box.x1 > area.x1 or box.y1 > area.y1:
            clipped.append(name)
        for other, b in boxes[i + 1:]:
            if min(box.x1, b.x1) > max(box.x0, b.x0) and min(box.y1, b.y1) > max(box.y0, b.y0):
                overlaps.append((name, other))
    positions = {'factor': tuple(map(float, factor_point)), 'left': {}, 'right': {}}
    for record in records:
        positions[record['side']][record['gene']] = tuple(map(float, record['point']))
    all_genes = set(groups[0]) | set(groups[1])
    reference = {'source': template['source'], 'path': template['path'],
        'sha256': template['sha256'], 'page_1based': template['page_1based'],
        'coordinate_system': 'Original PDF points, top-left origin; uniform affine transform',
        'original_slots': {'left': len(template['slots']['left']), 'right': len(template['slots']['right'])},
        'slot_mapping': edge_records,
        'slot_names_are_biological_targets': False,
        'runtime_pdf_or_Tmp_dependency': False,
        'local_text_offsets_points': template.get('current_label_offsets', {}).get(str(factor), {})}
    motif = str(kind).lower() in {'motif', 'motif_supported', 'early_motif_supported', 'regulon'}
    qa = {'ok': not (overlaps or clipped), 'factor': str(factor), 'kind': str(kind),
        'left_genes': len(groups[0]), 'right_genes': len(groups[1]),
        'gene_labels': len(records), 'edges': len(patches),
        'all_genes_connected': len(patches) == sum(map(len, groups)),
        'label_overlaps': overlaps, 'clipped_labels': clipped,
        'axes_width_inches': width / 72, 'axes_height_inches': height / 72,
        'font_size_points': float(font_size), 'factor_font_size_points': float(font_size),
        'highlighted_genes': sorted(highlights & all_genes),
        'requested_highlights_not_in_modules': sorted(highlights - all_genes),
        'highlight_semantics': 'Literature-discussed representatives; not statistical significance or stronger edges',
        'curve_geometry': 'Original filled Bezier ribbons; interpolation only for additional members',
        'paper_palette': {'left': '#2D7184', 'right': '#5AAD7F', 'red': highlight_color},
        'template_scale_points': float(scale),
        'reference_template': reference,
        'edge_semantics': 'Predicted motif-supported TF-target link' if motif else 'RNA module association'}
    return {'labeltexts': [text for _, text in named], 'edgepatches': patches,
        'positions': positions, 'factor_text': factor_text, 'edge_records': edge_records,
        'reference_template': reference, 'QA': qa}
```

```python
#7. =========== Literature-style joint B-C-D figure
def render_factor(result, factor, output_dir, seed=42):
    import matplotlib
    matplotlib.use('Agg')
    import matplotlib.pyplot as plt
    from matplotlib.colors import LinearSegmentedColormap, Normalize
    from matplotlib.lines import Line2D
    FIGSIZE = (19.2, 4.15)
    DPI = 600
    B_POINT_AREA = 1.5
    C_POINT_AREA = 3.1
    BACKGROUND_ALPHA = .20
    FOREGROUND_ALPHA = .90
    NETWORK_FONT = 11.4
    PALETTE = {'CVG': '#1A6D79', 'T': '#1A6D79', 'VGN': '#00008B',
        'AB': '#7A4E94', 'AC': '#3585BC', 'Ia': '#F37271',
        'Ib': '#44B9DE', 'Ic': '#97C8A8', 'II': '#E377C2'}
    LEFT_COLOR, RIGHT_COLOR = '#1A6D79', '#97C8A8'
    plt.rcParams.update({'font.family': 'Arial', 'font.size': 8, 'axes.linewidth': .65,
                         'xtick.direction': 'out', 'ytick.direction': 'out'})
    obs, xy = result['obs'], result['X_umap']
    modules = result['modules']
    chosen = result['selected_modules']
    titles = result['reviewed_annotations']['reviewed_program_titles_json']
    reviewed = list(titles.values()) if isinstance(titles, dict) else titles
    on_path = obs.seg.isin(('CVG', 'T', 'AC', 'Ic')).to_numpy()
    grid, grid_fit = result['fit_grid'], result['grid_fitted']
    chosen = sorted(chosen, key=lambda m: float(grid[np.nanargmax(grid_fit[:, modules[m]['job']])]))
    names = {}
    for m in chosen:
        gene_hash = hashlib.sha256('\n'.join(sorted(modules[m]['genes'])).encode()).hexdigest()
        found = [d for d in reviewed if d['TF'] == factor and d['gene_sha256'] == gene_hash]
        assert len(found) == 1, 'Review descriptive titles after changing gene sets'
        names[m] = found[0]['title']
    fig = plt.figure(figsize=FIGSIZE)
    text_items, curve_axes = [], []
    cm_b = LinearSegmentedColormap.from_list('activity', ['#D4D4D4', '#FFFFFF', '#F3151D'])
    bscore = result['B_raw'] if factor == 'Prox1' else result['fitted'][:, 0]
    if factor == 'Nhlh1':
        bmax, metric, btitle = math.ceil(np.nanmax(bscore) * 20) / 20, 'Regulon AUC', r'$\it{Nhlh1}(+)$ activity'
    elif factor == 'Prox1':
        bmax, metric, btitle = math.ceil(np.nanmax(bscore) * 2) / 2, 'RNA: log10(1 + CPM)', r'$\it{Prox1}$ RNA'
    else:
        bmax, metric, btitle = .70, 'Associated expression score', r'$\it{' + factor + '}$ associated program'
    lim = (xy.min(axis=0) - 1, xy.max(axis=0) + 1)
    def embedding(ax, values, cmap, vmax, mask=None):
        ax.scatter(xy[:, 0], xy[:, 1], s=B_POINT_AREA, c='#D4D4D4', alpha=.55, linewidths=0)
        mask = np.isfinite(values) if mask is None else mask & np.isfinite(values)
        order = np.flatnonzero(mask)[np.argsort(values[mask], kind='stable')]
        sc = ax.scatter(xy[order, 0], xy[order, 1], s=B_POINT_AREA, c=values[order],
            cmap=cmap, norm=Normalize(0, vmax, clip=True), alpha=.88, linewidths=0)
        ax.set_xlim(lim[0][0], lim[1][0]); ax.set_ylim(lim[0][1], lim[1][1])
        ax.set_aspect('equal'); ax.axis('off')
        return sc
    bax = fig.add_axes([.012, .17, .16, .70])
    sc = embedding(bax, bscore, cm_b, bmax)
    text_items.append(bax.set_title(btitle, fontsize=10.5, pad=6))
    cax = fig.add_axes([.058, .115, .068, .014])
    cb = fig.colorbar(sc, cax=cax, orientation='horizontal')
    cb.set_ticks([0, bmax]); cb.ax.tick_params(labelsize=6.5, length=2, pad=1)
    cb.outline.set_linewidth(.4)
    text_items.extend(cax.get_xticklabels())
    text_items.append(fig.text(.092, .048, metric, fontsize=7, ha='center'))
    rng = np.random.default_rng(seed)
    background = rng.permutation(len(obs)); foreground = rng.permutation(np.flatnonzero(result['primary_path']))
    colors = np.array([PALETTE[s] for s in obs.seg])
    if not chosen:
        for i in range(2):
            ax = fig.add_axes([.208, .565 if i == 0 else .175, .145, .315])
            curve_axes.append(ax)
            ax.set_xlim(0, 6.12); ax.set_ylim(0, .90)
            ax.set_xticks([0, 2, 4, 6]); ax.set_yticks([0, .45, .90])
            ax.tick_params(labelsize=7, length=2.5, width=.6, pad=2)
            text_items.append(ax.set_title('Program unavailable', fontsize=8.0, pad=4))
            text_items.append(ax.text(.5, .5, '\u2014', transform=ax.transAxes,
                                      ha='center', va='center', color='#999999', fontsize=16))
            if i == 0:
                ax.tick_params(labelbottom=False)
            else:
                text_items.append(ax.set_xlabel('Pseudotime', labelpad=2, fontsize=8))
    for i, m in enumerate(chosen):
        ax = fig.add_axes([.208, .565 if i == 0 else .175, .145, .315])
        curve_axes.append(ax)
        values = result['module_scores'][:, m]
        ax.scatter(obs.t.to_numpy()[background], values[background], s=C_POINT_AREA,
                   c=colors[background], alpha=BACKGROUND_ALPHA, linewidths=0)
        ax.scatter(obs.t.to_numpy()[foreground], values[foreground], s=C_POINT_AREA,
                   c=colors[foreground], alpha=FOREGROUND_ALPHA, linewidths=0)
        ax.plot(grid, grid_fit[:, modules[m]['job']], lw=1.05, color='black', zorder=5)
        if i == 0:
            ax.axvline(result['fork_t'], lw=.65, ls=(0, (4, 3)), color='black')
            ax.tick_params(labelbottom=False)
        else:
            text_items.append(ax.set_xlabel('Pseudotime', labelpad=2, fontsize=8))
        ax.set_xlim(0, 6.12); ax.set_ylim(0, .90)
        ax.set_xticks([0, 2, 4, 6]); ax.set_yticks([0, .45, .90])
        ax.tick_params(labelsize=7, length=2.5, width=.6, pad=2)
        text_items.append(ax.set_title(f'{names[m]}  (n={modules[m]["n_genes"]})', fontsize=8.0, pad=4))
    text_items.append(fig.text(.184, .52, 'Scaled gene-set RNA', rotation=90, va='center', ha='center', fontsize=7))
    network = None
    spatial_limits = []
    if len(chosen) == 2:
        # The two spatial maps show precisely the same modules as the two C panels.
        for i, (m, color, x) in enumerate(zip(chosen, (LEFT_COLOR, RIGHT_COLOR), (.368, .853))):
            dcol = int(np.flatnonzero(result['D_jobs'] == modules[m]['job'])[0])
            dscore = result['D_fitted'][:, dcol]
            dmax = math.ceil(np.nanmax(dscore[result['good']]) * 20) / 20
            spatial_limits.append(dmax)
            np.testing.assert_array_equal(dscore[on_path], result['fitted'][on_path, modules[m]['job']])
            dax = fig.add_axes([x, .17, .13, .70])
            cmap = LinearSegmentedColormap.from_list('program' + str(i), ['#D4D4D4', '#FFFFFF', color])
            scd = embedding(dax, dscore, cmap, dmax, mask=result['good'])
            barax = fig.add_axes([x + .034, .115, .062, .014])
            cbar = fig.colorbar(scd, cax=barax, orientation='horizontal')
            cbar.set_ticks([0, dmax]); cbar.ax.tick_params(labelsize=6.2, length=2, pad=1); cbar.outline.set_linewidth(.4)
            text_items.extend(barax.get_xticklabels())
        axn = fig.add_axes([.508, .15, .33, .72])
        network = draw_module_network(axn, factor, modules[chosen[0]]['genes'], modules[chosen[1]]['genes'],
            kind='motif' if factor == 'Nhlh1' else 'associated', left_color='#2D7184',
            right_color='#5AAD7F', font_size=NETWORK_FONT,
            highlight_genes=NETWORK_HIGHLIGHTS.get(factor, ()))
        text_items.extend(network['labeltexts'])
        text_items.append(fig.text(.488, .935, names[chosen[0]], fontsize=10.5, color=LEFT_COLOR, ha='center'))
        text_items.append(fig.text(.816, .935, names[chosen[1]], fontsize=10.5, color=RIGHT_COLOR, ha='center'))
        text_items.append(fig.text(.674, .048, 'Predicted motif-supported targets' if factor == 'Nhlh1' else 'RNA associations', fontsize=7, ha='center', color='#555555'))
    else:
        text_items.append(fig.text(.67, .52, 'No supported early program', ha='center', fontsize=12, color='#777777'))
    for x, letter in ((.007, 'b'), (.185, 'c'), (.359, 'd')):
        text_items.append(fig.text(x, .935, letter, fontsize=14, fontweight='bold'))
    handles = [Line2D([0], [0], marker='o', linestyle='None', color=PALETTE[s],
        label='VGN*' if s == 'VGN' else s, markersize=3) for s in ('CVG', 'VGN', 'T', 'AC', 'AB', 'Ia', 'Ib', 'Ic', 'II')]
    legend = fig.legend(handles=handles, loc='lower center', bbox_to_anchor=(.226, -.005),
        ncol=9, frameon=False, fontsize=6, handletextpad=.2, columnspacing=.65)
    text_items.extend(legend.get_texts())
    for ax in curve_axes:
        text_items.extend(ax.get_xticklabels()); text_items.extend(ax.get_yticklabels())
    fig.canvas.draw()
    renderer = fig.canvas.get_renderer()
    clipped = [t.get_text() for t in text_items if t.get_visible() and t.get_text() and
        (not fig.bbox.contains(*t.get_window_extent(renderer).p0) or not fig.bbox.contains(*t.get_window_extent(renderer).p1))]
    assert not clipped, clipped
    assert network is None or network['QA']['ok'], network['QA']
    output_dir = Path(output_dir); output_dir.mkdir(parents=True, exist_ok=True)
    path = output_dir / f'{factor}.png'
    fig.savefig(path, dpi=DPI)
    plt.close(fig)
    result['figure'] = {'filename': path.name, 'dpi': DPI, 'figsize_in': FIGSIZE,
        'C_D_module_ids': chosen, 'C_D_comparison_applicable': len(chosen) == 2,
        'C_D_gene_sets_equal': True if len(chosen) == 2 else None,
        'C_D_primary_fitted_equal': True if len(chosen) == 2 else None,
        'title_records': {str(m): names[m] for m in chosen}, 'clipped_texts': clipped,
        'network': None if network is None else {k: v for k, v in network.items() if k not in ('labeltexts', 'edgepatches', 'factor_text')},
        'palette': PALETTE, 'C_background_alpha': BACKGROUND_ALPHA, 'C_foreground_alpha': FOREGROUND_ALPHA,
        'VGN_asterisk': 'Candidate VGN tree arm, distinct from independently curated VGN reference',
        'D_colorbar_maxima': spatial_limits,
        'highlight_evidence': NETWORK_HIGHLIGHT_EVIDENCE.get(factor, {}),
        'highlight_semantics': 'Literature-discussed representatives; red is not significance, edge strength, direct binding, or fate determination.',
        'spatial_values': 'Module fitted RNA, not TF RNA or AUCell. Each colorbar maximum is the actual fitted maximum rounded up to0.05; stored fitted values are not clipped. Compare spatial patterns within a module, not saturation across modules.'}
    audit = result['association_audit'].set_index('gene')
    rows = []
    for side, m in enumerate(chosen):
        for gene in modules[m]['genes']:
            rho = float(audit.loc[gene, 'mean_library_rho'])
            rows.append({'factor': factor, 'gene': gene, 'module_id': m, 'side': 'left' if side == 0 else 'right',
                'program_title': names[m], 'evidence': 'motif-supported prediction' if factor == 'Nhlh1' else 'RNA association',
                'mean_library_RNA_rho': rho,
                'RNA_association_sign': 'not encoded for motif targets' if factor == 'Nhlh1' else ('positive' if rho > 0 else 'negative'),
                'arrowhead': False})
    result['network_edges'] = pd.DataFrame(rows)
    assert len(rows) == sum(modules[m]['n_genes'] for m in chosen)
    return path
```

## 8. One cumulative result container

Only Inputs.h5 and Candidates.h5 remain in the result folder. Each factor retains its own screening evidence, all folds/modules, predictions, edges, layout QA, methods and executable-source provenance. Atomic rewriting copies live groups and prevents deleted HDF5 data from accumulating. Earlier factors are preserved.

Appearance-only runs preserve the original analysis provenance and numerical nodes. Current figure geometry, red-label sources, image hash, and outside-network pixel checks are stored in each factor's render_provenance. The legacy figure hash in analysis provenance describes the earlier rendering.

```python

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")
#6. =========== One concise result container, preserving earlier factors
def write_result_value(group, name, value):
    if isinstance(value, pd.DataFrame):
        node = group.create_group(name)
        node.attrs['kind'] = 'dataframe'
        node.attrs['columns'] = json.dumps(value.columns.tolist())
        write_result_value(node, '_index', value.index.astype(str).to_numpy())
        for col in value:
            vals = value[col]
            if vals.dtype.kind in 'OUS':
                vals = vals.fillna('').astype(str).to_numpy()
            else:
                vals = vals.to_numpy()
            write_result_value(node, str(col), vals)
    elif isinstance(value, dict):
        node = group.create_group(name)
        node.attrs['kind'] = 'dict'
        for key, item in value.items():
            write_result_value(node, str(key), item)
    elif isinstance(value, (list, tuple)) and (not value or any(isinstance(x, (dict, list, tuple, pd.DataFrame, np.ndarray)) for x in value)):
        node = group.create_group(name)
        node.attrs['kind'] = 'list'
        for i, item in enumerate(value):
            write_result_value(node, str(i), item)
    elif value is None:
        node = group.create_dataset(name, data='', dtype=h5py.string_dtype())
        node.attrs['kind'] = 'none'
    elif isinstance(value, (str, Path)):
        group.create_dataset(name, data=str(value), dtype=h5py.string_dtype())
    else:
        a = np.asarray(value)
        if a.dtype.kind in 'OUS':
            group.create_dataset(name, data=a.astype(str).astype(object), dtype=h5py.string_dtype(),
                                 compression='gzip' if a.ndim else None)
        else:
            group.create_dataset(name, data=a, compression='gzip' if a.ndim else None,
                                 shuffle=bool(a.ndim))


def save_factor(result, path, factor, provenance):
    path = Path(path)
    # Set this storage path independently before running. Candidates-build.h5.
    temporary = ROOT
    temporary.parent.mkdir(parents=True, exist_ok=True)
    if temporary.exists():
        raise FileExistsError('Remove a failed candidate-build file only after inspecting it')
    # Rewrite live groups atomically so repeated runs do not accumulate HDF5 dead space.
    with h5py.File(temporary, 'w') as h:
        h.attrs['owner'] = 'candidate_regulors_v1'
        factors = h.require_group('factors')
        if path.exists():
            with h5py.File(path, 'r') as previous:
                assert previous.attrs.get('owner') == 'candidate_regulors_v1', 'Refuse to overwrite an unrelated file'
                for other in previous['factors']:
                    if other != factor:
                        previous.copy(previous['factors'][other], factors, name=other)
        write_result_value(factors, factor, result)
        write_result_value(factors[factor], 'provenance', provenance)
        write_result_value(h, 'latest_provenance', provenance)
    temporary.replace(path)




#8. =========== Load archived results without changing numerical data

def read_result_value(node):
    if isinstance(node, h5py.Group):
        kind = node.attrs.get('kind', 'dict')
        if kind == 'dataframe':
            columns = json.loads(node.attrs['columns'])
            table = pd.DataFrame({col: read_result_value(node[col]) for col in columns})
            table.index = read_result_value(node['_index'])
            return table
        if kind == 'list':
            return [read_result_value(node[key]) for key in sorted(node, key=int)]
        return {key: read_result_value(node[key]) for key in node}
    if node.attrs.get('kind') == 'none':
        return None
    return node.asstr()[()] if node.dtype.kind in 'OSU' else node[()]


def load_render_result(path, factor):
    fields = ('factor','obs','X_umap','modules','selected_modules','reviewed_annotations',
        'fit_grid','grid_fitted','B_raw','fitted','module_scores','primary_path','good',
        'D_jobs','D_fitted','fork_t','association_audit','methods','summary')
    with h5py.File(path, 'r') as h:
        group = h['factors/' + factor]
        result = {key: read_result_value(group[key]) for key in fields}
        origin = read_result_value(group['provenance'])
    return result, origin


def outside_network_sha256(path):
    from PIL import Image as PILImage
    with PILImage.open(path) as im:
        pixels = np.asarray(im)
        assert im.size == (11520, 2490)
        # Exact original axes[.508,.15,.33,.72], top-left pixel coordinates.
        x0, y0, x1, y1 = 5852, 323, 9654, 2117
        digest = hashlib.sha256()
        for region in (pixels[:y0], pixels[y1:], pixels[y0:y1, :x0], pixels[y0:y1, x1:]):
            digest.update(region.tobytes())
    return digest.hexdigest()


def save_render_updates(path, updates, work_dir):
    path, work_dir = Path(path), Path(work_dir)
    work_dir.mkdir(parents=True, exist_ok=True)
    # Set this storage path independently before running. Candidates-render.h5.
    temporary = ROOT
    if temporary.exists():
        raise FileExistsError('Inspect a failed render-build file before removing it')
    with h5py.File(path, 'r') as source, h5py.File(temporary, 'w') as target:
        assert source.attrs.get('owner') == 'candidate_regulors_v1'
        for key, value in source.attrs.items():
            target.attrs[key] = value
        for key in source:
            source.copy(source[key], target, name=key)
        for factor, record in updates.items():
            group = target['factors/' + factor]
            for key in ('figure', 'render_provenance'):
                if key in group:
                    del group[key]
                write_result_value(group, key, record[key])
    temporary.replace(path)
```

## 9. Execute candidates in order

Each factor is saved after its own checks. B metrics remain distinct: four positive-associated expression scores, Nhlh1 regulon AUC, and Prox1 native RNA. C/D use identical complete module lists. The displayed branch VGN* is the candidate tree arm.

```python

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")
#11. =========== Reuse saved analysis for local network edits, or rebuild explicitly
import gc
import nbformat
from IPython.display import display, Image

def file_sha256(path):
    value = hashlib.sha256()
    with Path(path).open('rb') as stream:
        for block in iter(lambda:stream.read(4 * 1024 * 1024), b''):
            value.update(block)
    return value.hexdigest()

# Set this storage path independently before running. 05_Candidate_regulors.ipynb. Use the supplied fig4_candidate_regulator_analysis.ipynb source.
NOTEBOOK_FILE = ROOT
source_cells = [c.source for c in nbformat.read(NOTEBOOK_FILE, as_version=4).cells if c.cell_type == 'code']
source_sha = hashlib.sha256('\n\n'.join(source_cells).encode()).hexdigest()
with h5py.File(INPUT_FILE, 'r') as h:
    count_file = PROJECT_ROOT / h.attrs['common_counts_file']
    upstream = json.loads(h['provenance_json'].asstr()[()])
input_sha, counts_sha = file_sha256(INPUT_FILE), file_sha256(count_file)
assert counts_sha == upstream['source_sha256']['counts']
versions = {p: importlib.metadata.version(p) for p in
    ('numpy','pandas','scipy','networkx','h5py','matplotlib','nbformat')}
expected_pools = {'Foxo6':65,'Zfpm2':70,'Neurod2':45,'Nhlh1':46,'Neurod1':200,'Prox1':0}
expected_modules = {'Foxo6':[23,19,17],'Zfpm2':[26,18,15,11],'Neurod2':[23,19],
    'Nhlh1':[19,19,8],'Neurod1':[36,35,33,30,24,22,20],'Prox1':[]}
run_summary, runtime_cleanup, render_updates = [], [], {}
for FACTOR in FACTORS:
    cached = False
    figure_path = FIGURE_DIR / f'{FACTOR}.png'
    if not RECOMPUTE_ANALYSIS and RESULT_FILE.exists():
        with h5py.File(RESULT_FILE, 'r') as h:
            key = f'factors/{FACTOR}/provenance'
            if key in h:
                origin = h[key]
                cached = (origin['input_sha256'].asstr()[()] == input_sha
                    and origin['formal_counts_sha256'].asstr()[()] == counts_sha)
    if cached and FACTOR == 'Prox1' and figure_path.exists():
        print('Prox1: RNA reference retained; no network exists to redraw')
        run_summary.append({'factor':FACTOR,'operation':'RNA reference retained','display_genes':0,'red_genes':0})
        display(Image(filename=str(figure_path),width=1600))
        continue
    previous_png = file_sha256(figure_path) if figure_path.exists() else ''
    previous_outside = outside_network_sha256(figure_path) if cached and figure_path.exists() else ''
    if cached:
        result, origin = load_render_result(RESULT_FILE, FACTOR)
        print(FACTOR + ': archived scores and fits reused; network-only redraw',flush=True)
    else:
        print('Analyzing ' + FACTOR,flush=True)
        result = analyze_factor(INPUT_FILE,FACTOR,seed=SEED)
        assert result['summary']['analysis_pool_genes'] == expected_pools[FACTOR]
        assert result['summary']['module_sizes'] == expected_modules[FACTOR]
        if FACTOR == 'Prox1':
            result['late_validation'] = validate_late_prox1(INPUT_FILE)
            assert not result['late_validation']['full']['selected'].any()
            assert not result['late_validation']['LOO']['train_selected'].any()
        result = fit_factor(result,WORK_DIR,RSCRIPT)
    figure_path = render_factor(result,FACTOR,FIGURE_DIR,seed=SEED)
    gene_count = sum(result['modules'][m]['n_genes'] for m in result['selected_modules'])
    assert len(result['network_edges']) == gene_count
    if len(result['selected_modules']):
        assert result['figure']['C_D_gene_sets_equal'] and result['figure']['C_D_primary_fitted_equal']
        assert result['figure']['network']['QA']['ok']
        assert not result['figure']['network']['QA']['requested_highlights_not_in_modules']
        assert set(result['figure']['network']['QA']['highlighted_genes']) == set(NETWORK_HIGHLIGHTS.get(FACTOR, ()))
    after_png = file_sha256(figure_path)
    after_outside = outside_network_sha256(figure_path) if cached else ''
    if previous_outside:
        assert after_outside == previous_outside, 'A B/C/D-side pixel changed outside the network axes'
    rendering = {'operation':'network-only rendering' if cached else 'full analysis and rendering',
        'reused_saved_scores_and_fits':cached,'input_sha256':input_sha,
        'formal_counts_sha256':counts_sha,'previous_png_sha256':previous_png,
        'figure_sha256':after_png,'outside_network_roi':(5852,323,9654,2117),
        'outside_network_sha256_before':previous_outside,'outside_network_sha256_after':after_outside,
        'outside_network_unchanged':bool(previous_outside and previous_outside == after_outside),
        'executed_code_cells':source_cells,'code_sha256':source_sha,'python':sys.version,
        'versions':versions,'highlight_evidence':NETWORK_HIGHLIGHT_EVIDENCE.get(FACTOR,{}),
        'highlight_meaning':'Existing module members selected for discussion; not significance, stronger edges, binding or fate proof',
        'layer_meaning':'Visual depth only, not inferred regulatory hierarchy'}
    if cached:
        rendering['analysis_source_sha256'] = origin['code_sha256']
        render_updates[FACTOR] = {'figure':result['figure'],'render_provenance':rendering}
    else:
        interpretation = ('Frozen Nhlh1 regulon AUC and motif-supported predicted targets' if FACTOR == 'Nhlh1'
            else 'Measured Prox1 RNA with no supported full-data early program' if FACTOR == 'Prox1'
            else 'Signed RNA associations; B uses the positive pool; C/D use complete selected modules')
        provenance = {'factor':FACTOR,'input_file':str(INPUT_FILE.relative_to(PROJECT_ROOT)),
            'input_sha256':input_sha,'formal_counts_file':str(count_file.relative_to(PROJECT_ROOT)),
            'formal_counts_sha256':counts_sha,'input_archive_provenance':upstream,
            'executed_code_cells':source_cells,'code_sha256':source_sha,'python':sys.version,
            'platform':platform.platform(),'versions':versions,'figure_sha256':after_png,
            'interpretation':interpretation}
        save_factor(result,RESULT_FILE,FACTOR,provenance)
        render_updates[FACTOR] = {'figure':result['figure'],'render_provenance':rendering}
    runtime_cleanup.extend(result.get('temporary_cleanup',[]))
    red_count = len(result['figure']['network']['QA']['highlighted_genes']) if gene_count else 0
    run_summary.append({'factor':FACTOR,'operation':'network redraw' if cached else 'full analysis',
        'display_genes':gene_count,'red_genes':red_count})
    print(FACTOR + ': ' + str(gene_count) + ' genes; ' + str(red_count) + ' red representatives',flush=True)
    display(Image(filename=str(figure_path),width=1600))
    del result
    gc.collect()
if render_updates:
    save_render_updates(RESULT_FILE,render_updates,WORK_DIR)
files = sorted(p.name for p in RESULT_FILE.parent.iterdir() if p.is_file())
assert files == ['Candidates.h5','Inputs.h5'],files
print('Result files:',files)
display(pd.DataFrame(run_summary))
```

```python

# Configure each file/directory placeholder independently; no storage addresses are shipped.
from pathlib import Path
ROOT = Path(".")

# Cleanup must never interpret the project/current directory as a temporary target.
def _release_cleanup_target(path):
    candidate = Path(path).resolve()
    current = Path.cwd().resolve()
    if candidate == current or candidate in current.parents or candidate == Path(candidate.anchor):
        raise RuntimeError("Configure the owned temporary cleanup path before running this section.")
    return path

#12. =========== Clean only consumed runtime files
removed = list(runtime_cleanup)
if WORK_DIR.exists():
    for path in sorted(_release_cleanup_target(WORK_DIR).rglob("*"), key=lambda p: len(p.parts), reverse=True):
        assert path.resolve().is_relative_to(WORK_DIR.resolve())
        if path.is_file():
            removed.append({"name":str(path.relative_to(WORK_DIR)),"bytes":path.stat().st_size})
            _release_cleanup_target(path).unlink()
        elif path.is_dir() and not any(_release_cleanup_target(path).iterdir()):
            _release_cleanup_target(path).rmdir()
    if not any(_release_cleanup_target(WORK_DIR).iterdir()):
        _release_cleanup_target(WORK_DIR).rmdir()
print("Consumed runtime bytes removed:", sum(x["bytes"] for x in removed))
```

## Original standalone source: fig4_validate_figure_output.py

This complete existing program retains its original command-line entry point and runtime assumptions. Use it in its original independent script context. Executing the cell in Jupyter would invoke the unchanged entry point with Jupyter arguments; this section is not a new notebook dispatcher.


```python
#!/usr/bin/env python3
"""Fail-closed validation for 600-DPI PNGs and editable figure PDFs.

The default PDF profile is intentionally strict. It is designed for figures that
must remain straightforward to edit after opening in Adobe Illustrator:

* no more than four explicit, nonzero-winding clipping paths per page;
* clipping paths are sibling, axis-aligned rectangles (never nested);
* every clipping path is explicitly scoped by q/Q and ends with W n;
* no raster images, soft/image masks, inline images, text clipping, shading,
  or uninspected recursive Form XObjects;
* live, embedded, Unicode-mapped Arial/SimSun text; and
* at least one painted vector path.

The program writes its report to stdout only. Poppler render files are created in
a task-local temporary directory and removed before exit unless the caller uses
``--render-dir``; that option always creates and retains a new unique child
directory, so existing files are never overwritten.
"""

from __future__ import annotations

import argparse
import math
import os
import re
import secrets
import shutil
import subprocess
import sys
import tempfile
from collections import Counter
from dataclasses import dataclass, field, replace
from pathlib import Path
from typing import Any, Iterable, Sequence

try:
    from PIL import Image
    from pypdf import PdfReader
    from pypdf.generic import ArrayObject, ContentStream, DictionaryObject, IndirectObject
except ImportError as exc:  # pragma: no cover - exercised only in a broken runtime
    print(f"[NOT CERTIFIED] Required Python dependency is unavailable: {exc}")
    raise SystemExit(1)


EPSILON = 1e-7
PATH_BUILDING = {b"m", b"l", b"c", b"v", b"y", b"h", b"re"}
PATH_ENDING = {b"S", b"s", b"f", b"F", b"f*", b"B", b"B*", b"b", b"b*", b"n"}
TEXT_SHOWING = {b"Tj", b"TJ", b"'", b'"'}
FILLING = {b"f", b"F", b"f*"}
STROKING = {b"S", b"s"}
FILL_STROKING = {b"B", b"B*", b"b", b"b*"}
INLINE_IMAGE_OPS = {b"BI", b"ID", b"EI", b"INLINE IMAGE"}
SUBSET_PREFIX = re.compile(r"^[A-Z]{6}\+")
CJK_RE = re.compile(
    "[\u3400-\u4dbf\u4e00-\u9fff\uf900-\ufaff\U00020000-\U0002fa1f]"
)
ASCII_ALNUM_RE = re.compile(r"[A-Za-z0-9]")


@dataclass
class Issue:
    code: str
    message: str
    outcome: str = "FAIL"


@dataclass
class ClipEvent:
    number: int
    scope: str
    depth: int
    bbox: tuple[float, float, float, float] | None
    kinds: Counter[str] = field(default_factory=Counter)


@dataclass
class FontUse:
    scope: str
    resource_name: str
    base_font: str
    subtype: str
    embedded: bool
    to_unicode: bool


@dataclass
class GraphicsState:
    ctm: tuple[float, float, float, float, float, float] = (1, 0, 0, 1, 0, 0)
    clip_depth: int = 0
    active_clips: tuple[int, ...] = ()
    font_resource: str | None = None
    font_object: Any = None
    text_render_mode: int = 0


@dataclass
class PdfPageStats:
    page_number: int
    crop_size_points: tuple[float, float] = (0.0, 0.0)
    media_size_points: tuple[float, float] = (0.0, 0.0)
    rotation: int = 0
    clip_events: list[ClipEvent] = field(default_factory=list)
    max_clip_depth: int = 0
    q_count: int = 0
    Q_count: int = 0
    form_calls: int = 0
    image_count: int = 0
    inline_image_count: int = 0
    text_show_count: int = 0
    text_object_count: int = 0
    vector_paint_count: int = 0
    vector_ops: Counter[str] = field(default_factory=Counter)
    font_uses: dict[tuple[str, str, str], FontUse] = field(default_factory=dict)


@dataclass
class ValidationResult:
    path: Path
    kind: str
    issues: list[Issue] = field(default_factory=list)
    details: list[str] = field(default_factory=list)

    @property
    def passed(self) -> bool:
        return not self.issues

    @property
    def status(self) -> str:
        if any(issue.outcome == "FAIL" for issue in self.issues):
            return "FAIL"
        if self.issues:
            return "NOT CERTIFIED"
        return "PASS"

    def fail(self, code: str, message: str) -> None:
        self.issues.append(Issue(code, message, "FAIL"))

    def not_certified(self, code: str, message: str) -> None:
        self.issues.append(Issue(code, message, "NOT CERTIFIED"))


def dereference(value: Any) -> Any:
    try:
        return value.get_object()
    except AttributeError:
        return value


def dictionary(value: Any) -> DictionaryObject | dict:
    value = dereference(value)
    return value if isinstance(value, (DictionaryObject, dict)) else {}


def numeric(value: Any) -> float:
    return float(dereference(value))


def normalize_font_name(value: Any) -> str:
    name = str(value or "").lstrip("/")
    return SUBSET_PREFIX.sub("", name)


def font_family(base_font: str) -> str | None:
    compact = re.sub(r"[^a-z0-9]", "", normalize_font_name(base_font).lower())
    if compact.startswith("arial"):
        return "Arial"
    if compact.startswith("simsun") or compact.startswith("nsimsun"):
        return "SimSun"
    return None


def font_descriptor(font_obj: Any) -> dict:
    font = dictionary(font_obj)
    subtype = str(font.get("/Subtype", ""))
    if subtype == "/Type0":
        descendants = dereference(font.get("/DescendantFonts", []))
        if descendants:
            descendant = dictionary(descendants[0])
            return dictionary(descendant.get("/FontDescriptor", {}))
    return dictionary(font.get("/FontDescriptor", {}))


def font_is_embedded(font_obj: Any) -> bool:
    descriptor = font_descriptor(font_obj)
    return any(key in descriptor for key in ("/FontFile", "/FontFile2", "/FontFile3"))


def object_identity(value: Any) -> tuple[Any, ...]:
    if isinstance(value, IndirectObject):
        return ("indirect", value.idnum, value.generation)
    obj = dereference(value)
    ref = getattr(obj, "indirect_reference", None)
    if isinstance(ref, IndirectObject):
        return ("indirect", ref.idnum, ref.generation)
    return ("direct", id(obj))


def matrix_multiply(
    first: tuple[float, float, float, float, float, float],
    second: tuple[float, float, float, float, float, float],
) -> tuple[float, float, float, float, float, float]:
    """Compose affine transforms so result(point) == first(second(point))."""
    a1, b1, c1, d1, e1, f1 = first
    a2, b2, c2, d2, e2, f2 = second
    return (
        a1 * a2 + c1 * b2,
        b1 * a2 + d1 * b2,
        a1 * c2 + c1 * d2,
        b1 * c2 + d1 * d2,
        a1 * e2 + c1 * f2 + e1,
        b1 * e2 + d1 * f2 + f1,
    )


def transform_point(
    matrix: tuple[float, float, float, float, float, float], x: float, y: float
) -> tuple[float, float]:
    a, b, c, d, e, f = matrix
    return (a * x + c * y + e, b * x + d * y + f)


def normalized_bbox(points: Sequence[tuple[float, float]]) -> tuple[float, float, float, float]:
    xs = [point[0] for point in points]
    ys = [point[1] for point in points]
    left, right = min(xs), max(xs)
    bottom, top = min(ys), max(ys)
    return (left, bottom, right - left, top - bottom)


def nearly_equal(left: float, right: float, tolerance: float) -> bool:
    return abs(left - right) <= tolerance


def boxes_match(
    observed: tuple[float, float, float, float],
    expected: tuple[float, float, float, float],
) -> bool:
    tolerance = max(0.5, max(abs(value) for value in expected[2:]) * 1e-4)
    return all(nearly_equal(a, b, tolerance) for a, b in zip(observed, expected))


def image_blank_reason(image: Image.Image) -> str | None:
    """Return a concrete blank-page reason without allocating a full RGBA copy."""
    extrema: Any = image.getextrema()
    if not isinstance(extrema, tuple):
        extrema = (extrema,)
    elif extrema and not isinstance(extrema[0], tuple):
        extrema = (extrema,)
    bands = image.getbands()
    if "A" in bands and extrema[bands.index("A")] == (0, 0):
        return "all pixels are fully transparent"
    if all(low == high for low, high in extrema):
        return "all pixels have one uniform color"
    return None


def parse_simple_rectangle(
    path_ops: Sequence[tuple[bytes, list[Any]]],
    ctm: tuple[float, float, float, float, float, float],
) -> tuple[tuple[float, float, float, float] | None, str | None]:
    if len(path_ops) == 1 and path_ops[0][0] == b"re":
        operands = path_ops[0][1]
        if len(operands) != 4:
            return None, "rectangle operator does not have four operands"
        x, y, width, height = (numeric(value) for value in operands)
        if abs(width) <= EPSILON or abs(height) <= EPSILON:
            return None, "rectangle has zero width or height"
        raw_points = [
            (x, y),
            (x + width, y),
            (x + width, y + height),
            (x, y + height),
        ]
    else:
        operators = [operator for operator, _ in path_ops]
        if not operators or operators[0] != b"m" or any(
            operator not in {b"m", b"l", b"h"} for operator in operators
        ):
            return None, "clip path is not one re rectangle or one m/l/h rectangle"
        if operators.count(b"m") != 1 or operators.count(b"h") > 1:
            return None, "clip path contains multiple subpaths"
        raw_points: list[tuple[float, float]] = []
        for operator, operands in path_ops:
            if operator in {b"m", b"l"}:
                if len(operands) != 2:
                    return None, "rectangle vertex does not have two coordinates"
                raw_points.append((numeric(operands[0]), numeric(operands[1])))
        if len(raw_points) == 5 and all(
            nearly_equal(a, b, 1e-6) for a, b in zip(raw_points[0], raw_points[-1])
        ):
            raw_points.pop()
        if len(raw_points) != 4 or (operators[-1] != b"h" and len(path_ops) != 5):
            return None, "m/l/h clip does not contain exactly four rectangle vertices"

    points = [transform_point(ctm, x, y) for x, y in raw_points]
    tolerance = max(1e-5, max(abs(value) for point in points for value in point) * 1e-7)
    for start, end in zip(points, points[1:] + points[:1]):
        dx, dy = abs(end[0] - start[0]), abs(end[1] - start[1])
        if not (dx <= tolerance < dy or dy <= tolerance < dx):
            return None, "clip rectangle is rotated, skewed, or self-degenerate"
    unique_x = []
    unique_y = []
    for x, y in points:
        if not any(nearly_equal(x, item, tolerance) for item in unique_x):
            unique_x.append(x)
        if not any(nearly_equal(y, item, tolerance) for item in unique_y):
            unique_y.append(y)
    if len(unique_x) != 2 or len(unique_y) != 2:
        return None, "clip path does not form one axis-aligned rectangle"
    return normalized_bbox(points), None


class PdfAuditor:
    def __init__(
        self,
        path: Path,
        result: ValidationResult,
        max_clips: int,
        expected_clips: int | None,
        declared_clip_boxes: list[tuple[float, float, float, float]],
        allow_raster_images: bool,
        allow_form_xobjects: bool,
        allow_no_text: bool,
        allow_no_vector: bool,
    ) -> None:
        self.path = path
        self.result = result
        self.max_clips = max_clips
        self.expected_clips = expected_clips
        self.declared_clip_boxes = declared_clip_boxes
        self.allow_raster_images = allow_raster_images
        self.allow_form_xobjects = allow_form_xobjects
        self.allow_no_text = allow_no_text
        self.allow_no_vector = allow_no_vector
        self.reader: PdfReader | None = None
        self._scanned_resources: set[tuple[Any, ...]] = set()
        self._reported_fonts: set[tuple[Any, ...]] = set()

    def audit(self) -> list[PdfPageStats]:
        try:
            self.reader = PdfReader(str(self.path), strict=True)
        except Exception as exc:
            self.result.not_certified("PDF_PARSE", f"PDF cannot be parsed strictly: {exc}")
            return []
        if self.reader.is_encrypted:
            self.result.not_certified("PDF_ENCRYPTED", "encrypted PDFs cannot be inspected")
            return []
        if not self.reader.pages:
            self.result.fail("PDF_EMPTY", "PDF contains no pages")
            return []

        page_stats: list[PdfPageStats] = []
        for page_number, page in enumerate(self.reader.pages, start=1):
            crop_box = page.cropbox
            media_box = page.mediabox
            user_unit = float(page.get("/UserUnit", 1) or 1)
            rotation = int(page.get("/Rotate", 0) or 0) % 360
            stats = PdfPageStats(
                page_number,
                crop_size_points=(
                    float(crop_box.width) * user_unit,
                    float(crop_box.height) * user_unit,
                ),
                media_size_points=(
                    float(media_box.width) * user_unit,
                    float(media_box.height) * user_unit,
                ),
                rotation=rotation,
            )
            page_stats.append(stats)
            resources = dictionary(page.get("/Resources", {}))
            page_boxes = [
                (
                    float(crop_box.left),
                    float(crop_box.bottom),
                    float(crop_box.width),
                    float(crop_box.height),
                ),
                (
                    float(media_box.left),
                    float(media_box.bottom),
                    float(media_box.width),
                    float(media_box.height),
                ),
            ]
            allowed_boxes = page_boxes + self.declared_clip_boxes
            contents = page.get_contents()
            if contents is None:
                self.result.fail("PDF_NO_CONTENT", f"page {page_number} has no content stream")
                continue
            try:
                content_stream = ContentStream(contents, self.reader)
            except Exception as exc:
                self.result.not_certified(
                    "CONTENT_PARSE", f"page {page_number} content stream cannot be parsed: {exc}"
                )
                continue
            self._scan_resources(resources, f"page {page_number}", stats)
            self._analyze_operations(
                content_stream.operations,
                resources,
                f"page {page_number}",
                stats,
                GraphicsState(),
                allowed_boxes,
                form_stack=set(),
                implicit_scope=False,
            )
            self._validate_page_text(page, stats)
            self._validate_page_summary(stats)
        return page_stats

    def _scan_resources(self, resources: dict, scope: str, stats: PdfPageStats) -> None:
        identity = object_identity(resources)
        if identity in self._scanned_resources:
            return
        self._scanned_resources.add(identity)

        ext_states = dictionary(resources.get("/ExtGState", {}))
        for name, reference in ext_states.items():
            ext_state = dictionary(reference)
            if "/SMask" in ext_state and str(dereference(ext_state["/SMask"])) != "/None":
                self.result.fail(
                    "SOFT_MASK", f"{scope} ExtGState {name} contains a non-None /SMask"
                )

        patterns = dictionary(resources.get("/Pattern", {}))
        if patterns:
            self.result.fail(
                "PATTERN_RESOURCE",
                f"{scope} contains Pattern resources; their Illustrator editability is not certified",
            )
        shadings = dictionary(resources.get("/Shading", {}))
        if shadings:
            self.result.fail(
                "SHADING_RESOURCE",
                f"{scope} contains Shading resources; their Illustrator editability is not certified",
            )

        xobjects = dictionary(resources.get("/XObject", {}))
        for name, reference in xobjects.items():
            xobject = dictionary(reference)
            subtype = str(xobject.get("/Subtype", ""))
            if "/SMask" in xobject:
                self.result.fail("SOFT_MASK", f"{scope} XObject {name} contains /SMask")
            if "/Mask" in xobject:
                self.result.fail("IMAGE_MASK", f"{scope} XObject {name} contains /Mask")
            if bool(xobject.get("/ImageMask", False)):
                self.result.fail("IMAGE_MASK", f"{scope} XObject {name} is an image mask")
            if subtype == "/Image":
                stats.image_count += 1
                if not self.allow_raster_images:
                    self.result.fail("RASTER_IMAGE", f"{scope} contains image XObject {name}")
            elif subtype == "/Form" and not self.allow_form_xobjects:
                self.result.fail(
                    "FORM_XOBJECT",
                    f"{scope} contains Form XObject {name}; use --allow-form-xobjects only for an intentionally reviewed Form workflow",
                )
            elif subtype not in {"/Form", ""}:
                self.result.fail(
                    "UNSUPPORTED_XOBJECT", f"{scope} contains unsupported XObject {name} ({subtype})"
                )

    def _record_clip_content(self, stats: PdfPageStats, state: GraphicsState, kind: str) -> None:
        for event_number in state.active_clips:
            stats.clip_events[event_number - 1].kinds[kind] += 1

    def _register_font(
        self, scope: str, resource_name: str, font_reference: Any, stats: PdfPageStats
    ) -> None:
        font = dictionary(font_reference)
        base_font = normalize_font_name(font.get("/BaseFont", ""))
        subtype = str(font.get("/Subtype", ""))
        embedded = font_is_embedded(font)
        to_unicode = "/ToUnicode" in font
        key = (scope, resource_name, base_font)
        stats.font_uses[key] = FontUse(
            scope, resource_name, base_font, subtype, embedded, to_unicode
        )
        identity = object_identity(font_reference)
        if identity in self._reported_fonts:
            return
        self._reported_fonts.add(identity)
        family = font_family(base_font)
        if family is None:
            self.result.fail(
                "FONT_FAMILY",
                f"{scope} font {resource_name} uses {base_font or '<unnamed>'}; only Arial and SimSun are allowed",
            )
        if not embedded:
            self.result.fail(
                "FONT_NOT_EMBEDDED",
                f"{scope} font {resource_name} ({base_font or '<unnamed>'}) is not embedded",
            )
        if not to_unicode:
            self.result.fail(
                "FONT_NO_TOUNICODE",
                f"{scope} font {resource_name} ({base_font or '<unnamed>'}) has no /ToUnicode map",
            )

    def _analyze_operations(
        self,
        operations: Sequence[tuple[list[Any], bytes]],
        resources: dict,
        scope: str,
        stats: PdfPageStats,
        initial_state: GraphicsState,
        allowed_boxes: list[tuple[float, float, float, float]],
        form_stack: set[tuple[Any, ...]],
        implicit_scope: bool,
    ) -> None:
        state = replace(initial_state)
        base_clip_depth = initial_state.clip_depth
        stack: list[GraphicsState] = []
        path_ops: list[tuple[bytes, list[Any]]] = []
        pending_clip = False
        in_text = False
        text_objects_at_start = stats.text_object_count

        for index, (operands, operator) in enumerate(operations):
            operands = list(operands)
            if operator == b"q":
                stats.q_count += 1
                stack.append(replace(state))
                continue
            if operator == b"Q":
                stats.Q_count += 1
                if not stack:
                    self.result.fail("Q_UNDERFLOW", f"{scope} has Q without a matching q at operation {index}")
                else:
                    state = stack.pop()
                path_ops = []
                pending_clip = False
                continue
            if operator == b"cm":
                if len(operands) != 6:
                    self.result.fail("BAD_CTM", f"{scope} has malformed cm at operation {index}")
                else:
                    matrix = tuple(numeric(value) for value in operands)
                    state.ctm = matrix_multiply(state.ctm, matrix)  # type: ignore[arg-type]
                continue

            if operator in PATH_BUILDING:
                path_ops.append((operator, operands))
                stats.vector_ops[operator.decode("ascii")] += 1
                continue
            if operator in {b"W", b"W*"}:
                if pending_clip:
                    self.result.fail(
                        "CLIP_PENDING", f"{scope} has consecutive clipping operators at operation {index}"
                    )
                if operator == b"W*":
                    self.result.fail(
                        "CLIP_EVEN_ODD",
                        f"{scope} uses W* at operation {index}; the strict profile requires nonzero-winding W",
                    )
                pending_clip = True
                continue
            if operator in PATH_ENDING:
                if pending_clip:
                    if operator != b"n":
                        self.result.fail(
                            "CLIP_PAINTED_PATH",
                            f"{scope} applies clipping with W/W* followed by {operator.decode('latin1')}; strict clipping must end with n",
                        )
                    bbox, reason = parse_simple_rectangle(path_ops, state.ctm)
                    event_number = len(stats.clip_events) + 1
                    if reason:
                        self.result.fail(
                            "CLIP_NOT_RECTANGLE",
                            f"{scope} clip {event_number} is invalid: {reason}",
                        )
                    elif bbox is not None and not any(
                        boxes_match(bbox, candidate) for candidate in allowed_boxes
                    ):
                        pretty = ", ".join(f"{value:.3f}" for value in bbox)
                        self.result.fail(
                            "CLIP_BOX_UNDECLARED",
                            f"{scope} clip {event_number} box ({pretty}) is neither the page box nor a declared --clip-box",
                        )
                    if not stack and not implicit_scope:
                        self.result.fail(
                            "CLIP_NOT_SCOPED",
                            f"{scope} clip {event_number} is not enclosed by an explicit q/Q scope",
                        )
                    state.clip_depth += 1
                    state.active_clips = state.active_clips + (event_number,)
                    stats.max_clip_depth = max(stats.max_clip_depth, state.clip_depth)
                    stats.clip_events.append(
                        ClipEvent(event_number, scope, state.clip_depth, bbox)
                    )
                    if state.clip_depth > 1:
                        self.result.fail(
                            "CLIP_NESTED",
                            f"{scope} clip {event_number} reaches effective clipping depth {state.clip_depth}",
                        )
                    pending_clip = False
                if operator in FILLING:
                    stats.vector_paint_count += 1
                    self._record_clip_content(stats, state, "vector-fill")
                elif operator in STROKING:
                    stats.vector_paint_count += 1
                    self._record_clip_content(stats, state, "vector-stroke")
                elif operator in FILL_STROKING:
                    stats.vector_paint_count += 1
                    self._record_clip_content(stats, state, "vector-fill+stroke")
                path_ops = []
                continue

            if operator == b"BT":
                if in_text:
                    self.result.fail("TEXT_NESTED", f"{scope} contains nested BT operators")
                in_text = True
                stats.text_object_count += 1
                continue
            if operator == b"ET":
                if not in_text:
                    self.result.fail("TEXT_ET_UNMATCHED", f"{scope} contains ET without BT")
                in_text = False
                continue
            if operator == b"Tr":
                try:
                    state.text_render_mode = int(numeric(operands[0]))
                except Exception:
                    self.result.fail("TEXT_RENDER_MODE", f"{scope} contains malformed Tr")
                    continue
                if state.text_render_mode in {4, 5, 6, 7}:
                    self.result.fail(
                        "TEXT_CLIPPING",
                        f"{scope} uses clipping text rendering mode {state.text_render_mode}",
                    )
                continue
            if operator == b"Tf":
                if len(operands) < 2:
                    self.result.fail("FONT_SELECT", f"{scope} contains malformed Tf")
                    continue
                resource_name = str(operands[0])
                fonts = dictionary(resources.get("/Font", {}))
                font_reference = fonts.get(resource_name)
                if font_reference is None:
                    self.result.fail(
                        "FONT_MISSING", f"{scope} selects missing font resource {resource_name}"
                    )
                    state.font_resource = None
                    state.font_object = None
                else:
                    state.font_resource = resource_name
                    state.font_object = font_reference
                continue
            if operator in TEXT_SHOWING:
                stats.text_show_count += 1
                if not in_text:
                    self.result.fail("TEXT_OUTSIDE_BT", f"{scope} shows text outside BT/ET")
                if state.text_render_mode in {3, 7}:
                    self.result.fail(
                        "INVISIBLE_TEXT", f"{scope} contains invisible text rendering mode {state.text_render_mode}"
                    )
                if state.text_render_mode in {4, 5, 6, 7}:
                    self.result.fail(
                        "TEXT_CLIPPING",
                        f"{scope} shows text in clipping rendering mode {state.text_render_mode}",
                    )
                if not state.font_resource or state.font_object is None:
                    self.result.fail("TEXT_NO_FONT", f"{scope} shows text without a resolved font")
                else:
                    self._register_font(scope, state.font_resource, state.font_object, stats)
                self._record_clip_content(stats, state, "text")
                continue

            if operator == b"Do":
                if not operands:
                    self.result.fail("XOBJECT_CALL", f"{scope} has malformed Do operator")
                    continue
                resource_name = str(operands[0])
                xobjects = dictionary(resources.get("/XObject", {}))
                reference = xobjects.get(resource_name)
                if reference is None:
                    self.result.fail(
                        "XOBJECT_MISSING", f"{scope} invokes missing XObject {resource_name}"
                    )
                    continue
                xobject = dictionary(reference)
                subtype = str(xobject.get("/Subtype", ""))
                if subtype == "/Image":
                    self._record_clip_content(stats, state, "image")
                    continue
                if subtype != "/Form":
                    self.result.fail(
                        "UNSUPPORTED_XOBJECT",
                        f"{scope} invokes unsupported XObject {resource_name} ({subtype})",
                    )
                    continue
                stats.form_calls += 1
                self._record_clip_content(stats, state, "form")
                identity = object_identity(reference)
                if identity in form_stack:
                    self.result.fail(
                        "FORM_RECURSION", f"{scope} has recursive Form XObject {resource_name}"
                    )
                    continue
                form_resources = dictionary(xobject.get("/Resources", resources))
                self._scan_resources(form_resources, f"{scope}/Form {resource_name}", stats)
                form_matrix_raw = dereference(xobject.get("/Matrix", [1, 0, 0, 1, 0, 0]))
                try:
                    form_matrix = tuple(numeric(value) for value in form_matrix_raw)
                    if len(form_matrix) != 6:
                        raise ValueError
                except Exception:
                    self.result.fail(
                        "FORM_MATRIX", f"{scope}/Form {resource_name} has malformed /Matrix"
                    )
                    form_matrix = (1, 0, 0, 1, 0, 0)
                form_state = replace(state)
                form_state.ctm = matrix_multiply(state.ctm, form_matrix)  # type: ignore[arg-type]
                try:
                    form_stream = ContentStream(dereference(reference), self.reader)
                except Exception as exc:
                    self.result.not_certified(
                        "FORM_PARSE", f"{scope}/Form {resource_name} cannot be parsed: {exc}"
                    )
                    continue
                self._analyze_operations(
                    form_stream.operations,
                    form_resources,
                    f"{scope}/Form {resource_name}",
                    stats,
                    form_state,
                    allowed_boxes,
                    form_stack | {identity},
                    implicit_scope=True,
                )
                continue

            if operator in INLINE_IMAGE_OPS:
                stats.inline_image_count += 1
                self.result.fail("INLINE_IMAGE", f"{scope} contains an inline image")
                self._record_clip_content(stats, state, "inline-image")
                continue
            if operator == b"sh":
                self.result.fail("SHADING", f"{scope} paints a shading object")
                self._record_clip_content(stats, state, "shading")
                continue

        if pending_clip:
            self.result.fail("CLIP_UNFINISHED", f"{scope} ends before its W/W* clip is applied")
        if path_ops:
            self.result.fail("PATH_UNFINISHED", f"{scope} ends with an unfinished path")
        if in_text:
            self.result.fail("TEXT_UNFINISHED", f"{scope} ends before ET")
        if stack:
            self.result.fail(
                "Q_UNBALANCED", f"{scope} has {len(stack)} q operator(s) without matching Q"
            )
        if state.clip_depth != base_clip_depth and not implicit_scope:
            self.result.fail(
                "CLIP_NOT_RESTORED",
                f"{scope} ends at clipping depth {state.clip_depth}, expected {base_clip_depth}",
            )
        if stats.text_object_count < text_objects_at_start:
            self.result.not_certified("INTERNAL_TEXT_COUNT", f"{scope} text-object accounting failed")

    def _validate_page_text(self, page: Any, stats: PdfPageStats) -> None:
        fragments: list[tuple[str, float, float, float, str]] = []
        parsed_font_names = {
            normalize_font_name(font.base_font) for font in stats.font_uses.values()
        }

        def visitor(text: str, cm: Sequence[float], tm: Sequence[float], font: Any, size: float) -> None:
            cleaned = text.strip()
            if not cleaned:
                return
            base_font = normalize_font_name(dictionary(font).get("/BaseFont", ""))
            try:
                x = float(cm[0]) * float(tm[4]) + float(cm[2]) * float(tm[5]) + float(cm[4])
                y = float(cm[1]) * float(tm[4]) + float(cm[3]) * float(tm[5]) + float(cm[5])
            except Exception:
                x = y = 0.0
            fragments.append((cleaned, x, y, float(size or 0), base_font))
            family = font_family(base_font)
            # pypdf can report a parent-page default font for text inside a Form
            # XObject. The content-stream pass above resolves the Form's actual
            # font. Apply script-specific checks only when the visitor font is a
            # font that the operation parser observed on a text-show operation.
            visitor_font_is_resolved = base_font in parsed_font_names
            if visitor_font_is_resolved and ASCII_ALNUM_RE.search(cleaned) and family != "Arial":
                self.result.fail(
                    "ASCII_FONT",
                    f"page {stats.page_number} text {cleaned!r} uses {base_font or '<unknown>'}, not Arial",
                )
            if visitor_font_is_resolved and CJK_RE.search(cleaned) and family != "SimSun":
                self.result.fail(
                    "CJK_FONT",
                    f"page {stats.page_number} Chinese text {cleaned!r} uses {base_font or '<unknown>'}, not SimSun",
                )

        try:
            page.extract_text(visitor_text=visitor)
        except Exception as exc:
            self.result.not_certified(
                "TEXT_EXTRACT", f"page {stats.page_number} live text cannot be decoded: {exc}"
            )
            return

        # Detect the common bad export in which a word or label becomes one text
        # object per character. Single tick labels are excluded by requiring a run.
        single_fragments = [
            item
            for item in fragments
            if len(item[0]) == 1 and (item[0].isalpha() or CJK_RE.search(item[0]))
        ]
        for index, current in enumerate(single_fragments):
            run = [current]
            for candidate in single_fragments[index + 1 :]:
                prior = run[-1]
                same_line = abs(candidate[2] - prior[2]) <= max(0.75, prior[3] * 0.35)
                same_font = candidate[4] == prior[4] and abs(candidate[3] - prior[3]) <= 0.25
                gap = candidate[1] - prior[1]
                nearby = -0.25 * max(prior[3], 1) <= gap <= 2.5 * max(prior[3], 1)
                if same_line and same_font and nearby:
                    run.append(candidate)
                    if len(run) >= 3:
                        word = "".join(item[0] for item in run)
                        self.result.fail(
                            "TEXT_FRAGMENTED",
                            f"page {stats.page_number} appears to split label {word!r} into individual character objects",
                        )
                        return
                elif candidate[2] < prior[2] - max(1.0, prior[3]):
                    break

    def _validate_page_summary(self, stats: PdfPageStats) -> None:
        clip_count = len(stats.clip_events)
        if clip_count > self.max_clips:
            self.result.fail(
                "TOO_MANY_CLIPS",
                f"page {stats.page_number} contains {clip_count} clipping paths; maximum is {self.max_clips}",
            )
        if self.expected_clips is not None and clip_count != self.expected_clips:
            self.result.fail(
                "CLIP_COUNT_EXPECTED",
                f"page {stats.page_number} contains {clip_count} clipping paths; expected exactly {self.expected_clips}",
            )
        if stats.max_clip_depth > 1:
            self.result.fail(
                "CLIP_DEPTH",
                f"page {stats.page_number} reaches clipping depth {stats.max_clip_depth}; maximum is 1",
            )
        for event in stats.clip_events:
            if not event.kinds:
                self.result.fail(
                    "EMPTY_CLIP",
                    f"page {stats.page_number} clip {event.number} contains no painted vector or live text content",
                )
        if stats.q_count != stats.Q_count:
            self.result.fail(
                "Q_COUNT",
                f"page {stats.page_number} recursively contains q={stats.q_count}, Q={stats.Q_count}",
            )
        if not self.allow_no_text and stats.text_show_count == 0:
            self.result.fail("NO_LIVE_TEXT", f"page {stats.page_number} contains no live text")
        if not self.allow_no_vector and stats.vector_paint_count == 0:
            self.result.fail("NO_VECTOR", f"page {stats.page_number} contains no painted vector path")


def validate_png(
    path: Path,
    target_dpi: float,
    dpi_tolerance: float,
    expected_size: tuple[float, float] | None,
) -> ValidationResult:
    result = ValidationResult(path, "PNG")
    if path.suffix.lower() != ".png":
        result.fail("PNG_EXTENSION", "filename does not end in .png")
    try:
        with Image.open(path) as image:
            if image.format != "PNG":
                result.fail("PNG_FORMAT", f"file content is {image.format}, not PNG")
            width, height = image.size
            dpi = image.info.get("dpi")
            image.load()
            blank_reason = image_blank_reason(image)
            if blank_reason:
                result.fail("PNG_BLANK", blank_reason)
    except Exception as exc:
        result.not_certified("PNG_DECODE", f"PNG cannot be decoded: {exc}")
        return result

    result.details.append(f"pixels={width}x{height}")
    if not dpi or len(dpi) < 2:
        result.fail("PNG_DPI_MISSING", "PNG has no two-axis physical-resolution metadata")
    else:
        x_dpi, y_dpi = float(dpi[0]), float(dpi[1])
        result.details.append(f"dpi={x_dpi:.3f}x{y_dpi:.3f}")
        if abs(x_dpi - target_dpi) > dpi_tolerance or abs(y_dpi - target_dpi) > dpi_tolerance:
            result.fail(
                "PNG_DPI",
                f"PNG resolution is {x_dpi:.3f}x{y_dpi:.3f} DPI; expected {target_dpi:g}+/-{dpi_tolerance:g}",
            )
        if expected_size:
            expected_width = expected_size[0] * target_dpi
            expected_height = expected_size[1] * target_dpi
            pixel_tolerance = max(1.0, target_dpi * 0.005)
            if abs(width - expected_width) > pixel_tolerance or abs(height - expected_height) > pixel_tolerance:
                result.fail(
                    "PNG_PIXEL_SIZE",
                    f"pixels {width}x{height} do not match {expected_size[0]:g}x{expected_size[1]:g} inches at {target_dpi:g} DPI",
                )
            result.details.append(
                f"expected-physical-size={expected_size[0]:g}x{expected_size[1]:g}in"
            )
        else:
            result.details.append(
                f"inferred-physical-size={width / x_dpi:.3f}x{height / y_dpi:.3f}in"
            )
    return result


def find_pdftoppm(explicit: str | None) -> str | None:
    if explicit:
        candidate = Path(explicit)
        return str(candidate) if candidate.is_file() else None
    return shutil.which("pdftoppm")


def _render_size_matches(
    observed: tuple[int, int], expected_points: tuple[float, float], dpi: float, rotation: int
) -> bool:
    width_points, height_points = expected_points
    if rotation in {90, 270}:
        width_points, height_points = height_points, width_points
    expected = (width_points * dpi / 72.0, height_points * dpi / 72.0)
    # Poppler normally rounds page boundaries outward. Accept at most one pixel
    # of normal raster-grid rounding in either dimension.
    return all(abs(actual - target) <= 1.05 for actual, target in zip(observed, expected))


def _prepare_render_directory(
    path: Path, result: ValidationResult, requested_base: Path | None
) -> tuple[Path | None, tempfile.TemporaryDirectory[str] | None]:
    if requested_base is not None:
        try:
            base = requested_base.expanduser().resolve(strict=False)
            if base.exists() and not base.is_dir():
                result.not_certified(
                    "RENDER_DIR_NOT_DIRECTORY", f"--render-dir is not a directory: {base}"
                )
                return None, None
            base.mkdir(parents=True, exist_ok=True)
            safe_stem = re.sub(r"[^A-Za-z0-9._-]+", "-", path.stem).strip("-.") or "figure"
            # Always create a normal child directory so Windows applies the
            # parent's inherited ACL. Path.mkdir(exist_ok=False) is atomic and
            # cannot overwrite an existing entry; the retry loop handles the
            # extremely unlikely random-name collision.
            output = None
            for _ in range(20):
                candidate = base / f"{safe_stem[:60]}-render-{secrets.token_hex(4)}"
                try:
                    candidate.mkdir(exist_ok=False)
                except FileExistsError:
                    continue
                output = candidate
                break
            if output is None:
                result.not_certified(
                    "RENDER_DIR_COLLISION",
                    "could not allocate a unique render child after 20 attempts",
                )
                return None, None
        except OSError as exc:
            result.not_certified(
                "RENDER_DIR_CREATE", f"cannot create a unique child under --render-dir: {exc}"
            )
            return None, None
        result.details.append(f"render-output-dir={output} (retained; unique child, no overwrite)")
        return output, None

    task_tmp_root = Path.cwd() / "tmp"
    try:
        task_tmp_root.mkdir(parents=True, exist_ok=True)
        temporary = tempfile.TemporaryDirectory(prefix="validate-figure-", dir=task_tmp_root)
    except OSError:
        try:
            temporary = tempfile.TemporaryDirectory(prefix="validate-figure-")
        except OSError as exc:
            result.not_certified("RENDER_DIR_CREATE", f"cannot create a temporary render directory: {exc}")
            return None, None
    return Path(temporary.name), temporary


def render_pdf(
    path: Path,
    result: ValidationResult,
    page_stats: Sequence[PdfPageStats],
    executable: str | None,
    render_dpi: float,
    render_dir: Path | None,
) -> None:
    pdftoppm = find_pdftoppm(executable)
    if not pdftoppm:
        result.not_certified("POPPLER_MISSING", "pdftoppm was not found; visual rendering is mandatory")
        return
    output_dir, temporary = _prepare_render_directory(path, result, render_dir)
    if output_dir is None:
        return
    try:
        prefix = output_dir / "page"
        command = [
            pdftoppm,
            "-png",
            "-r",
            f"{render_dpi:g}",
            "-f",
            "1",
            "-l",
            str(len(page_stats)),
            str(path),
            str(prefix),
        ]
        try:
            completed = subprocess.run(
                command,
                stdout=subprocess.PIPE,
                stderr=subprocess.PIPE,
                text=True,
                timeout=max(60, len(page_stats) * 30),
                check=False,
            )
        except Exception as exc:
            result.not_certified("POPPLER_RUN", f"Poppler rendering could not run: {exc}")
            return
        if completed.returncode != 0:
            detail = completed.stderr.strip() or f"exit code {completed.returncode}"
            result.not_certified("POPPLER_RENDER", f"Poppler failed to render the PDF: {detail}")
            return
        rendered: dict[int, Path] = {}
        for candidate in output_dir.glob("page-*.png"):
            match = re.fullmatch(r"page-(\d+)\.png", candidate.name, flags=re.IGNORECASE)
            if match:
                rendered[int(match.group(1))] = candidate
        if len(rendered) != len(page_stats) or set(rendered) != set(
            range(1, len(page_stats) + 1)
        ):
            result.not_certified(
                "POPPLER_PAGE_COUNT",
                f"Poppler produced page numbers {sorted(rendered)}, expected 1..{len(page_stats)}",
            )
            return
        for stats in page_stats:
            output = rendered[stats.page_number]
            try:
                with Image.open(output) as image:
                    image.verify()
                with Image.open(output) as image:
                    image.load()
                    observed_size = image.size
                    blank_reason = image_blank_reason(image)
            except Exception as exc:
                result.not_certified("POPPLER_OUTPUT", f"rendered page {output.name} is invalid: {exc}")
                continue
            if stats.rotation not in {0, 90, 180, 270}:
                result.not_certified(
                    "PAGE_ROTATION",
                    f"page {stats.page_number} has unsupported /Rotate={stats.rotation}",
                )
                matched_box = None
            elif _render_size_matches(
                observed_size, stats.crop_size_points, render_dpi, stats.rotation
            ):
                matched_box = "CropBox"
            elif _render_size_matches(
                observed_size, stats.media_size_points, render_dpi, stats.rotation
            ):
                matched_box = "MediaBox"
            else:
                matched_box = None
                result.fail(
                    "RENDER_PIXEL_SIZE",
                    f"page {stats.page_number} rendered to {observed_size[0]}x{observed_size[1]} px at {render_dpi:g} DPI; "
                    f"this matches neither CropBox {stats.crop_size_points[0]:.3f}x{stats.crop_size_points[1]:.3f} pt "
                    f"nor MediaBox {stats.media_size_points[0]:.3f}x{stats.media_size_points[1]:.3f} pt",
                )
            if blank_reason:
                result.fail(
                    "RENDER_BLANK",
                    f"page {stats.page_number} rendered blank: {blank_reason}",
                )
            match_text = matched_box or "no-box-match"
            content_text = "blank" if blank_reason else "nonblank"
            result.details.append(
                f"render-page={stats.page_number} pixels={observed_size[0]}x{observed_size[1]} "
                f"dpi={render_dpi:g} box={match_text} content={content_text}"
            )
        result.details.append(f"poppler-rendered-pages={len(rendered)}")
    finally:
        if temporary is not None:
            temporary.cleanup()


def validate_pdf(path: Path, args: argparse.Namespace) -> ValidationResult:
    result = ValidationResult(path, "PDF")
    if path.suffix.lower() != ".pdf":
        result.fail("PDF_EXTENSION", "filename does not end in .pdf")
    auditor = PdfAuditor(
        path,
        result,
        max_clips=args.max_clips,
        expected_clips=args.expected_clips,
        declared_clip_boxes=args.clip_box,
        allow_raster_images=args.allow_raster_images,
        allow_form_xobjects=args.allow_form_xobjects,
        allow_no_text=args.allow_no_text,
        allow_no_vector=args.allow_no_vector,
    )
    page_stats = auditor.audit()
    if page_stats:
        for stats in page_stats:
            fonts = sorted({font.base_font for font in stats.font_uses.values()})
            clip_summary = "; ".join(
                f"#{event.number} depth={event.depth} kinds={'+'.join(sorted(event.kinds)) or 'empty'}"
                for event in stats.clip_events
            )
            result.details.append(
                f"page={stats.page_number} clips={len(stats.clip_events)} max-depth={stats.max_clip_depth} "
                f"q/Q={stats.q_count}/{stats.Q_count} text-shows={stats.text_show_count} "
                f"vector-paints={stats.vector_paint_count} images={stats.image_count + stats.inline_image_count} "
                f"forms={stats.form_calls} fonts={','.join(fonts) or 'none'}"
            )
            if clip_summary:
                result.details.append(f"page={stats.page_number} clip-groups: {clip_summary}")
        render_pdf(
            path,
            result,
            page_stats,
            args.pdftoppm,
            args.render_dpi,
            args.render_dir,
        )
    return result


def parse_box(value: str) -> tuple[float, float, float, float]:
    try:
        fields = tuple(float(part.strip()) for part in value.split(","))
    except ValueError as exc:
        raise argparse.ArgumentTypeError("clip box must be x,y,width,height") from exc
    if len(fields) != 4 or fields[2] <= 0 or fields[3] <= 0:
        raise argparse.ArgumentTypeError("clip box must be x,y,width,height with positive size")
    return fields  # type: ignore[return-value]


def parse_size(value: str) -> tuple[float, float]:
    match = re.fullmatch(r"\s*([0-9]*\.?[0-9]+)\s*[xX×]\s*([0-9]*\.?[0-9]+)\s*", value)
    if not match:
        raise argparse.ArgumentTypeError("expected size must be WIDTHxHEIGHT in inches")
    return (float(match.group(1)), float(match.group(2)))


def positive_float(value: str) -> float:
    try:
        parsed = float(value)
    except ValueError as exc:
        raise argparse.ArgumentTypeError("value must be a positive number") from exc
    if not math.isfinite(parsed) or parsed <= 0:
        raise argparse.ArgumentTypeError("value must be a finite positive number")
    return parsed


def build_parser() -> argparse.ArgumentParser:
    parser = argparse.ArgumentParser(
        description="Validate 600-DPI PNGs or Illustrator-editable vector figure PDFs."
    )
    parser.add_argument("files", nargs="+", type=Path)
    parser.add_argument("--target-dpi", type=float, default=600.0)
    parser.add_argument("--dpi-tolerance", type=float, default=1.0)
    parser.add_argument(
        "--expected-size-in",
        type=parse_size,
        metavar="WIDTHxHEIGHT",
        help="also verify PNG pixel dimensions against a physical size in inches",
    )
    parser.add_argument("--max-clips", type=int, default=4, choices=range(0, 5))
    parser.add_argument(
        "--expected-clips",
        type=int,
        choices=range(0, 5),
        help="require exactly this many clipping paths on every PDF page (use 3 for the full-period regression profile)",
    )
    parser.add_argument(
        "--clip-box",
        type=parse_box,
        action="append",
        default=[],
        metavar="X,Y,W,H",
        help="declare an allowed non-page rectangular PDF clipping box (repeatable)",
    )
    parser.add_argument(
        "--allow-raster-images",
        action="store_true",
        help="allow intentional image XObjects; masks and inline images remain forbidden",
    )
    parser.add_argument(
        "--allow-form-xobjects",
        action="store_true",
        help="allow recursively inspected Form XObjects; the strict default rejects all Forms",
    )
    parser.add_argument("--allow-no-text", action="store_true")
    parser.add_argument("--allow-no-vector", action="store_true")
    parser.add_argument("--pdftoppm", help="explicit path to Poppler pdftoppm")
    parser.add_argument(
        "--render-dpi",
        type=positive_float,
        default=72.0,
        help="Poppler verification resolution (default: 72 DPI)",
    )
    parser.add_argument(
        "--render-dir",
        type=Path,
        help="retain rendered pages in a newly created unique child of DIR; existing contents are never overwritten",
    )
    return parser


def print_result(result: ValidationResult) -> None:
    print(f"[{result.status}] {result.path}")
    for detail in result.details:
        print(f"  INFO {detail}")
    for issue in result.issues:
        label = "ERROR" if issue.outcome == "FAIL" else "BLOCKER"
        print(f"  {label} [{issue.code}] {issue.message}")


def main(argv: Sequence[str] | None = None) -> int:
    parser = build_parser()
    args = parser.parse_args(argv)
    if args.expected_clips is not None and args.expected_clips > args.max_clips:
        parser.error("--expected-clips cannot exceed --max-clips")
    results: list[ValidationResult] = []
    for raw_path in args.files:
        path = raw_path.resolve()
        if not path.is_file():
            result = ValidationResult(path, "UNKNOWN")
            result.not_certified("FILE_MISSING", "file does not exist or is not a regular file")
        elif path.suffix.lower() == ".png":
            result = validate_png(path, args.target_dpi, args.dpi_tolerance, args.expected_size_in)
        elif path.suffix.lower() == ".pdf":
            result = validate_pdf(path, args)
        else:
            result = ValidationResult(path, "UNKNOWN")
            result.not_certified("FILE_TYPE", "only .png and .pdf are supported")
        results.append(result)
        print_result(result)
    statuses = Counter(result.status for result in results)
    passed = statuses["PASS"]
    print(
        f"SUMMARY passed={passed} failed={statuses['FAIL']} "
        f"not-certified={statuses['NOT CERTIFIED']} total={len(results)}"
    )
    return 0 if passed == len(results) else 1


if __name__ == "__main__":
    raise SystemExit(main())
```

# Supplementary Figure 4B: Neurod1

Set the paths marked in the code before running this notebook.

The program fits, module memberships, activity values and UMAP are read from
`Fig4B-D.h5`, produced by `05_fig4B-D_SGN.ipynb`. The renderer comes from that
notebook's `render_factor` function. The first export draws the C/D assembly component.
The original physical coordinates are translated to this panel's canvas without
rescaling points, text, curves or network ribbons.

Selected modules: [0, 6]; gene counts: [36, 20]. Existing fitted values are
used directly. Module selection, activity normalization and fitting are unchanged.

This panel was assembled from two source-native components. The first export
contains the original Neurod1 module curves, module-expression maps and network;
its old, unavailable activity view is omitted. The second export is the final
`Neurod1-B_1` paper-reference activity view, which replaces the left component
when assembling the row in Illustrator. Their original scores and display caps
are kept separate. The four output file paths must be set independently.

The replacement renderer is extracted from the `method/Project-render.py`
source embedded in the notebook's final paper-reference block. Its original
H5 cache is read directly from `Activity.h5`, without restoring another data file.


```python
#1. =========== Paths and fixed display settings
from pathlib import Path
import hashlib
import json
import math
import os
import re
import sys
import h5py
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from matplotlib import font_manager
from PIL import Image as PILImage

ROOT = Path('.')  # Set the working root before replacing the path placeholders.
# Uses Fig4B-D.h5 generated by the SGN program and AUCell/GAM steps in 05_fig4B-D_SGN.ipynb.
RESULT_FILE = ROOT  # Set the saved program-result H5 file path.
# Uses Candidates.h5 and Inputs.h5 generated by the candidate-regulator analysis.
UPSTREAM_MODULES = ROOT  # Set the Candidates.h5 path for the original input-hash check.
UPSTREAM_INPUTS = ROOT  # Set the Inputs.h5 path for the original input-hash check.
PNG_FILE = ROOT  # Set this panel's output PNG file path.
PDF_FILE = ROOT  # Set this panel's output PDF file path.
OUTPUT_DIR = ROOT  # Set the output directory path.
WORK_DIR = ROOT  # Set the temporary working directory path.
FONT_FILES = {
    'normal': ROOT,  # Set the Arial regular TTF file path.
    'italic': ROOT,  # Set the Arial italic TTF file path.
    'bold': ROOT,  # Set the Arial bold TTF file path.
    'bolditalic': ROOT,  # Set the Arial bold italic TTF file path.
}
os.environ['MPLCONFIGDIR'] = str(WORK_DIR)
sys.dont_write_bytecode = True
for font_file in FONT_FILES.values():
    font_manager.fontManager.addfont(str(font_file))

FACTORS = ('Foxo6', 'Zfpm2', 'Neurod2', 'Nhlh1', 'Neurod1')
SEED = 42

FACTOR = 'Neurod1'
PANEL_PARTS = ('C', 'D')
VIEW_X0, VIEW_X1 = 200.0, 1458.0
EXPECTED_MODULES = [0, 6]
EXPECTED_GENES = [36, 20]
```

```python
def file_sha256(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as stream:
        for block in iter(lambda: stream.read(4 * 1024 * 1024), b''):
            digest.update(block)
    return digest.hexdigest()

def read_result_value(node):
    if isinstance(node, h5py.Group):
        kind = node.attrs.get('kind', 'dict')
        if kind == 'dataframe':
            columns = json.loads(node.attrs['columns'])
            table = pd.DataFrame({col: read_result_value(node[col]) for col in columns})
            table.index = read_result_value(node['_index'])
            return table
        if kind == 'list':
            return [read_result_value(node[key]) for key in sorted(node, key=int)]
        return {key: read_result_value(node[key]) for key in node}
    if node.attrs.get('kind') == 'none':
        return None
    return node.asstr()[()] if node.dtype.kind in 'OSU' else node[()]

PROGRAM_SELECTION = {'Foxo6': [0, 1], 'Zfpm2': [0, 2], 'Neurod2': [1, 0],
                     'Nhlh1': [0, 2], 'Neurod1': [0, 6]}

PROGRAM_TITLES = {'Foxo6': ['Transient developmental genes', 'Broad early-rising genes', 'Early transition-associated genes'], 'Zfpm2': ['Neuronal morphogenesis genes', 'Broad early-rising genes', 'Transient developmental genes', 'Early-high genes, inverse-associated'], 'Neurod2': ['Broad early-rising genes', 'Transient developmental genes'], 'Nhlh1': ['Transient developmental genes', 'Weakly varying genes', 'Early transition-associated genes'], 'Neurod1': ['Postsynaptic receptor regulation genes, inverse-associated', 'Mid-trajectory peaking genes, inverse-associated', 'Late-rising genes (Calb1/Kcna1), inverse-associated', 'Progressively rising genes, mostly inverse-associated', 'Early-high genes', 'Late-rising genes (Cnr1/Kcnq2), inverse-associated', 'Transient differentiation-associated genes']}
```

```python
#2. =========== Read the frozen module fits and activity
assert file_sha256(RESULT_FILE) == '7d1c9030ae2052453521e474f77577f0ad2b5293f4717e37b3b4125b99ef2ff9', 'The accepted SGN result archive changed.'
expected_program_method = {
    'upstream_modules_sha256': file_sha256(UPSTREAM_MODULES),
    'upstream_inputs_sha256': file_sha256(UPSTREAM_INPUTS),
    'GAM_source_sha256': 'c0820327f2d4df7160ec3a92827c7415b150aab2d91eac5d76a7ad68467b9eb3',
    'GAM_k_max': 10, 'GAM_gamma': 1.5, 'GAM_grid_n': 250, 'GAM_method': 'GCV.Cp',
}
with h5py.File(RESULT_FILE, 'r') as handle:
    assert handle.attrs.get('schema') == 'Fig4BD_all_programs_v2'
    shared = read_result_value(handle['shared'])
    module_results = {tf: read_result_value(handle['factors/' + tf]) for tf in FACTORS}
for tf, saved in module_results.items():
    assert all(saved['program_provenance'].get(key) == value
               for key, value in expected_program_method.items()), tf
    saved.update(shared)
    assert list(saved['selected_modules']) == PROGRAM_SELECTION[tf]
    assert saved['module_scores'].shape[1] == len(saved['modules'])
    assert saved['fitted'].shape[1] == len(saved['modules']) + 1
assert sum(len(saved['modules']) for saved in module_results.values()) == 19
result = module_results[FACTOR]
assert len(result['obs']) == 7355 and result['X_umap'].shape == (7355, 2)
assert list(result['selected_modules']) == EXPECTED_MODULES
assert [result['modules'][m]['n_genes'] for m in EXPECTED_MODULES] == EXPECTED_GENES
```

```python
"""Strict vector PDF export for the Fig4 B/C/D Matplotlib canvas.

No analysis is performed. Each native Matplotlib label becomes one live Arial
text object, including the TF italic run in the two supported mathtext titles.
All geometry is written directly, without XObjects, glyph outlines or images.
The caller declares at most four data axes; their clip rectangles are collected
into one sibling viewport group per axis. Other artists must fit their axes.
"""
from pathlib import Path
import re
import sys
import math
import numpy as np
from matplotlib.backend_bases import RendererBase
from matplotlib.path import Path as MplPath

from reportlab.pdfgen import canvas
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont

_FONTS = {'normal':'FigArial','italic':'FigArialItalic',
          'bold':'FigArialBold','bolditalic':'FigArialBoldItalic'}

def _register():
    for key, file in FONT_FILES.items():
        if _FONTS[key] not in pdfmetrics.getRegisteredFontNames():
            pdfmetrics.registerFont(TTFont(_FONTS[key],str(file)))

def _font(prop):
    bold=prop.get_weight() in ('bold','heavy','black','semibold','demibold') or isinstance(prop.get_weight(),(int,float)) and prop.get_weight()>=600
    italic=prop.get_style() in ('italic','oblique')
    return _FONTS['bolditalic' if bold and italic else 'bold' if bold else 'italic' if italic else 'normal']

def _runs(text,prop,ismath):
    if not ismath:
        if '$' in text: raise ValueError('Unexpected dollar in plain label: '+text)
        return [(text,_font(prop))]
    # Current Fig4 titles use only this explicitly audited mathtext form.
    found=re.fullmatch(r'\$\\it\{([^{}]+)\}([^$]*)\$(.*)',text)
    if found is None: raise ValueError('Unsupported math label: '+text)
    tf,inside,after=found.groups()
    return [(tf,_FONTS['italic']),(inside+after,_FONTS['normal'])]

class _Recorder(RendererBase):
    def __init__(self,fig,agg,viewports,halo_labels):
        super().__init__()
        self.dpi=fig.dpi;self.unit=72/fig.dpi;self.agg=agg
        self.width,self.height=fig.bbox.size;self.ops=[];self.viewports=viewports;self.halo_labels=halo_labels
    def points_to_pixels(self,p): return p/self.unit
    def flipy(self):return False
    def get_canvas_width_height(self):return self.width,self.height
    def get_text_width_height_descent(self,s,prop,ismath):
        runs=_runs(s,prop,ismath)
        width=sum(pdfmetrics.stringWidth(t,f,prop.get_size_in_points()) for t,f in runs)/self.unit
        # Match the source Arial baseline/ascent geometry; width uses the same
        # embedded font as the PDF, including the semantic mixed italic title.
        _,h,d=self.agg.get_text_width_height_descent(re.sub(r'\$|\\it\{|\}', '',s),prop,False)
        return width,h,d
    def _clip(self,gc):
        if gc.get_clip_path()[0] is not None:
            raise ValueError('Nonrectangular artist clip is unsupported')
        b=gc.get_clip_rectangle()
        if b is None:return None
        for i,box in enumerate(self.viewports):
            if np.allclose(b.bounds,box,rtol=0,atol=1e-5):return i
        return None
    def draw_path(self,gc,path,transform,rgbFace=None):
        if gc.get_hatch():raise ValueError('Hatching/Pattern is forbidden')
        p=transform.transform_path(path)
        self.ops.append(('path',self._clip(gc),p,gc.get_rgb(),rgbFace,
                         gc.get_linewidth(),gc.get_alpha(),gc.get_forced_alpha(),
                         gc.get_dashes(),gc.get_capstyle(),gc.get_joinstyle()))
    def draw_text(self,gc,x,y,s,prop,angle,ismath=False,mtext=None):
        self.ops.append(('text',None,float(x),float(y),str(s),prop.copy(),float(angle),
                         ismath,gc.get_rgb(),gc.get_alpha(),id(mtext) in self.halo_labels))
    def draw_image(self,*args,**kwargs):raise ValueError('Raster image is forbidden')
    def start_rasterizing(self):pass
    def stop_rasterizing(self):pass
    def start_filter(self):raise ValueError('Raster filter is forbidden')
    def stop_filter(self,*args):raise ValueError('Raster filter is forbidden')

def _pdf_path(pdf,path,unit):
    out=pdf.beginPath();current=None;start=None
    for vertices,code in path.iter_segments(remove_nans=True,simplify=False,curves=True):
        v=np.asarray(vertices).reshape(-1,2)*unit
        if code==MplPath.MOVETO:
            current=v[0];start=current;out.moveTo(*current)
        elif code==MplPath.LINETO:
            current=v[0];out.lineTo(*current)
        elif code==MplPath.CURVE3:
            control,end=v
            a=current+(control-current)*2/3;b=end+(control-end)*2/3
            out.curveTo(*a,*b,*end);current=end
        elif code==MplPath.CURVE4:
            out.curveTo(*v.ravel());current=v[-1]
        elif code==MplPath.CLOSEPOLY:
            out.close();current=start
        elif code==MplPath.STOP:break
        else:raise ValueError('Unknown path opcode '+str(code))
    return out

def strict_editable_pdf(fig,destination,*,data_axes=(),live_halo_width=.20):
    """Write a strict vector PDF. Return declared clip boxes in PDF points.

    The artist canvas remains unchanged after return. Rasterization flags and
    white text path effects are replaced by a narrow live text fill+stroke
    (render mode 2) in the same single label object; no outlined font paths are
    emitted. The width is narrower than Agg's halo to preserve the black stems.
    """
    _register()
    if len(data_axes)>4:raise ValueError('At most four sibling data viewports')
    fig.canvas.draw();agg=fig.canvas.get_renderer()
    viewports=[tuple(ax.bbox.bounds) for ax in data_axes]
    effects=[]
    for artist in fig.findobj():
        if hasattr(artist,'get_path_effects'):
            old=artist.get_path_effects()
            if old:effects.append((artist,old));artist.set_path_effects([])
    rec=_Recorder(fig,agg,viewports,{id(a) for a,e in effects})
    try:fig.draw(rec)
    finally:
        for a,e in effects:a.set_path_effects(e)
    dst=Path(destination);dst.parent.mkdir(parents=True,exist_ok=True)
    pdf=canvas.Canvas(str(dst),pagesize=tuple(fig.get_size_inches()*72),
                      pageCompression=1,invariant=1,initialFontName=_FONTS['normal'])
    unit=rec.unit
    def emit(op):
        if op[0]=='text':
            _,_,x,y,s,prop,angle,ismath,color,alpha,halo=op
            if not s:return
            pdf.saveState();pdf.setFillColorRGB(*color[:3]);pdf.setFillAlpha(alpha)
            text=pdf.beginText();rad=math.radians(angle)
            if halo and live_halo_width>0:
                pdf.setStrokeColorRGB(1,1,1);pdf.setStrokeAlpha(alpha)
                pdf.setLineWidth(float(live_halo_width));text.setTextRenderMode(2)
            text.setTextTransform(math.cos(rad),math.sin(rad),-math.sin(rad),math.cos(rad),x*unit,y*unit)
            for value,font in _runs(s,prop,ismath):
                if value:text.setFont(font,prop.get_size_in_points());text.textOut(value)
            pdf.drawText(text);pdf.restoreState();return
        _,_,path,stroke,face,lw,alpha,forced,dashes,cap,join=op
        pdf.saveState();pdf.setLineWidth(lw)
        pdf.setLineCap({'butt':0,'round':1,'projecting':2}[cap])
        pdf.setLineJoin({'miter':0,'round':1,'bevel':2}[join])
        offset,pattern=dashes
        pdf.setDash([] if pattern is None else [float(v) for v in pattern],0.0 if offset is None else float(offset))
        if stroke is not None:
            pdf.setStrokeColorRGB(*stroke[:3]);pdf.setStrokeAlpha(alpha if forced else (stroke[3] if len(stroke)>3 else alpha))
        if face is not None:
            pdf.setFillColorRGB(*face[:3]);pdf.setFillAlpha(alpha if forced else (face[3] if len(face)>3 else alpha))
        pdf.drawPath(_pdf_path(pdf,path,unit),stroke=int(stroke is not None and lw>0),fill=int(face is not None),fillMode=1)
        pdf.restoreState()
    done=set()
    for op in rec.ops:
        key=op[1]
        if key is None:emit(op);continue
        if key in done:continue
        done.add(key);pdf.saveState()
        p=pdf.beginPath();p.rect(*(np.asarray(viewports[key])*unit))
        pdf.clipPath(p,stroke=0,fill=0,fillMode=1)
        for candidate in rec.ops:
            if candidate[1]==key:emit(candidate)
        pdf.restoreState()
    pdf.showPage();pdf.save()
    if done!=set(range(len(data_axes))):raise AssertionError('Declared clip has no artist')
    return {'expected_clips':len(done),'clip_boxes_pt':(np.asarray(viewports)*unit).tolist(),
            'editable_label_objects':sum(op[0]=='text' for op in rec.ops),
            'label_strings':[op[4] for op in rec.ops if op[0]=='text'],
            'vector_paths':sum(op[0]=='path' for op in rec.ops),
            'text_effects':'White live text fill+stroke, single object; stroke width '+str(live_halo_width)+' pt.',
            'halo_labels':len(effects)}
```

```python
"""Published Fig.3d vector ribbons with current module gene names substituted."""

import numpy as np
import matplotlib.patheffects as pe
from matplotlib.path import Path as MplPath
from matplotlib.patches import PathPatch
from matplotlib.font_manager import FontProperties
from matplotlib.textpath import TextPath


# Literature-discussed representatives, not significance or stronger-edge gates.

NETWORK_HIGHLIGHT_EVIDENCE = {'Foxo6': {'Nhlh1': {'reason': 'Highlighted in the reference paper as a neurogenesis-related member.', 'source': 'https://www.nature.com/articles/s41467-022-31580-1'}, 'Sox11': {'reason': 'Representative neuronal differentiation factor; general CNS evidence, not demonstrated SGN fate specification.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC1698453/'}}, 'Zfpm2': {'Isl1': {'reason': 'Representative with direct cochleovestibular neuron differentiation, migration and axon evidence; does not validate a Zfpm2-to-Isl1 edge.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC10883659/'}, 'Sox11': {'reason': 'Neuronal differentiation representative; general CNS evidence.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC1698453/'}}, 'Neurod2': {'Nhlh1': {'reason': 'Reference-paper neurogenesis representative already present in this module.', 'source': 'https://www.nature.com/articles/s41467-022-31580-1'}, 'Neurod6': {'reason': 'Reference-paper neurogenesis representative already present in this module.', 'source': 'https://www.nature.com/articles/s41467-022-31580-1'}, 'Isl1': {'reason': 'Direct cochleovestibular developmental evidence; does not establish the current association as direct regulation.', 'source': 'https://pmc.ncbi.nlm.nih.gov/articles/PMC10883659/'}}, 'Nhlh1': {'Nhlh2': {'reason': 'Developmental sensory-neuron expression representative; expression evidence is not functional necessity and Nhlh1 knockout effects are not transferred to Nhlh2.', 'source': 'https://pubmed.ncbi.nlm.nih.gov/17004922/'}}, 'Neurod1': {'Myt1l': {'reason': 'Representative neuronal identity factor; not evidence of direct Neurod1 regulation.', 'source': 'https://www.nature.com/articles/nature21722'}, 'Sema3a': {'reason': 'Representative with auditory-neuron connectivity/excitability evidence; not early SGN fate or direct Neurod1-target validation.', 'source': 'https://pubmed.ncbi.nlm.nih.gov/36284453/'}}, 'Prox1': {}}
NETWORK_HIGHLIGHTS = {tf: tuple(genes) for tf, genes in NETWORK_HIGHLIGHT_EVIDENCE.items()}

PAPER_NETWORK_TEMPLATE = {'source': 'Petitpre et al. 2022, Nature Communications, Fig.3d', 'path': str(ROOT), 'sha256': '8dda1b31fe727e1ee88d73932b6b0dd485ab070855ec39bd4d5587e1b2826bd4', 'page_1based': 6, 'bbox': [302.0, 208.0, 466.0, 286.0], 'center': [384.737473256552, 248.26325348336002], 'base_font_size': 4.34579991999999, 'red_font_size': 4.1339998199999854, 'red_rgb': [0.72941, 0.0, 0.1098], 'guides': [{'path': [['m', [383.54000854, 219.98101855999994]], ['l', [383.54000854, 221.68801856999994]]], 'pts': [[383.54000854, 219.98101855999994], [383.54000854, 221.68801856999994]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 219.98101855999994, 'bottom': 221.68801856999994, 'linewidth': 0.329, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 222.28698780000002]], ['l', [383.54000854, 242.96197558999995]]], 'pts': [[383.54000854, 222.28698780000002], [383.54000854, 242.96197558999995]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 222.28698780000002, 'bottom': 242.96197558999995, 'linewidth': 0.329, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[3.5955, 0.59925997], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 254.99102832999995]], ['l', [383.54000854, 256.5500283399999]]], 'pts': [[383.54000854, 254.99102832999995], [383.54000854, 256.5500283399999]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 254.99102832999995, 'bottom': 256.5500283399999, 'linewidth': 0.30000001, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 257.04101612]], ['l', [383.54000854, 274.01001025999994]]], 'pts': [[383.54000854, 257.04101612], [383.54000854, 274.01001025999994]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 257.04101612, 'bottom': 274.01001025999994, 'linewidth': 0.30000001, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[2.95099997, 0.49184], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}, {'path': [['m', [383.54000854, 274.25601245999997]], ['l', [383.54000854, 275.81401242]]], 'pts': [[383.54000854, 274.25601245999997], [383.54000854, 275.81401242]], 'x0': 383.54000854, 'x1': 383.54000854, 'top': 274.25601245999997, 'bottom': 275.81401242, 'linewidth': 0.30000001, 'stroking_color': [0.70196002, 0.70196002, 0.70196002], 'dash': [[], 0], 'stroke': True, 'fill': False, 'stroke_opacity': 1.0}], 'slots': {'left': [{'visual_slot': 'St18', 'point': [346.20866888772, 262.10511139336006], 'bbox': [341.73901367, 259.93221143336007, 350.67832410544, 264.27801135336006], 'path': [['m', [384.55599975, 247.91198780000002]], ['c', [382.62599981, 252.00998784], [377.81299972, 255.82098771000005], [372.05499935, 258.58798744]], ['c', [366.29899979000004, 261.35598804000006], [359.60300064, 263.08798744], [353.93199921, 263.086988]], ['c', [351.02399825000003, 263.086988], [348.38800049, 262.63198806], [346.2879982, 261.63998748000006]], ['l', [346.11899948, 261.99698782999997]], ['c', [348.29299927, 263.02298786000006], [350.98299789000004, 263.48098804000006], [353.93199921, 263.48198750000006]], ['c', [359.68199921, 263.48098804000006], [366.42000008, 261.7349877500001], [372.22699928000003, 258.94398738000007]], ['c', [378.03099965, 256.14898826], [382.9089998, 252.31698800000004], [384.91299975000004, 248.0809878]], ['l', [384.55599975, 247.91198780000002]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Nhlh1', 'point': [356.088734364884, 244.5250502825599], 'bbox': [350.6889038, 242.4580503725599, 361.48856492976796, 246.5920501925599], 'path': [['m', [384.61499023, 247.83898975]], ['c', [382.29399013, 249.61298979000003], [378.49899006, 250.44698978999998], [374.39698982, 250.44498968999994]], ['c', [367.40699005, 250.45298982999998], [359.5329895, 248.02498975000003], [356.52198982, 244.05298974000004]], ['l', [356.20698928999997, 244.29298974999995]], ['c', [359.38298987999997, 248.43698973000005], [367.31798935, 250.83198977999996], [374.39698982, 250.83998967000002]], ['c', [378.55099011, 250.83798981999996], [382.41299033, 250.01098967999997], [384.85399022999997, 248.15398975000005]], ['l', [384.61499023, 247.83898975]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Rd3', 'point': [322.756078766624, 226.05335357336003], 'bbox': [318.77011108, 223.88045361336003, 326.742046453248, 228.22625353336002], 'path': [['m', [384.6499939, 247.81799365999996]], ['c', [381.24699379000003, 249.42599369999994], [376.98399402, 250.16099358999998], [372.28799344000004, 250.16099358999998]], ['c', [363.14399339000005, 250.16199375999997], [352.3639946, 247.37599365999995], [343.0929947, 242.91899349999994]], ['c', [333.81999207, 238.46499397999992], [326.06099320000004, 232.33299401], [322.9339943, 225.7029938899999]], ['l', [322.57699204000005, 225.87199451999993]], ['c', [325.77799226, 232.63999320999994], [333.60199357000005, 238.79299404999995], [342.92199326, 243.27499344]], ['c', [352.24299240000005, 247.75399366], [363.06499482000004, 250.5549936399999], [372.28799344000004, 250.55599356999994]], ['c', [377.0249939, 250.55599356999994], [381.34199382, 249.81699360999994], [384.8189939, 248.17499365999993]], ['l', [384.6499939, 247.81799365999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'St6galnac5', 'point': [345.172961909844, 238.51819732336003], 'bbox': [334.2999878, 236.34529736336003, 356.045936019688, 240.69109728336002], 'path': [['m', [384.63101196, 247.82800341999996]], ['c', [381.88601208, 249.52700341999991], [377.98401213, 250.3030033199999], [373.67301178, 250.30200338999998]], ['c', [368.28301239, 250.3030033199999], [362.25501251000003, 249.09300339999993], [357.04801178, 247.01500339999996]], ['c', [351.83901215000003, 244.93900349], [347.45501328, 241.98800326999992], [345.33501053000003, 238.5520034]], ['l', [344.999012, 238.76000359]], ['c', [347.19801331, 242.31200361999993], [351.64701081000004, 245.28300334999994], [356.90201186, 247.38200341999993]], ['c', [362.15801239, 249.47900343999993], [368.22401238000003, 250.69700336999995], [373.67301178, 250.69700336999995]], ['c', [378.03101206, 250.69600344999992], [381.99501205, 249.92000341999994], [384.83901196000005, 248.16500342999996]], ['l', [384.63101196, 247.82800341999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ntrk1', 'point': [325.02116664444, 241.59320952335997], 'bbox': [319.83010864, 239.42030956335998, 330.21222464887995, 243.76610948335997], 'path': [['m', [384.61099243, 247.84198046999995]], ['c', [379.48399258, 251.94298029999993], [370.88299275, 253.86098052], [361.68199158, 253.85898066999994]], ['c', [347.34299088, 253.86598062999997], [331.55499268, 249.20198047999997], [325.17399216, 241.19598056999996]], ['l', [324.8659935, 241.44298030999994]], ['c', [331.40899277, 249.61398041999996], [347.2599907, 254.24698066999997], [361.68199158, 254.25398064]], ['c', [370.93699265000004, 254.25098038], [379.60499239, 252.34198046999995], [384.85799243, 248.1509804499999]], ['l', [384.61099243, 247.84198046999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Egf', 'point': [335.78853088996004, 244.15320709336004], 'bbox': [332.52700805, 241.98030713336004, 339.05005372992, 246.32610705336003], 'path': [['m', [384.60598754, 247.84698535999996]], ['c', [380.19398737, 251.64298536], [372.34598732, 253.4589853399999], [364.09498787, 253.45598555999993]], ['c', [352.93998718, 253.46198513000002], [341.05598831, 250.13998533999995], [335.93298721, 244.16198541999995]], ['l', [335.63398742, 244.4199853199999]], ['c', [340.91498564, 250.54998541999998], [352.86398697, 253.8449855], [364.09498787, 253.85098554]], ['c', [372.40198707, 253.84798528], [380.31898737, 252.04398537999998], [384.86298754, 248.14598535999994]], ['l', [384.60598754, 247.84698535999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Clvs1', 'point': [336.701124429984, 255.54914215335998], 'bbox': [331.26800537, 253.37624219335999, 342.134243489968, 257.72204211336], 'path': [['m', [384.57598876, 247.87902880000001]], ['c', [379.39998864999995, 254.92302894], [365.57698822, 259.28602838000006], [353.48098945, 259.27702855999996]], ['c', [346.81998825, 259.27902841], [340.68598937999997, 257.95902872], [336.81698988999995, 255.10202884]], ['l', [336.58198927999996, 255.42002867999997]], ['c', [340.56798934, 258.35502863], [346.76898955999997, 259.67102861], [353.48098945, 259.67202901999997]], ['c', [365.66898918, 259.66402863999997], [379.55198859999996, 255.33602857000005], [384.89398874, 248.11402880000003]], ['l', [384.57598876, 247.87902880000001]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Raver2', 'point': [351.98725317106005, 222.76331451336], 'bbox': [344.98291015, 220.59041455336, 358.99159619212, 224.93621447336], 'path': [['m', [384.70999145, 247.79998829]], ['c', [384.20099145, 247.86398828999995], [383.67499149, 247.89498829000001], [383.13399148, 247.89498829000001]], ['c', [377.3329916, 247.89898829000003], [369.91099167000004, 244.30598834], [363.79399109, 239.31598809000002]], ['c', [357.67199135000004, 234.33098842999993], [352.85699081, 227.94598818999998], [352.18299103000004, 222.47198918000004]], ['l', [351.79099274000004, 222.52098896999996]], ['c', [352.50099182, 228.17798855], [357.37999153000004, 234.59198811], [363.54499054, 239.62198783999997]], ['c', [369.71499157, 244.64698818], [377.17499161, 248.2849883], [383.13399148, 248.28998829]], ['c', [383.68999147, 248.28998829], [384.23199147, 248.25798828999996], [384.75899145, 248.19198827000002]], ['l', [384.70999145, 247.79998829]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Foxo6', 'point': [331.56035424882396, 219.52234771335998], 'bbox': [325.5209961, 217.34944775335998, 337.599712397648, 221.69524767335997], 'path': [['m', [384.67800903, 247.80700732999992]], ['c', [382.77300906, 248.37600732999988], [380.65000916, 248.64300734999995], [378.37800883, 248.64300734999995]], ['c', [369.85500908, 248.64500732999988], [359.24900818, 244.87600732999988], [350.2530098, 239.30200721999995]], ['c', [341.2550087, 233.73200752999992], [333.87100983, 226.34800769999993], [331.74401093, 219.2010064299999]], ['l', [331.36600876, 219.31400729999996]], ['c', [333.55200958, 226.62300731999994], [341.00000763, 234.03100730999995], [350.04500961, 239.63800765999997]], ['c', [359.09300995, 245.2410073499999], [369.74100876, 249.03600729999994], [378.37800883, 249.03800726999987]], ['c', [380.68100906, 249.03800726999987], [382.84200907, 248.76700729999993], [384.79100903, 248.18600732999994]], ['l', [384.67800903, 247.80700732999992]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Shh', 'point': [347.06160342884004, 250.46924712336], 'bbox': [343.1960144, 248.29634716336, 350.92719245768, 252.64214708336], 'path': [['m', [384.58700561, 247.86499072999993]], ['c', [380.7070055, 252.23799086999998], [371.95900536, 254.62499094999998], [363.6190052, 254.6199908399999]], ['c', [357.00400543, 254.6229906299999], [350.64800644, 253.12199068999996], [347.18900681, 250.03599072999998]], ['l', [346.92700577, 250.3309908099999]], ['c', [350.52200699, 253.52399063999997], [356.94600487, 255.01099062999992], [363.6190052, 255.01499082999987]], ['c', [372.03200531, 255.00899077999998], [380.84500551, 252.6469907899999], [384.88200559, 248.12799073999997]], ['l', [384.58700561, 247.86499072999993]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Cecr2', 'point': [341.931003113304, 228.58631256336002], 'bbox': [336.13500976, 226.41341260336003, 347.726996466608, 230.75921252336002], 'path': [['m', [384.6659851, 247.8109746]], ['c', [382.79998505, 248.50097460000006], [380.62498522, 248.82197463], [378.2609849, 248.82197463]], ['c', [371.64598465, 248.82397461000005], [363.56498527, 246.30497456], [356.6639843, 242.46397448000005]], ['c', [349.76098633, 238.62597418999997], [344.042984, 233.4569745], [342.11598587000003, 228.22997523000004]], ['l', [341.74598694, 228.36697435999997]], ['c', [343.73898697, 233.74597502000006], [349.5219841, 238.93797445999996], [356.47198486, 242.80897475000006]], ['c', [363.42398453, 246.67697466000004], [371.54798507, 249.21597456999996], [378.2609849, 249.21697460999997]], ['c', [380.65998506, 249.21697460999997], [382.87998509, 248.89197457], [384.8029851, 248.1819746]], ['l', [384.6659851, 247.8109746]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Neurod6', 'point': [355.255089154796, 233.17708885255996], 'bbox': [347.3269043, 231.11008894255997, 363.183274009592, 235.24408876255995], 'path': [['m', [384.6749878, 247.80798388999995]], ['c', [383.58998777, 248.14998388999993], [382.36598779, 248.3109839099999], [381.04898787999997, 248.3109839099999]], ['c', [376.45298768, 248.31298388999994], [370.73598768, 246.34498383999994], [365.88198854, 243.41198398999995]], ['c', [361.02498819, 240.48198367999998], [357.03598787, 236.5799842099999], [355.84898759, 232.79198408999991]], ['l', [355.47198869, 232.91098358]], ['c', [356.71998788, 236.85798408999995], [360.77398874, 240.78498413], [365.67798806, 243.7499838099999]], ['c', [370.5839882, 246.71098388999997], [376.34298803, 248.70398390999992], [381.04898787999997, 248.70598388999997]], ['c', [382.39798785, 248.70598388999997], [383.66198779, 248.54198391], [384.7939878, 248.1849838899999]], ['l', [384.6749878, 247.80798388999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ppp1r17', 'point': [338.6215513488, 268.92810942336007], 'bbox': [330.4079895, 266.7552094633601, 346.8351131976, 271.10100938336], 'path': [['m', [384.5499878, 247.92700244999992]], ['c', [382.46198775, 253.54200221999997], [376.30998803, 259.0780024699999], [368.88598824999997, 263.18700266999997]], ['c', [361.46598817, 267.2990021899999], [352.7789879, 269.9910016199999], [345.6579895, 269.9900021699999]], ['c', [343.08498765, 269.9900021699999], [340.71698762, 269.63900234999994], [338.68798829, 268.8820023699999]], ['l', [338.5499878, 269.2520032099999]], ['c', [340.63598634, 270.0300030899999], [343.04898835, 270.3850026299999], [345.6579895, 270.3850026299999]], ['c', [352.87598802, 270.3830018199999], [361.6049881, 267.6700024799999], [369.07798768, 263.53300238999987]], ['c', [376.54798795, 259.3910021899999], [382.76498784, 253.83200264999994], [384.9199878, 248.06500244999995]], ['l', [384.5499878, 247.92700244999992]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Klc3', 'point': [313.35505509218797, 245.62421539336003], 'bbox': [309.1281128, 243.45131543336004, 317.581997384376, 247.79711535336003], 'path': [['m', [384.6000061, 247.85199023999996]], ['c', [377.81400633, 254.16299011], [364.97900582, 257.2509904], [351.80200577, 257.24699068999996]], ['c', [336.5300064, 257.25198985], [320.80100631999994, 253.1039900899999], [313.49700928, 245.22599031999994]], ['l', [313.20700836, 245.49499013000002]], ['c', [320.66500472999996, 253.51199007999992], [336.46000671999997, 257.6369900799999], [351.80200577, 257.6419902]], ['c', [365.03900527999997, 257.63799049], [377.94200609999996, 254.56699039], [384.8690061, 248.14099023999995]], ['l', [384.6000061, 247.85199023999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ppp1r14a', 'point': [324.71179432656004, 252.19410064336], 'bbox': [315.2901001, 250.02120068336, 334.13348855312, 254.36700060336], 'path': [['m', [384.58499145, 247.86700488999998]], ['c', [378.29899168000003, 255.14800502000003], [363.83999062000004, 259.16400481999995], [350.22599030000003, 259.15800525]], ['c', [340.01898955, 259.1610045599999], [330.28799057000003, 256.90500498999995], [324.83999252, 252.18400480000003]], ['l', [324.58098984000003, 252.48300505999998]], ['c', [330.16299057000003, 257.3060050099999], [339.96199035, 259.5500045], [350.22599030000003, 259.55300476]], ['c', [363.91599083, 259.54700518999994], [378.43899155, 255.55800487], [384.88399145, 248.12600489]], ['l', [384.58499145, 247.86700488999998]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Klhl25', 'point': [331.8726626188, 232.49829986336], 'bbox': [325.83200073, 230.32539990336, 337.9133245076, 234.67119982335998], 'path': [['m', [384.64099121, 247.82299853999996]], ['c', [381.36599111, 249.59399855999993], [377.02899123, 250.39999844], [372.22899150999996, 250.39999844]], ['c', [364.69799041, 250.39999844], [356.03499030999996, 248.41699852], [348.56399154999997, 245.13799859999995]], ['c', [341.09199143, 241.86099863999993], [334.81698989999995, 237.27999832], [332.04199219, 232.13899852999998]], ['l', [331.69499207, 232.32699825999998]], ['c', [334.54698944, 237.5959983099999], [340.88599015, 242.19699861999993], [348.40499115, 245.4999985899999]], ['c', [355.92499161, 248.79999853999993], [364.62899208, 250.79399848999992], [372.22899150999996, 250.79499841999996]], ['c', [377.07199145, 250.79399848999992], [381.46899127999995, 249.98499847999994], [384.82899120999997, 248.16999853999994]], ['l', [384.64099121, 247.82299853999996]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Necab2', 'point': [318.41408894212, 267.16614654336], 'bbox': [310.92410278, 264.99324658336, 325.90407510424, 269.33904650336], 'path': [['m', [384.56201171, 247.90002489999995]], ['c', [381.04601168, 254.22502469999995], [373.26701163999996, 259.80202532], [364.01501083, 263.7800250099999]], ['c', [354.76401138, 267.76002551], [344.04501340999997, 270.1470246399999], [334.66701126, 270.1470246399999]], ['c', [328.40001297, 270.1470246399999], [322.73401259999997, 269.0790257499999], [318.50901031, 266.7280240099999]], ['l', [318.31700896999996, 267.07302523]], ['c', [322.63001251, 269.47102404], [328.35601042999997, 270.54102563999993], [334.66701126, 270.54202509999993]], ['c', [344.11101149999996, 270.54102563999993], [354.87101172999996, 268.1430249299999], [364.17101097, 264.1430249299999]], ['c', [373.47001171, 260.14002467], [381.31401180999995, 254.54302501999996], [384.90701171, 248.09302489999993]], ['l', [384.56201171, 247.90002489999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Dscaml1', 'point': [357.05660059441595, 256.80414702336], 'bbox': [348.60488891, 254.63124706335998, 365.50831227883197, 258.97704698335997], 'path': [['m', [384.55999755, 247.90399218999994]], ['c', [383.13399755, 250.61099219999994], [379.83199738999997, 253.06199218999996], [375.88999748, 254.81699228999992]], ['c', [371.94899749, 256.57499170999995], [367.37299728, 257.64499235999995], [363.42199707, 257.64399196]], ['c', [360.99599838, 257.64399196], [358.80799675, 257.24099208999996], [357.1479969, 256.36299181999993]], ['l', [356.96399687999997, 256.71199179999996]], ['c', [358.70599745, 257.63199187], [360.95299719999997, 258.0379920099999], [363.42199707, 258.03899241]], ['c', [367.44299698, 258.0379920099999], [372.05999755, 256.95699263999995], [376.04999733, 255.17799234999995]], ['c', [380.03899765, 253.3969922199999], [383.40699754999997, 250.92499208999993], [384.90899755, 248.08899218999989]], ['l', [384.55999755, 247.90399218999994]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Cacna1f', 'point': [326.950596540992, 275.43103912335994], 'bbox': [318.85610961, 273.25813916335994, 335.045083471984, 277.60393908335993], 'path': [['m', [384.54901123, 247.92901659999995]], ['c', [381.96901131000004, 255.09601640999995], [374.19801140000004, 262.2010169099999], [364.81201172000004, 267.49001741999996]], ['c', [355.42801095000004, 272.78301669999996], [344.43400956000005, 276.26701593999996], [335.4480095, 276.2660165]], ['c', [332.34601212, 276.2660165], [329.48501206000003, 275.85101747999994], [327.01301193, 274.95901727999996]], ['l', [326.87901307000004, 275.33101701999993]], ['c', [329.40600968, 276.24201631999995], [332.31101227, 276.6600175], [335.4480095, 276.66101695]], ['c', [344.53401185, 276.65901613999995], [355.57001113, 273.15401697999994], [365.00601197000003, 267.83401728]], ['c', [374.43801117000004, 262.51201677999995], [382.27501131, 255.38401651999993], [384.92001123, 248.06401659999995]], ['l', [384.54901123, 247.93001659999993]], ['l', [384.54901123, 247.92901659999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Kcne1l', 'point': [328.69355004772, 263.50513580335996], 'bbox': [322.05099487, 261.33223584335997, 335.33610522544, 265.67803576335996], 'path': [['m', [384.56298828, 247.8980107499999]], ['c', [381.58098817, 253.1150107499999], [375.09998798000004, 257.68301058999987], [367.39598847, 260.93001032999996]], ['c', [359.69298745000003, 264.17801142999986], [350.76998902, 266.11201144999995], [342.91898728, 266.1110100899999]], ['c', [337.41698838, 266.11201144999995], [332.44298935, 265.16000987999996], [328.78899002, 263.0680108199999]], ['l', [328.59298706000004, 263.4110102799999]], ['c', [332.33698655, 265.55301142999986], [337.37198638, 266.5060105499999], [342.91898728, 266.5060105499999]], ['c', [350.83398818, 266.5060105499999], [359.79698754000003, 264.5620102999999], [367.54898835, 261.2940106499999]], ['c', [375.29998875, 258.0230107499999], [381.84498835, 253.43501091999997], [384.90598828000003, 248.09501074999991]], ['l', [384.56298828, 247.8980107499999]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Mfap4', 'point': [348.57326316882404, 277.13901642336], 'bbox': [342.53390502, 274.96611646336, 354.612621317648, 279.31191638335997], 'path': [['m', [384.53799438, 247.97601367999994]], ['c', [383.88999437999996, 254.18201352999995], [378.54999447, 261.50101328999995], [371.71799468, 267.23701335]], ['c', [364.89199448, 272.97901392999995], [356.57699393999997, 277.14101458999994], [350.06999587999996, 277.13601351999995]], ['c', [349.56599425999997, 277.13601351999995], [349.07199477999995, 277.11101389999993], [348.59099577999996, 277.06001330999993]], ['l', [348.54999541999996, 277.4530129599999]], ['c', [349.04499434999997, 277.50501299999996], [349.55299377, 277.53101397999995], [350.06999587999996, 277.53101397999995]], ['c', [356.73999405, 277.52601290999996], [365.09299468, 273.3170132799999], [371.97199439999997, 267.5400128499999]], ['c', [378.84599447, 261.75701381], [384.2499944, 254.40801383999997], [384.93099437999996, 248.01701368]], ['l', [384.53799438, 247.97601367999994]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Kcnh6', 'point': [315.9832609866, 234.44025542336], 'bbox': [309.8230896, 232.26735546336, 322.14343237319997, 236.61315538335998], 'path': [['m', [384.625, 247.83197069999994]], ['c', [379.5079999, 251.23897075999992], [371.94699955, 252.78997087999994], [363.64100075, 252.78797054999995]], ['c', [345.40999985, 252.7979707799999], [323.6080017, 245.3139708199999], [316.1439972, 234.0659708999999]], ['l', [315.81500245, 234.28497076999997]], ['c', [323.44699860000003, 245.72697068999992], [345.3069992, 253.1729707799999], [363.64100075, 253.18297052999992]], ['c', [371.99499989, 253.1819706], [379.62200022, 251.63297080999996], [384.844, 248.1609706999999]], ['l', [384.625, 247.83197069999994]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Btbd17', 'point': [355.26389850322, 268.97010162336005], 'bbox': [348.37689208, 266.79720166336006, 362.15090492644, 271.14300158336], 'path': [['m', [384.54000854, 247.96197558999995]], ['c', [383.74100852, 252.56897545999993], [379.46600866, 257.77397585999995], [374.10800838, 261.7969756199999]], ['c', [368.75400829, 265.8239751], [362.32200813, 268.67597629], [357.28400803, 268.67197466999994]], ['c', [356.59200859, 268.67197466999994], [355.92700768, 268.61897517], [355.29400826, 268.50797510999996]], ['l', [355.22600937, 268.8969750499999]], ['c', [355.88400841, 269.01197482], [356.57200813, 269.06697512999995], [357.28400803, 269.06697512999995]], ['c', [362.46600914, 269.06297541999993], [368.93900871, 266.17297603], [374.34500884, 262.11197519999996]], ['c', [379.74700831999996, 258.0469756199999], [384.08700854, 252.81497574999992], [384.92900854, 248.02997558999994]], ['l', [384.54000854, 247.96197558999995]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Ebf2', 'point': [342.18483272487, 217.1121474525599], 'bbox': [337.93301391, 215.0451475425599, 346.43665153973996, 219.1791473625599], 'path': [['m', [384.70498657, 247.80102588]], ['c', [383.90998656, 247.92102588], [383.08198655, 247.97902588], [382.22698665, 247.97902588]], ['c', [374.84298611, 247.98302588], [365.42998696, 243.62802603], [357.62598610000003, 237.53002598]], ['c', [349.81798554, 231.43702556000005], [343.62198639, 223.59702541000001], [342.59298706000004, 216.73302507999995]], ['l', [342.20198822000003, 216.79202511000005]], ['c', [343.27098847, 223.83602572999996], [349.53098679000004, 231.70402576000004], [357.38298607, 237.84102585000005]], ['c', [365.23998642000004, 243.97302581999998], [374.69198704, 248.37002587999996], [382.22698665, 248.37402588999998]], ['c', [383.09998656, 248.37402588999998], [383.94798655, 248.31502588], [384.76398657000004, 248.19202587999996]], ['l', [384.70498657, 247.80102588]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Khdrbs3', 'point': [314.19896563104, 256.75110748336], 'bbox': [306.10708618, 254.57820752336, 322.29084508208, 258.92400744336], 'path': [['m', [384.57901, 247.87402393000002]], ['c', [377.0040102, 257.54302359999997], [357.98900985, 263.23302413], [340.85300827, 263.22602416000007]], ['c', [330.21900939, 263.22802403000003], [320.31101226, 261.03802348], [314.32000732, 256.32902385]], ['l', [314.07601165, 256.64002373000005]], ['c', [320.19001006999997, 261.43702365], [330.1660118, 263.61802436000005], [340.85300827, 263.62102367]], ['c', [358.07401083999997, 263.61302424999997], [377.15101003, 257.95502425000006], [384.89000998, 248.11802393000005]], ['l', [384.57901, 247.87402393000002]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}, {'visual_slot': 'Mfng', 'point': [337.58536149108, 279.76703766336], 'bbox': [332.75500488, 277.59413770336, 342.41571810216, 281.93993762336], 'path': [['m', [384.5409851, 247.95898487]], ['c', [383.1479851, 255.21098471999994], [376.32698536, 263.25498439], [367.8289852, 269.45598461]], ['c', [359.33598518, 275.66198398], [349.16998672, 280.03198672999997], [341.15998459, 280.02798511]], ['c', [339.92298507, 280.02798511], [338.73898697, 279.92398502], [337.61998367, 279.70898487]], ['l', [337.54498672, 280.09698535]], ['c', [338.69198608, 280.31798602], [339.90098572, 280.42298557], [341.15998459, 280.42298557]], ['c', [349.30898666, 280.4199834], [359.51598549, 276.01398517], [368.061985, 269.77498485]], ['c', [376.6039853, 263.53198482000005], [383.4899851, 255.46098471999994], [384.92898511, 248.03398487000004]], ['l', [384.5409851, 247.95898487]]], 'rgb': [0.17647, 0.44314, 0.51765], 'interpolated': False}], 'right': [{'visual_slot': 'Cnih3', 'point': [431.303848375488, 281.52503326335994], 'bbox': [425.62649536, 279.35213330335995, 436.98120139097597, 283.69793322335994], 'path': [['m', [384.7669983, 248.1909794999999]], ['c', [385.73099828, 248.02997949999997], [386.74099827000003, 247.95297949999986], [387.7879982, 247.95297949999986]], ['c', [395.82999802, 247.94897949999995], [406.06399918, 252.55297947999986], [414.5739975, 259.03397990999986]], ['c', [423.0889969, 265.50997973999995], [429.87899781, 273.8679795399999], [431.10899735, 281.2719807799999]], ['l', [431.49899674, 281.2069783399999]], ['c', [430.22599793, 273.6239800599999], [423.3729973, 265.23897982999995], [414.81399918, 258.7199797799999]], ['c', [406.2499981, 252.20497942999987], [395.97699833, 247.56097949999992], [387.7879982, 247.5569794999999]], ['c', [386.7209983, 247.5569794999999], [385.68999827, 247.63697947999992], [384.7019983, 247.8019794999999]], ['l', [384.7669983, 248.1909794999999]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Lypd1', 'point': [452.10173249480397, 232.17214500255994], 'bbox': [446.47039794, 230.10514509255995, 457.733067049608, 234.23914491255994], 'path': [['m', [384.90200805, 248.10101367999994]], ['c', [388.55800795, 242.22601367999994], [396.04500769000003, 237.22001411999997], [404.92800711, 233.6990132599999]], ['c', [413.80900763, 230.17501307999999], [424.08000945000003, 228.13001300999997], [433.29100798, 228.13001300999997]], ['c', [440.75800705, 228.12901355999998], [447.52600859, 229.47601367999994], [452.28601073, 232.44301367999992]], ['l', [452.49501036000004, 232.10801364999998]], ['c', [447.6360092, 229.08301403999997], [440.80500792, 227.73601390999988], [433.29100798, 227.7350144599999]], ['c', [424.02200698, 227.73601390999988], [413.71100806, 229.79001285999993], [404.78200721, 233.33101322999994]], ['c', [395.85300826, 236.87501384999996], [388.30200815, 241.90101387999994], [384.56700805, 247.89201368]], ['l', [384.90200805, 248.10101367999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'H3f3a', 'point': [415.66477391326396, 235.01331451336], 'bbox': [409.86660766, 232.84041455336, 421.46294016652797, 237.18621447336], 'path': [['m', [384.91799926, 248.07000782]], ['c', [386.35499928999997, 244.46000791999995], [390.40399931999997, 240.93900778], [395.28099917, 238.34400799000002]], ['c', [400.15499972, 235.74600747], [405.85299871999996, 234.06600810999998], [410.55399893, 234.06700757]], ['c', [412.43099974, 234.06700757], [414.1479988, 234.33500817000004], [415.58799933, 234.91100742000003]], ['l', [415.73499869, 234.54400780000003]], ['c', [414.23299979, 233.94400742000005], [412.46799849, 233.67300751000005], [410.55399893, 233.67200806000005]], ['c', [405.76099967, 233.67400792], [400.02099893999997, 235.37300828000002], [395.0949993, 237.99600747]], ['c', [390.17099903999997, 240.62200786999995], [386.05699931, 244.16500785000005], [384.55099924999996, 247.92300781999995]], ['l', [384.91799926, 248.07000782]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Vav3', 'point': [440.62342571883994, 260.32312896336], 'bbox': [435.67138671, 258.15022900336, 445.57546472767996, 262.49602892336], 'path': [['m', [384.8420105, 248.16302539999992]], ['c', [388.88401031, 245.55302549999988], [394.7550106, 244.36402537999993], [401.22200966, 244.36602546999995]], ['c', [408.72200966, 244.36502529999996], [417.02201080000003, 245.9600253399999], [424.2000122, 248.7360254099999]], ['c', [431.37900924, 251.5110254399999], [437.43201065, 255.47402526999997], [440.45401001, 260.16602564999994]], ['l', [440.78601074, 259.9510254999999]], ['c', [437.68500900000004, 255.1470255999999], [431.56700897, 251.16402531999995], [424.3420105, 248.36802539999996]], ['c', [417.11600876, 245.57302548999996], [408.77801132, 243.97102548999987], [401.22200966, 243.97102548999987]], ['c', [394.70801067, 243.9720254199999], [388.77201056, 245.15902545999995], [384.6280105, 247.83002539999995]], ['l', [384.8420105, 248.16302539999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Bhlhe22', 'point': [434.266184673216, 227.52430084336004], 'bbox': [426.29251098, 225.35140088336004, 442.239858366432, 229.69720080336003], 'path': [['m', [384.91699218, 248.0709843799999]], ['c', [387.26799225, 242.3239841699999], [393.7309923, 236.78998424999986], [401.49799155000005, 232.71598483999992]], ['c', [409.26199149, 228.63898517999996], [418.32599258000005, 226.01298381999993], [425.84599303000005, 226.01398517999996]], ['c', [428.96099089, 226.01398517999996], [431.80899048000003, 226.4649834799999], [434.18899153, 227.43998386999988]], ['l', [434.33899306, 227.07398463999994]], ['c', [431.89599228000003, 226.07398463999994], [428.99799346000003, 225.61998416999995], [425.84599303000005, 225.61898471999996]], ['c', [418.23599242, 225.62098361999995], [409.12999152000003, 228.26398517999996], [401.31399153, 232.36598445999994]], ['c', [393.49999236, 236.47098399999993], [386.97199224, 242.02698423999993], [384.55199218, 247.9219843799999]], ['l', [384.91699218, 248.0709843799999]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Runx1t1', 'point': [409.813794761016, 258.22223785336007], 'bbox': [401.72061157, 256.0493378933601, 417.90697795203204, 260.39513781336007], 'path': [['m', [384.81298828, 248.17797899999994]], ['c', [386.05998825, 247.63897902999997], [387.58398818, 247.38897902999997], [389.25998830000003, 247.38997901999994]], ['c', [393.00698852, 247.38797897999996], [397.50498866000004, 248.63897899999995], [401.36098861, 250.60197901999993]], ['c', [405.21898841, 252.56197881999992], [408.4289875, 255.24197911999988], [409.62998771, 258.03297853999993]], ['l', [409.99198913000004, 257.87597895]], ['c', [408.72098922000004, 254.94097900999998], [405.44498824000004, 252.2399787999999], [401.53998756, 250.24997901999996]], ['c', [397.63398838, 248.26297899999997], [393.09298801, 246.9959789599999], [389.25998830000003, 246.99397898999996]], ['c', [387.54498839, 246.99497903999998], [385.96998822, 247.24897897999995], [384.65598828000003, 247.81497899999988]], ['l', [384.81298828, 248.17797899999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Sema3a', 'point': [409.52446611104006, 275.58503082335994], 'bbox': [401.43258666, 273.41213086335995, 417.61634556208, 277.75793078335994], 'path': [['m', [384.7250061, 248.19397021999998]], ['c', [389.75300596999995, 248.42897022], [395.93800639999995, 252.48497009999994], [400.83200645, 257.79397059999997]], ['c', [405.73300551999995, 263.09697009], [409.34500692999995, 269.65097095], [409.33800696, 274.77796985]], ['c', [409.33800696, 274.94997072999996], [409.33400535, 275.12097026], [409.32600592999995, 275.28896953]], ['l', [409.72100637999995, 275.30797054]], ['c', [409.72900581, 275.13296938999997], [409.73400687, 274.95596935], [409.73400687, 274.77796985]], ['c', [409.7270069, 269.47097064], [406.06100654, 262.88196994], [401.12300681, 257.52596999]], ['c', [396.17800616, 252.17597032999993], [389.96200608, 248.05797022000002], [384.7440061, 247.79897021]], ['l', [384.7250061, 248.19397021999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Hgf', 'point': [435.41528772774404, 215.20630767336002], 'bbox': [432.03338623, 213.03340771336002, 438.79718922548807, 217.37920763336], 'path': [['m', [384.92800903, 248.03802539999992]], ['c', [386.55800903, 240.47002556999996], [393.83500861, 232.21202518999996], [402.8480091, 225.87502529999995]], ['c', [411.85600853, 219.5330262399999], [422.59600829, 215.10602619999997], [431.08900833, 215.10902454999996]], ['c', [432.59400939, 215.10902454999996], [434.02901076, 215.24702503999993], [435.37100983, 215.53802539999992]], ['l', [435.45401, 215.15202381999995]], ['c', [434.08100890000003, 214.85502674999987], [432.61800765, 214.71402408999995], [431.08900833, 214.71402408999995]], ['c', [422.46300887, 214.7170262599999], [411.68200874, 219.17802478999988], [402.62000846, 225.5520253399999]], ['c', [393.56300925, 231.9300255999999], [386.22200905, 240.21402551999995], [384.54200903, 247.95502539999995]], ['l', [384.92800903, 248.03802539999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Gpr12', 'point': [423.68097555884003, 212.90430572336004], 'bbox': [417.64248657, 210.73140576336004, 429.71946454768, 215.07720568336003], 'path': [['m', [384.93200683, 248.00701954]], ['c', [385.30900683, 240.84601977], [391.10500668, 232.11301948999994], [398.66500663, 225.20201924000003]], ['c', [406.22000692999995, 218.28601887000002], [415.53700637, 213.19101764000004], [422.85400770999996, 213.19701816999998]], ['c', [423.12900543, 213.19701816999998], [423.40000533, 213.20401814000002], [423.66900633999995, 213.21902134000004]], ['l', [423.6910057, 212.82402089000004]], ['c', [423.41400526999996, 212.80901768], [423.13500594, 212.80201770999997], [422.85400770999996, 212.80201770999997]], ['c', [415.35900687, 212.80801823000002], [406.00600622999997, 217.95601894000004], [398.39900682999996, 224.91101887000002]], ['c', [390.7970066, 231.87101986999994], [384.93900683, 240.63601973000004], [384.53700682, 247.98601954000003]], ['l', [384.93200683, 248.00701954]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Prr15', 'point': [444.7351342022, 249.10224272336006], 'bbox': [439.4223938, 246.92934276336007, 450.0478746044, 251.27514268336006], 'path': [['m', [384.87200927, 248.13800098000002]], ['c', [390.71300935, 242.44900086000007], [402.22200964999996, 239.60300114000006], [413.85200881, 239.60800125000003]], ['c', [426.15200804, 239.60300114000006], [438.58501051999997, 242.78600074999997], [444.59201049, 248.97500099]], ['l', [444.87500761999996, 248.70000095]], ['c', [438.71900939, 242.38000108000006], [426.21900939, 239.21700146], [413.85200881, 239.21300078000002]], ['c', [402.15900991999996, 239.21700146], [390.5820093, 242.04400111999996], [384.59700927, 247.85500098]], ['l', [384.87200927, 248.13800098000002]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Galntl6', 'point': [454.981455833256, 256.27924469336006], 'bbox': [448.09640502, 254.10634473336006, 461.866506646512, 258.45214465336005], 'path': [['m', [384.8580017, 248.15100146999998]], ['c', [390.88200186, 243.35300162], [400.94300078, 241.11600163000003], [411.7140026, 241.11900141]], ['c', [428.63400267, 241.11100150999994], [447.29700087, 246.63800145999994], [454.82199858, 256.11600162]], ['l', [455.13100432, 255.87000133000004]], ['c', [447.44300078, 246.22600151999995], [428.71700285, 240.73100139999997], [411.7140026, 240.7230015]], ['c', [400.88900183, 240.72600126999998], [390.76000165, 242.95500137], [384.6120017, 247.84200149000003]], ['l', [384.8580017, 248.15100146999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Acsl1', 'point': [450.3621483522, 267.26709869336], 'bbox': [445.04940795, 265.09419873336003, 455.6748887544, 269.43999865336], 'path': [['m', [384.82998657, 248.16900683999995]], ['c', [388.97398662, 245.88600683999994], [394.50498677, 244.84800673999996], [400.62298679, 244.84900690999996]], ['c', [409.92998695, 244.84800673999996], [420.58998489000004, 247.24500681999996], [429.7869873, 251.23000693999995]], ['c', [438.98598479000003, 255.2120070599999], [446.71798705000003, 260.78900672], [450.18898772, 267.0950074399999]], ['l', [450.53498840000003, 266.9050069]], ['c', [446.98698807, 260.47200728999996], [439.18998718, 254.87500668999996], [429.94398498000004, 250.86700679]], ['c', [420.69698714000003, 246.8620067999999], [409.99698638, 244.45400693], [400.62298679, 244.45400693]], ['c', [394.46098613000004, 244.45400693], [388.86998653, 245.49400688999992], [384.63898657000004, 247.82300685999996]], ['l', [384.82998657, 248.16900683999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Rnf122', 'point': [405.27320903102, 240.44025542336], 'bbox': [398.26669311, 238.26735546336, 412.27972495204, 242.61315538335998], 'path': [['m', [384.91500854, 248.07702685999993]], ['c', [385.90400855999997, 245.84802675999993], [388.50200844, 243.7160267999999], [391.62200830999996, 242.16302682999992]], ['c', [394.74100875, 240.60702705999995], [398.37800883999995, 239.62302638999995], [401.41900825, 239.6240267999999]], ['c', [402.84800909999996, 239.6240267999999], [404.14400864, 239.8410267999999], [405.19000815, 240.3080268099999]], ['l', [405.35100936, 239.94702674999996]], ['c', [404.23500823999996, 239.45002700999999], [402.88800811, 239.22902728999998], [401.41900825, 239.22902728999998]], ['c', [398.29500865, 239.23002673999997], [394.61500834, 240.23002673999997], [391.44600866999997, 241.81002665999995]], ['c', [388.27900861999996, 243.39202691999992], [385.61600851, 245.54502679999996], [384.55400854, 247.91602686]], ['l', [384.91500854, 248.07702685999993]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Gm2516', 'point': [427.71574959658005, 258.04615142336], 'bbox': [419.3815918, 255.87325146336002, 436.04990739316, 260.21905138336], 'path': [['m', [384.83999633, 248.16400195999995]], ['c', [387.87399624, 246.25100197999996], [392.22999619, 245.37700201999996], [397.03699587, 245.37800195]], ['c', [402.85299681000004, 245.37800195], [409.32699584, 246.65200199999992], [414.92299651, 248.85500193999997]], ['c', [420.51999663000004, 251.05600190999996], [425.23399733, 254.19100190999995], [427.54499816000003, 257.8650021699999]], ['l', [427.87899779, 257.65500212999996]], ['c', [425.48899459, 253.8650021699999], [420.71099471, 250.71000196], [415.06799697, 248.4880019599999]], ['c', [409.42399596, 246.26600195999993], [402.90999603, 244.9830019599999], [397.03699587, 244.9830019599999]], ['c', [392.18299626000004, 244.98400189999995], [387.76399635, 245.85700204], [384.62899633, 247.8290019599999]], ['l', [384.83999633, 248.16400195999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Syn3', 'point': [419.74662518883997, 281.57102325335995], 'bbox': [414.79458618, 279.39812329335996, 424.69866419768, 283.74392321335995], 'path': [['m', [384.7380066, 248.19397021999998]], ['l', [384.9970066, 248.19197022000003]], ['c', [391.77700663, 248.18497021999997], [400.38200665, 253.17097045000003], [407.29200745, 259.87496996999994]], ['c', [414.20900727000003, 266.57197047], [419.42800522, 274.98597003], [419.5470047, 281.71897174]], ['l', [419.94200516, 281.71197177]], ['c', [419.80500794, 274.78596927], [414.5240059, 266.33997012], [407.56700707, 259.59097052999994]], ['c', [400.60500622, 252.84797001999993], [391.96500683, 247.80297022000002], [384.9970066, 247.79597021999996]], ['l', [384.7310066, 247.79897021]], ['l', [384.7380066, 248.19397021999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Myt1l', 'point': [441.36555751443996, 237.00721100336], 'bbox': [436.17449951, 234.83431104336, 446.55661551887994, 239.18011096336], 'path': [['m', [384.89801025, 248.10699511999997]], ['c', [391.0340104, 239.04499484999997], [408.66101073000004, 233.09199477999994], [423.53001021, 233.10199501]], ['c', [430.56201171000004, 233.09999515999993], [436.97301101000005, 234.42899466999995], [441.25201035000003, 237.33699466999997]], ['l', [441.47401046000004, 237.009995]], ['c', [437.08800887, 234.0339951699999], [430.6110115, 232.70799495999995], [423.53001021, 232.70699550999996]], ['c', [408.5600109, 232.71599531999993], [390.87401008, 238.63099529], [384.57101025000003, 247.88499511999999]], ['l', [384.89801025, 248.10699511999997]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Sptb', 'point': [405.52768864773196, 251.24720122335998], 'bbox': [401.0586853, 249.07430126335998, 409.99669199546395, 253.42010118335998], 'path': [['m', [384.8500061, 248.15698292000002]], ['c', [386.46300613, 246.98998287000006], [389.01300619999995, 246.44098296000004], [391.79800604999997, 246.44298293999998]], ['c', [397.06900595999997, 246.43398288000003], [403.16800689999997, 248.40998292000006], [405.36400603, 251.47998287000007]], ['l', [405.68500517999996, 251.24898292]], ['c', [403.32200622, 247.99698292000005], [397.16300581999997, 246.05598284999996], [391.79800604999997, 246.04798296]], ['c', [388.96300600999996, 246.04898288000004], [386.34600615, 246.59398296000006], [384.6190061, 247.83598292]], ['l', [384.8500061, 248.15698292000002]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Chrm3', 'point': [432.674079539984, 247.93421295335997], 'bbox': [426.1545105, 245.76131299335998, 439.193648579968, 250.10711291335997], 'path': [['m', [384.875, 248.13501025999994]], ['c', [389.62799978, 243.3010101499999], [399.375, 240.82301045999998], [409.06900024, 240.8270101799999]], ['c', [418.48500061, 240.82301045999998], [427.84799957, 243.16101028999992], [432.53400039999997, 247.78801025999996]], ['l', [432.81200027, 247.50701024]], ['c', [427.97999954, 242.75501014999998], [418.54899978000003, 240.4370103099999], [409.06900024, 240.43201019999992]], ['c', [399.30900001, 240.4370103099999], [389.49599981, 242.89401008999994], [384.59400002, 247.8580102499999]], ['l', [384.875, 248.13501025999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Pou6f2', 'point': [418.55852739324, 246.85724762335997], 'bbox': [411.67260742, 244.68434766335997, 425.44444736648, 249.03014758335996], 'path': [['m', [384.88000488, 248.13000537999994]], ['c', [388.31400489, 244.38000537999994], [395.85000514, 242.3670053699999], [403.13300513, 242.37200547999998]], ['c', [409.28800582, 242.36900522999997], [415.26000404, 243.80500557999994], [418.41900634, 246.71700539999995]], ['l', [418.68700408, 246.42600538]], ['c', [415.3880043, 243.40200521999998], [409.34800529, 241.98100521999993], [403.13300513, 241.9770055099999]], ['c', [395.77900504, 241.98300554999992], [388.17700481, 243.97200537999993], [384.58900488, 247.8630053999999]], ['l', [384.88000488, 248.13000537999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Fam65b', 'point': [427.05886894326, 238.62024810336004], 'bbox': [419.08758544, 236.44734814336005, 435.03015244652, 240.79314806336004], 'path': [['m', [384.90200805, 248.10199023999996]], ['c', [387.17600797, 244.47999026999992], [391.82100819, 241.38799001999996], [397.33700846, 239.2169900199999]], ['c', [402.85200881000003, 237.04399062999994], [409.23300742000004, 235.78599025999995], [414.96200751000003, 235.78599025999995]], ['c', [419.68700789, 235.78498986], [423.96700667, 236.64299060999997], [426.95100783, 238.52199032]], ['l', [427.16100692, 238.18799069]], ['c', [424.07800673, 236.24999000999992], [419.73400878, 235.39199021000002], [414.96200751000003, 235.39098979999994]], ['c', [409.17500876, 235.39199021000002], [402.75600815, 236.65798999999993], [397.19200801, 238.84898998999995]], ['c', [391.63100815, 241.04299023], [386.9210081, 244.1539902899999], [384.56700805, 247.89099023999995]], ['l', [384.90200805, 248.10199023999996]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'D130009I18Rik', 'point': [402.146551924224, 230.45234039336003], 'bbox': [387.16571044, 228.27944043336004, 417.12739340844803, 232.62524035336003], 'path': [['m', [384.93200683, 247.99401905000002]], ['l', [384.93200683, 247.91501905000007]], ['c', [384.92500683, 244.48801901000002], [387.49200677, 240.09401895000008], [390.94700668999997, 236.57901908999997]], ['c', [394.39500712999995, 233.05701878000002], [398.73000715999996, 230.41701939000006], [402.14200590999997, 230.3850198]], ['l', [402.13800619999995, 229.99001934]], ['c', [398.53300665999996, 230.03901913000004], [394.16500663, 232.7410188], [390.66500663, 236.30201867000005]], ['c', [387.17100690999996, 239.87001944999997], [384.54400682, 244.29801895000003], [384.53700682, 247.91501905000007]], ['l', [384.53700682, 247.99901905000002]], ['l', [384.93200683, 247.99401905000002]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Pou4f1', 'point': [442.052851574745, 276.09286034256], 'bbox': [435.50149536, 274.02586043255997, 448.60420778949, 278.15986025256], 'path': [['m', [384.7999878, 248.18298388999995]], ['c', [387.19898772, 247.34298391999994], [389.95698785999997, 246.95198395], [392.93998813999997, 246.95198395]], ['c', [401.94998741, 246.94998384999997], [413.00698853, 250.5229837999999], [422.43498612, 255.92498399]], ['c', [431.8659897, 261.32298423], [439.66398621, 268.55798388999995], [442.1979866, 275.80698442999994]], ['l', [442.56998825, 275.67598391999996]], ['c', [439.97198868, 268.27198459], [432.1089859, 261.01398421], [422.63198853, 255.58198404999996]], ['c', [413.15098763, 250.15298391], [402.0519886, 246.55798388999995], [392.93998813999997, 246.55698384999994]], ['c', [389.92198801, 246.55698384999994], [387.12198782, 246.95098388999997], [384.6689878, 247.8099838899999]], ['l', [384.7999878, 248.18298388999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Tnfrsf19', 'point': [420.20380911328, 269.61512115335995], 'bbox': [412.23361206, 267.44222119335996, 428.17400616656, 271.78802111335995], 'path': [['m', [384.78201293, 248.18798876999995]], ['c', [385.84701298, 247.92298878999998], [387.00501297, 247.79698876999998], [388.23101281, 247.79698876999998]], ['c', [394.04501246, 247.79398876999994], [401.37701224, 250.6419887699999], [407.55301283, 254.77498866999997]], ['c', [413.73301314, 258.90298892], [418.75401305, 264.32298899], [420.00601195, 269.37698793999994]], ['l', [420.38901137, 269.2809891799999]], ['c', [419.08401106, 264.0589890499999], [413.99801253, 258.61398840999993], [407.77301215, 254.44698858999993]], ['c', [401.54401206, 250.28298879], [394.17101286, 247.40498876999993], [388.23101281, 247.40198876]], ['c', [386.97801303, 247.40198876], [385.78901289, 247.5309887699999], [384.68701293, 247.80498878999992]], ['l', [384.78201293, 248.18798876999995]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Dcbld2', 'point': [405.62584035548, 264.29914215336], 'bbox': [398.86160278, 262.12624219336, 412.39007793096, 266.47204211336], 'path': [['m', [384.76000976, 248.19201708999992]], ['c', [385.09900974, 248.14801708999994], [385.45100974, 248.1260170899999], [385.8120098, 248.1260170899999]], ['c', [389.46600961, 248.1210170899999], [394.16700934, 250.36601710999992], [398.0450096, 253.4950170599999]], ['c', [401.92700957, 256.61901713], [404.98200988, 260.6300168099999], [405.42700957, 264.05701686]], ['l', [405.81801032, 264.0050168099999]], ['c', [405.33700942, 260.3960175599999], [402.21800994, 256.3560166399999], [398.29300976, 253.18701696999995]], ['c', [394.36200999, 250.02401709999992], [389.62200974, 247.73501710999994], [385.8120098, 247.7310170899999]], ['c', [385.43500977, 247.7310170899999], [385.06600976, 247.75401708999993], [384.70900976, 247.80101708999996]], ['l', [384.76000976, 248.19201708999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Runx1', 'point': [436.936950439864, 269.23595725256], 'bbox': [430.9624939, 267.16895734256, 442.911406979728, 271.30295716256], 'path': [['m', [384.81298828, 248.17797899999994]], ['c', [387.45498823, 247.0409790399999], [390.66298818, 246.51697898999998], [394.18198871000004, 246.5179790399999]], ['c', [402.07898902, 246.51697898999998], [411.53098869, 249.15297901999998], [419.63998794, 253.28597879999995]], ['c', [427.750988, 257.4169793199999], [434.51298904000004, 263.05297899999994], [437.06198882, 268.9849781999999]], ['l', [437.42498779, 268.82897805999994]], ['c', [434.80498885000003, 262.75297879999994], [427.97698974, 257.09497880999993], [419.81898879, 252.93397903999994]], ['c', [411.65898895000004, 248.77697896999996], [402.16498756, 246.12397909999993], [394.18198871000004, 246.12297893999994]], ['c', [390.62398815, 246.12297893999994], [387.36498832, 246.64997899999992], [384.65698828, 247.81497899999988]], ['l', [384.81298828, 248.17797899999994]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Prkce', 'point': [455.57032022776, 243.81525542336], 'bbox': [450.01638793, 241.64235546336, 461.12425252552, 245.98815538335998], 'path': [['m', [384.8829956, 248.12701464999998]], ['c', [392.26999568, 239.73701430999995], [408.93899535, 235.17601444000002], [424.76999663000004, 235.18201494999994]], ['c', [437.09799575, 235.17901469999993], [448.91399382000003, 237.94301464], [455.43799590000003, 243.70301486999995]], ['l', [455.6989975, 243.40701485]], ['c', [449.03999328000003, 237.54101420999996], [437.1559944, 234.79001474999995], [424.76999663000004, 234.78701449999994]], ['c', [408.86499595000004, 234.79301501999998], [392.12999581, 239.32701444999998], [384.5869956, 247.86601465]], ['l', [384.8829956, 248.12701464999998]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Tshz1', 'point': [422.11932123108, 225.07129791336], 'bbox': [416.20251464, 222.89839795336002, 428.03612782216, 227.24419787336], 'path': [['m', [384.92700195, 248.04101612]], ['c', [386.1940019, 242.60901595999997], [391.51800203, 236.73001624999995], [398.09100149999995, 232.23801566999998]], ['c', [404.66000175, 227.74201633999996], [412.47300147, 224.62601521], [418.65600205, 224.62901547]], ['c', [419.86500166999997, 224.62901547], [421.01100157999997, 224.74701549999997], [422.07300184999997, 224.99601604999998]], ['l', [422.16300200999996, 224.61101582000003]], ['c', [421.06700133, 224.35401585], [419.89100264999996, 224.23301556], [418.65600205, 224.23301556]], ['c', [412.34500120999996, 224.23601582000003], [404.49000166999997, 227.38501598999994], [397.86800193, 231.91201640999998]], ['c', [391.2500019, 236.44401600000003], [385.86100195999995, 242.34801629000003], [384.54200195, 247.95101611999996]], ['l', [384.92700195, 248.04101612]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Sgms1', 'point': [412.09867355552, 217.09345367336005], 'bbox': [405.33660888, 214.92055371336005, 418.86073823104005, 219.26635363336004], 'path': [['m', [384.93200683, 247.98498583999992]], ['c', [384.91800682999997, 247.75398583999993], [384.91200683, 247.5199858399999], [384.91200683, 247.28398586999992]], ['c', [384.90600682999997, 241.5469856399999], [388.87700676, 234.21798563999994], [394.2800064, 228.27098513999988]], ['c', [399.67700671, 222.31798602999993], [406.50600622999997, 217.7479863299999], [412.10800743, 217.42798663999997]], ['l', [412.08400724999996, 217.03298617999997]], ['c', [406.29400633999995, 217.3789849399999], [399.43300723, 222.01098490999993], [393.98700713, 228.00498629999993]], ['c', [388.54700683, 234.00498629999993], [384.52200682999995, 241.36998606999987], [384.51700682999996, 247.28398586999992]], ['c', [384.51700682999996, 247.52798585999994], [384.52300683, 247.7689858399999], [384.53700682, 248.00798583999995]], ['l', [384.93200683, 247.98498583999992]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}, {'visual_slot': 'Crtac1', 'point': [445.5786032222, 222.91431549336005], 'bbox': [439.17941284, 220.74141553336005, 451.9777936044, 225.08721545336005], 'path': [['m', [384.91699218, 248.07202197999993]], ['c', [387.81799220000005, 241.01102210999989], [395.75599192000004, 234.23402167999996], [405.29099272, 229.24402286999998]], ['c', [414.82299231, 224.2520222799999], [425.94699095000004, 221.03902103999997], [435.18699263, 221.04102184999988]], ['c', [439.03699111000003, 221.04002239999988], [442.55899048000003, 221.59802104999994], [445.50099181, 222.80902149999997]], ['l', [445.65199279, 222.44302226999991]], ['c', [442.64599228000003, 221.20702220999988], [439.07499312000004, 220.64602137999998], [435.18699263, 220.64502193999988]], ['c', [425.8569908, 220.6470227499999], [414.69099234000004, 223.8770222799999], [405.10699271000004, 228.89402248]], ['c', [395.52599238000005, 233.91402197999992], [387.52299212, 240.71402217999992], [384.55199218, 247.92102197999998]], ['l', [384.91699218, 248.07202197999993]]], 'rgb': [0.35294, 0.67843002, 0.49804], 'interpolated': False}]}, 'current_label_offsets': {'Foxo6': {'Cplx2': [0, 1.15], 'Mfap4': [0, -1.15], 'Pcdh10': [0, 1.35], 'Nhlh1': [0, -1.35]}, 'Zfpm2': {'Shox2': [1.5, 0], 'Nhsl1': [-1.5, 0], 'Ank2': [1.35, 0], 'Soga3': [-1.35, 0], 'Elavl3': [0, 0.8], 'Epha5': [0, -0.8], 'Rab3a': [0, 1.15], 'Map1b': [0, -1.15], 'Mapk8ip1': [0, 1.55], 'Tubb3': [0, -1.55], '3110039M20Rik': [0, 1.9], 'Afap1': [0, -1.9]}, 'Neurod2': {'Neurod6': [0, 1.35], 'Myo6': [0, -1.35], 'Kcnmb2': [1.55, 0], 'Mreg': [-1.55, 0]}, 'Nhlh1': {'1700037H04Rik': [0, 1.55], 'Ypel3': [0, -1.55], 'Nhlh2': [1.5, 0], 'Trp53i11': [-1.5, 0]}, 'Neurod1': {'Jun': [0, 0.241], 'Ntm': [-3.244, 0], 'Vcan': [0.973, 0], 'Edil3': [-0.762, 0.728], 'Cd81': [1.065, 3.971], 'Lrrn3': [-5.092, 0], 'Uchl1': [-1.284, 0], 'Acsl1': [-3.441, -1.537], 'Olfm1': [1.6, 0], 'Aplp1': [-1.6, 0], 'Chst2': [0.762, 0], 'Elavl2': [0, -3.499], 'Pcsk2': [4.538, 0.475], 'Ckmt1': [0, 1.106], 'Scn3a': [-0.973, 0.431], 'Unc5c': [1.284, -0.479], 'Snx18': [0, -0.475], 'Ddah1': [0, 0.804], 'Gnao1': [0, 0.516], 'Ywhah': [0, -0.037], 'Tmeff2': [-1.065, 0], 'Rbms1': [11.147, 0], 'Pou6f2': [0, -0.804], 'Ctnna2': [5.092, 0], 'Spock1': [0, -0.35], 'Hsp90aa1': [0, -1.09], 'C1ql3': [0, 0.874], 'Olfm2': [-1.814, 0], 'Rbms3': [0, -0.874], 'Cacng5': [1.814, 0], 'Sncg': [0, 2.5], 'Mllt11': [0, -2.5]}}}

def draw_module_network(ax, factor, left_genes, right_genes, kind='associated',
                        left_color='#2D7184', right_color='#5AAD7F', font_size=11.4,
                        highlight_genes=tuple()):
    """Substitute current module genes into the published Fig.3d vector layout.

    Original slot names are visual provenance only. Every current input member
    retains exactly one ribbon; template geometry carries no biological weight.
    Side-color parameters apply to ribbons only; outer panels are untouched.
    """
    #1. =========== Original vector template and appearance
    padding_fraction = .025
    halo_width_pt = .85
    highlight_color = '#BA001C'
    template = PAPER_NETWORK_TEMPLATE
    groups = [list(map(str, left_genes)), list(map(str, right_genes))]
    highlights = set(map(str, highlight_genes))
    if any(len(set(g)) != len(g) for g in groups):
        raise ValueError('A displayed module contains duplicate genes.')
    if font_size <= 0:
        raise ValueError('font_size must be positive.')
    ax.set_axis_off(); ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.set_facecolor('white')
    fig = ax.figure
    fig.canvas.draw()
    renderer = fig.canvas.get_renderer()
    area = ax.get_window_extent(renderer)
    point_scale = 72.0 / fig.dpi
    width, height = area.width * point_scale, area.height * point_scale
    bbox = np.asarray(template['bbox'], dtype=float)
    scale = min(width * (1 - 2 * padding_fraction) / (bbox[2] - bbox[0]),
                height * (1 - 2 * padding_fraction) / (bbox[3] - bbox[1]))
    used = (bbox[2:] - bbox[:2]) * scale
    offset = (np.array([width, height]) - used) / 2
    def xy(points):
        a = np.asarray(points, dtype=float)
        result = (a - bbox[:2]) * scale + offset
        result[..., 1] = height - result[..., 1]
        return result / np.array([width, height])
    def angle(slot, side):
        delta = np.asarray(slot['point']) - np.asarray(template['center'])
        return np.arctan2(delta[1], -delta[0] if side == 'left' else delta[0])
    def topology(slot):
        return tuple(step[0] for step in slot['path'])
    def blend(a, b):
        path = [[sa[0]] + [((np.asarray(pa) + np.asarray(pb)) / 2).tolist()
                for pa, pb in zip(sa[1:], sb[1:])]
                for sa, sb in zip(a['path'], b['path'])]
        return {'visual_slot': [a['visual_slot'], b['visual_slot']],
            'point': ((np.asarray(a['point']) + np.asarray(b['point'])) / 2).tolist(),
            'bbox': ((np.asarray(a['bbox']) + np.asarray(b['bbox'])) / 2).tolist(),
            'path': path, 'rgb': a['rgb'], 'interpolated': True}
    def choose_slots(side, count):
        slots = list(template['slots'][side])
        if count <= len(slots):
            if count == 0: return []
            points = np.asarray([slot['point'] for slot in slots])
            # Preserve the actual top, bottom and outer boundary before filling
            # the remaining slots from the original two-dimensional layout.
            outer = np.argmin(points[:, 0]) if side == 'left' else np.argmax(points[:, 0])
            chosen = list(dict.fromkeys([int(np.argmin(points[:, 1])),
                                         int(np.argmax(points[:, 1])), int(outer)]))[:count]
            while len(chosen) < count:
                distances = np.linalg.norm((points[:, None] - points[chosen][None])
                                           * np.array([.8, 1.0]), axis=2).min(axis=1)
                distances[chosen] = -1
                chosen.append(int(np.argmax(distances)))
            return [slots[i] for i in chosen]
        # Additional members use local interpolation of neighboring original vector
        # ribbons with identical command topology; no ellipse is generated.
        while len(slots) < count:
            slots.sort(key=lambda s: angle(s, side))
            points = np.asarray([slot['point'] for slot in slots])
            candidates = []
            boxes = np.asarray([slot['bbox'] for slot in slots])
            for i, a in enumerate(slots):
                for j in range(i + 1, len(slots)):
                    b = slots[j]
                    if topology(a) != topology(b) or abs(angle(a, side) - angle(b, side)) > .65:
                        continue
                    box = (np.asarray(a['bbox']) + np.asarray(b['bbox'])) / 2
                    dx = np.maximum(0, np.maximum(box[0] - boxes[:, 2], boxes[:, 0] - box[2]))
                    dy = np.maximum(0, np.maximum(box[1] - boxes[:, 3], boxes[:, 1] - box[3]))
                    ox = np.maximum(0, np.minimum(box[2], boxes[:, 2]) - np.maximum(box[0], boxes[:, 0]))
                    oy = np.maximum(0, np.minimum(box[3], boxes[:, 3]) - np.maximum(box[1], boxes[:, 1]))
                    # Insert into genuine local whitespace between neighboring
                    # source ribbons, rather than halfway through existing words.
                    score = np.hypot(dx, dy).min() - 100 * (ox * oy).sum()
                    candidates.append((float(score), i, j))
            if not candidates:
                raise ValueError('No compatible original ribbon pair for interpolation.')
            _, i, j = max(candidates)
            slots.append(blend(slots[i], slots[j]))
        return slots
    def mpl_path(commands):
        vertices, codes = [], []
        for step in commands:
            code = {'m': MplPath.MOVETO, 'l': MplPath.LINETO, 'c': MplPath.CURVE4}[step[0]]
            vertices.extend(step[1:]); codes.extend([code] * len(step[1:]))
        vertices.append(vertices[0]); codes.append(MplPath.CLOSEPOLY)
        return MplPath(xy(vertices), codes)
    stroke = [pe.withStroke(linewidth=halo_width_pt, foreground='white')]
    #2. =========== Original gray guide, ribbons and replacement gene labels
    for guide in template['guides']:
        points = xy(guide['pts'])
        line, = ax.plot(points[:, 0], points[:, 1], transform=ax.transAxes,
            color=guide['stroking_color'], linewidth=guide['linewidth'] * scale,
            solid_capstyle='butt', zorder=.2, clip_on=True)
        dash, phase = guide['dash']
        if dash:
            line.set_linestyle((phase / guide['linewidth'],
                                tuple(float(v) / guide['linewidth'] for v in dash)))
    factor_point = xy(template['center'])
    factor_text = ax.text(*factor_point, str(factor), transform=ax.transAxes,
        ha='center', va='center', fontfamily='Arial', fontstyle='italic',
        fontweight='normal', fontsize=font_size, color='black', path_effects=stroke,
        zorder=6, clip_on=False)
    records, patches, edge_records = [], [], []
    for side, genes in zip(['left', 'right'], groups):
        slots = choose_slots(side, len(genes))
        texts = [ax.text(.5, .5, gene, transform=ax.transAxes, ha='center', va='center',
            fontfamily='Arial', fontstyle='italic',
            fontsize=font_size * (template['red_font_size'] / template['base_font_size'] if gene in highlights else 1),
            color=highlight_color if gene in highlights else 'black', path_effects=stroke,
            zorder=5, clip_on=False) for gene in genes]
        fig.canvas.draw(); renderer = fig.canvas.get_renderer()
        # Match long replacement names to the genuinely wider original text
        # slots; this preserves the published geometry rather than repacking it.
        gene_widths = [TextPath((0, 0), gene, prop=FontProperties(family='Arial', style='italic', size=text.get_fontsize())).get_extents().width for gene, text in zip(genes, texts)]
        gene_order = sorted(range(len(genes)), key=lambda i: (gene_widths[i], genes[i]))
        slot_order = sorted(range(len(slots)), key=lambda i: (slots[i]['bbox'][2] - slots[i]['bbox'][0], i))
        for gi, si in zip(gene_order, slot_order):
            gene, text, slot = genes[gi], texts[gi], slots[si]
            shift = template.get('current_label_offsets', {}).get(str(factor), {}).get(gene, (0, 0))
            point = xy(slot['point']) + np.asarray(shift) / np.array([width, height])
            text.set_position(point)
            patch = PathPatch(mpl_path(slot['path']), transform=ax.transAxes,
                facecolor=left_color if side == 'left' else right_color, edgecolor='none', linewidth=0, alpha=1,
                zorder=1, clip_on=True)
            ax.add_patch(patch); patches.append(patch)
            records.append({'side': side, 'gene': gene, 'text': text, 'point': point})
            edge_records.append({'side': side, 'gene': gene,
                'visual_slot': slot['visual_slot'], 'template_interpolated': slot['interpolated'],
                'label_offset_points': list(shift)})
    #3. =========== Report actual layout and the fixed visual provenance
    fig.canvas.draw(); renderer = fig.canvas.get_renderer()
    named = [(str(factor), factor_text)] + [(r['side'] + ':' + r['gene'], r['text']) for r in records]
    boxes = [(name, text.get_window_extent(renderer)) for name, text in named]
    overlaps, clipped = [], []
    for i, (name, box) in enumerate(boxes):
        if box.x0 < area.x0 or box.y0 < area.y0 or box.x1 > area.x1 or box.y1 > area.y1:
            clipped.append(name)
        for other, b in boxes[i + 1:]:
            if min(box.x1, b.x1) > max(box.x0, b.x0) and min(box.y1, b.y1) > max(box.y0, b.y0):
                overlaps.append((name, other))
    positions = {'factor': tuple(map(float, factor_point)), 'left': {}, 'right': {}}
    for record in records:
        positions[record['side']][record['gene']] = tuple(map(float, record['point']))
    all_genes = set(groups[0]) | set(groups[1])
    reference = {'source': template['source'], 'path': template['path'],
        'sha256': template['sha256'], 'page_1based': template['page_1based'],
        'coordinate_system': 'Original PDF points, top-left origin; uniform affine transform',
        'original_slots': {'left': len(template['slots']['left']), 'right': len(template['slots']['right'])},
        'slot_mapping': edge_records,
        'slot_names_are_biological_targets': False,
        'runtime_pdf_or_Tmp_dependency': False,
        'local_text_offsets_points': template.get('current_label_offsets', {}).get(str(factor), {})}
    motif = str(kind).lower() in {'motif', 'motif_supported', 'early_motif_supported', 'regulon'}
    qa = {'ok': not (overlaps or clipped), 'factor': str(factor), 'kind': str(kind),
        'left_genes': len(groups[0]), 'right_genes': len(groups[1]),
        'gene_labels': len(records), 'edges': len(patches),
        'all_genes_connected': len(patches) == sum(map(len, groups)),
        'label_overlaps': overlaps, 'clipped_labels': clipped,
        'axes_width_inches': width / 72, 'axes_height_inches': height / 72,
        'font_size_points': float(font_size), 'factor_font_size_points': float(font_size),
        'highlighted_genes': sorted(highlights & all_genes),
        'requested_highlights_not_in_modules': sorted(highlights - all_genes),
        'highlight_semantics': 'Literature-discussed representatives; not statistical significance or stronger edges',
        'curve_geometry': 'Original filled Bezier ribbons; interpolation only for additional members',
        'paper_palette': {'left': '#2D7184', 'right': '#5AAD7F', 'red': highlight_color},
        'template_scale_points': float(scale),
        'reference_template': reference,
        'edge_semantics': 'Predicted motif-supported TF-target link' if motif else 'RNA module association'}
    return {'labeltexts': [text for _, text in named], 'edgepatches': patches,
        'positions': positions, 'factor_text': factor_text, 'edge_records': edge_records,
        'reference_template': reference, 'QA': qa}


# SGN-1 display only; append after the existing network helper cell.
# Red marks approved literature representatives, independent of RNA association sign.
NETWORK_HIGHLIGHTS = {
    "Foxo6": ("Nhlh1", "Sox11"),
    "Zfpm2": ("Isl1", "Nhlh1"),
    "Neurod2": ("Nhlh1", "Neurod6", "Isl1"),
    "Nhlh1": ("Nhlh2",),
    "Neurod1": ("Myt1l", "Nhlh2"),
}

# Reuse approved gene-level evidence; neither copy validates a new TF-target edge.
NETWORK_HIGHLIGHT_EVIDENCE["Zfpm2"]["Nhlh1"] = dict(
    NETWORK_HIGHLIGHT_EVIDENCE["Foxo6"]["Nhlh1"]
)
NETWORK_HIGHLIGHT_EVIDENCE["Neurod1"]["Nhlh2"] = dict(
    NETWORK_HIGHLIGHT_EVIDENCE["Nhlh1"]["Nhlh2"]
)
NETWORK_HIGHLIGHT_EVIDENCE = {
    factor: {gene: dict(NETWORK_HIGHLIGHT_EVIDENCE[factor][gene]) for gene in genes}
    for factor, genes in NETWORK_HIGHLIGHTS.items()
}

# SGN-1 local text-only offsets after changing displayed modules.
# Points: positive x is right; positive y is up. Paths, slots and fonts are unchanged.
PAPER_NETWORK_TEMPLATE["current_label_offsets"]['Nhlh1'].update({'Trp53i11': [-1.5, 3.1]})
PAPER_NETWORK_TEMPLATE["current_label_offsets"]['Neurod1'].update({'Cfl2': [0.001, 3.74], 'Btg1': [0.0, 0.338], 'Rtn1': [0.383, -3.064], 'Dner': [-5.286, 0.0], 'Sncg': [-0.556, 2.5], 'Mapt': [0.729, 0.0], 'Myt1l': [-0.972, 3.867], 'Chgb': [2.286, -4.683], 'Nav3': [-3.032, 0.0], 'Nsg2': [-0.568, -0.122], 'Trim2': [-3.122, -1.942], 'Fmn1': [1.562, 0.0], 'Scrn1': [-1.562, 0.0], 'Mllt11': [0.972, -2.5], 'C1ql3': [0.0, -1.359], 'Olfm2': [1.901, 0.677], 'Vdac1': [0.0, 1.106], 'Cxxc4': [-0.729, 0.836], 'Clec2l': [0.568, -1.087], 'Scn3b': [0.0, -0.677], 'Rims3': [0.0, 1.006], 'Eps15': [0.0, 0.718], 'Lancl1': [0.0, -0.239], 'Trim67': [0.379, 0.0], 'Vstm2l': [4.693, 0.0], 'Hpcal1': [0.0, -1.006], 'Ncam1': [3.032, 0.0], 'Cacng5': [1.814, 0.122], 'Akap12': [0.0, 0.608], 'Slc17a6': [-2.493, 1.454], 'S100a10': [0.0, 0.582], 'St18': [1.171, 0.0], 'Actc1': [0.113, 0.0], 'Insm1': [-0.113, 0.0], 'Fam110a': [-1.171, 0.0]})
```

```python
#3. =========== Original plotting settings and selected canvas
REFERENCE_W, REFERENCE_H, LOGICAL_PPI = 1458, 313, 75

DPI = 600

REFERENCE_SHA256 = '72566b7a48fdf5247f9e6e9a6b364c70b018ecbb39b5ef43b2fd80641d304526'

C_DISPLAY_TITLES = {
    'Foxo6': {0:'Transient developmental', 1:'Broad early-rising', 2:'Early transition-associated'},
    'Zfpm2': {0:'Neuronal morphogenesis', 1:'Broad early-rising', 2:'Transient developmental', 3:'Early-high (inverse)'},
    'Neurod2': {0:'Broad early-rising', 1:'Transient developmental'},
    'Nhlh1': {0:'Transient developmental', 1:'Weakly varying', 2:'Early transition-associated'},
    'Neurod1': {0:'Postsynaptic (inverse)', 1:'Mid-trajectory peak (inverse)', 2:'Late-rising Calb1/Kcna1',
                3:'Progressively rising', 4:'Early-high', 5:'Late-rising Cnr1/Kcnq2', 6:'Transient differentiation'}}

VIEW_WIDTH = VIEW_X1 - VIEW_X0
FIGSIZE = (VIEW_WIDTH / LOGICAL_PPI, np.nextafter(REFERENCE_H / LOGICAL_PPI, np.inf))

def reference_box(x, y, width, height):
    return [(x - VIEW_X0) / VIEW_WIDTH, 1 - (y + height) / REFERENCE_H,
            width / VIEW_WIDTH, height / REFERENCE_H]

def reference_text(fig, x, baseline_y, text, **kwargs):
    return fig.text((x - VIEW_X0) / VIEW_WIDTH, 1 - baseline_y / REFERENCE_H,
                    text, va='baseline', **kwargs)

def render_factor(result, factor, output_dir, seed=42):
    import matplotlib
    matplotlib.use('Agg')
    import matplotlib.pyplot as plt
    from matplotlib.colors import LinearSegmentedColormap, Normalize
    from matplotlib.lines import Line2D
    B_POINT_AREA = 1.5
    C_POINT_AREA = 3.1
    BACKGROUND_ALPHA = .20
    FOREGROUND_ALPHA = .90
    NETWORK_FONT = 11.4
    PALETTE = {'CVG': '#1A6D79', 'T': '#1A6D79', 'VGN': '#00008B',
        'AB': '#7A4E94', 'AC': '#3585BC', 'Ia': '#F37271',
        'Ib': '#44B9DE', 'Ic': '#97C8A8', 'II': '#E377C2'}
    LEFT_COLOR, RIGHT_COLOR = '#1A6D79', '#97C8A8'
    plt.rcParams.update({'font.family': 'Arial', 'font.size': 8, 'axes.linewidth': .65,
                         'xtick.direction': 'out', 'ytick.direction': 'out'})
    assert factor in ('Foxo6', 'Zfpm2', 'Neurod2', 'Nhlh1', 'Neurod1'), factor
    obs, xy = result['obs'], result['X_umap']
    modules = result['modules']
    chosen = result['selected_modules']
    titles = result['reviewed_annotations']['reviewed_program_titles_json']
    reviewed = list(titles.values()) if isinstance(titles, dict) else titles
    on_path = obs.seg.isin(('CVG', 'T', 'AC', 'Ic')).to_numpy()
    grid, grid_fit = result['fit_grid'], result['grid_fitted']
    chosen = list(map(int, chosen))  # Fixed user order, never peak-sorted.
    assert len(chosen) == 2
    names = {}
    for m in chosen:
        gene_hash = hashlib.sha256('\n'.join(sorted(modules[m]['genes'])).encode()).hexdigest()
        found = [d for d in reviewed if d['TF'] == factor and d['gene_sha256'] == gene_hash]
        assert len(found) == 1, 'Review descriptive titles after changing gene sets'
        names[m] = found[0]['title']
    fig = plt.figure(figsize=FIGSIZE)
    text_items, curve_axes = [], []
    cm_b = LinearSegmentedColormap.from_list('activity', ['#D4D4D4', '#FFFFFF', '#F3151D'])
    activity = result['activity']
    status = str(activity['status'])
    if status not in ('available', 'not_estimable'):
        raise ValueError('Unknown activity status: ' + status)
    bscore = np.asarray(activity['fitted'], dtype=float)
    bmask = np.asarray(activity['display_mask'])
    assert bscore.shape == (len(obs),) and bmask.shape == (len(obs),)
    assert bmask.dtype.kind == 'b', 'Activity display_mask must be boolean'
    reason = str(activity['reason'])
    btitle = r'$\it{' + factor + '}$ (+) activity' if status == 'available' else r'$\it{' + factor + '}$ activity'
    if status == 'available':
        bmax = float(activity['display_vmax'])
        assert np.isfinite(bmax) and bmax > 0
        assert bmask.any() and np.isfinite(bscore[bmask]).all()
        metric = 'Regulon activity (AUC)'
    else:
        assert activity['display_vmax'] is None, 'Unavailable activity has no AUC color scale'
        assert not bmask.any(), 'Unavailable activity cannot display scored cells'
        assert np.isnan(bscore).all(), 'Unavailable activity must stay NaN, not zero'
        bmax, metric = None, 'Activity not estimable'
    lim = (xy.min(axis=0) - 1, xy.max(axis=0) + 1)
    def embedding(ax, values, cmap, vmax, mask=None):
        ax.scatter(xy[:, 0], xy[:, 1], s=B_POINT_AREA, c='#D4D4D4', alpha=.55, linewidths=0)
        sc = None
        if values is not None:
            mask = np.isfinite(values) if mask is None else mask & np.isfinite(values)
            order = np.flatnonzero(mask)[np.argsort(values[mask], kind='stable')]
            sc = ax.scatter(xy[order, 0], xy[order, 1], s=B_POINT_AREA, c=values[order],
                cmap=cmap, norm=Normalize(0, vmax, clip=True), alpha=.88, linewidths=0)
        ax.set_xlim(lim[0][0], lim[1][0]); ax.set_ylim(lim[0][1], lim[1][1])
        ax.set_aspect('equal'); ax.axis('off')
        return sc
    rng = np.random.default_rng(seed)
    background = rng.permutation(len(obs)); foreground = rng.permutation(np.flatnonzero(result['primary_path']))
    colors = np.array([PALETTE[s] for s in obs.seg])
    if not chosen:
        for i in range(2):
            ax = fig.add_axes(reference_box(318,51 if i == 0 else 179,220,103))
            curve_axes.append(ax)
            ax.set_xlim(0, 6.12); ax.set_ylim(0, .90)
            ax.set_xticks([0, 2, 4, 6]); ax.set_yticks([0, .45, .90])
            ax.tick_params(labelsize=7.68, length=2.5, width=.6, pad=2)
            text_items.append(ax.set_title('Program unavailable', fontsize=8.0, pad=4))
            text_items.append(ax.text(.5, .5, '\u2014', transform=ax.transAxes,
                                      ha='center', va='center', color='#999999', fontsize=16))
            if i == 0:
                ax.tick_params(labelbottom=False)
            else:
                text_items.append(ax.set_xlabel('Pseudotime', labelpad=2, fontsize=11.52))
    for i, m in enumerate(chosen):
        ax = fig.add_axes(reference_box(318,51 if i == 0 else 179,220,103))
        curve_axes.append(ax)
        values = result['module_scores'][:, m]
        ax.scatter(obs.t.to_numpy()[background], values[background], s=C_POINT_AREA,
                   c=colors[background], alpha=BACKGROUND_ALPHA, linewidths=0)
        ax.scatter(obs.t.to_numpy()[foreground], values[foreground], s=C_POINT_AREA,
                   c=colors[foreground], alpha=FOREGROUND_ALPHA, linewidths=0)
        ax.plot(grid, grid_fit[:, modules[m]['job']], lw=1.05, color='black', zorder=5)
        if i == 0:
            ax.axvline(result['fork_t'], lw=.65, ls=(0, (4, 3)), color='black')
            ax.tick_params(labelbottom=False)
        else:
            text_items.append(ax.set_xlabel('Pseudotime', labelpad=2, fontsize=11.52))
        ax.set_xlim(0, 6.12); ax.set_ylim(0, .90)
        ax.set_xticks([0, 2, 4, 6]); ax.set_yticks([0, .45, .90])
        ax.tick_params(labelsize=7.68, length=2.5, width=.6, pad=2)
        title = f'{C_DISPLAY_TITLES[factor][m]}  (n={modules[m]["n_genes"]})'
        text_items.append(reference_text(fig,428,46 if i == 0 else 174,title,fontsize=12.48,ha='center'))
    # The reference omits a vertical RNA-axis title; units remain in the archive.
    network = None
    spatial_limits = []
    if len(chosen) == 2:
        # The two spatial maps show precisely the same modules as the two C panels.
        for i, (m, color, x) in enumerate(zip(chosen, (LEFT_COLOR, RIGHT_COLOR), (586,1258))):
            dcol = int(np.flatnonzero(result['D_jobs'] == modules[m]['job'])[0])
            dscore = result['D_fitted'][:, dcol]
            dmax = math.ceil(np.nanmax(dscore[result['good']]) * 20) / 20
            spatial_limits.append(dmax)
            np.testing.assert_array_equal(dscore[on_path], result['fitted'][on_path, modules[m]['job']])
            dax = fig.add_axes(reference_box(x,91,198,155.3))
            cmap = LinearSegmentedColormap.from_list('program' + str(i), ['#D4D4D4', '#FFFFFF', color])
            scd = embedding(dax, dscore, cmap, dmax, mask=result['good'])
            barax = fig.add_axes(reference_box(636 if i == 0 else 1308,272,118,7))
            cbar = fig.colorbar(scd, cax=barax, orientation='horizontal')
            cbar.set_ticks([0, dmax]); cbar.ax.tick_params(labelsize=11.52, length=2, pad=1); cbar.outline.set_linewidth(.4)
            cbar.solids.set_edgecolor('face'); cbar.solids.set_linewidth(.12)
            text_items.extend(barax.get_xticklabels())
        axn = fig.add_axes(reference_box(780,62,480,224))
        network = draw_module_network(axn, factor, modules[chosen[0]]['genes'], modules[chosen[1]]['genes'],
            kind='motif' if factor == 'Nhlh1' else 'associated', left_color='#2D7184',
            right_color='#5AAD7F', font_size=NETWORK_FONT,
            highlight_genes=NETWORK_HIGHLIGHTS.get(factor, ()))
        text_items.extend(network['labeltexts'])
        left_title = 'Postsynaptic receptor regulation (inverse)' if factor == 'Neurod1' and chosen[0] == 0 else names[chosen[0]]
        text_items.append(reference_text(fig,783,40,left_title,fontsize=15.36,color='#00585D',ha='center'))
        text_items.append(reference_text(fig,1228,40,names[chosen[1]],fontsize=15.36,color='#006600',ha='center'))
        # Edge evidence is recorded in the notebook/archive, as in the supplied layout.
    else:
        text_items.append(fig.text(.67, .52, 'No supported early program', ha='center', fontsize=12, color='#777777'))
    for x, letter in ((x, letter) for x, letter in ((16, 'B'), (284, 'C'), (548, 'D')) if letter in PANEL_PARTS):
        text_items.append(reference_text(fig,x,34,letter,fontsize=34.56,fontweight='normal'))
    legend_labels = {'CVG':'CVG','VGN':'VGN*','T':'SGN trunk','AC':'Ib/Ic shared','AB':'Ia/II shared',
                     'Ia':'Ia','Ib':'Ib','Ic':'Ic','II':'II'}
    legend_records = []
    for segment, y in zip(('CVG','VGN','T','AC','AB','Ia','Ib','Ic','II'),(54,67,80,94,107,121,134,147.5,161)):
        label = legend_labels[segment]
        text = fig.text((291-VIEW_X0)/VIEW_WIDTH,1-y/REFERENCE_H,label,ha='right',va='center',fontsize=12.48)
        text_items.append(text)
        fig.canvas.draw()
        logical_width = text.get_window_extent(fig.canvas.get_renderer()).width * LOGICAL_PPI / fig.dpi
        marker_x = 291 - logical_width - 8
        fig.add_artist(Line2D([(marker_x-VIEW_X0)/VIEW_WIDTH],[1-y/REFERENCE_H],transform=fig.transFigure,
            marker='o',linestyle='None',color=PALETTE[segment],markersize=7.68,markeredgewidth=0))
        legend_records.append({'segment':segment,'label':label,'marker_xy':[marker_x,y],'text_right_x':291})
    for ax in curve_axes:
        text_items.extend(ax.get_xticklabels()); text_items.extend(ax.get_yticklabels())
    fig.canvas.draw()
    renderer = fig.canvas.get_renderer()
    clipped = [t.get_text() for t in text_items if t.get_visible() and t.get_text() and
        (not fig.bbox.contains(*t.get_window_extent(renderer).p0) or not fig.bbox.contains(*t.get_window_extent(renderer).p1))]
    assert not clipped, clipped
    assert network is None or network['QA']['ok'], network['QA']
    output_dir = Path(output_dir); output_dir.mkdir(parents=True, exist_ok=True)
    path = PNG_FILE
    fig.savefig(path, dpi=DPI)
    pdf_path = PDF_FILE
    pdf_info = strict_editable_pdf(fig, pdf_path, data_axes=curve_axes)
    plt.close(fig)
    result['figure'] = {'filename': path.name, 'filename_png': path.name, 'filename_pdf': pdf_path.name,
        'pdf_export': pdf_info, 'dpi': DPI, 'figsize_in': FIGSIZE,
        'B_activity_status': status, 'B_activity_reason': reason,
        'B_data': ('Motif-supported regulon AUCell GAM; no RNA or associated-program substitution' if status == 'available' else 'Unavailable activity; gray embedding only; no activity values plotted'),
        'B_display_mask_cells': int(bmask.sum()), 'B_display_vmax': bmax,
        'C_D_module_ids': chosen, 'C_D_comparison_applicable': len(chosen) == 2,
        'C_D_gene_sets_equal': True if len(chosen) == 2 else None,
        'C_D_primary_fitted_equal': True if len(chosen) == 2 else None,
        'title_records': {str(m): names[m] for m in chosen}, 'clipped_texts': clipped,
        'network': None if network is None else {k: v for k, v in network.items() if k not in ('labeltexts', 'edgepatches', 'factor_text')},
        'palette': PALETTE, 'C_background_alpha': BACKGROUND_ALPHA, 'C_foreground_alpha': FOREGROUND_ALPHA,
        'VGN_asterisk': 'Candidate VGN tree arm, distinct from independently curated VGN reference',
        'D_colorbar_maxima': spatial_limits,
        'C_display_titles': {str(m):C_DISPLAY_TITLES[factor][m] for m in chosen},
        'reference_layout': {'width':REFERENCE_W,'height':REFERENCE_H,'logical_ppi':LOGICAL_PPI,'sha256':REFERENCE_SHA256,
            'C_axes_xywh':[[318,51,220,103],[318,179,220,103]],'network_xywh':[780,62,480,224],
            'legend':legend_records,'font_sizes_pt':{'panel':34.56,'B_D_title':15.36,'C_title':12.48,'C_tick':7.68,'network':NETWORK_FONT}},
        'highlight_evidence': NETWORK_HIGHLIGHT_EVIDENCE.get(factor, {}),
        'highlight_semantics': 'Literature-discussed representatives; red is not significance, edge strength, direct binding, or fate determination.',
        'spatial_values': 'Module fitted RNA, not TF RNA or AUCell. Each colorbar maximum is the actual fitted maximum rounded up to0.05; stored fitted values are not clipped. Compare spatial patterns within a module, not saturation across modules.'}
    audit = result['association_audit'].set_index('gene')
    rows = []
    for side, m in enumerate(chosen):
        for gene in modules[m]['genes']:
            rho = float(audit.loc[gene, 'mean_library_rho'])
            rows.append({'factor': factor, 'gene': gene, 'module_id': m, 'side': 'left' if side == 0 else 'right',
                'program_title': names[m], 'evidence': 'motif-supported prediction' if factor == 'Nhlh1' else 'RNA association',
                'mean_library_RNA_rho': rho,
                'RNA_association_sign': 'not encoded for motif targets' if factor == 'Nhlh1' else ('positive' if rho > 0 else 'negative'),
                'arrowhead': False})
    result['network_edges'] = pd.DataFrame(rows)
    assert len(rows) == sum(modules[m]['n_genes'] for m in chosen)
    return path
```

```python
APPROVED_SCORE_SOURCES=('GSE178931','GSE231513','GSE165502')
EXCLUDED_SCORE_SOURCES=('GSE195500',)

def _paper_plain(x):
    if isinstance(x,bytes):return x.decode('utf-8')
    if isinstance(x,np.generic):return _paper_plain(x.item())
    if isinstance(x,np.ndarray):return [_paper_plain(v) for v in x.tolist()]
    if isinstance(x,dict):return {str(k):_paper_plain(v) for k,v in x.items()}
    if isinstance(x,(tuple,list)):return [_paper_plain(v) for v in x]
    if isinstance(x,Path):return str(x)
    return x

def _paper_json(x):return json.dumps(_paper_plain(x),sort_keys=True,ensure_ascii=False,allow_nan=True)

def _paper_get(dataset):
    if dataset.dtype.kind in 'OSU':v=dataset.asstr()[()]
    else:
        v=dataset[()]
        if isinstance(v,np.ndarray) and v.dtype==np.uint8:v=v.tobytes().decode('utf-8')
    return json.loads(v)

def _paper_nodes(file,prefix='',ignore=()):
    out={}
    with h5py.File(file,'r') as h:
        start=h[prefix] if prefix else h
        def visit(name,o):
            if name=='integrity_json' or any(name==v or name.startswith(v+'/') for v in ignore):return
            d=hashlib.sha256();d.update(_paper_json(dict(o.attrs)).encode())
            if isinstance(o,h5py.Dataset):
                d.update(str(o.dtype).encode());d.update(str(o.shape).encode())
                if o.dtype.kind in 'OSU':d.update(_paper_json(o.asstr()[()]).encode())
                elif not o.shape:d.update(np.asarray(o[()]).tobytes())
                else:
                    step=max(1,1048576//max(1,int(np.prod(o.shape[1:]))*o.dtype.itemsize))
                    for i in range(0,len(o),step):d.update(np.ascontiguousarray(o[i:i+step]).tobytes())
            out[name]=d.hexdigest()
        visit('/',start);start.visititems(visit)
    return out

def _paper_scope_contract(methods,rank_genes):
    assert methods['score_sources']==list(APPROVED_SCORE_SOURCES) and methods['excluded_sources']==list(EXCLUDED_SCORE_SOURCES)
    assert methods['signature_full_genes']==191 and methods['signature_trimmed'] is False
    assert methods['min_full_signature_coverage']==.8 and methods['original_full_four_source_gate_failed'] is True
    coverage=methods['coverage'];full=coverage['four11471'];umi=coverage['GSE195500'];common=coverage['eligible_common']
    assert full['full']==umi['full']==common['full']==191
    assert full['matched']==133 and full['coverage']<.8 and umi['matched']==140 and umi['coverage']<.8
    assert common['sources']==list(APPROVED_SCORE_SOURCES) and common['rank_genes']==rank_genes and common['coverage']>=.8
    assert all(coverage[source]['coverage']>=.8 for source in APPROVED_SCORE_SOURCES)
    assert common['matched']==179 and methods['rank_genes']==rank_genes
    assert methods['pyscenic_version']=='0.11.2' and methods['noweights'] is True and methods['normalize'] is False
    assert methods['auc_threshold']==.05 and methods['seed']==42
    return dict(original_full_source_failed=True,four_source=full,excluded_UMI=umi,eligible_common=common,eligible_sources=list(APPROVED_SCORE_SOURCES),excluded_sources=list(EXCLUDED_SCORE_SOURCES),canonical_signature_targets=191,threshold=.8,signature_trimmed=False)

def _paper_check_project_fit(file):
    with h5py.File(file,'r') as archive:
        h = archive['B_paper/project/fit']
        assert h.attrs.get('complete') and not bool(h.attrs.get('test_fixture',False))
        assert list(h['TFs'].asstr()[:])==['Neurod1(+)'] and len(h['cell_ids'])==7355
        valid=h['valid_fit'][:].reshape(-1);fit=h['fitted'][:]
        assert fit.shape==(7355,1) and valid.dtype.kind=='b' and valid.any()
        assert np.isfinite(fit[valid]).all() and np.isnan(fit[~valid]).all()
        ref=h['display_reference_mask'][:].astype(bool);eligible=h['display_eligible_reference_mask'][:].astype(bool)
        assert ref.sum()==975;np.testing.assert_array_equal(eligible,ref&valid)
        assert eligible.any() and float(h['display_vmax'][()])==float(fit[eligible,0].max()) and float(h['display_vmin'][()])==float(fit[valid,0].min())
        methods=_paper_get(h['methods_json']);assert methods['full_signature_preserved'] is True
        _paper_scope_contract(methods['score_methods'],len(h['genes']))
        scope=h['source_scope'][:];assert scope.dtype.kind=='b' and not (valid&~scope).any()
        for key in ('raw_auc','gated_activity','fitted'):assert np.isnan(h[key][:][~scope]).all()
        assert methods['display_reference']['policy']=='reference_maturation_exact_ID_mapping' and methods['project_maturation_knee'] is None
        assert methods['score_methods']['noweights'] is True and methods['score_methods']['normalize'] is False
    return methods
```

```python
#7. =========== Original paper-reference activity renderer
PROJECT_FIGSIZE=(4.0,343/75)

PROJECT_DPI=600

PROJECT_POINT_AREA=1.5

PROJECT_POINT_ALPHA=.88

PROJECT_TITLE_SIZE=15.36

PROJECT_METRIC_SIZE=11.52

PROJECT_PALETTE={'CVG':'#1A6D79','T':'#1A6D79','VGN':'#00008B','AB':'#7A4E94','AC':'#3585BC','Ia':'#F37271','Ib':'#44B9DE','Ic':'#97C8A8','II':'#E377C2'}

PROJECT_NATIVE_ORDER=('CVG','VGN','T','AC','AB','Ia','Ib','Ic','II')

PROJECT_NATIVE_LABELS={'CVG':'CVG','VGN':'VGN*','T':'SGN trunk','AC':'Ib/Ic shared','AB':'Ia/II shared','Ia':'Ia','Ib':'Ib','Ic':'Ic','II':'II'}

PROJECT_PAPER_COLORS=('#D4D4D4','#FFFFFF','#F3151D')

PROJECT_CMAP_SOURCE=dict(file='make_fig3.ipynb',definition_cell='Cell4 in author make_fig3.py export; zero-based notebook cell3',RGB=[[212,212,212],[255,255,255],[243,21,29]],N=256,used_by='Cell9 final fig3-B export; zero-based notebook cell8',earlier_generic_cmap='lightgrey/white/red is overridden before final B')

def plain(value):
    if isinstance(value,np.generic):return value.item()
    if isinstance(value,np.ndarray):return value.tolist()
    if isinstance(value,Path):return str(value)
    if isinstance(value,dict):return {str(k):plain(v) for k,v in value.items()}
    if isinstance(value,(tuple,list)):return [plain(v) for v in value]
    return value

def sha(path):
    h=hashlib.sha256()
    with open(path,'rb') as f:
        for b in iter(lambda:f.read(1048576),b''):h.update(b)
    return h.hexdigest()

def decode(node):
    if isinstance(node,h5py.Dataset):return node.asstr()[()] if node.dtype.kind in 'OSU' else node[()]
    if 'categories' in node and 'codes' in node:
        c=decode(node['categories']);ix=node['codes'][:];return np.asarray([c[i] if i>=0 else '' for i in ix])
    return {k:decode(v) for k,v in node.items()}

def verify_input(fit_file,shared_file,*,pilot=False):
    with h5py.File(fit_file,'r') as archive:
        h = archive['B_paper/project/fit']
        assert h.attrs.get('complete'), 'Only a complete project fit is renderable'
        assert not bool(h.attrs.get('test_fixture',False)) and bool(h.attrs.get('scientific_Neurod1_result',True)),'A test fixture cannot be labelled as a Neurod1 result'
        if 'TFs' in h:
            assert list(h['TFs'].asstr()[:])==['Neurod1(+)'],'Project activity must be exactly Neurod1(+)'
        else:assert pilot,'A production fit must declare the exact TFs axis'
        cells=h['cell_ids'].asstr()[:];values=h['fitted'][:];valid=h['valid_fit'][:];xy=h['X_umap'][:]
        assert len(cells)==7355 and len(set(cells))==7355 and values.shape==(7355,1) and xy.shape==(7355,2)
        assert valid.shape in ((7355,),(7355,1)) and valid.dtype.kind=='b','valid_fit must be an explicit boolean mask'
        valid=valid.reshape(-1);values=values[:,0]
        assert np.isfinite(xy).all() and valid.any()
        assert np.isfinite(values[valid]).all() and np.isnan(values[~valid]).all(),'Unestimated cells must remain NaN'
        seg=decode(h['seg']) if 'seg' in h else (decode(h['obs/seg']) if 'obs/seg' in h else None)
        scope=h['source_scope'][:] if 'source_scope' in h else None
        if not pilot:
            assert scope is not None and scope.shape==(7355,) and scope.dtype.kind=='b'
            assert not (valid&~scope).any()
            for key in ('raw_auc','gated_activity','fitted'):assert np.isnan(h[key][:][~scope]).all(),'Unestimated UMI activity must remain NaN'
    with h5py.File(shared_file,'r') as h:
        np.testing.assert_array_equal(cells,h['shared/obs/_index'].asstr()[:])
        np.testing.assert_array_equal(xy,h['shared/X_umap'][:],err_msg='Original project UMAP coordinates changed')
        native=np.asarray(decode(h['shared/obs/seg'])).astype(str)
        if not pilot:np.testing.assert_array_equal(scope,np.isin(h['shared/obs/study_id'].asstr()[:],['GSE178931','GSE231513','GSE165502']))
    if seg is not None:np.testing.assert_array_equal(np.asarray(seg).astype(str),native,err_msg='Original native labels changed')
    assert len(native)==7355 and set(native).issubset(PROJECT_NATIVE_LABELS)
    return cells,values,valid,xy,native

def render_project(fit_file,output_dir,*,vmax,cap_source,shared_file=None,work_dir=None,file_stem='Neurod1-B_1',vmin=None,validate=True,qa_overlay=False,pilot=False):
    """Render unchanged project UMAP. vmax and cap_source have no defaults.
    vmin follows scvelo's data-minimum default unless the caller explicitly
    supplies it. No subgroup or Ic selection occurs anywhere in this renderer.
    """
    fit_file,output_dir=map(Path,[fit_file,output_dir]);root=ROOT
    shared_file=Path(shared_file or RESULT_FILE)
    work=Path(work_dir or PAPER_WORK_DIR).resolve()
    output_dir=output_dir.resolve();work.mkdir(parents=True,exist_ok=True);output_dir.mkdir(parents=True,exist_ok=True)
    assert work!=output_dir and not work.is_relative_to(output_dir),'Validation/runtime must stay outside the figure output directory'
    assert isinstance(cap_source,(str,dict)) and bool(cap_source) and (not isinstance(cap_source,str) or cap_source.strip()),'Explicit reviewed cap provenance is required'
    vmax=float(vmax);assert np.isfinite(vmax),'Explicit finite vmax is required; this function never estimates a cap'
    # A project fit may archive its predeclared external-reference display cap.
    # Verify the caller's explicit cap; never derive a new threshold here.
    with h5py.File(fit_file,'r') as archive:
        input_fit = archive['B_paper/project/fit']
        if 'display_vmax' in input_fit:
            assert vmax==float(input_fit['display_vmax'][()]),'Explicit cap differs from the fit archive predeclared reference cap'
        scope_count=int(input_fit['source_scope'][:].sum()) if 'source_scope' in input_fit else None
        scope_methods=json.loads(input_fit['methods_json'].asstr()[()]) if 'methods_json' in input_fit else {}
        scope_methods=scope_methods.get('score_methods',scope_methods)
    os.environ['MPLCONFIGDIR']=str(PAPER_WORK_DIR)
    cells,values,valid,xy,native=verify_input(fit_file,shared_file,pilot=pilot)
    lo_value=float(values[valid].min());vmin=lo_value if vmin is None else float(vmin)
    assert np.isfinite(vmin) and vmax>vmin
    import matplotlib
    matplotlib.use('Agg')
    import matplotlib.pyplot as plt
    from matplotlib.lines import Line2D
    from matplotlib.colors import LinearSegmentedColormap,Normalize
    w,h=300,343
    def box(x,y,ww,hh):return [x/w,1-(y+hh)/h,ww/w,hh/h]
    def text(fig,x,y,label,**kw):
        kw.setdefault('va','baseline');return fig.text(x/w,1-y/h,label,**kw)
    cmap=LinearSegmentedColormap.from_list('paper_final_B',PROJECT_PAPER_COLORS,N=256)
    norm=Normalize(vmin=vmin,vmax=vmax,clip=True)
    with plt.rc_context({'font.family':'Arial','font.size':8,'axes.linewidth':.65}):
        fig=plt.figure(figsize=PROJECT_FIGSIZE);ax=fig.add_axes(box(19,52,262,206))
        ax.scatter(xy[~valid,0],xy[~valid,1],s=PROJECT_POINT_AREA,c='#D4D4D4',alpha=.45,linewidths=0)
        order=np.flatnonzero(valid)[np.argsort(values[valid],kind='stable')]
        sc=ax.scatter(xy[order,0],xy[order,1],s=PROJECT_POINT_AREA,c=values[order],cmap=cmap,norm=norm,alpha=PROJECT_POINT_ALPHA,linewidths=0)
        lower=xy.min(0)-1;upper=xy.max(0)+1;ax.set_xlim(lower[0],upper[0]);ax.set_ylim(lower[1],upper[1]);ax.set_aspect('equal');ax.axis('off')
        title=r'$\it{Neurod1}$(+) activity'+(' (PILOT)' if pilot else '')
        labels=[text(fig,150,40,title,ha='center',fontsize=PROJECT_TITLE_SIZE)]
        # Native colors describe C/annotation QA only; the B activity map uses one continuous scale.
        labels.append(text(fig,150,271,'Regulon activity (AUC)',ha='center',fontsize=PROJECT_METRIC_SIZE))
        cax=fig.add_axes(box(91,278,118,8));cb=fig.colorbar(sc,cax=cax,orientation='horizontal')
        cb.set_ticks([vmin,vmax]);cb.set_ticklabels([f'{vmin:.4g}',f'> {vmax:.4g}']);cb.ax.tick_params(labelsize=11.52,length=2,pad=1);cb.outline.set_linewidth(.4)
        cb.solids.set_edgecolor('face');cb.solids.set_linewidth(.12);labels.extend(cax.get_xticklabels())
        labels.append(text(fig,150,318,'PILOT: activity layout validation' if pilot else 'RNA-gated regulon activity',ha='center',fontsize=7.2))
        labels.append(text(fig,150,332,'PILOT: layout only; not the new paper-method result' if pilot else 'Early + SS2 scored; UMI not estimable',ha='center',fontsize=6.9,color='#B00020' if pilot else 'black'))
        fig.canvas.draw();renderer=fig.canvas.get_renderer()
        clipped=[t.get_text() for t in labels if not (fig.bbox.contains(*t.get_window_extent(renderer).p0) and fig.bbox.contains(*t.get_window_extent(renderer).p1))];assert not clipped,clipped
        stem=file_stem+('-PILOT' if pilot else '');png=PROJECT_PNG_FILE;pdf=PROJECT_PDF_FILE
        fig.savefig(png,dpi=PROJECT_DPI)
        pdf_info=strict_editable_pdf(fig,pdf,data_axes=());plt.close(fig)
    overlay=None
    info=dict(PNG=str(png),PDF=str(pdf),PNG_sha256=sha(png),PDF_sha256=sha(pdf),pixels=[round(PROJECT_FIGSIZE[0]*PROJECT_DPI),round(PROJECT_FIGSIZE[1]*PROJECT_DPI)],dpi=PROJECT_DPI,
        cohort_cells=7355,source_scope_cells=scope_count,score_sources=scope_methods.get('score_sources'),excluded_sources=scope_methods.get('excluded_sources'),source_coverage_audit=scope_methods.get('coverage'),partial_projection=True,unscored_activity_is_NaN_not_zero=True,valid_fit=int(valid.sum()),not_estimated=int((~valid).sum()),native_labels_unchanged=True,original_UMAP_exact=True,
        title='Neurod1(+) activity',pilot=pilot,vmin=vmin,vmin_rule='minimum of supplied valid fitted values' if vmin==lo_value else 'explicit caller-supplied vmin',vmax=vmax,cap_source=plain(cap_source),
        cap_was_estimated_by_renderer=False,display_saturated=int((values[valid]>vmax).sum()),no_Ic_specific_threshold_or_order=True,
        actual_fitted_values_unchanged=True,cmap=list(PROJECT_PAPER_COLORS),cmap_source=PROJECT_CMAP_SOURCE,point_area=PROJECT_POINT_AREA,alpha=PROJECT_POINT_ALPHA,
        native_legend_displayed=False,native_palette=PROJECT_PALETTE,native_display_labels=PROJECT_NATIVE_LABELS,PDF_export=pdf_info,clipped_labels=clipped,QA_PNG=str(overlay) if overlay else None,
        input_fit_sha256=sha(fit_file),original_shared_sha256=sha(shared_file),worker_source_sha256=PROJECT_RENDER_SOURCE_SHA256,validation_work_dir=str(work))
    print(json.dumps({k:info[k] for k in ('PNG','PDF','valid_fit','not_estimated','vmin','vmax','cap_source','validation') if k in info},ensure_ascii=False))
    return info
```

```python
#8. =========== Export the replacement Neurod1 activity view
paper_component = render_project(
    ACTIVITY_FILE, PROJECT_OUTPUT_DIR,
    vmax=paper_vmax, vmin=paper_vmin,
    cap_source=paper_fit_methods['display_reference'],
    shared_file=RESULT_FILE, work_dir=PAPER_WORK_DIR,
    file_stem='Neurod1-B_1', qa_overlay=False, pilot=False, validate=False
)
```
